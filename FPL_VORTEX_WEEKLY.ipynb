```python
# ============================================================
# CELL 0 — FPL VORTEX WEEKLY MANUAL CONTROLS
# ============================================================
# Edit this form, then choose Runtime → Run all.
# WEEKLY is standalone and does not read DO_NOT_DELETE.ipynb.

# @title 🎛️ FPL VORTEX WEEKLY CONTROLS — set options, then Run all { display-mode: "form" }

# @markdown ### 👤 FPL ENTRY
# @markdown **FPL Team ID — your FPL entry/team number**
FPL_TEAM_ID = 145784  # @param {type:"integer"}

# Compatibility aliases used by downstream GW Review cells.
TEAM_ID = int(FPL_TEAM_ID)
ENTRY_ID = int(FPL_TEAM_ID)

# @markdown ---
# @markdown ### 🔌 DATA SOURCES
# @markdown **Raw market-odds benchmark capture — stored separately; NEVER blended into V1**
ENABLE_MARKET_ODDS = True  # @param {type:"boolean"}


# @markdown **Free player artwork fallback — TheSportsDB V1 public API, no secret required**
ENABLE_PLAYER_IMAGE_FALLBACK = True  # @param {type:"boolean"}

# @markdown **Optional: comma-separated player names whose FPL photo is old/wrong-kit — force free fallback**
FORCE_IMAGE_FALLBACK_PLAYERS = "Joao Pedro, "  # @param {type:"string"}

# @markdown ---
# @markdown ### 🎬 MP4 OUTPUT
# @markdown **DRAFT = 854×480 at 5 FPS • FINAL = 3840×2160 at 30 FPS**
MP4_QUALITY = "FINAL"  # @param ["DRAFT", "FINAL"]

if MP4_QUALITY == "DRAFT":
    VIDEO_WIDTH, VIDEO_HEIGHT, VIDEO_FPS = 854, 480, 5
else:
    VIDEO_WIDTH, VIDEO_HEIGHT, VIDEO_FPS = 3840, 2160, 30

# @markdown ---
# @markdown ### 🎞️ SLIDES TO GENERATE
# @markdown **Uncheck any trusted slide you do not want rendered into MP4.**
SLIDE_01_INTRO = True  # @param {type:"boolean"}
SLIDE_02_GW_REVIEW = True  # @param {type:"boolean"}
SLIDE_03_FDR = True  # @param {type:"boolean"}
SLIDE_04_PROJECTED_GOALS = True  # @param {type:"boolean"}
SLIDE_05_CLEAN_SHEETS = True  # @param {type:"boolean"}
SLIDE_06_OUTRO = True  # @param {type:"boolean"}

SLIDE_SELECTION = {
    "00_intro": SLIDE_01_INTRO,
    "gw_review": SLIDE_02_GW_REVIEW,
    "02_fdr": SLIDE_03_FDR,
    "03_projected_goals": SLIDE_04_PROJECTED_GOALS,
    "04_clean_sheets": SLIDE_05_CLEAN_SHEETS,
    "05_outro": SLIDE_06_OUTRO,
}

SELECTED_SCENES = [
    key for key, enabled in SLIDE_SELECTION.items()
    if enabled
]

if not SELECTED_SCENES:
    raise ValueError("Select at least one slide in Cell 0.")

# ============================================================
# STARTUP + GOOGLE DRIVE + ONE CLEAN OUTPUT TREE
# ============================================================

from pathlib import Path
import os
import shutil
import time

RUNNING_IN_GITHUB = bool(os.getenv("GITHUB_ACTIONS"))
RUNNING_IN_COLAB = False

if RUNNING_IN_GITHUB:
    DRIVE_MOUNT = Path(os.environ.get("FPL_VORTEX_LOCAL_ROOT", "output")).resolve()
    MY_DRIVE = DRIVE_MOUNT / "MyDrive"
    MY_DRIVE.mkdir(parents=True, exist_ok=True)
else:
    try:
        from google.colab import drive
        RUNNING_IN_COLAB = True
    except Exception as exc:
        raise RuntimeError(
            "FPL VORTEX requires either Google Colab or GitHub Actions execution."
        ) from exc
    DRIVE_MOUNT = Path("/content/drive")
    MY_DRIVE = DRIVE_MOUNT / "MyDrive"

FPL_VORTEX_DATA_ROOT = Path(
    os.environ.get("FPL_VORTEX_DATA_ROOT", str(MY_DRIVE / "FPL_VORTEX_DATA"))
)
ELEMENTS_ROOT = Path(
    os.environ.get("FPL_VORTEX_ELEMENTS_ROOT", str(MY_DRIVE / "elements"))
)


def vx_prepare_output_tree():
    """Safely mount/reuse Drive and rebuild only FIRST VIDEO."""
    if RUNNING_IN_COLAB and not MY_DRIVE.is_dir():
        try:
            drive.mount(
                str(DRIVE_MOUNT),
                force_remount=False,
                timeout_ms=300_000,
            )
        except Exception as exc:
            if not MY_DRIVE.is_dir():
                msg = str(exc)
                if "credential propagation" in msg.lower():
                    raise RuntimeError(
                        "Google Colab could not pass credentials to Drive. "
                        "No FPL VORTEX output was deleted. Reconnect a hosted "
                        "Colab runtime, authorize Drive, then run all again."
                    ) from None
                raise RuntimeError(
                    "Google Drive mount failed. No FPL VORTEX output was deleted."
                ) from exc

    if not MY_DRIVE.is_dir():
        raise RuntimeError(
            f"FPL VORTEX storage root is not available: {MY_DRIVE}. "
            "No FPL VORTEX output was deleted."
        )

    project_root = MY_DRIVE / "FPL_VORTEX"
    output_root = project_root / "FIRST VIDEO"
    expected = MY_DRIVE / "FPL_VORTEX" / "FIRST VIDEO"

    if output_root.resolve(strict=False) != expected.resolve(strict=False):
        raise RuntimeError(f"Unsafe output target refused: {output_root}")

    # Verify Drive write access BEFORE deleting any previous output.
    try:
        project_root.mkdir(parents=True, exist_ok=True)
        probe = project_root / ".vx_drive_write_probe.tmp"
        probe.write_text("FPL_VORTEX_OK", encoding="utf-8")
        if probe.read_text(encoding="utf-8") != "FPL_VORTEX_OK":
            raise OSError("Drive write probe returned unexpected content")
        probe.unlink(missing_ok=True)
    except Exception as exc:
        raise RuntimeError(
            "Google Drive is mounted but not safely writable. "
            "Existing 'FIRST VIDEO' was NOT deleted."
        ) from exc

    # Delete the ENTIRE previous output root, including every file and
    # subfolder, then recreate one empty approved output tree. Google Drive
    # filesystem deletes can settle asynchronously, so verify disappearance
    # before creating the new run tree.
    def _vx_delete_tree_fully(path, label):
        path = Path(path)
        for attempt in range(1, 6):
            if not path.exists():
                return
            try:
                shutil.rmtree(path)
            except FileNotFoundError:
                return
            except Exception:
                if attempt >= 5:
                    raise
            if not path.exists():
                return
            time.sleep(0.35 * attempt)
        raise RuntimeError(f"Complete {label} deletion failed: {path}")

    _vx_delete_tree_fully(output_root, "output-root")
    output_root.mkdir(parents=False, exist_ok=False)

    mp4_dir = output_root / "MP4"
    mp3_dir = output_root / "MP3"
    slide_dir = output_root / "SLIDE"
    data_dir = output_root / "DATA"

    for directory in (mp4_dir, mp3_dir, slide_dir, data_dir):
        directory.mkdir(parents=False, exist_ok=False)

    approved_output_names = {"MP4", "MP3", "SLIDE", "DATA"}
    created_output_names = {path.name for path in output_root.iterdir()}
    if created_output_names != approved_output_names:
        raise RuntimeError(
            "Fresh output-tree verification failed: "
            f"expected {sorted(approved_output_names)}, got {sorted(created_output_names)}"
        )

    if RUNNING_IN_GITHUB:
        work_root = Path(
            os.environ.get(
                "FPL_VORTEX_WORK_ROOT",
                str(Path(os.environ.get("RUNNER_TEMP", ".")) / "FPL_VORTEX_FIRST_DAY_WORK"),
            )
        ).resolve()
    else:
        work_root = Path("/content/FPL_VORTEX_FIRST_DAY_WORK")
    _vx_delete_tree_fully(work_root, "temporary-root")
    work_root.mkdir(parents=True, exist_ok=False)

    html_work = work_root / "html"
    timing_work = work_root / "timing"
    audio_work = work_root / "audio"
    video_work = work_root / "video"
    preview_work = work_root / "preview"
    module_dir = work_root / "vortex"

    for directory in (
        html_work, timing_work, audio_work, video_work, preview_work, module_dir
    ):
        directory.mkdir(parents=False, exist_ok=False)

    return {
        "PROJECT_ROOT": project_root,
        "OUTPUT_ROOT": output_root,
        "MP4_DIR": mp4_dir,
        "MP3_DIR": mp3_dir,
        "SLIDE_DIR": slide_dir,
        "DATA_DIR": data_dir,
        "WORK_ROOT": work_root,
        "HTML_WORK_DIR": html_work,
        "TIMING_WORK_DIR": timing_work,
        "AUDIO_WORK_DIR": audio_work,
        "VIDEO_WORK_DIR": video_work,
        "PREVIEW_WORK_DIR": preview_work,
        "MODULE_DIR": module_dir,
    }


VX_PATHS = vx_prepare_output_tree()
PROJECT_ROOT     = VX_PATHS["PROJECT_ROOT"]
OUTPUT_ROOT      = VX_PATHS["OUTPUT_ROOT"]
MP4_DIR          = VX_PATHS["MP4_DIR"]
MP3_DIR          = VX_PATHS["MP3_DIR"]
SLIDE_DIR        = VX_PATHS["SLIDE_DIR"]
DATA_DIR         = VX_PATHS["DATA_DIR"]
WORK_ROOT        = VX_PATHS["WORK_ROOT"]
HTML_WORK_DIR    = VX_PATHS["HTML_WORK_DIR"]
TIMING_WORK_DIR  = VX_PATHS["TIMING_WORK_DIR"]
AUDIO_WORK_DIR   = VX_PATHS["AUDIO_WORK_DIR"]
VIDEO_WORK_DIR   = VX_PATHS["VIDEO_WORK_DIR"]
PREVIEW_WORK_DIR = VX_PATHS["PREVIEW_WORK_DIR"]
MODULE_DIR       = VX_PATHS["MODULE_DIR"]

# Compatibility dictionary only. P0 remains the sole folder owner.
FOLDERS = {
    "MP4": MP4_DIR,
    "MP3": MP3_DIR,
    "SLIDE": SLIDE_DIR,
    "DATA": DATA_DIR,
}

print("✅ FPL VORTEX STARTED")
print("✅ Storage workspace verified writable before output cleanup")
print(f"📁 CLEAN OUTPUT ROOT: {OUTPUT_ROOT}")
print("📂 Output folders: MP4 • MP3 • SLIDE • DATA")
print("🧹 Entire previous 'FIRST VIDEO' root deleted and recreated clean")
print(f"🧪 Temporary HTML/timing/render files stay in {WORK_ROOT} only")
print(f"📊 Raw market-odds benchmark capture: {'ON' if ENABLE_MARKET_ODDS else 'OFF'}")
print("🔒 Market odds are benchmark data only in V1 — no blending or calibration")
print(f"🖼️ Free player artwork fallback: {'ON' if ENABLE_PLAYER_IMAGE_FALLBACK else 'OFF'}")
print(f"🧩 Forced image fallbacks: {FORCE_IMAGE_FALLBACK_PLAYERS or 'none'}")
print(f"🎬 MP4 quality: {MP4_QUALITY} — {VIDEO_WIDTH}×{VIDEO_HEIGHT} @ {VIDEO_FPS} FPS")
print(f"🎞 Slides selected: {len(SELECTED_SCENES)} — {SELECTED_SCENES}")
print("🔒 WEEKLY IS STANDALONE — no runtime dependency on DO_NOT_DELETE.ipynb")
```

```python
# ============================================================
# CELL 1 — PHASE 0 CLEAN-ROOM AUDIT + DATA CONTRACT
# NO FORECASTING LOGIC IS ALLOWED IN THIS CELL
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import inspect
import hashlib
import json
import re
import sys
import importlib
import pandas as pd

def vx_frame_digest(frame):
    """Stable SHA-256 digest used by downstream immutability guards."""
    if not isinstance(frame, pd.DataFrame):
        return "NOT_A_FRAME"
    ordered = frame.copy().reindex(sorted(frame.columns), axis=1)
    payload = ordered.to_json(
        orient="split", date_format="iso", double_precision=15
    )
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()


# ------------------------------------------------------------
# RERUN-SAFE RUNTIME RESET
# ------------------------------------------------------------
# Colab keeps globals when Runtime -> Run all is used without disconnecting.
# Remove only this notebook's stale downstream Phase/model outputs before the
# clean-room audit is rebuilt; manual Cell 0 controls and Drive paths survive.
def vx_phase0_reset_stale_runtime():
    namespace = globals()
    stale_exact = {
        "CURRENT_SQUAD_DECISIONS",
        "MODEL_CONTROLS",
        "RAW_TABLES",
        "RAW_TABLE_STATUS",
        "ANALYSIS_SCENE_DATA",
        # Canonical live-data/review handoff from an earlier Colab run.
        # These are rebuilt only after Cell 3 loads the freshly written
        # fpl_data module and the live API snapshot passes its contract.
        "data",
        "players",
        "teams",
        "events",
        "fixtures",
        "NEXT_GW",
        "LATEST_COMPLETED_GW",
        "REVIEW_GW",
        "GW_REVIEW",
        "GW_REVIEW_AVAILABLE",
        "GW_REVIEW_PACKAGE",
        "review_package",
        "fetch_fpl_data",
        "fetch_gw_review_snapshot",
        "fetch_event_live",
        "projected_goals_model_only",
        "fixture_probabilities_model_only",
        "fixture_fdr_model_only",
        "player_expected_points_fixture",
        "player_expected_points_gameweek",
        "player_expected_points_horizon",
        "_ceil",
        "_safe",
        "_diff",
        # Phase-0 raw tables from a prior run must not survive before Cell 3.
        "fixtures_raw",
        "match_results_raw",
        "team_match_xg_raw",
        "player_match_raw",
        "injuries_raw",
        "suspensions_raw",
        "lineups_raw",
        "transfers_raw",
        "market_odds_raw",
        # Forecast/model outputs owned by later cells.
        "team_priors",
        "team_strength_signals",
        "projected_goals_baseline",
        "team_strength_updated",
        "fixture_context_phase2",
        "fixture_goal_expectations_model",
        "projected_goals_model_only",
        "poisson_probability_matrix",
        "clean_sheet_probabilities_model_only",
        "fdr_table",
        "fixture_projection_df",
        "team_projection_df",
        "goalkeeper_fixture_fdr_df",
        "player_fixture_xpts_df",
        "player_horizon_xpts_df",
        # Render/narration handoffs can also survive a failed Run all.
        "RENDERED",
        "SCENES",
        "narr",
        "manifest",
        # Exact module build/load state from the previous run.
        "module_code",
        "file_path",
        "FPL_DATA_MODULE_SHA256",
        "_fpl_module",
        "_fpl_spec",
        "_loaded_module_file",
    }
    stale_fragments = (
        "captain_rankings",
        "captain_score",
        "squad_decisions",
        "decision_pool",
    )
    phase_name = re.compile(r"^_?(?:vx_)?phase\d+(?:_|$)", flags=re.I)
    removed = []

    # Cell 2 rewrites fpl_data.py. Remove Python's old module object here so a
    # failed or partial rerun can never leak last run's loader into this run.
    if sys.modules.pop("fpl_data", None) is not None:
        removed.append("sys.modules:fpl_data")
    importlib.invalidate_caches()

    for name in list(namespace):
        if name.startswith("__") or name == "vx_phase0_reset_stale_runtime":
            continue
        lower_name = name.lower()
        if (
            name in stale_exact
            or phase_name.match(name)
            or any(fragment in lower_name for fragment in stale_fragments)
        ):
            namespace.pop(name, None)
            removed.append(name)

    return sorted(removed)


PHASE0_RUNTIME_RESET_REMOVED = vx_phase0_reset_stale_runtime()
print(
    "✅ Rerun-safe Phase 0 reset:",
    f"removed {len(PHASE0_RUNTIME_RESET_REMOVED)} stale downstream globals",
)


# ------------------------------------------------------------
# PHASE 0 CONTROL PLANE
# ------------------------------------------------------------
# Lower-case names are intentional: these are the canonical provenance controls
# every later model layer must inherit rather than silently recreating.
model_version = "VORTEX-V1-PHASE0-CLEANROOM"
data_snapshot_timestamp = None
forecast_timestamp = None
gameweek = None
fixture_id = None
source_name = None
source_timestamp = None
is_pre_deadline_data = None

PHASE0_CONTROL_FIELDS = (
    "model_version",
    "data_snapshot_timestamp",
    "forecast_timestamp",
    "gameweek",
    "fixture_id",
    "source_name",
    "source_timestamp",
    "is_pre_deadline_data",
)

PHASE0_RAW_TABLE_NAMES = (
    "fixtures_raw",
    "match_results_raw",
    "team_match_xg_raw",
    "player_match_raw",
    "injuries_raw",
    "suspensions_raw",
    "lineups_raw",
    "transfers_raw",
    "market_odds_raw",
)

# Build the forbidden names from pieces so the audit definition does not itself
# contain a forbidden literal when the notebook source is scanned.
def _vx_phase0_forbidden_tokens():
    return (
        "team_h_" + "difficulty",
        "team_a_" + "difficulty",
        "old_" + "fdr",
        "old_" + "projected_goals",
        "old_" + "clean_sheet_odds",
        "old_" + "xpts",
        "old_" + "captain_score",
        "old_" + "squad_score",
    )


def _vx_phase0_legacy_markers():
    # Old generated-model artefacts and variable families that must not quietly
    # survive in cache files, helper outputs, summaries or cards.
    return (
        "fixture_" + "projection",
        "team_" + "projection",
        "model_fixture_" + "fdr",
        "player_fixture_" + "xpts",
        "player_horizon_" + "xpts",
        "goalkeeper_fixture_" + "fdr",
        "squad_" + "decisions",
        "captain_" + "score",
    )


def _vx_phase0_review_forbidden_phrases():
    return (
        "projected " + "goals",
        "f" + "dr",
        "clean-sheet " + "odds",
        "clean sheet " + "odds",
        "x" + "pts",
        "buy/sell " + "recommendations",
        "buy " + "recommendation",
        "sell " + "recommendation",
        "captain " + "recommendation",
    )


GW_REVIEW_ALLOWED_FACTS = (
    "Goals",
    "xG",
    "NPxG",
    "xGA",
    "Shots",
    "Big chances",
    "Possession",
    "Line-ups",
    "Minutes",
    "Cards",
    "Set pieces",
    "DEFCON actions",
    "Actual FPL points",
)


def _vx_phase0_token_hits(text, *, include_legacy=True):
    value = str(text or "").lower()

    # Hard-forbidden Phase-0 names remain substring checks because the old_*
    # namespace must never survive inside a larger cache/helper/file label.
    hits = {
        token
        for token in _vx_phase0_forbidden_tokens()
        if token.lower() in value
    }

    if include_legacy:
        # Legacy model markers are exact identifier/key names, not substrings.
        # This still blocks retired fields such as "captain_score" and
        # "squad_decisions", while allowing current-model fields such as
        # "safe_captain_score", "ceiling_captain_score", and
        # "CURRENT_SQUAD_DECISIONS" that are produced legitimately downstream.
        for token in _vx_phase0_legacy_markers():
            pattern = rf"(?<![a-z0-9_]){re.escape(token.lower())}(?![a-z0-9_])"
            if re.search(pattern, value):
                hits.add(token)

    return sorted(hits)


def vx_phase0_assert_no_forbidden_columns(frame, label):
    if not isinstance(frame, pd.DataFrame):
        raise TypeError(f"{label} must be a pandas DataFrame")
    hits = []
    for column in frame.columns:
        token_hits = _vx_phase0_token_hits(column)
        if token_hits:
            hits.append(f"column:{column}:{','.join(token_hits)}")
    if hits:
        raise RuntimeError(
            f"PHASE 0 CLEAN-ROOM FAIL — {label} contains banned model fields: {hits}"
        )


def _vx_phase0_scan_mapping(value, label, violations, depth=0):
    if depth > 6:
        return
    if isinstance(value, pd.DataFrame):
        for col in value.columns:
            for hit in _vx_phase0_token_hits(col):
                violations.append(f"{label}.column:{col}:{hit}")
        return
    if isinstance(value, pd.Series):
        if value.name is not None:
            for hit in _vx_phase0_token_hits(value.name):
                violations.append(f"{label}.series:{value.name}:{hit}")
        return
    if isinstance(value, dict):
        for key, item in list(value.items())[:5000]:
            for hit in _vx_phase0_token_hits(key):
                violations.append(f"{label}.key:{key}:{hit}")
            if isinstance(item, (dict, list, tuple, set, pd.DataFrame, pd.Series)):
                _vx_phase0_scan_mapping(item, f"{label}.{key}", violations, depth + 1)
        return
    if isinstance(value, (list, tuple, set)):
        for idx, item in enumerate(list(value)[:5000]):
            if isinstance(item, (dict, list, tuple, set, pd.DataFrame, pd.Series)):
                _vx_phase0_scan_mapping(item, f"{label}[{idx}]", violations, depth + 1)


def _vx_phase0_scan_files(roots):
    violations = []
    text_suffixes = {".csv", ".json", ".jsonl", ".txt", ".py", ".html", ".js", ".md", ".vtt"}
    pickle_suffixes = {".pkl", ".pickle", ".joblib", ".sav"}
    columnar_suffixes = {".parquet", ".feather"}

    for root in roots:
        root = Path(root)
        if not root.exists():
            continue
        for path in root.rglob("*"):
            if not path.is_file():
                continue
            name_hits = _vx_phase0_token_hits(path.name)
            for hit in name_hits:
                violations.append(f"file_name:{path}:{hit}")

            suffix = path.suffix.lower()
            if suffix in pickle_suffixes:
                # V1 clean-room must not inherit opaque pickled model state.
                violations.append(f"pickled_model_file:{path}")
                continue

            if suffix in columnar_suffixes:
                try:
                    frame = pd.read_parquet(path) if suffix == ".parquet" else pd.read_feather(path)
                    for col in frame.columns:
                        for hit in _vx_phase0_token_hits(col):
                            violations.append(f"columnar:{path}:{col}:{hit}")
                except Exception as exc:
                    violations.append(f"unreadable_columnar:{path}:{type(exc).__name__}")
                continue

            if suffix in text_suffixes:
                try:
                    # Limit protects the notebook from accidentally loading a huge cache.
                    text = path.read_text(encoding="utf-8", errors="ignore")[:8_000_000]
                except Exception:
                    continue
                # Ignore embedded image bytes so random base64 characters cannot
                # trigger a false model-field match in HTML/player-card files.
                text = re.sub(
                    r"data:[^\"']+;base64,[A-Za-z0-9+/=\r\n]+",
                    "<DATA_URI>",
                    text,
                    flags=re.I,
                )
                for hit in _vx_phase0_token_hits(text):
                    violations.append(f"file_content:{path}:{hit}")

                # Any generated GW-review text/HTML must remain retrospective.
                if "review" in re.split(r"[^a-z0-9]+", path.stem.lower()):
                    low = text.lower()
                    for phrase in _vx_phase0_review_forbidden_phrases():
                        if phrase in {"fdr", "xpts"}:
                            found = re.search(rf"\b{re.escape(phrase)}\b", low) is not None
                        else:
                            found = phrase in low
                        if found:
                            violations.append(f"gw_review_forward_output:{path}:{phrase}")
    return violations


def _vx_phase0_scan_chart_objects(namespace):
    violations = []
    try:
        from matplotlib.figure import Figure
    except Exception:
        Figure = ()
    if not Figure:
        return violations
    for name, obj in list(namespace.items()):
        if isinstance(obj, Figure):
            for ax in obj.axes:
                texts = [ax.get_title(), ax.get_xlabel(), ax.get_ylabel()]
                texts += [t.get_text() for t in ax.texts]
                legend = ax.get_legend()
                if legend:
                    texts += [t.get_text() for t in legend.get_texts()]
                for text in texts:
                    for hit in _vx_phase0_token_hits(text):
                        violations.append(f"chart:{name}:{hit}")
    return violations


def vx_phase0_assert_cleanroom(stage="runtime"):
    """Hard gate covering globals, DataFrames, helpers, caches, exports and charts."""
    violations = []
    namespace = globals()

    ignored_names = {
        "PHASE0_CONTROL_FIELDS",
        "PHASE0_RAW_TABLE_NAMES",
        "GW_REVIEW_ALLOWED_FACTS",
    }

    for name, obj in list(namespace.items()):
        if name.startswith("__") or name in ignored_names or name.startswith("_vx_phase0") or name.startswith("vx_phase0"):
            continue

        for hit in _vx_phase0_token_hits(name):
            violations.append(f"global_name:{name}:{hit}")

        if isinstance(obj, (pd.DataFrame, pd.Series, dict, list, tuple, set)):
            _vx_phase0_scan_mapping(obj, f"global:{name}", violations)

        # Hidden helper/colour/card/summary/API mapping usage is caught here.
        if callable(obj):
            try:
                source = inspect.getsource(obj)
            except Exception:
                source = ""
            for hit in _vx_phase0_token_hits(source):
                violations.append(f"helper_source:{name}:{hit}")

    roots = []
    for root_name in ("DATA_DIR", "WORK_ROOT", "MODULE_DIR"):
        root = namespace.get(root_name)
        if root:
            roots.append(Path(root))
    violations.extend(_vx_phase0_scan_files(roots))
    violations.extend(_vx_phase0_scan_chart_objects(namespace))

    if violations:
        preview = "\n".join(f" - {x}" for x in sorted(set(violations))[:80])
        raise RuntimeError(
            f"PHASE 0 CLEAN-ROOM FAIL at {stage}. Remove every stale dependency before continuing:\n{preview}"
        )
    return True


def vx_phase0_assert_review_output(payload, label="GW_REVIEW_PACKAGE"):
    """GW review may describe completed events only; no forward model output."""
    text = json.dumps(payload, ensure_ascii=False, default=str).lower()
    hits = []
    for phrase in _vx_phase0_review_forbidden_phrases():
        if phrase in {"fdr", "xpts"}:
            found = re.search(rf"\b{re.escape(phrase)}\b", text) is not None
        else:
            found = phrase in text
        if found:
            hits.append(phrase)
    if hits:
        raise RuntimeError(
            f"PHASE 0 GW REVIEW FAIL — {label} contains forward-looking output: {sorted(set(hits))}"
        )
    return True


def vx_phase0_empty_raw_table(data_columns=()):
    cols = list(dict.fromkeys(list(data_columns) + list(PHASE0_CONTROL_FIELDS)))
    return pd.DataFrame(columns=cols)


# Placeholders establish the clean-room contract before raw APIs are loaded.
# They are populated only by explicit Layer-0 ingestion downstream.
fixtures_raw = vx_phase0_empty_raw_table(("id", "event", "team_h", "team_a", "kickoff_time", "finished"))
match_results_raw = vx_phase0_empty_raw_table(("id", "event", "team_h", "team_a", "team_h_score", "team_a_score", "finished"))
team_match_xg_raw = vx_phase0_empty_raw_table(("season", "match_date", "team_id", "opponent_id", "venue", "npxg_for", "npxga", "big_chances_for", "big_chances_against", "set_piece_xg_for", "set_piece_xg_against", "penalty_xg_for", "penalty_xg_against", "red_card_minutes_for", "red_card_minutes_against"))
player_match_raw = vx_phase0_empty_raw_table(("player_id", "team_id", "opponent_id", "minutes", "goals", "xg", "npxg", "xga", "shots", "big_chances", "yellow_cards", "red_cards", "actual_fpl_points"))
injuries_raw = vx_phase0_empty_raw_table(("player_id", "team_id", "status", "news", "chance_of_playing_this_round", "chance_of_playing_next_round", "news_added"))
suspensions_raw = vx_phase0_empty_raw_table(("player_id", "team_id", "status", "news", "news_added"))
lineups_raw = vx_phase0_empty_raw_table(("player_id", "team_id", "is_starting", "position", "minutes"))
transfers_raw = vx_phase0_empty_raw_table(("player_id", "team_id", "transfers_in_event", "transfers_out_event", "selected_by_percent"))
market_odds_raw = vx_phase0_empty_raw_table(("home_team", "away_team", "market", "selection", "price", "bookmaker", "last_update"))

RAW_TABLES = {name: globals()[name] for name in PHASE0_RAW_TABLE_NAMES}

# Startup gate catches stale globals when a user reruns the clean notebook in an
# old Colab runtime without restarting it.
vx_phase0_assert_cleanroom("startup")

print("✅ PHASE 0 CLEAN-ROOM POLICY LOADED")
print("✅ Nine raw-table contracts created")
print("✅ Forecast timestamp intentionally EMPTY in Phase 0")
print("✅ Hard gates armed for globals, cached DataFrames, exports, pickles, charts, helpers, cards and summaries")
```

```python
# ============================================================
# FPL VORTEX WEEKLY — CELL 2
# CANONICAL OFFICIAL FPL DATA MODULE
# ============================================================
# ONE loader owns bootstrap + fixture data for the whole notebook.
# GW Review entry/history/picks/live are fetched once into one snapshot.
# ============================================================

from pathlib import Path
import hashlib

MODULE_PATH = Path(MODULE_DIR)

module_code = r'''
from __future__ import annotations

from typing import Any
import copy
import requests
import pandas as pd
import time
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

FPL_API_BASE = "https://fantasy.premierleague.com/api"

# Run-wide immutable FPL API snapshot. Cell 2 rewrites/reloads this module once at
# the start of every Run all, so this cache is automatically fresh for each run.
# The first response for each official endpoint is locked and every later caller
# receives a deep copy of that exact payload.
_VX_RUN_JSON_CACHE: dict[str, Any] = {}


def _vx_http_session() -> requests.Session:
    retry = Retry(
        total=3,
        connect=3,
        read=3,
        backoff_factor=0.6,
        status_forcelist=(429, 500, 502, 503, 504),
        allowed_methods=frozenset({"GET"}),
        raise_on_status=False,
    )
    adapter = HTTPAdapter(max_retries=retry)
    session = requests.Session()
    session.mount("https://", adapter)
    session.headers.update({
        "User-Agent": "FPL-VORTEX/2.0",
        "Accept": "application/json",
    })
    return session


def _vx_get_json(session: requests.Session, path: str) -> Any:
    # One Run all = one immutable official FPL snapshot per endpoint. The first
    # request is forced fresh; every later request in this runtime reuses it.
    cache_key = str(path)
    if cache_key in _VX_RUN_JSON_CACHE:
        return copy.deepcopy(_VX_RUN_JSON_CACHE[cache_key])

    sep = "&" if "?" in path else "?"
    url = f"{FPL_API_BASE}{path}{sep}_vx={time.time_ns()}"
    response = session.get(
        url,
        timeout=30,
        headers={"Cache-Control": "no-cache, no-store", "Pragma": "no-cache"},
    )
    response.raise_for_status()
    payload = response.json()
    _VX_RUN_JSON_CACHE[cache_key] = copy.deepcopy(payload)
    return copy.deepcopy(payload)


def _vx_sanitize_fixture_rows(rows):
    # Provider-owned difficulty labels are discarded at the ingestion boundary.
    banned_keys = ("team_h_" + "difficulty", "team_a_" + "difficulty")
    clean = []
    for item in rows or []:
        row = dict(item or {})
        for key in banned_keys:
            row.pop(key, None)
        clean.append(row)
    return clean


def _vx_require_columns(frame: pd.DataFrame, required: set[str], label: str) -> None:
    missing = sorted(required - set(frame.columns))
    if missing:
        raise ValueError(f"{label} missing required columns: {missing}")


def fetch_fpl_data() -> dict[str, Any]:
    # ONE canonical bootstrap/fixtures snapshot for the notebook.
    session = _vx_http_session()
    bootstrap = _vx_get_json(session, "/bootstrap-static/")
    fixture_rows = _vx_sanitize_fixture_rows(_vx_get_json(session, "/fixtures/"))

    players = pd.DataFrame(bootstrap["elements"]).copy()
    teams = pd.DataFrame(bootstrap["teams"]).copy()
    events = pd.DataFrame(bootstrap["events"]).copy()
    element_types = pd.DataFrame(bootstrap.get("element_types", [])).copy()
    fixtures = pd.DataFrame(fixture_rows).copy()

    _vx_require_columns(players, {"id", "team", "element_type", "now_cost"}, "players")
    _vx_require_columns(teams, {"id", "name", "short_name"}, "teams")
    _vx_require_columns(
        events,
        {"id", "deadline_time", "is_previous", "is_current", "is_next", "finished", "data_checked"},
        "events",
    )
    _vx_require_columns(
        fixtures,
        {
            "id", "team_h", "team_a", "event", "started", "minutes",
            "finished", "finished_provisional", "team_h_score", "team_a_score",
        },
        "fixtures",
    )

    if players["id"].duplicated().any():
        raise ValueError("Duplicate FPL player IDs found")
    if teams["id"].duplicated().any():
        raise ValueError("Duplicate FPL team IDs found")
    if fixtures["id"].duplicated().any():
        raise ValueError("Duplicate FPL fixture IDs found")

    valid_team_ids = set(teams["id"].astype(int).tolist())
    referenced_team_ids = set(fixtures["team_h"].dropna().astype(int)) | set(fixtures["team_a"].dropna().astype(int))
    unknown = referenced_team_ids - valid_team_ids
    if unknown:
        raise ValueError(f"Fixtures reference unknown teams: {sorted(unknown)}")

    team_name_map = teams.set_index("id")["name"].to_dict()
    team_short_map = teams.set_index("id")["short_name"].to_dict()

    if not element_types.empty and {"id", "singular_name_short"}.issubset(element_types.columns):
        position_map = element_types.set_index("id")["singular_name_short"].to_dict()
    else:
        position_map = {1: "GKP", 2: "DEF", 3: "MID", 4: "FWD"}

    position_map = {
        int(k): ("GK" if str(v).upper() == "GKP" else str(v).upper())
        for k, v in position_map.items()
    }

    players["team_name"] = players["team"].map(team_name_map)
    players["team_short"] = players["team"].map(team_short_map)
    players["price"] = pd.to_numeric(players["now_cost"], errors="coerce") / 10.0
    players["position"] = players["element_type"].map(position_map)

    # ------------------------------------------------------------------
    # ONE CANONICAL OFFICIAL STATUS CONTRACT
    # ------------------------------------------------------------------
    # The live FPL payload can pass through short transition windows where
    # event flags and fixture flags update at different moments. No single flag
    # is accepted as proof of completion. Every completed fixture must satisfy
    # all four pieces of provider evidence already present in /fixtures/:
    #   1) started == True
    #   2) minutes >= 90
    #   3) finished OR finished_provisional
    #   4) both final-score fields are present
    # A review Gameweek additionally needs a terminal event lifecycle signal
    # (finished, data_checked, or is_previous) AND every fixture in that event
    # must satisfy the full fixture contract above.
    def _flag(frame: pd.DataFrame, column: str) -> pd.Series:
        if column not in frame.columns:
            raise ValueError(f"Official FPL payload missing required flag: {column}")
        return frame[column].fillna(False).astype(bool)

    event_finished_signal = _flag(events, "finished")
    event_checked_signal = _flag(events, "data_checked")
    event_previous_signal = _flag(events, "is_previous")
    event_current_signal = _flag(events, "is_current")
    event_next_signal = _flag(events, "is_next")

    fixture_started_signal = _flag(fixtures, "started")
    fixture_finished_signal = _flag(fixtures, "finished")
    fixture_provisional_signal = _flag(fixtures, "finished_provisional")
    fixture_minutes = pd.to_numeric(fixtures["minutes"], errors="coerce").fillna(0.0)
    fixture_minutes_complete = fixture_minutes.ge(90.0)
    fixture_has_score = fixtures["team_h_score"].notna() & fixtures["team_a_score"].notna()
    fixture_status_complete = fixture_finished_signal | fixture_provisional_signal
    fixture_complete = (
        fixture_started_signal
        & fixture_minutes_complete
        & fixture_status_complete
        & fixture_has_score
    )

    # Persist the atomic fixture checks so downstream cells consume the same
    # answer and never reimplement status logic locally.
    fixtures["vx_started"] = fixture_started_signal
    fixtures["vx_minutes_complete"] = fixture_minutes_complete
    fixtures["vx_status_complete"] = fixture_status_complete
    fixtures["vx_score_complete"] = fixture_has_score
    fixtures["vx_completed"] = fixture_complete

    fixture_completion_by_gw: dict[int, bool] = {}
    fixture_counts_by_gw: dict[int, int] = {}
    fixture_state_counts_by_gw: dict[int, dict[str, int]] = {}
    fixture_rows_with_event = fixtures.loc[fixtures["event"].notna()].copy()

    for gw_value, gw_rows in fixture_rows_with_event.groupby("event"):
        gw = int(gw_value)
        idx = gw_rows.index
        fixture_counts_by_gw[gw] = int(len(gw_rows))
        fixture_completion_by_gw[gw] = bool(
            len(gw_rows) > 0 and fixtures.loc[idx, "vx_completed"].all()
        )
        fixture_state_counts_by_gw[gw] = {
            "fixtures": int(len(gw_rows)),
            "started": int(fixtures.loc[idx, "vx_started"].sum()),
            "minutes_90_plus": int(fixtures.loc[idx, "vx_minutes_complete"].sum()),
            "status_complete": int(fixtures.loc[idx, "vx_status_complete"].sum()),
            "scores_present": int(fixtures.loc[idx, "vx_score_complete"].sum()),
            "completed": int(fixtures.loc[idx, "vx_completed"].sum()),
        }

    event_finished_gws = set(events.loc[event_finished_signal, "id"].astype(int).tolist())
    event_data_checked_gws = set(events.loc[event_checked_signal, "id"].astype(int).tolist())
    event_previous_gws = set(events.loc[event_previous_signal, "id"].astype(int).tolist())
    event_current_gws = set(events.loc[event_current_signal, "id"].astype(int).tolist())
    event_terminal_gws = event_finished_gws | event_data_checked_gws | event_previous_gws
    fixture_finished_gws = {gw for gw, done in fixture_completion_by_gw.items() if done}

    # ------------------------------------------------------------------
    # TRANSITIONAL CURRENT-GW PATH
    # ------------------------------------------------------------------
    # The live FPL API can sit in a legitimate transitional state where the
    # just-completed Gameweek is still flagged is_current=True — finished,
    # data_checked and is_previous all still False — while the following
    # Gameweek has already become is_next=True. event_terminal_gws alone
    # cannot see a GW in this state, even when every one of its fixtures has
    # actually finished.
    #
    # Recognise that state as a second, independently-verified route into
    # review eligibility, gated on:
    #   1) event.is_current is True
    #   2) the event's own official deadline has already passed (so an
    #      upcoming GW mislabelled early can never qualify)
    #   3) the event has at least one fixture on record
    #   4) every fixture belonging to that event satisfies the existing
    #      fixture-evidence contract above (started, 90+ minutes, finished
    #      OR finished_provisional, both scores present)
    # This never widens the fixture-evidence contract itself and never
    # touches the terminal-flag route below — it only adds one more way to
    # reach the same review_ready_gws set once fixture evidence agrees.
    event_deadline_series = pd.to_datetime(events["deadline_time"], errors="coerce", utc=True)
    event_deadline_passed_signal = event_deadline_series.notna() & event_deadline_series.le(
        pd.Timestamp.now(tz="UTC")
    )
    transitional_current_gws = {
        gw for gw in set(
            events.loc[event_current_signal & event_deadline_passed_signal, "id"].astype(int).tolist()
        )
        if fixture_counts_by_gw.get(gw, 0) > 0 and fixture_completion_by_gw.get(gw, False)
    }

    review_eligible_gws = event_terminal_gws | transitional_current_gws

    review_ready_gws = sorted(
        int(gw)
        for gw in review_eligible_gws
        if fixture_completion_by_gw.get(int(gw), False)
    )
    latest_completed_gw = max(review_ready_gws) if review_ready_gws else None
    events["vx_review_ready"] = pd.to_numeric(events["id"], errors="coerce").isin(review_ready_gws)

    completion_source_by_gw: dict[int, list[str]] = {}
    for gw in review_ready_gws:
        sources: list[str] = []
        if gw in event_finished_gws:
            sources.append("event.finished")
        if gw in event_data_checked_gws:
            sources.append("event.data_checked")
        if gw in event_previous_gws:
            sources.append("event.is_previous")
        if gw in transitional_current_gws:
            sources.append("event.is_current+deadline_passed")
        sources.append("fixtures.started+90m+status+scores")
        completion_source_by_gw[int(gw)] = sources

    # ------------------------------------------------------------------
    # NEXT GAMEWEEK — OFFICIAL POINTER FIRST, DEADLINE-ONLY TRANSITION
    # ------------------------------------------------------------------
    # Prefer FPL's unique is_next row. During a short provider transition where
    # is_next is temporarily absent, use only the earliest future official
    # deadline. If neither exists, stop instead of guessing max(event) or any
    # fixed Gameweek number.
    next_rows = events.loc[event_next_signal].copy()
    if len(next_rows) > 1:
        raise RuntimeError(
            "Official FPL payload contains multiple is_next events; refusing to guess NEXT_GW."
        )

    if len(next_rows) == 1:
        next_gw = int(next_rows.iloc[0]["id"])
        next_gw_source = "event.is_next"
    else:
        deadline_series = pd.to_datetime(events["deadline_time"], errors="coerce", utc=True)
        now_utc = pd.Timestamp.now(tz="UTC")
        future_mask = deadline_series.notna() & deadline_series.gt(now_utc)
        future_events = events.loc[future_mask].copy()
        if future_events.empty:
            raise RuntimeError(
                "Official FPL payload has no is_next event and no future deadline. "
                "NEXT_GW cannot be resolved without guessing."
            )
        future_events["_vx_deadline"] = deadline_series.loc[future_events.index]
        future_events = future_events.sort_values(["_vx_deadline", "id"])
        next_gw = int(future_events.iloc[0]["id"])
        next_gw_source = "earliest_future_official_deadline"

    if next_gw in set(review_ready_gws):
        raise RuntimeError(
            f"Official FPL status contradiction: NEXT_GW={next_gw} is already review-ready."
        )

    return {
        "bootstrap": bootstrap,
        "players": players,
        "teams": teams,
        "events": events,
        "element_types": element_types,
        "fixtures": fixtures,
        "next_gw": next_gw,
        "next_gw_source": next_gw_source,
        "latest_completed_gw": latest_completed_gw,
        "review_ready_gws": review_ready_gws,
        "completion_source_by_gw": completion_source_by_gw,
        "event_finished_gws": sorted(event_finished_gws),
        "event_data_checked_gws": sorted(event_data_checked_gws),
        "event_previous_gws": sorted(event_previous_gws),
        "event_current_gws": sorted(event_current_gws),
        "event_terminal_gws": sorted(event_terminal_gws),
        "transitional_current_gws": sorted(transitional_current_gws),
        "fixture_finished_gws": sorted(fixture_finished_gws),
        "fixture_completion_by_gw": fixture_completion_by_gw,
        "fixture_counts_by_gw": fixture_counts_by_gw,
        "fixture_state_counts_by_gw": fixture_state_counts_by_gw,
        "team_name_map": team_name_map,
        "team_short_map": team_short_map,
        "position_map": position_map,
    }


def fetch_gw_review_snapshot(
    entry_id: int,
    review_gw: int,
    *,
    bootstrap: dict[str, Any] | None = None,
    fixtures: list[dict[str, Any]] | None = None,
) -> dict[str, Any]:
    # ONE immutable entry-specific GW Review snapshot.
    # Reuses the canonical bootstrap/full fixture snapshot when supplied,
    # so P9/P10 never refetch those endpoints.
    entry_id = int(entry_id)
    review_gw = int(review_gw)
    if entry_id <= 0:
        raise ValueError("entry_id must be positive")
    if not 1 <= review_gw <= 38:
        raise ValueError("review_gw must be in 1..38")

    session = _vx_http_session()
    if bootstrap is None:
        bootstrap = _vx_get_json(session, "/bootstrap-static/")
    if fixtures is None:
        fixtures = _vx_sanitize_fixture_rows(_vx_get_json(session, "/fixtures/"))

    return {
        "entry_id": entry_id,
        "gw": review_gw,
        "bootstrap": bootstrap,
        "entry": _vx_get_json(session, f"/entry/{entry_id}/"),
        "history": _vx_get_json(session, f"/entry/{entry_id}/history/"),
        "picks": _vx_get_json(session, f"/entry/{entry_id}/event/{review_gw}/picks/"),
        "live": _vx_get_json(session, f"/event/{review_gw}/live/"),
        "fixtures": fixtures,
    }


def fetch_event_live(event_id: int, *, cache_path: str | None = None) -> dict[str, Any]:
    """Fetch one official FPL event-live payload through the canonical retry session.

    If cache_path is supplied, completed historical payloads can be reused without
    another network call. This is the canonical event-live fetch helper for descriptive historical data.
    """
    event_id = int(event_id)
    if not 1 <= event_id <= 38:
        raise ValueError("event_id must be in 1..38")

    cache = None
    if cache_path:
        from pathlib import Path
        cache = Path(cache_path)
        if cache.exists():
            try:
                import json
                return json.loads(cache.read_text(encoding="utf-8"))
            except Exception:
                pass

    session = _vx_http_session()
    payload = _vx_get_json(session, f"/event/{event_id}/live/")

    if cache is not None:
        import json
        cache.parent.mkdir(parents=True, exist_ok=True)
        cache.write_text(json.dumps(payload, ensure_ascii=False), encoding="utf-8")

    return payload


def fetch_recent_player_match_snapshot(
    events,
    *,
    lookback_gws: int = 6,
    cache_dir: str | None = None,
) -> pd.DataFrame:
    """Return completed historical player-Gameweek raw statistics from official event-live.

    Only FINISHED gameweeks are used. This is raw descriptive evidence only; no forecast is produced.
    """
    frame = events.copy() if isinstance(events, pd.DataFrame) else pd.DataFrame(events)
    if frame.empty or "id" not in frame.columns:
        return pd.DataFrame()

    if "vx_review_ready" not in frame.columns:
        raise RuntimeError(
            "fetch_recent_player_match_snapshot requires canonical events.vx_review_ready; "
            "run fetch_fpl_data() from this module first."
        )
    completed = frame.loc[frame["vx_review_ready"].fillna(False).astype(bool)].copy()

    if completed.empty:
        return pd.DataFrame()

    completed["id"] = pd.to_numeric(completed["id"], errors="coerce")
    completed = completed.dropna(subset=["id"]).sort_values("id")
    gws = completed["id"].astype(int).tolist()[-max(1, int(lookback_gws)):]

    rows: list[dict[str, Any]] = []
    cache_root = None
    if cache_dir:
        from pathlib import Path
        cache_root = Path(cache_dir)
        cache_root.mkdir(parents=True, exist_ok=True)

    for gw in gws:
        cache_path = str(cache_root / f"gw_{gw:02d}_live.json") if cache_root else None
        payload = fetch_event_live(gw, cache_path=cache_path)
        for element in payload.get("elements", []):
            stats = dict(element.get("stats") or {})
            rows.append({
                "event": int(gw),
                "player_id": int(element.get("id") or 0),
                **stats,
            })

    result = pd.DataFrame(rows)
    if not result.empty:
        result = result.loc[result["player_id"] > 0].sort_values(["player_id", "event"]).reset_index(drop=True)
    return result

'''

file_path = MODULE_PATH / "fpl_data.py"
file_path.write_text(module_code, encoding="utf-8")
FPL_DATA_MODULE_SHA256 = hashlib.sha256(module_code.encode("utf-8")).hexdigest()
_disk_module_sha256 = hashlib.sha256(file_path.read_bytes()).hexdigest()
if _disk_module_sha256 != FPL_DATA_MODULE_SHA256:
    raise RuntimeError("fpl_data.py write verification failed; refusing to load a stale module.")

print("✅ CANONICAL FPL MODULE CREATED")
print(file_path)
print(f"✅ MODULE SHA256: {FPL_DATA_MODULE_SHA256[:16]}…")
```

```python
# ============================================================
# FPL VORTEX WEEKLY — VISUAL ENRICHMENT ONLY
# FREE PLAYER ARTWORK FALLBACK • NO PREDICTION DATA
# TheSportsDB V1 uses public free key 123 — no Colab Secret required.
# STANDALONE: never reads DO_NOT_DELETE.ipynb
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from difflib import SequenceMatcher
import json, re, time, requests

CACHE_DIR = Path(globals().get("DATA_DIR", Path("output/.work/data")))
CACHE_DIR.mkdir(parents=True, exist_ok=True)

THESPORTSDB_BASE = "https://www.thesportsdb.com/api/v1/json/123"

now_utc = datetime.now(timezone.utc)
EPL_SEASON = now_utc.year if now_utc.month >= 7 else now_utc.year - 1

def vx_normalize_name(value):
    return re.sub(r"[^a-z0-9]", "", str(value or "").lower())

def _cache_path(name):
    safe = re.sub(r"[^A-Za-z0-9_.-]+", "_", name)
    return CACHE_DIR / safe

def _read_cache(name, max_age_hours=None):
    path = _cache_path(name)
    if not path.exists():
        return None
    if max_age_hours is not None:
        age_hours = (time.time() - path.stat().st_mtime) / 3600
        if age_hours > max_age_hours:
            return None
    try:
        return json.loads(path.read_text(encoding="utf-8"))
    except Exception:
        return None

def _write_cache(name, payload):
    path = _cache_path(name)
    path.write_text(json.dumps(payload, ensure_ascii=False, indent=2), encoding="utf-8")
    return payload

# ============================================================
# FREE PLAYER ARTWORK — THESPORTSDB V1
# ============================================================
# Public free key: 123. Rate limit: 30 requests/minute.
# On-demand queries are cached for 30 days.
# FPL is primary. TheSportsDB is used only for missing/broken FPL images or manual wrong-kit overrides.
# ============================================================
ENABLE_PLAYER_IMAGE_FALLBACK_LOCAL = bool(globals().get("ENABLE_PLAYER_IMAGE_FALLBACK", True))
THESPORTSDB_CC_ONLY = True
_THESPORTSDB_LAST_CALL = 0.0

FORCE_IMAGE_FALLBACK_SET = {
    vx_normalize_name(x) for x in str(globals().get("FORCE_IMAGE_FALLBACK_PLAYERS", "") or "").split(",")
    if vx_normalize_name(x)
}

def _fpl_full_name(player):
    if hasattr(player, "to_dict"):
        player = player.to_dict()
    return " ".join(
        str(player.get(k, "") or "").strip()
        for k in ("first_name", "second_name")
        if str(player.get(k, "") or "").strip()
    ).strip()

def _fpl_team_name(player):
    if hasattr(player, "to_dict"):
        player = player.to_dict()
    try:
        team_id = int(player.get("team"))
    except Exception:
        return ""
    obj = globals().get("teams")
    try:
        if hasattr(obj, "loc"):
            row = obj.loc[obj["id"].astype(int) == team_id]
            if not row.empty:
                return str(row.iloc[0].get("name", "") or "")
        elif isinstance(obj, list):
            for t in obj:
                if int(t.get("id", -1)) == team_id:
                    return str(t.get("name", "") or "")
    except Exception:
        pass
    return ""

def fpl_photo_url(fpl_player):
    if hasattr(fpl_player, "to_dict"):
        fpl_player = fpl_player.to_dict()
    code = fpl_player.get("code")
    if code:
        try:
            code = int(code)
        except Exception:
            pass
        return "https://resources.premierleague.com/premierleague/" + f"photos/players/250x250/p{code}.png"
    return ""

def _thesportsdb_search(name, max_age_hours=720):
    """Search one player. Cached for 30 days and throttled for the free limit."""
    global _THESPORTSDB_LAST_CALL
    name = str(name or "").strip()
    if not name or not ENABLE_PLAYER_IMAGE_FALLBACK_LOCAL:
        return None
    cache_name = f"thesportsdb_player_{vx_normalize_name(name)}.json"
    cached = _read_cache(cache_name, max_age_hours=max_age_hours)
    if cached is not None:
        return cached
    elapsed = time.time() - _THESPORTSDB_LAST_CALL
    if elapsed < 2.05:
        time.sleep(2.05 - elapsed)
    try:
        r = requests.get(f"{THESPORTSDB_BASE}/searchplayers.php", params={"p": name}, timeout=20)
        _THESPORTSDB_LAST_CALL = time.time()
        r.raise_for_status()
        payload = r.json()
    except Exception as exc:
        payload = {"player": None, "_error": f"{type(exc).__name__}: {exc}"}
    return _write_cache(cache_name, payload)

def _cc_allowed(row):
    if not THESPORTSDB_CC_ONLY:
        return True
    flag = str(row.get("strCreativeCommons", "") or "").strip().lower()
    return bool(flag) and flag not in {"0", "false", "no", "none", "null"}

def match_thesportsdb_player(fpl_player):
    if hasattr(fpl_player, "to_dict"):
        fpl_player = fpl_player.to_dict()
    full_name = _fpl_full_name(fpl_player)
    web_name = str(fpl_player.get("web_name", "") or "").strip()
    team_name = _fpl_team_name(fpl_player)
    queries = []
    for q in (full_name, web_name):
        if q and q not in queries:
            queries.append(q)
    for q in queries:
        payload = _thesportsdb_search(q)
        rows = (payload or {}).get("player") or []
        if isinstance(rows, dict):
            rows = [rows]
        for row in rows:
            sport = str(row.get("strSport", "") or "").lower()
            if sport and sport not in {"soccer", "football"}:
                continue
            api_name = str(row.get("strPlayer", "") or "")
            name_score = max(
                SequenceMatcher(None, vx_normalize_name(full_name), vx_normalize_name(api_name)).ratio() if full_name else 0.0,
                SequenceMatcher(None, vx_normalize_name(web_name), vx_normalize_name(api_name)).ratio() if web_name else 0.0,
            )
            api_team = str(row.get("strTeam", "") or "")
            team_score = SequenceMatcher(None, vx_normalize_name(team_name), vx_normalize_name(api_team)).ratio() if team_name and api_team else 0.0
            if name_score >= 0.72 or (name_score >= 0.62 and team_score >= 0.65):
                return row
    return None

def thesportsdb_player_artwork(fpl_player):
    row = match_thesportsdb_player(fpl_player)
    if not row or not _cc_allowed(row):
        return {"cutout": "", "thumb": "", "fanart": "", "creative_commons": "", "source": ""}
    fanart = next((
        str(row.get(k, "") or "").strip()
        for k in ("strFanart1", "strFanart2", "strFanart3", "strFanart4")
        if str(row.get(k, "") or "").strip()
    ), "")
    return {
        "cutout": str(row.get("strCutout", "") or "").strip(),
        "thumb": str(row.get("strThumb", "") or "").strip(),
        "fanart": fanart,
        "creative_commons": str(row.get("strCreativeCommons", "") or "").strip(),
        "source": "TheSportsDB",
    }

def _forced_image_fallback(fpl_player):
    if hasattr(fpl_player, "to_dict"):
        fpl_player = fpl_player.to_dict()
    names = {
        vx_normalize_name(_fpl_full_name(fpl_player)),
        vx_normalize_name(str(fpl_player.get("web_name", "") or "")),
    }
    return bool(FORCE_IMAGE_FALLBACK_SET.intersection(names))

def _official_fpl_image_usable(url, max_age_hours=168):
    """Validate the official FPL/PL image once, then cache the result for 7 days."""
    if not url:
        return False
    cache_name = f"fpl_image_check_{vx_normalize_name(url)}.json"
    cached = _read_cache(cache_name, max_age_hours=max_age_hours)
    if isinstance(cached, dict) and "ok" in cached:
        return bool(cached["ok"])
    ok = False
    try:
        with requests.get(url, stream=True, timeout=12) as r:
            ctype = str(r.headers.get("content-type", "") or "").lower()
            first = next(r.iter_content(chunk_size=512), b"")
            ok = (r.status_code == 200 and ctype.startswith("image/") and len(first) > 32)
    except Exception:
        ok = False
    _write_cache(cache_name, {"ok": ok, "checked_at": datetime.now(timezone.utc).isoformat()})
    return ok

def resolve_player_photo(fpl_player, prefer_cutout=True):
    """
    FPL FIRST.
    Use the official FPL/PL player image whenever it is valid.
    Use TheSportsDB only when the FPL image is missing/broken or the player is
    explicitly listed in FORCE_IMAGE_FALLBACK_PLAYERS because the visible kit is stale/wrong.
    """
    official = fpl_photo_url(fpl_player)
    force_fallback = _forced_image_fallback(fpl_player)

    if not force_fallback and _official_fpl_image_usable(official):
        return official

    if ENABLE_PLAYER_IMAGE_FALLBACK_LOCAL:
        art = thesportsdb_player_artwork(fpl_player)
        if prefer_cutout and art.get("cutout"):
            return art["cutout"]
        if art.get("thumb"):
            return art["thumb"]
        if art.get("cutout"):
            return art["cutout"]

    # If the only problem is that an image could not be validated, still return
    # the official URL as the final graceful fallback rather than leaving blank.
    return official or ""

if ENABLE_PLAYER_IMAGE_FALLBACK_LOCAL:
    print("✅ Free player artwork fallback ON — TheSportsDB V1 (cached, CC-tagged artwork preferred)")
else:
    print("ℹ️ Free player artwork fallback disabled in Cell 0.")

print("✅ VISUAL ENRICHMENT READY — no prediction source loaded")


# ============================================================
# SHARED VISUAL ASSETS — SINGLE SOURCE OF TRUTH
# ============================================================
# Team metadata, club crests and VORTEX logo are loaded ONCE here.
# Later slide/card cells consume these globals and must not rebuild them.

from concurrent.futures import ThreadPoolExecutor, as_completed
from io import BytesIO
import base64
from PIL import Image

for _required in ("teams", "MY_DRIVE", "PROJECT_ROOT"):
    if _required not in globals():
        raise NameError(f"Run Cell 0 and official FPL-data cells first: missing {_required}")

def vx_bytes_to_data_uri(data, mime_type="image/png"):
    return f"data:{mime_type};base64," + base64.b64encode(data).decode("utf-8")

def vx_trimmed_logo_data_uri(path):
    image = Image.open(path).convert("RGBA")
    bbox = image.getchannel("A").getbbox()
    if bbox:
        left, top, right, bottom = bbox
        pad = max(8, int(max(right-left, bottom-top) * 0.025))
        image = image.crop((max(0,left-pad), max(0,top-pad), min(image.width,right+pad), min(image.height,bottom+pad)))
    buffer = BytesIO(); image.save(buffer, format="PNG", optimize=True)
    return vx_bytes_to_data_uri(buffer.getvalue(), "image/png")

team_meta_by_id = {}
for _, _row in teams.iterrows():
    _id = int(_row["id"])
    team_meta_by_id[_id] = {
        "id": _id, "code": int(_row["code"]), "name": str(_row["name"]),
        "short": str(_row["short_name"]).upper().strip(),
        "badge": "",
    }

_vx_crest_session = requests.Session()
def vx_download_team_badge(meta):
    code = int(meta["code"])
    for url in (
        f"https://resources.premierleague.com/premierleague/badges/100/t{code}.png",
        f"https://resources.premierleague.com/premierleague/badges/50/t{code}.png",
    ):
        try:
            response = _vx_crest_session.get(url, timeout=8)
            if response.status_code == 200 and len(response.content) > 300:
                return int(meta["id"]), vx_bytes_to_data_uri(response.content, "image/png")
        except Exception:
            pass
    return int(meta["id"]), ""

print("⏳ Loading canonical club crests once...")
with ThreadPoolExecutor(max_workers=10) as _executor:
    _jobs=[_executor.submit(vx_download_team_badge,m) for m in team_meta_by_id.values()]
    for _job in as_completed(_jobs):
        _id,_badge=_job.result(); team_meta_by_id[_id]["badge"]=_badge
team_meta_by_short={m["short"]:m for m in team_meta_by_id.values()}
print("✅ Canonical club crests:",sum(1 for m in team_meta_by_id.values() if m["badge"]),f"/{len(team_meta_by_id)}")

_vx_logo_candidates=[Path(MY_DRIVE)/"elements"/"logo.png",Path(PROJECT_ROOT)/"elements"/"logo.png",Path(PROJECT_ROOT)/"logo.png"]
_vx_logo_path=next((x for x in _vx_logo_candidates if x.exists()),None)
if _vx_logo_path is None:
    _hits=list(Path(MY_DRIVE).rglob("logo.png")); _vx_logo_path=_hits[0] if _hits else None
if _vx_logo_path:
    vortex_logo_uri=vx_trimmed_logo_data_uri(_vx_logo_path)
    vortex_logo_html='<img class="vortex-logo" src="'+vortex_logo_uri+'" alt="FPL VORTEX">'
    print("✅ Canonical VORTEX logo:",_vx_logo_path)
else:
    vortex_logo_uri=""; vortex_logo_html='<div class="logo-fallback">FPL <span>VORTEX</span></div>'
    print("⚠️ logo.png not found — using canonical text fallback")


# Premier League broadcast lockup — ONE shared source for every slide.
# The exact pl.png colours are preserved. Only the lion area is cropped from
# that source asset; the words are set once in a single readable line.
_pl_logo_candidates = [
    Path(MY_DRIVE) / "elements" / "pl.png",
    Path(PROJECT_ROOT) / "elements" / "pl.png",
    Path(PROJECT_ROOT) / "pl.png",
]
_pl_logo_path = next((p for p in _pl_logo_candidates if p.exists()), None)
if _pl_logo_path is None:
    _pl_hits = list(Path(MY_DRIVE).rglob("pl.png"))
    _pl_logo_path = _pl_hits[0] if _pl_hits else None

_pl_season_start = str(EPL_SEASON)
_pl_season_end = str(EPL_SEASON + 1)
if _pl_logo_path:
    PL_LOGO_URI = vx_trimmed_logo_data_uri(_pl_logo_path)
    print("✅ Canonical Premier League artwork loaded in original colours:", _pl_logo_path)
else:
    PL_LOGO_URI = ""
    print("⚠️ pl.png not found — using readable Premier League fallback")

_pl_mark_html = (
    '<div class="plLogoCrop"><img class="plLogoOriginal" '
    f'src="{PL_LOGO_URI}" alt="Premier League"></div>'
    if PL_LOGO_URI
    else '<div class="plLogoFallback">PL</div>'
)
PL_BUG_HTML = (
    '<div class="plBug" aria-label="Premier League season '
    + _pl_season_start + '/' + _pl_season_end + '">'
    + '<div class="plBugPanel">'
    + _pl_mark_html
    + '<div class="plCopy">'
    + '<div class="plWordmark">Premier League</div>'
    + '<div class="plSeason">'
    + f'<span>{_pl_season_start}</span><span>/</span><span>{_pl_season_end}</span>'
    + '</div></div></div></div>'
)

# 3840×2160 geometry (validated against all six slide templates):
#   panel 540×104 at y=4 -> bottom y=108
#   Intro GW pill: y=152..236  -> 44 px vertical clearance
#   Outro GW pill: y≈156..231 -> ~48 px vertical clearance
#   GW Review pitch begins y=105 -> only a 3 px decorative-edge overlap
#   Matrix header meta is shifted left in Cell 14B, leaving >100 px horizontal clearance.
PL_BUG_CSS = r"""
.plBug{
  position:absolute;
  top:var(--pl-bug-top,4px);
  right:var(--pl-bug-right,30px);
  z-index:45;
  pointer-events:none;
}
.plBugPanel{
  width:var(--pl-bug-width,540px);
  height:104px;
  display:flex;
  align-items:center;
  gap:16px;
  padding:8px 18px 8px 16px;
  overflow:hidden;
  background:#ffffff;
  border:2px solid rgba(61,25,91,.18);
  border-radius:14px;
  box-shadow:0 10px 28px rgba(0,0,0,.30);
}
.plLogoCrop{
  position:relative;
  flex:0 0 88px;
  width:88px;
  height:88px;
  overflow:hidden;
}
.plLogoOriginal{
  position:absolute;
  left:-3px;
  top:-22px;
  width:auto;
  height:150px;
  max-width:none!important;
  max-height:none!important;
  object-fit:initial!important;
  filter:none!important;
}
.plLogoFallback{
  flex:0 0 92px;
  width:92px;
  height:92px;
  display:flex;
  align-items:center;
  justify-content:center;
  border-radius:50%;
  background:#3d195b;
  color:#ffffff;
  font-size:36px;
  font-weight:1000;
  line-height:1;
}
.plCopy{
  min-width:0;
  display:flex;
  flex-direction:column;
  align-items:flex-start;
  justify-content:center;
}
.plWordmark{
  color:#3d195b;
  font-size:50px;
  font-weight:1000;
  line-height:.95;
  letter-spacing:-1.5px;
  white-space:nowrap;
}
.plSeason{
  margin-top:4px;
  display:inline-flex!important;
  align-items:baseline;
  gap:.10em;
  color:#3d195b;
  font-size:34px;
  font-weight:1000;
  line-height:1;
  letter-spacing:.5px;
  white-space:nowrap;
}
.plSeason span{color:#3d195b;}
""".strip()
```

```python
# ============================================================
# CELL 5 — PHASE 1A
# PRESEASON PRIOR + TEAM-STRENGTH ENGINE
# MODEL-ONLY BASELINE • NO DOWNSTREAM DECISION LOGIC
# ============================================================
# Contract:
#   Layer 0 raw descriptive evidence
#       -> preseason priors
#       -> current-season signals
#       -> team-strength state
#
# This cell NEVER creates fixture ratings, player expected points,
# captaincy, squad decisions, or market-calibrated probabilities.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import math
import re
import numpy as np
import pandas as pd

if "vx_phase0_assert_cleanroom" not in globals():
    raise RuntimeError("Run Cell 1 PHASE 0 before Phase 1.")
vx_phase0_assert_cleanroom("before Phase 1")

# ----------------------------------------------------------------
# PHASE 1 MODEL VERSION + SEASON CLOCK
# ----------------------------------------------------------------
model_version = "VORTEX-V2-PHASE1-COMPONENT-TEAM-STRENGTH"

def _vx_season_label(ts=None):
    t = pd.Timestamp(ts or datetime.now(timezone.utc))
    if t.tzinfo is None:
        t = t.tz_localize("UTC")
    y = int(t.year)
    start = y if int(t.month) >= 7 else y - 1
    return f"{start}-{str(start + 1)[-2:]}"

CURRENT_SEASON = _vx_season_label(data_snapshot_timestamp)
_start_year = int(CURRENT_SEASON.split("-")[0])
PREVIOUS_SEASON = f"{_start_year - 1}-{str(_start_year)[-2:]}"

# ----------------------------------------------------------------
# REQUIRED RAW MATCH EVIDENCE
# ----------------------------------------------------------------
PHASE1_MATCH_FIELDS = (
    "season",
    "match_date",
    "team_id",
    "opponent_id",
    "venue",
    "npxg_for",
    "npxga",
    "big_chances_for",
    "big_chances_against",
    "set_piece_xg_for",
    "set_piece_xg_against",
    "penalty_xg_for",
    "penalty_xg_against",
    "red_card_minutes_for",
    "red_card_minutes_against",
)

PHASE1_PRIOR_FIELDS = (
    "team_id",
    "prior_attack_overall",
    "prior_defence_overall",
    "prior_attack_home",
    "prior_defence_home",
    "prior_attack_away",
    "prior_defence_away",
    "prior_penalty_attack_signal",
    "prior_penalty_defence_signal",
    "prior_uncertainty",
    "promoted_attack_prior_adjustment",
    "promoted_defence_prior_adjustment",
    "promoted_team_uncertainty",
)

PHASE1_SIGNAL_FIELDS = (
    "team_id",
    "season_attack_signal",
    "season_defence_signal",
    "season_home_attack_signal",
    "season_home_defence_signal",
    "season_away_attack_signal",
    "season_away_defence_signal",
    "season_penalty_attack_signal",
    "season_penalty_defence_signal",
    "opponent_adjusted_npxg_for",
    "opponent_adjusted_npxga",
    "recent_form_weight",
    "sample_size",
)

# ----------------------------------------------------------------
# APPROVED RAW INPUT FILE DISCOVERY
# ----------------------------------------------------------------
# Phase 0 owns the raw-table names. Phase 1 may hydrate the empty
# team_match_xg_raw table only from an explicit descriptive raw file.
# No alternative strength proxy is substituted when it is absent.
PHASE1_RAW_INPUT_DIR = Path(PROJECT_ROOT) / "MODEL_INPUTS"
PHASE1_RAW_INPUT_DIR.mkdir(parents=True, exist_ok=True)

# Preferred Drive evidence discovered during the clean-room audit.
# This source has full 2025/26 Premier League match coverage plus current-season
# folders. We intentionally ignore any rating/Elo products that may coexist in
# that repository; only descriptive matches/shots/incidents are eligible here.
PHASE1_FPL_CORE_ROOT = Path(FPL_VORTEX_DATA_ROOT) / "00_SOURCE_ARCHIVE" / "FPL_CORE" / "data"
PHASE1_FPL_CORE_ADAPTER_ERROR = None
PHASE1_FPL_CORE_SOURCE_USED = False
# Reset cross-season identity evidence on every Phase-1 execution so a rerun in
# the same notebook session cannot inherit stale team identity state.
PHASE1_PREVIOUS_SEASON_TEAM_NAMES = []
PHASE1_CONTINUING_CURRENT_TEAM_IDS = []


def _vx_norm_team_name(value):
    text = str(value or "").strip().lower()
    for a, b in (("&", "and"), ("-", " "), ("_", " "), (".", ""), ("'", "")):
        text = text.replace(a, b)
    text = " ".join(text.split())
    aliases = {
        "man city": "manchester city",
        "man utd": "manchester united",
        "man united": "manchester united",
        "spurs": "tottenham hotspur",
        "tottenham": "tottenham hotspur",
        "wolves": "wolverhampton wanderers",
        "wolverhampton": "wolverhampton wanderers",
        "west ham": "west ham united",
        "brighton": "brighton and hove albion",
        "newcastle": "newcastle united",
        "nottingham forest": "nottingham forest",
        "nottm forest": "nottingham forest",
        "leeds": "leeds united",
    }
    return aliases.get(text, text)


def _vx_pick_col(frame, aliases, label, required=True):
    lut = {str(c).strip().lower(): c for c in frame.columns}
    for name in aliases:
        if str(name).lower() in lut:
            return lut[str(name).lower()]
    if required:
        raise RuntimeError(f"FPL_CORE adapter: cannot prove column for {label}; tried {aliases}")
    return None


def _vx_read_many_csv(paths):
    frames = []
    for path in sorted(set(Path(x) for x in paths)):
        try:
            f = pd.read_csv(path)
        except Exception as exc:
            raise RuntimeError(f"FPL_CORE adapter cannot read {path}: {type(exc).__name__}: {exc}")
        if not f.empty:
            f = f.copy()
            f["__source_file"] = str(path)
            frames.append(f)
    return pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()


def _vx_fpl_core_season_dir(start_year):
    return PHASE1_FPL_CORE_ROOT / f"{start_year}-{start_year + 1}"


def _vx_fpl_core_pl_root(start_year):
    return _vx_fpl_core_season_dir(start_year) / "By Tournament" / "Premier League"


def _vx_provider_team_key(value):
    """Canonical key for provider team identifiers such as 39, 39.0 or '39'."""
    if value is None or (isinstance(value, float) and np.isnan(value)):
        return None
    try:
        num = float(value)
        if np.isfinite(num) and abs(num - round(num)) < 1e-9:
            return str(int(round(num)))
    except Exception:
        pass
    text = str(value).strip()
    return text if text else None


def _vx_fpl_core_season_team_maps(start_year):
    """Read the season-local teams.csv and expose explicit code/id -> club-name maps.

    FPL_CORE match files use provider identifiers that can differ from official
    FPL's current-season `id`. 2025/26 `matches.csv`, for example, stores the
    `code` field (e.g. Wolves=39, Man City=43) in home_team/away_team.
    We therefore prove the key space from the season-local teams table rather
    than treating those numbers as cross-season identities.
    """
    season_dir = _vx_fpl_core_season_dir(start_year)
    candidates = [season_dir / "teams.csv"]
    if not candidates[0].exists():
        # Safe fallback to a season snapshot when the master table is absent.
        candidates.extend(sorted((season_dir / "By Tournament" / "Premier League").rglob("teams.csv")))
        candidates.extend(sorted((season_dir / "By Gameweek").rglob("teams.csv")))
    team_file = next((p for p in candidates if p.exists()), None)
    if team_file is None:
        raise RuntimeError(f"FPL_CORE adapter: no season-local teams.csv for {start_year}-{start_year + 1}")

    tf = pd.read_csv(team_file)
    if tf.empty:
        raise RuntimeError(f"FPL_CORE adapter: empty teams.csv at {team_file}")
    name_col = _vx_pick_col(tf, ("name", "team_name", "fotmob_name"), "season team name")
    code_col = _vx_pick_col(tf, ("code",), "season team code", required=False)
    id_col = _vx_pick_col(tf, ("id", "team_id"), "season team id", required=False)
    if code_col is None and id_col is None:
        raise RuntimeError(f"FPL_CORE adapter: teams.csv has neither code nor id at {team_file}")

    def _build_map(key_col):
        out = {}
        if key_col is None:
            return out
        for _, tr in tf.iterrows():
            k = _vx_provider_team_key(tr.get(key_col))
            nm = str(tr.get(name_col, "")).strip()
            if k is None or not nm:
                continue
            # A single key must never point to two clubs inside the same key space.
            if k in out and _vx_norm_team_name(out[k]) != _vx_norm_team_name(nm):
                raise RuntimeError(
                    f"FPL_CORE adapter: duplicate season team key {k!r} in {key_col}: "
                    f"{out[k]!r} vs {nm!r}"
                )
            out[k] = nm
        return out

    code_map = _build_map(code_col)
    id_map = _build_map(id_col)
    return {"code": code_map, "id": id_map, "source": team_file}


def _vx_resolve_match_team_map(matches, home_col, away_col, start_year):
    """Prove whether match home/away identifiers use season team code or id."""
    maps = _vx_fpl_core_season_team_maps(start_year)
    raw_vals = pd.concat([matches[home_col], matches[away_col]], ignore_index=True)
    keys = {_vx_provider_team_key(v) for v in raw_vals.dropna().tolist()}
    keys.discard(None)
    if not keys:
        raise RuntimeError("FPL_CORE adapter: match file contains no team identifiers")

    coverage = {}
    for label in ("code", "id"):
        m = maps[label]
        coverage[label] = sum(k in m for k in keys) / float(len(keys)) if m else 0.0

    best = max(coverage, key=coverage.get)
    best_cov = float(coverage[best])
    if best_cov < 0.95:
        raise RuntimeError(
            "FPL_CORE adapter: cannot prove match-team identifier space from season teams.csv; "
            f"code_coverage={coverage['code']:.3f}, id_coverage={coverage['id']:.3f}, "
            f"sample_keys={sorted(keys)[:10]}"
        )
    chosen = maps[best]
    unresolved = sorted(k for k in keys if k not in chosen)
    if unresolved:
        raise RuntimeError(
            f"FPL_CORE adapter: unresolved match team identifiers using {best}: {unresolved[:10]}"
        )
    return chosen, best, maps["source"]


def _vx_synthetic_legacy_team_id(name):
    # Stable negative ID for previous-season opponents not present in the
    # current FPL team list. It is used only for opponent adjustment; it never
    # becomes a current-team output row.
    import hashlib
    digest = hashlib.sha1(_vx_norm_team_name(name).encode("utf-8")).hexdigest()[:8]
    return -int(digest, 16)


def _vx_fpl_core_side_from_shot(row, home_name, away_name, team_col=None, side_col=None):
    if side_col is not None:
        v = row.get(side_col)
        if isinstance(v, (bool, np.bool_)):
            return "H" if bool(v) else "A"
        t = str(v).strip().lower()
        if t in {"h", "home", "1", "true"}:
            return "H"
        if t in {"a", "away", "0", "false"}:
            return "A"
    if team_col is not None:
        n = _vx_norm_team_name(row.get(team_col))
        if n == _vx_norm_team_name(home_name):
            return "H"
        if n == _vx_norm_team_name(away_name):
            return "A"
    return None


def _vx_try_build_fpl_core_team_match_raw():
    """Build Phase-1 team-match evidence from canonical FPL_CORE match aggregates.

    The trusted match table already exposes non-penalty xG, set-play xG, total
    xG and big chances. Using those canonical aggregates removes the fragile
    dependency on shot-level situation parsing. Incidents are consulted only
    when a finished match actually contains a red card, because Phase 1 needs
    the duration of the numerical disadvantage for its context weight.

    Cross-season numeric team identifiers are never treated as identities.
    Each season's matches are resolved through that season's teams.csv and are
    then re-keyed to the current official FPL team IDs by normalized club name.
    Relegated historical opponents receive stable negative IDs used only for
    opponent adjustment.
    """
    global PHASE1_PREVIOUS_SEASON_TEAM_NAMES, PHASE1_CONTINUING_CURRENT_TEAM_IDS

    if not PHASE1_FPL_CORE_ROOT.exists():
        return None, None

    current_name_to_id = {}
    for _, tr in teams.iterrows():
        tid = int(tr["id"])
        for nm in (tr.get("name"), tr.get("short_name")):
            if nm is not None and str(nm).strip():
                current_name_to_id[_vx_norm_team_name(nm)] = tid

    def _current_or_legacy_id(name):
        norm = _vx_norm_team_name(name)
        return int(current_name_to_id.get(norm, _vx_synthetic_legacy_team_id(norm)))

    def _as_float(value):
        try:
            out = float(value)
        except Exception:
            return np.nan
        return out if np.isfinite(out) else np.nan

    def _boolish(value):
        if isinstance(value, (bool, np.bool_)):
            return bool(value)
        text = str(value).strip().lower()
        return text in {"1", "true", "yes", "y", "finished", "complete", "completed"}

    def _incident_minute(value, added=None):
        """Return a football minute including stoppage time when separately stored."""
        if value is None or (isinstance(value, float) and np.isnan(value)):
            return np.nan
        if isinstance(value, (int, float, np.integer, np.floating)):
            base = float(value)
        else:
            text = str(value).strip()
            mm = re.match(r"^(\d+)(?:\s*\+\s*(\d+))?", text)
            if not mm:
                return np.nan
            base = float(mm.group(1)) + float(mm.group(2) or 0.0)
        if added is not None and not (isinstance(added, float) and np.isnan(added)):
            try:
                add_v = float(added)
                if np.isfinite(add_v) and add_v > 0 and "+" not in str(value):
                    base += add_v
            except Exception:
                pass
        return base

    all_team_rows = []
    previous_season_team_names = set()
    skipped_red_timing = []

    for start_year in (_start_year - 1, _start_year):
        season_label = f"{start_year}-{str(start_year + 1)[-2:]}"
        pl_root = _vx_fpl_core_pl_root(start_year)
        if not pl_root.exists():
            if start_year == _start_year:
                print(f"ℹ️ Phase 1: no current-season Premier League detail folder at {pl_root}; prior-only mode.")
                continue
            raise RuntimeError(f"FPL_CORE adapter: missing Premier League folder {pl_root}")

        match_files = list(pl_root.rglob("matches.csv"))
        if not match_files:
            if start_year == _start_year:
                print(f"ℹ️ Phase 1: no completed current-season matches under {pl_root}; prior-only mode.")
                continue
            raise RuntimeError(f"FPL_CORE adapter: no matches.csv under {pl_root}")

        matches = _vx_read_many_csv(match_files)
        if matches.empty:
            if start_year == _start_year:
                print(f"ℹ️ Phase 1: current-season match table is empty; prior-only mode.")
                continue
            raise RuntimeError(f"FPL_CORE adapter: previous-season matches.csv files are empty under {pl_root}")

        # Canonical match fields. These are documented FPL_CORE match aggregates
        # and are materially safer than reconstructing the same quantities from
        # shot labels.
        m_id = _vx_pick_col(matches, ("match_id", "id"), "match id")
        m_date = _vx_pick_col(matches, ("kickoff_time", "match_date", "start_time", "date"), "match date")
        m_home = _vx_pick_col(matches, ("home_team", "home_team_name", "home_name"), "home team")
        m_away = _vx_pick_col(matches, ("away_team", "away_team_name", "away_name"), "away team")
        m_npxg_h = _vx_pick_col(matches, ("home_non_penalty_xg", "home_npxg"), "home non-penalty xG")
        m_npxg_a = _vx_pick_col(matches, ("away_non_penalty_xg", "away_npxg"), "away non-penalty xG")
        m_total_h = _vx_pick_col(matches, ("home_expected_goals_xg", "home_xg"), "home total xG")
        m_total_a = _vx_pick_col(matches, ("away_expected_goals_xg", "away_xg"), "away total xG")
        m_set_h = _vx_pick_col(matches, ("home_xg_set_play", "home_set_piece_xg", "home_set_play_xg"), "home set-play xG")
        m_set_a = _vx_pick_col(matches, ("away_xg_set_play", "away_set_piece_xg", "away_set_play_xg"), "away set-play xG")
        m_hbc = _vx_pick_col(matches, ("home_big_chances", "home_big_chance"), "home big chances")
        m_abc = _vx_pick_col(matches, ("away_big_chances", "away_big_chance"), "away big chances")
        m_finished = _vx_pick_col(matches, ("finished", "is_finished", "complete", "completed"), "finished flag", required=False)
        m_red_h = _vx_pick_col(matches, ("home_red_cards", "home_red_card"), "home red-card count", required=False)
        m_red_a = _vx_pick_col(matches, ("away_red_cards", "away_red_card"), "away red-card count", required=False)

        season_team_lookup, season_team_key_space, season_team_source = _vx_resolve_match_team_map(
            matches, m_home, m_away, start_year
        )
        if start_year == _start_year - 1:
            previous_season_team_names.update(
                _vx_norm_team_name(nm) for nm in season_team_lookup.values() if str(nm).strip()
            )
            print(
                f"✅ Phase 1 identity: {season_label} matches use teams.csv `{season_team_key_space}` keys; "
                f"{len(previous_season_team_names)} previous-season PL clubs resolved."
            )

        # A GW directory can be either gameweek-only or cumulative. Sort by the
        # numeric GW encoded in the source path and retain the latest occurrence.
        def _source_gw(path_text):
            mm = re.search(r"(?:^|[/\\])GW(\d+)(?:[/\\]|$)", str(path_text), flags=re.I)
            return int(mm.group(1)) if mm else -1

        matches = matches.copy()
        matches["__match_id_text"] = matches[m_id].astype(str)
        matches["__source_gw"] = matches["__source_file"].map(_source_gw)
        matches = matches.sort_values(["__match_id_text", "__source_gw"]).drop_duplicates(
            subset=["__match_id_text"], keep="last"
        )

        # Read incidents only as a red-card timing source. Current-season
        # incidents may not exist yet; that is harmless when no finished match
        # in the available sample has a red card.
        incident_files = list(pl_root.rglob("incidents.csv"))
        incidents = _vx_read_many_csv(incident_files) if incident_files else pd.DataFrame()
        i_mid = i_min = i_added = i_side = i_card = None
        if not incidents.empty:
            incidents = incidents.drop_duplicates(
                subset=[c for c in incidents.columns if c != "__source_file"]
            ).copy()
            i_mid = _vx_pick_col(incidents, ("match_id",), "incident match id")
            i_min = _vx_pick_col(incidents, ("minute", "incident_minute", "time"), "incident minute")
            i_added = _vx_pick_col(incidents, ("added_time", "stoppage_time"), "incident added time", required=False)
            i_side = _vx_pick_col(incidents, ("team_side", "side", "home_away", "is_home"), "incident team side")
            i_card = _vx_pick_col(
                incidents,
                ("card_type", "card_colour", "card_color", "card", "incident_detail", "detail"),
                "incident card type",
                required=False,
            )
            incidents[i_mid] = incidents[i_mid].astype(str)

        season_rows_before = len(all_team_rows)
        for _, mr in matches.iterrows():
            if m_finished is not None and not _boolish(mr.get(m_finished)):
                continue

            mid = str(mr[m_id])
            home_key = _vx_provider_team_key(mr[m_home])
            away_key = _vx_provider_team_key(mr[m_away])
            if home_key not in season_team_lookup or away_key not in season_team_lookup:
                raise RuntimeError(
                    f"FPL_CORE adapter: unresolved season team key for match_id={mid}: "
                    f"home={home_key!r}, away={away_key!r}, key_space={season_team_key_space}"
                )
            home_name = str(season_team_lookup[home_key])
            away_name = str(season_team_lookup[away_key])
            match_date = pd.to_datetime(mr[m_date], utc=True, errors="coerce")
            if pd.isna(match_date):
                continue

            vals = {
                "hnp": _as_float(mr[m_npxg_h]),
                "anp": _as_float(mr[m_npxg_a]),
                "htot": _as_float(mr[m_total_h]),
                "atot": _as_float(mr[m_total_a]),
                "hset": _as_float(mr[m_set_h]),
                "aset": _as_float(mr[m_set_a]),
                "hbc": _as_float(mr[m_hbc]),
                "abc": _as_float(mr[m_abc]),
            }
            # Missing aggregate statistics mean the fixture is not yet usable as
            # descriptive evidence. Do not turn missing values into zeroes.
            if any(not np.isfinite(v) for v in vals.values()):
                continue
            if any(v < -1e-9 for v in vals.values()):
                raise RuntimeError(f"FPL_CORE adapter: negative descriptive match statistic for match_id={mid}")

            # Total xG and NPxG are the same provider measure; their difference
            # isolates penalty xG. Allow only tiny rounding noise below zero.
            hpen_delta = vals["htot"] - vals["hnp"]
            apen_delta = vals["atot"] - vals["anp"]
            if hpen_delta < -0.03 or apen_delta < -0.03:
                raise RuntimeError(
                    f"FPL_CORE adapter: total xG < NPxG beyond rounding tolerance for match_id={mid}: "
                    f"home_delta={hpen_delta:.3f}, away_delta={apen_delta:.3f}"
                )
            hpen = float(max(0.0, hpen_delta))
            apen = float(max(0.0, apen_delta))

            red_count_h = int(max(0.0, _as_float(mr[m_red_h]))) if m_red_h is not None and np.isfinite(_as_float(mr[m_red_h])) else 0
            red_count_a = int(max(0.0, _as_float(mr[m_red_a]))) if m_red_a is not None and np.isfinite(_as_float(mr[m_red_a])) else 0
            red_h = red_a = 0.0

            if red_count_h or red_count_a:
                # Red-card duration is the only item not recoverable from the
                # aggregate match row. If timing cannot be proven, omit this one
                # distorted match rather than fabricating a minute or failing the
                # entire 380-match season.
                if incidents.empty or i_card is None:
                    skipped_red_timing.append(mid)
                    continue
                inc = incidents.loc[incidents[i_mid].eq(mid)].copy()
                if inc.empty:
                    skipped_red_timing.append(mid)
                    continue
                card_text = (
                    inc[i_card].astype(str).str.lower()
                    .str.replace(r"[_-]+", " ", regex=True)
                    .str.replace(r"\s+", " ", regex=True)
                    .str.strip()
                )
                red_mask = card_text.str.contains(
                    r"\bred\b|\bsecond yellow\b|\b2nd yellow\b|\byellow red\b", regex=True, na=False
                )
                reds = inc.loc[red_mask].copy()
                if len(reds) < red_count_h + red_count_a:
                    skipped_red_timing.append(mid)
                    continue
                reds["__minute"] = [
                    _incident_minute(v, a if i_added is not None else None)
                    for v, a in zip(
                        reds[i_min].tolist(),
                        reds[i_added].tolist() if i_added is not None else [None] * len(reds),
                    )
                ]
                reds["__side"] = reds.apply(
                    lambda r: _vx_fpl_core_side_from_shot(r, home_name, away_name, None, i_side), axis=1
                )
                if reds["__minute"].isna().any() or reds["__side"].isna().any():
                    skipped_red_timing.append(mid)
                    continue
                actual_h = int(reds["__side"].eq("H").sum())
                actual_a = int(reds["__side"].eq("A").sum())
                if actual_h < red_count_h or actual_a < red_count_a:
                    skipped_red_timing.append(mid)
                    continue
                red_h = float(sum(max(0.0, 90.0 - float(x)) for x in reds.loc[reds["__side"].eq("H"), "__minute"]))
                red_a = float(sum(max(0.0, 90.0 - float(x)) for x in reds.loc[reds["__side"].eq("A"), "__minute"]))

            hid = _current_or_legacy_id(home_name)
            aid = _current_or_legacy_id(away_name)
            source_path = Path(str(mr.get("__source_file", "")))
            if source_path.exists():
                source_ts = datetime.fromtimestamp(source_path.stat().st_mtime, tz=timezone.utc).isoformat()
            else:
                source_ts = data_snapshot_timestamp

            common = {
                "season": season_label,
                "match_date": match_date.isoformat(),
                "source_name": "FPL_CORE_CANONICAL_MATCH_AGGREGATES_INCIDENTS",
                "source_timestamp": source_ts,
            }
            all_team_rows.extend([
                {
                    **common,
                    "team_id": hid,
                    "opponent_id": aid,
                    "team_name": home_name,
                    "opponent_name": away_name,
                    "venue": "H",
                    "npxg_for": vals["hnp"],
                    "npxga": vals["anp"],
                    "big_chances_for": vals["hbc"],
                    "big_chances_against": vals["abc"],
                    "set_piece_xg_for": vals["hset"],
                    "set_piece_xg_against": vals["aset"],
                    "penalty_xg_for": hpen,
                    "penalty_xg_against": apen,
                    "red_card_minutes_for": red_h,
                    "red_card_minutes_against": red_a,
                },
                {
                    **common,
                    "team_id": aid,
                    "opponent_id": hid,
                    "team_name": away_name,
                    "opponent_name": home_name,
                    "venue": "A",
                    "npxg_for": vals["anp"],
                    "npxga": vals["hnp"],
                    "big_chances_for": vals["abc"],
                    "big_chances_against": vals["hbc"],
                    "set_piece_xg_for": vals["aset"],
                    "set_piece_xg_against": vals["hset"],
                    "penalty_xg_for": apen,
                    "penalty_xg_against": hpen,
                    "red_card_minutes_for": red_a,
                    "red_card_minutes_against": red_h,
                },
            ])

        n_team_rows = len(all_team_rows) - season_rows_before
        if start_year == _start_year - 1 and n_team_rows < 600:
            # 300 usable matches is a deliberately broad integrity floor; a
            # normal previous PL season should yield close to 760 team-match rows.
            raise RuntimeError(
                f"FPL_CORE adapter: only {n_team_rows // 2} usable previous-season PL matches; "
                "expected at least 300"
            )
        if start_year == _start_year and n_team_rows == 0:
            print("ℹ️ Phase 1: no usable completed current-season aggregate rows yet; prior-only mode.")

    if not all_team_rows:
        raise RuntimeError("FPL_CORE adapter produced no completed-match evidence")

    frame = pd.DataFrame(all_team_rows)
    PHASE1_PREVIOUS_SEASON_TEAM_NAMES = sorted(previous_season_team_names)
    PHASE1_CONTINUING_CURRENT_TEAM_IDS = sorted({
        int(tr["id"])
        for _, tr in teams.iterrows()
        if (
            _vx_norm_team_name(tr.get("name")) in previous_season_team_names
            or _vx_norm_team_name(tr.get("short_name")) in previous_season_team_names
        )
    })
    if len(PHASE1_CONTINUING_CURRENT_TEAM_IDS) < 14:
        raise RuntimeError(
            "FPL_CORE adapter: cross-season roster continuity remains unproven after teams.csv mapping; "
            f"matched_current_clubs={len(PHASE1_CONTINUING_CURRENT_TEAM_IDS)}, "
            f"previous_names={PHASE1_PREVIOUS_SEASON_TEAM_NAMES}"
        )

    if skipped_red_timing:
        print(
            f"ℹ️ Phase 1: omitted {len(set(skipped_red_timing))} red-card-distorted match(es) "
            "whose exact send-off timing could not be proven."
        )

    missing = [c for c in PHASE1_MATCH_FIELDS if c not in frame.columns]
    if missing:
        raise RuntimeError(f"FPL_CORE adapter output missing required fields: {missing}")
    return frame, PHASE1_FPL_CORE_ROOT


def _vx_read_phase1_match_source():
    global PHASE1_FPL_CORE_ADAPTER_ERROR, PHASE1_FPL_CORE_SOURCE_USED
    try:
        core_frame, core_root = _vx_try_build_fpl_core_team_match_raw()
        if core_frame is not None:
            frame = core_frame.copy()
            for c in PHASE0_CONTROL_FIELDS:
                if c not in frame.columns:
                    if c == "model_version":
                        frame[c] = model_version
                    elif c == "data_snapshot_timestamp":
                        frame[c] = data_snapshot_timestamp
                    elif c == "forecast_timestamp":
                        frame[c] = None
                    elif c == "gameweek":
                        frame[c] = pd.Series([pd.NA] * len(frame), dtype="Int64")
                    elif c == "fixture_id":
                        frame[c] = pd.Series([pd.NA] * len(frame), dtype="Int64")
                    elif c == "source_name":
                        frame[c] = "FPL_CORE_CANONICAL_MATCH_AGGREGATES_INCIDENTS"
                    elif c == "source_timestamp":
                        frame[c] = data_snapshot_timestamp
                    elif c == "is_pre_deadline_data":
                        frame[c] = pd.NA
            vx_phase0_assert_no_forbidden_columns(frame, "Phase 1 FPL_CORE adapter")
            PHASE1_FPL_CORE_SOURCE_USED = True
            return frame, core_root
    except Exception as exc:
        PHASE1_FPL_CORE_ADAPTER_ERROR = f"{type(exc).__name__}: {exc}"
        print("⛔ Phase 1 FPL_CORE adapter error:", PHASE1_FPL_CORE_ADAPTER_ERROR)

    for p in _MATCH_CANDIDATES:
        if not p.exists():
            continue
        if p.suffix.lower() == ".parquet":
            frame = pd.read_parquet(p)
        elif p.suffix.lower() == ".csv":
            frame = pd.read_csv(p)
        else:
            frame = pd.read_json(p)
        frame = frame.copy()
        missing = [c for c in PHASE1_MATCH_FIELDS if c not in frame.columns]
        if missing:
            raise RuntimeError(
                f"PHASE 1 raw match source {p.name} missing required fields: {missing}"
            )
        # Add provenance controls if the source file does not already carry them.
        for c in PHASE0_CONTROL_FIELDS:
            if c not in frame.columns:
                if c == "model_version":
                    frame[c] = model_version
                elif c == "data_snapshot_timestamp":
                    frame[c] = data_snapshot_timestamp
                elif c == "forecast_timestamp":
                    frame[c] = None
                elif c == "gameweek":
                    frame[c] = pd.Series([pd.NA] * len(frame), dtype="Int64")
                elif c == "fixture_id":
                    frame[c] = pd.Series([pd.NA] * len(frame), dtype="Int64")
                elif c == "source_name":
                    frame[c] = f"RAW_INPUT:{p.name}"
                elif c == "source_timestamp":
                    frame[c] = datetime.fromtimestamp(p.stat().st_mtime, tz=timezone.utc).isoformat()
                elif c == "is_pre_deadline_data":
                    frame[c] = pd.NA
        vx_phase0_assert_no_forbidden_columns(frame, f"Phase 1 raw match source {p.name}")
        return frame, p
    return None, None

_loaded_match_frame, PHASE1_MATCH_SOURCE_PATH = _vx_read_phase1_match_source()
if _loaded_match_frame is not None:
    team_match_xg_raw = _loaded_match_frame.copy()
    RAW_TABLES["team_match_xg_raw"] = team_match_xg_raw

# ----------------------------------------------------------------
# TEAM CONTEXT CONTRACT
# ----------------------------------------------------------------
# These are descriptive pre-season/context inputs, expressed as LOG-STRENGTH
# adjustments. 0.00 = no known adjustment. Positive attack delta strengthens
# attack. Positive defence delta means MORE expected concession (weaker defence).
#
# Manual context remains authoritative. For unattended Run-all, an untouched
# machine-created template is upgraded to an EXPLICIT provenance-labelled
# baseline. Continuing PL teams get zero known squad/manager/XI adjustment plus
# uncertainty; promoted teams are identified from previous-season PL roster
# continuity and receive conservative pre-validation promoted priors.
PHASE1_CONTEXT_FIELDS = (
    "team_id",
    "promoted_team",
    "promoted_attack_prior_adjustment",
    "promoted_defence_prior_adjustment",
    "promoted_team_uncertainty",
    "squad_arrivals_attack_delta",
    "squad_arrivals_defence_delta",
    "squad_departures_attack_delta",
    "squad_departures_defence_delta",
    "manager_attack_delta",
    "manager_defence_delta",
    "expected_xi_attack_delta",
    "expected_xi_defence_delta",
    "context_uncertainty",
    "context_complete",
    "context_source",
    "context_source_timestamp",
)

PHASE1_CONTEXT_PATH = PHASE1_RAW_INPUT_DIR / "team_context_raw.csv"
PHASE1_AUTO_PROMOTED_ATTACK_ADJUSTMENT = float(math.log(0.82))
PHASE1_AUTO_PROMOTED_DEFENCE_ADJUSTMENT = float(math.log(1.18))
PHASE1_AUTO_PROMOTED_UNCERTAINTY = 0.30
PHASE1_AUTO_BASE_CONTEXT_UNCERTAINTY = 0.08

_PHASE1_CONTEXT_NUMERIC_FIELDS = (
    "promoted_attack_prior_adjustment",
    "promoted_defence_prior_adjustment",
    "promoted_team_uncertainty",
    "squad_arrivals_attack_delta",
    "squad_arrivals_defence_delta",
    "squad_departures_attack_delta",
    "squad_departures_defence_delta",
    "manager_attack_delta",
    "manager_defence_delta",
    "expected_xi_attack_delta",
    "expected_xi_defence_delta",
    "context_uncertainty",
)

def _vx_context_template():
    base = teams[["id", "name", "short_name"]].rename(columns={"id": "team_id"}).copy()
    base["promoted_team"] = pd.NA
    for c in _PHASE1_CONTEXT_NUMERIC_FIELDS:
        base[c] = np.nan
    base["context_complete"] = False
    base["context_source"] = ""
    base["context_source_timestamp"] = ""
    return base

def _vx_context_is_untouched_template(frame):
    """True only for the machine-created blank template; partial edits are preserved."""
    if not isinstance(frame, pd.DataFrame) or frame.empty:
        return False
    required = {"team_id", "promoted_team", "context_complete", "context_source", *_PHASE1_CONTEXT_NUMERIC_FIELDS}
    if not required.issubset(frame.columns):
        return False
    complete_text = frame["context_complete"].astype(str).str.strip().str.lower()
    if (~complete_text.isin({"false", "0", "no", "n", "", "nan", "<na>"})).any():
        return False
    if frame["context_source"].fillna("").astype(str).str.strip().ne("").any():
        return False
    # Promoted/not-promoted flags alone are descriptive scaffold metadata.
    # They do not make this a manually calibrated context file. This allows a
    # persisted machine template with only those flags filled to be safely
    # completed from trusted FPL_CORE roster continuity. Any numeric adjustment
    # or provenance below still protects genuine manual work from overwrite.
    promoted_text = frame["promoted_team"].fillna("").astype(str).str.strip().str.lower()
    _allowed_promoted_scaffold = {"", "nan", "<na>", "none", "true", "false", "1", "0", "yes", "no", "y", "n"}
    if (~promoted_text.isin(_allowed_promoted_scaffold)).any():
        return False
    for c in _PHASE1_CONTEXT_NUMERIC_FIELDS:
        if pd.to_numeric(frame[c], errors="coerce").notna().any():
            return False
    return True

def _vx_auto_context_from_match_evidence(match_frame):
    if not isinstance(match_frame, pd.DataFrame) or match_frame.empty:
        return None, "NO_TRUSTED_MATCH_EVIDENCE"
    prev = match_frame.loc[match_frame["season"].astype(str).eq(PREVIOUS_SEASON)].copy()
    if prev.empty:
        return None, f"NO_PREVIOUS_SEASON_EVIDENCE_{PREVIOUS_SEASON}"

    current_ids = set(pd.to_numeric(teams["id"], errors="coerce").dropna().astype(int))

    # Prefer the explicit previous-season normalized club-name set captured by
    # the FPL_CORE adapter. This avoids the cross-season-ID bug that previously
    # classified every current club as unmatched. Fall back to IDs only for a
    # manually supplied raw table that does not carry names.
    explicit_continuing = set(
        int(x) for x in globals().get("PHASE1_CONTINUING_CURRENT_TEAM_IDS", [])
        if pd.notna(x)
    )
    if explicit_continuing:
        prev_ids = explicit_continuing & current_ids
    elif "team_name" in prev.columns:
        prev_names = set(prev["team_name"].dropna().astype(str).map(_vx_norm_team_name))
        prev_ids = set()
        for _, _tr in teams.iterrows():
            if (
                _vx_norm_team_name(_tr.get("name")) in prev_names
                or _vx_norm_team_name(_tr.get("short_name")) in prev_names
            ):
                prev_ids.add(int(_tr["id"]))
    else:
        prev_ids = set(pd.to_numeric(prev["team_id"], errors="coerce").dropna().astype(int)) & current_ids

    # A normal PL rollover retains about 17 clubs. These broad rails catch a
    # broken team-name map rather than accidentally labelling many clubs promoted.
    if len(prev_ids) < 14:
        return None, f"ROSTER_CONTINUITY_UNPROVEN_ONLY_{len(prev_ids)}_CURRENT_TEAMS_MATCHED"
    promoted_ids = current_ids - prev_ids
    if len(promoted_ids) > 6:
        return None, f"ROSTER_CONTINUITY_IMPLAUSIBLE_{len(promoted_ids)}_PROMOTED_IDS"

    ts = datetime.now(timezone.utc).isoformat()
    base = teams[["id", "name", "short_name"]].rename(columns={"id": "team_id"}).copy()
    base["team_id"] = pd.to_numeric(base["team_id"], errors="raise").astype(int)
    base["promoted_team"] = base["team_id"].isin(promoted_ids)
    base["promoted_attack_prior_adjustment"] = np.where(
        base["promoted_team"], PHASE1_AUTO_PROMOTED_ATTACK_ADJUSTMENT, 0.0
    )
    base["promoted_defence_prior_adjustment"] = np.where(
        base["promoted_team"], PHASE1_AUTO_PROMOTED_DEFENCE_ADJUSTMENT, 0.0
    )
    base["promoted_team_uncertainty"] = np.where(
        base["promoted_team"], PHASE1_AUTO_PROMOTED_UNCERTAINTY, 0.0
    )
    for c in (
        "squad_arrivals_attack_delta",
        "squad_arrivals_defence_delta",
        "squad_departures_attack_delta",
        "squad_departures_defence_delta",
        "manager_attack_delta",
        "manager_defence_delta",
        "expected_xi_attack_delta",
        "expected_xi_defence_delta",
    ):
        base[c] = 0.0
    base["context_uncertainty"] = PHASE1_AUTO_BASE_CONTEXT_UNCERTAINTY
    base["context_complete"] = True
    base["context_source"] = "AUTO_BASELINE_FPL_CORE_ROSTER_CONTINUITY_V1"
    base["context_source_timestamp"] = ts
    return base, f"AUTO_BASELINE_READY_PROMOTED_IDS={sorted(promoted_ids)}"

if PHASE1_CONTEXT_PATH.exists():
    team_context_raw = pd.read_csv(PHASE1_CONTEXT_PATH)
else:
    team_context_raw = _vx_context_template()
    team_context_raw.to_csv(PHASE1_CONTEXT_PATH, index=False)

if _vx_context_is_untouched_template(team_context_raw):
    _auto_context, _auto_context_status = _vx_auto_context_from_match_evidence(team_match_xg_raw)
    if _auto_context is not None:
        team_context_raw = _auto_context
        team_context_raw.to_csv(PHASE1_CONTEXT_PATH, index=False)
        print(f"✅ Phase 1 explicit auto-context persisted: {_auto_context_status}")
    else:
        print(f"⚠️ Phase 1 blank context template retained: {_auto_context_status}")

_missing_context_cols = [c for c in PHASE1_CONTEXT_FIELDS if c not in team_context_raw.columns]
if _missing_context_cols:
    raise RuntimeError(f"PHASE 1 team context missing fields: {_missing_context_cols}")

# ----------------------------------------------------------------
# INPUT VALIDATION
# ----------------------------------------------------------------
def _vx_phase1_validate_match_frame(frame):
    if not isinstance(frame, pd.DataFrame) or frame.empty:
        return False, ["team_match_xg_raw has no approved rows"]
    missing = [c for c in PHASE1_MATCH_FIELDS if c not in frame.columns]
    issues = [f"missing field: {c}" for c in missing]
    if issues:
        return False, issues

    f = frame.copy()
    for c in (
        "team_id", "opponent_id", "npxg_for", "npxga",
        "big_chances_for", "big_chances_against",
        "set_piece_xg_for", "set_piece_xg_against",
        "penalty_xg_for", "penalty_xg_against",
        "red_card_minutes_for", "red_card_minutes_against",
    ):
        f[c] = pd.to_numeric(f[c], errors="coerce")

    if f["team_id"].isna().any() or f["opponent_id"].isna().any():
        issues.append("team_id/opponent_id contains null or non-numeric rows")
    if (~f["venue"].astype(str).str.upper().isin(["H", "A"])).any():
        issues.append("venue must be H or A")
    for c in ("npxg_for", "npxga", "set_piece_xg_for", "set_piece_xg_against", "penalty_xg_for", "penalty_xg_against"):
        if f[c].isna().any() or (f[c] < 0).any():
            issues.append(f"{c} contains missing/negative values")
    for c in ("big_chances_for", "big_chances_against", "red_card_minutes_for", "red_card_minutes_against"):
        if f[c].isna().any() or (f[c] < 0).any():
            issues.append(f"{c} contains missing/negative values")

    return len(issues) == 0, issues

def _vx_bool(value):
    if value is True or value is False:
        return bool(value)
    if value is None or (isinstance(value, float) and np.isnan(value)):
        return False
    text = str(value).strip().lower()
    if text in {"true", "1", "yes", "y"}:
        return True
    if text in {"false", "0", "no", "n", ""}:
        return False
    raise ValueError(f"Cannot parse boolean value: {value!r}")

def _vx_phase1_validate_context(frame):
    issues = []
    if not isinstance(frame, pd.DataFrame) or frame.empty:
        return False, ["team_context_raw is empty"]
    expected_ids = set(pd.to_numeric(teams["id"], errors="coerce").dropna().astype(int))
    got_ids = set(pd.to_numeric(frame["team_id"], errors="coerce").dropna().astype(int))
    if expected_ids != got_ids:
        issues.append(f"context team IDs mismatch: missing={sorted(expected_ids-got_ids)}, extra={sorted(got_ids-expected_ids)}")
    if "context_complete" not in frame.columns:
        issues.append("context_complete field is missing")
    else:
        try:
            complete_flags = frame["context_complete"].map(_vx_bool)
            if not complete_flags.all():
                issues.append("every team must have context_complete=True")
        except Exception as exc:
            issues.append(f"context_complete contains invalid boolean values: {exc}")
    numeric_context = [
        c for c in PHASE1_CONTEXT_FIELDS
        if c.endswith("_delta") or c.endswith("_adjustment") or c.endswith("_uncertainty")
    ]
    for c in numeric_context:
        vals = pd.to_numeric(frame[c], errors="coerce")
        if vals.isna().any():
            issues.append(f"{c} contains missing/non-numeric values")
    if "context_source" in frame.columns:
        if frame["context_source"].fillna("").astype(str).str.strip().eq("").any():
            issues.append("context_source must be non-empty for every team")
    if "context_source_timestamp" in frame.columns:
        parsed_ts = pd.to_datetime(frame["context_source_timestamp"], utc=True, errors="coerce")
        if parsed_ts.isna().any():
            issues.append("context_source_timestamp contains missing/invalid values")
    return len(issues) == 0, issues

_match_ok, _match_issues = _vx_phase1_validate_match_frame(team_match_xg_raw)
_context_ok, _context_issues = _vx_phase1_validate_context(team_context_raw)

# ----------------------------------------------------------------
# RED-CARD CONTEXT + EXACT 90-DAY HALF-LIFE MATCH RECENCY
# ----------------------------------------------------------------
def _vx_red_context_weight(row):
    # NPxG from heavily non-11v11 game states is downweighted, never deleted.
    rf = float(row.get("red_card_minutes_for", 0.0) or 0.0)
    ra = float(row.get("red_card_minutes_against", 0.0) or 0.0)
    distorted = min(90.0, abs(rf - ra))
    return float(max(0.35, 1.0 - 0.65 * distorted / 90.0))

def _vx_time_decay_weight(match_date, anchor_ts, half_life_days=90.0):
    d = pd.Timestamp(match_date)
    a = pd.Timestamp(anchor_ts)
    if d.tzinfo is None:
        d = d.tz_localize("UTC")
    else:
        d = d.tz_convert("UTC")
    if a.tzinfo is None:
        a = a.tz_localize("UTC")
    else:
        a = a.tz_convert("UTC")
    age_days = max(0.0, (a - d).total_seconds() / 86400.0)
    # Exact Phase-2 recency contract: w_match = 0.5 ** (days_since_match / 90).
    return float(0.5 ** (age_days / float(half_life_days)))

# ----------------------------------------------------------------
# OPPONENT-ADJUSTED PRESEASON PRIOR
# ----------------------------------------------------------------
def _vx_weighted_mean(values, weights):
    v = np.asarray(values, dtype=float)
    w = np.asarray(weights, dtype=float)
    ok = np.isfinite(v) & np.isfinite(w) & (w > 0)
    if not ok.any():
        return np.nan
    return float(np.sum(v[ok] * w[ok]) / np.sum(w[ok]))

def _vx_shrink(raw_strength, n_eff, prior_equiv=4.0):
    if not np.isfinite(raw_strength):
        return 1.0
    n_eff = max(0.0, float(n_eff))
    return float((raw_strength * n_eff + 1.0 * prior_equiv) / (n_eff + prior_equiv))

def _vx_build_prior(match_frame, context_frame):
    f = match_frame.loc[match_frame["season"].astype(str).eq(PREVIOUS_SEASON)].copy()
    if f.empty:
        raise RuntimeError(f"No previous-season rows found for {PREVIOUS_SEASON}")

    f["match_date"] = pd.to_datetime(f["match_date"], utc=True, errors="coerce")
    if f["match_date"].isna().any():
        raise RuntimeError("Previous-season match_date contains invalid rows")

    num_cols = [
        "team_id","opponent_id","npxg_for","npxga","big_chances_for","big_chances_against",
        "set_piece_xg_for","set_piece_xg_against","penalty_xg_for","penalty_xg_against",
        "red_card_minutes_for","red_card_minutes_against"
    ]
    for c in num_cols:
        f[c] = pd.to_numeric(f[c], errors="raise")
    f["team_id"] = f["team_id"].astype(int)
    f["opponent_id"] = f["opponent_id"].astype(int)
    f["venue"] = f["venue"].astype(str).str.upper()
    f["_w"] = f.apply(_vx_red_context_weight, axis=1)

    mu_home = _vx_weighted_mean(f.loc[f["venue"].eq("H"), "npxg_for"], f.loc[f["venue"].eq("H"), "_w"])
    mu_away = _vx_weighted_mean(f.loc[f["venue"].eq("A"), "npxg_for"], f.loc[f["venue"].eq("A"), "_w"])
    if not np.isfinite(mu_home) or not np.isfinite(mu_away) or mu_home <= 0 or mu_away <= 0:
        raise RuntimeError("Cannot estimate previous-season venue NPxG baselines")

    # Penalties are a separate scoring component. NPxG remains the non-penalty
    # process baseline; penalty xG is added back explicitly downstream.
    pen_home = _vx_weighted_mean(f.loc[f["venue"].eq("H"), "penalty_xg_for"], f.loc[f["venue"].eq("H"), "_w"])
    pen_away = _vx_weighted_mean(f.loc[f["venue"].eq("A"), "penalty_xg_for"], f.loc[f["venue"].eq("A"), "_w"])
    pen_for_league = _vx_weighted_mean(f["penalty_xg_for"], f["_w"])
    pen_against_league = _vx_weighted_mean(f["penalty_xg_against"], f["_w"])
    if any((not np.isfinite(x) or x < 0) for x in (pen_home, pen_away, pen_for_league, pen_against_league)):
        raise RuntimeError("Cannot estimate previous-season penalty-xG baselines")

    team_ids = sorted(set(pd.to_numeric(teams["id"], errors="coerce").dropna().astype(int)))
    ah = {t: 1.0 for t in team_ids}
    dh = {t: 1.0 for t in team_ids}
    aa = {t: 1.0 for t in team_ids}
    da = {t: 1.0 for t in team_ids}

    # Fixed-point opponent adjustment. Attack >1 means stronger scoring process.
    # Defence >1 means weaker defence / more concession.
    for _ in range(20):
        new_ah, new_dh, new_aa, new_da = {}, {}, {}, {}
        for t in team_ids:
            h = f.loc[(f["team_id"].eq(t)) & (f["venue"].eq("H"))]
            a = f.loc[(f["team_id"].eq(t)) & (f["venue"].eq("A"))]

            h_att_ratio = h["npxg_for"].to_numpy() / np.maximum(
                0.05, mu_home * np.array([da.get(int(o), 1.0) for o in h["opponent_id"]])
            )
            h_def_ratio = h["npxga"].to_numpy() / np.maximum(
                0.05, mu_away * np.array([aa.get(int(o), 1.0) for o in h["opponent_id"]])
            )
            a_att_ratio = a["npxg_for"].to_numpy() / np.maximum(
                0.05, mu_away * np.array([dh.get(int(o), 1.0) for o in a["opponent_id"]])
            )
            a_def_ratio = a["npxga"].to_numpy() / np.maximum(
                0.05, mu_home * np.array([ah.get(int(o), 1.0) for o in a["opponent_id"]])
            )

            new_ah[t] = _vx_shrink(_vx_weighted_mean(h_att_ratio, h["_w"]), h["_w"].sum(), 4.0)
            new_dh[t] = _vx_shrink(_vx_weighted_mean(h_def_ratio, h["_w"]), h["_w"].sum(), 4.0)
            new_aa[t] = _vx_shrink(_vx_weighted_mean(a_att_ratio, a["_w"]), a["_w"].sum(), 4.0)
            new_da[t] = _vx_shrink(_vx_weighted_mean(a_def_ratio, a["_w"]), a["_w"].sum(), 4.0)

        # Geometric recentering prevents scale drift during iteration.
        def _recenter(d):
            vals = np.array([max(0.20, float(v)) for v in d.values()], dtype=float)
            gm = float(np.exp(np.mean(np.log(vals))))
            return {k: float(max(0.20, min(3.0, v / gm))) for k, v in d.items()}
        ah, dh, aa, da = map(_recenter, (new_ah, new_dh, new_aa, new_da))

    ctx = context_frame.copy()
    ctx["team_id"] = pd.to_numeric(ctx["team_id"], errors="raise").astype(int)
    ctx = ctx.set_index("team_id")

    rows = []
    for t in team_ids:
        team_rows = f.loc[f["team_id"].eq(t)]
        n = float(team_rows["_w"].sum())

        # Major squad, manager and expected-XI changes are applied exactly once
        # to the preseason prior as log-strength deltas.
        c = ctx.loc[t]
        att_delta = (
            float(c["squad_arrivals_attack_delta"])
            + float(c["squad_departures_attack_delta"])
            + float(c["manager_attack_delta"])
            + float(c["expected_xi_attack_delta"])
        )
        def_delta = (
            float(c["squad_arrivals_defence_delta"])
            + float(c["squad_departures_defence_delta"])
            + float(c["manager_defence_delta"])
            + float(c["expected_xi_defence_delta"])
        )

        promoted = _vx_bool(c["promoted_team"])
        promoted_att = float(c["promoted_attack_prior_adjustment"]) if promoted else 0.0
        promoted_def = float(c["promoted_defence_prior_adjustment"]) if promoted else 0.0
        promoted_unc = float(c["promoted_team_uncertainty"]) if promoted else 0.0

        # Temporary promoted adjustment lives ONLY here in the prior.
        ah_t = ah[t] * math.exp(att_delta + promoted_att)
        aa_t = aa[t] * math.exp(att_delta + promoted_att)
        dh_t = dh[t] * math.exp(def_delta + promoted_def)
        da_t = da[t] * math.exp(def_delta + promoted_def)

        overall_att = math.sqrt(max(0.05, ah_t) * max(0.05, aa_t))
        overall_def = math.sqrt(max(0.05, dh_t) * max(0.05, da_t))

        # Known injuries/suspensions enter uncertainty. Their strength impact
        # must be represented explicitly through expected-XI deltas, never by
        # a guessed per-player penalty.
        injury_count = int(
            pd.to_numeric(injuries_raw.get("team_id", pd.Series(dtype=float)), errors="coerce").eq(t).sum()
        )
        suspension_count = int(
            pd.to_numeric(suspensions_raw.get("team_id", pd.Series(dtype=float)), errors="coerce").eq(t).sum()
        )
        availability_unc = min(0.20, 0.025 * injury_count + 0.04 * suspension_count)
        sampling_unc = min(0.35, 1.0 / math.sqrt(max(1.0, n)))
        prior_unc = float(min(
            1.0,
            max(0.02, sampling_unc + float(c["context_uncertainty"]) + promoted_unc + availability_unc)
        ))

        # Penalty opportunity is sparse, so its separate attack/defence signals
        # receive heavy shrinkage and conservative caps. They never contaminate
        # the non-penalty attack/defence strength.
        pen_att_raw = float(_vx_weighted_mean(team_rows["penalty_xg_for"], team_rows["_w"]) / max(0.01, pen_for_league))
        pen_def_raw = float(_vx_weighted_mean(team_rows["penalty_xg_against"], team_rows["_w"]) / max(0.01, pen_against_league))
        pen_att_prior = float(np.clip(_vx_shrink(pen_att_raw, n, 12.0), 0.60, 1.40))
        pen_def_prior = float(np.clip(_vx_shrink(pen_def_raw, n, 12.0), 0.60, 1.40))

        rows.append({
            "team_id": t,
            "prior_attack_overall": overall_att,
            "prior_defence_overall": overall_def,
            "prior_attack_home": ah_t,
            "prior_defence_home": dh_t,
            "prior_attack_away": aa_t,
            "prior_defence_away": da_t,
            "prior_penalty_attack_signal": pen_att_prior,
            "prior_penalty_defence_signal": pen_def_prior,
            "prior_uncertainty": prior_unc,
            "promoted_attack_prior_adjustment": promoted_att,
            "promoted_defence_prior_adjustment": promoted_def,
            "promoted_team_uncertainty": promoted_unc,
            "previous_season_effective_matches": n,
        })

    prior = pd.DataFrame(rows)
    return prior, {
        "league_home_npxg": float(mu_home),
        "league_away_npxg": float(mu_away),
        "league_home_penalty_xg": float(max(0.0, pen_home)),
        "league_away_penalty_xg": float(max(0.0, pen_away)),
    }

# ----------------------------------------------------------------
# CURRENT-SEASON SIGNALS
# ----------------------------------------------------------------
def _vx_safe_ratio(value, denom, floor=0.05):
    return float(value) / max(float(floor), float(denom))

def _vx_build_current_signals(match_frame, prior, venue_baselines):
    f = match_frame.loc[match_frame["season"].astype(str).eq(CURRENT_SEASON)].copy()
    team_ids = sorted(set(pd.to_numeric(teams["id"], errors="coerce").dropna().astype(int)))
    p = prior.set_index("team_id")
    anchor = forecast_timestamp or data_snapshot_timestamp or datetime.now(timezone.utc).isoformat()

    if f.empty:
        # Valid preseason state: no current-season matches yet.
        rows = []
        for t in team_ids:
            rows.append({
                "team_id": t,
                "season_attack_signal": 1.0,
                "season_defence_signal": 1.0,
                "season_home_attack_signal": 1.0,
                "season_home_defence_signal": 1.0,
                "season_away_attack_signal": 1.0,
                "season_away_defence_signal": 1.0,
                "season_penalty_attack_signal": 1.0,
                "season_penalty_defence_signal": 1.0,
                "opponent_adjusted_npxg_for": np.nan,
                "opponent_adjusted_npxga": np.nan,
                "recent_form_weight": 0.0,
                "sample_size": 0.0,
                "home_sample_size": 0.0,
                "away_sample_size": 0.0,
                "home_recent_form_weight": 0.0,
                "away_recent_form_weight": 0.0,
            })
        return pd.DataFrame(rows)

    f["match_date"] = pd.to_datetime(f["match_date"], utc=True, errors="coerce")
    if f["match_date"].isna().any():
        raise RuntimeError("Current-season match_date contains invalid rows")
    for c in (
        "team_id","opponent_id","npxg_for","npxga","big_chances_for","big_chances_against",
        "set_piece_xg_for","set_piece_xg_against","penalty_xg_for","penalty_xg_against",
        "red_card_minutes_for","red_card_minutes_against"
    ):
        f[c] = pd.to_numeric(f[c], errors="raise")
    f["team_id"] = f["team_id"].astype(int)
    f["opponent_id"] = f["opponent_id"].astype(int)
    f["venue"] = f["venue"].astype(str).str.upper()
    f["_red_w"] = f.apply(_vx_red_context_weight, axis=1)
    f["_time_w"] = f["match_date"].map(lambda x: _vx_time_decay_weight(x, anchor))
    f["_w"] = f["_red_w"] * f["_time_w"]

    mu_h = float(venue_baselines["league_home_npxg"])
    mu_a = float(venue_baselines["league_away_npxg"])

    # League feature baselines are estimated from the CURRENT raw season only.
    # They normalize unlike units (counts vs xG) before component blending.
    def _league_mean(col):
        val = _vx_weighted_mean(f[col], f["_w"])
        return max(0.05, float(val)) if np.isfinite(val) else 0.05

    bc_for_mu = _league_mean("big_chances_for")
    bc_against_mu = _league_mean("big_chances_against")
    sp_for_mu = _league_mean("set_piece_xg_for")
    sp_against_mu = _league_mean("set_piece_xg_against")
    pen_for_observed = _vx_weighted_mean(f["penalty_xg_for"], f["_w"])
    pen_against_observed = _vx_weighted_mean(f["penalty_xg_against"], f["_w"])
    pen_for_mu = max(0.05, float(pen_for_observed)) if np.isfinite(pen_for_observed) else 0.05
    pen_against_mu = max(0.05, float(pen_against_observed)) if np.isfinite(pen_against_observed) else 0.05
    use_penalty_attack = bool(np.isfinite(pen_for_observed) and pen_for_observed >= 0.03)
    use_penalty_defence = bool(np.isfinite(pen_against_observed) and pen_against_observed >= 0.03)

    def _row_adjusted(row):
        t_opp = int(row["opponent_id"])
        venue = str(row["venue"])
        if venue == "H":
            opp_def = float(p.loc[t_opp, "prior_defence_away"])
            opp_att = float(p.loc[t_opp, "prior_attack_away"])
            mu_for, mu_against = mu_h, mu_a
        else:
            opp_def = float(p.loc[t_opp, "prior_defence_home"])
            opp_att = float(p.loc[t_opp, "prior_attack_home"])
            mu_for, mu_against = mu_a, mu_h
        return pd.Series({
            "_adj_npxg_for": float(row["npxg_for"]) / max(0.20, opp_def),
            "_adj_npxga": float(row["npxga"]) / max(0.20, opp_att),
            "_npxg_att_ratio": _vx_safe_ratio(row["npxg_for"], mu_for * opp_def),
            "_npxg_def_ratio": _vx_safe_ratio(row["npxga"], mu_against * opp_att),
        })

    f = pd.concat([f, f.apply(_row_adjusted, axis=1)], axis=1)

    def _signal_subset(g, team_id):
        if g.empty:
            return {
                "att": 1.0, "def": 1.0, "pen_att": 1.0, "pen_def": 1.0,
                "adj_for": np.nan, "adj_against": np.nan,
                "n_eff": 0.0, "n_matches": 0, "recency_mean": 0.0,
            }
        w = g["_w"]
        n_eff = float(w.sum())
        n_matches = int(len(g))
        recency_mean = float(pd.to_numeric(g["_time_w"], errors="coerce").fillna(0.0).mean())

        # NPxG dominates. Big chances and set-piece xG are secondary process
        # features. Penalties are kept separate, strongly shrunk and capped.
        npxg_att = _vx_weighted_mean(g["_npxg_att_ratio"], w)
        npxg_def = _vx_weighted_mean(g["_npxg_def_ratio"], w)
        bc_att = _vx_safe_ratio(_vx_weighted_mean(g["big_chances_for"], w), bc_for_mu)
        bc_def = _vx_safe_ratio(_vx_weighted_mean(g["big_chances_against"], w), bc_against_mu)
        sp_att = _vx_safe_ratio(_vx_weighted_mean(g["set_piece_xg_for"], w), sp_for_mu)
        sp_def = _vx_safe_ratio(_vx_weighted_mean(g["set_piece_xg_against"], w), sp_against_mu)

        # Penalty opportunity is noisy: shrink 75% toward league average and cap.
        if use_penalty_attack:
            pen_att_raw = _vx_safe_ratio(_vx_weighted_mean(g["penalty_xg_for"], w), pen_for_mu)
            pen_att = float(np.clip(1.0 + 0.25 * (pen_att_raw - 1.0), 0.80, 1.20))
        else:
            pen_att = 1.0
        if use_penalty_defence:
            pen_def_raw = _vx_safe_ratio(_vx_weighted_mean(g["penalty_xg_against"], w), pen_against_mu)
            pen_def = float(np.clip(1.0 + 0.25 * (pen_def_raw - 1.0), 0.80, 1.20))
        else:
            pen_def = 1.0

        def _geom(parts):
            total = 0.0
            for val, weight in parts:
                total += weight * math.log(max(0.20, min(3.0, float(val))))
            return float(math.exp(total))

        # Core team strength is NON-PENALTY process strength. The weights sum
        # to one and penalties remain a separate component.
        attack = _geom(((npxg_att, 0.70), (bc_att, 0.15), (sp_att, 0.15)))
        defence = _geom(((npxg_def, 0.70), (bc_def, 0.15), (sp_def, 0.15)))

        return {
            "att": attack,
            "def": defence,
            "pen_att": pen_att,
            "pen_def": pen_def,
            "adj_for": _vx_weighted_mean(g["_adj_npxg_for"], w),
            "adj_against": _vx_weighted_mean(g["_adj_npxga"], w),
            "n_eff": n_eff,
            "n_matches": n_matches,
            "recency_mean": recency_mean,
        }

    rows = []
    for t in team_ids:
        g = f.loc[f["team_id"].eq(t)]
        gh = g.loc[g["venue"].eq("H")]
        ga = g.loc[g["venue"].eq("A")]
        overall = _signal_subset(g, t)
        home = _signal_subset(gh, t)
        away = _signal_subset(ga, t)

        # Venue-specific signals are estimated with the exact same recency-weighted
        # match evidence, then shrunk toward the overall current-season signal
        # when the venue sample is thin. The PRIOR/CURRENT blend is deliberately
        # NOT done here; Phase 2 owns that update with w = n / (n + 8).
        def _venue_signal(v, o, n_matches):
            venue_share = min(1.0, float(n_matches) / 6.0)
            return float(math.exp(
                venue_share * math.log(max(0.20, v))
                + (1.0 - venue_share) * math.log(max(0.20, o))
            ))

        rows.append({
            "team_id": t,
            "season_attack_signal": overall["att"],
            "season_defence_signal": overall["def"],
            "season_home_attack_signal": _venue_signal(home["att"], overall["att"], home["n_matches"]),
            "season_home_defence_signal": _venue_signal(home["def"], overall["def"], home["n_matches"]),
            "season_away_attack_signal": _venue_signal(away["att"], overall["att"], away["n_matches"]),
            "season_away_defence_signal": _venue_signal(away["def"], overall["def"], away["n_matches"]),
            "season_penalty_attack_signal": float(overall["pen_att"]),
            "season_penalty_defence_signal": float(overall["pen_def"]),
            "opponent_adjusted_npxg_for": overall["adj_for"],
            "opponent_adjusted_npxga": overall["adj_against"],
            "recent_form_weight": overall["recency_mean"],
            "sample_size": overall["n_matches"],
            "home_sample_size": home["n_matches"],
            "away_sample_size": away["n_matches"],
            "home_recent_form_weight": home["recency_mean"],
            "away_recent_form_weight": away["recency_mean"],
            "recency_effective_sample_size": overall["n_eff"],
        })

    return pd.DataFrame(rows)

# ----------------------------------------------------------------
# BUILD STATUS
# ----------------------------------------------------------------
PHASE1_INPUT_ISSUES = list(_match_issues) + list(_context_issues)
PHASE1_READY = bool(_match_ok and _context_ok)

team_priors = pd.DataFrame(columns=PHASE1_PRIOR_FIELDS)
team_strength_signals = pd.DataFrame(columns=PHASE1_SIGNAL_FIELDS)
PHASE1_VENUE_BASELINES = {}

if PHASE1_READY:
    team_priors, PHASE1_VENUE_BASELINES = _vx_build_prior(team_match_xg_raw, team_context_raw)
    team_strength_signals = _vx_build_current_signals(
        team_match_xg_raw, team_priors, PHASE1_VENUE_BASELINES
    )

    _prior_missing = [c for c in PHASE1_PRIOR_FIELDS if c not in team_priors.columns]
    _signal_missing = [c for c in PHASE1_SIGNAL_FIELDS if c not in team_strength_signals.columns]
    if _prior_missing or _signal_missing:
        raise RuntimeError(
            f"PHASE 1 output contract failure: priors={_prior_missing}, signals={_signal_missing}"
        )
    if len(team_priors) != len(teams) or len(team_strength_signals) != len(teams):
        raise RuntimeError("PHASE 1 must produce exactly one prior/signal row per current team")

    print("✅ PHASE 1A TEAM-STRENGTH STATE BUILT")
    print(f"   Previous season: {PREVIOUS_SEASON}")
    print(f"   Current season:  {CURRENT_SEASON}")
    print(f"   Match source:    {PHASE1_MATCH_SOURCE_PATH or 'in-memory approved raw table'}")
    if PHASE1_FPL_CORE_SOURCE_USED:
        print("   Source priority: FPL_CORE canonical match aggregates + incidents (red timing only)")
else:
    print("⛔ PHASE 1A BLOCKED — trusted inputs are incomplete.")
    for issue in PHASE1_INPUT_ISSUES:
        print("   •", issue)
    if PHASE1_FPL_CORE_ADAPTER_ERROR:
        print("   Preferred FPL_CORE source was found but not silently coerced:")
        print("   •", PHASE1_FPL_CORE_ADAPTER_ERROR)
    print("   Template:", PHASE1_CONTEXT_PATH)
    print("   Manual fallback input:", PHASE1_RAW_INPUT_DIR / "team_match_xg_raw.csv")
    print("   No team strengths or forecast numbers were fabricated.")

# Do not scan generated Phase-1 model objects with the Phase-0 no-model scanner;
# instead, keep the provider-owned banned-field gate explicit at the raw boundary.
vx_phase0_assert_no_forbidden_columns(team_match_xg_raw, "Phase 1 team_match_xg_raw")
```

```python
# ============================================================
# CELL 6 — PHASE 1B
# MODEL-ONLY FIXTURE PROJECTED-GOALS PRIOR-ONLY BASELINE
# OUTPUT CONTRACT: projected_home_goals + projected_away_goals
# ============================================================
# Inputs:
#   team_priors + raw fixtures
# Phase 2 is the ONLY owner of the GW2+ prior/current-season blend.
# Outputs:
#   projected_home_goals
#   projected_away_goals
#
# No score matrix, clean-sheet probability, fixture rating, player
# expected points, transfer, captaincy, squad or line-up logic exists here.
# ============================================================

from datetime import datetime, timezone
import math
import numpy as np
import pandas as pd

PHASE1_PROJECTION_FIELDS = (
    "fixture_id",
    "gameweek",
    "home_team_id",
    "away_team_id",
    "projected_home_nonpenalty_goals",
    "projected_home_penalty_goals",
    "projected_away_nonpenalty_goals",
    "projected_away_penalty_goals",
    "projected_home_goals",
    "projected_away_goals",
    "model_version",
    "data_snapshot_timestamp",
    "forecast_timestamp",
    "source_name",
    "source_timestamp",
    "is_pre_deadline_data",
)

projected_goals_baseline = pd.DataFrame(columns=PHASE1_PROJECTION_FIELDS)

if not globals().get("PHASE1_READY", False):
    print("⛔ PHASE 1B BLOCKED — Phase 1A inputs are not complete.")
    print("   No projected home/away goals were emitted.")
else:
    forecast_timestamp = datetime.now(timezone.utc).isoformat()

    p = team_priors.set_index("team_id")

    def _vx_team_state(team_id, venue):
        """Phase-1 baseline = preseason prior only. No current-season shortcut."""
        t = int(team_id)
        if t not in p.index:
            raise RuntimeError(f"Missing Phase 1 prior for team_id={t}")
        if venue == "H":
            return float(p.loc[t, "prior_attack_home"]), float(p.loc[t, "prior_defence_home"])
        return float(p.loc[t, "prior_attack_away"]), float(p.loc[t, "prior_defence_away"])

    rows = []
    fixture_frame = fixtures_raw.copy()
    if "vx_completed" not in fixture_frame.columns:
        raise RuntimeError("Phase 1B requires canonical fixtures_raw.vx_completed from Cell 3.")
    fixture_frame = fixture_frame.loc[~fixture_frame["vx_completed"].fillna(False).astype(bool)].copy()

    # Forecast only fixtures with a known gameweek.
    fixture_frame["event"] = pd.to_numeric(fixture_frame["event"], errors="coerce")
    fixture_frame = fixture_frame.loc[fixture_frame["event"].notna()].copy()

    mu_h = float(PHASE1_VENUE_BASELINES["league_home_npxg"])
    mu_a = float(PHASE1_VENUE_BASELINES["league_away_npxg"])
    pen_h = float(PHASE1_VENUE_BASELINES["league_home_penalty_xg"])
    pen_a = float(PHASE1_VENUE_BASELINES["league_away_penalty_xg"])

    for _, fx in fixture_frame.iterrows():
        home_id = int(fx["team_h"])
        away_id = int(fx["team_a"])

        home_att, home_def = _vx_team_state(home_id, "H")
        away_att, away_def = _vx_team_state(away_id, "A")

        # Explicit total-goal baseline = non-penalty scoring + penalty scoring.
        # The promoted-team adjustment is already embedded once in the preseason
        # non-penalty prior and is NOT applied again here.
        home_np = float(mu_h * home_att * away_def)
        away_np = float(mu_a * away_att * home_def)
        home_pen = float(pen_h * float(p.loc[home_id, "prior_penalty_attack_signal"]) * float(p.loc[away_id, "prior_penalty_defence_signal"]))
        away_pen = float(pen_a * float(p.loc[away_id, "prior_penalty_attack_signal"]) * float(p.loc[home_id, "prior_penalty_defence_signal"]))
        home_mu = float(home_np + home_pen)
        away_mu = float(away_np + away_pen)

        # These are process means, not score predictions. Bounds are only
        # numerical safety rails; hitting them is treated as a warning below.
        home_mu = float(np.clip(home_mu, 0.20, 4.00))
        away_mu = float(np.clip(away_mu, 0.20, 4.00))

        gw = int(fx["event"])
        rows.append({
            "fixture_id": int(fx["id"]),
            "gameweek": gw,
            "home_team_id": home_id,
            "away_team_id": away_id,
            "projected_home_nonpenalty_goals": home_np,
            "projected_home_penalty_goals": home_pen,
            "projected_away_nonpenalty_goals": away_np,
            "projected_away_penalty_goals": away_pen,
            "projected_home_goals": home_mu,
            "projected_away_goals": away_mu,
            "model_version": model_version,
            "data_snapshot_timestamp": data_snapshot_timestamp,
            "forecast_timestamp": forecast_timestamp,
            "source_name": "VORTEX_PHASE1_PRIOR_ONLY_BASELINE",
            "source_timestamp": forecast_timestamp,
            "is_pre_deadline_data": _phase0_is_pre_deadline(gw) if "_phase0_is_pre_deadline" in globals() else pd.NA,
        })

    projected_goals_baseline = pd.DataFrame(rows, columns=PHASE1_PROJECTION_FIELDS)

    if projected_goals_baseline.empty:
        raise RuntimeError("PHASE 1B produced no future fixtures")

    for c in ("projected_home_goals", "projected_away_goals"):
        if projected_goals_baseline[c].isna().any():
            raise RuntimeError(f"PHASE 1B null values in {c}")
        if (~projected_goals_baseline[c].between(0.20, 4.00)).any():
            raise RuntimeError(f"PHASE 1B out-of-range values in {c}")

    if projected_goals_baseline["fixture_id"].duplicated().any():
        raise RuntimeError("PHASE 1B duplicate fixture IDs")

    # Save only the model-only baseline output. No downstream model artefacts.
    PHASE1_OUTPUT_PATH = Path(DATA_DIR) / "phase1_projected_goals_baseline.json"
    PHASE1_OUTPUT_PATH.write_text(
        projected_goals_baseline.to_json(orient="records", date_format="iso", indent=2),
        encoding="utf-8",
    )

    print("✅ PHASE 1B PRIOR-ONLY BASELINE BUILT")
    print(f"   Fixtures forecast: {len(projected_goals_baseline)}")
    print("   No current-season blend is applied here; Phase 2 owns GW2+ updates.")
    print("   Output columns: projected_home_goals, projected_away_goals")
    print("   Saved:", PHASE1_OUTPUT_PATH)
```

```python
# ============================================================
# CELL 7 — PHASE 2
# GAMEWEEK TEAM-STRENGTH UPDATE + EXPLICIT TEAM-NEWS OVERRIDES
# MODEL-ONLY • NO ODDS • NO ELO • NO FIXTURE RATING • NO PLAYER XPTS
# ============================================================
# Contract:
#   updated_strength = (1-w) * prior + w * current-season signal
#   w = n / (n + 8)
# where n is the team's ACTUAL number of completed current-season matches.
#
# Match recency is already inside the Phase-1 current-season signal:
#   match_weight = 0.5 ** (days_since_match / 90)
# Phase 2 NEVER replaces n with the recency-weighted effective sample size.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import math
import numpy as np
import pandas as pd

if "PHASE1_READY" not in globals():
    raise RuntimeError("Run Phase 1 before Phase 2.")

model_version = "VORTEX-V2-PHASE2-COMPONENT-TEAM-UPDATE"
forecast_timestamp = datetime.now(timezone.utc).isoformat()

# ----------------------------------------------------------------
# HARD EXCLUSIONS FOR THIS PHASE
# ----------------------------------------------------------------
# Phase 2 consumes only Phase-1 priors/signals, descriptive raw match evidence,
# future fixture identity, and explicit pre-deadline team-news adjustments.
# External market prices, third-party rating blends and player-point models are
# intentionally outside this phase.
PHASE2_NEWS_FIELDS = (
    "attack_injury_adjustment",
    "defence_injury_adjustment",
    "goalkeeper_adjustment",
    "manager_change_adjustment",
    "penalty_taker_adjustment",
    "rest_days_adjustment",
    "lineup_confidence_adjustment",
)

PHASE2_STRENGTH_FIELDS = (
    "team_id",
    "completed_matches",
    "prior_weight",
    "current_season_weight",
    "updated_attack_overall",
    "updated_defence_overall",
    "updated_attack_home",
    "updated_defence_home",
    "updated_attack_away",
    "updated_defence_away",
    "updated_penalty_attack_signal",
    "updated_penalty_defence_signal",
    "recent_form_weight",
    "recency_effective_sample_size",
)

PHASE2_CONTEXT_FIELDS = (
    "fixture_id",
    "gameweek",
    "home_team_id",
    "away_team_id",
    "home_attack_context_delta",
    "home_defence_context_delta",
    "away_attack_context_delta",
    "away_defence_context_delta",
    "home_penalty_taker_context_delta",
    "away_penalty_taker_context_delta",
    "context_home_log_adjustment",
    "context_away_log_adjustment",
    "context_home_multiplier",
    "context_away_multiplier",
    "context_home_penalty_multiplier",
    "context_away_penalty_multiplier",
    "model_version",
    "data_snapshot_timestamp",
    "forecast_timestamp",
    "source_name",
    "source_timestamp",
    "is_pre_deadline_data",
)

# ----------------------------------------------------------------
# EXACT GAMEWEEK UPDATE WEIGHT
# ----------------------------------------------------------------
def vx_phase2_current_weight(n):
    n = int(n)
    if n < 0:
        raise ValueError("completed match count cannot be negative")
    return float(n / (n + 8.0)) if n else 0.0


# Contract tests: forecasting GW means n = GW - 1 only in a normal schedule.
# The live model below does NOT assume this; it counts actual completed matches.
_PHASE2_WEIGHT_CHECKS = {
    1: 1 / 9,   # GW2
    2: 2 / 10,  # GW3
    3: 3 / 11,  # GW4
    4: 4 / 12,  # GW5
    5: 5 / 13,  # GW6
    6: 6 / 14,  # GW7
    7: 7 / 15,  # GW8
    8: 8 / 16,  # GW9
    9: 9 / 17,  # GW10
    12: 12 / 20, # GW13
    15: 15 / 23, # GW16
    20: 20 / 28, # GW21
    37: 37 / 45, # GW38 forecast
}
for _n, _expected in _PHASE2_WEIGHT_CHECKS.items():
    _actual = vx_phase2_current_weight(_n)
    if not math.isclose(_actual, _expected, rel_tol=0.0, abs_tol=1e-12):
        raise RuntimeError(f"Phase 2 weighting contract failed for n={_n}: {_actual} != {_expected}")


def _vx_phase2_linear_blend(prior_value, signal_value, n):
    """Exact arithmetic blend required by the Phase-2 contract."""
    p = float(prior_value)
    s = float(signal_value)
    if not np.isfinite(p) or not np.isfinite(s) or p <= 0 or s <= 0:
        raise RuntimeError(f"Invalid strength inputs prior={p}, signal={s}")
    w = vx_phase2_current_weight(n)
    return float((1.0 - w) * p + w * s)


# ----------------------------------------------------------------
# COMPLETED-MATCH COUNTS — ACTUAL CURRENT-SEASON MATCHES PER TEAM
# ----------------------------------------------------------------
def _vx_phase2_completed_matches(frame):
    if not isinstance(frame, pd.DataFrame) or frame.empty:
        return {}
    f = frame.loc[frame["season"].astype(str).eq(CURRENT_SEASON)].copy()
    if f.empty:
        return {}
    f["match_date"] = pd.to_datetime(f["match_date"], utc=True, errors="coerce")
    _anchor = pd.Timestamp(forecast_timestamp)
    if _anchor.tzinfo is None:
        _anchor = _anchor.tz_localize("UTC")
    else:
        _anchor = _anchor.tz_convert("UTC")
    # Only evidence available before the forecast timestamp can contribute.
    f = f.loc[f["match_date"].notna() & f["match_date"].le(_anchor)].copy()
    f["team_id"] = pd.to_numeric(f["team_id"], errors="coerce")
    f = f.loc[f["team_id"].notna()].copy()
    return f.groupby(f["team_id"].astype(int), sort=False).size().astype(int).to_dict()


team_strength_updated = pd.DataFrame(columns=PHASE2_STRENGTH_FIELDS)
fixture_context_phase2 = pd.DataFrame(columns=PHASE2_CONTEXT_FIELDS)
PHASE2_READY = False

if not bool(PHASE1_READY):
    print("⛔ PHASE 2 BLOCKED — Phase 1 trusted inputs are incomplete.")
    print("   Weight/update/news logic is installed, but no forecast numbers were emitted.")
else:
    _prior = team_priors.set_index("team_id")
    _signal = team_strength_signals.set_index("team_id")
    _completed = _vx_phase2_completed_matches(team_match_xg_raw)
    _team_ids = sorted(set(pd.to_numeric(teams["id"], errors="coerce").dropna().astype(int)))

    _rows = []
    for _t in _team_ids:
        if _t not in _prior.index or _t not in _signal.index:
            raise RuntimeError(f"Phase 2 missing Phase-1 state for team_id={_t}")
        _n = int(_completed.get(_t, 0))
        _w = vx_phase2_current_weight(_n)
        _pw = 1.0 - _w

        _rows.append({
            "team_id": _t,
            "completed_matches": _n,
            "prior_weight": _pw,
            "current_season_weight": _w,
            "updated_attack_overall": _vx_phase2_linear_blend(
                _prior.loc[_t, "prior_attack_overall"], _signal.loc[_t, "season_attack_signal"], _n
            ),
            "updated_defence_overall": _vx_phase2_linear_blend(
                _prior.loc[_t, "prior_defence_overall"], _signal.loc[_t, "season_defence_signal"], _n
            ),
            "updated_attack_home": _vx_phase2_linear_blend(
                _prior.loc[_t, "prior_attack_home"], _signal.loc[_t, "season_home_attack_signal"], _n
            ),
            "updated_defence_home": _vx_phase2_linear_blend(
                _prior.loc[_t, "prior_defence_home"], _signal.loc[_t, "season_home_defence_signal"], _n
            ),
            "updated_attack_away": _vx_phase2_linear_blend(
                _prior.loc[_t, "prior_attack_away"], _signal.loc[_t, "season_away_attack_signal"], _n
            ),
            "updated_defence_away": _vx_phase2_linear_blend(
                _prior.loc[_t, "prior_defence_away"], _signal.loc[_t, "season_away_defence_signal"], _n
            ),
            "updated_penalty_attack_signal": _vx_phase2_linear_blend(
                _prior.loc[_t, "prior_penalty_attack_signal"], _signal.loc[_t, "season_penalty_attack_signal"], _n
            ),
            "updated_penalty_defence_signal": _vx_phase2_linear_blend(
                _prior.loc[_t, "prior_penalty_defence_signal"], _signal.loc[_t, "season_penalty_defence_signal"], _n
            ),
            "recent_form_weight": float(_signal.loc[_t, "recent_form_weight"]),
            "recency_effective_sample_size": float(
                _signal.loc[_t, "recency_effective_sample_size"]
                if "recency_effective_sample_size" in _signal.columns else _n
            ),
        })

    team_strength_updated = pd.DataFrame(_rows, columns=PHASE2_STRENGTH_FIELDS)
    if len(team_strength_updated) != len(_team_ids):
        raise RuntimeError("Phase 2 must produce exactly one updated-strength row per current team")
    if team_strength_updated["team_id"].duplicated().any():
        raise RuntimeError("Phase 2 duplicate team rows")

    # ------------------------------------------------------------
    # EXPLICIT TEAM-NEWS OVERRIDE TABLE
    # ------------------------------------------------------------
    # All adjustment fields are LOG multipliers. 0.00 is neutral.
    # +0.10 is approximately +10.5% to the affected goal process.
    # -0.10 is approximately -9.5%.
    #
    # Semantics:
    #   attack_injury_adjustment   -> own attacking process
    #   defence_injury_adjustment  -> defensive weakness; + means concede more
    #   goalkeeper_adjustment      -> defensive weakness; + means concede more
    #   manager_change_adjustment  -> net team strength; + helps attack and defence
    #   penalty_taker_adjustment   -> penalty component only
    #   rest_days_adjustment       -> net team strength; + helps attack and defence
    #   lineup_confidence_adjustment -> net team strength; + helps attack and defence
    #
    # Therefore positive net-strength values are ADDED to attack and SUBTRACTED
    # from defensive weakness. Every non-zero active override requires provenance.
    # Reuse the approved persistent raw-input root created upstream in Phase 1.
    # Override inputs are model inputs, not final video DATA artifacts.
    PHASE2_NEWS_PATH = Path(PHASE1_RAW_INPUT_DIR) / "team_news_overrides.csv"
    PHASE2_NEWS_PATH.parent.mkdir(parents=True, exist_ok=True)

    # One-time migration for a runtime started before this path correction.
    # Preserve the persistent file when it already exists; otherwise move the
    # current-run template out of DATA so final output QA stays clean.
    _phase2_legacy_news_path = Path(DATA_DIR) / "RAW_INPUT" / "team_news_overrides.csv"
    if _phase2_legacy_news_path.exists():
        if not PHASE2_NEWS_PATH.exists():
            _phase2_legacy_news_path.replace(PHASE2_NEWS_PATH)
        else:
            _phase2_legacy_news_path.unlink()
        _phase2_legacy_news_dir = _phase2_legacy_news_path.parent
        if _phase2_legacy_news_dir.exists() and not any(_phase2_legacy_news_dir.iterdir()):
            _phase2_legacy_news_dir.rmdir()

    _future = fixtures_raw.copy()
    if "vx_completed" not in _future.columns:
        raise RuntimeError("Phase 2 requires canonical fixtures_raw.vx_completed from Cell 3.")
    _future = _future.loc[~_future["vx_completed"].fillna(False).astype(bool)].copy()
    _future["event"] = pd.to_numeric(_future["event"], errors="coerce")
    _future = _future.loc[_future["event"].notna()].copy()

    def _vx_phase2_news_template():
        _out = []
        for _, _fx in _future.iterrows():
            _fid = int(_fx["id"])
            _gw = int(_fx["event"])
            _h = int(_fx["team_h"])
            _a = int(_fx["team_a"])
            for _team, _opp, _venue in ((_h, _a, "H"), (_a, _h, "A")):
                _r = {
                    "fixture_id": _fid,
                    "gameweek": _gw,
                    "team_id": _team,
                    "opponent_id": _opp,
                    "venue": _venue,
                    "override_active": False,
                    "reason": "",
                    "source_name": "",
                    "source_timestamp": "",
                    "is_pre_deadline_data": False,
                }
                for _c in PHASE2_NEWS_FIELDS:
                    _r[_c] = 0.0
                _out.append(_r)
        return pd.DataFrame(_out)

    if PHASE2_NEWS_PATH.exists():
        team_news_overrides = pd.read_csv(PHASE2_NEWS_PATH)
        _template = _vx_phase2_news_template()
        # Preserve entered rows, append newly appearing future fixtures only.
        _key = ["fixture_id", "team_id"]
        if not team_news_overrides.empty and all(c in team_news_overrides.columns for c in _key):
            _existing = set(map(tuple, team_news_overrides[_key].astype(int).to_numpy()))
            _missing_rows = _template.loc[
                ~_template[_key].apply(lambda r: tuple(r.astype(int)), axis=1).isin(_existing)
            ]
            if not _missing_rows.empty:
                team_news_overrides = pd.concat([team_news_overrides, _missing_rows], ignore_index=True)
                team_news_overrides.to_csv(PHASE2_NEWS_PATH, index=False)
    else:
        team_news_overrides = _vx_phase2_news_template()
        team_news_overrides.to_csv(PHASE2_NEWS_PATH, index=False)

    _required_news_cols = {
        "fixture_id", "gameweek", "team_id", "opponent_id", "venue",
        "override_active", "reason", "source_name", "source_timestamp",
        "is_pre_deadline_data", *PHASE2_NEWS_FIELDS,
    }
    _missing = sorted(_required_news_cols - set(team_news_overrides.columns))
    if _missing:
        raise RuntimeError(f"Phase 2 team-news table missing fields: {_missing}")

    for _c in PHASE2_NEWS_FIELDS:
        team_news_overrides[_c] = pd.to_numeric(team_news_overrides[_c], errors="coerce")
        if team_news_overrides[_c].isna().any():
            raise RuntimeError(f"Phase 2 team-news field {_c} contains non-numeric values")
        if (team_news_overrides[_c].abs() > 0.50).any():
            raise RuntimeError(f"Phase 2 team-news field {_c} exceeds the ±0.50 safety rail")

    def _vx_phase2_bool(v):
        if isinstance(v, (bool, np.bool_)):
            return bool(v)
        if pd.isna(v):
            return False
        return str(v).strip().lower() in {"1", "true", "yes", "y"}

    team_news_overrides["override_active"] = team_news_overrides["override_active"].map(_vx_phase2_bool)
    team_news_overrides["is_pre_deadline_data"] = team_news_overrides["is_pre_deadline_data"].map(_vx_phase2_bool)

    if team_news_overrides[["fixture_id", "team_id"]].duplicated().any():
        raise RuntimeError("Phase 2 team-news table has duplicate fixture/team rows")

    _active = team_news_overrides.loc[team_news_overrides["override_active"]].copy()
    if not _active.empty:
        _nonzero = _active[list(PHASE2_NEWS_FIELDS)].abs().sum(axis=1) > 0
        _active = _active.loc[_nonzero].copy()
        if not _active.empty:
            if (~_active["is_pre_deadline_data"]).any():
                raise RuntimeError("Every active non-zero team-news override must be pre-deadline data")
            if _active["reason"].astype(str).str.strip().eq("").any():
                raise RuntimeError("Every active non-zero team-news override requires a reason")
            if _active["source_name"].astype(str).str.strip().eq("").any():
                raise RuntimeError("Every active non-zero team-news override requires source_name")
            _src_ts = pd.to_datetime(_active["source_timestamp"], utc=True, errors="coerce")
            if _src_ts.isna().any():
                raise RuntimeError("Every active non-zero team-news override requires a valid source_timestamp")
            _fc_ts = pd.Timestamp(forecast_timestamp)
            if (_src_ts > _fc_ts).any():
                raise RuntimeError("Team-news override source timestamp cannot be after forecast timestamp")

    _news_idx = team_news_overrides.set_index(["fixture_id", "team_id"], drop=False)

    def _vx_phase2_news_effect(fixture_id, team_id):
        _key = (int(fixture_id), int(team_id))
        if _key not in _news_idx.index:
            return {"attack_delta": 0.0, "defence_delta": 0.0, "penalty_delta": 0.0}
        _r = _news_idx.loc[_key]
        if isinstance(_r, pd.DataFrame):
            raise RuntimeError(f"Duplicate team-news rows for fixture/team {_key}")
        if not bool(_r["override_active"]):
            # Inactive rows MUST be neutral; hidden values are not tolerated.
            if float(_r[list(PHASE2_NEWS_FIELDS)].abs().sum()) > 1e-12:
                raise RuntimeError(f"Inactive team-news row contains non-zero adjustment for {_key}")
            return {"attack_delta": 0.0, "defence_delta": 0.0, "penalty_delta": 0.0}

        _net = (
            float(_r["manager_change_adjustment"])
            + float(_r["rest_days_adjustment"])
            + float(_r["lineup_confidence_adjustment"])
        )
        _attack = (
            float(_r["attack_injury_adjustment"])
            + _net
        )
        _penalty = float(_r["penalty_taker_adjustment"])
        # Defence strength is encoded as weakness (>1 = concedes more), so a
        # positive net team-strength adjustment reduces defensive weakness.
        _defence = (
            float(_r["defence_injury_adjustment"])
            + float(_r["goalkeeper_adjustment"])
            - _net
        )
        return {
            "attack_delta": float(_attack),
            "defence_delta": float(_defence),
            "penalty_delta": float(_penalty),
        }

    # ------------------------------------------------------------
    # EXPLICIT FIXTURE CONTEXT HAND-OFF — NO GOAL EXPECTATION HERE
    # ------------------------------------------------------------
    # Phase 2 owns the team-strength update and the explicit team-news
    # transformation only. Phase 3 is the sole owner of fixture lambdas.
    _ctx_rows = []
    for _, _fx in _future.iterrows():
        _fid = int(_fx["id"])
        _gw = int(_fx["event"])
        _h = int(_fx["team_h"])
        _a = int(_fx["team_a"])

        _hn = _vx_phase2_news_effect(_fid, _h)
        _an = _vx_phase2_news_effect(_fid, _a)

        # C_context is side-specific because home and away can have different
        # injuries, goalkeeper status, manager effects, rest and lineup news.
        # Own attack context combines with the opponent's defensive weakness.
        _home_log_context = float(_hn["attack_delta"] + _an["defence_delta"])
        _away_log_context = float(_an["attack_delta"] + _hn["defence_delta"])
        _home_context = float(math.exp(_home_log_context))
        _away_context = float(math.exp(_away_log_context))
        # Generic team/opponent context affects both scoring components; only the
        # explicit penalty-taker delta is restricted to the penalty component.
        _home_penalty_context = float(math.exp(_home_log_context + float(_hn["penalty_delta"])))
        _away_penalty_context = float(math.exp(_away_log_context + float(_an["penalty_delta"])))

        if not all(np.isfinite(x) for x in (_home_context, _away_context, _home_penalty_context, _away_penalty_context)):
            raise RuntimeError(f"Invalid Phase 2 context multiplier for fixture {_fid}")
        if min(_home_context, _away_context, _home_penalty_context, _away_penalty_context) <= 0:
            raise RuntimeError(f"Non-positive Phase 2 context multiplier for fixture {_fid}")

        _ctx_rows.append({
            "fixture_id": _fid,
            "gameweek": _gw,
            "home_team_id": _h,
            "away_team_id": _a,
            "home_attack_context_delta": float(_hn["attack_delta"]),
            "home_defence_context_delta": float(_hn["defence_delta"]),
            "away_attack_context_delta": float(_an["attack_delta"]),
            "away_defence_context_delta": float(_an["defence_delta"]),
            "home_penalty_taker_context_delta": float(_hn["penalty_delta"]),
            "away_penalty_taker_context_delta": float(_an["penalty_delta"]),
            "context_home_log_adjustment": _home_log_context,
            "context_away_log_adjustment": _away_log_context,
            "context_home_multiplier": _home_context,
            "context_away_multiplier": _away_context,
            "context_home_penalty_multiplier": _home_penalty_context,
            "context_away_penalty_multiplier": _away_penalty_context,
            "model_version": model_version,
            "data_snapshot_timestamp": data_snapshot_timestamp,
            "forecast_timestamp": forecast_timestamp,
            "source_name": "VORTEX_PHASE2_EXPLICIT_CONTEXT",
            "source_timestamp": forecast_timestamp,
            "is_pre_deadline_data": _phase0_is_pre_deadline(_gw) if "_phase0_is_pre_deadline" in globals() else pd.NA,
        })

    fixture_context_phase2 = pd.DataFrame(_ctx_rows, columns=PHASE2_CONTEXT_FIELDS)
    if fixture_context_phase2.empty:
        raise RuntimeError("Phase 2 produced no future fixture context rows")
    if fixture_context_phase2["fixture_id"].duplicated().any():
        raise RuntimeError("Phase 2 duplicate fixture context rows")

    PHASE2_STRENGTH_PATH = Path(DATA_DIR) / "phase2_team_strength_updated.json"
    PHASE2_CONTEXT_PATH = Path(DATA_DIR) / "phase2_fixture_context_explicit.json"
    PHASE2_STRENGTH_PATH.write_text(
        team_strength_updated.to_json(orient="records", date_format="iso", indent=2),
        encoding="utf-8",
    )
    PHASE2_CONTEXT_PATH.write_text(
        fixture_context_phase2.to_json(orient="records", date_format="iso", indent=2),
        encoding="utf-8",
    )

    PHASE2_READY = True
    print("✅ PHASE 2 TEAM-STRENGTH UPDATE BUILT")
    print("   Exact update: (1-w)*Prior + w*Current, w=n/(n+8)")
    print("   Match recency: 0.5 ** (days_since_match / 90)")
    print(f"   Updated teams: {len(team_strength_updated)}")
    print(f"   Explicit fixture contexts: {len(fixture_context_phase2)}")
    print("   Team-news fields are explicit and provenance-gated:")
    for _c in PHASE2_NEWS_FIELDS:
        print("     -", _c)
    print("   News template:", PHASE2_NEWS_PATH)
    print("   Strength output:", PHASE2_STRENGTH_PATH)
    print("   Context output:", PHASE2_CONTEXT_PATH)
    print("   No fixture lambda is calculated in Phase 2; Phase 3 owns that step.")

# Keep the Phase-0 provider-field ban active at the raw boundary.
vx_phase0_assert_no_forbidden_columns(team_match_xg_raw, "Phase 2 team_match_xg_raw")
```

```python
# ============================================================
# CELL 7A — PHASE 3
# FIXTURE PROJECTED-GOALS ENGINE
# MODEL-ONLY • NO POISSON • NO CLEAN-SHEET OUTPUT • NO PLAYER XPTS
# ============================================================
# Contract:
#   lambda_home = lambda_home_nonpenalty + lambda_home_penalty
#   lambda_away = lambda_away_nonpenalty + lambda_away_penalty
#   nonpenalty = league NPxG baseline × attack × defence × context
#   penalty    = league penalty-xG baseline × penalty attack × penalty defence × penalty context
#
# League non-penalty and penalty venue baselines are built separately in Phase 1.
# A / D are the updated strengths from Phase 2.
# C_context is the explicit Phase-2 team-news context multiplier.
#
# This cell does NOT read GW-review tables directly and does NOT use market
# odds, PlayerElo, official fixture difficulty, score probabilities or player
# expected-points logic.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import math
import numpy as np
import pandas as pd

if "PHASE2_READY" not in globals():
    raise RuntimeError("Run Phase 2 before Phase 3.")

model_version = "VORTEX-V2-PHASE3-COMPONENT-TOTAL-GOAL-LAMBDA"
forecast_timestamp = datetime.now(timezone.utc).isoformat()

PHASE3_OUTPUT_FIELDS = (
    "fixture_id",
    "gameweek",
    "home_team_id",
    "away_team_id",
    "league_home_scoring_baseline",
    "league_away_scoring_baseline",
    "league_home_nonpenalty_baseline",
    "league_away_nonpenalty_baseline",
    "league_home_penalty_baseline",
    "league_away_penalty_baseline",
    "home_attack_strength",
    "away_defensive_weakness",
    "away_attack_strength",
    "home_defensive_weakness",
    "home_penalty_attack_signal",
    "away_penalty_defence_signal",
    "away_penalty_attack_signal",
    "home_penalty_defence_signal",
    "context_home_multiplier",
    "context_away_multiplier",
    "context_home_penalty_multiplier",
    "context_away_penalty_multiplier",
    "model_lambda_home_nonpenalty",
    "model_lambda_home_penalty",
    "model_lambda_away_nonpenalty",
    "model_lambda_away_penalty",
    "model_lambda_home",
    "model_lambda_away",
    "model_lambda_home_low",
    "model_lambda_home_high",
    "model_lambda_away_low",
    "model_lambda_away_high",
    "home_lambda_uncertainty",
    "away_lambda_uncertainty",
    "model_confidence",
    "uncertainty_status",
    "model_version",
    "data_snapshot_timestamp",
    "forecast_timestamp",
    "source_name",
    "source_timestamp",
    "is_pre_deadline_data",
)

fixture_goal_expectations_model = pd.DataFrame(columns=PHASE3_OUTPUT_FIELDS)
projected_goals_model_only = pd.DataFrame(columns=PHASE3_OUTPUT_FIELDS)
PHASE3_READY = False


def _vx_phase3_current_sampling_uncertainty(row):
    """Transparent current-season sampling uncertainty from recency-effective N."""
    w = float(row["current_season_weight"])
    if w <= 0:
        return 0.0
    n_eff = float(row["recency_effective_sample_size"])
    if not np.isfinite(n_eff) or n_eff <= 0:
        n_eff = float(row["completed_matches"])
    return float(min(1.0, 1.0 / math.sqrt(max(1.0, n_eff))))


def _vx_phase3_team_uncertainty(team_id, state, priors):
    """Blend prior uncertainty and current sampling uncertainty using Phase-2 weights."""
    t = int(team_id)
    if t not in state.index or t not in priors.index:
        raise RuntimeError(f"Phase 3 missing uncertainty state for team_id={t}")
    row = state.loc[t]
    prior_u = float(priors.loc[t, "prior_uncertainty"])
    if not np.isfinite(prior_u) or prior_u < 0:
        raise RuntimeError(f"Invalid prior uncertainty for team_id={t}")
    prior_w = float(row["prior_weight"])
    current_w = float(row["current_season_weight"])
    current_u = _vx_phase3_current_sampling_uncertainty(row)
    # Root-sum-square keeps independent prior/current uncertainty contributions
    # explicit and avoids pretending the pre-validation range is a calibrated CI.
    return float(math.sqrt((prior_w * prior_u) ** 2 + (current_w * current_u) ** 2))


def _vx_phase3_lambda_range(lam, uncertainty):
    """Multiplicative pre-validation range; not a claimed probability interval."""
    lam = float(lam)
    u = float(uncertainty)
    if not np.isfinite(lam) or lam <= 0 or not np.isfinite(u) or u < 0:
        raise RuntimeError(f"Invalid lambda/range inputs lambda={lam}, uncertainty={u}")
    return float(lam * math.exp(-u)), float(lam * math.exp(u))


if not bool(PHASE2_READY):
    print("⛔ PHASE 3 BLOCKED — Phase 2 is not ready.")
    print("   No fixture lambdas were emitted.")
else:
    _required_strength = {
        "team_id", "completed_matches", "prior_weight", "current_season_weight",
        "updated_attack_home", "updated_defence_home",
        "updated_attack_away", "updated_defence_away",
        "updated_penalty_attack_signal", "updated_penalty_defence_signal",
        "recency_effective_sample_size",
    }
    _missing_strength = sorted(_required_strength - set(team_strength_updated.columns))
    if _missing_strength:
        raise RuntimeError(f"Phase 3 missing Phase-2 strength fields: {_missing_strength}")

    _required_context = {
        "fixture_id", "gameweek", "home_team_id", "away_team_id",
        "context_home_multiplier", "context_away_multiplier",
        "context_home_penalty_multiplier", "context_away_penalty_multiplier",
    }
    _missing_context = sorted(_required_context - set(fixture_context_phase2.columns))
    if _missing_context:
        raise RuntimeError(f"Phase 3 missing Phase-2 context fields: {_missing_context}")

    if not isinstance(PHASE1_VENUE_BASELINES, dict):
        raise RuntimeError("Phase 3 requires Phase-1 league venue baselines")
    _L_H = float(PHASE1_VENUE_BASELINES.get("league_home_npxg", np.nan))
    _L_A = float(PHASE1_VENUE_BASELINES.get("league_away_npxg", np.nan))
    _P_H = float(PHASE1_VENUE_BASELINES.get("league_home_penalty_xg", np.nan))
    _P_A = float(PHASE1_VENUE_BASELINES.get("league_away_penalty_xg", np.nan))
    if not all(np.isfinite(x) for x in (_L_H, _L_A, _P_H, _P_A)) or min(_L_H, _L_A) <= 0 or min(_P_H, _P_A) < 0:
        raise RuntimeError("Phase 3 league non-penalty/penalty scoring baselines are invalid")

    _state = team_strength_updated.set_index("team_id")
    _priors = team_priors.set_index("team_id")
    if _state.index.duplicated().any() or _priors.index.duplicated().any():
        raise RuntimeError("Phase 3 requires unique team rows")

    _rows = []
    for _, _fx in fixture_context_phase2.iterrows():
        _fid = int(_fx["fixture_id"])
        _gw = int(_fx["gameweek"])
        _h = int(_fx["home_team_id"])
        _a = int(_fx["away_team_id"])
        if _h not in _state.index or _a not in _state.index:
            raise RuntimeError(f"Phase 3 missing updated strength for fixture {_fid}")

        _A_home = float(_state.loc[_h, "updated_attack_home"])
        _D_away = float(_state.loc[_a, "updated_defence_away"])
        _A_away = float(_state.loc[_a, "updated_attack_away"])
        _D_home = float(_state.loc[_h, "updated_defence_home"])
        _PAtt_home = float(_state.loc[_h, "updated_penalty_attack_signal"])
        _PDef_away = float(_state.loc[_a, "updated_penalty_defence_signal"])
        _PAtt_away = float(_state.loc[_a, "updated_penalty_attack_signal"])
        _PDef_home = float(_state.loc[_h, "updated_penalty_defence_signal"])
        _C_home = float(_fx["context_home_multiplier"])
        _C_away = float(_fx["context_away_multiplier"])
        _CP_home = float(_fx["context_home_penalty_multiplier"])
        _CP_away = float(_fx["context_away_penalty_multiplier"])

        _components = [_A_home, _D_away, _A_away, _D_home, _PAtt_home, _PDef_away, _PAtt_away, _PDef_home, _C_home, _C_away, _CP_home, _CP_away]
        if any((not np.isfinite(x) or x <= 0) for x in _components):
            raise RuntimeError(f"Phase 3 invalid positive multiplier for fixture {_fid}: {_components}")

        # Total expected goals are built from disjoint components. This fixes the
        # NPxG-to-scoreline mismatch: NPxG drives only non-penalty scoring, while
        # penalty xG is added explicitly. No external benchmark blend is applied.
        _lambda_home_np = float(_L_H * _A_home * _D_away * _C_home)
        _lambda_away_np = float(_L_A * _A_away * _D_home * _C_away)
        _lambda_home_pen = float(_P_H * _PAtt_home * _PDef_away * _CP_home)
        _lambda_away_pen = float(_P_A * _PAtt_away * _PDef_home * _CP_away)
        _lambda_home = float(_lambda_home_np + _lambda_home_pen)
        _lambda_away = float(_lambda_away_np + _lambda_away_pen)
        if not np.isfinite(_lambda_home) or not np.isfinite(_lambda_away):
            raise RuntimeError(f"Phase 3 non-finite lambda for fixture {_fid}")
        if _lambda_home <= 0 or _lambda_away <= 0:
            raise RuntimeError(f"Phase 3 non-positive lambda for fixture {_fid}")

        _u_home_team = _vx_phase3_team_uncertainty(_h, _state, _priors)
        _u_away_team = _vx_phase3_team_uncertainty(_a, _state, _priors)
        # Multiplicative attack/defence uncertainties combine on the log scale.
        _u_home_lambda = float(math.sqrt(_u_home_team ** 2 + _u_away_team ** 2))
        _u_away_lambda = float(math.sqrt(_u_away_team ** 2 + _u_home_team ** 2))
        _home_low, _home_high = _vx_phase3_lambda_range(_lambda_home, _u_home_lambda)
        _away_low, _away_high = _vx_phase3_lambda_range(_lambda_away, _u_away_lambda)

        # 0–1 information-confidence index only; NOT a calibrated probability.
        # Phase 5 validation is responsible for empirical calibration.
        _confidence = float(math.exp(-max(_u_home_lambda, _u_away_lambda)))

        _rows.append({
            "fixture_id": _fid,
            "gameweek": _gw,
            "home_team_id": _h,
            "away_team_id": _a,
            "league_home_scoring_baseline": float(_L_H + _P_H),
            "league_away_scoring_baseline": float(_L_A + _P_A),
            "league_home_nonpenalty_baseline": _L_H,
            "league_away_nonpenalty_baseline": _L_A,
            "league_home_penalty_baseline": _P_H,
            "league_away_penalty_baseline": _P_A,
            "home_attack_strength": _A_home,
            "away_defensive_weakness": _D_away,
            "away_attack_strength": _A_away,
            "home_defensive_weakness": _D_home,
            "home_penalty_attack_signal": _PAtt_home,
            "away_penalty_defence_signal": _PDef_away,
            "away_penalty_attack_signal": _PAtt_away,
            "home_penalty_defence_signal": _PDef_home,
            "context_home_multiplier": _C_home,
            "context_away_multiplier": _C_away,
            "context_home_penalty_multiplier": _CP_home,
            "context_away_penalty_multiplier": _CP_away,
            "model_lambda_home_nonpenalty": _lambda_home_np,
            "model_lambda_home_penalty": _lambda_home_pen,
            "model_lambda_away_nonpenalty": _lambda_away_np,
            "model_lambda_away_penalty": _lambda_away_pen,
            "model_lambda_home": _lambda_home,
            "model_lambda_away": _lambda_away,
            "model_lambda_home_low": _home_low,
            "model_lambda_home_high": _home_high,
            "model_lambda_away_low": _away_low,
            "model_lambda_away_high": _away_high,
            "home_lambda_uncertainty": _u_home_lambda,
            "away_lambda_uncertainty": _u_away_lambda,
            "model_confidence": _confidence,
            "uncertainty_status": "PREVALIDATION_STRUCTURAL_RANGE_NOT_COVERAGE",
            "model_version": model_version,
            "data_snapshot_timestamp": data_snapshot_timestamp,
            "forecast_timestamp": forecast_timestamp,
            "source_name": "VORTEX_PHASE3_MODEL_ONLY",
            "source_timestamp": forecast_timestamp,
            "is_pre_deadline_data": _phase0_is_pre_deadline(_gw) if "_phase0_is_pre_deadline" in globals() else pd.NA,
        })

    fixture_goal_expectations_model = pd.DataFrame(_rows, columns=PHASE3_OUTPUT_FIELDS)
    if fixture_goal_expectations_model.empty:
        raise RuntimeError("Phase 3 produced no future fixture expectations")
    if fixture_goal_expectations_model["fixture_id"].duplicated().any():
        raise RuntimeError("Phase 3 duplicate fixture IDs")

    for _c in (
        "model_lambda_home", "model_lambda_away",
        "model_lambda_home_low", "model_lambda_home_high",
        "model_lambda_away_low", "model_lambda_away_high",
        "model_confidence",
    ):
        if fixture_goal_expectations_model[_c].isna().any():
            raise RuntimeError(f"Phase 3 null values in {_c}")

    if not (fixture_goal_expectations_model["model_lambda_home_low"] <= fixture_goal_expectations_model["model_lambda_home"]).all():
        raise RuntimeError("Phase 3 home low bound exceeds lambda")
    if not (fixture_goal_expectations_model["model_lambda_home"] <= fixture_goal_expectations_model["model_lambda_home_high"]).all():
        raise RuntimeError("Phase 3 home high bound below lambda")
    if not (fixture_goal_expectations_model["model_lambda_away_low"] <= fixture_goal_expectations_model["model_lambda_away"]).all():
        raise RuntimeError("Phase 3 away low bound exceeds lambda")
    if not (fixture_goal_expectations_model["model_lambda_away"] <= fixture_goal_expectations_model["model_lambda_away_high"]).all():
        raise RuntimeError("Phase 3 away high bound below lambda")
    if not fixture_goal_expectations_model["model_confidence"].between(0.0, 1.0, inclusive="both").all():
        raise RuntimeError("Phase 3 model_confidence must be between 0 and 1")

    # Canonical hand-off to the later score-probability layer.
    projected_goals_model_only = fixture_goal_expectations_model.copy()

    PHASE3_OUTPUT_PATH = Path(DATA_DIR) / "phase3_fixture_goal_expectations_model_only.json"
    PHASE3_OUTPUT_PATH.write_text(
        fixture_goal_expectations_model.to_json(orient="records", date_format="iso", indent=2),
        encoding="utf-8",
    )

    PHASE3_READY = True
    print("✅ PHASE 3 FIXTURE PROJECTED-GOALS ENGINE BUILT")
    print("   lambda_home = L_H × A_home × D_away × C_context_home")
    print("   lambda_away = L_A × A_away × D_home × C_context_away")
    print(f"   L_H={_L_H:.4f} | L_A={_L_A:.4f}")
    print(f"   Forecast fixtures: {len(fixture_goal_expectations_model)}")
    print("   Low/high fields are pre-validation structural ranges, not calibrated coverage.")
    print("   model_confidence is a 0–1 information index, not a probability.")
    print("   No Poisson, clean-sheet or player expected-points calculation runs here.")
    print("   Saved:", PHASE3_OUTPUT_PATH)
```

```python
# ============================================================
# FPL VORTEX — SHARED PRODUCTION HELPERS
# P0 is the ONLY owner of Drive output folders.
# ============================================================

from pathlib import Path

for _required_path_name in (
    "OUTPUT_ROOT", "MP4_DIR", "MP3_DIR", "SLIDE_DIR", "DATA_DIR"
):
    if _required_path_name not in globals():
        raise RuntimeError(f"Run Cell 0 first — missing {_required_path_name}")

VIDEO_FOLDERS = {
    "MP4": Path(MP4_DIR),
    "MP3": Path(MP3_DIR),
    "SLIDE": Path(SLIDE_DIR),
    "DATA": Path(DATA_DIR),
}


# ============================================================
# SHARED FOOTER METADATA — ONE DYNAMIC SOURCE
# ============================================================
from datetime import datetime
from zoneinfo import ZoneInfo

VX_FOOTER_UPDATED_DATE = datetime.now(
    ZoneInfo("America/New_York")
).strftime("%d/%m/%Y")

print(f"✅ Footer data stamp: Updated: {VX_FOOTER_UPDATED_DATE}")


# ============================================================
# EDGE-TTS NARRATION PROFILE — SINGLE SOURCE OF TRUTH
# ============================================================
# RyanNeural remains the project voice.
# Optimized for a modern faceless YouTube/FPL analysis pace:
# energetic enough to hold attention, but controlled around dense numbers.
# Keep volume flat here; final loudness belongs in audio mastering.

VOICE = "en-GB-RyanNeural"

# FPL VORTEX two-tone Ryan delivery:
# - Professor (Calm): clear, confident, analytical explanation.
# - Match Impact: a short energetic lift for important numbers / reveals.
#
# Target delivery:
# - Intro: fast, confident hook.
# - Analysis: energetic but easy to follow.
# - Key Stat: strongest lift for rankings, numbers and reveals.
# - Warning: slightly slower than normal analysis for clarity.
# - Outro: quick, energetic close.
#
# IMPORTANT:
# Do NOT slow Ryan to -10%.
# These rates are designed for roughly 165–175 WPM final narration,
# depending on punctuation and sentence structure.

MOODS = {
    "intro":    {"rate": "+13%", "pitch": "-4Hz", "volume": "+0%"},
    "analysis": {"rate": "+10%", "pitch": "-4Hz", "volume": "+0%"},
    "key_stat": {"rate": "+15%", "pitch": "+0Hz", "volume": "+0%"},
    "warning":  {"rate":  "+7%", "pitch": "-4Hz", "volume": "+0%"},
    "outro":    {"rate": "+11%", "pitch": "-4Hz", "volume": "+0%"},
}

_REQUIRED_TTS_MOODS = {"intro", "analysis", "key_stat", "warning", "outro"}

if set(MOODS) != _REQUIRED_TTS_MOODS:
    raise RuntimeError(
        "MOODS must contain exactly: "
        + ", ".join(sorted(_REQUIRED_TTS_MOODS))
    )

for _mood_name, _profile in MOODS.items():
    for _field in ("rate", "pitch", "volume"):
        if _field not in _profile:
            raise RuntimeError(f"MOODS[{_mood_name!r}] missing {_field!r}")

def vx_tts_profile(mood="analysis"):
    if mood not in MOODS:
        raise KeyError(f"Unknown TTS mood: {mood}")
    return dict(MOODS[mood])

print(f"✅ TTS voice: {VOICE}")

print("✅ TTS moods: " + " • ".join(
    f"{name} {profile['rate']} / {profile['pitch']}"
    for name, profile in MOODS.items()
))

print("✅ SHARED PRODUCTION HELPERS READY")
print("✅ Output folders are owned only by Cell 0")
# ============================================================
# SHARED PRODUCTION HELPERS — SINGLE SOURCE OF TRUTH
# ============================================================
import json, re, subprocess, unicodedata, math


# ============================================================
# EDGE-TTS LOADER + WORD-BOUNDARY STREAM — SINGLE SOURCE OF TRUTH
# ============================================================
# edge-tts releases before and after the boundary constructor option are both
# supported. The stream itself remains the timing authority, so an already
# running Colab never has to disconnect merely because its package is older.
import asyncio as _vx_asyncio
import importlib as _vx_importlib
import inspect as _vx_inspect
import sys as _vx_sys

def _vx_edge_tts_supports_boundary_option(module):
    try:
        return "boundary" in _vx_inspect.signature(module.Communicate).parameters
    except (TypeError, ValueError, AttributeError):
        return False

def vx_load_edge_tts():
    """Load edge-tts once; install it only when the package is absent."""
    try:
        return _vx_importlib.import_module("edge_tts")
    except ImportError:
        subprocess.run(
            [_vx_sys.executable, "-m", "pip", "install", "-q", "edge-tts"],
            check=True,
        )
        _vx_importlib.invalidate_caches()
        return _vx_importlib.import_module("edge_tts")

async def vx_edge_tts_stream(text, profile, attempts=3):
    """Return MP3 bytes plus WordBoundary timing across edge-tts versions."""
    module = vx_load_edge_tts()
    failures = []
    supports_boundary = _vx_edge_tts_supports_boundary_option(module)

    for attempt in range(1, int(attempts) + 1):
        audio = bytearray()
        words = []
        try:
            kwargs = {
                "rate": str(profile["rate"]),
                "pitch": str(profile["pitch"]),
                "volume": str(profile["volume"]),
            }
            if supports_boundary:
                kwargs["boundary"] = "WordBoundary"

            communicate = module.Communicate(str(text), VOICE, **kwargs)
            async for chunk in communicate.stream():
                if chunk.get("type") == "audio":
                    audio.extend(chunk.get("data", b""))
                elif chunk.get("type") == "WordBoundary":
                    words.append({
                        "text": str(chunk.get("text", "")),
                        "start": float(chunk.get("offset", 0)) / 10_000_000,
                        "duration": float(chunk.get("duration", 0)) / 10_000_000,
                    })

            if audio and words:
                return bytes(audio), words
            raise RuntimeError(
                f"empty TTS payload (audio_bytes={len(audio)}, word_boundaries={len(words)})"
            )
        except Exception as exc:
            failures.append(f"attempt {attempt}: {type(exc).__name__}: {exc}")
            if attempt < int(attempts):
                print(f"⚠️ Ryan TTS retry {attempt}/{attempts - 1}: {type(exc).__name__}")
                await _vx_asyncio.sleep(1.5 * attempt)

    raise RuntimeError(
        "Ryan TTS failed after package-compatible setup and retries: "
        + " | ".join(failures)
    )


def vx_media_duration(path):
    r=subprocess.run(["ffprobe","-v","error","-show_entries","format=duration","-of","default=noprint_wrappers=1:nokey=1",str(path)],capture_output=True,text=True,check=True)
    return float(r.stdout.strip())


def vx_frame_plan(source_duration, fps):
    """Quantize a narration duration to the nearest constant-FPS frame grid.

    Constant-frame-rate MP4 duration is an integer number of frames. Using
    ceil() always biases low-FPS Draft renders long; nearest-frame quantization
    keeps the unavoidable error within half a frame without loosening QA.
    """
    duration = float(source_duration)
    rate = int(fps)
    if not math.isfinite(duration) or duration <= 0:
        raise ValueError(f"Invalid source duration: {source_duration!r}")
    if rate <= 0:
        raise ValueError(f"Invalid FPS: {fps!r}")

    frames = max(1, int(math.floor(duration * rate + 0.5)))
    grid_duration = frames / rate
    quantization_error = grid_duration - duration
    max_quantization_error = 0.5 / rate

    if abs(quantization_error) > max_quantization_error + 1e-9:
        raise RuntimeError(
            "Frame-grid quantization exceeded half a frame: "
            f"duration={duration:.6f}, fps={rate}, frames={frames}"
        )

    return {
        "frames": frames,
        "grid_duration": grid_duration,
        "source_duration": duration,
        "quantization_error": quantization_error,
        "max_quantization_error": max_quantization_error,
    }


def vx_video_info(path):
    r=subprocess.run(["ffprobe","-v","error","-select_streams","v:0","-show_entries","stream=width,height,avg_frame_rate","-of","json",str(path)],capture_output=True,text=True,check=True)
    return json.loads(r.stdout)["streams"][0]

async def vx_apply_render_scale(page, output_w, output_h, design_w=3840, design_h=2160):
    sx=float(output_w)/float(design_w); sy=float(output_h)/float(design_h)
    css=f"""html{{width:{int(output_w)}px!important;height:{int(output_h)}px!important;min-width:{int(output_w)}px!important;min-height:{int(output_h)}px!important;max-width:{int(output_w)}px!important;max-height:{int(output_h)}px!important;margin:0!important;padding:0!important;overflow:hidden!important;background:#000!important;}}body{{position:absolute!important;left:0!important;top:0!important;width:{int(design_w)}px!important;height:{int(design_h)}px!important;min-width:{int(design_w)}px!important;min-height:{int(design_h)}px!important;margin:0!important;padding:0!important;transform:scale({sx},{sy})!important;transform-origin:0 0!important;overflow:hidden!important;}}*,*::before,*::after{{animation:none!important;transition:none!important;scroll-behavior:auto!important;}}"""
    await page.add_style_tag(content=css)

def vx_norm_text(value):
    text=unicodedata.normalize("NFKD",str(value or ""))
    text="".join(ch for ch in text if not unicodedata.combining(ch))
    return re.sub(r"[^a-z0-9]+"," ",text.lower()).strip()

def vx_phrase_span(words, phrase, *, required=False, context=""):
    """Return exact Ryan WordBoundary start/span for a spoken phrase."""
    target=vx_norm_text(phrase).split()
    if not target:
        if required:
            raise RuntimeError(f"Empty sync phrase in {context or 'animation sync'}.")
        return None

    flat=[]
    for boundary_index,w in enumerate(words or []):
        for token in vx_norm_text(w.get("text","")).split():
            flat.append((token,boundary_index))

    tokens=[token for token,_ in flat]
    for i in range(len(tokens)-len(target)+1):
        if tokens[i:i+len(target)] != target:
            continue
        first_boundary=flat[i][1]
        last_boundary=flat[i+len(target)-1][1]
        first=words[first_boundary]
        last=words[last_boundary]
        start=float(first["start"])
        end=float(last["start"])+float(last.get("duration",0.0))
        return start,max(0.01,end-start)

    if required:
        label=f" in {context}" if context else ""
        raise RuntimeError(
            f"Ryan exact-sync phrase not found{label}: {phrase!r}. "
            f"Received {len(words or [])} WordBoundary events."
        )
    return None

def vx_phrase_time(words, phrase, fallback=0.0, *, required=False, context=""):
    span=vx_phrase_span(words,phrase,required=required,context=context)
    return float(span[0]) if span is not None else fallback

print("✅ Shared production helpers ready")


# ============================================================
# FIX 2 — SHARED 4K BRANDING / READABILITY POLISH
# ONE helper is reused by every presentation HTML.
# It changes presentation only — never data/model calculations.
# ============================================================

VX_BRAND_POLISH_CSS = r"""
<style id="vx-brand-polish">
:root{
  --vx-brand-gold:#ffd447;
  --vx-brand-cyan:#35e6ff;
  --vx-brand-white:#f7fbff;
}

/* Two-colour channel lockup */
.channelName{
  display:flex!important;
  align-items:baseline!important;
  gap:20px!important;
  font-size:104px!important;
  line-height:.94!important;
  font-weight:1000!important;
  letter-spacing:1.5px!important;
  color:var(--vx-brand-white)!important;
  text-shadow:none!important;
}
.channelName .vx-brand-fpl,
.vx-brand-inline .vx-brand-fpl,
.pc-footer-brand .vx-brand-fpl{
  color:var(--vx-brand-gold)!important;
  text-shadow:0 5px 20px rgba(255,212,71,.22)!important;
}
.channelName .vx-brand-vortex,
.vx-brand-inline .vx-brand-vortex,
.pc-footer-brand .vx-brand-vortex{
  color:var(--vx-brand-cyan)!important;
  font-style:italic!important;
  letter-spacing:2px!important;
  text-shadow:0 0 24px rgba(53,230,255,.20)!important;
}
.channelSub{
  font-size:34px!important;
  font-weight:950!important;
  letter-spacing:5.5px!important;
  color:#d4e2ec!important;
}

/* Bigger, cleaner channel logo on every slide family */
.logoBox{
  width:790px!important;
  height:285px!important;
}
.logoBox .vortex-logo{
  max-width:770px!important;
  max-height:275px!important;
  filter:
    drop-shadow(0 12px 28px rgba(0,0,0,.46))
    drop-shadow(0 0 24px rgba(53,230,255,.12))!important;
}
.outroScene .logoBox{
  width:720px!important;
  height:250px!important;
}
.outroScene .logoBox .vortex-logo{
  max-width:700px!important;
  max-height:240px!important;
}
.outroScene .channelName{
  font-size:96px!important;
}
.outroScene .channelSub{
  font-size:30px!important;
  letter-spacing:5px!important;
}

.brand.brandImage{
  width:1320px!important;
  height:380px!important;
}
.brand.brandImage .vortex-logo{
  max-width:1280px!important;
  max-height:360px!important;
  filter:
    drop-shadow(0 14px 22px rgba(0,0,0,.48))
    drop-shadow(0 0 22px rgba(255,212,71,.10))!important;
}

.logo-area{
  width:900px!important;
  height:180px!important;
  overflow:visible!important;
}
.logo-area .vortex-logo{
  max-width:820px!important;
  max-height:178px!important;
  filter:
    drop-shadow(0 8px 20px rgba(0,0,0,.42))
    drop-shadow(0 0 26px rgba(53,230,255,.12))!important;
}

/* Global small-text readability uplift; large headings retain their own sizes */
.footer{
  font-size:38px!important;
}
.footerLeft strong{
  font-size:45px!important;
}
.footerRight .updated{
  font-size:36px!important;
}
.footerRight .season{
  font-size:31px!important;
}
.footerSocial{
  font-size:40px!important;
}
.footer-section{
  font-size:35px!important;
}
.header-kicker{
  font-size:40px!important;
}
.header-season{
  font-size:34px!important;
}
.subtitle{
  font-size:34px!important;
  line-height:1.16!important;
}

.panelTitle .smallTitle{
  font-size:28px!important;
}
.metricLabel{
  font-size:36px!important;
}
.metricSub{
  font-size:29px!important;
}
.chipName{
  font-size:25px!important;
}
.chipStatus{
  font-size:22px!important;
}
.pitchHeader .gw{
  font-size:56px!important;
}
.pitchHeader .team{
  font-size:31px!important;
}
.playerName{
  font-size:34px!important;
}
.playerMeta{
  font-size:30px!important;
}
.tinyScore{
  font-size:18px!important;
}
.points{
  font-size:48px!important;
}
.miniName{
  font-size:33px!important;
}
.miniMeta{
  font-size:30px!important;
}
.miniPts{
  font-size:44px!important;
}

/* Outro phone/1080p legibility */
.outroScene .panelTag{
  font-size:34px!important;
}
.outroScene .todayPanel .lineText{
  font-size:48px!important;
}
.outroScene .todayPanel .lineSub{
  font-size:42px!important;
}
.outroScene .todayPanel .takeawaySub{
  font-size:38px!important;
}
.outroScene .tomorrowPanel .tom-sub{
  font-size:33px!important;
}
.outroScene .tomorrowPanel .lineText{
  font-size:54px!important;
}


/* Unified footer-note readability across every slide family */
.footer,
.vortex-footer{
  font-size:38px!important;
  line-height:1.05!important;
}
.footerLeft,
.footerRight,
.footerSocial,
.footer-section,
.social-group{
  font-size:38px!important;
  line-height:1.05!important;
  font-weight:950!important;
}
.footer-label{
  font-size:33px!important;
  font-weight:950!important;
}
.footer-value{
  font-size:38px!important;
  font-weight:1000!important;
}
.footerLeft strong{
  font-size:45px!important;
}
.footerRight .updated{
  font-size:36px!important;
}
.footerRight .season{
  font-size:31px!important;
}

/* Player hero-card footer branding */
.pc-footer-brand{
  font-size:clamp(8px,1.35cqw,20px)!important;
  letter-spacing:.02em!important;
}
.pc-footer-copy,
.pc-footer-season{
  font-size:clamp(7px,1.05cqw,16px)!important;
}
</style>
"""

def vx_polish_branding(html):
    """Apply the shared logo/type polish to one complete slide HTML string."""
    text = str(html)

    # Exact two-colour channel-name treatment where the visible lockup exists.
    text = text.replace(
        '<div class="channelName">FPL VORTEX</div>',
        '<div class="channelName"><span class="vx-brand-fpl">FPL</span>'
        '<span class="vx-brand-vortex">VORTEX</span></div>',
    )
    text = text.replace(
        '<div class="pc-footer-brand">FPL VORTEX</div>',
        '<div class="pc-footer-brand vx-brand-inline">'
        '<span class="vx-brand-fpl">FPL</span> '
        '<span class="vx-brand-vortex">VORTEX</span></div>',
    )
    text = text.replace(
        '<strong>FPL VORTEX</strong>',
        '<strong class="vx-brand-inline"><span class="vx-brand-fpl">FPL</span> '
        '<span class="vx-brand-vortex">VORTEX</span></strong>',
    )
    text = text.replace(
        'FPL VORTEX • 2026/27',
        '<span class="vx-brand-inline"><span class="vx-brand-fpl">FPL</span> '
        '<span class="vx-brand-vortex">VORTEX</span></span> • 2026/27',
    )
    text = text.replace(
        'FPL VORTEX &nbsp;|&nbsp;',
        '<span class="vx-brand-inline"><span class="vx-brand-fpl">FPL</span> '
        '<span class="vx-brand-vortex">VORTEX</span></span> &nbsp;|&nbsp;',
    )

    if 'id="vx-brand-polish"' not in text:
        if "</head>" not in text:
            raise RuntimeError("Brand polish requires a complete HTML document with </head>.")
        text = text.replace("</head>", VX_BRAND_POLISH_CSS + "\n</head>", 1)

    return text

print("✅ FIX 2 shared branding polish ready")
print("✅ Logo + typography presentation only — model/data logic unchanged")
```

```python
# ============================================================
# CELL 8 — PHASE 4A
# BASELINE POISSON SCORE-PROBABILITY ENGINE
# MODEL-ONLY • CONSUMES PHASE 3 LAMBDAS ONLY
# NO MARKET ODDS • NO PLAYERELO • NO DIXON–COLES • NO PLAYER XPTS
# ============================================================
# Baseline independent Poisson:
#   P(X=k) = exp(-lambda) * lambda**k / k!
#
# Stored inspection matrix: exact score cells 0–0 through 8–8.
# Scalar probabilities use exact closed forms where available and an adaptive
# Poisson support for W/D/L so the 8-goal display cutoff does not distort tails.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import math
import numpy as np
import pandas as pd

if "PHASE3_READY" not in globals():
    raise RuntimeError("Run Phase 3 before Phase 4A.")

PHASE4_POISSON_MODEL_VERSION = "VORTEX-V1-PHASE4-BASELINE-POISSON"
PHASE4_POISSON_MAX_GOALS = 8
PHASE4_CALCULATION_TIMESTAMP = datetime.now(timezone.utc).isoformat()

PHASE4_PROBABILITY_FIELDS = (
    "fixture_id",
    "gameweek",
    "home_team_id",
    "away_team_id",
    "model_lambda_home",
    "model_lambda_away",
    "home_win_probability",
    "draw_probability",
    "away_win_probability",
    "home_scores_probability",
    "away_scores_probability",
    "home_scores_2plus_probability",
    "away_scores_2plus_probability",
    "home_clean_sheet_probability",
    "away_clean_sheet_probability",
    "both_teams_score_probability",
    "over_1_5_goals_probability",
    "over_2_5_goals_probability",
    "poisson_matrix_covered_mass_0_to_8",
    "model_confidence",
    "model_version",
    "data_snapshot_timestamp",
    "forecast_timestamp",
    "source_name",
    "source_timestamp",
    "is_pre_deadline_data",
)

PHASE4_SCORE_MATRIX_FIELDS = (
    "fixture_id",
    "gameweek",
    "home_team_id",
    "away_team_id",
    "home_goals",
    "away_goals",
    "scoreline",
    "probability",
    "model_lambda_home",
    "model_lambda_away",
    "model_version",
    "forecast_timestamp",
)

fixture_probabilities_model_only = pd.DataFrame(columns=PHASE4_PROBABILITY_FIELDS)
poisson_probability_matrix = pd.DataFrame(columns=PHASE4_SCORE_MATRIX_FIELDS)
PHASE4_POISSON_READY = False


def _vx_phase4_poisson_pmf(lam, k):
    """Numerically stable Poisson PMF without adding a SciPy dependency."""
    lam = float(lam)
    k = int(k)
    if not np.isfinite(lam) or lam <= 0:
        raise RuntimeError(f"Poisson lambda must be finite and > 0, got {lam}")
    if k < 0:
        return 0.0
    return float(math.exp(-lam + k * math.log(lam) - math.lgamma(k + 1)))


def _vx_phase4_poisson_vector(lam, max_k):
    return np.array(
        [_vx_phase4_poisson_pmf(lam, k) for k in range(int(max_k) + 1)],
        dtype=float,
    )


def _vx_phase4_adaptive_poisson_vector(lam, tail_tolerance=1e-12):
    """
    Build enough Poisson support that omitted marginal tail mass is negligible.
    This is used only for W/D/L aggregation; the persisted score matrix remains
    exactly 0..8 as requested.
    """
    lam = float(lam)
    if not np.isfinite(lam) or lam <= 0:
        raise RuntimeError(f"Poisson lambda must be finite and > 0, got {lam}")

    max_k = max(20, int(math.ceil(lam + 10.0 * math.sqrt(lam + 1.0))))
    hard_cap = max(80, max_k)

    while True:
        probs = _vx_phase4_poisson_vector(lam, max_k)
        mass = float(probs.sum())
        if mass >= 1.0 - float(tail_tolerance):
            return probs
        if max_k >= hard_cap:
            if mass < 1.0 - 1e-9:
                raise RuntimeError(
                    f"Poisson tail support insufficient for lambda={lam:.6f}; mass={mass:.12f}"
                )
            return probs
        max_k = min(hard_cap, max_k + 10)


def _vx_phase4_wdl_probabilities(lambda_home, lambda_away):
    """
    Tail-safe independent-Poisson W/D/L.
    Rows are home goals; columns are away goals.
    """
    home = _vx_phase4_adaptive_poisson_vector(lambda_home)
    away = _vx_phase4_adaptive_poisson_vector(lambda_away)
    matrix = np.outer(home, away)
    mass = float(matrix.sum())
    if not np.isfinite(mass) or mass <= 0:
        raise RuntimeError("Invalid adaptive Poisson score mass")
    # Only numerical tail residue is normalized away (<~1e-12 by construction).
    matrix = matrix / mass
    p_home = float(np.tril(matrix, -1).sum())
    p_draw = float(np.trace(matrix))
    p_away = float(np.triu(matrix, 1).sum())
    return p_home, p_draw, p_away


def _vx_phase4_closed_form_probabilities(lambda_home, lambda_away):
    """
    Exact independent-Poisson marginals for scoring, CS, BTTS and O2.5.
    """
    lh = float(lambda_home)
    la = float(lambda_away)

    p_h0 = math.exp(-lh)
    p_a0 = math.exp(-la)

    home_scores = 1.0 - p_h0
    away_scores = 1.0 - p_a0
    home_2plus = 1.0 - p_h0 * (1.0 + lh)
    away_2plus = 1.0 - p_a0 * (1.0 + la)

    home_cs = p_a0
    away_cs = p_h0
    btts = home_scores * away_scores

    total_lambda = lh + la
    p_total_0_to_1 = math.exp(-total_lambda) * (1.0 + total_lambda)
    p_total_0_to_2 = math.exp(-total_lambda) * (
        1.0 + total_lambda + (total_lambda ** 2) / 2.0
    )
    over_1_5 = 1.0 - p_total_0_to_1
    over_2_5 = 1.0 - p_total_0_to_2

    return {
        "home_scores_probability": float(home_scores),
        "away_scores_probability": float(away_scores),
        "home_scores_2plus_probability": float(home_2plus),
        "away_scores_2plus_probability": float(away_2plus),
        "home_clean_sheet_probability": float(home_cs),
        "away_clean_sheet_probability": float(away_cs),
        "both_teams_score_probability": float(btts),
        "over_1_5_goals_probability": float(over_1_5),
        "over_2_5_goals_probability": float(over_2_5),
    }


if not bool(PHASE3_READY):
    print("⛔ PHASE 4A BLOCKED — Phase 3 is not ready.")
    print("   No Poisson probabilities were emitted.")
else:
    _required = {
        "fixture_id", "gameweek", "home_team_id", "away_team_id",
        "model_lambda_home", "model_lambda_away", "model_confidence",
        "model_version", "data_snapshot_timestamp", "forecast_timestamp",
        "is_pre_deadline_data",
    }
    _missing = sorted(_required - set(projected_goals_model_only.columns))
    if _missing:
        raise RuntimeError(f"Phase 4A missing Phase-3 fields: {_missing}")
    if projected_goals_model_only.empty:
        raise RuntimeError("Phase 4A received no Phase-3 fixtures")
    if projected_goals_model_only["fixture_id"].duplicated().any():
        raise RuntimeError("Phase 4A requires unique Phase-3 fixture IDs")

    _probability_rows = []
    _matrix_rows = []

    for _, _fx in projected_goals_model_only.iterrows():
        _fid = int(_fx["fixture_id"])
        _gw = int(_fx["gameweek"])
        _home_id = int(_fx["home_team_id"])
        _away_id = int(_fx["away_team_id"])
        _lh = float(_fx["model_lambda_home"])
        _la = float(_fx["model_lambda_away"])

        if not np.isfinite(_lh) or not np.isfinite(_la) or _lh <= 0 or _la <= 0:
            raise RuntimeError(f"Phase 4A invalid lambda for fixture {_fid}: {_lh}, {_la}")

        # Persisted 0..8 score matrix: no renormalization, so each cell remains
        # the true independent-Poisson score probability.
        _home_0_8 = _vx_phase4_poisson_vector(_lh, PHASE4_POISSON_MAX_GOALS)
        _away_0_8 = _vx_phase4_poisson_vector(_la, PHASE4_POISSON_MAX_GOALS)
        _matrix_0_8 = np.outer(_home_0_8, _away_0_8)
        _covered_mass = float(_matrix_0_8.sum())

        for _hg in range(PHASE4_POISSON_MAX_GOALS + 1):
            for _ag in range(PHASE4_POISSON_MAX_GOALS + 1):
                _matrix_rows.append({
                    "fixture_id": _fid,
                    "gameweek": _gw,
                    "home_team_id": _home_id,
                    "away_team_id": _away_id,
                    "home_goals": _hg,
                    "away_goals": _ag,
                    "scoreline": f"{_hg}-{_ag}",
                    "probability": float(_matrix_0_8[_hg, _ag]),
                    "model_lambda_home": _lh,
                    "model_lambda_away": _la,
                    "model_version": PHASE4_POISSON_MODEL_VERSION,
                    "forecast_timestamp": _fx["forecast_timestamp"],
                })

        _p_home, _p_draw, _p_away = _vx_phase4_wdl_probabilities(_lh, _la)
        _closed = _vx_phase4_closed_form_probabilities(_lh, _la)

        _probability_rows.append({
            "fixture_id": _fid,
            "gameweek": _gw,
            "home_team_id": _home_id,
            "away_team_id": _away_id,
            "model_lambda_home": _lh,
            "model_lambda_away": _la,
            "home_win_probability": _p_home,
            "draw_probability": _p_draw,
            "away_win_probability": _p_away,
            **_closed,
            "poisson_matrix_covered_mass_0_to_8": _covered_mass,
            "model_confidence": float(_fx["model_confidence"]),
            "model_version": PHASE4_POISSON_MODEL_VERSION,
            "data_snapshot_timestamp": _fx["data_snapshot_timestamp"],
            "forecast_timestamp": _fx["forecast_timestamp"],
            "source_name": "VORTEX_PHASE4_BASELINE_POISSON",
            "source_timestamp": PHASE4_CALCULATION_TIMESTAMP,
            "is_pre_deadline_data": _fx["is_pre_deadline_data"],
        })

    fixture_probabilities_model_only = pd.DataFrame(
        _probability_rows, columns=PHASE4_PROBABILITY_FIELDS
    )
    poisson_probability_matrix = pd.DataFrame(
        _matrix_rows, columns=PHASE4_SCORE_MATRIX_FIELDS
    )

    if fixture_probabilities_model_only.empty or poisson_probability_matrix.empty:
        raise RuntimeError("Phase 4A produced empty probability outputs")

    _prob_cols = [
        "home_win_probability", "draw_probability", "away_win_probability",
        "home_scores_probability", "away_scores_probability",
        "home_scores_2plus_probability", "away_scores_2plus_probability",
        "home_clean_sheet_probability", "away_clean_sheet_probability",
        "both_teams_score_probability", "over_1_5_goals_probability",
        "over_2_5_goals_probability", "poisson_matrix_covered_mass_0_to_8",
    ]
    for _c in _prob_cols:
        _s = pd.to_numeric(fixture_probabilities_model_only[_c], errors="coerce")
        if _s.isna().any() or not _s.between(0.0, 1.0, inclusive="both").all():
            raise RuntimeError(f"Phase 4A probability field out of range: {_c}")

    _wdl_sum = (
        fixture_probabilities_model_only["home_win_probability"]
        + fixture_probabilities_model_only["draw_probability"]
        + fixture_probabilities_model_only["away_win_probability"]
    )
    if not np.allclose(_wdl_sum.to_numpy(dtype=float), 1.0, atol=1e-10, rtol=0.0):
        raise RuntimeError("Phase 4A W/D/L probabilities do not sum to 1")

    # Required clean-sheet identities.
    _expected_home_cs = np.exp(
        -fixture_probabilities_model_only["model_lambda_away"].to_numpy(dtype=float)
    )
    _expected_away_cs = np.exp(
        -fixture_probabilities_model_only["model_lambda_home"].to_numpy(dtype=float)
    )
    if not np.allclose(
        fixture_probabilities_model_only["home_clean_sheet_probability"].to_numpy(dtype=float),
        _expected_home_cs,
        atol=1e-12,
        rtol=0.0,
    ):
        raise RuntimeError("Phase 4A home clean-sheet identity failed")
    if not np.allclose(
        fixture_probabilities_model_only["away_clean_sheet_probability"].to_numpy(dtype=float),
        _expected_away_cs,
        atol=1e-12,
        rtol=0.0,
    ):
        raise RuntimeError("Phase 4A away clean-sheet identity failed")

    _matrix_group_sizes = poisson_probability_matrix.groupby("fixture_id").size()
    _expected_cells = (PHASE4_POISSON_MAX_GOALS + 1) ** 2
    if not (_matrix_group_sizes == _expected_cells).all():
        raise RuntimeError("Phase 4A score matrix must contain exactly 81 cells per fixture")

    PHASE4_MATRIX_PATH = Path(DATA_DIR) / "phase4_poisson_probability_matrix_0_to_8.json"
    PHASE4_PROBABILITY_PATH = Path(DATA_DIR) / "phase4_fixture_probabilities_model_only.json"

    PHASE4_MATRIX_PATH.write_text(
        poisson_probability_matrix.to_json(orient="records", date_format="iso", indent=2),
        encoding="utf-8",
    )
    PHASE4_PROBABILITY_PATH.write_text(
        fixture_probabilities_model_only.to_json(orient="records", date_format="iso", indent=2),
        encoding="utf-8",
    )

    PHASE4_POISSON_READY = True
    print("✅ PHASE 4A BASELINE POISSON BUILT")
    print(f"   Stored score matrix: 0–0 through {PHASE4_POISSON_MAX_GOALS}–{PHASE4_POISSON_MAX_GOALS}")
    print(f"   Matrix cells per fixture: {_expected_cells}")
    print("   W/D/L uses adaptive tail-safe Poisson support.")
    print("   CS identities: P(Home CS)=exp(-lambda_away), P(Away CS)=exp(-lambda_home)")
    print("   No market odds, PlayerElo, Dixon–Coles or player xPts used.")
    print("   Saved:", PHASE4_MATRIX_PATH)
    print("   Saved:", PHASE4_PROBABILITY_PATH)
```

```python
# ============================================================
# CELL 9 — PHASE 4B
# MODEL-ONLY CLEAN-SHEET + CONTINUOUS ATTACK/DEFENCE FDR OUTPUT
# CONSUMES PHASE 4A PROBABILITIES ONLY
# NO FDR AVERAGING FOR PLAYER DECISIONS • PRESENTATION BANDS ONLY
# ============================================================
# Attack FDR:
#   1 + 4 * (3.2 - lambda_for) / 2.8
#
# Defence FDR:
#   1 + 4 * (exp(-0.4) - P(CS)) / (exp(-0.4) - exp(-3.2))
#
# The 1..5 scale has natural endpoint lambdas 3.2 (easy) and 0.4 (hard)
# for attack, with the corresponding CS endpoint probabilities for defence.
# Values outside those endpoints are boundary-clipped to [1, 5] only so the
# defined FDR scale remains valid; this is not a new strength adjustment.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import math
import numpy as np
import pandas as pd

if "PHASE4_POISSON_READY" not in globals():
    raise RuntimeError("Run Phase 4A before Phase 4B.")

PHASE4_OUTPUT_MODEL_VERSION = "VORTEX-V1-PHASE4-POISSON-CS-FDR"
PHASE4_OUTPUT_TIMESTAMP = datetime.now(timezone.utc).isoformat()

PHASE4_FDR_BANDS = (
    (1.00, 1.80, "Dark green"),
    (1.80, 2.60, "Green"),
    (2.60, 3.40, "Grey"),
    (3.40, 4.20, "Orange"),
    (4.20, 5.0000001, "Red"),
)

PHASE4_FDR_OUTPUT_FIELDS = (
    "fixture_id",
    "gameweek",
    "home_team_id",
    "away_team_id",
    "model_lambda_home",
    "model_lambda_away",
    "home_clean_sheet_probability",
    "away_clean_sheet_probability",
    "attack_fdr_home",
    "defence_fdr_home",
    "attack_fdr_away",
    "defence_fdr_away",
    "attack_fdr_home_colour_band",
    "defence_fdr_home_colour_band",
    "attack_fdr_away_colour_band",
    "defence_fdr_away_colour_band",
    "model_confidence",
    "model_version",
    "data_snapshot_timestamp",
    "forecast_timestamp",
    "source_name",
    "source_timestamp",
    "is_pre_deadline_data",
)

fixture_fdr_model_only = pd.DataFrame(columns=PHASE4_FDR_OUTPUT_FIELDS)
clean_sheet_probabilities_model_only = pd.DataFrame()
PHASE4_READY = False


def _vx_phase4_clip_fdr(value):
    value = float(value)
    if not np.isfinite(value):
        raise RuntimeError(f"Non-finite FDR value: {value}")
    return float(min(5.0, max(1.0, value)))


def vx_phase4_attack_fdr(lambda_for):
    """
    Continuous attack difficulty. Lower = easier attacking fixture.
    Formula requested for V1; output boundary-clipped to the defined 1..5 scale.
    """
    lam = float(lambda_for)
    if not np.isfinite(lam) or lam <= 0:
        raise RuntimeError(f"Attack FDR requires finite positive lambda, got {lam}")
    raw = 1.0 + 4.0 * (3.2 - lam) / 2.8
    return _vx_phase4_clip_fdr(raw)


def vx_phase4_defence_fdr(clean_sheet_probability):
    """
    Continuous defensive difficulty. Lower = easier clean-sheet environment.
    """
    p_cs = float(clean_sheet_probability)
    if not np.isfinite(p_cs) or not 0.0 <= p_cs <= 1.0:
        raise RuntimeError(f"Defence FDR requires CS probability in [0,1], got {p_cs}")
    easy_cs = math.exp(-0.4)
    hard_cs = math.exp(-3.2)
    raw = 1.0 + 4.0 * (easy_cs - p_cs) / (easy_cs - hard_cs)
    return _vx_phase4_clip_fdr(raw)


def vx_phase4_fdr_colour_band(fdr_value):
    """Presentation-only colour band applied after continuous FDR is calculated."""
    value = float(fdr_value)
    if not np.isfinite(value) or not 1.0 <= value <= 5.0:
        raise RuntimeError(f"FDR colour band requires value in [1,5], got {value}")
    if value < 1.80:
        return "Dark green"
    if value < 2.60:
        return "Green"
    if value < 3.40:
        return "Grey"
    if value < 4.20:
        return "Orange"
    return "Red"


if not bool(PHASE4_POISSON_READY):
    print("⛔ PHASE 4B BLOCKED — baseline Poisson is not ready.")
    print("   No clean-sheet/FDR outputs were emitted.")
else:
    _required = {
        "fixture_id", "gameweek", "home_team_id", "away_team_id",
        "model_lambda_home", "model_lambda_away",
        "home_clean_sheet_probability", "away_clean_sheet_probability",
        "model_confidence", "data_snapshot_timestamp", "forecast_timestamp",
        "is_pre_deadline_data",
    }
    _missing = sorted(_required - set(fixture_probabilities_model_only.columns))
    if _missing:
        raise RuntimeError(f"Phase 4B missing Phase-4A fields: {_missing}")
    if fixture_probabilities_model_only.empty:
        raise RuntimeError("Phase 4B received no fixture probabilities")

    _rows = []
    for _, _fx in fixture_probabilities_model_only.iterrows():
        _lh = float(_fx["model_lambda_home"])
        _la = float(_fx["model_lambda_away"])
        _home_cs = float(_fx["home_clean_sheet_probability"])
        _away_cs = float(_fx["away_clean_sheet_probability"])

        # Attack FDR uses the team's own scoring lambda.
        _att_h = vx_phase4_attack_fdr(_lh)
        _att_a = vx_phase4_attack_fdr(_la)

        # Defence FDR uses the team's own CS probability.
        _def_h = vx_phase4_defence_fdr(_home_cs)
        _def_a = vx_phase4_defence_fdr(_away_cs)

        _rows.append({
            "fixture_id": int(_fx["fixture_id"]),
            "gameweek": int(_fx["gameweek"]),
            "home_team_id": int(_fx["home_team_id"]),
            "away_team_id": int(_fx["away_team_id"]),
            "model_lambda_home": _lh,
            "model_lambda_away": _la,
            "home_clean_sheet_probability": _home_cs,
            "away_clean_sheet_probability": _away_cs,
            "attack_fdr_home": _att_h,
            "defence_fdr_home": _def_h,
            "attack_fdr_away": _att_a,
            "defence_fdr_away": _def_a,
            "attack_fdr_home_colour_band": vx_phase4_fdr_colour_band(_att_h),
            "defence_fdr_home_colour_band": vx_phase4_fdr_colour_band(_def_h),
            "attack_fdr_away_colour_band": vx_phase4_fdr_colour_band(_att_a),
            "defence_fdr_away_colour_band": vx_phase4_fdr_colour_band(_def_a),
            "model_confidence": float(_fx["model_confidence"]),
            "model_version": PHASE4_OUTPUT_MODEL_VERSION,
            "data_snapshot_timestamp": _fx["data_snapshot_timestamp"],
            "forecast_timestamp": _fx["forecast_timestamp"],
            "source_name": "VORTEX_PHASE4_MODEL_ONLY_CS_FDR",
            "source_timestamp": PHASE4_OUTPUT_TIMESTAMP,
            "is_pre_deadline_data": _fx["is_pre_deadline_data"],
        })

    fixture_fdr_model_only = pd.DataFrame(_rows, columns=PHASE4_FDR_OUTPUT_FIELDS)

    if fixture_fdr_model_only.empty:
        raise RuntimeError("Phase 4B produced no FDR rows")
    if fixture_fdr_model_only["fixture_id"].duplicated().any():
        raise RuntimeError("Phase 4B duplicate fixture IDs")

    _fdr_cols = [
        "attack_fdr_home", "defence_fdr_home",
        "attack_fdr_away", "defence_fdr_away",
    ]
    for _c in _fdr_cols:
        _s = pd.to_numeric(fixture_fdr_model_only[_c], errors="coerce")
        if _s.isna().any() or not _s.between(1.0, 5.0, inclusive="both").all():
            raise RuntimeError(f"Phase 4B FDR out of [1,5]: {_c}")

    # Formula boundary checks: lambda=3.2 -> attack FDR 1, lambda=0.4 -> 5.
    if not math.isclose(vx_phase4_attack_fdr(3.2), 1.0, abs_tol=1e-12):
        raise RuntimeError("Phase 4B attack-FDR easy endpoint failed")
    if not math.isclose(vx_phase4_attack_fdr(0.4), 5.0, abs_tol=1e-12):
        raise RuntimeError("Phase 4B attack-FDR hard endpoint failed")
    if not math.isclose(vx_phase4_defence_fdr(math.exp(-0.4)), 1.0, abs_tol=1e-12):
        raise RuntimeError("Phase 4B defence-FDR easy endpoint failed")
    if not math.isclose(vx_phase4_defence_fdr(math.exp(-3.2)), 5.0, abs_tol=1e-12):
        raise RuntimeError("Phase 4B defence-FDR hard endpoint failed")

    # Exact requested colour boundaries.
    _band_tests = {
        1.00: "Dark green",
        1.79: "Dark green",
        1.80: "Green",
        2.59: "Green",
        2.60: "Grey",
        3.39: "Grey",
        3.40: "Orange",
        4.19: "Orange",
        4.20: "Red",
        5.00: "Red",
    }
    for _value, _expected in _band_tests.items():
        _actual = vx_phase4_fdr_colour_band(_value)
        if _actual != _expected:
            raise RuntimeError(
                f"Phase 4B colour boundary failed at {_value}: {_actual} != {_expected}"
            )

    # Separate CS table for downstream validation; still model-only.
    clean_sheet_probabilities_model_only = fixture_fdr_model_only[[
        "fixture_id", "gameweek", "home_team_id", "away_team_id",
        "model_lambda_home", "model_lambda_away",
        "home_clean_sheet_probability", "away_clean_sheet_probability",
        "model_confidence", "model_version", "data_snapshot_timestamp",
        "forecast_timestamp", "source_name", "source_timestamp",
        "is_pre_deadline_data",
    ]].copy()

    PHASE4_CS_PATH = Path(DATA_DIR) / "phase4_clean_sheet_probabilities_model_only.json"
    PHASE4_FDR_PATH = Path(DATA_DIR) / "phase4_continuous_attack_defence_fdr_model_only.json"

    PHASE4_CS_PATH.write_text(
        clean_sheet_probabilities_model_only.to_json(
            orient="records", date_format="iso", indent=2
        ),
        encoding="utf-8",
    )
    PHASE4_FDR_PATH.write_text(
        fixture_fdr_model_only.to_json(
            orient="records", date_format="iso", indent=2
        ),
        encoding="utf-8",
    )

    PHASE4_READY = True
    print("✅ PHASE 4B CLEAN-SHEET + CONTINUOUS FDR OUTPUT BUILT")
    print("   Attack FDR and Defence FDR remain separate — no averaging is calculated.")
    print("   Colour bands are presentation-only and applied after continuous FDR.")
    print("   1.00–1.79 Dark green | 1.80–2.59 Green | 2.60–3.39 Grey")
    print("   3.40–4.19 Orange | 4.20–5.00 Red")
    print("   No market odds, PlayerElo, Dixon–Coles or player xPts used.")
    print("   Saved:", PHASE4_CS_PATH)
    print("   Saved:", PHASE4_FDR_PATH)
```

```python
# ============================================================
# CELL 12 — PHASE 5
# MARKET-ODDS SNAPSHOT + DEADLINE-SAFE BENCHMARK
# RAW ODDS STAY SEPARATE • NO MARKET BLEND IN V1
# ============================================================
# Phase 5 does exactly four things:
#   1) preserve raw provider odds snapshots independently of model outputs;
#   2) convert decimal odds to implied probabilities and remove bookmaker margin;
#   3) compare deadline-safe market probabilities with the Phase-4 model;
#   4) fit MARKET-ONLY expected goals from de-vigged 1X2 + O/U 2.5 solely
#      to derive a market-implied score distribution / clean-sheet benchmark.
#
# It does NOT modify projected goals, Poisson probabilities, CS probabilities,
# Attack/Defence FDR, or any player/squad decision output.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from difflib import SequenceMatcher
import hashlib
import json
import math
import os
import re
import requests
import numpy as np
import pandas as pd

PHASE5_MODEL_VERSION = "VORTEX-V1-PHASE5-MARKET-BENCHMARK"
PHASE5_MARKET_BLEND_ACTIVE = False
PHASE5_READY = False
PHASE5_CAPTURE_STATUS = "NOT_RUN"
PHASE5_BENCHMARK_STATUS = "NOT_RUN"

if PHASE5_MARKET_BLEND_ACTIVE:
    raise RuntimeError("V1 clean-room violation: Phase 5 market blending must remain OFF.")

# Exact user-requested raw odds fields come first. Provenance/mapping fields follow.
PHASE5_RAW_ODDS_FIELDS = (
    "fixture_id",
    "odds_snapshot_timestamp",
    "bookmaker_or_exchange",
    "home_win_odds",
    "draw_odds",
    "away_win_odds",
    "over_1_5_odds",
    "over_2_5_odds",
    "under_2_5_odds",
    "both_teams_score_yes_odds",
    "both_teams_score_no_odds",
    "market_margin",
    "is_before_fpl_deadline",
    "gameweek",
    "home_team_id",
    "away_team_id",
    "home_team",
    "away_team",
    "kickoff_time",
    "fpl_deadline_time",
    "bookmaker_last_update",
    "over_2_5_market_margin",
    "both_teams_score_market_margin",
    "fixture_match_method",
    "source_name",
    "source_timestamp",
    "data_snapshot_timestamp",
    "forecast_timestamp",
    "model_version",
)

PHASE5_BENCHMARK_FIELDS = (
    "fixture_id",
    "gameweek",
    "market_snapshot_timestamp",
    "market_bookmaker_count",
    "market_goal_line_bookmaker_count",
    "market_consensus_method",
    "model_home_win_probability",
    "market_home_win_probability",
    "home_win_probability_delta_model_minus_market",
    "model_draw_probability",
    "market_draw_probability",
    "draw_probability_delta_model_minus_market",
    "model_away_win_probability",
    "market_away_win_probability",
    "away_win_probability_delta_model_minus_market",
    "model_over_2_5_goals_probability",
    "market_over_2_5_probability",
    "over_2_5_probability_delta_model_minus_market",
    "model_home_clean_sheet_probability",
    "market_home_clean_sheet_probability",
    "home_clean_sheet_delta_model_minus_market",
    "model_away_clean_sheet_probability",
    "market_away_clean_sheet_probability",
    "away_clean_sheet_delta_model_minus_market",
    "market_lambda_home",
    "market_lambda_away",
    "market_fit_rmse",
    "market_fit_status",
    "market_h2h_median_margin",
    "market_goal_line_median_margin",
    "model_version",
    "forecast_timestamp",
    "source_name",
    "source_timestamp",
    "is_pre_deadline_data",
)


def _vx_phase5_frame_digest(frame):
    """Small immutable-state guard proving Phase 5 did not rewrite model outputs."""
    if not isinstance(frame, pd.DataFrame):
        return None
    payload = frame.to_json(orient="split", date_format="iso", double_precision=15)
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()


_PHASE5_MODEL_STATE_BEFORE = {
    "projected_goals_model_only": _vx_phase5_frame_digest(
        globals().get("projected_goals_model_only")
    ),
    "fixture_probabilities_model_only": _vx_phase5_frame_digest(
        globals().get("fixture_probabilities_model_only")
    ),
    "fixture_fdr_model_only": _vx_phase5_frame_digest(
        globals().get("fixture_fdr_model_only")
    ),
}


def _vx_phase5_get_secret(name):
    value = str(os.getenv(name, "") or "").strip()
    if value:
        return value
    try:
        from google.colab import userdata
    except Exception:
        return ""
    try:
        return str(userdata.get(name) or "").strip()
    except Exception as exc:
        exc_name = type(exc).__name__
        print(f"⚠️ Could not read Colab secret '{name}': {exc_name}: {exc}")
        if exc_name == "NotebookAccessError":
            print(
                f"   '{name}' exists in your Colab secrets but this notebook "
                "has not been granted access to it yet."
            )
            print(
                "   Open the key icon in the left sidebar, find "
                f"'{name}', and toggle 'Notebook access' ON for this notebook."
            )
        elif exc_name == "SecretNotFoundError":
            print(
                f"   No secret named '{name}' exists in this Colab account's "
                "secret manager. Add it via the key icon in the left sidebar."
            )
        return ""


def _vx_phase5_norm(value):
    return re.sub(r"[^a-z0-9]+", "", str(value or "").lower())


def _vx_phase5_utc(value):
    parsed = pd.to_datetime(value, utc=True, errors="coerce")
    return parsed


def _vx_phase5_decimal(value):
    try:
        number = float(value)
    except Exception:
        return np.nan
    return number if np.isfinite(number) and number > 1.0 else np.nan


def _vx_phase5_devig_threeway(home_odds, draw_odds, away_odds):
    odds = np.array(
        [_vx_phase5_decimal(home_odds), _vx_phase5_decimal(draw_odds), _vx_phase5_decimal(away_odds)],
        dtype=float,
    )
    if not np.isfinite(odds).all():
        return None
    raw = 1.0 / odds
    total = float(raw.sum())
    if total <= 0:
        return None
    return {
        "p_home": float(raw[0] / total),
        "p_draw": float(raw[1] / total),
        "p_away": float(raw[2] / total),
        "margin": float(total - 1.0),
    }


def _vx_phase5_devig_twoway(odds_a, odds_b):
    odds = np.array(
        [_vx_phase5_decimal(odds_a), _vx_phase5_decimal(odds_b)],
        dtype=float,
    )
    if not np.isfinite(odds).all():
        return None
    raw = 1.0 / odds
    total = float(raw.sum())
    if total <= 0:
        return None
    return {
        "p_a": float(raw[0] / total),
        "p_b": float(raw[1] / total),
        "margin": float(total - 1.0),
    }


def _vx_phase5_team_lookup(teams_frame):
    if not isinstance(teams_frame, pd.DataFrame) or teams_frame.empty:
        raise RuntimeError("Phase 5 requires the canonical FPL teams table.")

    exact = {}
    official_names = {}
    for _, row in teams_frame.iterrows():
        team_id = int(row["id"])
        official_names[team_id] = str(row.get("name", "") or "")
        for candidate in (row.get("name"), row.get("short_name")):
            key = _vx_phase5_norm(candidate)
            if key:
                exact[key] = team_id

    # Explicit provider/FPL naming variants only. No football-strength information.
    aliases = {
        "tottenhamhotspur": ("spurs", "tottenham"),
        "tottenham": ("spurs", "tottenhamhotspur"),
        "wolverhamptonwanderers": ("wolves", "wolverhampton"),
        "wolverhampton": ("wolves", "wolverhamptonwanderers"),
        "brightonandhovealbion": ("brighton",),
        "afcbournemouth": ("bournemouth",),
        "nottinghamforest": ("nottmforest", "nottinghamforest"),
        "nottmforest": ("nottinghamforest",),
        "westhamunited": ("westham",),
        "newcastleunited": ("newcastle",),
        "manchesterunited": ("manunited", "manutd"),
        "manchestercity": ("mancity",),
        "leedsunited": ("leeds",),
        "sunderlandafc": ("sunderland",),
        "sheffieldunited": ("sheffieldutd",),
    }

    def resolve(name):
        key = _vx_phase5_norm(name)
        if not key:
            return None, "UNMAPPED"

        if key in exact:
            return int(exact[key]), "EXACT"

        for alias_key in aliases.get(key, ()):
            if alias_key in exact:
                return int(exact[alias_key]), "EXPLICIT_ALIAS"

        # Conservative fallback: accept only a very high-confidence unique match.
        scores = []
        for official_key, team_id in exact.items():
            score = SequenceMatcher(None, key, official_key).ratio()
            scores.append((float(score), int(team_id)))
        scores.sort(reverse=True)
        if scores:
            best_score, best_id = scores[0]
            second_score = scores[1][0] if len(scores) > 1 else 0.0
            if best_score >= 0.92 and (best_score - second_score) >= 0.05:
                return best_id, f"FUZZY_{best_score:.3f}"

        return None, "UNMAPPED"

    return resolve, official_names


def _vx_phase5_fixture_context(fixtures_frame, events_frame):
    if not isinstance(fixtures_frame, pd.DataFrame) or fixtures_frame.empty:
        raise RuntimeError("Phase 5 requires fixtures_raw.")
    if not isinstance(events_frame, pd.DataFrame) or events_frame.empty:
        raise RuntimeError("Phase 5 requires FPL events/deadlines.")

    deadline_map = {}
    if {"id", "deadline_time"}.issubset(events_frame.columns):
        for _, row in events_frame.iterrows():
            gw = pd.to_numeric(row.get("id"), errors="coerce")
            deadline = _vx_phase5_utc(row.get("deadline_time"))
            if pd.notna(gw) and pd.notna(deadline):
                deadline_map[int(gw)] = deadline

    lookup = {}
    for _, row in fixtures_frame.iterrows():
        home_id = pd.to_numeric(row.get("team_h"), errors="coerce")
        away_id = pd.to_numeric(row.get("team_a"), errors="coerce")
        fid = pd.to_numeric(row.get("fixture_id", row.get("id")), errors="coerce")
        gw = pd.to_numeric(row.get("gameweek", row.get("event")), errors="coerce")
        if pd.isna(home_id) or pd.isna(away_id) or pd.isna(fid):
            continue
        gw_int = int(gw) if pd.notna(gw) else None
        lookup[(int(home_id), int(away_id))] = {
            "fixture_id": int(fid),
            "gameweek": gw_int,
            "kickoff_time": row.get("kickoff_time"),
            "deadline_time": deadline_map.get(gw_int),
        }
    return lookup


def _vx_phase5_request_odds(api_key):
    """
    Current EPL benchmark snapshot from The Odds API.
    Raw response is preserved separately. No prices are blended into V1.
    """
    url = "https://api.the-odds-api.com/v4/sports/soccer_epl/odds/"
    attempts = (
        {
            "regions": "uk,eu",
            "markets": "h2h,totals",
            "oddsFormat": "decimal",
            "dateFormat": "iso",
        },
        {
            "regions": "uk",
            "markets": "h2h,totals",
            "oddsFormat": "decimal",
            "dateFormat": "iso",
        },
    )
    last_error = None
    for params in attempts:
        response = requests.get(
            url,
            params={"apiKey": api_key, **params},
            timeout=30,
        )
        if response.ok:
            return response.json(), dict(response.headers), params
        try:
            body = response.json()
            detail = body.get("message") or body.get("error") or body.get("error_code")
        except Exception:
            detail = (response.text or "")[:240]
        last_error = RuntimeError(
            f"The Odds API HTTP {response.status_code}"
            + (f" — {detail}" if detail else "")
        )
    raise last_error or RuntimeError("The Odds API request failed.")


def _vx_phase5_extract_price(outcomes, predicate):
    for outcome in outcomes or []:
        if predicate(outcome):
            return _vx_phase5_decimal(outcome.get("price"))
    return np.nan


def _vx_phase5_normalize_payload(payload, capture_timestamp):
    resolve_team, official_names = _vx_phase5_team_lookup(teams)
    fixture_lookup = _vx_phase5_fixture_context(fixtures_raw, events)

    rows = []
    mapping_audit = []
    for game in payload or []:
        market_home_name = str(game.get("home_team", "") or "")
        market_away_name = str(game.get("away_team", "") or "")
        home_id, home_method = resolve_team(market_home_name)
        away_id, away_method = resolve_team(market_away_name)

        if home_id is None or away_id is None:
            mapping_audit.append({
                "home_team": market_home_name,
                "away_team": market_away_name,
                "status": "UNMAPPED_TEAM",
                "home_method": home_method,
                "away_method": away_method,
            })
            continue

        fx = fixture_lookup.get((int(home_id), int(away_id)))
        if not fx:
            mapping_audit.append({
                "home_team": market_home_name,
                "away_team": market_away_name,
                "status": "NO_FPL_FIXTURE_MATCH",
                "home_team_id": int(home_id),
                "away_team_id": int(away_id),
            })
            continue

        deadline = fx.get("deadline_time")
        capture_dt = _vx_phase5_utc(capture_timestamp)
        is_pre_deadline = bool(
            pd.notna(capture_dt)
            and deadline is not None
            and pd.notna(deadline)
            and capture_dt <= deadline
        )

        for bookmaker in game.get("bookmakers", []) or []:
            bookmaker_name = str(
                bookmaker.get("title")
                or bookmaker.get("key")
                or "UNKNOWN_BOOKMAKER"
            )
            bookmaker_last_update = bookmaker.get("last_update")
            markets = bookmaker.get("markets", []) or []

            h2h_outcomes = []
            totals_outcomes = []
            btts_outcomes = []

            for market in markets:
                key = str(market.get("key", "") or "").lower()
                outcomes = market.get("outcomes", []) or []
                if key == "h2h":
                    h2h_outcomes.extend(outcomes)
                elif key == "totals":
                    totals_outcomes.extend(outcomes)
                elif key in {"btts", "both_teams_to_score"}:
                    btts_outcomes.extend(outcomes)

            game_home_norm = _vx_phase5_norm(market_home_name)
            game_away_norm = _vx_phase5_norm(market_away_name)

            home_win_odds = _vx_phase5_extract_price(
                h2h_outcomes,
                lambda x: _vx_phase5_norm(x.get("name")) == game_home_norm,
            )
            away_win_odds = _vx_phase5_extract_price(
                h2h_outcomes,
                lambda x: _vx_phase5_norm(x.get("name")) == game_away_norm,
            )
            draw_odds = _vx_phase5_extract_price(
                h2h_outcomes,
                lambda x: str(x.get("name", "") or "").strip().lower() == "draw",
            )

            def total_price(side, point):
                return _vx_phase5_extract_price(
                    totals_outcomes,
                    lambda x: (
                        str(x.get("name", "") or "").strip().lower() == side
                        and math.isclose(
                            float(x.get("point")),
                            float(point),
                            rel_tol=0.0,
                            abs_tol=1e-9,
                        )
                    ) if x.get("point") is not None else False,
                )

            over_1_5_odds = total_price("over", 1.5)
            over_2_5_odds = total_price("over", 2.5)
            under_2_5_odds = total_price("under", 2.5)

            btts_yes_odds = _vx_phase5_extract_price(
                btts_outcomes,
                lambda x: str(x.get("name", "") or "").strip().lower() in {"yes", "over"},
            )
            btts_no_odds = _vx_phase5_extract_price(
                btts_outcomes,
                lambda x: str(x.get("name", "") or "").strip().lower() in {"no", "under"},
            )

            h2h_fair = _vx_phase5_devig_threeway(
                home_win_odds, draw_odds, away_win_odds
            )
            totals_fair = _vx_phase5_devig_twoway(
                over_2_5_odds, under_2_5_odds
            )
            btts_fair = _vx_phase5_devig_twoway(
                btts_yes_odds, btts_no_odds
            )

            rows.append({
                "fixture_id": int(fx["fixture_id"]),
                "odds_snapshot_timestamp": capture_timestamp,
                "bookmaker_or_exchange": bookmaker_name,
                "home_win_odds": home_win_odds,
                "draw_odds": draw_odds,
                "away_win_odds": away_win_odds,
                "over_1_5_odds": over_1_5_odds,
                "over_2_5_odds": over_2_5_odds,
                "under_2_5_odds": under_2_5_odds,
                "both_teams_score_yes_odds": btts_yes_odds,
                "both_teams_score_no_odds": btts_no_odds,
                "market_margin": h2h_fair["margin"] if h2h_fair else np.nan,
                "is_before_fpl_deadline": is_pre_deadline,
                "gameweek": fx.get("gameweek"),
                "home_team_id": int(home_id),
                "away_team_id": int(away_id),
                "home_team": official_names.get(int(home_id), market_home_name),
                "away_team": official_names.get(int(away_id), market_away_name),
                "kickoff_time": fx.get("kickoff_time"),
                "fpl_deadline_time": (
                    deadline.isoformat() if deadline is not None and pd.notna(deadline) else None
                ),
                "bookmaker_last_update": bookmaker_last_update,
                "over_2_5_market_margin": totals_fair["margin"] if totals_fair else np.nan,
                "both_teams_score_market_margin": btts_fair["margin"] if btts_fair else np.nan,
                "fixture_match_method": f"{home_method}|{away_method}",
                "source_name": "THE_ODDS_API:SOCCER_EPL",
                "source_timestamp": bookmaker_last_update or capture_timestamp,
                "data_snapshot_timestamp": capture_timestamp,
                "forecast_timestamp": None,
                "model_version": PHASE5_MODEL_VERSION,
            })

    return (
        pd.DataFrame(rows, columns=PHASE5_RAW_ODDS_FIELDS),
        pd.DataFrame(mapping_audit),
    )


def _vx_phase5_persist_snapshot(raw_frame, provider_payload, capture_timestamp, request_params):
    """
    Persistent path is OUTSIDE FIRST VIDEO, so Cell 0 cleanup cannot erase
    historical market snapshots. This is the deadline-safe archive for Phase 7.
    """
    persistent_dir = Path(PROJECT_ROOT) / "RAW_DATA" / "MARKET_ODDS"
    persistent_dir.mkdir(parents=True, exist_ok=True)

    stamp_dt = _vx_phase5_utc(capture_timestamp)
    stamp = (
        stamp_dt.strftime("%Y%m%dT%H%M%SZ")
        if pd.notna(stamp_dt)
        else datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%SZ")
    )

    provider_path = persistent_dir / f"provider_snapshot_{stamp}.json"
    normalized_path = persistent_dir / f"market_odds_snapshot_{stamp}.json"
    history_path = persistent_dir / "market_odds_history.jsonl"

    provider_path.write_text(
        json.dumps(
            {
                "capture_timestamp": capture_timestamp,
                "source": "THE_ODDS_API:SOCCER_EPL",
                "request": {k: v for k, v in request_params.items() if k != "apiKey"},
                "payload": provider_payload,
            },
            indent=2,
            ensure_ascii=False,
        ),
        encoding="utf-8",
    )

    normalized_path.write_text(
        raw_frame.to_json(orient="records", date_format="iso", indent=2),
        encoding="utf-8",
    )

    records = json.loads(raw_frame.to_json(orient="records", date_format="iso"))
    with history_path.open("a", encoding="utf-8") as handle:
        for record in records:
            handle.write(json.dumps(record, ensure_ascii=False) + "\n")

    return persistent_dir, provider_path, normalized_path, history_path


def _vx_phase5_load_history(history_path):
    if not Path(history_path).exists():
        return pd.DataFrame(columns=PHASE5_RAW_ODDS_FIELDS)

    rows = []
    with Path(history_path).open("r", encoding="utf-8") as handle:
        for line_number, line in enumerate(handle, start=1):
            line = line.strip()
            if not line:
                continue
            try:
                row = json.loads(line)
            except Exception as exc:
                raise RuntimeError(
                    f"Corrupt market-odds history at line {line_number}: {exc}"
                ) from exc
            rows.append(row)

    frame = pd.DataFrame(rows)
    for field in PHASE5_RAW_ODDS_FIELDS:
        if field not in frame.columns:
            frame[field] = np.nan
    return frame.loc[:, PHASE5_RAW_ODDS_FIELDS].copy()


def _vx_phase5_market_metrics(lambda_home, lambda_away):
    lh = float(lambda_home)
    la = float(lambda_away)
    if lh <= 0 or la <= 0 or not np.isfinite([lh, la]).all():
        raise ValueError("Market lambdas must be finite and positive.")

    if "_vx_phase4_wdl_probabilities" not in globals():
        raise RuntimeError("Phase 5 requires the validated Phase-4 Poisson helper.")

    p_home, p_draw, p_away = _vx_phase4_wdl_probabilities(lh, la)
    total_lambda = lh + la
    p_over_2_5 = 1.0 - math.exp(-total_lambda) * (
        1.0 + total_lambda + (total_lambda ** 2) / 2.0
    )
    return {
        "p_home": float(p_home),
        "p_draw": float(p_draw),
        "p_away": float(p_away),
        "p_over_2_5": float(p_over_2_5),
        "home_cs": float(math.exp(-la)),
        "away_cs": float(math.exp(-lh)),
    }


def _vx_phase5_fit_market_lambdas(p_home, p_draw, p_away, p_over_2_5):
    """
    MARKET-ONLY 1X2 + goal-line fit.
    No model lambda enters the objective or starting values.
    """
    target = np.array(
        [p_home, p_draw, p_away, p_over_2_5],
        dtype=float,
    )
    if not np.isfinite(target).all():
        return {
            "lambda_home": np.nan,
            "lambda_away": np.nan,
            "rmse": np.nan,
            "status": "MISSING_1X2_OR_OU25",
        }

    if not (0.0 < target[0] < 1.0 and 0.0 < target[1] < 1.0
            and 0.0 < target[2] < 1.0 and 0.0 < target[3] < 1.0):
        return {
            "lambda_home": np.nan,
            "lambda_away": np.nan,
            "rmse": np.nan,
            "status": "INVALID_TARGET_PROBABILITY",
        }

    try:
        from scipy.optimize import minimize
    except Exception:
        return {
            "lambda_home": np.nan,
            "lambda_away": np.nan,
            "rmse": np.nan,
            "status": "SCIPY_UNAVAILABLE",
        }

    def objective(x):
        lh, la = float(x[0]), float(x[1])
        metrics = _vx_phase5_market_metrics(lh, la)
        wdl_sq = (
            (metrics["p_home"] - target[0]) ** 2
            + (metrics["p_draw"] - target[1]) ** 2
            + (metrics["p_away"] - target[2]) ** 2
        ) / 3.0
        goal_line_sq = (metrics["p_over_2_5"] - target[3]) ** 2
        # Equal block weight: 1X2 block + O/U 2.5 block.
        return float(wdl_sq + goal_line_sq)

    starts = (
        (1.55, 1.25),
        (1.20, 1.20),
        (2.00, 1.00),
        (1.00, 2.00),
        (2.00, 2.00),
    )

    best = None
    for start in starts:
        result = minimize(
            objective,
            x0=np.array(start, dtype=float),
            method="L-BFGS-B",
            bounds=((0.05, 5.50), (0.05, 5.50)),
            options={"maxiter": 500, "ftol": 1e-14},
        )
        if np.isfinite(result.fun) and (best is None or result.fun < best.fun):
            best = result

    if best is None:
        return {
            "lambda_home": np.nan,
            "lambda_away": np.nan,
            "rmse": np.nan,
            "status": "FIT_FAILED",
        }

    lh, la = map(float, best.x)
    fitted = _vx_phase5_market_metrics(lh, la)
    residuals = np.array([
        fitted["p_home"] - target[0],
        fitted["p_draw"] - target[1],
        fitted["p_away"] - target[2],
        fitted["p_over_2_5"] - target[3],
    ])
    rmse = float(np.sqrt(np.mean(residuals ** 2)))

    return {
        "lambda_home": lh,
        "lambda_away": la,
        "rmse": rmse,
        "status": "OK" if bool(best.success) else "BEST_FINITE_NONCONVERGED",
    }


def _vx_phase5_build_consensus(history_frame, model_frame):
    if history_frame.empty or model_frame.empty:
        return pd.DataFrame(columns=PHASE5_BENCHMARK_FIELDS)

    h = history_frame.copy()
    h["fixture_id"] = pd.to_numeric(h["fixture_id"], errors="coerce")
    h["is_before_fpl_deadline"] = h["is_before_fpl_deadline"].fillna(False).astype(bool)
    h["_snapshot_dt"] = pd.to_datetime(
        h["odds_snapshot_timestamp"], utc=True, errors="coerce"
    )
    h = h.loc[
        h["fixture_id"].notna()
        & h["is_before_fpl_deadline"]
        & h["_snapshot_dt"].notna()
    ].copy()

    if h.empty:
        return pd.DataFrame(columns=PHASE5_BENCHMARK_FIELDS)

    h["fixture_id"] = h["fixture_id"].astype(int)

    # For each bookmaker, use the latest snapshot that was actually captured
    # before the FPL deadline. No post-deadline prices enter the benchmark.
    h = h.sort_values("_snapshot_dt")
    latest = h.drop_duplicates(
        subset=["fixture_id", "bookmaker_or_exchange"],
        keep="last",
    )

    consensus_rows = []
    for fixture_id_value, group in latest.groupby("fixture_id"):
        fair_h2h = []
        fair_ou25 = []

        for _, row in group.iterrows():
            h2h = _vx_phase5_devig_threeway(
                row.get("home_win_odds"),
                row.get("draw_odds"),
                row.get("away_win_odds"),
            )
            if h2h:
                fair_h2h.append(h2h)

            ou25 = _vx_phase5_devig_twoway(
                row.get("over_2_5_odds"),
                row.get("under_2_5_odds"),
            )
            if ou25:
                fair_ou25.append(ou25)

        if not fair_h2h:
            continue

        p_home = float(np.median([x["p_home"] for x in fair_h2h]))
        p_draw = float(np.median([x["p_draw"] for x in fair_h2h]))
        p_away = float(np.median([x["p_away"] for x in fair_h2h]))
        p_sum = p_home + p_draw + p_away
        if p_sum <= 0:
            continue
        p_home, p_draw, p_away = (
            p_home / p_sum,
            p_draw / p_sum,
            p_away / p_sum,
        )

        p_over_2_5 = (
            float(np.median([x["p_a"] for x in fair_ou25]))
            if fair_ou25
            else np.nan
        )

        fit = _vx_phase5_fit_market_lambdas(
            p_home, p_draw, p_away, p_over_2_5
        )

        market_home_cs = np.nan
        market_away_cs = np.nan
        if np.isfinite(fit["lambda_home"]) and np.isfinite(fit["lambda_away"]):
            market_home_cs = float(math.exp(-fit["lambda_away"]))
            market_away_cs = float(math.exp(-fit["lambda_home"]))

        consensus_rows.append({
            "fixture_id": int(fixture_id_value),
            "market_snapshot_timestamp": group["_snapshot_dt"].max().isoformat(),
            "market_bookmaker_count": int(len(fair_h2h)),
            "market_goal_line_bookmaker_count": int(len(fair_ou25)),
            "market_consensus_method": "MEDIAN_OF_DEVIGGED_BOOKMAKER_PROBABILITIES",
            "market_home_win_probability": p_home,
            "market_draw_probability": p_draw,
            "market_away_win_probability": p_away,
            "market_over_2_5_probability": p_over_2_5,
            "market_home_clean_sheet_probability": market_home_cs,
            "market_away_clean_sheet_probability": market_away_cs,
            "market_lambda_home": fit["lambda_home"],
            "market_lambda_away": fit["lambda_away"],
            "market_fit_rmse": fit["rmse"],
            "market_fit_status": fit["status"],
            "market_h2h_median_margin": float(
                np.median([x["margin"] for x in fair_h2h])
            ),
            "market_goal_line_median_margin": (
                float(np.median([x["margin"] for x in fair_ou25]))
                if fair_ou25 else np.nan
            ),
        })

    consensus = pd.DataFrame(consensus_rows)
    if consensus.empty:
        return pd.DataFrame(columns=PHASE5_BENCHMARK_FIELDS)

    model = model_frame.copy()
    required_model = {
        "fixture_id", "gameweek",
        "home_win_probability", "draw_probability", "away_win_probability",
        "over_2_5_goals_probability",
        "home_clean_sheet_probability", "away_clean_sheet_probability",
        "forecast_timestamp",
    }
    missing = sorted(required_model - set(model.columns))
    if missing:
        raise RuntimeError(f"Phase 5 model benchmark missing Phase-4 fields: {missing}")

    model["fixture_id"] = pd.to_numeric(model["fixture_id"], errors="raise").astype(int)
    merged = model.merge(consensus, on="fixture_id", how="inner", validate="one_to_one")

    rows = []
    for _, row in merged.iterrows():
        model_home = float(row["home_win_probability"])
        model_draw = float(row["draw_probability"])
        model_away = float(row["away_win_probability"])
        model_o25 = float(row["over_2_5_goals_probability"])
        model_home_cs = float(row["home_clean_sheet_probability"])
        model_away_cs = float(row["away_clean_sheet_probability"])

        market_home = float(row["market_home_win_probability"])
        market_draw = float(row["market_draw_probability"])
        market_away = float(row["market_away_win_probability"])
        market_o25 = pd.to_numeric(row["market_over_2_5_probability"], errors="coerce")
        market_home_cs = pd.to_numeric(
            row["market_home_clean_sheet_probability"], errors="coerce"
        )
        market_away_cs = pd.to_numeric(
            row["market_away_clean_sheet_probability"], errors="coerce"
        )

        rows.append({
            "fixture_id": int(row["fixture_id"]),
            "gameweek": int(row["gameweek"]),
            "market_snapshot_timestamp": row["market_snapshot_timestamp"],
            "market_bookmaker_count": int(row["market_bookmaker_count"]),
            "market_goal_line_bookmaker_count": int(row["market_goal_line_bookmaker_count"]),
            "market_consensus_method": row["market_consensus_method"],
            "model_home_win_probability": model_home,
            "market_home_win_probability": market_home,
            "home_win_probability_delta_model_minus_market": model_home - market_home,
            "model_draw_probability": model_draw,
            "market_draw_probability": market_draw,
            "draw_probability_delta_model_minus_market": model_draw - market_draw,
            "model_away_win_probability": model_away,
            "market_away_win_probability": market_away,
            "away_win_probability_delta_model_minus_market": model_away - market_away,
            "model_over_2_5_goals_probability": model_o25,
            "market_over_2_5_probability": market_o25,
            "over_2_5_probability_delta_model_minus_market": (
                model_o25 - float(market_o25) if pd.notna(market_o25) else np.nan
            ),
            "model_home_clean_sheet_probability": model_home_cs,
            "market_home_clean_sheet_probability": market_home_cs,
            "home_clean_sheet_delta_model_minus_market": (
                model_home_cs - float(market_home_cs)
                if pd.notna(market_home_cs) else np.nan
            ),
            "model_away_clean_sheet_probability": model_away_cs,
            "market_away_clean_sheet_probability": market_away_cs,
            "away_clean_sheet_delta_model_minus_market": (
                model_away_cs - float(market_away_cs)
                if pd.notna(market_away_cs) else np.nan
            ),
            "market_lambda_home": row["market_lambda_home"],
            "market_lambda_away": row["market_lambda_away"],
            "market_fit_rmse": row["market_fit_rmse"],
            "market_fit_status": row["market_fit_status"],
            "market_h2h_median_margin": row["market_h2h_median_margin"],
            "market_goal_line_median_margin": row["market_goal_line_median_margin"],
            "model_version": PHASE5_MODEL_VERSION,
            "forecast_timestamp": row["forecast_timestamp"],
            "source_name": "MARKET_BENCHMARK_ONLY_NO_BLEND",
            "source_timestamp": row["market_snapshot_timestamp"],
            "is_pre_deadline_data": True,
        })

    return pd.DataFrame(rows, columns=PHASE5_BENCHMARK_FIELDS)


# ------------------------------------------------------------------
# CAPTURE CURRENT RAW SNAPSHOT
# ------------------------------------------------------------------
PHASE5_PERSISTENT_DIR = Path(PROJECT_ROOT) / "RAW_DATA" / "MARKET_ODDS"
PHASE5_PERSISTENT_DIR.mkdir(parents=True, exist_ok=True)
PHASE5_HISTORY_PATH = PHASE5_PERSISTENT_DIR / "market_odds_history.jsonl"

market_odds_raw = pd.DataFrame(columns=PHASE5_RAW_ODDS_FIELDS)
market_odds_mapping_audit = pd.DataFrame()
market_odds_benchmark = pd.DataFrame(columns=PHASE5_BENCHMARK_FIELDS)

_capture_timestamp = datetime.now(timezone.utc).isoformat()
_enable_market = bool(globals().get("ENABLE_MARKET_ODDS", True))
_api_key = _vx_phase5_get_secret("ODDS_API_KEY") if _enable_market else ""

if _enable_market and _api_key:
    try:
        _provider_payload, _provider_headers, _request_params = _vx_phase5_request_odds(
            _api_key
        )
        market_odds_raw, market_odds_mapping_audit = _vx_phase5_normalize_payload(
            _provider_payload,
            _capture_timestamp,
        )

        (
            PHASE5_PERSISTENT_DIR,
            PHASE5_PROVIDER_SNAPSHOT_PATH,
            PHASE5_NORMALIZED_SNAPSHOT_PATH,
            PHASE5_HISTORY_PATH,
        ) = _vx_phase5_persist_snapshot(
            market_odds_raw,
            _provider_payload,
            _capture_timestamp,
            _request_params,
        )

        PHASE5_CAPTURE_STATUS = (
            "LIVE_SNAPSHOT_CAPTURED"
            if not market_odds_raw.empty
            else "LIVE_PROVIDER_RESPONSE_NO_MAPPED_ROWS"
        )
        _requests_remaining = (
            _provider_headers.get("x-requests-remaining")
            or _provider_headers.get("X-Requests-Remaining")
        )
    except Exception as exc:
        PHASE5_CAPTURE_STATUS = f"CAPTURE_ERROR:{type(exc).__name__}"
        print(f"⚠️ Phase 5 market snapshot unavailable: {type(exc).__name__}: {exc}")
        print("   Model-only Phase 4 outputs remain untouched.")
else:
    PHASE5_CAPTURE_STATUS = (
        "DISABLED"
        if not _enable_market
        else "NO_ODDS_API_KEY"
    )
    print("ℹ️ Phase 5 live odds capture not run.")
    print("   ENABLE_MARKET_ODDS is OFF or ODDS_API_KEY is unavailable.")

# Keep the Phase-0 canonical raw-table registry pointing at the current schema.
if "RAW_TABLES" in globals() and isinstance(RAW_TABLES, dict):
    RAW_TABLES["market_odds_raw"] = market_odds_raw
if "RAW_TABLE_STATUS" in globals() and isinstance(RAW_TABLE_STATUS, dict):
    RAW_TABLE_STATUS["market_odds_raw"] = PHASE5_CAPTURE_STATUS
if "vx_phase0_assert_no_forbidden_columns" in globals():
    vx_phase0_assert_no_forbidden_columns(market_odds_raw, "market_odds_raw Phase 5")

# ------------------------------------------------------------------
# BUILD DEADLINE-SAFE MARKET BENCHMARK FROM PERSISTENT HISTORY
# ------------------------------------------------------------------
market_odds_history_raw = _vx_phase5_load_history(PHASE5_HISTORY_PATH)

if (
    bool(globals().get("PHASE4_READY", False))
    and isinstance(globals().get("fixture_probabilities_model_only"), pd.DataFrame)
    and not fixture_probabilities_model_only.empty
):
    market_odds_benchmark = _vx_phase5_build_consensus(
        market_odds_history_raw,
        fixture_probabilities_model_only,
    )

    if market_odds_benchmark.empty:
        PHASE5_BENCHMARK_STATUS = "NO_DEADLINE_SAFE_MATCHED_MARKET_ROWS"
    else:
        PHASE5_BENCHMARK_STATUS = "READY"
        PHASE5_READY = True
else:
    PHASE5_BENCHMARK_STATUS = "BLOCKED_PHASE4_NOT_READY"

# Current-run copies live in the disposable output tree; the historical archive
# above remains persistent and is never erased by FIRST VIDEO cleanup.
PHASE5_OUTPUT_DIR = Path(DATA_DIR) / "PHASE5_MARKET_BENCHMARK"
PHASE5_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

PHASE5_CURRENT_RAW_PATH = PHASE5_OUTPUT_DIR / "phase5_market_odds_raw_current.json"
PHASE5_HISTORY_EXPORT_PATH = PHASE5_OUTPUT_DIR / "phase5_market_odds_history_deadline_safe.json"
PHASE5_BENCHMARK_PATH = PHASE5_OUTPUT_DIR / "phase5_model_vs_market_benchmark.json"
PHASE5_MAPPING_AUDIT_PATH = PHASE5_OUTPUT_DIR / "phase5_market_fixture_mapping_audit.json"

PHASE5_CURRENT_RAW_PATH.write_text(
    market_odds_raw.to_json(orient="records", date_format="iso", indent=2),
    encoding="utf-8",
)

_deadline_safe_history_export = market_odds_history_raw.loc[
    market_odds_history_raw.get(
        "is_before_fpl_deadline",
        pd.Series(False, index=market_odds_history_raw.index),
    ).fillna(False).astype(bool)
].copy()

PHASE5_HISTORY_EXPORT_PATH.write_text(
    _deadline_safe_history_export.to_json(
        orient="records", date_format="iso", indent=2
    ),
    encoding="utf-8",
)
PHASE5_BENCHMARK_PATH.write_text(
    market_odds_benchmark.to_json(
        orient="records", date_format="iso", indent=2
    ),
    encoding="utf-8",
)
PHASE5_MAPPING_AUDIT_PATH.write_text(
    market_odds_mapping_audit.to_json(
        orient="records", date_format="iso", indent=2
    ),
    encoding="utf-8",
)

# ------------------------------------------------------------------
# HARD NO-BLEND / NO-MUTATION ASSERTIONS
# ------------------------------------------------------------------
_PHASE5_MODEL_STATE_AFTER = {
    "projected_goals_model_only": _vx_phase5_frame_digest(
        globals().get("projected_goals_model_only")
    ),
    "fixture_probabilities_model_only": _vx_phase5_frame_digest(
        globals().get("fixture_probabilities_model_only")
    ),
    "fixture_fdr_model_only": _vx_phase5_frame_digest(
        globals().get("fixture_fdr_model_only")
    ),
}

if _PHASE5_MODEL_STATE_AFTER != _PHASE5_MODEL_STATE_BEFORE:
    raise RuntimeError(
        "PHASE 5 FAIL — market benchmark mutated model-only Phase 3/4 outputs."
    )

_forbidden_blend_columns = {
    "final_lambda_home",
    "final_lambda_away",
    "blended_lambda_home",
    "blended_lambda_away",
    "market_blended_fdr",
}
for _frame_name, _frame in {
    "market_odds_raw": market_odds_raw,
    "market_odds_benchmark": market_odds_benchmark,
}.items():
    _hits = sorted(_forbidden_blend_columns.intersection(set(_frame.columns)))
    if _hits:
        raise RuntimeError(
            f"PHASE 5 FAIL — {_frame_name} contains forbidden blend fields: {_hits}"
        )

print("✅ PHASE 5 MARKET BENCHMARK LAYER BUILT")
print(f"   Live capture status: {PHASE5_CAPTURE_STATUS}")
print(f"   Benchmark status:    {PHASE5_BENCHMARK_STATUS}")
print(f"   Current raw rows:    {len(market_odds_raw)}")
print(f"   Archived raw rows:   {len(market_odds_history_raw)}")
print(f"   Benchmark fixtures:  {len(market_odds_benchmark)}")
if "_requests_remaining" in globals() and _requests_remaining is not None:
    print(f"   Odds API requests remaining: {_requests_remaining}")
print("🔒 Decimal odds are de-vigged before comparison.")
print("🔒 Only snapshots captured before the FPL deadline are benchmark-eligible.")
print("🔒 Market lambdas are fit from market 1X2 + O/U 2.5 only.")
print("🔒 NO market blend, NO market-adjusted FDR, NO player xPts.")
print("📦 Persistent raw archive:", PHASE5_PERSISTENT_DIR)
print("📊 Current benchmark:", PHASE5_BENCHMARK_PATH)
```

```python
# ============================================================
# CELL 13 — PHASE 6
# DIXON–COLES LOW-SCORE CORRECTION — VALIDATION-GATED
# CONSUMES PHASE-3 LAMBDAS / PHASE-4 POISSON ONLY
# NO TEAM-STRENGTH CHANGES • NO MARKET BLEND • NO PLAYER EXPECTED-POINTS
# ============================================================
# Required order:
#   1) baseline Poisson exists;
#   2) frozen pre-deadline historical forecasts are validated;
#   3) fit one global Dixon–Coles rho on an earlier training window;
#   4) compare Poisson vs Dixon–Coles on a later holdout window;
#   5) activate Dixon–Coles only if the holdout metrics improve.
#
# Dixon–Coles changes only the joint low-score probabilities 0-0, 0-1,
# 1-0 and 1-1. It never changes team strengths or model lambdas.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import math
import numpy as np
import pandas as pd

if "PHASE4_POISSON_READY" not in globals() or not bool(PHASE4_POISSON_READY):
    _blocked = [name for name, flag in (("Phase 1A", "PHASE1_READY"), ("Phase 2", "PHASE2_READY"), ("Phase 3", "PHASE3_READY")) if not bool(globals().get(flag, False))]
    raise RuntimeError("Phase 4A did not build; root upstream block: " + (", ".join(_blocked) if _blocked else "Phase 4A itself"))
if "fixture_probabilities_model_only" not in globals():
    raise RuntimeError("Phase 6 requires fixture_probabilities_model_only from Phase 4A.")
if "poisson_probability_matrix" not in globals():
    raise RuntimeError("Phase 6 requires poisson_probability_matrix from Phase 4A.")
if fixture_probabilities_model_only.empty or poisson_probability_matrix.empty:
    raise RuntimeError("Phase 6 received empty Phase-4 Poisson outputs.")

PHASE6_MODEL_VERSION = "VORTEX-V1-PHASE6-DIXON-COLES"
PHASE6_BASELINE_MODEL_VERSION = str(
    globals().get("PHASE4_POISSON_MODEL_VERSION", "VORTEX-V1-PHASE4-BASELINE-POISSON")
)
PHASE6_TIMESTAMP = datetime.now(timezone.utc).isoformat()
PHASE6_MIN_TRAIN_FIXTURES = 60
PHASE6_MIN_VALIDATION_FIXTURES = 20
PHASE6_TRAIN_FRACTION = 0.75
PHASE6_METRIC_TOLERANCE = 1e-4
PHASE6_RHO_GRID = np.round(np.arange(-0.20, 0.2001, 0.005), 3)

PHASE6_READY = False
PHASE6_DIXON_COLES_ACTIVE = False
PHASE6_VALIDATION_STATUS = "NOT_RUN"
PHASE6_SELECTION_REASON = "NOT_EVALUATED"
rho_parameter = 0.0

PHASE6_MATRIX_FIELDS = (
    "fixture_id",
    "gameweek",
    "home_team_id",
    "away_team_id",
    "home_goals",
    "away_goals",
    "scoreline",
    "probability",
    "poisson_probability",
    "tau_correction",
    "rho_parameter",
    "model_lambda_home",
    "model_lambda_away",
    "model_version",
    "forecast_timestamp",
    "validation_approved",
)

PHASE6_FIXTURE_FIELDS = (
    "fixture_id",
    "gameweek",
    "home_team_id",
    "away_team_id",
    "model_lambda_home",
    "model_lambda_away",
    "rho_parameter",
    "home_win_probability",
    "draw_probability",
    "away_win_probability",
    "both_teams_score_probability",
    "over_2_5_goals_probability",
    "home_clean_sheet_probability",
    "away_clean_sheet_probability",
    "dixon_coles_matrix_covered_mass_0_to_8",
    "validation_approved",
    "model_version",
    "forecast_timestamp",
)

PHASE6_VALIDATION_FIELDS = (
    "candidate_rho",
    "train_fixture_count",
    "validation_fixture_count",
    "poisson_scoreline_log_loss",
    "dixon_coles_scoreline_log_loss",
    "scoreline_log_loss_improvement",
    "poisson_wdl_log_loss",
    "dixon_coles_wdl_log_loss",
    "wdl_log_loss_improvement",
    "poisson_multiclass_brier",
    "dixon_coles_multiclass_brier",
    "multiclass_brier_improvement",
    "validation_approved",
    "selection_reason",
    "model_version",
    "calculated_at",
)

dixon_coles_probability_matrix = pd.DataFrame(columns=PHASE6_MATRIX_FIELDS)
dixon_coles_fixture_probabilities = pd.DataFrame(columns=PHASE6_FIXTURE_FIELDS)
phase6_validation_report = pd.DataFrame(columns=PHASE6_VALIDATION_FIELDS)


def _vx_phase6_frame_digest(frame):
    """Stable in-run digest used to prove Phase 6 did not mutate upstream outputs."""
    if not isinstance(frame, pd.DataFrame):
        return None
    h = hashlib.sha256()
    h.update("|".join(map(str, frame.columns)).encode("utf-8"))
    h.update(pd.util.hash_pandas_object(frame, index=True).values.tobytes())
    return h.hexdigest()


_phase6_upstream_before = {
    "projected_goals_model_only": _vx_phase6_frame_digest(globals().get("projected_goals_model_only")),
    "fixture_probabilities_model_only": _vx_phase6_frame_digest(fixture_probabilities_model_only),
    "poisson_probability_matrix": _vx_phase6_frame_digest(poisson_probability_matrix),
    "fixture_fdr_model_only": _vx_phase6_frame_digest(globals().get("fixture_fdr_model_only")),
}


def vx_phase6_tau(home_goals, away_goals, lambda_home, lambda_away, rho):
    """Classic Dixon–Coles tau correction for the four low-score cells only."""
    hg = int(home_goals)
    ag = int(away_goals)
    lh = float(lambda_home)
    la = float(lambda_away)
    r = float(rho)

    if not all(np.isfinite(x) for x in (lh, la, r)) or lh <= 0 or la <= 0:
        raise RuntimeError("Dixon–Coles requires finite positive lambdas and finite rho.")

    if hg == 0 and ag == 0:
        tau = 1.0 - (lh * la * r)
    elif hg == 0 and ag == 1:
        tau = 1.0 + (lh * r)
    elif hg == 1 and ag == 0:
        tau = 1.0 + (la * r)
    elif hg == 1 and ag == 1:
        tau = 1.0 - r
    else:
        tau = 1.0

    if not np.isfinite(tau) or tau <= 0.0:
        raise RuntimeError(
            f"Invalid Dixon–Coles tau={tau} for score {hg}-{ag}, "
            f"lambdas=({lh:.6f},{la:.6f}), rho={r:.6f}"
        )
    return float(tau)


def _vx_phase6_rho_valid_for_lambdas(rho, lambda_home, lambda_away):
    """All four corrected low-score cells must remain strictly positive."""
    try:
        for hg, ag in ((0, 0), (0, 1), (1, 0), (1, 1)):
            vx_phase6_tau(hg, ag, lambda_home, lambda_away, rho)
        return True
    except Exception:
        return False


def _vx_phase6_rho_valid_for_frame(rho, frame):
    if frame is None or frame.empty:
        return False
    for _, row in frame.iterrows():
        if not _vx_phase6_rho_valid_for_lambdas(
            rho, row["model_lambda_home"], row["model_lambda_away"]
        ):
            return False
    return True


def _vx_phase6_poisson_score_probability(lambda_home, lambda_away, home_goals, away_goals):
    return float(
        _vx_phase4_poisson_pmf(float(lambda_home), int(home_goals))
        * _vx_phase4_poisson_pmf(float(lambda_away), int(away_goals))
    )


def _vx_phase6_dc_score_probability(lambda_home, lambda_away, home_goals, away_goals, rho):
    base = _vx_phase6_poisson_score_probability(
        lambda_home, lambda_away, home_goals, away_goals
    )
    tau = vx_phase6_tau(home_goals, away_goals, lambda_home, lambda_away, rho)
    return float(base * tau)


def _vx_phase6_joint_matrix(lambda_home, lambda_away, rho, max_goals):
    lh = float(lambda_home)
    la = float(lambda_away)
    max_goals = int(max_goals)
    home = _vx_phase4_poisson_vector(lh, max_goals)
    away = _vx_phase4_poisson_vector(la, max_goals)
    matrix = np.outer(home, away)
    tau = np.ones_like(matrix, dtype=float)
    for hg, ag in ((0, 0), (0, 1), (1, 0), (1, 1)):
        tau[hg, ag] = vx_phase6_tau(hg, ag, lh, la, rho)
    dc = matrix * tau
    return matrix, dc, tau


def _vx_phase6_adaptive_joint(lambda_home, lambda_away, rho):
    """Tail-safe joint matrix for W/D/L and BTTS aggregation."""
    home = _vx_phase4_adaptive_poisson_vector(float(lambda_home))
    away = _vx_phase4_adaptive_poisson_vector(float(lambda_away))
    n = max(len(home), len(away))
    if len(home) < n:
        home = _vx_phase4_poisson_vector(float(lambda_home), n - 1)
    if len(away) < n:
        away = _vx_phase4_poisson_vector(float(lambda_away), n - 1)

    base = np.outer(home, away)
    dc = base.copy()
    for hg, ag in ((0, 0), (0, 1), (1, 0), (1, 1)):
        dc[hg, ag] *= vx_phase6_tau(
            hg, ag, float(lambda_home), float(lambda_away), float(rho)
        )

    mass = float(dc.sum())
    if not np.isfinite(mass) or mass <= 0:
        raise RuntimeError("Invalid Dixon–Coles adaptive matrix mass.")
    # The analytical DC correction preserves total mass. This tiny normalization
    # removes only the negligible adaptive Poisson tail truncation.
    dc = dc / mass
    return dc


def _vx_phase6_fixture_probabilities(lambda_home, lambda_away, rho):
    matrix = _vx_phase6_adaptive_joint(lambda_home, lambda_away, rho)
    p_home = float(np.tril(matrix, -1).sum())
    p_draw = float(np.trace(matrix))
    p_away = float(np.triu(matrix, 1).sum())
    p_btts = float(matrix[1:, 1:].sum())

    # Dixon–Coles preserves each team's Poisson marginal distribution exactly.
    home_cs = float(math.exp(-float(lambda_away)))
    away_cs = float(math.exp(-float(lambda_home)))
    total_lambda = float(lambda_home) + float(lambda_away)
    over_2_5 = float(
        1.0 - math.exp(-total_lambda) * (1.0 + total_lambda + total_lambda**2 / 2.0)
    )

    return {
        "home_win_probability": p_home,
        "draw_probability": p_draw,
        "away_win_probability": p_away,
        "both_teams_score_probability": p_btts,
        "over_2_5_goals_probability": over_2_5,
        "home_clean_sheet_probability": home_cs,
        "away_clean_sheet_probability": away_cs,
    }


def _vx_phase6_prepare_history(frame):
    """Accept only frozen historical forecasts with realised scorelines."""
    if frame is None:
        return pd.DataFrame()
    h = frame.copy() if isinstance(frame, pd.DataFrame) else pd.DataFrame(frame)
    required = {
        "forecast_timestamp",
        "fixture_id",
        "model_lambda_home",
        "model_lambda_away",
        "actual_home_goals",
        "actual_away_goals",
    }
    if h.empty or not required.issubset(h.columns):
        return pd.DataFrame()

    if "is_pre_deadline_data" in h.columns:
        h = h.loc[h["is_pre_deadline_data"].fillna(False).astype(bool)].copy()

    for c in (
        "fixture_id", "model_lambda_home", "model_lambda_away",
        "actual_home_goals", "actual_away_goals",
    ):
        h[c] = pd.to_numeric(h[c], errors="coerce")
    h["forecast_timestamp"] = pd.to_datetime(h["forecast_timestamp"], utc=True, errors="coerce")
    h = h.dropna(subset=list(required)).copy()
    h = h.loc[
        (h["model_lambda_home"] > 0)
        & (h["model_lambda_away"] > 0)
        & (h["actual_home_goals"] >= 0)
        & (h["actual_away_goals"] >= 0)
    ].copy()
    if h.empty:
        return h

    h["fixture_id"] = h["fixture_id"].astype(int)
    h["actual_home_goals"] = h["actual_home_goals"].astype(int)
    h["actual_away_goals"] = h["actual_away_goals"].astype(int)
    h = (
        h.sort_values(["forecast_timestamp", "fixture_id"])
        .drop_duplicates("fixture_id", keep="last")
        .reset_index(drop=True)
    )
    return h


def _vx_phase6_load_history():
    """Phase 7 will own the permanent frozen-forecast archive; Phase 6 only reads it."""
    for name in (
        "phase7_validation_history",
        "forecast_validation_history",
        "frozen_forecast_history",
    ):
        obj = globals().get(name)
        prepared = _vx_phase6_prepare_history(obj)
        if not prepared.empty:
            return prepared, f"global:{name}"

    history_root = Path(PROJECT_ROOT) / "MODEL_HISTORY"
    candidates = (
        history_root / "phase7_frozen_forecast_history.jsonl",
        history_root / "phase7_frozen_forecast_history.json",
        history_root / "forecast_validation_history.jsonl",
        history_root / "forecast_validation_history.json",
    )
    for path in candidates:
        if not path.exists():
            continue
        try:
            if path.suffix.lower() == ".jsonl":
                raw = pd.read_json(path, lines=True)
            else:
                raw = pd.read_json(path)
            prepared = _vx_phase6_prepare_history(raw)
            if not prepared.empty:
                return prepared, str(path)
        except Exception:
            continue
    return pd.DataFrame(), "none"


def _vx_phase6_row_outcome_probabilities(row, rho):
    probs = _vx_phase6_fixture_probabilities(
        row["model_lambda_home"], row["model_lambda_away"], rho
    )
    hg = int(row["actual_home_goals"])
    ag = int(row["actual_away_goals"])
    if hg > ag:
        actual_index = 0
    elif hg == ag:
        actual_index = 1
    else:
        actual_index = 2
    vector = np.array(
        [
            probs["home_win_probability"],
            probs["draw_probability"],
            probs["away_win_probability"],
        ],
        dtype=float,
    )
    return vector, actual_index


def _vx_phase6_metrics(frame, rho):
    eps = 1e-15
    score_losses = []
    wdl_losses = []
    brier_scores = []

    for _, row in frame.iterrows():
        p_score = _vx_phase6_dc_score_probability(
            row["model_lambda_home"],
            row["model_lambda_away"],
            row["actual_home_goals"],
            row["actual_away_goals"],
            rho,
        )
        score_losses.append(-math.log(max(eps, min(1.0, p_score))))

        vector, actual_index = _vx_phase6_row_outcome_probabilities(row, rho)
        p_actual = float(vector[actual_index])
        wdl_losses.append(-math.log(max(eps, min(1.0, p_actual))))
        target = np.zeros(3, dtype=float)
        target[actual_index] = 1.0
        brier_scores.append(float(np.sum((vector - target) ** 2)))

    return {
        "scoreline_log_loss": float(np.mean(score_losses)),
        "wdl_log_loss": float(np.mean(wdl_losses)),
        "multiclass_brier": float(np.mean(brier_scores)),
    }


def _vx_phase6_fit_and_validate(history):
    """Chronological train/holdout selection. No same-sample activation."""
    h = _vx_phase6_prepare_history(history)
    n = len(h)
    min_total = PHASE6_MIN_TRAIN_FIXTURES + PHASE6_MIN_VALIDATION_FIXTURES
    if n < min_total:
        return None, (
            f"WAITING_FOR_HISTORY: need at least {min_total} frozen completed fixtures; have {n}"
        )

    split = int(math.floor(n * PHASE6_TRAIN_FRACTION))
    split = max(PHASE6_MIN_TRAIN_FIXTURES, split)
    split = min(split, n - PHASE6_MIN_VALIDATION_FIXTURES)
    train = h.iloc[:split].copy()
    valid = h.iloc[split:].copy()

    usable_rhos = [
        float(r) for r in PHASE6_RHO_GRID
        if _vx_phase6_rho_valid_for_frame(float(r), train)
        and _vx_phase6_rho_valid_for_frame(float(r), valid)
    ]
    if 0.0 not in usable_rhos:
        usable_rhos.append(0.0)
    usable_rhos = sorted(set(usable_rhos))

    scored = []
    for r in usable_rhos:
        train_metrics = _vx_phase6_metrics(train, r)
        scored.append((train_metrics["scoreline_log_loss"], abs(r), r))
    scored.sort(key=lambda x: (x[0], x[1], x[2]))
    candidate_rho = float(scored[0][2])

    poisson = _vx_phase6_metrics(valid, 0.0)
    dc = _vx_phase6_metrics(valid, candidate_rho)

    score_improvement = poisson["scoreline_log_loss"] - dc["scoreline_log_loss"]
    wdl_improvement = poisson["wdl_log_loss"] - dc["wdl_log_loss"]
    brier_improvement = poisson["multiclass_brier"] - dc["multiclass_brier"]

    approved = bool(
        abs(candidate_rho) > 1e-12
        and score_improvement > PHASE6_METRIC_TOLERANCE
        and wdl_improvement >= -PHASE6_METRIC_TOLERANCE
        and brier_improvement >= -PHASE6_METRIC_TOLERANCE
    )

    if approved:
        reason = "APPROVED_OUT_OF_SAMPLE"
    elif abs(candidate_rho) <= 1e-12:
        reason = "REJECTED_RHO_ZERO_BEST"
    elif score_improvement <= PHASE6_METRIC_TOLERANCE:
        reason = "REJECTED_NO_SCORELINE_LOG_LOSS_GAIN"
    elif wdl_improvement < -PHASE6_METRIC_TOLERANCE:
        reason = "REJECTED_WDL_LOG_LOSS_WORSE"
    else:
        reason = "REJECTED_MULTICLASS_BRIER_WORSE"

    result = {
        "candidate_rho": candidate_rho,
        "train_fixture_count": int(len(train)),
        "validation_fixture_count": int(len(valid)),
        "poisson_scoreline_log_loss": poisson["scoreline_log_loss"],
        "dixon_coles_scoreline_log_loss": dc["scoreline_log_loss"],
        "scoreline_log_loss_improvement": score_improvement,
        "poisson_wdl_log_loss": poisson["wdl_log_loss"],
        "dixon_coles_wdl_log_loss": dc["wdl_log_loss"],
        "wdl_log_loss_improvement": wdl_improvement,
        "poisson_multiclass_brier": poisson["multiclass_brier"],
        "dixon_coles_multiclass_brier": dc["multiclass_brier"],
        "multiclass_brier_improvement": brier_improvement,
        "validation_approved": approved,
        "selection_reason": reason,
        "model_version": PHASE6_MODEL_VERSION,
        "calculated_at": PHASE6_TIMESTAMP,
    }
    return result, reason


# ------------------------------------------------------------
# VALIDATION GATE
# ------------------------------------------------------------
_phase6_history, PHASE6_HISTORY_SOURCE = _vx_phase6_load_history()
_phase6_result, PHASE6_SELECTION_REASON = _vx_phase6_fit_and_validate(_phase6_history)

if _phase6_result is None:
    PHASE6_VALIDATION_STATUS = "WAITING_FOR_PHASE7_FROZEN_HISTORY"
    rho_parameter = 0.0
    PHASE6_DIXON_COLES_ACTIVE = False
else:
    phase6_validation_report = pd.DataFrame(
        [_phase6_result], columns=PHASE6_VALIDATION_FIELDS
    )
    rho_parameter = float(_phase6_result["candidate_rho"])
    PHASE6_DIXON_COLES_ACTIVE = bool(_phase6_result["validation_approved"])
    PHASE6_VALIDATION_STATUS = (
        "APPROVED" if PHASE6_DIXON_COLES_ACTIVE else "REJECTED_KEEP_POISSON"
    )

# If validation has not approved the candidate, rho=0 is used for the stored
# candidate matrix. That makes it mathematically identical to baseline Poisson
# and guarantees there is no unvalidated low-score correction in production.
_phase6_applied_rho = float(rho_parameter if PHASE6_DIXON_COLES_ACTIVE else 0.0)


# ------------------------------------------------------------
# BUILD CURRENT-FIXTURE DIXON–COLES CANDIDATE MATRIX
# ------------------------------------------------------------
_phase6_matrix_rows = []
_phase6_fixture_rows = []
_phase6_max_goals = int(globals().get("PHASE4_POISSON_MAX_GOALS", 8))
_phase4_by_fixture = fixture_probabilities_model_only.set_index("fixture_id", drop=False)

for fid, base_group in poisson_probability_matrix.groupby("fixture_id", sort=False):
    fid = int(fid)
    if fid not in _phase4_by_fixture.index:
        raise RuntimeError(f"Phase 6 fixture {fid} missing Phase-4 fixture probabilities.")
    base_fx = _phase4_by_fixture.loc[fid]
    if isinstance(base_fx, pd.DataFrame):
        raise RuntimeError(f"Phase 6 requires unique fixture ID {fid} in Phase 4A.")

    lh = float(base_fx["model_lambda_home"])
    la = float(base_fx["model_lambda_away"])
    if not _vx_phase6_rho_valid_for_lambdas(_phase6_applied_rho, lh, la):
        raise RuntimeError(f"Validated rho is not admissible for current fixture {fid}.")

    base_matrix, dc_matrix, tau_matrix = _vx_phase6_joint_matrix(
        lh, la, _phase6_applied_rho, _phase6_max_goals
    )

    base_mass = float(base_matrix.sum())
    dc_mass = float(dc_matrix.sum())
    if not math.isclose(base_mass, dc_mass, abs_tol=1e-12, rel_tol=0.0):
        raise RuntimeError(f"Dixon–Coles mass-preservation check failed for fixture {fid}.")

    gw = int(base_fx["gameweek"])
    home_id = int(base_fx["home_team_id"])
    away_id = int(base_fx["away_team_id"])
    forecast_ts = base_fx["forecast_timestamp"]

    for hg in range(_phase6_max_goals + 1):
        for ag in range(_phase6_max_goals + 1):
            _phase6_matrix_rows.append({
                "fixture_id": fid,
                "gameweek": gw,
                "home_team_id": home_id,
                "away_team_id": away_id,
                "home_goals": hg,
                "away_goals": ag,
                "scoreline": f"{hg}-{ag}",
                "probability": float(dc_matrix[hg, ag]),
                "poisson_probability": float(base_matrix[hg, ag]),
                "tau_correction": float(tau_matrix[hg, ag]),
                "rho_parameter": _phase6_applied_rho,
                "model_lambda_home": lh,
                "model_lambda_away": la,
                "model_version": PHASE6_MODEL_VERSION,
                "forecast_timestamp": forecast_ts,
                "validation_approved": bool(PHASE6_DIXON_COLES_ACTIVE),
            })

    probs = _vx_phase6_fixture_probabilities(lh, la, _phase6_applied_rho)
    _phase6_fixture_rows.append({
        "fixture_id": fid,
        "gameweek": gw,
        "home_team_id": home_id,
        "away_team_id": away_id,
        "model_lambda_home": lh,
        "model_lambda_away": la,
        "rho_parameter": _phase6_applied_rho,
        **probs,
        "dixon_coles_matrix_covered_mass_0_to_8": dc_mass,
        "validation_approved": bool(PHASE6_DIXON_COLES_ACTIVE),
        "model_version": PHASE6_MODEL_VERSION,
        "forecast_timestamp": forecast_ts,
    })

dixon_coles_probability_matrix = pd.DataFrame(
    _phase6_matrix_rows, columns=PHASE6_MATRIX_FIELDS
)
dixon_coles_fixture_probabilities = pd.DataFrame(
    _phase6_fixture_rows, columns=PHASE6_FIXTURE_FIELDS
)

if dixon_coles_probability_matrix.empty or dixon_coles_fixture_probabilities.empty:
    raise RuntimeError("Phase 6 produced empty current-fixture outputs.")

_expected_cells = (_phase6_max_goals + 1) ** 2
_group_sizes = dixon_coles_probability_matrix.groupby("fixture_id").size()
if not (_group_sizes == _expected_cells).all():
    raise RuntimeError("Phase 6 score matrix does not contain the expected cells per fixture.")

# rho=0 must reproduce baseline Poisson exactly. This is the required locked
# fallback whenever Dixon–Coles has not won out-of-sample validation.
if not PHASE6_DIXON_COLES_ACTIVE:
    _merged = dixon_coles_probability_matrix.merge(
        poisson_probability_matrix[["fixture_id", "home_goals", "away_goals", "probability"]],
        on=["fixture_id", "home_goals", "away_goals"],
        how="left",
        suffixes=("_dc", "_phase4"),
        validate="one_to_one",
    )
    if _merged["probability_phase4"].isna().any() or not np.allclose(
        _merged["probability_dc"].to_numpy(dtype=float),
        _merged["probability_phase4"].to_numpy(dtype=float),
        atol=1e-14,
        rtol=0.0,
    ):
        raise RuntimeError("Phase 6 locked Poisson fallback is not identical to Phase 4A.")

# Dixon–Coles must not alter the Poisson marginals used for clean sheets or the
# total-goals distribution used for Over 2.5. Assert that identity explicitly.
_phase6_compare = dixon_coles_fixture_probabilities.merge(
    fixture_probabilities_model_only[
        [
            "fixture_id",
            "home_clean_sheet_probability",
            "away_clean_sheet_probability",
            "over_2_5_goals_probability",
        ]
    ],
    on="fixture_id",
    how="left",
    suffixes=("_dc", "_poisson"),
    validate="one_to_one",
)
for c in (
    "home_clean_sheet_probability",
    "away_clean_sheet_probability",
    "over_2_5_goals_probability",
):
    if not np.allclose(
        _phase6_compare[f"{c}_dc"].to_numpy(dtype=float),
        _phase6_compare[f"{c}_poisson"].to_numpy(dtype=float),
        atol=1e-12,
        rtol=0.0,
    ):
        raise RuntimeError(f"Phase 6 illegally changed Poisson marginal output: {c}")

# Selected score probabilities are explicit and separate. Phase 4A is never
# overwritten. Until validation approves Dixon–Coles, selection remains Poisson.
score_probability_matrix_selected = (
    dixon_coles_probability_matrix.copy()
    if PHASE6_DIXON_COLES_ACTIVE
    else poisson_probability_matrix.copy()
)
PHASE6_SELECTED_SCORE_MODEL = (
    "DIXON_COLES" if PHASE6_DIXON_COLES_ACTIVE else "BASELINE_POISSON"
)

# ------------------------------------------------------------
# SAVE CURRENT-RUN INSPECTION OUTPUTS
# ------------------------------------------------------------
PHASE6_OUTPUT_DIR = Path(DATA_DIR)
PHASE6_MATRIX_PATH = PHASE6_OUTPUT_DIR / "phase6_dixon_coles_probability_matrix_0_to_8.json"
PHASE6_FIXTURE_PATH = PHASE6_OUTPUT_DIR / "phase6_dixon_coles_fixture_probabilities.json"
PHASE6_VALIDATION_PATH = PHASE6_OUTPUT_DIR / "phase6_poisson_vs_dixon_coles_validation.json"

PHASE6_MATRIX_PATH.write_text(
    dixon_coles_probability_matrix.to_json(orient="records", date_format="iso", indent=2),
    encoding="utf-8",
)
PHASE6_FIXTURE_PATH.write_text(
    dixon_coles_fixture_probabilities.to_json(orient="records", date_format="iso", indent=2),
    encoding="utf-8",
)
PHASE6_VALIDATION_PATH.write_text(
    phase6_validation_report.to_json(orient="records", date_format="iso", indent=2),
    encoding="utf-8",
)

# Only persist an activation state after an out-of-sample win. A rejected or
# untested rho is never persisted as production truth.
if PHASE6_DIXON_COLES_ACTIVE:
    _state_dir = Path(PROJECT_ROOT) / "MODEL_HISTORY"
    _state_dir.mkdir(parents=True, exist_ok=True)
    PHASE6_VALIDATED_STATE_PATH = _state_dir / "phase6_dixon_coles_validated_state.json"
    PHASE6_VALIDATED_STATE_PATH.write_text(
        json.dumps(
            {
                "model_version": PHASE6_MODEL_VERSION,
                "baseline_model_version": PHASE6_BASELINE_MODEL_VERSION,
                "rho_parameter": float(rho_parameter),
                "validation_status": PHASE6_VALIDATION_STATUS,
                "selection_reason": PHASE6_SELECTION_REASON,
                "history_source": PHASE6_HISTORY_SOURCE,
                "validation_report": _phase6_result,
            },
            ensure_ascii=False,
            indent=2,
            default=str,
        ),
        encoding="utf-8",
    )

# ------------------------------------------------------------
# HARD NO-MUTATION CHECK
# ------------------------------------------------------------
_phase6_upstream_after = {
    "projected_goals_model_only": _vx_phase6_frame_digest(globals().get("projected_goals_model_only")),
    "fixture_probabilities_model_only": _vx_phase6_frame_digest(fixture_probabilities_model_only),
    "poisson_probability_matrix": _vx_phase6_frame_digest(poisson_probability_matrix),
    "fixture_fdr_model_only": _vx_phase6_frame_digest(globals().get("fixture_fdr_model_only")),
}
if _phase6_upstream_before != _phase6_upstream_after:
    raise RuntimeError("Phase 6 clean-room violation: an upstream model output was mutated.")

if "vx_phase0_assert_cleanroom" in globals():
    vx_phase0_assert_cleanroom("after Phase 6 Dixon-Coles candidate")

PHASE6_READY = True
print("✅ PHASE 6 DIXON–COLES CANDIDATE ENGINE BUILT")
print(f"   Validation status: {PHASE6_VALIDATION_STATUS}")
print(f"   History source: {PHASE6_HISTORY_SOURCE}")
print(f"   Selection: {PHASE6_SELECTED_SCORE_MODEL}")
print(f"   rho_parameter used now: {_phase6_applied_rho:.3f}")
if not PHASE6_DIXON_COLES_ACTIVE:
    print("🔒 Dixon–Coles is NOT active: baseline Poisson remains production truth.")
    print("   Phase 7 must provide enough frozen pre-deadline completed forecasts for out-of-sample validation.")
else:
    print("✅ Dixon–Coles beat baseline Poisson on the chronological holdout and is active for score probabilities.")
print("🔒 Team strengths and projected-goal lambdas were not changed.")
print("🔒 Clean-sheet probabilities and FDR inputs remain the Phase-4 model-only marginals.")
print("🔒 No market odds, PlayerElo or player expected-points input is used here.")
print("   Saved:", PHASE6_MATRIX_PATH)
print("   Saved:", PHASE6_FIXTURE_PATH)
print("   Saved:", PHASE6_VALIDATION_PATH)
```

```python
# ============================================================
# CELL 13A — PHASE 7
# DEADLINE-SAFE FORECAST FREEZE + OUT-OF-SAMPLE VALIDATION
# NO PLAYER EXPECTED-POINTS • NO CAPTAINCY • NO SQUAD DECISIONS
# ============================================================
# Contract:
#   1) freeze the latest model-only forecast available BEFORE each FPL deadline;
#   2) never rewrite a frozen forecast after its deadline;
#   3) attach realised scores only after the fixture is officially finished;
#   4) validate projected goals, clean sheets, W/D/L, FDR bands, confidence,
#      market benchmarks and historical market-lambda blend candidates;
#   5) create evidence only — Phase 7 does NOT activate a market blend and
#      does NOT unlock any player expected-points or decision model.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import math
import numpy as np
import pandas as pd

if "PHASE4_POISSON_READY" not in globals() or not bool(PHASE4_POISSON_READY):
    raise RuntimeError("Run Phase 4A baseline Poisson before Phase 7.")
if "PHASE4_READY" not in globals() or not bool(PHASE4_READY):
    raise RuntimeError("Run Phase 4B clean-sheet/FDR outputs before Phase 7.")
if "fixture_probabilities_model_only" not in globals() or fixture_probabilities_model_only.empty:
    raise RuntimeError("Phase 7 requires non-empty Phase-4 model probabilities.")
if "fixture_fdr_model_only" not in globals() or fixture_fdr_model_only.empty:
    raise RuntimeError("Phase 7 requires non-empty Phase-4 FDR outputs.")

PHASE7_MODEL_VERSION = "VORTEX-V1-PHASE7-VALIDATION"
PHASE7_TIMESTAMP = datetime.now(timezone.utc).isoformat()
PHASE7_MIN_COMPLETED_FIXTURES = 80
PHASE7_MARKET_MIN_TRAIN_FIXTURES = 60
PHASE7_MARKET_MIN_VALIDATION_FIXTURES = 20
PHASE7_MARKET_TRAIN_FRACTION = 0.75
PHASE7_BLEND_ALPHAS = (1.00, 0.90, 0.80, 0.70, 0.60, 0.50, 0.00)
PHASE7_MARKET_BLEND_ACTIVE = False
PHASE7_READY = False
PHASE7_VALIDATION_EVIDENCE_READY = False
PHASE7_READY_FOR_PLAYER_MODEL = False
PHASE7_VALIDATION_STATUS = "NOT_RUN"
PHASE7_MARKET_BLEND_STATUS = "NOT_RUN"

if PHASE7_MARKET_BLEND_ACTIVE:
    raise RuntimeError("Phase 7 clean-room violation: historical blend testing must not activate a live blend.")

# ----------------------------------------------------------------
# PERSISTENT ARCHIVE — survives the disposable FIRST VIDEO output tree.
# ----------------------------------------------------------------
PHASE7_HISTORY_DIR = Path(PROJECT_ROOT) / "MODEL_HISTORY"
PHASE7_HISTORY_DIR.mkdir(parents=True, exist_ok=True)
PHASE7_HISTORY_PATH = PHASE7_HISTORY_DIR / "phase7_frozen_forecast_history.jsonl"
PHASE7_SUMMARY_PATH = PHASE7_HISTORY_DIR / "phase7_validation_summary.json"

PHASE7_FREEZE_FIELDS = (
    "forecast_timestamp",
    "data_cutoff_timestamp",
    "fixture_id",
    "gameweek",
    "fpl_deadline_timestamp",
    "model_version",
    "input_data_version",
    "probability_model_version",
    "home_team",
    "away_team",
    "model_lambda_home",
    "model_lambda_away",
    "home_win_probability",
    "draw_probability",
    "away_win_probability",
    "over_1_5_probability",
    "over_2_5_probability",
    "both_teams_score_probability",
    "home_team_scores_1plus_probability",
    "home_team_scores_2plus_probability",
    "away_team_scores_1plus_probability",
    "away_team_scores_2plus_probability",
    "home_clean_sheet_probability",
    "away_clean_sheet_probability",
    "home_cs_probability",
    "away_cs_probability",
    "attack_fdr_home",
    "defence_fdr_home",
    "attack_fdr_away",
    "defence_fdr_away",
    "model_confidence",
    "home_prior_weight",
    "home_current_season_weight",
    "away_prior_weight",
    "away_current_season_weight",
    "home_team_strength_inputs_json",
    "away_team_strength_inputs_json",
    "home_team_news_adjustments_json",
    "away_team_news_adjustments_json",
    "market_snapshot_timestamp",
    "market_home_win_probability",
    "market_draw_probability",
    "market_away_win_probability",
    "market_over_2_5_probability",
    "market_home_clean_sheet_probability",
    "market_away_clean_sheet_probability",
    "market_lambda_home",
    "market_lambda_away",
    "actual_home_goals",
    "actual_away_goals",
    "final_result",
    "actual_home_clean_sheet",
    "actual_away_clean_sheet",
    "actual_clean_sheets",
    "is_pre_deadline_data",
    "freeze_status",
    "data_snapshot_timestamp",
    "source_name",
    "source_timestamp",
)

PHASE7_METRIC_FIELDS = (
    "metric_group",
    "metric_name",
    "value",
    "sample_size",
    "model_version",
    "calculated_at",
)


def _vx_phase7_frame_digest(frame):
    if frame is None:
        return "NONE"
    if not isinstance(frame, pd.DataFrame):
        return hashlib.sha256(repr(frame).encode("utf-8")).hexdigest()
    if frame.empty:
        return "EMPTY:" + ",".join(map(str, frame.columns))
    stable = frame.copy()
    stable = stable.reindex(sorted(stable.columns), axis=1)
    try:
        stable = stable.sort_values(list(stable.columns), kind="stable").reset_index(drop=True)
    except Exception:
        stable = stable.reset_index(drop=True)
    payload = stable.to_json(orient="split", date_format="iso", default_handler=str)
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()


_phase7_upstream_before = {
    "projected_goals_model_only": _vx_phase7_frame_digest(globals().get("projected_goals_model_only")),
    "fixture_probabilities_model_only": _vx_phase7_frame_digest(fixture_probabilities_model_only),
    "poisson_probability_matrix": _vx_phase7_frame_digest(globals().get("poisson_probability_matrix")),
    "fixture_fdr_model_only": _vx_phase7_frame_digest(fixture_fdr_model_only),
    "market_odds_benchmark": _vx_phase7_frame_digest(globals().get("market_odds_benchmark")),
    "dixon_coles_probability_matrix": _vx_phase7_frame_digest(globals().get("dixon_coles_probability_matrix")),
}


def _vx_phase7_utc(value):
    ts = pd.Timestamp(value)
    if pd.isna(ts):
        return pd.NaT
    if ts.tzinfo is None:
        return ts.tz_localize("UTC")
    return ts.tz_convert("UTC")


def _vx_phase7_deadline_map():
    out = {}
    frame = globals().get("events")
    if not isinstance(frame, pd.DataFrame) or frame.empty or "id" not in frame.columns:
        return out
    deadline_col = "deadline_time" if "deadline_time" in frame.columns else None
    if deadline_col is None:
        return out
    for _, row in frame.iterrows():
        try:
            gw = int(row["id"])
            ts = _vx_phase7_utc(row[deadline_col])
            if not pd.isna(ts):
                out[gw] = ts
        except Exception:
            continue
    return out


PHASE7_DEADLINES = _vx_phase7_deadline_map()


def _vx_phase7_load_history(path):
    if not Path(path).exists():
        return pd.DataFrame(columns=PHASE7_FREEZE_FIELDS)
    try:
        h = pd.read_json(path, lines=True)
    except Exception:
        return pd.DataFrame(columns=PHASE7_FREEZE_FIELDS)
    for c in PHASE7_FREEZE_FIELDS:
        if c not in h.columns:
            h[c] = pd.NA
    return h.loc[:, PHASE7_FREEZE_FIELDS].copy()


def _vx_phase7_market_lookup():
    frame = globals().get("market_odds_benchmark")
    if not isinstance(frame, pd.DataFrame) or frame.empty or "fixture_id" not in frame.columns:
        return {}
    m = frame.copy()
    if "is_pre_deadline_data" in m.columns:
        m = m.loc[m["is_pre_deadline_data"].fillna(False).astype(bool)].copy()
    if m.empty:
        return {}
    m["fixture_id"] = pd.to_numeric(m["fixture_id"], errors="coerce")
    m = m.dropna(subset=["fixture_id"]).copy()
    m["fixture_id"] = m["fixture_id"].astype(int)
    if "market_snapshot_timestamp" in m.columns:
        m["_ts"] = pd.to_datetime(m["market_snapshot_timestamp"], utc=True, errors="coerce")
        m = m.sort_values(["fixture_id", "_ts"]).drop_duplicates("fixture_id", keep="last")
    else:
        m = m.drop_duplicates("fixture_id", keep="last")
    return {int(r["fixture_id"]): r for _, r in m.iterrows()}


def _vx_phase7_team_name_map():
    frame = globals().get("teams")
    if not isinstance(frame, pd.DataFrame) or frame.empty or not {"id", "name"}.issubset(frame.columns):
        return {}
    out = {}
    for _, row in frame.iterrows():
        try:
            out[int(row["id"])] = str(row["name"])
        except Exception:
            continue
    return out


def _vx_phase7_input_data_version():
    parts = []
    for name in (
        "team_match_xg_raw", "team_priors", "team_strength_signals",
        "team_strength_updated", "fixture_context_phase2", "team_news_overrides",
    ):
        parts.append(f"{name}:{_vx_phase7_frame_digest(globals().get(name))}")
    return "INPUT-" + hashlib.sha256("|".join(parts).encode("utf-8")).hexdigest()[:20]


def _vx_phase7_data_cutoff(forecast_ts):
    candidates = []
    for name in (
        "team_match_xg_raw", "team_priors", "team_strength_signals",
        "team_strength_updated", "fixture_context_phase2", "team_news_overrides",
    ):
        frame = globals().get(name)
        if not isinstance(frame, pd.DataFrame) or frame.empty:
            continue
        for col in ("source_timestamp", "data_snapshot_timestamp"):
            if col not in frame.columns:
                continue
            values = pd.to_datetime(frame[col], utc=True, errors="coerce").dropna()
            candidates.extend([x for x in values.tolist() if x <= forecast_ts])
    global_snapshot = pd.to_datetime(globals().get("data_snapshot_timestamp"), utc=True, errors="coerce")
    if not pd.isna(global_snapshot) and global_snapshot <= forecast_ts:
        candidates.append(global_snapshot)
    return max(candidates) if candidates else pd.NaT


def _vx_phase7_strength_lookup():
    frame = globals().get("team_strength_updated")
    if not isinstance(frame, pd.DataFrame) or frame.empty or "team_id" not in frame.columns:
        return {}
    return {int(r["team_id"]): r for _, r in frame.iterrows() if not pd.isna(r.get("team_id"))}


def _vx_phase7_news_lookup():
    frame = globals().get("team_news_overrides")
    if not isinstance(frame, pd.DataFrame) or frame.empty or not {"fixture_id", "team_id"}.issubset(frame.columns):
        return {}
    out = {}
    for _, r in frame.iterrows():
        try:
            out[(int(r["fixture_id"]), int(r["team_id"]))] = r
        except Exception:
            continue
    return out


def _vx_phase7_jsonable_dict(row, fields):
    if row is None:
        return json.dumps({}, sort_keys=True)
    obj = {}
    for field in fields:
        value = row.get(field, pd.NA)
        if pd.isna(value):
            obj[field] = None
        elif isinstance(value, (np.integer, int)):
            obj[field] = int(value)
        elif isinstance(value, (np.floating, float)):
            obj[field] = float(value)
        elif isinstance(value, (np.bool_, bool)):
            obj[field] = bool(value)
        else:
            obj[field] = str(value)
    return json.dumps(obj, ensure_ascii=False, sort_keys=True)


def _vx_phase7_current_freeze_rows():
    probs = fixture_probabilities_model_only.copy()
    fdr = fixture_fdr_model_only.copy()
    required_prob = {
        "fixture_id", "gameweek", "model_lambda_home", "model_lambda_away",
        "home_win_probability", "draw_probability", "away_win_probability",
        "home_scores_probability", "away_scores_probability",
        "home_scores_2plus_probability", "away_scores_2plus_probability",
        "both_teams_score_probability", "over_1_5_goals_probability",
        "over_2_5_goals_probability", "home_clean_sheet_probability",
        "away_clean_sheet_probability", "model_confidence", "model_version",
        "forecast_timestamp", "home_team_id", "away_team_id",
    }
    required_fdr = {
        "fixture_id", "attack_fdr_home", "defence_fdr_home",
        "attack_fdr_away", "defence_fdr_away",
    }
    if not required_prob.issubset(probs.columns):
        raise RuntimeError(f"Phase 7 missing probability fields: {sorted(required_prob - set(probs.columns))}")
    if not required_fdr.issubset(fdr.columns):
        raise RuntimeError(f"Phase 7 missing FDR fields: {sorted(required_fdr - set(fdr.columns))}")

    fdr_small = fdr.loc[:, list(required_fdr)].drop_duplicates("fixture_id", keep="last")
    merged = probs.merge(fdr_small, on="fixture_id", how="left", validate="one_to_one")
    market = _vx_phase7_market_lookup()
    team_names = _vx_phase7_team_name_map()
    strengths = _vx_phase7_strength_lookup()
    news = _vx_phase7_news_lookup()
    input_version = _vx_phase7_input_data_version()
    rows = []
    for _, r in merged.iterrows():
        fid = int(r["fixture_id"])
        gw = int(r["gameweek"])
        home_id = int(r["home_team_id"])
        away_id = int(r["away_team_id"])
        forecast_ts = _vx_phase7_utc(r["forecast_timestamp"])
        deadline_ts = PHASE7_DEADLINES.get(gw)
        # A frozen forecast is valid only when the forecast itself existed before
        # the FPL deadline. Unknown deadlines are not silently treated as valid.
        pre_deadline = bool(
            deadline_ts is not None
            and not pd.isna(forecast_ts)
            and forecast_ts < deadline_ts
        )
        if not pre_deadline:
            continue

        m = market.get(fid)
        market_ts = pd.NaT
        market_values = {}
        if m is not None:
            market_ts = pd.to_datetime(m.get("market_snapshot_timestamp"), utc=True, errors="coerce")
            if not pd.isna(market_ts) and market_ts < deadline_ts:
                market_values = {
                    "market_home_win_probability": m.get("market_home_win_probability", pd.NA),
                    "market_draw_probability": m.get("market_draw_probability", pd.NA),
                    "market_away_win_probability": m.get("market_away_win_probability", pd.NA),
                    "market_over_2_5_probability": m.get("market_over_2_5_probability", pd.NA),
                    "market_home_clean_sheet_probability": m.get("market_home_clean_sheet_probability", pd.NA),
                    "market_away_clean_sheet_probability": m.get("market_away_clean_sheet_probability", pd.NA),
                    "market_lambda_home": m.get("market_lambda_home", pd.NA),
                    "market_lambda_away": m.get("market_lambda_away", pd.NA),
                }
            else:
                market_ts = pd.NaT

        home_strength = strengths.get(home_id)
        away_strength = strengths.get(away_id)
        home_news = news.get((fid, home_id))
        away_news = news.get((fid, away_id))
        data_cutoff_ts = _vx_phase7_data_cutoff(forecast_ts)
        strength_fields = (
            "completed_matches", "prior_weight", "current_season_weight",
            "updated_attack_overall", "updated_defence_overall",
            "updated_attack_home", "updated_defence_home",
            "updated_attack_away", "updated_defence_away",
            "recent_form_weight", "recency_effective_sample_size",
        )
        news_fields = tuple(globals().get("PHASE2_NEWS_FIELDS", ())) + (
            "override_active", "reason", "source_name", "source_timestamp", "is_pre_deadline_data",
        )

        row = {
            "forecast_timestamp": forecast_ts.isoformat(),
            "data_cutoff_timestamp": data_cutoff_ts.isoformat() if not pd.isna(data_cutoff_ts) else pd.NA,
            "fixture_id": fid,
            "gameweek": gw,
            "fpl_deadline_timestamp": deadline_ts.isoformat(),
            "model_version": str(globals().get("PHASE4_OUTPUT_MODEL_VERSION", "VORTEX-V1-PHASE4-POISSON-CS-FDR")),
            "input_data_version": input_version,
            "probability_model_version": str(r["model_version"]),
            "home_team": team_names.get(home_id, str(home_id)),
            "away_team": team_names.get(away_id, str(away_id)),
            "model_lambda_home": float(r["model_lambda_home"]),
            "model_lambda_away": float(r["model_lambda_away"]),
            "home_win_probability": float(r["home_win_probability"]),
            "draw_probability": float(r["draw_probability"]),
            "away_win_probability": float(r["away_win_probability"]),
            "over_1_5_probability": float(r["over_1_5_goals_probability"]),
            "over_2_5_probability": float(r["over_2_5_goals_probability"]),
            "both_teams_score_probability": float(r["both_teams_score_probability"]),
            "home_team_scores_1plus_probability": float(r["home_scores_probability"]),
            "home_team_scores_2plus_probability": float(r["home_scores_2plus_probability"]),
            "away_team_scores_1plus_probability": float(r["away_scores_probability"]),
            "away_team_scores_2plus_probability": float(r["away_scores_2plus_probability"]),
            "home_clean_sheet_probability": float(r["home_clean_sheet_probability"]),
            "away_clean_sheet_probability": float(r["away_clean_sheet_probability"]),
            "home_cs_probability": float(r["home_clean_sheet_probability"]),
            "away_cs_probability": float(r["away_clean_sheet_probability"]),
            "attack_fdr_home": float(r["attack_fdr_home"]),
            "defence_fdr_home": float(r["defence_fdr_home"]),
            "attack_fdr_away": float(r["attack_fdr_away"]),
            "defence_fdr_away": float(r["defence_fdr_away"]),
            "model_confidence": float(r["model_confidence"]),
            "home_prior_weight": float(home_strength.get("prior_weight")) if home_strength is not None else pd.NA,
            "home_current_season_weight": float(home_strength.get("current_season_weight")) if home_strength is not None else pd.NA,
            "away_prior_weight": float(away_strength.get("prior_weight")) if away_strength is not None else pd.NA,
            "away_current_season_weight": float(away_strength.get("current_season_weight")) if away_strength is not None else pd.NA,
            "home_team_strength_inputs_json": _vx_phase7_jsonable_dict(home_strength, strength_fields),
            "away_team_strength_inputs_json": _vx_phase7_jsonable_dict(away_strength, strength_fields),
            "home_team_news_adjustments_json": _vx_phase7_jsonable_dict(home_news, news_fields),
            "away_team_news_adjustments_json": _vx_phase7_jsonable_dict(away_news, news_fields),
            "market_snapshot_timestamp": market_ts.isoformat() if not pd.isna(market_ts) else pd.NA,
            "actual_home_goals": pd.NA,
            "actual_away_goals": pd.NA,
            "final_result": pd.NA,
            "actual_home_clean_sheet": pd.NA,
            "actual_away_clean_sheet": pd.NA,
            "actual_clean_sheets": pd.NA,
            "is_pre_deadline_data": True,
            "freeze_status": "LATEST_PRE_DEADLINE_RUN",
            "data_snapshot_timestamp": r.get("data_snapshot_timestamp", globals().get("data_snapshot_timestamp", pd.NA)),
            "source_name": "VORTEX_PHASE7_FROZEN_FORECAST",
            "source_timestamp": PHASE7_TIMESTAMP,
            **{k: market_values.get(k, pd.NA) for k in (
                "market_home_win_probability", "market_draw_probability", "market_away_win_probability",
                "market_over_2_5_probability", "market_home_clean_sheet_probability",
                "market_away_clean_sheet_probability", "market_lambda_home", "market_lambda_away",
            )},
        }
        rows.append(row)
    return pd.DataFrame(rows, columns=PHASE7_FREEZE_FIELDS)


def _vx_phase7_upsert_latest_predeadline(existing, current):
    e = existing.copy() if isinstance(existing, pd.DataFrame) else pd.DataFrame(columns=PHASE7_FREEZE_FIELDS)
    c = current.copy() if isinstance(current, pd.DataFrame) else pd.DataFrame(columns=PHASE7_FREEZE_FIELDS)
    if e.empty:
        combo = c.copy()
    elif c.empty:
        combo = e.copy()
    else:
        combo = pd.concat([e, c], ignore_index=True, sort=False)
    if combo.empty:
        return pd.DataFrame(columns=PHASE7_FREEZE_FIELDS)
    for col in PHASE7_FREEZE_FIELDS:
        if col not in combo.columns:
            combo[col] = pd.NA
    combo["fixture_id"] = pd.to_numeric(combo["fixture_id"], errors="coerce")
    combo["_forecast_ts"] = pd.to_datetime(combo["forecast_timestamp"], utc=True, errors="coerce")
    combo = combo.dropna(subset=["fixture_id", "model_version", "_forecast_ts"]).copy()
    combo["fixture_id"] = combo["fixture_id"].astype(int)
    # Keep the latest valid pre-deadline run for each fixture/model version.
    combo = (
        combo.sort_values(["fixture_id", "model_version", "_forecast_ts"])
        .drop_duplicates(["fixture_id", "model_version"], keep="last")
        .drop(columns=["_forecast_ts"])
        .reset_index(drop=True)
    )
    return combo.loc[:, PHASE7_FREEZE_FIELDS]


def _vx_phase7_attach_actuals(history):
    h = history.copy()
    fx = globals().get("fixtures_raw")
    if not isinstance(fx, pd.DataFrame) or fx.empty:
        return h
    id_col = "id" if "id" in fx.columns else ("fixture_id" if "fixture_id" in fx.columns else None)
    if id_col is None:
        return h
    score_h = "team_h_score" if "team_h_score" in fx.columns else None
    score_a = "team_a_score" if "team_a_score" in fx.columns else None
    if score_h is None or score_a is None:
        return h
    if "vx_completed" not in fx.columns:
        raise RuntimeError("Phase 7 actuals require canonical fixtures_raw.vx_completed from Cell 3.")
    finished_col = "vx_completed"
    actual_map = {}
    for _, r in fx.iterrows():
        try:
            fid = int(r[id_col])
        except Exception:
            continue
        finished = bool(r.get(finished_col, False)) if finished_col else False
        hg = pd.to_numeric(pd.Series([r.get(score_h)]), errors="coerce").iloc[0]
        ag = pd.to_numeric(pd.Series([r.get(score_a)]), errors="coerce").iloc[0]
        if not finished or pd.isna(hg) or pd.isna(ag):
            continue
        hg, ag = int(hg), int(ag)
        actual_map[fid] = {
            "actual_home_goals": hg,
            "actual_away_goals": ag,
            "final_result": ("HOME_WIN" if hg > ag else ("AWAY_WIN" if ag > hg else "DRAW")),
            "actual_home_clean_sheet": int(ag == 0),
            "actual_away_clean_sheet": int(hg == 0),
            "actual_clean_sheets": int(ag == 0) + int(hg == 0),
        }
    if not actual_map:
        return h
    for idx, row in h.iterrows():
        try:
            actual = actual_map.get(int(row["fixture_id"]))
        except Exception:
            actual = None
        if actual:
            for k, v in actual.items():
                h.at[idx, k] = v
    return h


def _vx_phase7_write_history(frame, path):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_suffix(path.suffix + ".tmp")
    frame.to_json(tmp, orient="records", lines=True, date_format="iso")
    tmp.replace(path)


# ----------------------------------------------------------------
# FREEZE CURRENT PRE-DEADLINE FORECASTS, THEN ATTACH FINISHED OUTCOMES.
# No hindsight forecast creation is allowed.
# ----------------------------------------------------------------
_phase7_existing = _vx_phase7_load_history(PHASE7_HISTORY_PATH)
_phase7_current = _vx_phase7_current_freeze_rows()
phase7_frozen_forecast_history = _vx_phase7_upsert_latest_predeadline(_phase7_existing, _phase7_current)
phase7_frozen_forecast_history = _vx_phase7_attach_actuals(phase7_frozen_forecast_history)
_vx_phase7_write_history(phase7_frozen_forecast_history, PHASE7_HISTORY_PATH)

# Compatibility name consumed by Phase 6 on the NEXT run. Phase 7 never invents
# earlier forecasts; it only exposes genuinely frozen rows from this archive.
phase7_validation_history = phase7_frozen_forecast_history.copy()
forecast_validation_history = phase7_frozen_forecast_history.copy()


def _vx_phase7_completed_history(frame):
    if frame is None or frame.empty:
        return pd.DataFrame(columns=PHASE7_FREEZE_FIELDS)
    h = frame.copy()
    numeric = (
        "fixture_id", "model_lambda_home", "model_lambda_away",
        "home_win_probability", "draw_probability", "away_win_probability",
        "home_cs_probability", "away_cs_probability", "attack_fdr_home",
        "defence_fdr_home", "attack_fdr_away", "defence_fdr_away",
        "model_confidence", "actual_home_goals", "actual_away_goals",
        "actual_home_clean_sheet", "actual_away_clean_sheet",
    )
    for c in numeric:
        if c in h.columns:
            h[c] = pd.to_numeric(h[c], errors="coerce")
    h["forecast_timestamp"] = pd.to_datetime(h["forecast_timestamp"], utc=True, errors="coerce")
    h["fpl_deadline_timestamp"] = pd.to_datetime(h["fpl_deadline_timestamp"], utc=True, errors="coerce")
    h = h.loc[h["is_pre_deadline_data"].fillna(False).astype(bool)].copy()
    h = h.loc[h["forecast_timestamp"] < h["fpl_deadline_timestamp"]].copy()
    h = h.dropna(subset=[
        "fixture_id", "model_lambda_home", "model_lambda_away",
        "actual_home_goals", "actual_away_goals",
    ]).copy()
    if h.empty:
        return h
    h["fixture_id"] = h["fixture_id"].astype(int)
    h["actual_home_goals"] = h["actual_home_goals"].astype(int)
    h["actual_away_goals"] = h["actual_away_goals"].astype(int)
    return h.sort_values(["forecast_timestamp", "fixture_id"]).reset_index(drop=True)


def _vx_phase7_poisson_pmf(k, lam):
    k = int(k)
    lam = float(lam)
    if k < 0 or not np.isfinite(lam) or lam <= 0:
        return 0.0
    return float(math.exp(-lam) * (lam ** k) / math.factorial(k))


def _vx_phase7_scoreline_probability(lh, la, hg, ag):
    return _vx_phase7_poisson_pmf(int(hg), float(lh)) * _vx_phase7_poisson_pmf(int(ag), float(la))


def _vx_phase7_poisson_deviance(y, mu):
    y = float(y)
    mu = max(1e-12, float(mu))
    if y <= 0:
        return float(2.0 * mu)
    return float(2.0 * (y * math.log(y / mu) - (y - mu)))


def _vx_phase7_fixture_wdl_from_lambdas(lh, la):
    lh, la = float(lh), float(la)
    max_goal = max(12, int(math.ceil(max(lh, la) + 8.0 * math.sqrt(max(lh, la, 1e-9)) + 8.0)))
    ph = np.array([_vx_phase7_poisson_pmf(k, lh) for k in range(max_goal + 1)], dtype=float)
    pa = np.array([_vx_phase7_poisson_pmf(k, la) for k in range(max_goal + 1)], dtype=float)
    matrix = np.outer(ph, pa)
    mass = float(matrix.sum())
    if mass <= 0:
        raise RuntimeError("Phase 7 WDL matrix has zero mass.")
    matrix /= mass
    return float(np.tril(matrix, -1).sum()), float(np.trace(matrix)), float(np.triu(matrix, 1).sum())


def _vx_phase7_row_wdl_target(row):
    hg, ag = int(row["actual_home_goals"]), int(row["actual_away_goals"])
    if hg > ag:
        return np.array([1.0, 0.0, 0.0])
    if hg == ag:
        return np.array([0.0, 1.0, 0.0])
    return np.array([0.0, 0.0, 1.0])


def _vx_phase7_baseline_metrics(frame):
    eps = 1e-15
    if frame is None or frame.empty:
        return {}
    dev, mae, score_ll, wdl_ll, wdl_brier, cs_brier = [], [], [], [], [], []
    for _, r in frame.iterrows():
        lh, la = float(r["model_lambda_home"]), float(r["model_lambda_away"])
        hg, ag = int(r["actual_home_goals"]), int(r["actual_away_goals"])
        dev.append((_vx_phase7_poisson_deviance(hg, lh) + _vx_phase7_poisson_deviance(ag, la)) / 2.0)
        mae.extend([abs(hg - lh), abs(ag - la)])
        p_score = _vx_phase7_scoreline_probability(lh, la, hg, ag)
        score_ll.append(-math.log(max(eps, min(1.0, p_score))))

        pred = np.array([
            float(r["home_win_probability"]),
            float(r["draw_probability"]),
            float(r["away_win_probability"]),
        ], dtype=float)
        pred = pred / pred.sum()
        target = _vx_phase7_row_wdl_target(r)
        actual_index = int(np.argmax(target))
        wdl_ll.append(-math.log(max(eps, min(1.0, pred[actual_index]))))
        wdl_brier.append(float(np.sum((pred - target) ** 2)))

        home_actual = float(int(ag == 0))
        away_actual = float(int(hg == 0))
        cs_brier.extend([
            (float(r["home_cs_probability"]) - home_actual) ** 2,
            (float(r["away_cs_probability"]) - away_actual) ** 2,
        ])
    return {
        "projected_goals_poisson_deviance": float(np.mean(dev)),
        "projected_goals_mae": float(np.mean(mae)),
        "projected_goals_scoreline_log_loss": float(np.mean(score_ll)),
        "wdl_log_loss": float(np.mean(wdl_ll)),
        "wdl_multiclass_brier": float(np.mean(wdl_brier)),
        "clean_sheet_brier": float(np.mean(cs_brier)),
    }


def _vx_phase7_cs_reliability(frame):
    rows = []
    if frame is None or frame.empty:
        return pd.DataFrame(columns=["bin_low", "bin_high", "sample_size", "mean_predicted_probability", "observed_clean_sheet_rate", "calibration_gap"])
    side_rows = []
    for _, r in frame.iterrows():
        side_rows.append((float(r["home_cs_probability"]), float(int(r["actual_away_goals"] == 0))))
        side_rows.append((float(r["away_cs_probability"]), float(int(r["actual_home_goals"] == 0))))
    d = pd.DataFrame(side_rows, columns=["pred", "actual"])
    d["pred"] = d["pred"].clip(0.0, 1.0)
    edges = np.linspace(0.0, 1.0, 11)
    d["bin"] = pd.cut(d["pred"], bins=edges, include_lowest=True, right=True, duplicates="drop")
    for interval, g in d.groupby("bin", observed=False):
        if g.empty:
            continue
        rows.append({
            "bin_low": float(interval.left),
            "bin_high": float(interval.right),
            "sample_size": int(len(g)),
            "mean_predicted_probability": float(g["pred"].mean()),
            "observed_clean_sheet_rate": float(g["actual"].mean()),
            "calibration_gap": float(g["pred"].mean() - g["actual"].mean()),
        })
    return pd.DataFrame(rows)


def _vx_phase7_fdr_band_validation(frame):
    columns = [
        "fdr_type", "colour_band", "sample_size", "average_fdr",
        "average_goals_for", "blank_rate", "average_goals_conceded", "clean_sheet_rate",
    ]
    if frame is None or frame.empty:
        return pd.DataFrame(columns=columns)
    band_fn = globals().get("vx_phase4_fdr_colour_band")
    if band_fn is None:
        raise RuntimeError("Phase 7 requires the Phase-4 FDR colour-band helper.")
    attack_rows, defence_rows = [], []
    for _, r in frame.iterrows():
        hg, ag = int(r["actual_home_goals"]), int(r["actual_away_goals"])
        attack_rows.extend([
            {"fdr": float(r["attack_fdr_home"]), "goals_for": hg},
            {"fdr": float(r["attack_fdr_away"]), "goals_for": ag},
        ])
        defence_rows.extend([
            {"fdr": float(r["defence_fdr_home"]), "goals_conceded": ag, "cs": int(ag == 0)},
            {"fdr": float(r["defence_fdr_away"]), "goals_conceded": hg, "cs": int(hg == 0)},
        ])
    out = []
    a = pd.DataFrame(attack_rows)
    a["colour_band"] = a["fdr"].map(band_fn)
    for band, g in a.groupby("colour_band", sort=False):
        out.append({
            "fdr_type": "ATTACK",
            "colour_band": band,
            "sample_size": int(len(g)),
            "average_fdr": float(g["fdr"].mean()),
            "average_goals_for": float(g["goals_for"].mean()),
            "blank_rate": float((g["goals_for"] == 0).mean()),
            "average_goals_conceded": pd.NA,
            "clean_sheet_rate": pd.NA,
        })
    d = pd.DataFrame(defence_rows)
    d["colour_band"] = d["fdr"].map(band_fn)
    for band, g in d.groupby("colour_band", sort=False):
        out.append({
            "fdr_type": "DEFENCE",
            "colour_band": band,
            "sample_size": int(len(g)),
            "average_fdr": float(g["fdr"].mean()),
            "average_goals_for": pd.NA,
            "blank_rate": pd.NA,
            "average_goals_conceded": float(g["goals_conceded"].mean()),
            "clean_sheet_rate": float(g["cs"].mean()),
        })
    return pd.DataFrame(out, columns=columns)


def _vx_phase7_confidence_validation(frame):
    columns = [
        "confidence_band", "sample_size", "mean_model_confidence",
        "mean_goal_mae", "mean_scoreline_log_loss", "mean_wdl_brier", "mean_cs_brier",
    ]
    if frame is None or frame.empty:
        return pd.DataFrame(columns=columns), "INSUFFICIENT_HISTORY"
    rows = []
    eps = 1e-15
    for _, r in frame.iterrows():
        lh, la = float(r["model_lambda_home"]), float(r["model_lambda_away"])
        hg, ag = int(r["actual_home_goals"]), int(r["actual_away_goals"])
        pred = np.array([float(r["home_win_probability"]), float(r["draw_probability"]), float(r["away_win_probability"])])
        pred = pred / pred.sum()
        target = _vx_phase7_row_wdl_target(r)
        rows.append({
            "confidence": float(r["model_confidence"]),
            "goal_mae": (abs(hg - lh) + abs(ag - la)) / 2.0,
            "score_ll": -math.log(max(eps, min(1.0, _vx_phase7_scoreline_probability(lh, la, hg, ag)))),
            "wdl_brier": float(np.sum((pred - target) ** 2)),
            "cs_brier": ((float(r["home_cs_probability"]) - float(int(ag == 0))) ** 2 + (float(r["away_cs_probability"]) - float(int(hg == 0))) ** 2) / 2.0,
        })
    d = pd.DataFrame(rows)
    d["confidence"] = d["confidence"].clip(0.0, 1.0)
    edges = np.linspace(0.0, 1.0, 6)
    d["confidence_band"] = pd.cut(d["confidence"], bins=edges, include_lowest=True, right=True)
    out = []
    for band, g in d.groupby("confidence_band", observed=False):
        if g.empty:
            continue
        out.append({
            "confidence_band": str(band),
            "sample_size": int(len(g)),
            "mean_model_confidence": float(g["confidence"].mean()),
            "mean_goal_mae": float(g["goal_mae"].mean()),
            "mean_scoreline_log_loss": float(g["score_ll"].mean()),
            "mean_wdl_brier": float(g["wdl_brier"].mean()),
            "mean_cs_brier": float(g["cs_brier"].mean()),
        })
    out_df = pd.DataFrame(out, columns=columns)
    if len(d) < PHASE7_MIN_COMPLETED_FIXTURES or d["confidence"].nunique() < 5:
        status = "INSUFFICIENT_SAMPLE_OR_CONFIDENCE_VARIATION"
    else:
        corr = float(d[["confidence", "goal_mae"]].corr(method="spearman").iloc[0, 1])
        status = "LOW_CONFIDENCE_FAILS_MORE" if np.isfinite(corr) and corr < 0 else "CONFIDENCE_NOT_YET_MONOTONIC"
    return out_df, status


def _vx_phase7_market_metrics(frame, prefix="market"):
    eps = 1e-15
    if frame is None or frame.empty:
        return {}
    ll, brier, cs_brier, over_brier, score_ll, goal_mae = [], [], [], [], [], []
    for _, r in frame.iterrows():
        p = np.array([
            float(r[f"{prefix}_home_win_probability"]),
            float(r[f"{prefix}_draw_probability"]),
            float(r[f"{prefix}_away_win_probability"]),
        ], dtype=float)
        if not np.all(np.isfinite(p)) or p.sum() <= 0:
            continue
        p /= p.sum()
        target = _vx_phase7_row_wdl_target(r)
        idx = int(np.argmax(target))
        ll.append(-math.log(max(eps, min(1.0, p[idx]))))
        brier.append(float(np.sum((p - target) ** 2)))
        hg, ag = int(r["actual_home_goals"]), int(r["actual_away_goals"])
        hcs = r.get(f"{prefix}_home_clean_sheet_probability", pd.NA)
        acs = r.get(f"{prefix}_away_clean_sheet_probability", pd.NA)
        if pd.notna(hcs) and pd.notna(acs):
            cs_brier.extend([
                (float(hcs) - float(int(ag == 0))) ** 2,
                (float(acs) - float(int(hg == 0))) ** 2,
            ])
        over = r.get(f"{prefix}_over_2_5_probability", pd.NA)
        if pd.notna(over):
            over_brier.append((float(over) - float(int(hg + ag >= 3))) ** 2)
        lh, la = r.get(f"{prefix}_lambda_home", pd.NA), r.get(f"{prefix}_lambda_away", pd.NA)
        if pd.notna(lh) and pd.notna(la) and float(lh) > 0 and float(la) > 0:
            score_ll.append(-math.log(max(eps, min(1.0, _vx_phase7_scoreline_probability(float(lh), float(la), hg, ag)))))
            goal_mae.extend([abs(hg - float(lh)), abs(ag - float(la))])
    return {
        "wdl_log_loss": float(np.mean(ll)) if ll else np.nan,
        "wdl_multiclass_brier": float(np.mean(brier)) if brier else np.nan,
        "clean_sheet_brier": float(np.mean(cs_brier)) if cs_brier else np.nan,
        "over_2_5_brier": float(np.mean(over_brier)) if over_brier else np.nan,
        "scoreline_log_loss": float(np.mean(score_ll)) if score_ll else np.nan,
        "goal_mae": float(np.mean(goal_mae)) if goal_mae else np.nan,
        "sample_size": int(len(frame)),
    }


def _vx_phase7_market_overlap(frame):
    if frame is None or frame.empty:
        return pd.DataFrame()
    h = frame.copy()
    needed = [
        "market_home_win_probability", "market_draw_probability", "market_away_win_probability",
        "market_lambda_home", "market_lambda_away",
    ]
    for c in needed:
        h[c] = pd.to_numeric(h[c], errors="coerce")
    return h.dropna(subset=needed).copy()


def _vx_phase7_blend_probabilities(lh_model, la_model, lh_market, la_market, alpha):
    lh = float(alpha) * float(lh_model) + (1.0 - float(alpha)) * float(lh_market)
    la = float(alpha) * float(la_model) + (1.0 - float(alpha)) * float(la_market)
    ph, pd_, pa = _vx_phase7_fixture_wdl_from_lambdas(lh, la)
    return lh, la, ph, pd_, pa


def _vx_phase7_blend_metrics(frame, alpha):
    eps = 1e-15
    score_ll, wdl_ll, brier = [], [], []
    for _, r in frame.iterrows():
        lh, la, ph, pd_, pa = _vx_phase7_blend_probabilities(
            r["model_lambda_home"], r["model_lambda_away"],
            r["market_lambda_home"], r["market_lambda_away"], alpha,
        )
        hg, ag = int(r["actual_home_goals"]), int(r["actual_away_goals"])
        score_ll.append(-math.log(max(eps, min(1.0, _vx_phase7_scoreline_probability(lh, la, hg, ag)))))
        pred = np.array([ph, pd_, pa])
        target = _vx_phase7_row_wdl_target(r)
        idx = int(np.argmax(target))
        wdl_ll.append(-math.log(max(eps, min(1.0, pred[idx]))))
        brier.append(float(np.sum((pred - target) ** 2)))
    return {
        "scoreline_log_loss": float(np.mean(score_ll)),
        "wdl_log_loss": float(np.mean(wdl_ll)),
        "multiclass_brier": float(np.mean(brier)),
    }


def _vx_phase7_market_blend_experiment(frame):
    columns = [
        "alpha_model_weight", "split", "sample_size", "scoreline_log_loss",
        "wdl_log_loss", "multiclass_brier",
    ]
    h = _vx_phase7_market_overlap(frame)
    minimum = PHASE7_MARKET_MIN_TRAIN_FIXTURES + PHASE7_MARKET_MIN_VALIDATION_FIXTURES
    if len(h) < minimum:
        return pd.DataFrame(columns=columns), {
            "status": f"WAITING_FOR_MARKET_HISTORY: need {minimum}, have {len(h)}",
            "selected_alpha": None,
            "validation_approved": False,
        }
    h = h.sort_values(["forecast_timestamp", "fixture_id"]).reset_index(drop=True)
    split = int(math.floor(len(h) * PHASE7_MARKET_TRAIN_FRACTION))
    split = max(PHASE7_MARKET_MIN_TRAIN_FIXTURES, split)
    split = min(split, len(h) - PHASE7_MARKET_MIN_VALIDATION_FIXTURES)
    train, valid = h.iloc[:split].copy(), h.iloc[split:].copy()
    rows, train_scores = [], []
    for a in PHASE7_BLEND_ALPHAS:
        mt = _vx_phase7_blend_metrics(train, a)
        mv = _vx_phase7_blend_metrics(valid, a)
        rows.append({"alpha_model_weight": float(a), "split": "TRAIN", "sample_size": int(len(train)), **mt})
        rows.append({"alpha_model_weight": float(a), "split": "HOLDOUT", "sample_size": int(len(valid)), **mv})
        train_scores.append((mt["scoreline_log_loss"], abs(1.0 - float(a)), float(a)))
    train_scores.sort(key=lambda x: (x[0], x[1], -x[2]))
    selected = float(train_scores[0][2])
    out = pd.DataFrame(rows, columns=columns)
    hold = out.loc[out["split"] == "HOLDOUT"].set_index("alpha_model_weight")
    chosen = hold.loc[selected]
    model_only = hold.loc[1.0]
    approved = bool(
        selected < 1.0
        and float(chosen["scoreline_log_loss"]) < float(model_only["scoreline_log_loss"])
        and float(chosen["wdl_log_loss"]) <= float(model_only["wdl_log_loss"]) + 1e-4
        and float(chosen["multiclass_brier"]) <= float(model_only["multiclass_brier"]) + 1e-4
    )
    result = {
        "status": "HOLDOUT_COMPLETE",
        "selected_alpha": selected,
        "validation_approved": approved,
        "train_fixture_count": int(len(train)),
        "holdout_fixture_count": int(len(valid)),
        "holdout_scoreline_log_loss_gain_vs_model": float(model_only["scoreline_log_loss"] - chosen["scoreline_log_loss"]),
        "holdout_wdl_log_loss_gain_vs_model": float(model_only["wdl_log_loss"] - chosen["wdl_log_loss"]),
        "holdout_multiclass_brier_gain_vs_model": float(model_only["multiclass_brier"] - chosen["multiclass_brier"]),
        "live_blend_activated": False,
    }
    return out, result


# ----------------------------------------------------------------
# VALIDATION REPORTS — only completed, genuinely frozen forecasts qualify.
# ----------------------------------------------------------------
phase7_completed_forecasts = _vx_phase7_completed_history(phase7_frozen_forecast_history)
_phase7_n = int(len(phase7_completed_forecasts))
_phase7_baseline = _vx_phase7_baseline_metrics(phase7_completed_forecasts)

phase7_validation_metrics = pd.DataFrame([
    {
        "metric_group": "BASELINE_MODEL",
        "metric_name": name,
        "value": value,
        "sample_size": _phase7_n,
        "model_version": str(globals().get("PHASE4_OUTPUT_MODEL_VERSION", "VORTEX-V1-PHASE4-POISSON-CS-FDR")),
        "calculated_at": PHASE7_TIMESTAMP,
    }
    for name, value in _phase7_baseline.items()
], columns=PHASE7_METRIC_FIELDS)

phase7_cs_reliability = _vx_phase7_cs_reliability(phase7_completed_forecasts)
phase7_fdr_band_validation = _vx_phase7_fdr_band_validation(phase7_completed_forecasts)
phase7_confidence_validation, PHASE7_CONFIDENCE_STATUS = _vx_phase7_confidence_validation(phase7_completed_forecasts)

# Market benchmark: compare only completed forecasts that had a deadline-safe
# market snapshot frozen at the same time. No odds are fed back into V1 outputs.
_phase7_market = _vx_phase7_market_overlap(phase7_completed_forecasts)
_phase7_model_market_frame = _phase7_market.copy()
if not _phase7_model_market_frame.empty:
    _phase7_model_market_frame["model_home_win_probability"] = _phase7_model_market_frame["home_win_probability"]
    _phase7_model_market_frame["model_draw_probability"] = _phase7_model_market_frame["draw_probability"]
    _phase7_model_market_frame["model_away_win_probability"] = _phase7_model_market_frame["away_win_probability"]
    _phase7_model_market_frame["model_home_clean_sheet_probability"] = _phase7_model_market_frame["home_cs_probability"]
    _phase7_model_market_frame["model_away_clean_sheet_probability"] = _phase7_model_market_frame["away_cs_probability"]
    _phase7_model_market_frame["model_lambda_home"] = _phase7_model_market_frame["model_lambda_home"]
    _phase7_model_market_frame["model_lambda_away"] = _phase7_model_market_frame["model_lambda_away"]

_phase7_model_market_metrics = _vx_phase7_market_metrics(_phase7_model_market_frame, prefix="model") if not _phase7_model_market_frame.empty else {}
_phase7_market_metrics = _vx_phase7_market_metrics(_phase7_model_market_frame, prefix="market") if not _phase7_model_market_frame.empty else {}
phase7_model_vs_market_validation = pd.DataFrame([
    {"source": "PURE_MODEL", **_phase7_model_market_metrics},
    {"source": "PURE_MARKET", **_phase7_market_metrics},
]) if _phase7_model_market_frame is not None and not _phase7_model_market_frame.empty else pd.DataFrame(
    columns=["source", "wdl_log_loss", "wdl_multiclass_brier", "clean_sheet_brier", "over_2_5_brier", "scoreline_log_loss", "goal_mae", "sample_size"]
)

phase7_market_blend_experiment, PHASE7_MARKET_BLEND_RESULT = _vx_phase7_market_blend_experiment(phase7_completed_forecasts)
PHASE7_MARKET_BLEND_STATUS = str(PHASE7_MARKET_BLEND_RESULT.get("status", "NOT_RUN"))

# Refresh the Phase-6 comparison evidence using the newly updated frozen archive,
# but do NOT mutate the already-selected score model in this run. Phase 6 will
# consume this same persistent history at the start of the next clean Run all.
phase7_dixon_coles_validation_refresh = pd.DataFrame()
if "_vx_phase6_fit_and_validate" in globals():
    try:
        _p7_dc_result, _p7_dc_reason = _vx_phase6_fit_and_validate(phase7_completed_forecasts)
        if _p7_dc_result is not None:
            phase7_dixon_coles_validation_refresh = pd.DataFrame([_p7_dc_result])
    except Exception as exc:
        phase7_dixon_coles_validation_refresh = pd.DataFrame([{
            "validation_approved": False,
            "selection_reason": f"PHASE7_REFRESH_ERROR:{type(exc).__name__}",
        }])

# Evidence is considered mature enough for review at 80 completed frozen
# fixtures, matching the 60-train + 20-holdout minimum used by Phase 6.
PHASE7_VALIDATION_EVIDENCE_READY = bool(_phase7_n >= PHASE7_MIN_COMPLETED_FIXTURES)
PHASE7_VALIDATION_STATUS = (
    "EVIDENCE_READY_FOR_REVIEW"
    if PHASE7_VALIDATION_EVIDENCE_READY
    else f"ACCUMULATING_FROZEN_HISTORY:{_phase7_n}/{PHASE7_MIN_COMPLETED_FIXTURES}"
)

# Phase 7 deliberately does NOT auto-unlock the player model. The validation
# evidence must first be reviewed as a whole; this prevents one metric or one
# short run from silently promoting the downstream player/decision layers.
PHASE7_READY_FOR_PLAYER_MODEL = False

# ----------------------------------------------------------------
# SAVE CURRENT-RUN VALIDATION OUTPUTS + PERSISTENT SUMMARY
# ----------------------------------------------------------------
PHASE7_OUTPUT_DIR = Path(DATA_DIR)
PHASE7_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
PHASE7_HISTORY_EXPORT_PATH = PHASE7_OUTPUT_DIR / "phase7_frozen_forecast_history.json"
PHASE7_METRICS_PATH = PHASE7_OUTPUT_DIR / "phase7_validation_metrics.json"
PHASE7_CS_RELIABILITY_PATH = PHASE7_OUTPUT_DIR / "phase7_clean_sheet_reliability.json"
PHASE7_FDR_VALIDATION_PATH = PHASE7_OUTPUT_DIR / "phase7_fdr_band_validation.json"
PHASE7_CONFIDENCE_PATH = PHASE7_OUTPUT_DIR / "phase7_model_confidence_validation.json"
PHASE7_MARKET_COMPARE_PATH = PHASE7_OUTPUT_DIR / "phase7_model_vs_market_validation.json"
PHASE7_MARKET_BLEND_PATH = PHASE7_OUTPUT_DIR / "phase7_market_blend_experiment.json"
PHASE7_DC_REFRESH_PATH = PHASE7_OUTPUT_DIR / "phase7_dixon_coles_validation_refresh.json"

phase7_frozen_forecast_history.to_json(PHASE7_HISTORY_EXPORT_PATH, orient="records", date_format="iso", indent=2)
phase7_validation_metrics.to_json(PHASE7_METRICS_PATH, orient="records", date_format="iso", indent=2)
phase7_cs_reliability.to_json(PHASE7_CS_RELIABILITY_PATH, orient="records", date_format="iso", indent=2)
phase7_fdr_band_validation.to_json(PHASE7_FDR_VALIDATION_PATH, orient="records", date_format="iso", indent=2)
phase7_confidence_validation.to_json(PHASE7_CONFIDENCE_PATH, orient="records", date_format="iso", indent=2)
phase7_model_vs_market_validation.to_json(PHASE7_MARKET_COMPARE_PATH, orient="records", date_format="iso", indent=2)
phase7_market_blend_experiment.to_json(PHASE7_MARKET_BLEND_PATH, orient="records", date_format="iso", indent=2)
phase7_dixon_coles_validation_refresh.to_json(PHASE7_DC_REFRESH_PATH, orient="records", date_format="iso", indent=2)

# Reliability chart is generated only when realised frozen forecasts exist.
PHASE7_CS_RELIABILITY_CHART_PATH = PHASE7_OUTPUT_DIR / "phase7_clean_sheet_reliability.png"
if not phase7_cs_reliability.empty:
    try:
        import matplotlib.pyplot as plt
        fig, ax = plt.subplots(figsize=(7, 6))
        ax.plot([0, 1], [0, 1], linestyle="--", linewidth=1.2)
        ax.plot(
            phase7_cs_reliability["mean_predicted_probability"],
            phase7_cs_reliability["observed_clean_sheet_rate"],
            marker="o",
            linewidth=1.8,
        )
        ax.set_xlabel("Predicted clean-sheet probability")
        ax.set_ylabel("Observed clean-sheet rate")
        ax.set_title("VORTEX clean-sheet reliability")
        ax.set_xlim(0, 1)
        ax.set_ylim(0, 1)
        fig.tight_layout()
        fig.savefig(PHASE7_CS_RELIABILITY_CHART_PATH, dpi=160)
        plt.close(fig)
    except Exception:
        pass

_phase7_summary = {
    "phase7_model_version": PHASE7_MODEL_VERSION,
    "calculated_at": PHASE7_TIMESTAMP,
    "frozen_forecast_rows": int(len(phase7_frozen_forecast_history)),
    "completed_frozen_fixtures": _phase7_n,
    "validation_status": PHASE7_VALIDATION_STATUS,
    "validation_evidence_ready": bool(PHASE7_VALIDATION_EVIDENCE_READY),
    "ready_for_player_model": False,
    "baseline_metrics": _phase7_baseline,
    "confidence_status": PHASE7_CONFIDENCE_STATUS,
    "market_overlap_completed_fixtures": int(len(_phase7_market)),
    "market_blend_status": PHASE7_MARKET_BLEND_STATUS,
    "market_blend_result": PHASE7_MARKET_BLEND_RESULT,
    "market_blend_active": False,
    "dixon_coles_live_state_mutated": False,
}
PHASE7_SUMMARY_PATH.write_text(json.dumps(_phase7_summary, ensure_ascii=False, indent=2, default=str), encoding="utf-8")

# ----------------------------------------------------------------
# HARD NO-MUTATION + CLEAN-ROOM CHECKS
# ----------------------------------------------------------------
_phase7_upstream_after = {
    "projected_goals_model_only": _vx_phase7_frame_digest(globals().get("projected_goals_model_only")),
    "fixture_probabilities_model_only": _vx_phase7_frame_digest(fixture_probabilities_model_only),
    "poisson_probability_matrix": _vx_phase7_frame_digest(globals().get("poisson_probability_matrix")),
    "fixture_fdr_model_only": _vx_phase7_frame_digest(fixture_fdr_model_only),
    "market_odds_benchmark": _vx_phase7_frame_digest(globals().get("market_odds_benchmark")),
    "dixon_coles_probability_matrix": _vx_phase7_frame_digest(globals().get("dixon_coles_probability_matrix")),
}
if _phase7_upstream_before != _phase7_upstream_after:
    raise RuntimeError("Phase 7 clean-room violation: an upstream model output was mutated.")

if "vx_phase0_assert_cleanroom" in globals():
    vx_phase0_assert_cleanroom("after Phase 7 validation")

PHASE7_READY = True
print("✅ PHASE 7 DEADLINE-SAFE VALIDATION ENGINE BUILT")
print(f"   Frozen forecast rows: {_phase7_summary['frozen_forecast_rows']}")
print(f"   Completed frozen fixtures: {_phase7_n}")
print(f"   Validation status: {PHASE7_VALIDATION_STATUS}")
print(f"   Model-confidence status: {PHASE7_CONFIDENCE_STATUS}")
print(f"   Completed market-overlap fixtures: {len(_phase7_market)}")
print(f"   Market-blend experiment: {PHASE7_MARKET_BLEND_STATUS}")
print("🔒 No live market blend was activated.")
print("🔒 Phase 6 live selection was not mutated in this run.")
print("🔒 Player expected-points, captaincy, transfers and squad decisions remain blocked.")
print("📦 Persistent frozen history:", PHASE7_HISTORY_PATH)
print("📊 Validation summary:", PHASE7_SUMMARY_PATH)
```

```python
# ============================================================
# CELL 13B — PHASE 8
# MARKET + TEAM ELO EXTERNAL BENCHMARKS
# BENCHMARK-ONLY • ZERO BLEND WEIGHTS • NO PLAYER EXPECTED POINTS
# ============================================================

from pathlib import Path
import hashlib
import json
import math
import re
import numpy as np
import pandas as pd

# ----------------------------------------------------------------
# HARD POLICY — EXTERNAL BENCHMARKS DO NOT TOUCH V1 MODEL OUTPUTS
# ----------------------------------------------------------------
MARKET_BLEND_WEIGHT = 0.00
ELO_BLEND_WEIGHT = 0.00
PHASE8_BLEND_ACTIVE = False
PHASE8_VERSION = "VORTEX-V1-PHASE8-EXTERNAL-BENCHMARK"
PHASE8_FPL_CORE_ROOT = Path(FPL_VORTEX_DATA_ROOT) / "00_SOURCE_ARCHIVE" / "FPL_CORE" / "data"

if MARKET_BLEND_WEIGHT != 0.00:
    raise RuntimeError("Phase 8 policy violation: MARKET_BLEND_WEIGHT must remain 0.00.")
if ELO_BLEND_WEIGHT != 0.00:
    raise RuntimeError("Phase 8 policy violation: ELO_BLEND_WEIGHT must remain 0.00.")
if PHASE8_BLEND_ACTIVE:
    raise RuntimeError("Phase 8 policy violation: external benchmark blending is disabled.")

_REQUIRED_MODEL_FRAMES = (
    "projected_goals_model_only",
    "fixture_probabilities_model_only",
    "fixture_fdr_model_only",
)

for _name in _REQUIRED_MODEL_FRAMES:
    _frame = globals().get(_name)
    if not isinstance(_frame, pd.DataFrame) or _frame.empty:
        raise RuntimeError(f"Phase 8 requires non-empty {_name} from Phases 3–4.")

def _vx_phase8_frame_digest(frame):
    """Stable digest used only to prove Phase 8 did not alter model-only state."""
    if not isinstance(frame, pd.DataFrame):
        return "NOT_A_DATAFRAME"
    tmp = frame.copy()
    tmp = tmp.reindex(sorted(tmp.columns), axis=1)
    payload = tmp.to_json(orient="split", date_format="iso", double_precision=15)
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()

PHASE8_MODEL_STATE_BEFORE = {
    name: _vx_phase8_frame_digest(globals()[name])
    for name in _REQUIRED_MODEL_FRAMES
}

# ----------------------------------------------------------------
# OFFICIAL TEAM-ID → FPL CORE TEAM-CODE MAP
# FPL Core match tables use the official FPL team code, not FPL team id.
# ----------------------------------------------------------------
if not isinstance(globals().get("teams"), pd.DataFrame) or globals()["teams"].empty:
    raise RuntimeError("Phase 8 requires the official FPL teams table.")

_team_meta = globals()["teams"].copy()
for _col in ("id", "code", "name"):
    if _col not in _team_meta.columns:
        raise RuntimeError(f"Phase 8 teams table missing required column: {_col}")

_team_meta["id"] = pd.to_numeric(_team_meta["id"], errors="raise").astype(int)
_team_meta["code"] = pd.to_numeric(_team_meta["code"], errors="raise").astype(int)
_PHASE8_TEAM_ID_TO_CODE = _team_meta.set_index("id")["code"].to_dict()
_PHASE8_TEAM_ID_TO_NAME = _team_meta.set_index("id")["name"].astype(str).to_dict()

# ----------------------------------------------------------------
# SOURCE SNAPSHOTS
# - Previous seasons: final GW snapshot where available.
# - Current season: latest gameweek snapshot + GW0 if available.
# - Completed rows are filtered to kickoff < forecast_timestamp.
# ----------------------------------------------------------------
def _vx_phase8_season_dir(start_year):
    return PHASE8_FPL_CORE_ROOT / f"{int(start_year)}-{int(start_year)+1}"

def _vx_phase8_gw_number(path):
    m = re.search(r"GW(\d+)$", path.parent.name, flags=re.I)
    return int(m.group(1)) if m else -1

def _vx_phase8_pick_snapshot_files(season_start, current_start, current_gw):
    season_dir = _vx_phase8_season_dir(season_start)
    by_gw = season_dir / "By Gameweek"
    if not by_gw.exists():
        return []

    candidates = list(by_gw.glob("GW*/matches.csv"))
    if not candidates:
        return []

    by_num = {}
    for p in candidates:
        n = _vx_phase8_gw_number(p)
        if n >= 0:
            by_num[n] = p

    if int(season_start) < int(current_start):
        # Final cumulative snapshot is enough for a completed past season.
        if 38 in by_num:
            return [by_num[38]]
        return [by_num[max(by_num)]]

    # Current season: GW0 may contain preseason evidence; current GW snapshot
    # contains the latest available season state. We still filter strictly by
    # kickoff timestamp before using a completed match.
    selected = []
    if 0 in by_num:
        selected.append(by_num[0])

    eligible = [n for n in by_num if n <= max(1, int(current_gw))]
    if eligible:
        latest = by_num[max(eligible)]
        if latest not in selected:
            selected.append(latest)
    return selected

def _vx_phase8_read_match_snapshot(path):
    try:
        frame = pd.read_csv(path)
    except Exception as exc:
        raise RuntimeError(
            f"Phase 8 cannot read FPL Core snapshot {path}: "
            f"{type(exc).__name__}: {exc}"
        )
    required = {
        "kickoff_time", "home_team", "away_team",
        "home_team_elo", "away_team_elo",
        "home_score", "away_score", "finished", "match_id",
    }
    missing = sorted(required - set(frame.columns))
    if missing:
        raise RuntimeError(
            f"Phase 8 FPL Core snapshot {path} missing columns: {missing}"
        )
    frame = frame.copy()
    frame["_phase8_source_path"] = str(path)
    frame["_phase8_snapshot_gw"] = _vx_phase8_gw_number(path)
    return frame

def _vx_phase8_load_elo_history():
    forecast_ts = pd.to_datetime(
        globals().get("forecast_timestamp"), utc=True, errors="coerce"
    )
    if pd.isna(forecast_ts):
        raise RuntimeError("Phase 8 requires a valid forecast_timestamp.")

    current_label = str(globals().get("CURRENT_SEASON", ""))
    m = re.match(r"^(\d{4})-", current_label)
    if not m:
        raise RuntimeError("Phase 8 cannot resolve CURRENT_SEASON.")
    current_start = int(m.group(1))
    _gw_num = pd.to_numeric(globals().get("gameweek", 1), errors="coerce")
    current_gw = int(_gw_num) if pd.notna(_gw_num) and float(_gw_num) >= 0 else 1

    frames = []
    # Three completed seasons + current season gives enough calibration depth
    # without turning Elo into a model input.
    for season_start in range(current_start - 3, current_start + 1):
        for p in _vx_phase8_pick_snapshot_files(
            season_start, current_start, current_gw
        ):
            try:
                f = _vx_phase8_read_match_snapshot(p)
                f["_phase8_season_start"] = season_start
                frames.append(f)
            except Exception as exc:
                print(f"⚠️ Phase 8 Elo snapshot skipped: {exc}")

    if not frames:
        return pd.DataFrame()

    hist = pd.concat(frames, ignore_index=True, sort=False)
    hist["_kickoff"] = pd.to_datetime(hist["kickoff_time"], utc=True, errors="coerce")
    hist = hist.loc[hist["_kickoff"].notna() & (hist["_kickoff"] < forecast_ts)].copy()

    # Normalize boolean semantics.
    finished = hist["finished"]
    if finished.dtype == bool:
        hist["_finished"] = finished
    else:
        hist["_finished"] = (
            finished.astype(str).str.strip().str.lower()
            .isin({"true", "1", "yes"})
        )

    # A cumulative snapshot can repeat the same match. Keep the latest source GW.
    hist["_phase8_snapshot_gw"] = pd.to_numeric(
        hist["_phase8_snapshot_gw"], errors="coerce"
    ).fillna(-1)
    hist = hist.sort_values(["_kickoff", "_phase8_snapshot_gw"])
    hist = hist.drop_duplicates(subset=["match_id"], keep="last").reset_index(drop=True)
    return hist

phase8_elo_history_raw = _vx_phase8_load_elo_history()

# ----------------------------------------------------------------
# EMPIRICAL TEAM-ELO → 1X2 BENCHMARK CALIBRATION
# This does NOT alter team strengths or lambdas.
# It learns only the mapping from historical Team Elo difference to H/D/A.
# ----------------------------------------------------------------
def _vx_phase8_is_premier_league(frame):
    if "tournament" not in frame.columns:
        return pd.Series(False, index=frame.index)
    s = frame["tournament"].astype(str).str.strip().str.lower()
    return s.eq("prem") | s.str.contains("premier league", regex=False)

def _vx_phase8_prepare_elo_calibration(frame):
    if frame.empty:
        return pd.DataFrame()

    f = frame.loc[frame["_finished"] & _vx_phase8_is_premier_league(frame)].copy()
    numeric_cols = (
        "home_team_elo", "away_team_elo", "home_score", "away_score"
    )
    for c in numeric_cols:
        f[c] = pd.to_numeric(f[c], errors="coerce")
    f = f.dropna(subset=list(numeric_cols)).copy()
    if f.empty:
        return f

    f["elo_diff_scaled"] = (
        f["home_team_elo"].astype(float) - f["away_team_elo"].astype(float)
    ) / 400.0

    f["result_class"] = np.where(
        f["home_score"] > f["away_score"], 0,
        np.where(f["home_score"] == f["away_score"], 1, 2),
    ).astype(int)
    return f

def _vx_phase8_softmax(z):
    z = np.asarray(z, dtype=float)
    z = z - np.max(z, axis=1, keepdims=True)
    ez = np.exp(z)
    return ez / ez.sum(axis=1, keepdims=True)

def _vx_phase8_fit_elo_probability_map(calibration):
    minimum = 120
    if len(calibration) < minimum:
        return {
            "status": f"WAITING_FOR_ELO_CALIBRATION_HISTORY: need {minimum}, have {len(calibration)}",
            "coefficients": None,
            "sample_size": int(len(calibration)),
            "iterations": 0,
            "loss": np.nan,
        }

    x = calibration["elo_diff_scaled"].to_numpy(dtype=float)
    X = np.column_stack([np.ones(len(x)), x])
    y = calibration["result_class"].to_numpy(dtype=int)
    if set(np.unique(y).tolist()) != {0, 1, 2}:
        return {
            "status": "ELO_CALIBRATION_MISSING_ONE_OR_MORE_RESULT_CLASSES",
            "coefficients": None,
            "sample_size": int(len(calibration)),
            "iterations": 0,
            "loss": np.nan,
        }
    Y = np.eye(3, dtype=float)[y]

    B = np.zeros((2, 3), dtype=float)
    learning_rate = 0.08
    l2 = 0.015
    max_iter = 5000
    tol = 1e-9
    prev_loss = np.inf

    for iteration in range(1, max_iter + 1):
        P = _vx_phase8_softmax(X @ B)
        eps = 1e-15
        loss = -float(np.mean(np.sum(Y * np.log(np.clip(P, eps, 1.0)), axis=1)))
        # Penalize slope only; home/draw base rates are learned freely.
        loss += 0.5 * l2 * float(np.sum(B[1:, :] ** 2))

        grad = (X.T @ (P - Y)) / len(X)
        grad[1:, :] += l2 * B[1:, :]
        B -= learning_rate * grad

        if abs(prev_loss - loss) < tol:
            break
        prev_loss = loss

    P = _vx_phase8_softmax(X @ B)
    final_loss = -float(
        np.mean(np.sum(Y * np.log(np.clip(P, 1e-15, 1.0)), axis=1))
    )
    return {
        "status": "READY",
        "coefficients": B,
        "sample_size": int(len(calibration)),
        "iterations": int(iteration),
        "loss": final_loss,
    }

def _vx_phase8_predict_elo_1x2(home_elo, away_elo, fit):
    if fit.get("status") != "READY" or fit.get("coefficients") is None:
        return (np.nan, np.nan, np.nan)
    h = float(home_elo)
    a = float(away_elo)
    if not (np.isfinite(h) and np.isfinite(a)):
        return (np.nan, np.nan, np.nan)
    x = np.array([[1.0, (h - a) / 400.0]], dtype=float)
    p = _vx_phase8_softmax(x @ fit["coefficients"])[0]
    return tuple(float(v) for v in p)

phase8_elo_calibration_history = _vx_phase8_prepare_elo_calibration(
    phase8_elo_history_raw
)
PHASE8_ELO_FIT = _vx_phase8_fit_elo_probability_map(
    phase8_elo_calibration_history
)

# ----------------------------------------------------------------
# LATEST PRE-FORECAST TEAM ELO OBSERVATION
# Use the latest completed-match Team Elo observation before forecast.
# Missing stays missing; Phase 8 never fabricates a promoted-team Elo.
# ----------------------------------------------------------------
def _vx_phase8_latest_team_elo(frame):
    if frame.empty:
        return pd.DataFrame(
            columns=["team_code", "team_elo", "elo_observed_timestamp", "elo_source"]
        )

    finished = frame.loc[frame["_finished"]].copy()
    rows = []
    for side in ("home", "away"):
        team_col = f"{side}_team"
        elo_col = f"{side}_team_elo"
        if team_col not in finished.columns or elo_col not in finished.columns:
            continue
        t = pd.DataFrame({
            "team_code": pd.to_numeric(finished[team_col], errors="coerce"),
            "team_elo": pd.to_numeric(finished[elo_col], errors="coerce"),
            "elo_observed_timestamp": finished["_kickoff"],
            "elo_source": "LATEST_COMPLETED_MATCH_TEAM_ELO",
        })
        rows.append(t)

    if not rows:
        return pd.DataFrame()

    obs = pd.concat(rows, ignore_index=True)
    obs = obs.dropna(subset=["team_code", "team_elo", "elo_observed_timestamp"]).copy()
    obs["team_code"] = obs["team_code"].astype(int)
    obs = obs.sort_values("elo_observed_timestamp")
    return obs.groupby("team_code", as_index=False).tail(1).reset_index(drop=True)

phase8_latest_team_elo = _vx_phase8_latest_team_elo(phase8_elo_history_raw)

# ----------------------------------------------------------------
# CURRENT MODEL FIXTURES + TEAM ELO BENCHMARK
# ----------------------------------------------------------------
_phase8_model = globals()["fixture_probabilities_model_only"].copy()
for c in ("fixture_id", "home_team_id", "away_team_id"):
    _phase8_model[c] = pd.to_numeric(_phase8_model[c], errors="raise").astype(int)

_phase8_model["home_team_code"] = _phase8_model["home_team_id"].map(
    _PHASE8_TEAM_ID_TO_CODE
)
_phase8_model["away_team_code"] = _phase8_model["away_team_id"].map(
    _PHASE8_TEAM_ID_TO_CODE
)
_phase8_model["home_team"] = _phase8_model["home_team_id"].map(
    _PHASE8_TEAM_ID_TO_NAME
)
_phase8_model["away_team"] = _phase8_model["away_team_id"].map(
    _PHASE8_TEAM_ID_TO_NAME
)

if _phase8_model[["home_team_code", "away_team_code"]].isna().any().any():
    raise RuntimeError("Phase 8 cannot map one or more FPL team ids to FPL Core team codes.")

_elo_lookup = (
    phase8_latest_team_elo.set_index("team_code")
    if not phase8_latest_team_elo.empty
    else pd.DataFrame()
)

_elo_rows = []
_forecast_ts = pd.to_datetime(
    globals().get("forecast_timestamp"), utc=True, errors="coerce"
)

for _, r in _phase8_model.iterrows():
    hc = int(r["home_team_code"])
    ac = int(r["away_team_code"])

    home_elo = np.nan
    away_elo = np.nan
    home_obs = pd.NaT
    away_obs = pd.NaT

    if isinstance(_elo_lookup, pd.DataFrame) and not _elo_lookup.empty:
        if hc in _elo_lookup.index:
            rr = _elo_lookup.loc[hc]
            home_elo = float(rr["team_elo"])
            home_obs = pd.to_datetime(
                rr["elo_observed_timestamp"], utc=True, errors="coerce"
            )
        if ac in _elo_lookup.index:
            rr = _elo_lookup.loc[ac]
            away_elo = float(rr["team_elo"])
            away_obs = pd.to_datetime(
                rr["elo_observed_timestamp"], utc=True, errors="coerce"
            )

    p_home, p_draw, p_away = _vx_phase8_predict_elo_1x2(
        home_elo, away_elo, PHASE8_ELO_FIT
    )

    elo_status = "READY"
    if PHASE8_ELO_FIT.get("status") != "READY":
        elo_status = str(PHASE8_ELO_FIT.get("status"))
    elif not (np.isfinite(home_elo) and np.isfinite(away_elo)):
        elo_status = "MISSING_TEAM_ELO"

    oldest_obs = pd.NaT
    if not pd.isna(home_obs) and not pd.isna(away_obs):
        oldest_obs = min(home_obs, away_obs)
    elo_age_days = (
        float((_forecast_ts - oldest_obs).total_seconds() / 86400.0)
        if not pd.isna(_forecast_ts) and not pd.isna(oldest_obs)
        else np.nan
    )

    _elo_rows.append({
        "fixture_id": int(r["fixture_id"]),
        "gameweek": int(r["gameweek"]),
        "home_team_id": int(r["home_team_id"]),
        "away_team_id": int(r["away_team_id"]),
        "home_team": str(r["home_team"]),
        "away_team": str(r["away_team"]),
        "home_team_code": hc,
        "away_team_code": ac,
        "team_elo_home": home_elo,
        "team_elo_away": away_elo,
        "team_elo_difference_home_minus_away": (
            float(home_elo - away_elo)
            if np.isfinite(home_elo) and np.isfinite(away_elo)
            else np.nan
        ),
        "team_elo_home_win_probability": p_home,
        "team_elo_draw_probability": p_draw,
        "team_elo_away_win_probability": p_away,
        "elo_home_observed_timestamp": (
            home_obs.isoformat() if not pd.isna(home_obs) else pd.NA
        ),
        "elo_away_observed_timestamp": (
            away_obs.isoformat() if not pd.isna(away_obs) else pd.NA
        ),
        "elo_rating_age_days": elo_age_days,
        "elo_calibration_sample_size": int(PHASE8_ELO_FIT.get("sample_size", 0)),
        "elo_calibration_log_loss": PHASE8_ELO_FIT.get("loss", np.nan),
        "elo_calibration_status": PHASE8_ELO_FIT.get("status"),
        "elo_benchmark_status": elo_status,
        "elo_source_name": "FPL_CORE_CLUBELO_TEAM_RATINGS",
        "elo_probability_mapping": "EMPIRICAL_1X2_SOFTMAX_FROM_HISTORICAL_TEAM_ELO_DIFFERENCE",
        "forecast_timestamp": globals().get("forecast_timestamp"),
        "model_version": globals().get("model_version"),
        "phase8_version": PHASE8_VERSION,
    })

team_elo_benchmark = pd.DataFrame(_elo_rows)

if not team_elo_benchmark.empty:
    ready = team_elo_benchmark["elo_benchmark_status"].eq("READY")
    if ready.any():
        s = (
            team_elo_benchmark.loc[
                ready,
                [
                    "team_elo_home_win_probability",
                    "team_elo_draw_probability",
                    "team_elo_away_win_probability",
                ],
            ].sum(axis=1)
        )
        if not np.allclose(s.to_numpy(dtype=float), 1.0, atol=1e-9):
            raise RuntimeError("Phase 8 Team Elo 1X2 probabilities do not sum to 1.")

# ----------------------------------------------------------------
# CONSOLIDATED EXTERNAL BENCHMARK TABLE
# Market benchmark comes from Phase 5 and remains untouched.
# Team Elo is probability-only; no Elo projected-goal shortcut is invented.
# ----------------------------------------------------------------
PHASE8_BENCHMARK_FIELDS = [
    "fixture_id", "gameweek", "home_team", "away_team",
    "model_lambda_home", "model_lambda_away",
    "model_home_win_probability", "model_draw_probability",
    "model_away_win_probability",
    "market_snapshot_timestamp",
    "market_lambda_home", "market_lambda_away",
    "market_home_win_probability", "market_draw_probability",
    "market_away_win_probability",
    "team_elo_home", "team_elo_away",
    "team_elo_home_win_probability", "team_elo_draw_probability",
    "team_elo_away_win_probability",
    "model_minus_market_lambda_home",
    "model_minus_market_lambda_away",
    "model_minus_market_home_win_probability",
    "model_minus_market_draw_probability",
    "model_minus_market_away_win_probability",
    "model_minus_team_elo_home_win_probability",
    "model_minus_team_elo_draw_probability",
    "model_minus_team_elo_away_win_probability",
    "market_benchmark_available",
    "team_elo_benchmark_available",
    "MARKET_BLEND_WEIGHT", "ELO_BLEND_WEIGHT",
    "benchmark_policy", "forecast_timestamp", "model_version", "phase8_version",
]

_phase8_external_rows = []
_market = globals().get("market_odds_benchmark")
if not isinstance(_market, pd.DataFrame):
    _market = pd.DataFrame()

_market_idx = {}
if not _market.empty and "fixture_id" in _market.columns:
    _m = _market.copy()
    _m["fixture_id"] = pd.to_numeric(_m["fixture_id"], errors="coerce")
    _m = _m.dropna(subset=["fixture_id"]).copy()
    _m["fixture_id"] = _m["fixture_id"].astype(int)
    _market_idx = {int(r["fixture_id"]): r for _, r in _m.iterrows()}

_elo_idx = {
    int(r["fixture_id"]): r
    for _, r in team_elo_benchmark.iterrows()
}

for _, r in _phase8_model.iterrows():
    fid = int(r["fixture_id"])
    m = _market_idx.get(fid)
    e = _elo_idx.get(fid)

    def _num(obj, key):
        if obj is None:
            return np.nan
        v = pd.to_numeric(obj.get(key, np.nan), errors="coerce")
        return float(v) if pd.notna(v) else np.nan

    mh = float(r["model_lambda_home"])
    ma = float(r["model_lambda_away"])
    mwh = float(r["home_win_probability"])
    mwd = float(r["draw_probability"])
    mwa = float(r["away_win_probability"])

    market_lh = _num(m, "market_lambda_home")
    market_la = _num(m, "market_lambda_away")
    market_ph = _num(m, "market_home_win_probability")
    market_pd = _num(m, "market_draw_probability")
    market_pa = _num(m, "market_away_win_probability")

    elo_ph = _num(e, "team_elo_home_win_probability")
    elo_pd = _num(e, "team_elo_draw_probability")
    elo_pa = _num(e, "team_elo_away_win_probability")

    _phase8_external_rows.append({
        "fixture_id": fid,
        "gameweek": int(r["gameweek"]),
        "home_team": str(r["home_team"]),
        "away_team": str(r["away_team"]),
        "model_lambda_home": mh,
        "model_lambda_away": ma,
        "model_home_win_probability": mwh,
        "model_draw_probability": mwd,
        "model_away_win_probability": mwa,
        "market_snapshot_timestamp": (
            m.get("market_snapshot_timestamp", pd.NA) if m is not None else pd.NA
        ),
        "market_lambda_home": market_lh,
        "market_lambda_away": market_la,
        "market_home_win_probability": market_ph,
        "market_draw_probability": market_pd,
        "market_away_win_probability": market_pa,
        "team_elo_home": _num(e, "team_elo_home"),
        "team_elo_away": _num(e, "team_elo_away"),
        "team_elo_home_win_probability": elo_ph,
        "team_elo_draw_probability": elo_pd,
        "team_elo_away_win_probability": elo_pa,
        "model_minus_market_lambda_home": (
            mh - market_lh if np.isfinite(market_lh) else np.nan
        ),
        "model_minus_market_lambda_away": (
            ma - market_la if np.isfinite(market_la) else np.nan
        ),
        "model_minus_market_home_win_probability": (
            mwh - market_ph if np.isfinite(market_ph) else np.nan
        ),
        "model_minus_market_draw_probability": (
            mwd - market_pd if np.isfinite(market_pd) else np.nan
        ),
        "model_minus_market_away_win_probability": (
            mwa - market_pa if np.isfinite(market_pa) else np.nan
        ),
        "model_minus_team_elo_home_win_probability": (
            mwh - elo_ph if np.isfinite(elo_ph) else np.nan
        ),
        "model_minus_team_elo_draw_probability": (
            mwd - elo_pd if np.isfinite(elo_pd) else np.nan
        ),
        "model_minus_team_elo_away_win_probability": (
            mwa - elo_pa if np.isfinite(elo_pa) else np.nan
        ),
        "market_benchmark_available": bool(
            np.isfinite(market_ph) and np.isfinite(market_pd) and np.isfinite(market_pa)
        ),
        "team_elo_benchmark_available": bool(
            np.isfinite(elo_ph) and np.isfinite(elo_pd) and np.isfinite(elo_pa)
        ),
        "MARKET_BLEND_WEIGHT": MARKET_BLEND_WEIGHT,
        "ELO_BLEND_WEIGHT": ELO_BLEND_WEIGHT,
        "benchmark_policy": "EXTERNAL_COMPARISON_ONLY_NO_BLEND",
        "forecast_timestamp": globals().get("forecast_timestamp"),
        "model_version": globals().get("model_version"),
        "phase8_version": PHASE8_VERSION,
    })

phase8_external_benchmark = pd.DataFrame(
    _phase8_external_rows, columns=PHASE8_BENCHMARK_FIELDS
)

# ----------------------------------------------------------------
# PERSIST BENCHMARK SNAPSHOT FOR PHASE 9
# No hindsight backfill: append only the current pre-deadline run.
# ----------------------------------------------------------------
PHASE8_OUTPUT_DIR = Path(DATA_DIR) / "PHASE8_EXTERNAL_BENCHMARK"
PHASE8_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

PHASE8_TEAM_ELO_PATH = PHASE8_OUTPUT_DIR / "phase8_team_elo_benchmark.json"
PHASE8_EXTERNAL_COMPARE_PATH = PHASE8_OUTPUT_DIR / "phase8_model_vs_market_vs_team_elo.json"
PHASE8_STATUS_PATH = PHASE8_OUTPUT_DIR / "phase8_status.json"

team_elo_benchmark.to_json(
    PHASE8_TEAM_ELO_PATH, orient="records", date_format="iso", indent=2
)
phase8_external_benchmark.to_json(
    PHASE8_EXTERNAL_COMPARE_PATH, orient="records", date_format="iso", indent=2
)

# Permanent history lives outside disposable rendering output.
PHASE8_HISTORY_DIR = Path(PROJECT_ROOT) / "MODEL_HISTORY" / "PHASE8_EXTERNAL_BENCHMARK"
PHASE8_HISTORY_DIR.mkdir(parents=True, exist_ok=True)
PHASE8_HISTORY_PATH = PHASE8_HISTORY_DIR / "team_elo_benchmark_history.jsonl"

def _vx_phase8_deadline_map():
    events_frame = globals().get("events")
    if not isinstance(events_frame, pd.DataFrame) or events_frame.empty:
        return {}
    if "id" not in events_frame.columns or "deadline_time" not in events_frame.columns:
        return {}
    out = {}
    for _, rr in events_frame.iterrows():
        gw = pd.to_numeric(rr["id"], errors="coerce")
        ts = pd.to_datetime(rr["deadline_time"], utc=True, errors="coerce")
        if pd.notna(gw) and not pd.isna(ts):
            out[int(gw)] = ts
    return out

_deadline_map = _vx_phase8_deadline_map()
_history_records = []
for _, rr in team_elo_benchmark.iterrows():
    gw = int(rr["gameweek"])
    deadline = _deadline_map.get(gw)
    if deadline is None or pd.isna(_forecast_ts) or not (_forecast_ts < deadline):
        continue
    rec = rr.to_dict()
    rec["deadline_timestamp"] = deadline.isoformat()
    rec["is_pre_deadline_data"] = True
    _history_records.append(rec)

def _vx_phase8_json_safe(obj):
    if isinstance(obj, dict):
        return {k: _vx_phase8_json_safe(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple)):
        return [_vx_phase8_json_safe(v) for v in obj]
    if obj is pd.NA or obj is None:
        return None
    if isinstance(obj, pd.Timestamp):
        return obj.isoformat()
    if isinstance(obj, (np.integer,)):
        return int(obj)
    if isinstance(obj, (np.floating, float)):
        return None if not np.isfinite(float(obj)) else float(obj)
    if isinstance(obj, (np.bool_,)):
        return bool(obj)
    return obj

if _history_records:
    # De-duplicate this exact forecast run before appending.
    existing_keys = set()
    if PHASE8_HISTORY_PATH.exists():
        try:
            with PHASE8_HISTORY_PATH.open("r", encoding="utf-8") as fh:
                for line in fh:
                    line = line.strip()
                    if not line:
                        continue
                    obj = json.loads(line)
                    existing_keys.add((
                        str(obj.get("forecast_timestamp")),
                        int(obj.get("fixture_id")),
                        str(obj.get("phase8_version")),
                    ))
        except Exception as exc:
            raise RuntimeError(
                f"Phase 8 cannot safely read existing benchmark history: {exc}"
            )

    with PHASE8_HISTORY_PATH.open("a", encoding="utf-8") as fh:
        for rec in _history_records:
            key = (
                str(rec.get("forecast_timestamp")),
                int(rec.get("fixture_id")),
                str(rec.get("phase8_version")),
            )
            if key in existing_keys:
                continue
            fh.write(json.dumps(_vx_phase8_json_safe(rec), sort_keys=True) + "\n")
            existing_keys.add(key)

# ----------------------------------------------------------------
# FINAL INTEGRITY GATE
# ----------------------------------------------------------------
PHASE8_MODEL_STATE_AFTER = {
    name: _vx_phase8_frame_digest(globals()[name])
    for name in _REQUIRED_MODEL_FRAMES
}

if PHASE8_MODEL_STATE_BEFORE != PHASE8_MODEL_STATE_AFTER:
    raise RuntimeError(
        "Phase 8 integrity failure: an external benchmark changed model-only outputs."
    )

PHASE8_STATUS = {
    "phase8_version": PHASE8_VERSION,
    "market_blend_weight": MARKET_BLEND_WEIGHT,
    "elo_blend_weight": ELO_BLEND_WEIGHT,
    "blend_active": False,
    "model_output_source": "PURE_MODEL_ONLY",
    "market_benchmark_rows": int(
        phase8_external_benchmark["market_benchmark_available"].sum()
        if not phase8_external_benchmark.empty else 0
    ),
    "team_elo_benchmark_rows": int(
        phase8_external_benchmark["team_elo_benchmark_available"].sum()
        if not phase8_external_benchmark.empty else 0
    ),
    "team_elo_calibration_rows": int(PHASE8_ELO_FIT.get("sample_size", 0)),
    "team_elo_calibration_status": str(PHASE8_ELO_FIT.get("status")),
    "history_rows_appended_this_run": int(len(_history_records)),
    "projected_goals_changed": False,
    "clean_sheet_probabilities_changed": False,
    "fixture_rating_changed": False,
    "player_expected_points_built": False,
}
PHASE8_STATUS_PATH.write_text(
    json.dumps(PHASE8_STATUS, indent=2, default=str),
    encoding="utf-8",
)

# Phase-0 clean-room scan must still pass all new benchmark tables.
vx_phase0_assert_no_forbidden_columns(
    team_elo_benchmark, "Phase 8 Team Elo benchmark"
)
vx_phase0_assert_no_forbidden_columns(
    phase8_external_benchmark, "Phase 8 external benchmark"
)

print("✅ PHASE 8 — MARKET + TEAM ELO BENCHMARKING BUILT")
print(f"   MARKET_BLEND_WEIGHT = {MARKET_BLEND_WEIGHT:.2f}")
print(f"   ELO_BLEND_WEIGHT    = {ELO_BLEND_WEIGHT:.2f}")
print("   Model outputs remain PURE MODEL ONLY.")
print(f"   Market benchmark fixtures: {PHASE8_STATUS['market_benchmark_rows']}")
print(f"   Team Elo benchmark fixtures: {PHASE8_STATUS['team_elo_benchmark_rows']}")
print(
    "   Team Elo calibration:",
    PHASE8_STATUS["team_elo_calibration_status"],
    f"(n={PHASE8_STATUS['team_elo_calibration_rows']})",
)
print("   No player expected-points or FPL decision logic was built.")
print("   Saved:", PHASE8_EXTERNAL_COMPARE_PATH)
```

```python
# ============================================================
# CELL 13C — PHASE 9
# OUT-OF-SAMPLE BLEND TESTING + CALIBRATION REVIEW
# BASELINE IS IMMUTABLE • NO LIVE BLEND ACTIVATION • NO PLAYER XPTS
# ============================================================
# Consultant contract implemented here:
#   - keep pure model as the production baseline;
#   - test market and Team Elo only on genuine pre-deadline frozen history;
#   - choose/tune on earlier fixtures and judge on later holdout fixtures;
#   - never overwrite the baseline model;
#   - never activate a blend merely because it exists;
#   - Team Elo remains a probability benchmark, not an invented xG engine.
# ============================================================

import json
import math
import hashlib
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd

PHASE9_VERSION = "VORTEX-V1.1-PHASE9-ROLLING-OOS-CALIBRATION"
PHASE9_TIMESTAMP = datetime.now(timezone.utc).isoformat()

# Production remains the model-only baseline. These are evaluation candidates,
# not live weights.
PHASE9_PRODUCTION_MODEL_LABEL = "FDR_V1.0_MODEL_ONLY"
PHASE9_MARKET_MODEL_WEIGHTS = tuple(round(x, 2) for x in np.arange(1.00, 0.49, -0.05))
PHASE9_ELO_MODEL_WEIGHTS = tuple(round(x, 2) for x in np.arange(1.00, 0.59, -0.05))
PHASE9_COMBINED_WEIGHTS = {"model": 0.80, "market": 0.10, "team_elo": 0.10}
PHASE9_MIN_TRAIN_FIXTURES = 60
PHASE9_MIN_HOLDOUT_FIXTURES = 20
PHASE9_TARGET_TRAIN_FRACTION = 0.75
PHASE9_MIN_WALK_FORWARD_FOLDS = 3
PHASE9_WALK_FORWARD_FIXTURES = 20
PHASE9_BOOTSTRAP_REPEATS = 2000
PHASE9_BOOTSTRAP_SEED = 20260824
PHASE9_REQUIRED_FOLD_WIN_RATE = 2.0 / 3.0
PHASE9_LIVE_BLEND_ACTIVE = False
PHASE9_READY_FOR_PLAYER_MODEL = False

# Phase 8 benchmark weights must still be zero. Phase 9 does not change them.
if float(globals().get("MARKET_BLEND_WEIGHT", 0.0)) != 0.0:
    raise RuntimeError("Phase 9 policy violation: Phase 8 MARKET_BLEND_WEIGHT must remain 0.00.")
if float(globals().get("ELO_BLEND_WEIGHT", 0.0)) != 0.0:
    raise RuntimeError("Phase 9 policy violation: Phase 8 ELO_BLEND_WEIGHT must remain 0.00.")
if PHASE9_LIVE_BLEND_ACTIVE:
    raise RuntimeError("Phase 9 policy violation: blend testing cannot activate a live blend.")

# ----------------------------------------------------------------
# IMMUTABILITY GATE — Phase 9 may read these objects but never mutate them.
# ----------------------------------------------------------------
def _vx_phase9_digest(obj):
    if obj is None:
        return "NONE"
    if isinstance(obj, pd.DataFrame):
        if obj.empty:
            return "EMPTY:" + ",".join(map(str, obj.columns))
        stable = obj.copy()
        stable = stable.reindex(sorted(stable.columns), axis=1)
        try:
            stable = stable.sort_values(list(stable.columns), kind="stable").reset_index(drop=True)
        except Exception:
            stable = stable.reset_index(drop=True)
        payload = stable.to_json(orient="split", date_format="iso", default_handler=str)
    else:
        payload = repr(obj)
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()

_PHASE9_IMMUTABLE_NAMES = (
    "projected_goals_model_only",
    "fixture_probabilities_model_only",
    "fixture_fdr_model_only",
    "market_odds_benchmark",
    "team_elo_benchmark",
    "phase8_external_benchmark",
)
_PHASE9_STATE_BEFORE = {
    name: _vx_phase9_digest(globals().get(name)) for name in _PHASE9_IMMUTABLE_NAMES
}
_PHASE9_WEIGHT_STATE_BEFORE = (
    float(globals().get("MARKET_BLEND_WEIGHT", 0.0)),
    float(globals().get("ELO_BLEND_WEIGHT", 0.0)),
)

# ----------------------------------------------------------------
# INPUT 1 — completed Phase-7 frozen forecasts only.
# No hindsight forecasts are manufactured in this phase.
# ----------------------------------------------------------------
def _vx_phase9_completed_phase7():
    frame = globals().get("phase7_completed_forecasts")
    if isinstance(frame, pd.DataFrame):
        h = frame.copy()
    else:
        frame = globals().get("phase7_frozen_forecast_history")
        if isinstance(frame, pd.DataFrame):
            h = frame.copy()
        else:
            path = globals().get("PHASE7_HISTORY_PATH")
            if path is not None and Path(path).exists():
                try:
                    h = pd.read_json(path, lines=True)
                except Exception:
                    h = pd.DataFrame()
            else:
                h = pd.DataFrame()

    if h.empty:
        return h

    required = [
        "fixture_id", "forecast_timestamp", "model_lambda_home", "model_lambda_away",
        "home_win_probability", "draw_probability", "away_win_probability",
        "home_cs_probability", "away_cs_probability",
        "actual_home_goals", "actual_away_goals",
    ]
    missing = [c for c in required if c not in h.columns]
    if missing:
        raise RuntimeError(f"Phase 9 requires Phase-7 frozen fields: {missing}")

    for c in [
        "model_lambda_home", "model_lambda_away",
        "home_win_probability", "draw_probability", "away_win_probability",
        "home_cs_probability", "away_cs_probability",
        "market_lambda_home", "market_lambda_away",
        "market_home_win_probability", "market_draw_probability", "market_away_win_probability",
        "market_home_clean_sheet_probability", "market_away_clean_sheet_probability",
        "actual_home_goals", "actual_away_goals",
    ]:
        if c not in h.columns:
            h[c] = np.nan
        h[c] = pd.to_numeric(h[c], errors="coerce")

    h["fixture_id"] = pd.to_numeric(h["fixture_id"], errors="coerce")
    h["forecast_timestamp"] = pd.to_datetime(h["forecast_timestamp"], utc=True, errors="coerce")
    if "fpl_deadline_timestamp" in h.columns:
        h["fpl_deadline_timestamp"] = pd.to_datetime(h["fpl_deadline_timestamp"], utc=True, errors="coerce")
    else:
        h["fpl_deadline_timestamp"] = pd.NaT
    if "gameweek" not in h.columns:
        h["gameweek"] = np.nan
    h["gameweek"] = pd.to_numeric(h["gameweek"], errors="coerce")

    h = h.dropna(subset=[
        "fixture_id", "forecast_timestamp", "model_lambda_home", "model_lambda_away",
        "actual_home_goals", "actual_away_goals",
    ]).copy()
    h["fixture_id"] = h["fixture_id"].astype(int)
    h["actual_home_goals"] = h["actual_home_goals"].astype(int)
    h["actual_away_goals"] = h["actual_away_goals"].astype(int)

    # Keep one genuinely frozen forecast per fixture. Phase 7 owns the freeze
    # policy; Phase 9 simply uses its latest stored row for that fixture.
    h = h.sort_values(["fixture_id", "forecast_timestamp"], kind="stable")
    h = h.drop_duplicates(subset=["fixture_id"], keep="last").reset_index(drop=True)
    return h

phase9_completed_forecasts = _vx_phase9_completed_phase7()

# ----------------------------------------------------------------
# INPUT 2 — deadline-safe Phase-8 Team Elo benchmark history.
# Match each frozen Phase-7 forecast to the latest Elo snapshot that existed
# no later than that forecast. Missing remains missing.
# ----------------------------------------------------------------
def _vx_phase9_load_elo_history():
    frame = globals().get("phase8_team_elo_history")
    if isinstance(frame, pd.DataFrame):
        e = frame.copy()
    else:
        path = globals().get("PHASE8_HISTORY_PATH")
        if path is not None and Path(path).exists():
            try:
                e = pd.read_json(path, lines=True)
            except Exception:
                e = pd.DataFrame()
        else:
            e = pd.DataFrame()
    if e.empty:
        return e

    needed = [
        "fixture_id", "forecast_timestamp",
        "team_elo_home_win_probability", "team_elo_draw_probability",
        "team_elo_away_win_probability",
    ]
    for c in needed:
        if c not in e.columns:
            e[c] = np.nan
    e["fixture_id"] = pd.to_numeric(e["fixture_id"], errors="coerce")
    e["forecast_timestamp"] = pd.to_datetime(e["forecast_timestamp"], utc=True, errors="coerce")
    for c in needed[2:]:
        e[c] = pd.to_numeric(e[c], errors="coerce")
    if "is_pre_deadline_data" in e.columns:
        e = e.loc[e["is_pre_deadline_data"].fillna(False).astype(bool)].copy()
    if "elo_benchmark_status" in e.columns:
        e = e.loc[e["elo_benchmark_status"].astype(str).eq("READY")].copy()
    e = e.dropna(subset=needed).copy()
    if e.empty:
        return e
    e["fixture_id"] = e["fixture_id"].astype(int)
    return e.sort_values(["fixture_id", "forecast_timestamp"], kind="stable").reset_index(drop=True)


def _vx_phase9_attach_elo(forecasts, elo_history):
    out = forecasts.copy()
    for c in [
        "team_elo_home_win_probability", "team_elo_draw_probability",
        "team_elo_away_win_probability", "team_elo_benchmark_timestamp",
    ]:
        if c not in out.columns:
            out[c] = np.nan if "timestamp" not in c else pd.Series(pd.NaT, index=out.index, dtype="datetime64[ns, UTC]")
    if out.empty or elo_history.empty:
        return out

    groups = {fid: g for fid, g in elo_history.groupby("fixture_id", sort=False)}
    for idx, row in out.iterrows():
        fid = int(row["fixture_id"])
        g = groups.get(fid)
        if g is None or g.empty:
            continue
        cutoff = row["forecast_timestamp"]
        eligible = g.loc[g["forecast_timestamp"] <= cutoff]
        if eligible.empty:
            continue
        rr = eligible.iloc[-1]
        out.at[idx, "team_elo_home_win_probability"] = float(rr["team_elo_home_win_probability"])
        out.at[idx, "team_elo_draw_probability"] = float(rr["team_elo_draw_probability"])
        out.at[idx, "team_elo_away_win_probability"] = float(rr["team_elo_away_win_probability"])
        out.at[idx, "team_elo_benchmark_timestamp"] = rr["forecast_timestamp"]
    return out

phase9_elo_history = _vx_phase9_load_elo_history()
phase9_validation_frame = _vx_phase9_attach_elo(phase9_completed_forecasts, phase9_elo_history)

# ----------------------------------------------------------------
# CORE METRICS
# ----------------------------------------------------------------
def _vx_phase9_poisson_pmf(k, lam):
    if not np.isfinite(lam) or lam <= 0:
        return np.nan
    return math.exp(-float(lam)) * (float(lam) ** int(k)) / math.factorial(int(k))


def _vx_phase9_scoreline_probability(lh, la, hg, ag):
    return float(_vx_phase9_poisson_pmf(int(hg), float(lh)) * _vx_phase9_poisson_pmf(int(ag), float(la)))


def _vx_phase9_wdl_from_lambdas(lh, la):
    lh, la = float(lh), float(la)
    # Adaptive tail: retain effectively all probability mass.
    max_goal = max(12, int(math.ceil(max(lh, la) + 10.0 * math.sqrt(max(lh, la, 1.0)))))
    hp = np.array([_vx_phase9_poisson_pmf(k, lh) for k in range(max_goal + 1)], dtype=float)
    ap = np.array([_vx_phase9_poisson_pmf(k, la) for k in range(max_goal + 1)], dtype=float)
    mat = np.outer(hp, ap)
    mass = float(mat.sum())
    if not np.isfinite(mass) or mass <= 0:
        return np.nan, np.nan, np.nan
    mat = mat / mass
    return float(np.tril(mat, -1).sum()), float(np.trace(mat)), float(np.triu(mat, 1).sum())


def _vx_phase9_poisson_deviance_component(y, mu):
    y, mu = float(y), float(mu)
    if not np.isfinite(mu) or mu <= 0:
        return np.nan
    if y <= 0:
        return 2.0 * mu
    return 2.0 * (mu - y + y * math.log(y / mu))


def _vx_phase9_target_wdl(hg, ag):
    if hg > ag:
        return np.array([1.0, 0.0, 0.0], dtype=float)
    if hg == ag:
        return np.array([0.0, 1.0, 0.0], dtype=float)
    return np.array([0.0, 0.0, 1.0], dtype=float)


def _vx_phase9_binary_ece(predicted, observed, bins=10):
    p = np.asarray(predicted, dtype=float)
    y = np.asarray(observed, dtype=float)
    mask = np.isfinite(p) & np.isfinite(y)
    p, y = p[mask], y[mask]
    if len(p) == 0:
        return np.nan
    edges = np.linspace(0.0, 1.0, int(bins) + 1)
    total = len(p)
    ece = 0.0
    for i in range(len(edges) - 1):
        lo, hi = edges[i], edges[i + 1]
        m = (p >= lo) & ((p < hi) if i < len(edges) - 2 else (p <= hi))
        if not np.any(m):
            continue
        ece += (np.sum(m) / total) * abs(float(np.mean(p[m])) - float(np.mean(y[m])))
    return float(ece)


def _vx_phase9_wdl_calibration_ece(prob_rows, target_rows):
    if not prob_rows:
        return np.nan
    return _vx_phase9_binary_ece(np.concatenate(prob_rows), np.concatenate(target_rows), bins=10)


def _vx_phase9_full_metrics(frame, lambda_builder):
    """Metrics for candidates that define home/away lambdas."""
    eps = 1e-15
    dev, mae, score_ll, wdl_ll, wdl_brier, cs_brier = [], [], [], [], [], []
    wdl_probs, wdl_targets, cs_probs, cs_targets = [], [], [], []
    for _, r in frame.iterrows():
        lh, la = lambda_builder(r)
        if not (np.isfinite(lh) and np.isfinite(la) and lh > 0 and la > 0):
            continue
        hg, ag = int(r["actual_home_goals"]), int(r["actual_away_goals"])
        ph, pd_, pa = _vx_phase9_wdl_from_lambdas(lh, la)
        pred = np.array([ph, pd_, pa], dtype=float)
        target = _vx_phase9_target_wdl(hg, ag)
        result_idx = int(np.argmax(target))

        dev.append((_vx_phase9_poisson_deviance_component(hg, lh) + _vx_phase9_poisson_deviance_component(ag, la)) / 2.0)
        mae.append((abs(hg - lh) + abs(ag - la)) / 2.0)
        score_p = _vx_phase9_scoreline_probability(lh, la, hg, ag)
        score_ll.append(-math.log(max(eps, min(1.0, score_p))))
        wdl_ll.append(-math.log(max(eps, min(1.0, pred[result_idx]))))
        wdl_brier.append(float(np.sum((pred - target) ** 2)))

        home_cs = float(math.exp(-la))
        away_cs = float(math.exp(-lh))
        home_obs = 1.0 if ag == 0 else 0.0
        away_obs = 1.0 if hg == 0 else 0.0
        cs_brier.extend([(home_cs - home_obs) ** 2, (away_cs - away_obs) ** 2])
        cs_probs.extend([home_cs, away_cs])
        cs_targets.extend([home_obs, away_obs])
        wdl_probs.append(pred)
        wdl_targets.append(target)

    n = len(score_ll)
    return {
        "sample_size": int(n),
        "poisson_deviance": float(np.mean(dev)) if dev else np.nan,
        "projected_goal_mae": float(np.mean(mae)) if mae else np.nan,
        "scoreline_log_loss": float(np.mean(score_ll)) if score_ll else np.nan,
        "wdl_log_loss": float(np.mean(wdl_ll)) if wdl_ll else np.nan,
        "wdl_multiclass_brier": float(np.mean(wdl_brier)) if wdl_brier else np.nan,
        "clean_sheet_brier": float(np.mean(cs_brier)) if cs_brier else np.nan,
        "wdl_calibration_ece": _vx_phase9_wdl_calibration_ece(wdl_probs, wdl_targets),
        "clean_sheet_calibration_ece": _vx_phase9_binary_ece(cs_probs, cs_targets, bins=10),
    }


def _vx_phase9_probability_metrics(frame, probability_builder):
    """W/D/L-only metrics for Team Elo probability candidates."""
    eps = 1e-15
    ll, brier, probs, targets = [], [], [], []
    for _, r in frame.iterrows():
        p = np.asarray(probability_builder(r), dtype=float)
        if p.shape != (3,) or not np.all(np.isfinite(p)) or np.any(p < 0):
            continue
        s = float(p.sum())
        if s <= 0:
            continue
        p = p / s
        target = _vx_phase9_target_wdl(int(r["actual_home_goals"]), int(r["actual_away_goals"]))
        idx = int(np.argmax(target))
        ll.append(-math.log(max(eps, min(1.0, p[idx]))))
        brier.append(float(np.sum((p - target) ** 2)))
        probs.append(p)
        targets.append(target)
    return {
        "sample_size": int(len(ll)),
        "wdl_log_loss": float(np.mean(ll)) if ll else np.nan,
        "wdl_multiclass_brier": float(np.mean(brier)) if brier else np.nan,
        "wdl_calibration_ece": _vx_phase9_wdl_calibration_ece(probs, targets),
    }

# ----------------------------------------------------------------
# CHRONOLOGICAL SPLIT — never split a single FPL deadline across train/holdout.
# ----------------------------------------------------------------
def _vx_phase9_chronological_split(frame):
    minimum = PHASE9_MIN_TRAIN_FIXTURES + PHASE9_MIN_HOLDOUT_FIXTURES
    if frame is None or frame.empty or len(frame) < minimum:
        n = 0 if frame is None else int(len(frame))
        return None, None, {
            "status": f"WAITING_FOR_HISTORY: need {minimum}, have {n}",
            "train_size": 0,
            "holdout_size": 0,
            "split_key": None,
        }

    h = frame.copy()
    if "fpl_deadline_timestamp" in h.columns:
        deadline = pd.to_datetime(h["fpl_deadline_timestamp"], utc=True, errors="coerce")
    else:
        deadline = pd.Series(pd.NaT, index=h.index)
    fallback = pd.to_datetime(h["forecast_timestamp"], utc=True, errors="coerce")
    h["_split_time"] = deadline.where(deadline.notna(), fallback)
    h = h.dropna(subset=["_split_time"]).sort_values(["_split_time", "fixture_id"], kind="stable").reset_index(drop=True)
    if len(h) < minimum:
        return None, None, {
            "status": f"WAITING_FOR_HISTORY: need {minimum}, have {len(h)} with timestamps",
            "train_size": 0,
            "holdout_size": 0,
            "split_key": None,
        }

    groups = h.groupby("_split_time", sort=True).size().reset_index(name="n")
    groups["cum_n"] = groups["n"].cumsum()
    target = PHASE9_TARGET_TRAIN_FRACTION * len(h)
    candidates = groups.loc[
        (groups["cum_n"] >= PHASE9_MIN_TRAIN_FIXTURES)
        & ((len(h) - groups["cum_n"]) >= PHASE9_MIN_HOLDOUT_FIXTURES)
    ].copy()
    if candidates.empty:
        return None, None, {
            "status": "WAITING_FOR_DEADLINE_BOUNDARY_SPLIT",
            "train_size": 0,
            "holdout_size": 0,
            "split_key": None,
        }
    candidates["distance"] = (candidates["cum_n"] - target).abs()
    chosen = candidates.sort_values(["distance", "_split_time"], kind="stable").iloc[0]
    boundary = chosen["_split_time"]
    train = h.loc[h["_split_time"] <= boundary].drop(columns=["_split_time"]).copy()
    holdout = h.loc[h["_split_time"] > boundary].drop(columns=["_split_time"]).copy()
    return train, holdout, {
        "status": "READY",
        "train_size": int(len(train)),
        "holdout_size": int(len(holdout)),
        "split_key": pd.Timestamp(boundary).isoformat(),
    }

# ----------------------------------------------------------------
# MARKET-LAMBDA EXPERIMENT
# 100/0 baseline, then 90/10, 80/20, 70/30. Pure market is reported as
# an external benchmark but is not a tunable production candidate here.
# ----------------------------------------------------------------
def _vx_phase9_market_overlap(frame):
    if frame.empty:
        return frame.copy()
    needed = ["market_lambda_home", "market_lambda_away"]
    h = frame.copy()
    for c in needed:
        h[c] = pd.to_numeric(h[c], errors="coerce")
    return h.dropna(subset=needed).copy()


def _vx_phase9_market_lambda_builder(model_weight):
    mw = float(model_weight)
    return lambda r: (
        mw * float(r["model_lambda_home"]) + (1.0 - mw) * float(r["market_lambda_home"]),
        mw * float(r["model_lambda_away"]) + (1.0 - mw) * float(r["market_lambda_away"]),
    )


def _vx_phase9_market_experiment(frame):
    overlap = _vx_phase9_market_overlap(frame)
    train, holdout, split_info = _vx_phase9_chronological_split(overlap)
    rows = []
    result = {
        "status": split_info["status"],
        "selected_model_weight": None,
        "selected_market_weight": None,
        "holdout_approved": False,
        **split_info,
    }
    if train is None:
        return pd.DataFrame(), result

    train_scores = []
    for mw in PHASE9_MARKET_MODEL_WEIGHTS:
        for split_name, part in (("TRAIN", train), ("HOLDOUT", holdout)):
            metrics = _vx_phase9_full_metrics(part, _vx_phase9_market_lambda_builder(mw))
            rows.append({
                "candidate": f"MODEL_{int(round(mw*100))}_MARKET_{int(round((1-mw)*100))}",
                "model_weight": float(mw),
                "market_weight": float(1.0 - mw),
                "split": split_name,
                **metrics,
            })
        mt = _vx_phase9_full_metrics(train, _vx_phase9_market_lambda_builder(mw))
        train_scores.append((
            mt["scoreline_log_loss"], mt["poisson_deviance"],
            mt["wdl_multiclass_brier"], abs(1.0 - mw), -mw, mw,
        ))

    # Pure market benchmark, evaluated on the same split but never auto-selected.
    for split_name, part in (("TRAIN", train), ("HOLDOUT", holdout)):
        metrics = _vx_phase9_full_metrics(part, _vx_phase9_market_lambda_builder(0.0))
        rows.append({
            "candidate": "PURE_MARKET_BENCHMARK",
            "model_weight": 0.0,
            "market_weight": 1.0,
            "split": split_name,
            **metrics,
        })

    train_scores.sort(key=lambda x: x[:-1])
    selected = float(train_scores[0][-1])
    out = pd.DataFrame(rows)
    hold = out.loc[out["split"].eq("HOLDOUT")].set_index("model_weight")
    chosen = hold.loc[selected]
    baseline = hold.loc[1.0]

    approved = bool(
        selected < 1.0
        and float(chosen["scoreline_log_loss"]) < float(baseline["scoreline_log_loss"]) - 1e-4
        and float(chosen["poisson_deviance"]) <= float(baseline["poisson_deviance"]) + 1e-4
        and float(chosen["projected_goal_mae"]) <= float(baseline["projected_goal_mae"]) + 1e-3
        and float(chosen["wdl_multiclass_brier"]) <= float(baseline["wdl_multiclass_brier"]) + 1e-4
        and float(chosen["clean_sheet_brier"]) <= float(baseline["clean_sheet_brier"]) + 1e-4
        and float(chosen["wdl_calibration_ece"]) <= float(baseline["wdl_calibration_ece"]) + 0.01
    )
    result.update({
        "status": "HOLDOUT_COMPLETE",
        "selected_model_weight": selected,
        "selected_market_weight": float(1.0 - selected),
        "holdout_approved": approved,
        "holdout_scoreline_log_loss_gain_vs_model": float(baseline["scoreline_log_loss"] - chosen["scoreline_log_loss"]),
        "holdout_poisson_deviance_gain_vs_model": float(baseline["poisson_deviance"] - chosen["poisson_deviance"]),
        "holdout_goal_mae_gain_vs_model": float(baseline["projected_goal_mae"] - chosen["projected_goal_mae"]),
        "holdout_wdl_brier_gain_vs_model": float(baseline["wdl_multiclass_brier"] - chosen["wdl_multiclass_brier"]),
        "holdout_cs_brier_gain_vs_model": float(baseline["clean_sheet_brier"] - chosen["clean_sheet_brier"]),
        "live_blend_activated": False,
    })
    return out, result

# ----------------------------------------------------------------
# TEAM-ELO PROBABILITY EXPERIMENT
# Team Elo has no invented lambdas. Therefore Elo tests W/D/L calibration only.
# ----------------------------------------------------------------
def _vx_phase9_elo_overlap(frame):
    if frame.empty:
        return frame.copy()
    needed = [
        "team_elo_home_win_probability", "team_elo_draw_probability",
        "team_elo_away_win_probability",
    ]
    h = frame.copy()
    for c in needed:
        h[c] = pd.to_numeric(h[c], errors="coerce")
    return h.dropna(subset=needed).copy()


def _vx_phase9_model_wdl(r):
    return np.array([
        float(r["home_win_probability"]),
        float(r["draw_probability"]),
        float(r["away_win_probability"]),
    ], dtype=float)


def _vx_phase9_market_wdl(r):
    return np.array([
        float(r["market_home_win_probability"]),
        float(r["market_draw_probability"]),
        float(r["market_away_win_probability"]),
    ], dtype=float)


def _vx_phase9_elo_wdl(r):
    return np.array([
        float(r["team_elo_home_win_probability"]),
        float(r["team_elo_draw_probability"]),
        float(r["team_elo_away_win_probability"]),
    ], dtype=float)


def _vx_phase9_elo_probability_builder(model_weight):
    mw = float(model_weight)
    return lambda r: mw * _vx_phase9_model_wdl(r) + (1.0 - mw) * _vx_phase9_elo_wdl(r)


def _vx_phase9_elo_experiment(frame):
    overlap = _vx_phase9_elo_overlap(frame)
    train, holdout, split_info = _vx_phase9_chronological_split(overlap)
    rows = []
    result = {
        "status": split_info["status"],
        "selected_model_weight": None,
        "selected_elo_weight": None,
        "holdout_approved": False,
        **split_info,
    }
    if train is None:
        return pd.DataFrame(), result

    train_scores = []
    for mw in PHASE9_ELO_MODEL_WEIGHTS:
        for split_name, part in (("TRAIN", train), ("HOLDOUT", holdout)):
            metrics = _vx_phase9_probability_metrics(part, _vx_phase9_elo_probability_builder(mw))
            rows.append({
                "candidate": f"MODEL_{int(round(mw*100))}_TEAM_ELO_{int(round((1-mw)*100))}",
                "model_weight": float(mw),
                "elo_weight": float(1.0 - mw),
                "split": split_name,
                **metrics,
            })
        mt = _vx_phase9_probability_metrics(train, _vx_phase9_elo_probability_builder(mw))
        train_scores.append((mt["wdl_log_loss"], mt["wdl_multiclass_brier"], mt["wdl_calibration_ece"], abs(1-mw), -mw, mw))

    for split_name, part in (("TRAIN", train), ("HOLDOUT", holdout)):
        metrics = _vx_phase9_probability_metrics(part, _vx_phase9_elo_probability_builder(0.0))
        rows.append({
            "candidate": "PURE_TEAM_ELO_BENCHMARK",
            "model_weight": 0.0,
            "elo_weight": 1.0,
            "split": split_name,
            **metrics,
        })

    train_scores.sort(key=lambda x: x[:-1])
    selected = float(train_scores[0][-1])
    out = pd.DataFrame(rows)
    hold = out.loc[out["split"].eq("HOLDOUT")].set_index("model_weight")
    chosen = hold.loc[selected]
    baseline = hold.loc[1.0]
    approved = bool(
        selected < 1.0
        and float(chosen["wdl_log_loss"]) < float(baseline["wdl_log_loss"]) - 1e-4
        and float(chosen["wdl_multiclass_brier"]) <= float(baseline["wdl_multiclass_brier"]) + 1e-4
        and float(chosen["wdl_calibration_ece"]) <= float(baseline["wdl_calibration_ece"]) + 0.01
    )
    result.update({
        "status": "HOLDOUT_COMPLETE",
        "selected_model_weight": selected,
        "selected_elo_weight": float(1.0 - selected),
        "holdout_approved": approved,
        "holdout_wdl_log_loss_gain_vs_model": float(baseline["wdl_log_loss"] - chosen["wdl_log_loss"]),
        "holdout_wdl_brier_gain_vs_model": float(baseline["wdl_multiclass_brier"] - chosen["wdl_multiclass_brier"]),
        "live_blend_activated": False,
    })
    return out, result

# ----------------------------------------------------------------
# FIXED COMBINED TEST: 80% model / 10% market / 10% Team Elo.
# This is a W/D/L probability test only. It cannot silently create blended xG.
# ----------------------------------------------------------------
def _vx_phase9_combined_overlap(frame):
    h = _vx_phase9_elo_overlap(frame)
    if h.empty:
        return h
    needed = [
        "market_home_win_probability", "market_draw_probability",
        "market_away_win_probability",
    ]
    for c in needed:
        h[c] = pd.to_numeric(h[c], errors="coerce")
    return h.dropna(subset=needed).copy()


def _vx_phase9_combined_builder(r):
    w = PHASE9_COMBINED_WEIGHTS
    return w["model"] * _vx_phase9_model_wdl(r) + w["market"] * _vx_phase9_market_wdl(r) + w["team_elo"] * _vx_phase9_elo_wdl(r)


def _vx_phase9_combined_experiment(frame):
    overlap = _vx_phase9_combined_overlap(frame)
    train, holdout, split_info = _vx_phase9_chronological_split(overlap)
    result = {
        "status": split_info["status"],
        "weights": dict(PHASE9_COMBINED_WEIGHTS),
        "holdout_approved": False,
        **split_info,
    }
    if train is None:
        return pd.DataFrame(), result

    rows = []
    for split_name, part in (("TRAIN", train), ("HOLDOUT", holdout)):
        model_m = _vx_phase9_probability_metrics(part, _vx_phase9_model_wdl)
        combo_m = _vx_phase9_probability_metrics(part, _vx_phase9_combined_builder)
        rows.append({"candidate": "PURE_MODEL", "split": split_name, **model_m})
        rows.append({"candidate": "MODEL_80_MARKET_10_TEAM_ELO_10", "split": split_name, **combo_m})

    out = pd.DataFrame(rows)
    hold = out.loc[out["split"].eq("HOLDOUT")].set_index("candidate")
    base = hold.loc["PURE_MODEL"]
    combo = hold.loc["MODEL_80_MARKET_10_TEAM_ELO_10"]
    approved = bool(
        float(combo["wdl_log_loss"]) < float(base["wdl_log_loss"]) - 1e-4
        and float(combo["wdl_multiclass_brier"]) <= float(base["wdl_multiclass_brier"]) + 1e-4
        and float(combo["wdl_calibration_ece"]) <= float(base["wdl_calibration_ece"]) + 0.01
    )
    result.update({
        "status": "HOLDOUT_COMPLETE",
        "holdout_approved": approved,
        "holdout_wdl_log_loss_gain_vs_model": float(base["wdl_log_loss"] - combo["wdl_log_loss"]),
        "holdout_wdl_brier_gain_vs_model": float(base["wdl_multiclass_brier"] - combo["wdl_multiclass_brier"]),
        "live_blend_activated": False,
    })
    return out, result

# ----------------------------------------------------------------
# DIRECT EXTERNAL W/D/L BENCHMARKS
# Compare pure model with the original de-vigged market 1X2 probabilities and
# with Team Elo probabilities. These are comparison metrics, not blend inputs.
# ----------------------------------------------------------------
def _vx_phase9_external_wdl_benchmark(frame):
    rows = []

    market = frame.copy()
    market_needed = [
        "market_home_win_probability", "market_draw_probability",
        "market_away_win_probability",
    ]
    for c in market_needed:
        market[c] = pd.to_numeric(market[c], errors="coerce")
    market = market.dropna(subset=market_needed).copy()
    if not market.empty:
        rows.append({
            "comparison_group": "MODEL_VS_MARKET_1X2",
            "source": "PURE_MODEL",
            **_vx_phase9_probability_metrics(market, _vx_phase9_model_wdl),
        })
        rows.append({
            "comparison_group": "MODEL_VS_MARKET_1X2",
            "source": "PURE_MARKET_1X2",
            **_vx_phase9_probability_metrics(market, _vx_phase9_market_wdl),
        })

    elo = _vx_phase9_elo_overlap(frame)
    if not elo.empty:
        rows.append({
            "comparison_group": "MODEL_VS_TEAM_ELO",
            "source": "PURE_MODEL",
            **_vx_phase9_probability_metrics(elo, _vx_phase9_model_wdl),
        })
        rows.append({
            "comparison_group": "MODEL_VS_TEAM_ELO",
            "source": "PURE_TEAM_ELO",
            **_vx_phase9_probability_metrics(elo, _vx_phase9_elo_wdl),
        })

    return pd.DataFrame(rows)

# ----------------------------------------------------------------
# ACCURACY UPGRADE — ROLLING-ORIGIN STABILITY
# Borrowed conceptually from the older weekly notebook, but rebuilt against the
# current deadline-safe Phase-7 archive and the hardened Phase-9 metric contract.
# A candidate must improve across multiple future-only folds and pass a
# deterministic bootstrap lower-bound test. One lucky holdout can never promote it.
# ----------------------------------------------------------------
def _vx_phase9_rolling_origin_splits(frame):
    if frame is None or frame.empty:
        return []
    h = frame.copy()
    if "fpl_deadline_timestamp" in h.columns:
        deadline = pd.to_datetime(h["fpl_deadline_timestamp"], utc=True, errors="coerce")
    else:
        deadline = pd.Series(pd.NaT, index=h.index)
    fallback = pd.to_datetime(h["forecast_timestamp"], utc=True, errors="coerce")
    h["_split_time"] = deadline.where(deadline.notna(), fallback)
    h = h.dropna(subset=["_split_time"]).sort_values(
        ["_split_time", "fixture_id"], kind="stable"
    ).reset_index(drop=True)
    if h.empty:
        return []

    boundaries = list(h["_split_time"].drop_duplicates().sort_values())
    folds = []
    train_end = None
    for boundary in boundaries:
        if int((h["_split_time"] <= boundary).sum()) >= PHASE9_MIN_TRAIN_FIXTURES:
            train_end = boundary
            break
    if train_end is None:
        return []

    while True:
        later = [x for x in boundaries if x > train_end]
        if not later:
            break
        validation_end = None
        for boundary in later:
            n_val = int(((h["_split_time"] > train_end) & (h["_split_time"] <= boundary)).sum())
            if n_val >= PHASE9_WALK_FORWARD_FIXTURES:
                validation_end = boundary
                break
        if validation_end is None:
            break
        train = h.loc[h["_split_time"] <= train_end].drop(columns=["_split_time"]).copy()
        validation = h.loc[
            (h["_split_time"] > train_end) & (h["_split_time"] <= validation_end)
        ].drop(columns=["_split_time"]).copy()
        folds.append((train, validation, {
            "fold": len(folds) + 1,
            "train_size": int(len(train)),
            "validation_size": int(len(validation)),
            "train_end": pd.Timestamp(train_end).isoformat(),
            "validation_end": pd.Timestamp(validation_end).isoformat(),
        }))
        train_end = validation_end
    return folds


def _vx_phase9_bootstrap_lower(values):
    x = np.asarray(values, dtype=float)
    x = x[np.isfinite(x)]
    if len(x) < PHASE9_MIN_WALK_FORWARD_FOLDS:
        return np.nan
    rng = np.random.default_rng(PHASE9_BOOTSTRAP_SEED)
    means = np.empty(PHASE9_BOOTSTRAP_REPEATS, dtype=float)
    for i in range(PHASE9_BOOTSTRAP_REPEATS):
        means[i] = float(np.mean(rng.choice(x, size=len(x), replace=True)))
    return float(np.quantile(means, 0.05))


def _vx_phase9_walk_forward_lambda(frame, weights):
    folds = _vx_phase9_rolling_origin_splits(frame)
    rows = []
    for _, validation, info in folds:
        for mw in weights:
            metrics = _vx_phase9_full_metrics(validation, _vx_phase9_market_lambda_builder(mw))
            rows.append({**info, "model_weight": float(mw), **metrics})
    out = pd.DataFrame(rows)
    result = {
        "status": f"WAITING_FOR_WALK_FORWARD_FOLDS: need {PHASE9_MIN_WALK_FORWARD_FOLDS}, have {len(folds)}",
        "folds": int(len(folds)), "selected_model_weight": None,
        "selected_market_weight": None, "approved": False,
    }
    if len(folds) < PHASE9_MIN_WALK_FORWARD_FOLDS or out.empty:
        return out, result

    aggregate = out.groupby("model_weight", as_index=False).agg(
        scoreline_log_loss=("scoreline_log_loss", "mean"),
        poisson_deviance=("poisson_deviance", "mean"),
        wdl_multiclass_brier=("wdl_multiclass_brier", "mean"),
        clean_sheet_brier=("clean_sheet_brier", "mean"),
    )
    aggregate = aggregate.sort_values(
        ["scoreline_log_loss", "poisson_deviance", "wdl_multiclass_brier", "model_weight"],
        ascending=[True, True, True, False], kind="stable",
    )
    selected = float(aggregate.iloc[0]["model_weight"])
    base = out.loc[out["model_weight"].eq(1.0)].set_index("fold")
    chosen = out.loc[out["model_weight"].eq(selected)].set_index("fold")
    common = base.index.intersection(chosen.index)
    ll_gain = (base.loc[common, "scoreline_log_loss"] - chosen.loc[common, "scoreline_log_loss"]).to_numpy(float)
    brier_gain = (base.loc[common, "wdl_multiclass_brier"] - chosen.loc[common, "wdl_multiclass_brier"]).to_numpy(float)
    cs_gain = (base.loc[common, "clean_sheet_brier"] - chosen.loc[common, "clean_sheet_brier"]).to_numpy(float)
    win_rate = float(np.mean(ll_gain > 0.0)) if len(ll_gain) else 0.0
    lower = _vx_phase9_bootstrap_lower(ll_gain)
    approved = bool(
        selected < 1.0
        and len(ll_gain) >= PHASE9_MIN_WALK_FORWARD_FOLDS
        and float(np.mean(ll_gain)) > 1e-4
        and float(np.mean(brier_gain)) >= -1e-4
        and float(np.mean(cs_gain)) >= -1e-4
        and np.isfinite(lower) and lower > 0.0
        and win_rate >= PHASE9_REQUIRED_FOLD_WIN_RATE
    )
    result.update({
        "status": "WALK_FORWARD_COMPLETE",
        "selected_model_weight": selected,
        "selected_market_weight": float(1.0 - selected),
        "mean_scoreline_log_loss_gain_vs_model": float(np.mean(ll_gain)),
        "bootstrap_95pct_lower_gain": float(lower),
        "fold_win_rate": win_rate,
        "mean_wdl_brier_gain_vs_model": float(np.mean(brier_gain)),
        "mean_clean_sheet_brier_gain_vs_model": float(np.mean(cs_gain)),
        "approved": approved,
    })
    return out, result


def _vx_phase9_walk_forward_probabilities(frame, builders):
    folds = _vx_phase9_rolling_origin_splits(frame)
    rows = []
    for _, validation, info in folds:
        for label, builder in builders.items():
            metrics = _vx_phase9_probability_metrics(validation, builder)
            rows.append({**info, "candidate": str(label), **metrics})
    out = pd.DataFrame(rows)
    result = {
        "status": f"WAITING_FOR_WALK_FORWARD_FOLDS: need {PHASE9_MIN_WALK_FORWARD_FOLDS}, have {len(folds)}",
        "folds": int(len(folds)), "selected_candidate": None, "approved": False,
    }
    if len(folds) < PHASE9_MIN_WALK_FORWARD_FOLDS or out.empty:
        return out, result
    aggregate = out.groupby("candidate", as_index=False).agg(
        wdl_log_loss=("wdl_log_loss", "mean"),
        wdl_multiclass_brier=("wdl_multiclass_brier", "mean"),
        wdl_calibration_ece=("wdl_calibration_ece", "mean"),
    ).sort_values(
        ["wdl_log_loss", "wdl_multiclass_brier", "wdl_calibration_ece", "candidate"],
        kind="stable",
    )
    selected = str(aggregate.iloc[0]["candidate"])
    base = out.loc[out["candidate"].eq("PURE_MODEL")].set_index("fold")
    chosen = out.loc[out["candidate"].eq(selected)].set_index("fold")
    common = base.index.intersection(chosen.index)
    ll_gain = (base.loc[common, "wdl_log_loss"] - chosen.loc[common, "wdl_log_loss"]).to_numpy(float)
    brier_gain = (base.loc[common, "wdl_multiclass_brier"] - chosen.loc[common, "wdl_multiclass_brier"]).to_numpy(float)
    ece_gain = (base.loc[common, "wdl_calibration_ece"] - chosen.loc[common, "wdl_calibration_ece"]).to_numpy(float)
    win_rate = float(np.mean(ll_gain > 0.0)) if len(ll_gain) else 0.0
    lower = _vx_phase9_bootstrap_lower(ll_gain)
    approved = bool(
        selected != "PURE_MODEL"
        and len(ll_gain) >= PHASE9_MIN_WALK_FORWARD_FOLDS
        and float(np.mean(ll_gain)) > 1e-4
        and float(np.mean(brier_gain)) >= -1e-4
        and float(np.mean(ece_gain)) >= -0.005
        and np.isfinite(lower) and lower > 0.0
        and win_rate >= PHASE9_REQUIRED_FOLD_WIN_RATE
    )
    result.update({
        "status": "WALK_FORWARD_COMPLETE",
        "selected_candidate": selected,
        "mean_wdl_log_loss_gain_vs_model": float(np.mean(ll_gain)),
        "bootstrap_95pct_lower_gain": float(lower),
        "fold_win_rate": win_rate,
        "mean_wdl_brier_gain_vs_model": float(np.mean(brier_gain)),
        "mean_calibration_ece_gain_vs_model": float(np.mean(ece_gain)),
        "approved": approved,
    })
    return out, result

# ----------------------------------------------------------------
# RUN PHASE-9 EXPERIMENTS
# ----------------------------------------------------------------
phase9_market_blend_validation, PHASE9_MARKET_RESULT = _vx_phase9_market_experiment(phase9_validation_frame)
phase9_elo_blend_validation, PHASE9_ELO_RESULT = _vx_phase9_elo_experiment(phase9_validation_frame)
phase9_combined_blend_validation, PHASE9_COMBINED_RESULT = _vx_phase9_combined_experiment(phase9_validation_frame)
phase9_market_walk_forward_validation, PHASE9_MARKET_WALK_FORWARD_RESULT = _vx_phase9_walk_forward_lambda(
    _vx_phase9_market_overlap(phase9_validation_frame), PHASE9_MARKET_MODEL_WEIGHTS
)
phase9_elo_walk_forward_validation, PHASE9_ELO_WALK_FORWARD_RESULT = _vx_phase9_walk_forward_probabilities(
    _vx_phase9_elo_overlap(phase9_validation_frame),
    {"PURE_MODEL": _vx_phase9_model_wdl, **{
        f"MODEL_{int(round(w*100))}_TEAM_ELO_{int(round((1-w)*100))}": _vx_phase9_elo_probability_builder(w)
        for w in PHASE9_ELO_MODEL_WEIGHTS if w < 1.0
    }}
)
phase9_combined_walk_forward_validation, PHASE9_COMBINED_WALK_FORWARD_RESULT = _vx_phase9_walk_forward_probabilities(
    _vx_phase9_combined_overlap(phase9_validation_frame),
    {"PURE_MODEL": _vx_phase9_model_wdl, "MODEL_80_MARKET_10_TEAM_ELO_10": _vx_phase9_combined_builder}
)
phase9_external_wdl_benchmark = _vx_phase9_external_wdl_benchmark(phase9_validation_frame)
phase9_model_only_baseline_metrics = _vx_phase9_full_metrics(
    phase9_validation_frame, lambda r: (float(r["model_lambda_home"]), float(r["model_lambda_away"]))
) if not phase9_validation_frame.empty else {}

# FDR calibration is intentionally not recalculated from any external blend.
# Phase 9 carries forward the model-only Phase-7 band validation as evidence.
_phase7_fdr = globals().get("phase7_fdr_band_validation")
phase9_model_only_fdr_band_validation = (
    _phase7_fdr.copy() if isinstance(_phase7_fdr, pd.DataFrame) else pd.DataFrame()
)

# Candidate qualification is evidence only. Production remains pure model.
PHASE9_QUALIFIED_CANDIDATES = []
if (PHASE9_MARKET_RESULT.get("holdout_approved") and PHASE9_MARKET_WALK_FORWARD_RESULT.get("approved")):
    PHASE9_QUALIFIED_CANDIDATES.append({
        "type": "MARKET_LAMBDA_BLEND",
        "model_weight": PHASE9_MARKET_RESULT.get("selected_model_weight"),
        "market_weight": PHASE9_MARKET_RESULT.get("selected_market_weight"),
    })
if (PHASE9_ELO_RESULT.get("holdout_approved") and PHASE9_ELO_WALK_FORWARD_RESULT.get("approved")):
    PHASE9_QUALIFIED_CANDIDATES.append({
        "type": "TEAM_ELO_WDL_BLEND",
        "model_weight": PHASE9_ELO_RESULT.get("selected_model_weight"),
        "elo_weight": PHASE9_ELO_RESULT.get("selected_elo_weight"),
    })
if (PHASE9_COMBINED_RESULT.get("holdout_approved") and PHASE9_COMBINED_WALK_FORWARD_RESULT.get("approved")):
    PHASE9_QUALIFIED_CANDIDATES.append({
        "type": "COMBINED_WDL_BLEND",
        **PHASE9_COMBINED_WEIGHTS,
    })

PHASE9_EVIDENCE_READY = bool(
    PHASE9_MARKET_RESULT.get("status") == "HOLDOUT_COMPLETE"
    or PHASE9_ELO_RESULT.get("status") == "HOLDOUT_COMPLETE"
    or PHASE9_COMBINED_RESULT.get("status") == "HOLDOUT_COMPLETE"
)
PHASE9_VALIDATION_STATUS = (
    "OOS_EVIDENCE_READY_FOR_REVIEW"
    if PHASE9_EVIDENCE_READY
    else "ACCUMULATING_DEADLINE_SAFE_EXTERNAL_BENCHMARK_HISTORY"
)

# Even a statistically qualifying candidate is NOT promoted automatically.
PHASE9_LIVE_BLEND_ACTIVE = False
PHASE9_READY_FOR_PLAYER_MODEL = False
PHASE9_PRODUCTION_SOURCE = "PURE_MODEL_ONLY"

# ----------------------------------------------------------------
# PERSIST REPORTS — validation evidence only.
# ----------------------------------------------------------------
PHASE9_OUTPUT_DIR = Path(DATA_DIR) / "PHASE9_BLEND_VALIDATION"
PHASE9_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
PHASE9_MARKET_PATH = PHASE9_OUTPUT_DIR / "phase9_market_blend_validation.json"
PHASE9_ELO_PATH = PHASE9_OUTPUT_DIR / "phase9_team_elo_blend_validation.json"
PHASE9_COMBINED_PATH = PHASE9_OUTPUT_DIR / "phase9_combined_blend_validation.json"
PHASE9_FDR_PATH = PHASE9_OUTPUT_DIR / "phase9_model_only_fdr_band_validation.json"
PHASE9_EXTERNAL_WDL_PATH = PHASE9_OUTPUT_DIR / "phase9_model_vs_market_vs_team_elo_wdl.json"
PHASE9_SUMMARY_PATH = PHASE9_OUTPUT_DIR / "phase9_validation_summary.json"
PHASE9_MARKET_WALK_FORWARD_PATH = PHASE9_OUTPUT_DIR / "phase9_market_walk_forward_validation.json"
PHASE9_ELO_WALK_FORWARD_PATH = PHASE9_OUTPUT_DIR / "phase9_team_elo_walk_forward_validation.json"
PHASE9_COMBINED_WALK_FORWARD_PATH = PHASE9_OUTPUT_DIR / "phase9_combined_walk_forward_validation.json"

phase9_market_blend_validation.to_json(PHASE9_MARKET_PATH, orient="records", date_format="iso", indent=2)
phase9_elo_blend_validation.to_json(PHASE9_ELO_PATH, orient="records", date_format="iso", indent=2)
phase9_combined_blend_validation.to_json(PHASE9_COMBINED_PATH, orient="records", date_format="iso", indent=2)
phase9_model_only_fdr_band_validation.to_json(PHASE9_FDR_PATH, orient="records", date_format="iso", indent=2)
phase9_external_wdl_benchmark.to_json(PHASE9_EXTERNAL_WDL_PATH, orient="records", date_format="iso", indent=2)
phase9_market_walk_forward_validation.to_json(PHASE9_MARKET_WALK_FORWARD_PATH, orient="records", date_format="iso", indent=2)
phase9_elo_walk_forward_validation.to_json(PHASE9_ELO_WALK_FORWARD_PATH, orient="records", date_format="iso", indent=2)
phase9_combined_walk_forward_validation.to_json(PHASE9_COMBINED_WALK_FORWARD_PATH, orient="records", date_format="iso", indent=2)

PHASE9_SUMMARY = {
    "phase9_version": PHASE9_VERSION,
    "calculated_at": PHASE9_TIMESTAMP,
    "production_model_label": PHASE9_PRODUCTION_MODEL_LABEL,
    "production_source": PHASE9_PRODUCTION_SOURCE,
    "live_blend_active": False,
    "phase8_market_blend_weight": float(globals().get("MARKET_BLEND_WEIGHT", 0.0)),
    "phase8_elo_blend_weight": float(globals().get("ELO_BLEND_WEIGHT", 0.0)),
    "completed_frozen_forecasts": int(len(phase9_completed_forecasts)),
    "model_only_baseline_metrics": phase9_model_only_baseline_metrics,
    "market_overlap_fixtures": int(len(_vx_phase9_market_overlap(phase9_validation_frame))),
    "team_elo_overlap_fixtures": int(len(_vx_phase9_elo_overlap(phase9_validation_frame))),
    "combined_overlap_fixtures": int(len(_vx_phase9_combined_overlap(phase9_validation_frame))),
    "validation_status": PHASE9_VALIDATION_STATUS,
    "market_experiment": PHASE9_MARKET_RESULT,
    "team_elo_experiment": PHASE9_ELO_RESULT,
    "combined_experiment": PHASE9_COMBINED_RESULT,
    "market_walk_forward": PHASE9_MARKET_WALK_FORWARD_RESULT,
    "team_elo_walk_forward": PHASE9_ELO_WALK_FORWARD_RESULT,
    "combined_walk_forward": PHASE9_COMBINED_WALK_FORWARD_RESULT,
    "qualified_candidates_for_manual_review": PHASE9_QUALIFIED_CANDIDATES,
    "automatic_promotion_allowed": False,
    "ready_for_player_model": False,
    "notes": [
        "Pure model remains production baseline.",
        "Market lambdas are the only external objects allowed in projected-goal blend experiments.",
        "Team Elo is evaluated only in W/D/L probability space; no Elo expected-goals shortcut is invented.",
        "FDR band validation remains model-only until a future blend is explicitly approved and versioned.",
    ],
}
PHASE9_SUMMARY_PATH.write_text(
    json.dumps(PHASE9_SUMMARY, ensure_ascii=False, indent=2, default=str), encoding="utf-8"
)

# Permanent validation-state summary for later manual promotion/release review.
PHASE9_HISTORY_DIR = Path(PROJECT_ROOT) / "MODEL_HISTORY" / "PHASE9_BLEND_VALIDATION"
PHASE9_HISTORY_DIR.mkdir(parents=True, exist_ok=True)
PHASE9_HISTORY_SUMMARY_PATH = PHASE9_HISTORY_DIR / "latest_phase9_validation_summary.json"
PHASE9_HISTORY_SUMMARY_PATH.write_text(
    json.dumps(PHASE9_SUMMARY, ensure_ascii=False, indent=2, default=str), encoding="utf-8"
)

# ----------------------------------------------------------------
# FINAL INTEGRITY GATE
# ----------------------------------------------------------------
_PHASE9_STATE_AFTER = {
    name: _vx_phase9_digest(globals().get(name)) for name in _PHASE9_IMMUTABLE_NAMES
}
_PHASE9_WEIGHT_STATE_AFTER = (
    float(globals().get("MARKET_BLEND_WEIGHT", 0.0)),
    float(globals().get("ELO_BLEND_WEIGHT", 0.0)),
)
if _PHASE9_STATE_BEFORE != _PHASE9_STATE_AFTER:
    raise RuntimeError("Phase 9 integrity failure: blend validation mutated an upstream model/benchmark table.")
if _PHASE9_WEIGHT_STATE_BEFORE != _PHASE9_WEIGHT_STATE_AFTER or _PHASE9_WEIGHT_STATE_AFTER != (0.0, 0.0):
    raise RuntimeError("Phase 9 integrity failure: Phase-8 live blend weights changed.")

if "vx_phase0_assert_cleanroom" in globals():
    vx_phase0_assert_cleanroom("after Phase 9 blend validation")

print("✅ PHASE 9 — OUT-OF-SAMPLE BLEND TESTING + CALIBRATION BUILT")
print(f"   Completed frozen forecasts: {len(phase9_completed_forecasts)}")
print(f"   Market overlap: {PHASE9_SUMMARY['market_overlap_fixtures']}")
print(f"   Team Elo overlap: {PHASE9_SUMMARY['team_elo_overlap_fixtures']}")
print(f"   Combined overlap: {PHASE9_SUMMARY['combined_overlap_fixtures']}")
print(f"   Validation status: {PHASE9_VALIDATION_STATUS}")
print(f"   Qualified candidates for manual review: {len(PHASE9_QUALIFIED_CANDIDATES)}")
print(f"   Market walk-forward: {PHASE9_MARKET_WALK_FORWARD_RESULT.get('status')}")
print(f"   Team Elo walk-forward: {PHASE9_ELO_WALK_FORWARD_RESULT.get('status')}")
print(f"   Combined walk-forward: {PHASE9_COMBINED_WALK_FORWARD_RESULT.get('status')}")
print("🔒 Production remains PURE MODEL ONLY.")
print("🔒 MARKET_BLEND_WEIGHT = 0.00 and ELO_BLEND_WEIGHT = 0.00 remain unchanged.")
print("🔒 No market/Elo blend is automatically promoted.")
print("🔒 No player expected-points, transfers, captaincy or squad decisions are built here.")
print("📊 Phase 9 summary:", PHASE9_SUMMARY_PATH)
```

```python
# ============================================================
# CELL 13D — PHASE 10
# PLAYER EXPECTED-POINTS ENGINE
# TEAM FORECAST -> PLAYER OPPORTUNITY -> PLAYER EXPECTED POINTS
# NO TRANSFER RANKINGS • NO CAPTAIN RANKINGS • NO SQUAD OPTIMISATION
# ============================================================
# Production policy:
#   - consumes model-only Phase 3/4 team outputs;
#   - market and Team Elo remain external benchmarks (live weights must be 0.00);
#   - player popularity/ownership is never an xPts input;
#   - official FPL fixture difficulty is never an input;
#   - calculations may be produced as RESEARCH_ONLY while team-model validation
#     is still accumulating, but cannot feed Phase 11 until manually approved.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import math
import re
import numpy as np
import pandas as pd

PHASE10_VERSION = "VORTEX-V2.2-PLAYER-XPTS-HARDENED-DISTRIBUTION"
PHASE10_TIMESTAMP = datetime.now(timezone.utc).isoformat()
PHASE10_FPL_CORE_ROOT = Path(FPL_VORTEX_DATA_ROOT) / "00_SOURCE_ARCHIVE" / "FPL_CORE" / "data"
PHASE10_RECENCY_HALF_LIFE_DAYS = 90.0
PHASE10_RATE_SHRINKAGE_MINUTES = 900.0
PHASE10_MAX_ROLE_OBSERVATIONS = 24
PHASE10_MAX_BONUS_OBSERVATIONS = 16

# Explicit manual release gate. Building/calculating the research engine is allowed;
# promoting it into decision tools is not.
PHASE10_MANUAL_TEAM_MODEL_APPROVAL = bool(globals().get("PHASE10_MANUAL_TEAM_MODEL_APPROVAL", False))
PHASE10_TEAM_VALIDATION_EVIDENCE = bool(globals().get("PHASE7_VALIDATION_EVIDENCE_READY", False))
PHASE10_BLEND_REVIEW_COMPLETE = bool(globals().get("PHASE9_EVIDENCE_READY", False))
PHASE10_PRODUCTION_ELIGIBLE = bool(
    PHASE10_MANUAL_TEAM_MODEL_APPROVAL and PHASE10_TEAM_VALIDATION_EVIDENCE
)
PHASE10_OUTPUT_MODE = (
    "PRODUCTION_ELIGIBLE_MODEL_ONLY"
    if PHASE10_PRODUCTION_ELIGIBLE
    else "RESEARCH_ONLY_WAITING_FOR_TEAM_MODEL_VALIDATION_APPROVAL"
)
PHASE10_READY = False
PHASE10_READY_FOR_DECISION_ENGINE = False

# The V1 team forecast remains model-only. Phase 10 must not silently promote a
# Phase-9 blend candidate.
if float(globals().get("MARKET_BLEND_WEIGHT", 0.0)) != 0.0:
    raise RuntimeError("Phase 10 policy violation: MARKET_BLEND_WEIGHT must remain 0.00.")
if float(globals().get("ELO_BLEND_WEIGHT", 0.0)) != 0.0:
    raise RuntimeError("Phase 10 policy violation: ELO_BLEND_WEIGHT must remain 0.00.")
if bool(globals().get("PHASE9_LIVE_BLEND_ACTIVE", False)):
    raise RuntimeError("Phase 10 policy violation: Phase-9 live blend must remain disabled.")

for _name in (
    "projected_goals_model_only",
    "fixture_probabilities_model_only",
    "fixture_fdr_model_only",
):
    _frame = globals().get(_name)
    if not isinstance(_frame, pd.DataFrame) or _frame.empty:
        raise RuntimeError(f"Phase 10 requires non-empty {_name} from Phases 3–4.")

if not isinstance(globals().get("players"), pd.DataFrame) or globals()["players"].empty:
    raise RuntimeError("Phase 10 requires the canonical current FPL players table.")
if not isinstance(globals().get("teams"), pd.DataFrame) or globals()["teams"].empty:
    raise RuntimeError("Phase 10 requires the canonical current FPL teams table.")

# ----------------------------------------------------------------
# UPSTREAM IMMUTABILITY GATE
# ----------------------------------------------------------------
_PHASE10_IMMUTABLE = (
    "projected_goals_model_only",
    "fixture_probabilities_model_only",
    "fixture_fdr_model_only",
)
_PHASE10_STATE_BEFORE = {k: vx_frame_digest(globals()[k]) for k in _PHASE10_IMMUTABLE}

# ----------------------------------------------------------------
# 2026/27 FPL SCORING CONTRACT
# ----------------------------------------------------------------
PHASE10_GOAL_POINTS = {"GK": 10.0, "DEF": 6.0, "MID": 5.0, "FWD": 4.0}
PHASE10_ASSIST_POINTS = 3.0
PHASE10_CS_POINTS = {"GK": 4.0, "DEF": 4.0, "MID": 1.0, "FWD": 0.0}
PHASE10_DEFCON_THRESHOLDS = {"DEF": 10.0, "MID": 12.0, "FWD": 12.0}
PHASE10_DEFCON_POINTS = 2.0
PHASE10_YELLOW_POINTS = -1.0
PHASE10_RED_POINTS = -3.0
PHASE10_OWN_GOAL_POINTS = -2.0
PHASE10_PENALTY_MISS_POINTS = -2.0
PHASE10_PENALTY_SAVE_POINTS = 5.0

# ----------------------------------------------------------------
# CURRENT PLAYER MASTER — stable player_code is the cross-season key.
# ----------------------------------------------------------------
_phase10_players = globals()["players"].copy()
_required_current = {"id", "code", "team", "position", "price"}
_missing_current = sorted(_required_current - set(_phase10_players.columns))
if _missing_current:
    raise RuntimeError(f"Phase 10 current player table missing fields: {_missing_current}")

_phase10_players["player_id"] = pd.to_numeric(_phase10_players["id"], errors="coerce").astype("Int64")
_phase10_players["player_code"] = pd.to_numeric(_phase10_players["code"], errors="coerce").astype("Int64")
_phase10_players["team_id"] = pd.to_numeric(_phase10_players["team"], errors="coerce").astype("Int64")
_phase10_players["position"] = _phase10_players["position"].astype(str).str.upper().replace({"GKP": "GK"})
_phase10_players["price"] = pd.to_numeric(_phase10_players["price"], errors="coerce")
_phase10_players = _phase10_players.dropna(subset=["player_id", "player_code", "team_id"]).copy()
_phase10_players[["player_id", "player_code", "team_id"]] = _phase10_players[["player_id", "player_code", "team_id"]].astype(int)

_team_name = globals()["teams"].set_index("id")["name"].to_dict()
_phase10_players["team_name"] = _phase10_players["team_id"].map(_team_name)
_phase10_players["player_name"] = _phase10_players.get("web_name", _phase10_players.get("second_name", _phase10_players["player_id"].astype(str))).astype(str)

# ----------------------------------------------------------------
# FPL CORE SOURCE ADAPTER
# Uses all competitions for role/minutes evidence, but Premier League only for
# attacking rates. player_code joins season-specific FPL player_ids safely.
# ----------------------------------------------------------------
def _vx_phase10_core_dir(season_label):
    text = str(season_label)
    m = re.match(r"^(\d{4})-(\d{2}|\d{4})$", text)
    if not m:
        return PHASE10_FPL_CORE_ROOT / text
    y = int(m.group(1))
    return PHASE10_FPL_CORE_ROOT / f"{y}-{y+1}"

_phase10_current_label = globals().get("CURRENT_SEASON", None)
_phase10_previous_label = globals().get("PREVIOUS_SEASON", None)
if _phase10_current_label is None:
    _now = pd.Timestamp(PHASE10_TIMESTAMP)
    _start = int(_now.year if _now.month >= 7 else _now.year - 1)
    _phase10_current_label = f"{_start}-{str(_start+1)[-2:]}"
if _phase10_previous_label is None:
    _start = int(str(_phase10_current_label).split("-")[0])
    _phase10_previous_label = f"{_start-1}-{str(_start)[-2:]}"

PHASE10_SOURCE_SEASONS = (str(_phase10_previous_label), str(_phase10_current_label))


def _vx_phase10_read_csv(path):
    path = Path(path)
    if not path.exists() or path.stat().st_size <= 1:
        return pd.DataFrame()
    try:
        return pd.read_csv(path, low_memory=False)
    except Exception as exc:
        print(f"⚠️ Phase 10 skipped unreadable file {path}: {exc}")
        return pd.DataFrame()


def _vx_phase10_player_map(season_dir):
    p = Path(season_dir) / "players.csv"
    m = _vx_phase10_read_csv(p)
    needed = {"player_code", "player_id", "team_code", "position"}
    if m.empty or not needed.issubset(m.columns):
        return pd.DataFrame(columns=["player_code", "season_player_id", "season_team_code", "historical_position"])
    out = m[["player_code", "player_id", "team_code", "position"]].copy()
    out = out.rename(columns={
        "player_id": "season_player_id",
        "team_code": "season_team_code",
        "position": "historical_position",
    })
    for c in ("player_code", "season_player_id", "season_team_code"):
        out[c] = pd.to_numeric(out[c], errors="coerce")
    out = out.dropna(subset=["player_code", "season_player_id"]).copy()
    out[["player_code", "season_player_id"]] = out[["player_code", "season_player_id"]].astype(int)
    out["historical_position"] = out["historical_position"].astype(str).str.upper().map({
        "GOALKEEPER": "GK", "DEFENDER": "DEF", "MIDFIELDER": "MID", "FORWARD": "FWD",
        "GK": "GK", "DEF": "DEF", "MID": "MID", "FWD": "FWD",
    }).fillna(out["historical_position"].astype(str).str.upper())
    return out.drop_duplicates("season_player_id", keep="last")


def _vx_phase10_match_date_map(matches):
    if matches.empty or "match_id" not in matches.columns:
        return pd.DataFrame(columns=["match_id", "match_date"])
    date_col = "kickoff_time" if "kickoff_time" in matches.columns else None
    if date_col is None:
        return pd.DataFrame(columns=["match_id", "match_date"])
    out = matches[["match_id", date_col]].copy().rename(columns={date_col: "match_date"})
    out["match_date"] = pd.to_datetime(out["match_date"], utc=True, errors="coerce")
    return out.dropna(subset=["match_id"]).drop_duplicates("match_id", keep="last")


def _vx_phase10_load_core_evidence():
    pm_frames, lineup_frames, shot_frames, gw_frames = [], [], [], []
    source_files = []

    for season_label in PHASE10_SOURCE_SEASONS:
        season_dir = _vx_phase10_core_dir(season_label)
        pmap = _vx_phase10_player_map(season_dir)
        if pmap.empty:
            continue
        by_gw = season_dir / "By Gameweek"
        if not by_gw.exists():
            continue
        gw_dirs = sorted(
            [p for p in by_gw.glob("GW*") if p.is_dir()],
            key=lambda p: int(re.sub(r"\D", "", p.name) or 999),
        )
        for gw_dir in gw_dirs:
            gw_num = int(re.sub(r"\D", "", gw_dir.name) or 0)
            pm = _vx_phase10_read_csv(gw_dir / "playermatchstats.csv")
            lu = _vx_phase10_read_csv(gw_dir / "lineups.csv")
            sh = _vx_phase10_read_csv(gw_dir / "shots.csv")
            mt = _vx_phase10_read_csv(gw_dir / "matches.csv")
            pg = _vx_phase10_read_csv(gw_dir / "player_gameweek_stats.csv")
            dates = _vx_phase10_match_date_map(mt)

            if not pm.empty and {"player_id", "match_id"}.issubset(pm.columns):
                x = pm.copy()
                x["season_player_id"] = pd.to_numeric(x["player_id"], errors="coerce")
                x = x.merge(pmap, on="season_player_id", how="inner")
                x = x.merge(dates, on="match_id", how="left")
                x["season_label"] = season_label
                x["gameweek"] = gw_num
                pm_frames.append(x)
                source_files.append(str(gw_dir / "playermatchstats.csv"))

            if not lu.empty and {"player_id", "match_id"}.issubset(lu.columns):
                x = lu.copy()
                x["season_player_id"] = pd.to_numeric(x["player_id"], errors="coerce")
                x = x.merge(pmap, on="season_player_id", how="inner")
                x = x.merge(dates, on="match_id", how="left")
                x["season_label"] = season_label
                x["gameweek"] = gw_num
                lineup_frames.append(x)
                source_files.append(str(gw_dir / "lineups.csv"))

            if not sh.empty and {"player_id", "match_id", "xg"}.issubset(sh.columns):
                x = sh.copy()
                x["season_player_id"] = pd.to_numeric(x["player_id"], errors="coerce")
                x = x.merge(pmap, on="season_player_id", how="inner")
                x = x.merge(dates, on="match_id", how="left")
                x["season_label"] = season_label
                x["gameweek"] = gw_num
                shot_frames.append(x)
                source_files.append(str(gw_dir / "shots.csv"))

            if not pg.empty and "id" in pg.columns:
                x = pg.copy()
                x["season_player_id"] = pd.to_numeric(x["id"], errors="coerce")
                x = x.merge(pmap, on="season_player_id", how="inner")
                x["season_label"] = season_label
                x["gameweek"] = pd.to_numeric(x.get("gw", gw_num), errors="coerce").fillna(gw_num).astype(int)
                gw_frames.append(x)
                source_files.append(str(gw_dir / "player_gameweek_stats.csv"))

    pm_all = pd.concat(pm_frames, ignore_index=True, sort=False) if pm_frames else pd.DataFrame()
    lineups_all = pd.concat(lineup_frames, ignore_index=True, sort=False) if lineup_frames else pd.DataFrame()
    shots_all = pd.concat(shot_frames, ignore_index=True, sort=False) if shot_frames else pd.DataFrame()
    player_gw_all = pd.concat(gw_frames, ignore_index=True, sort=False) if gw_frames else pd.DataFrame()

    if not pm_all.empty:
        pm_all = pm_all.drop_duplicates(["season_label", "match_id", "player_code"], keep="last")
    if not lineups_all.empty:
        lineups_all = lineups_all.drop_duplicates(["season_label", "match_id", "player_code"], keep="last")
    if not shots_all.empty and "shot_index" in shots_all.columns:
        shots_all = shots_all.drop_duplicates(["season_label", "match_id", "shot_index"], keep="last")
    if not player_gw_all.empty:
        player_gw_all = player_gw_all.drop_duplicates(["season_label", "gameweek", "player_code"], keep="last")

    return pm_all, lineups_all, shots_all, player_gw_all, sorted(set(source_files))


phase10_player_match_all, phase10_lineups_all, phase10_shots_all, phase10_player_gw_all, PHASE10_SOURCE_FILES = _vx_phase10_load_core_evidence()

# League FPL-assist probability per actual goal. xA determines WHO is likely to
# assist; this rate determines HOW MANY scored goals receive an official FPL assist.
def _vx_phase10_assist_per_goal_rate():
    g = phase10_player_gw_all.copy()
    if g.empty or not {"goals_scored", "assists"}.issubset(g.columns):
        return np.nan
    goals = float(pd.to_numeric(g["goals_scored"], errors="coerce").fillna(0.0).sum())
    assists = float(pd.to_numeric(g["assists"], errors="coerce").fillna(0.0).sum())
    if goals <= 0:
        return np.nan
    return float(np.clip(assists / goals, 0.0, 1.0))

PHASE10_FPL_ASSIST_PER_GOAL_RATE = _vx_phase10_assist_per_goal_rate()
if not np.isfinite(PHASE10_FPL_ASSIST_PER_GOAL_RATE):
    # No fabricated assist rate: player xPts will remain blocked until approved
    # historical official FPL goal/assist evidence exists.
    PHASE10_FPL_ASSIST_PER_GOAL_RATE = np.nan

# Strict forecast cutoff: no post-forecast player evidence can enter the model.
_phase10_cutoff = pd.Timestamp(projected_goals_model_only["forecast_timestamp"].dropna().iloc[0])
if _phase10_cutoff.tzinfo is None:
    _phase10_cutoff = _phase10_cutoff.tz_localize("UTC")
else:
    _phase10_cutoff = _phase10_cutoff.tz_convert("UTC")

for _frame_name in ("phase10_player_match_all", "phase10_lineups_all", "phase10_shots_all"):
    _f = globals()[_frame_name]
    if not _f.empty and "match_date" in _f.columns:
        _f["match_date"] = pd.to_datetime(_f["match_date"], utc=True, errors="coerce")
        globals()[_frame_name] = _f.loc[_f["match_date"].isna() | (_f["match_date"] <= _phase10_cutoff)].copy()

# ----------------------------------------------------------------
# BUILD MATCH-LEVEL ROLE + ATTACK EVIDENCE
# All competitions may inform role/minutes. PL only informs scoring rates.
# ----------------------------------------------------------------
def _vx_phase10_is_prem(match_id):
    return "-prem-" in str(match_id).lower()

if not phase10_player_match_all.empty:
    _pm = phase10_player_match_all.copy()
    for c in ("minutes_played", "xg", "xa", "saves", "defensive_contributions"):
        if c not in _pm.columns:
            _pm[c] = 0.0
        _pm[c] = pd.to_numeric(_pm[c], errors="coerce").fillna(0.0)
else:
    _pm = pd.DataFrame(columns=["season_label", "match_id", "player_code", "match_date", "minutes_played", "xg", "xa", "saves", "defensive_contributions"])

# Penalty xG is taken directly from shot-level situation labels; no assumed fixed
# penalty xG constant is subtracted.
_pen = pd.DataFrame(columns=["season_label", "match_id", "player_code", "penalty_xg", "penalty_attempts"])
if not phase10_shots_all.empty and {"situation", "xg", "player_code", "match_id", "season_label"}.issubset(phase10_shots_all.columns):
    _shots = phase10_shots_all.copy()
    _shots = _shots.loc[_shots["match_id"].map(_vx_phase10_is_prem)].copy()
    _shots["xg"] = pd.to_numeric(_shots["xg"], errors="coerce").fillna(0.0)
    _shots["is_penalty"] = _shots["situation"].astype(str).str.lower().eq("penalty")
    _ps = _shots.loc[_shots["is_penalty"]].copy()
    if not _ps.empty:
        _pen = _ps.groupby(["season_label", "match_id", "player_code"], as_index=False).agg(
            penalty_xg=("xg", "sum"), penalty_attempts=("xg", "size")
        )

_pm = _pm.merge(_pen, on=["season_label", "match_id", "player_code"], how="left")
_pm["penalty_xg"] = pd.to_numeric(_pm.get("penalty_xg", 0.0), errors="coerce").fillna(0.0)
_pm["penalty_attempts"] = pd.to_numeric(_pm.get("penalty_attempts", 0.0), errors="coerce").fillna(0.0)
_pm["npxg"] = (pd.to_numeric(_pm["xg"], errors="coerce").fillna(0.0) - _pm["penalty_xg"]).clip(lower=0.0)
_pm["is_premier_league"] = _pm["match_id"].map(_vx_phase10_is_prem)

# Lineup is the base for role evidence because it includes unused substitutes.
if not phase10_lineups_all.empty:
    _role = phase10_lineups_all[[c for c in ["season_label", "match_id", "player_code", "match_date", "is_starting", "lineup_status"] if c in phase10_lineups_all.columns]].copy()
    _role = _role.merge(
        _pm[["season_label", "match_id", "player_code", "minutes_played", "historical_position"]].drop_duplicates(["season_label", "match_id", "player_code"]),
        on=["season_label", "match_id", "player_code"], how="left"
    )
    _role["minutes_played"] = pd.to_numeric(_role.get("minutes_played", 0.0), errors="coerce").fillna(0.0)
    _role["is_starting"] = _role.get("is_starting", False).map(lambda v: str(v).strip().lower() in {"true", "1", "yes", "y"})
else:
    _role = _pm[["season_label", "match_id", "player_code", "match_date", "minutes_played", "historical_position"]].copy()
    _role["is_starting"] = (
        pd.to_numeric(_pm.get("start_min", np.nan), errors="coerce").fillna(999).eq(0)
        & _role["minutes_played"].gt(0)
    )

_role["appeared"] = pd.to_numeric(_role["minutes_played"], errors="coerce").fillna(0.0).gt(0)
_role["played_60"] = pd.to_numeric(_role["minutes_played"], errors="coerce").fillna(0.0).ge(60)
_role["match_date"] = pd.to_datetime(_role.get("match_date"), utc=True, errors="coerce")


def _vx_phase10_recency_weight(ts):
    if pd.isna(ts):
        return 0.25
    days = max(0.0, float((_phase10_cutoff - pd.Timestamp(ts)).total_seconds() / 86400.0))
    return float(0.5 ** (days / PHASE10_RECENCY_HALF_LIFE_DAYS))

_role["recency_weight"] = _role["match_date"].map(_vx_phase10_recency_weight)
_pm["match_date"] = pd.to_datetime(_pm.get("match_date"), utc=True, errors="coerce")
_pm["recency_weight"] = _pm["match_date"].map(_vx_phase10_recency_weight)

# ----------------------------------------------------------------
# POSITION PRIORS DERIVED FROM PL EVIDENCE (not hard-coded player talent).
# ----------------------------------------------------------------
def _vx_phase10_weighted_rate(frame, numer, denom="minutes_played", scale=90.0):
    if frame.empty:
        return np.nan
    w = pd.to_numeric(frame.get("recency_weight", 1.0), errors="coerce").fillna(0.0).to_numpy(float)
    n = pd.to_numeric(frame.get(numer, 0.0), errors="coerce").fillna(0.0).to_numpy(float)
    d = pd.to_numeric(frame.get(denom, 0.0), errors="coerce").fillna(0.0).to_numpy(float)
    den = float(np.sum(w * d))
    return float(scale * np.sum(w * n) / den) if den > 0 else np.nan

_pm_pl = _pm.loc[_pm["is_premier_league"]].copy()
_phase10_pos_prior_rows = []
for _pos in ("GK", "DEF", "MID", "FWD"):
    g = _pm_pl.loc[_pm_pl["historical_position"].astype(str).str.upper().map({
        "GOALKEEPER":"GK", "DEFENDER":"DEF", "MIDFIELDER":"MID", "FORWARD":"FWD",
        "GK":"GK", "DEF":"DEF", "MID":"MID", "FWD":"FWD",
    }).eq(_pos)].copy() if not _pm_pl.empty else pd.DataFrame()
    _phase10_pos_prior_rows.append({
        "position": _pos,
        "npxg_per90_prior": _vx_phase10_weighted_rate(g, "npxg"),
        "xa_per90_prior": _vx_phase10_weighted_rate(g, "xa"),
        "penalty_xg_per90_prior": _vx_phase10_weighted_rate(g, "penalty_xg"),
        "saves_per90_prior": _vx_phase10_weighted_rate(g, "saves") if _pos == "GK" else 0.0,
    })
phase10_position_rate_priors = pd.DataFrame(_phase10_pos_prior_rows).set_index("position")
for _c in ["npxg_per90_prior", "xa_per90_prior", "penalty_xg_per90_prior", "saves_per90_prior"]:
    if _c in phase10_position_rate_priors.columns:
        phase10_position_rate_priors[_c] = pd.to_numeric(phase10_position_rate_priors[_c], errors="coerce").fillna(0.0)

# ----------------------------------------------------------------
# PLAYER-ROLE OVERRIDES — explicit/provenance-gated, never silent.
# ----------------------------------------------------------------
# Reuse the approved persistent raw-input root created upstream in Phase 1.
# Player-role overrides are model inputs, not final video DATA artifacts.
PHASE10_OVERRIDE_DIR = Path(PHASE1_RAW_INPUT_DIR)
PHASE10_OVERRIDE_DIR.mkdir(parents=True, exist_ok=True)
PHASE10_OVERRIDE_PATH = PHASE10_OVERRIDE_DIR / "player_role_overrides.csv"

# One-time migration for a runtime started before this path correction.
_phase10_legacy_override_path = Path(DATA_DIR) / "RAW_INPUT" / "player_role_overrides.csv"
if _phase10_legacy_override_path.exists():
    if not PHASE10_OVERRIDE_PATH.exists():
        _phase10_legacy_override_path.replace(PHASE10_OVERRIDE_PATH)
    else:
        _phase10_legacy_override_path.unlink()
    _phase10_legacy_override_dir = _phase10_legacy_override_path.parent
    if _phase10_legacy_override_dir.exists() and not any(_phase10_legacy_override_dir.iterdir()):
        _phase10_legacy_override_dir.rmdir()
PHASE10_OVERRIDE_FIELDS = (
    "start_probability_override",
    "expected_minutes_override",
    "npxg_per90_override",
    "xa_per90_override",
    "penalty_taker_probability_override",
    "defcon_probability_override",
    "bonus_probability_override",
)


def _vx_phase10_override_template():
    rows = []
    for _, r in _phase10_players.iterrows():
        row = {
            "player_id": int(r["player_id"]),
            "player_name": str(r["player_name"]),
            "team_id": int(r["team_id"]),
            "override_active": False,
            "set_piece_role_override": "",
            "reason": "",
            "source_name": "",
            "source_timestamp": "",
            "is_pre_deadline_data": False,
        }
        for c in PHASE10_OVERRIDE_FIELDS:
            row[c] = np.nan
        rows.append(row)
    return pd.DataFrame(rows)

if PHASE10_OVERRIDE_PATH.exists():
    phase10_player_overrides = pd.read_csv(PHASE10_OVERRIDE_PATH)
    template = _vx_phase10_override_template()
    if "player_id" not in phase10_player_overrides.columns:
        raise RuntimeError("Phase 10 override file is missing player_id")
    existing = set(pd.to_numeric(phase10_player_overrides["player_id"], errors="coerce").dropna().astype(int))
    missing = template.loc[~template["player_id"].isin(existing)]
    if not missing.empty:
        phase10_player_overrides = pd.concat([phase10_player_overrides, missing], ignore_index=True, sort=False)
        phase10_player_overrides.to_csv(PHASE10_OVERRIDE_PATH, index=False)
else:
    phase10_player_overrides = _vx_phase10_override_template()
    phase10_player_overrides.to_csv(PHASE10_OVERRIDE_PATH, index=False)

for c in PHASE10_OVERRIDE_FIELDS:
    if c not in phase10_player_overrides.columns:
        phase10_player_overrides[c] = np.nan
    phase10_player_overrides[c] = pd.to_numeric(phase10_player_overrides[c], errors="coerce")
for c in ("override_active", "is_pre_deadline_data"):
    if c not in phase10_player_overrides.columns:
        phase10_player_overrides[c] = False
    phase10_player_overrides[c] = phase10_player_overrides[c].map(lambda v: str(v).strip().lower() in {"true", "1", "yes", "y"})

_active_ov = phase10_player_overrides.loc[phase10_player_overrides["override_active"]].copy()
if not _active_ov.empty:
    if (~_active_ov["is_pre_deadline_data"]).any():
        raise RuntimeError("Every active Phase-10 player override must be pre-deadline data.")
    for c in ("reason", "source_name", "source_timestamp"):
        if _active_ov.get(c, "").astype(str).str.strip().eq("").any():
            raise RuntimeError(f"Every active Phase-10 player override requires {c}.")
    _ov_ts = pd.to_datetime(_active_ov["source_timestamp"], utc=True, errors="coerce")
    if _ov_ts.isna().any() or (_ov_ts > _phase10_cutoff).any():
        raise RuntimeError("Phase-10 player override timestamps must be valid and no later than forecast time.")

_override_idx = phase10_player_overrides.set_index("player_id", drop=False)

# ----------------------------------------------------------------
# CURRENT AVAILABILITY + SET-PIECE ROLE
# ----------------------------------------------------------------
def _vx_phase10_availability(row):
    chance = pd.to_numeric(pd.Series([row.get("chance_of_playing_next_round", np.nan)]), errors="coerce").iloc[0]
    if np.isfinite(chance):
        return float(np.clip(chance / 100.0, 0.0, 1.0))
    status = str(row.get("status", "a")).lower().strip()
    return {"a": 1.0, "d": 0.75, "i": 0.0, "s": 0.0, "u": 0.0}.get(status, 0.75)


def _vx_phase10_set_piece_role(row):
    bits = []
    fields = (
        ("PEN", "penalties_order"),
        ("DFK", "direct_freekicks_order"),
        ("COR", "corners_and_indirect_freekicks_order"),
    )
    for label, col in fields:
        value = pd.to_numeric(pd.Series([row.get(col, np.nan)]), errors="coerce").iloc[0]
        if np.isfinite(value) and value > 0:
            bits.append(f"{label}{int(value)}")
    return "|".join(bits) if bits else "NONE"

# ----------------------------------------------------------------
# PLAYER EVIDENCE FEATURES
# ----------------------------------------------------------------
def _vx_phase10_weighted_mean(values, weights):
    v = np.asarray(values, dtype=float)
    w = np.asarray(weights, dtype=float)
    mask = np.isfinite(v) & np.isfinite(w) & (w > 0)
    if not np.any(mask):
        return np.nan
    return float(np.average(v[mask], weights=w[mask]))


def _vx_phase10_shrunk_rate(player_frame, numer, position, prior_col):
    prior = float(phase10_position_rate_priors.loc[position, prior_col]) if position in phase10_position_rate_priors.index else 0.0
    obs = _vx_phase10_weighted_rate(player_frame, numer)
    w = pd.to_numeric(player_frame.get("recency_weight", 1.0), errors="coerce").fillna(0.0).to_numpy(float) if not player_frame.empty else np.array([])
    mins = pd.to_numeric(player_frame.get("minutes_played", 0.0), errors="coerce").fillna(0.0).to_numpy(float) if not player_frame.empty else np.array([])
    eff_minutes = float(np.sum(w * mins)) if len(w) else 0.0
    reliability = float(eff_minutes / (eff_minutes + PHASE10_RATE_SHRINKAGE_MINUTES)) if eff_minutes > 0 else 0.0
    if not np.isfinite(obs):
        obs = prior
    return float(reliability * obs + (1.0 - reliability) * prior), eff_minutes, reliability


def _vx_phase10_recent_gw(player_code):
    if phase10_player_gw_all.empty or "player_code" not in phase10_player_gw_all.columns:
        return pd.DataFrame()
    g = phase10_player_gw_all.loc[pd.to_numeric(phase10_player_gw_all["player_code"], errors="coerce").eq(int(player_code))].copy()
    if g.empty:
        return g
    season_order = {str(_phase10_previous_label): 0, str(_phase10_current_label): 1}
    g["_season_order"] = g["season_label"].astype(str).map(season_order).fillna(-1)
    g["gameweek"] = pd.to_numeric(g["gameweek"], errors="coerce").fillna(-1)
    g = g.sort_values(["_season_order", "gameweek"], kind="stable").tail(PHASE10_MAX_BONUS_OBSERVATIONS)
    g["_recent_weight"] = np.power(0.85, np.arange(len(g)-1, -1, -1, dtype=float))
    return g


# Current-regime BPS prior by position. Previous-season bonus is never used
# because the 2026/27 Bonus Points System changed. Once current-season evidence
# exists, thin player samples shrink toward a same-regime positional prior.
_phase10_current_bonus_prior = {}
if not phase10_player_gw_all.empty and {"season_label", "player_code", "bonus", "minutes"}.issubset(phase10_player_gw_all.columns):
    _cb = phase10_player_gw_all.loc[
        phase10_player_gw_all["season_label"].astype(str).eq(str(_phase10_current_label))
    ].copy()
    if not _cb.empty:
        _pmap = _phase10_players[["player_code", "position"]].drop_duplicates("player_code")
        _cb = _cb.merge(_pmap, on="player_code", how="left")
        _cb["minutes"] = pd.to_numeric(_cb["minutes"], errors="coerce").fillna(0.0)
        _cb["bonus"] = pd.to_numeric(_cb["bonus"], errors="coerce").fillna(0.0)
        _cb = _cb.loc[_cb["minutes"] > 0].copy()
        for _pos, _gpos in _cb.groupby("position", dropna=True):
            _phase10_current_bonus_prior[str(_pos)] = {
                "mean_points": float(_gpos["bonus"].mean()),
                "probability": float((_gpos["bonus"] > 0).mean()),
                "n": int(len(_gpos)),
            }

def _vx_phase10_player_features(row):
    pid = int(row["player_id"])
    pcode = int(row["player_code"])
    pos = str(row["position"])
    avail = _vx_phase10_availability(row)

    rg = _role.loc[pd.to_numeric(_role.get("player_code", pd.Series(dtype=float)), errors="coerce").eq(pcode)].copy() if not _role.empty else pd.DataFrame()
    if not rg.empty:
        rg = rg.sort_values("match_date", kind="stable").tail(PHASE10_MAX_ROLE_OBSERVATIONS)
        rw = pd.to_numeric(rg["recency_weight"], errors="coerce").fillna(0.0).to_numpy(float)
        start_rate = _vx_phase10_weighted_mean(rg["is_starting"].astype(float), rw)
        appearance_rate = _vx_phase10_weighted_mean(rg["appeared"].astype(float), rw)
        p60_rate = _vx_phase10_weighted_mean(rg["played_60"].astype(float), rw)
        minute_mean = _vx_phase10_weighted_mean(pd.to_numeric(rg["minutes_played"], errors="coerce").fillna(0.0), rw)
        role_n = int(len(rg))
    else:
        # Current official FPL starts/minutes are a descriptive fallback once the
        # season has begun. Popularity and FPL expected-points fields are not used.
        team_matches = 0
        _ts = globals().get("team_strength_updated")
        if isinstance(_ts, pd.DataFrame) and not _ts.empty and {"team_id", "completed_matches"}.issubset(_ts.columns):
            hit = _ts.loc[pd.to_numeric(_ts["team_id"], errors="coerce").eq(int(row["team_id"]))]
            if not hit.empty:
                team_matches = int(pd.to_numeric(hit.iloc[0]["completed_matches"], errors="coerce") or 0)
        if team_matches > 0:
            fpl_starts = float(pd.to_numeric(pd.Series([row.get("starts", 0.0)]), errors="coerce").fillna(0.0).iloc[0])
            fpl_minutes = float(pd.to_numeric(pd.Series([row.get("minutes", 0.0)]), errors="coerce").fillna(0.0).iloc[0])
            start_rate = float(np.clip(fpl_starts / team_matches, 0.0, 1.0))
            minute_mean = float(np.clip(fpl_minutes / team_matches, 0.0, 90.0))
            appearance_rate = float(np.clip(max(start_rate, minute_mean / 90.0), 0.0, 1.0))
            p60_rate = float(np.clip(start_rate * min(1.0, minute_mean / 60.0), 0.0, 1.0))
            role_n = int(team_matches)
        else:
            start_rate = appearance_rate = p60_rate = minute_mean = 0.0
            role_n = 0

    start_probability = float(np.clip((start_rate if np.isfinite(start_rate) else 0.0) * avail, 0.0, 1.0))
    appearance_probability = float(np.clip((appearance_rate if np.isfinite(appearance_rate) else 0.0) * avail, 0.0, 1.0))
    sixty_probability = float(np.clip((p60_rate if np.isfinite(p60_rate) else 0.0) * avail, 0.0, 1.0))
    expected_minutes = float(np.clip((minute_mean if np.isfinite(minute_mean) else 0.0) * avail, 0.0, 90.0))

    pg = _pm_pl.loc[pd.to_numeric(_pm_pl.get("player_code", pd.Series(dtype=float)), errors="coerce").eq(pcode)].copy() if not _pm_pl.empty else pd.DataFrame()
    npxg90, attack_minutes, attack_reliability = _vx_phase10_shrunk_rate(pg, "npxg", pos, "npxg_per90_prior")
    xa90, _, _ = _vx_phase10_shrunk_rate(pg, "xa", pos, "xa_per90_prior")
    penxg90, _, _ = _vx_phase10_shrunk_rate(pg, "penalty_xg", pos, "penalty_xg_per90_prior")
    saves90, _, _ = _vx_phase10_shrunk_rate(pg, "saves", pos, "saves_per90_prior") if pos == "GK" else (0.0, attack_minutes, attack_reliability)

    # Empirical defensive-contribution probability conditional on playing, shrunk
    # toward the position population when player evidence is thin.
    threshold = PHASE10_DEFCON_THRESHOLDS.get(pos, np.nan)
    defcon_cond = 0.0
    if np.isfinite(threshold):
        pp = pg.loc[pd.to_numeric(pg.get("minutes_played", 0), errors="coerce").gt(0)].copy()
        if not pp.empty:
            ww = pd.to_numeric(pp["recency_weight"], errors="coerce").fillna(0.0).to_numpy(float)
            yy = pd.to_numeric(pp.get("defensive_contributions", 0), errors="coerce").fillna(0.0).ge(threshold).astype(float).to_numpy()
            raw = _vx_phase10_weighted_mean(yy, ww)
            # Position population prior from the same PL evidence.
            pop = _pm_pl.loc[_pm_pl["historical_position"].astype(str).str.upper().map({
                "GOALKEEPER":"GK", "DEFENDER":"DEF", "MIDFIELDER":"MID", "FORWARD":"FWD",
                "GK":"GK", "DEF":"DEF", "MID":"MID", "FWD":"FWD",
            }).eq(pos) & pd.to_numeric(_pm_pl.get("minutes_played", 0), errors="coerce").gt(0)].copy()
            if not pop.empty:
                py = pd.to_numeric(pop.get("defensive_contributions", 0), errors="coerce").fillna(0.0).ge(threshold).astype(float)
                prior_p = float(py.mean())
            else:
                prior_p = 0.0
            n_eff = float(np.sum(ww) / max(float(np.max(ww)), 1e-9)) if len(ww) else 0.0
            raw = prior_p if not np.isfinite(raw) else raw
            defcon_cond = float((n_eff * raw + 6.0 * prior_p) / (n_eff + 6.0))
    defcon_probability = float(np.clip(appearance_probability * defcon_cond, 0.0, 1.0))

    # Official per-GW FPL outcomes inform cards/negative events across the recent
    # evidence window. BONUS is different: the 2026/27 BPS regime changed, so
    # previous-season bonus is never mixed into the current-regime expectation.
    gw = _vx_phase10_recent_gw(pcode)
    bonus_cond = yellow_cond = red_cond = own_cond = pmiss_cond = psave_cond = 0.0
    bonus_points_cond = 0.0
    if not gw.empty:
        gm = pd.to_numeric(gw.get("minutes", 0), errors="coerce").fillna(0.0)
        active = gm.gt(0)
        if active.any():
            gg = gw.loc[active].copy()
            ww = pd.to_numeric(gg["_recent_weight"], errors="coerce").fillna(0.0).to_numpy(float)
            def _m(col, default=0.0):
                v = pd.to_numeric(gg.get(col, default), errors="coerce").fillna(default).to_numpy(float)
                return _vx_phase10_weighted_mean(v, ww)
            yellow_cond = _m("yellow_cards")
            red_cond = _m("red_cards")
            own_cond = _m("own_goals")
            pmiss_cond = _m("penalties_missed")
            psave_cond = _m("penalties_saved")
            yellow_cond = float(yellow_cond) if np.isfinite(yellow_cond) else 0.0
            red_cond = float(red_cond) if np.isfinite(red_cond) else 0.0
            own_cond = float(own_cond) if np.isfinite(own_cond) else 0.0
            pmiss_cond = float(pmiss_cond) if np.isfinite(pmiss_cond) else 0.0
            psave_cond = float(psave_cond) if np.isfinite(psave_cond) else 0.0

        current_bonus = gw.loc[gw["season_label"].astype(str).eq(str(_phase10_current_label))].copy()
        if not current_bonus.empty:
            cbm = pd.to_numeric(current_bonus.get("minutes", 0), errors="coerce").fillna(0.0)
            current_bonus = current_bonus.loc[cbm.gt(0)].copy()
        if not current_bonus.empty:
            cbw = pd.to_numeric(current_bonus["_recent_weight"], errors="coerce").fillna(0.0).to_numpy(float)
            bvals = pd.to_numeric(current_bonus.get("bonus", 0), errors="coerce").fillna(0.0).to_numpy(float)
            raw_bonus_points = _vx_phase10_weighted_mean(bvals, cbw)
            raw_bonus_prob = _vx_phase10_weighted_mean((bvals > 0).astype(float), cbw)
            n_eff_bonus = float(np.sum(cbw) / max(float(np.max(cbw)), 1e-9)) if len(cbw) else 0.0
            prior = _phase10_current_bonus_prior.get(pos, {})
            prior_points = float(prior.get("mean_points", 0.0))
            prior_prob = float(prior.get("probability", 0.0))
            prior_n = 6.0 if prior else 0.0
            if np.isfinite(raw_bonus_points):
                bonus_points_cond = float(
                    (n_eff_bonus * float(raw_bonus_points) + prior_n * prior_points)
                    / max(n_eff_bonus + prior_n, 1e-9)
                )
            if np.isfinite(raw_bonus_prob):
                bonus_cond = float(
                    (n_eff_bonus * float(raw_bonus_prob) + prior_n * prior_prob)
                    / max(n_eff_bonus + prior_n, 1e-9)
                )
            bonus_points_cond = float(np.clip(bonus_points_cond, 0.0, 3.0))
            bonus_cond = float(np.clip(bonus_cond, 0.0, 1.0))
        elif pos in _phase10_current_bonus_prior:
            # A player with no current-season appearance inherits only the
            # same-regime positional prior; no 2025/26 bonus leaks across rules.
            prior = _phase10_current_bonus_prior[pos]
            bonus_points_cond = float(np.clip(prior["mean_points"], 0.0, 3.0))
            bonus_cond = float(np.clip(prior["probability"], 0.0, 1.0))

    # Penalty-taker probability proxy from current FPL role order + availability.
    pen_order = pd.to_numeric(pd.Series([row.get("penalties_order", np.nan)]), errors="coerce").iloc[0]
    penalty_role_score = float(math.exp(-1.5 * max(0.0, float(pen_order) - 1.0)) * max(appearance_probability, 0.05)) if np.isfinite(pen_order) and pen_order > 0 else 0.0
    if penalty_role_score <= 0 and penxg90 > 0:
        penalty_role_score = float(penxg90 * max(appearance_probability, 0.05))

    set_piece_role = _vx_phase10_set_piece_role(row)

    # Explicit player override, if activated and provenance-safe.
    if pid in _override_idx.index:
        ov = _override_idx.loc[pid]
        if isinstance(ov, pd.DataFrame):
            raise RuntimeError(f"Duplicate Phase-10 override rows for player_id={pid}")
        if bool(ov.get("override_active", False)):
            values = {
                "start_probability": "start_probability_override",
                "expected_minutes": "expected_minutes_override",
                "npxg_per90": "npxg_per90_override",
                "xa_per90": "xa_per90_override",
                "defcon_probability": "defcon_probability_override",
                "bonus_probability": "bonus_probability_override",
            }
            local = {
                "start_probability": start_probability,
                "expected_minutes": expected_minutes,
                "npxg_per90": npxg90,
                "xa_per90": xa90,
                "defcon_probability": defcon_probability,
                "bonus_probability": appearance_probability * bonus_cond,
            }
            for key, col in values.items():
                val = pd.to_numeric(pd.Series([ov.get(col, np.nan)]), errors="coerce").iloc[0]
                if np.isfinite(val):
                    local[key] = float(val)
            start_probability = float(np.clip(local["start_probability"], 0.0, 1.0))
            expected_minutes = float(np.clip(local["expected_minutes"], 0.0, 90.0))
            npxg90 = max(0.0, float(local["npxg_per90"]))
            xa90 = max(0.0, float(local["xa_per90"]))
            defcon_probability = float(np.clip(local["defcon_probability"], 0.0, 1.0))
            bonus_probability = float(np.clip(local["bonus_probability"], 0.0, 1.0))
            # Keep 60-min probability coherent with an explicit minutes override.
            sixty_probability = float(np.clip(max(sixty_probability, start_probability * min(1.0, expected_minutes / 60.0)), 0.0, 1.0))
            appearance_probability = float(np.clip(max(appearance_probability, start_probability, expected_minutes / 90.0), 0.0, 1.0))
            pto = pd.to_numeric(pd.Series([ov.get("penalty_taker_probability_override", np.nan)]), errors="coerce").iloc[0]
            if np.isfinite(pto):
                penalty_role_score = float(max(0.0, pto))
            spr = str(ov.get("set_piece_role_override", "")).strip()
            if spr:
                set_piece_role = spr
        else:
            bonus_probability = float(np.clip(appearance_probability * bonus_cond, 0.0, 1.0))
    else:
        bonus_probability = float(np.clip(appearance_probability * bonus_cond, 0.0, 1.0))

    # Evidence confidence combines role depth and attacking-rate minutes.
    role_conf = 1.0 - math.exp(-role_n / 8.0)
    rate_conf = float(attack_minutes / (attack_minutes + PHASE10_RATE_SHRINKAGE_MINUTES)) if attack_minutes > 0 else 0.0
    player_evidence_confidence = float(np.clip(0.55 * role_conf + 0.45 * rate_conf, 0.0, 1.0))

    return {
        "player_id": pid,
        "player_code": pcode,
        "player_name": str(row["player_name"]),
        "team_id": int(row["team_id"]),
        "team_name": str(row.get("team_name", "")),
        "position": pos,
        "price": float(row["price"]) if np.isfinite(row["price"]) else np.nan,
        "availability_probability": avail,
        "start_probability": start_probability,
        "appearance_probability": appearance_probability,
        "expected_minutes": expected_minutes,
        "sixty_minute_probability": sixty_probability,
        "npxg_per90": float(max(0.0, npxg90)),
        "xa_per90": float(max(0.0, xa90)),
        "penalty_xg_per90": float(max(0.0, penxg90)),
        "saves_per90": float(max(0.0, saves90)),
        "penalty_role_score": float(max(0.0, penalty_role_score)),
        "penalty_taker_probability": 0.0,  # normalized by team below
        "set_piece_role": set_piece_role,
        "defcon_probability": defcon_probability,
        "bonus_probability": bonus_probability,
        "bonus_points_per_appearance": float(max(0.0, bonus_points_cond)),
        "yellow_cards_per_appearance": float(max(0.0, yellow_cond)),
        "red_cards_per_appearance": float(max(0.0, red_cond)),
        "own_goals_per_appearance": float(max(0.0, own_cond)),
        "penalty_misses_per_appearance": float(max(0.0, pmiss_cond)),
        "penalty_saves_per_appearance": float(max(0.0, psave_cond)),
        "role_observations": role_n,
        "attack_effective_minutes": float(attack_minutes),
        "attack_rate_reliability": float(attack_reliability),
        "player_evidence_confidence": player_evidence_confidence,
    }

phase10_player_features = pd.DataFrame([
    _vx_phase10_player_features(r) for _, r in _phase10_players.iterrows()
])

# Normalize penalty order proxy within current team. This is a role probability,
# conditional on the team receiving a penalty; it is not a penalty-award forecast.
for _tid, _idx in phase10_player_features.groupby("team_id").groups.items():
    scores = pd.to_numeric(phase10_player_features.loc[_idx, "penalty_role_score"], errors="coerce").fillna(0.0)
    # A designated taker must also be expected to appear. This matters for late
    # injury/rotation overrides and keeps the conditional role probability live.
    scores = scores * pd.to_numeric(phase10_player_features.loc[_idx, "appearance_probability"], errors="coerce").fillna(0.0).clip(0.0, 1.0)
    total = float(scores.sum())
    if total > 0:
        phase10_player_features.loc[_idx, "penalty_taker_probability"] = scores / total

# ----------------------------------------------------------------
# EXPECTED-POINTS MATH HELPERS
# ----------------------------------------------------------------
def _vx_phase10_expected_floor_poisson(mu, divisor, tail=30):
    mu = float(max(0.0, mu))
    probs = [math.exp(-mu)]
    for k in range(1, int(tail) + 1):
        probs.append(probs[-1] * mu / k)
    # Remaining tail is negligible for football-count means at chosen support.
    return float(sum((k // int(divisor)) * probs[k] for k in range(len(probs))))


def _vx_phase10_prob_ge_poisson(mu, threshold, tail=40):
    mu = float(max(0.0, mu))
    p = math.exp(-mu)
    cdf = p if threshold > 0 else 0.0
    for k in range(1, int(threshold)):
        p = p * mu / k
        cdf += p
    return float(np.clip(1.0 - cdf, 0.0, 1.0))

# Basic mathematical self-checks.
if not math.isclose(_vx_phase10_expected_floor_poisson(0.0, 2), 0.0, abs_tol=1e-12):
    raise RuntimeError("Phase 10 Poisson floor helper self-test failed")
if not (0.0 <= _vx_phase10_prob_ge_poisson(2.0, 3) <= 1.0):
    raise RuntimeError("Phase 10 Poisson tail helper self-test failed")

# ----------------------------------------------------------------
# FIXTURE-LEVEL PLAYER ALLOCATION
# ----------------------------------------------------------------
_prob_idx = fixture_probabilities_model_only.set_index("fixture_id", drop=False)
_feature_by_team = {int(t): g.copy() for t, g in phase10_player_features.groupby("team_id", sort=False)}

def _vx_phase10_minutes_distribution(player_code, appearance_probability, sixty_probability, expected_minutes):
    """Three-state minutes PMF matching current appearance/60+/mean constraints.

    Nonlinear FPL scoring is integrated over this PMF instead of evaluating at
    mean minutes. Historical minutes set the typical short/long minute locations;
    current availability/overrides set the probability masses.
    """
    app = float(np.clip(appearance_probability, 0.0, 1.0))
    p60 = float(np.clip(sixty_probability, 0.0, app))
    target = float(np.clip(expected_minutes, 0.0, 90.0))
    p_short = max(0.0, app - p60)
    p_long = p60

    rg = _role.loc[pd.to_numeric(_role.get("player_code", pd.Series(dtype=float)), errors="coerce").eq(int(player_code))].copy() if not _role.empty else pd.DataFrame()
    short_m, long_m = 30.0, 75.0
    if not rg.empty:
        mins_hist = pd.to_numeric(rg.get("minutes_played", 0), errors="coerce").fillna(0.0)
        ww = pd.to_numeric(rg.get("recency_weight", 1.0), errors="coerce").fillna(0.0)
        sh = (mins_hist > 0) & (mins_hist < 60)
        lg = mins_hist >= 60
        if sh.any():
            val = _vx_phase10_weighted_mean(mins_hist.loc[sh], ww.loc[sh])
            if np.isfinite(val): short_m = float(np.clip(val, 1.0, 59.0))
        if lg.any():
            val = _vx_phase10_weighted_mean(mins_hist.loc[lg], ww.loc[lg])
            if np.isfinite(val): long_m = float(np.clip(val, 60.0, 90.0))

    # Reconcile the support values to the current expected-minutes constraint.
    if p_long > 1e-12:
        long_m = float((target - p_short * short_m) / p_long)
        if long_m < 60.0 or long_m > 90.0:
            long_m = float(np.clip(long_m, 60.0, 90.0))
            if p_short > 1e-12:
                short_m = float(np.clip((target - p_long * long_m) / p_short, 1.0, 59.0))
    elif p_short > 1e-12:
        short_m = float(np.clip(target / p_short, 1.0, 59.0))

    pmf = []
    p0 = max(0.0, 1.0 - app)
    if p0 > 1e-12: pmf.append((0.0, p0))
    if p_short > 1e-12: pmf.append((short_m, p_short))
    if p_long > 1e-12: pmf.append((long_m, p_long))
    total = sum(p for _, p in pmf)
    if total <= 0:
        return [(0.0, 1.0)]
    return [(float(m), float(p / total)) for m, p in pmf]


PHASE10_FIXTURE_FIELDS = (
    "fixture_id", "gameweek", "player_id", "player_code", "player_name", "team_id", "team_name", "opponent_team_id",
    "position", "price", "venue", "start_probability", "appearance_probability", "expected_minutes", "sixty_minute_probability",
    "npxg_per90", "xa_per90", "goal_share", "assist_share", "penalty_taker_probability", "set_piece_role",
    "team_nonpenalty_projected_goals", "team_penalty_projected_goals", "expected_nonpenalty_goals", "expected_penalty_goals",
    "fpl_assist_per_goal_rate", "team_projected_goals", "opponent_projected_goals", "team_clean_sheet_probability", "player_clean_sheet_probability",
    "goalkeeper_three_plus_saves_probability", "defcon_probability", "bonus_probability", "card_probability",
    "appearance_points", "goal_points", "assist_points", "clean_sheet_points", "goals_conceded_deduction",
    "save_points", "penalty_save_points", "defcon_points", "bonus_points", "negative_event_deductions",
    "fixture_expected_points", "calculation_status", "team_model_confidence", "player_evidence_confidence", "player_model_confidence",
    "output_mode", "model_version", "forecast_timestamp", "source_name", "is_pre_deadline_data",
)

_phase10_fixture_rows = []
for _, fx in projected_goals_model_only.iterrows():
    fid = int(fx["fixture_id"])
    gw = int(fx["gameweek"])
    if fid not in _prob_idx.index:
        raise RuntimeError(f"Phase 10 missing Phase-4 probability row for fixture {fid}")
    pr = _prob_idx.loc[fid]
    if isinstance(pr, pd.DataFrame):
        raise RuntimeError(f"Phase 10 duplicate probability rows for fixture {fid}")

    for venue, tid, oid, lam_for, lam_nonpen, lam_pen, lam_against, cs_prob in (
        ("H", int(fx["home_team_id"]), int(fx["away_team_id"]), float(fx["model_lambda_home"]), float(fx["model_lambda_home_nonpenalty"]), float(fx["model_lambda_home_penalty"]), float(fx["model_lambda_away"]), float(pr["home_clean_sheet_probability"])),
        ("A", int(fx["away_team_id"]), int(fx["home_team_id"]), float(fx["model_lambda_away"]), float(fx["model_lambda_away_nonpenalty"]), float(fx["model_lambda_away_penalty"]), float(fx["model_lambda_home"]), float(pr["away_clean_sheet_probability"])),
    ):
        team_players = _feature_by_team.get(tid, pd.DataFrame()).copy()
        if team_players.empty:
            continue

        # Non-penalty attacking rates allocate only the non-penalty team lambda.
        # The separate team penalty lambda is allocated by live penalty-taker role.
        team_players["_goal_weight"] = (
            pd.to_numeric(team_players["npxg_per90"], errors="coerce").fillna(0.0)
            * pd.to_numeric(team_players["expected_minutes"], errors="coerce").fillna(0.0) / 90.0
        )
        team_players["_assist_weight"] = (
            pd.to_numeric(team_players["xa_per90"], errors="coerce").fillna(0.0)
            * pd.to_numeric(team_players["expected_minutes"], errors="coerce").fillna(0.0) / 90.0
        )
        goal_total = float(team_players["_goal_weight"].sum())
        assist_total = float(team_players["_assist_weight"].sum())
        penalty_role_total = float(pd.to_numeric(team_players["penalty_taker_probability"], errors="coerce").fillna(0.0).sum())
        allocation_ready = bool(goal_total > 0 and assist_total > 0 and np.isfinite(PHASE10_FPL_ASSIST_PER_GOAL_RATE))

        for _, p in team_players.iterrows():
            pos = str(p["position"])
            mins = float(np.clip(p["expected_minutes"], 0.0, 90.0))
            p60 = float(np.clip(p["sixty_minute_probability"], 0.0, 1.0))
            app = float(np.clip(p["appearance_probability"], 0.0, 1.0))
            np_goal_share = float(p["_goal_weight"] / goal_total) if goal_total > 0 else 0.0
            assist_share = float(p["_assist_weight"] / assist_total) if assist_total > 0 else 0.0
            mu_np_goal = float(lam_nonpen * np_goal_share) if allocation_ready else np.nan
            pen_share = float(p["penalty_taker_probability"]) if penalty_role_total > 0 else 0.0
            mu_pen_goal = float(lam_pen * pen_share) if allocation_ready else np.nan
            mu_goal = float(mu_np_goal + mu_pen_goal) if allocation_ready else np.nan
            goal_share = float(mu_goal / lam_for) if allocation_ready and lam_for > 0 else 0.0
            # Official FPL assists cannot exceed one per scored goal. The empirical
            # league assist-per-goal rate determines the total; xA allocates it.
            mu_assist = float(lam_nonpen * PHASE10_FPL_ASSIST_PER_GOAL_RATE * assist_share) if allocation_ready else np.nan

            appearance_points = float(app + p60)
            goal_points = float(mu_goal * PHASE10_GOAL_POINTS.get(pos, 0.0)) if np.isfinite(mu_goal) else np.nan
            assist_points = float(mu_assist * PHASE10_ASSIST_POINTS) if np.isfinite(mu_assist) else np.nan

            minutes_pmf = _vx_phase10_minutes_distribution(int(p["player_code"]), app, p60, mins)
            no_concede_uncond = 0.0
            clean_sheet_points = 0.0
            gc_deduction = 0.0
            save_points = 0.0
            save_prob = 0.0
            for m_played, p_m in minutes_pmf:
                if m_played <= 0 or p_m <= 0:
                    continue
                no_concede = math.exp(-lam_against * min(1.0, m_played / 90.0))
                no_concede_uncond += p_m * no_concede
                if m_played >= 60.0:
                    clean_sheet_points += p_m * no_concede * PHASE10_CS_POINTS.get(pos, 0.0)
                if pos in {"GK", "DEF"}:
                    gc_mu = float(lam_against * m_played / 90.0)
                    gc_deduction -= p_m * _vx_phase10_expected_floor_poisson(gc_mu, 2)
                if pos == "GK":
                    save_mu = float(max(0.0, p["saves_per90"]) * m_played / 90.0)
                    save_prob += p_m * _vx_phase10_prob_ge_poisson(save_mu, 3)
                    save_points += p_m * _vx_phase10_expected_floor_poisson(save_mu, 3)
            player_cs_prob = float(no_concede_uncond / app) if app > 1e-12 else 0.0

            penalty_save_points = 0.0
            if pos == "GK":
                penalty_save_points = float(app * max(0.0, p["penalty_saves_per_appearance"]) * PHASE10_PENALTY_SAVE_POINTS)

            defcon_points = float(PHASE10_DEFCON_POINTS * np.clip(p["defcon_probability"], 0.0, 1.0))
            bonus_points = float(app * max(0.0, p["bonus_points_per_appearance"]))

            yellow = float(app * max(0.0, p["yellow_cards_per_appearance"]))
            red = float(app * max(0.0, p["red_cards_per_appearance"]))
            own = float(app * max(0.0, p["own_goals_per_appearance"]))
            pmiss = float(app * max(0.0, p["penalty_misses_per_appearance"]))
            negative = float(
                yellow * PHASE10_YELLOW_POINTS
                + red * PHASE10_RED_POINTS
                + own * PHASE10_OWN_GOAL_POINTS
                + pmiss * PHASE10_PENALTY_MISS_POINTS
            )
            card_probability = float(np.clip(app * min(1.0, max(0.0, p["yellow_cards_per_appearance"] + p["red_cards_per_appearance"])), 0.0, 1.0))

            if allocation_ready:
                xpts = float(
                    appearance_points + goal_points + assist_points + clean_sheet_points
                    + gc_deduction + save_points + penalty_save_points + defcon_points
                    + bonus_points + negative
                )
                calculation_status = "READY"
            else:
                xpts = np.nan
                calculation_status = "BLOCKED_INSUFFICIENT_PLAYER_ROLE_ATTACK_EVIDENCE"
            team_conf = float(fx.get("model_confidence", np.nan))
            player_conf = float(np.clip(
                math.sqrt(max(0.0, team_conf) * max(0.0, float(p["player_evidence_confidence"])))
                if np.isfinite(team_conf) else float(p["player_evidence_confidence"]),
                0.0, 1.0,
            ))

            _phase10_fixture_rows.append({
                "fixture_id": fid, "gameweek": gw,
                "player_id": int(p["player_id"]), "player_code": int(p["player_code"]), "player_name": str(p["player_name"]),
                "team_id": tid, "team_name": str(p["team_name"]), "opponent_team_id": oid,
                "position": pos, "price": float(p["price"]), "venue": venue,
                "start_probability": float(p["start_probability"]), "appearance_probability": app,
                "expected_minutes": mins, "sixty_minute_probability": p60,
                "npxg_per90": float(p["npxg_per90"]), "xa_per90": float(p["xa_per90"]),
                "goal_share": goal_share, "assist_share": assist_share,
                "penalty_taker_probability": float(p["penalty_taker_probability"]), "set_piece_role": str(p["set_piece_role"]),
                "team_nonpenalty_projected_goals": lam_nonpen, "team_penalty_projected_goals": lam_pen,
                "expected_nonpenalty_goals": mu_np_goal, "expected_penalty_goals": mu_pen_goal,
                "fpl_assist_per_goal_rate": float(PHASE10_FPL_ASSIST_PER_GOAL_RATE),
                "team_projected_goals": lam_for, "opponent_projected_goals": lam_against,
                "team_clean_sheet_probability": cs_prob, "player_clean_sheet_probability": player_cs_prob,
                "goalkeeper_three_plus_saves_probability": save_prob,
                "defcon_probability": float(p["defcon_probability"]), "bonus_probability": float(p["bonus_probability"]),
                "card_probability": card_probability,
                "appearance_points": appearance_points, "goal_points": goal_points, "assist_points": assist_points,
                "clean_sheet_points": clean_sheet_points, "goals_conceded_deduction": gc_deduction,
                "save_points": save_points, "penalty_save_points": penalty_save_points,
                "defcon_points": defcon_points, "bonus_points": bonus_points,
                "negative_event_deductions": negative, "fixture_expected_points": xpts, "calculation_status": calculation_status,
                "team_model_confidence": team_conf, "player_evidence_confidence": float(p["player_evidence_confidence"]),
                "player_model_confidence": player_conf,
                "output_mode": PHASE10_OUTPUT_MODE, "model_version": PHASE10_VERSION,
                "forecast_timestamp": str(fx.get("forecast_timestamp", PHASE10_TIMESTAMP)),
                "source_name": "VORTEX_PHASE10_PLAYER_MODEL:FPL_CORE+FPL_API+PHASE3_4",
                "is_pre_deadline_data": fx.get("is_pre_deadline_data", pd.NA),
            })

phase10_player_fixture_expected_points = pd.DataFrame(_phase10_fixture_rows, columns=PHASE10_FIXTURE_FIELDS)
if phase10_player_fixture_expected_points.empty:
    raise RuntimeError("Phase 10 produced no player-fixture expected-points rows.")


# ----------------------------------------------------------------
# PLAYER RETURN DISTRIBUTION — deterministic Monte Carlo diagnostics
# ----------------------------------------------------------------
# This does NOT replace fixture_expected_points. It adds an explicit return
# distribution for captaincy/risk analysis while preserving the analytic mean
# as the canonical xPts forecast.
PHASE10_RETURN_SIMULATIONS = 1600
PHASE10_RETURN_SEED = 2627
PHASE10_RETURN_MODEL = "APPROX_JOINT_EVENT_MONTE_CARLO_V1"
_phase10_feature_idx = phase10_player_features.set_index("player_id", drop=False)
_phase10_return_samples = {}

def _vx_phase10_return_stats(samples):
    a = np.asarray(samples, dtype=float)
    if a.size == 0 or not np.isfinite(a).any():
        return {
            "return_mc_mean": np.nan, "return_mc_std": np.nan,
            "return_p10": np.nan, "return_p25": np.nan, "return_p50": np.nan,
            "return_p75": np.nan, "return_p90": np.nan,
            "prob_10_plus": np.nan, "prob_15_plus": np.nan,
        }
    return {
        "return_mc_mean": float(np.mean(a)),
        "return_mc_std": float(np.std(a, ddof=0)),
        "return_p10": float(np.quantile(a, 0.10)),
        "return_p25": float(np.quantile(a, 0.25)),
        "return_p50": float(np.quantile(a, 0.50)),
        "return_p75": float(np.quantile(a, 0.75)),
        "return_p90": float(np.quantile(a, 0.90)),
        "prob_10_plus": float(np.mean(a >= 10.0)),
        "prob_15_plus": float(np.mean(a >= 15.0)),
    }

def _vx_phase10_simulate_fixture_return(row):
    pid = int(row["player_id"]); pcode = int(row["player_code"]); fid = int(row["fixture_id"])
    feat = _phase10_feature_idx.loc[pid] if pid in _phase10_feature_idx.index else None
    if isinstance(feat, pd.DataFrame):
        feat = feat.iloc[0]
    rng = np.random.default_rng(PHASE10_RETURN_SEED + fid * 100003 + pid * 97)
    n = int(PHASE10_RETURN_SIMULATIONS)
    app = float(np.clip(row["appearance_probability"], 0.0, 1.0))
    p60 = float(np.clip(row["sixty_minute_probability"], 0.0, app))
    em = float(np.clip(row["expected_minutes"], 0.0, 90.0))
    pmf = _vx_phase10_minutes_distribution(pcode, app, p60, em)
    mins_vals = np.asarray([m for m, _ in pmf], dtype=float)
    probs = np.asarray([p for _, p in pmf], dtype=float)
    probs = probs / probs.sum()
    mins = rng.choice(mins_vals, size=n, p=probs)
    played = mins > 0
    sixty = mins >= 60.0
    pts = played.astype(float) + sixty.astype(float)

    expected_frac = float(sum((m / 90.0) * p for m, p in pmf))
    mu_goal = float(row.get("expected_nonpenalty_goals", 0.0) or 0.0) + float(row.get("expected_penalty_goals", 0.0) or 0.0)
    mu_assist = float(row.get("assist_points", 0.0) or 0.0) / max(PHASE10_ASSIST_POINTS, 1e-9)
    if expected_frac > 1e-12:
        goal_rate90 = max(0.0, mu_goal / expected_frac)
        assist_rate90 = max(0.0, mu_assist / expected_frac)
        goals = rng.poisson(goal_rate90 * mins / 90.0)
        assists = rng.poisson(assist_rate90 * mins / 90.0)
    else:
        goals = np.zeros(n, dtype=int); assists = np.zeros(n, dtype=int)
    pos = str(row["position"])
    pts += goals * PHASE10_GOAL_POINTS.get(pos, 0.0)
    pts += assists * PHASE10_ASSIST_POINTS

    lam_against = float(row.get("opponent_projected_goals", 0.0) or 0.0)
    conceded = rng.poisson(np.maximum(0.0, lam_against * mins / 90.0))
    if PHASE10_CS_POINTS.get(pos, 0.0) > 0:
        pts += (sixty & (conceded == 0)).astype(float) * PHASE10_CS_POINTS[pos]
    if pos in {"GK", "DEF"}:
        pts -= np.floor(conceded / 2.0)

    if feat is not None and pos == "GK":
        saves90 = float(max(0.0, feat.get("saves_per90", 0.0) or 0.0))
        saves = rng.poisson(saves90 * mins / 90.0)
        pts += np.floor(saves / 3.0)
        psave_rate = float(max(0.0, feat.get("penalty_saves_per_appearance", 0.0) or 0.0))
        pts += rng.poisson(psave_rate, size=n) * played.astype(int) * PHASE10_PENALTY_SAVE_POINTS

    if feat is not None:
        # DEFCON and bonus are sampled conditional on appearing. Bonus evidence is
        # current-regime only; before enough 2026/27 evidence exists its mean is zero,
        # avoiding contamination from the previous BPS rules.
        defcon_uncond = float(np.clip(feat.get("defcon_probability", 0.0) or 0.0, 0.0, 1.0))
        defcon_cond = float(np.clip(defcon_uncond / app, 0.0, 1.0)) if app > 1e-12 else 0.0
        pts += (rng.random(n) < defcon_cond).astype(float) * played.astype(float) * PHASE10_DEFCON_POINTS

        bonus_uncond = float(np.clip(feat.get("bonus_probability", 0.0) or 0.0, 0.0, 1.0))
        bonus_cond = float(np.clip(bonus_uncond / app, 0.0, 1.0)) if app > 1e-12 else 0.0
        bonus_mean_app = float(np.clip(feat.get("bonus_points_per_appearance", 0.0) or 0.0, 0.0, 3.0))
        bonus_hit = (rng.random(n) < bonus_cond) & played
        if bonus_cond > 1e-12:
            award_mean = float(np.clip(bonus_mean_app / bonus_cond, 1.0, 3.0))
            lo = int(math.floor(award_mean)); hi = int(math.ceil(award_mean))
            if hi == lo:
                award = np.full(n, lo, dtype=float)
            else:
                award = lo + (rng.random(n) < (award_mean - lo)).astype(float)
            pts += bonus_hit.astype(float) * award

        # Rare negatives are sampled as count processes conditional on playing.
        for col, score in (
            ("yellow_cards_per_appearance", PHASE10_YELLOW_POINTS),
            ("red_cards_per_appearance", PHASE10_RED_POINTS),
            ("own_goals_per_appearance", PHASE10_OWN_GOAL_POINTS),
            ("penalty_misses_per_appearance", PHASE10_PENALTY_MISS_POINTS),
        ):
            rate = float(max(0.0, feat.get(col, 0.0) or 0.0))
            pts += rng.poisson(rate, size=n) * played.astype(int) * float(score)

    return pts.astype(float)

_phase10_dist_rows = []
for _i, _r in phase10_player_fixture_expected_points.iterrows():
    if str(_r.get("calculation_status", "")) != "READY":
        _phase10_dist_rows.append(_vx_phase10_return_stats([]))
        continue
    _s = _vx_phase10_simulate_fixture_return(_r)
    _phase10_return_samples[(int(_r["player_id"]), int(_r["fixture_id"]))] = _s
    _phase10_dist_rows.append(_vx_phase10_return_stats(_s))

_phase10_dist_df = pd.DataFrame(_phase10_dist_rows, index=phase10_player_fixture_expected_points.index)
for _c in _phase10_dist_df.columns:
    phase10_player_fixture_expected_points[_c] = _phase10_dist_df[_c]
phase10_player_fixture_expected_points["return_distribution_model"] = PHASE10_RETURN_MODEL

# Allocation sanity: non-penalty expected goals must fully allocate the non-penalty
# team component. A missing penalty-taker role may leave only the penalty component
# unallocated rather than fabricating an equal-share taker.
_ready_share = phase10_player_fixture_expected_points.loc[phase10_player_fixture_expected_points["calculation_status"].eq("READY")].copy()
if not _ready_share.empty:
    _np_check = _ready_share.groupby(["fixture_id", "team_id"], as_index=False).agg(
        allocated_np=("expected_nonpenalty_goals", "sum"),
        target_np=("team_nonpenalty_projected_goals", "first"),
        combined_share=("goal_share", "sum"),
    )
    if not np.allclose(_np_check["allocated_np"].to_numpy(float), _np_check["target_np"].to_numpy(float), atol=1e-8):
        raise RuntimeError("Phase 10 non-penalty goal allocation does not reconcile to the team component.")
    if (_np_check["combined_share"] > 1.00000001).any():
        raise RuntimeError("Phase 10 combined player goal shares exceed the team total.")

# ----------------------------------------------------------------
# GAMEWEEK + NEXT-3 / NEXT-5 AGGREGATION
# No ranking or recommendation is created here.
# ----------------------------------------------------------------
phase10_player_gameweek_expected_points = (
    phase10_player_fixture_expected_points.groupby(
        ["player_id", "player_code", "player_name", "team_id", "team_name", "position", "price", "gameweek"],
        as_index=False,
    ).agg(
        gameweek_expected_points=("fixture_expected_points", lambda s: s.sum(min_count=1)),
        expected_minutes=("expected_minutes", "sum"),
        fixture_count=("fixture_id", "nunique"),
        player_model_confidence=("player_model_confidence", "mean"),
    )
)


# Combine fixture-level Monte Carlo samples inside each Gameweek. This handles
# double Gameweeks by summing independently simulated fixture returns.
_gw_dist_rows = []
for _keys, _g in phase10_player_fixture_expected_points.groupby(
    ["player_id", "gameweek"], sort=False
):
    _pid, _gw_num = int(_keys[0]), int(_keys[1])
    _parts = [
        _phase10_return_samples.get((_pid, int(fid)))
        for fid in pd.to_numeric(_g["fixture_id"], errors="coerce").dropna().astype(int)
    ]
    _parts = [x for x in _parts if isinstance(x, np.ndarray) and len(x) == PHASE10_RETURN_SIMULATIONS]
    if _parts:
        _total = np.sum(np.vstack(_parts), axis=0)
        _stats = _vx_phase10_return_stats(_total)
    else:
        _stats = _vx_phase10_return_stats([])
    _stats.update({"player_id": _pid, "gameweek": _gw_num})
    _gw_dist_rows.append(_stats)
_phase10_gw_dist = pd.DataFrame(_gw_dist_rows)
if not _phase10_gw_dist.empty:
    phase10_player_gameweek_expected_points = phase10_player_gameweek_expected_points.merge(
        _phase10_gw_dist, on=["player_id", "gameweek"], how="left", validate="one_to_one"
    )
phase10_player_gameweek_expected_points["return_distribution_model"] = PHASE10_RETURN_MODEL

_current_gw = int(globals().get("NEXT_GW", pd.to_numeric(phase10_player_gameweek_expected_points["gameweek"], errors="coerce").min()))
_future_gws = sorted(g for g in pd.to_numeric(phase10_player_gameweek_expected_points["gameweek"], errors="coerce").dropna().astype(int).unique() if g >= _current_gw)
_next3_gws = _future_gws[:3]
_next5_gws = _future_gws[:5]

_horizon_rows = []
for pid, g in phase10_player_gameweek_expected_points.groupby("player_id", sort=False):
    meta = g.iloc[0]
    gw_map = g.set_index("gameweek")["gameweek_expected_points"].to_dict()
    final_raw = gw_map.get(_current_gw, np.nan)
    final_gw = float(final_raw) if np.isfinite(final_raw) else np.nan
    vals3 = [gw_map.get(x, np.nan) for x in _next3_gws]
    vals5 = [gw_map.get(x, np.nan) for x in _next5_gws]
    next3 = float(np.sum(vals3)) if vals3 and all(np.isfinite(v) for v in vals3) else np.nan
    next5 = float(np.sum(vals5)) if vals5 and all(np.isfinite(v) for v in vals5) else np.nan
    price = float(meta["price"]) if np.isfinite(meta["price"]) and float(meta["price"]) > 0 else np.nan
    _horizon_rows.append({
        "player_id": int(pid), "player_code": int(meta["player_code"]), "player_name": str(meta["player_name"]),
        "team_id": int(meta["team_id"]), "team_name": str(meta["team_name"]), "position": str(meta["position"]),
        "price": price,
        "final_gameweek_xpts": final_gw,
        "next_3_xpts": next3,
        "next_5_xpts": next5,
        "points_per_million": float(final_gw / price) if np.isfinite(price) and price > 0 else np.nan,
        "next_5_points_per_million": float(next5 / price) if np.isfinite(price) and price > 0 else np.nan,
        "captain_expected_value": float(2.0 * final_gw) if np.isfinite(final_gw) else np.nan,
        "captain_incremental_expected_value": final_gw,
        "return_mc_mean": float(g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw), "return_mc_mean"].iloc[0]) if "return_mc_mean" in g.columns and not g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw)].empty else np.nan,
        "return_mc_std": float(g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw), "return_mc_std"].iloc[0]) if "return_mc_std" in g.columns and not g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw)].empty else np.nan,
        "return_p10": float(g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw), "return_p10"].iloc[0]) if "return_p10" in g.columns and not g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw)].empty else np.nan,
        "return_p25": float(g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw), "return_p25"].iloc[0]) if "return_p25" in g.columns and not g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw)].empty else np.nan,
        "return_p50": float(g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw), "return_p50"].iloc[0]) if "return_p50" in g.columns and not g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw)].empty else np.nan,
        "return_p75": float(g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw), "return_p75"].iloc[0]) if "return_p75" in g.columns and not g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw)].empty else np.nan,
        "return_p90": float(g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw), "return_p90"].iloc[0]) if "return_p90" in g.columns and not g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw)].empty else np.nan,
        "prob_10_plus": float(g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw), "prob_10_plus"].iloc[0]) if "prob_10_plus" in g.columns and not g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw)].empty else np.nan,
        "prob_15_plus": float(g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw), "prob_15_plus"].iloc[0]) if "prob_15_plus" in g.columns and not g.loc[pd.to_numeric(g["gameweek"], errors="coerce").eq(_current_gw)].empty else np.nan,
        "return_distribution_model": PHASE10_RETURN_MODEL,
    "current_bps_position_priors": _phase10_current_bonus_prior,
        "next_3_gameweeks": ",".join(map(str, _next3_gws)),
        "next_5_gameweeks": ",".join(map(str, _next5_gws)),
        "player_model_confidence": float(g["player_model_confidence"].mean()),
        "output_mode": PHASE10_OUTPUT_MODE,
        "model_version": PHASE10_VERSION,
        "forecast_timestamp": PHASE10_TIMESTAMP,
    })
phase10_player_horizon_expected_points = pd.DataFrame(_horizon_rows)

# Canonical Phase-10 handoff. Phase 11 must inspect the release gate before use.
player_expected_points_fixture = phase10_player_fixture_expected_points.copy()
player_expected_points_gameweek = phase10_player_gameweek_expected_points.copy()
player_expected_points_horizon = phase10_player_horizon_expected_points.copy()

# ----------------------------------------------------------------
# PERSISTENCE
# ----------------------------------------------------------------
PHASE10_OUTPUT_DIR = Path(DATA_DIR) / "PHASE10_PLAYER_EXPECTED_POINTS"
PHASE10_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
PHASE10_FEATURE_PATH = PHASE10_OUTPUT_DIR / "phase10_player_features.json"
PHASE10_FIXTURE_PATH = PHASE10_OUTPUT_DIR / "phase10_player_fixture_expected_points.json"
PHASE10_GAMEWEEK_PATH = PHASE10_OUTPUT_DIR / "phase10_player_gameweek_expected_points.json"
PHASE10_HORIZON_PATH = PHASE10_OUTPUT_DIR / "phase10_player_horizon_expected_points.json"
PHASE10_SUMMARY_PATH = PHASE10_OUTPUT_DIR / "phase10_summary.json"

phase10_player_features.to_json(PHASE10_FEATURE_PATH, orient="records", date_format="iso", indent=2)
phase10_player_fixture_expected_points.to_json(PHASE10_FIXTURE_PATH, orient="records", date_format="iso", indent=2)
phase10_player_gameweek_expected_points.to_json(PHASE10_GAMEWEEK_PATH, orient="records", date_format="iso", indent=2)
phase10_player_horizon_expected_points.to_json(PHASE10_HORIZON_PATH, orient="records", date_format="iso", indent=2)

PHASE10_SUMMARY = {
    "phase10_version": PHASE10_VERSION,
    "calculated_at": PHASE10_TIMESTAMP,
    "output_mode": PHASE10_OUTPUT_MODE,
    "production_eligible": PHASE10_PRODUCTION_ELIGIBLE,
    "manual_team_model_approval": PHASE10_MANUAL_TEAM_MODEL_APPROVAL,
    "team_validation_evidence_ready": PHASE10_TEAM_VALIDATION_EVIDENCE,
    "phase9_external_blend_evidence_ready": PHASE10_BLEND_REVIEW_COMPLETE,
    "live_market_blend_weight": float(globals().get("MARKET_BLEND_WEIGHT", 0.0)),
    "live_elo_blend_weight": float(globals().get("ELO_BLEND_WEIGHT", 0.0)),
    "player_features": int(len(phase10_player_features)),
    "fixture_rows": int(len(phase10_player_fixture_expected_points)),
    "gameweek_rows": int(len(phase10_player_gameweek_expected_points)),
    "horizon_rows": int(len(phase10_player_horizon_expected_points)),
    "source_seasons": list(PHASE10_SOURCE_SEASONS),
    "source_files_read": int(len(PHASE10_SOURCE_FILES)),
    "role_evidence_rows": int(len(_role)),
    "return_simulations_per_fixture": PHASE10_RETURN_SIMULATIONS,
    "return_distribution_model": PHASE10_RETURN_MODEL,
    "premier_league_attack_rows": int(len(_pm_pl)),
    "next_3_gameweeks": _next3_gws,
    "next_5_gameweeks": _next5_gws,
    "decision_engine_unlocked": False,
    "notes": [
        "Team forecast -> player opportunity -> expected points only.",
        "Player ownership/popularity is not an expected-points input.",
        "Non-penalty goal shares use expected-minutes-weighted NPxG; the separate penalty component uses live penalty-taker probability.",
        "NPxG and penalty xG remain separate from team lambda through player allocation; FPL assists use an empirical league assist-per-goal rate capped at one.",
        "All-competition lineups/minutes inform role; nonlinear CS/GC/save scoring is integrated over a minutes probability distribution.",
        "Thin player rates shrink toward empirically derived position priors.",
        "Market/Team Elo weights remain zero and no external blend is promoted here.",
        "A deterministic Monte Carlo return distribution supplies p10/p25/p50/p75/p90 and 10+/15+ haul probabilities without replacing analytic xPts.",
        "No transfer, captain ranking, XI, bench or squad recommendation is created in Phase 10.",
    ],
}
PHASE10_SUMMARY_PATH.write_text(json.dumps(PHASE10_SUMMARY, indent=2, default=str), encoding="utf-8")

# ----------------------------------------------------------------
# FINAL INTEGRITY + RELEASE GATE
# ----------------------------------------------------------------
_PHASE10_STATE_AFTER = {k: vx_frame_digest(globals()[k]) for k in _PHASE10_IMMUTABLE}
if _PHASE10_STATE_BEFORE != _PHASE10_STATE_AFTER:
    raise RuntimeError("Phase 10 integrity failure: player model mutated an upstream team-model table.")
if float(globals().get("MARKET_BLEND_WEIGHT", 0.0)) != 0.0 or float(globals().get("ELO_BLEND_WEIGHT", 0.0)) != 0.0:
    raise RuntimeError("Phase 10 integrity failure: benchmark weights changed.")

PHASE10_READY = True
PHASE10_READY_FOR_DECISION_ENGINE = bool(PHASE10_PRODUCTION_ELIGIBLE)

if "vx_phase0_assert_cleanroom" in globals():
    vx_phase0_assert_cleanroom("after Phase 10 player expected-points engine")

print("✅ PHASE 10 — PLAYER EXPECTED-POINTS ENGINE BUILT")
print(f"   Output mode: {PHASE10_OUTPUT_MODE}")
print(f"   Player features: {len(phase10_player_features)}")
print(f"   Player-fixture rows: {len(phase10_player_fixture_expected_points)}")
print(f"   Player-GW rows: {len(phase10_player_gameweek_expected_points)}")
print(f"   Next-3 GWs: {_next3_gws} | Next-5 GWs: {_next5_gws}")
print("🔒 Team projected goals / CS / FDR remain unchanged.")
print("🔒 Market and Team Elo live blend weights remain 0.00.")
print("🔒 No transfer/captain ranking, XI, bench or squad optimisation is built here.")
if not PHASE10_READY_FOR_DECISION_ENGINE:
    print("⛔ Phase 11 release gate remains CLOSED until genuine team-model validation is reviewed and manually approved.")
print("📊 Phase 10 summary:", PHASE10_SUMMARY_PATH)
```

```python
# ============================================================
# CELL 13E — PHASE 11
# FPL DECISION ENGINE
# PLAYER xPTS -> XI / BENCH / TRANSFERS / CAPTAINCY / CHIPS / SQUADS
# RESEARCH-ONLY UNTIL PHASE-10 RELEASE GATE + CURRENT SQUAD STATE ARE READY
# ============================================================
# Policy:
#   - consumes Phase-10 player expected points only;
#   - never writes back into team lambdas, CS probabilities or FDR;
#   - market and Team Elo remain benchmark-only with live weights at 0.00;
#   - ownership / elite EO are decision overlays only, never xPts inputs;
#   - official provider fixture labels are never used;
#   - no post-deadline hindsight is allowed to create a pre-deadline decision.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from itertools import combinations
import json
import math
import re
import numpy as np
import pandas as pd

PHASE11_VERSION = "VORTEX-V3.1-FPL-DECISION-ENGINE-DISTRIBUTION-AWARE"
PHASE11_TIMESTAMP = datetime.now(timezone.utc).isoformat()
PHASE11_CURRENT_GW = int(globals().get("NEXT_GW"))
PHASE11_LINEUP_WEIGHTS = (1.00, 0.90, 0.80, 0.70, 0.60)
PHASE11_SELL_MIN_NEXT5_GAIN = 2.0
PHASE11_SELL_MIN_UTILITY = 0.50
PHASE11_BUY_MIN_UTILITY = 0.50
PHASE11_DIFFERENTIAL_OWNERSHIP_MAX = 10.0
PHASE11_PROTECTION_EO_MIN = 60.0
PHASE11_MILP_TIME_LIMIT_SECONDS = 45.0
PHASE11_BENCH_MILP_WEIGHT = 0.03

# ------------------------------
# HARD UPSTREAM POLICY GATES
# ------------------------------
if float(globals().get("MARKET_BLEND_WEIGHT", 0.0)) != 0.0:
    raise RuntimeError("Phase 11 policy violation: MARKET_BLEND_WEIGHT must remain 0.00.")
if float(globals().get("ELO_BLEND_WEIGHT", 0.0)) != 0.0:
    raise RuntimeError("Phase 11 policy violation: ELO_BLEND_WEIGHT must remain 0.00.")
if bool(globals().get("PHASE9_LIVE_BLEND_ACTIVE", False)):
    raise RuntimeError("Phase 11 policy violation: Phase-9 live blend must remain disabled.")

for _name in (
    "player_expected_points_fixture",
    "player_expected_points_gameweek",
    "player_expected_points_horizon",
):
    _frame = globals().get(_name)
    if not isinstance(_frame, pd.DataFrame) or _frame.empty:
        raise RuntimeError(f"Phase 11 requires non-empty {_name} from Phase 10.")

for _name in (
    "projected_goals_model_only",
    "fixture_probabilities_model_only",
    "fixture_fdr_model_only",
):
    _frame = globals().get(_name)
    if not isinstance(_frame, pd.DataFrame) or _frame.empty:
        raise RuntimeError(f"Phase 11 requires non-empty {_name} from the team model.")

PHASE11_UPSTREAM_RELEASED = bool(globals().get("PHASE10_READY_FOR_DECISION_ENGINE", False))
PHASE11_PHASE10_OUTPUT_MODE = str(globals().get("PHASE10_OUTPUT_MODE", "UNKNOWN"))
PHASE11_READY = False
PHASE11_ACTIONABLE = False
PHASE11_READY_FOR_PRESENTATION = False

# ------------------------------
# IMMUTABILITY SNAPSHOT
# ------------------------------
_PHASE11_IMMUTABLE = (
    "projected_goals_model_only",
    "fixture_probabilities_model_only",
    "fixture_fdr_model_only",
    "player_expected_points_fixture",
    "player_expected_points_gameweek",
    "player_expected_points_horizon",
)
_PHASE11_STATE_BEFORE = {k: vx_frame_digest(globals()[k]) for k in _PHASE11_IMMUTABLE}

# ------------------------------
# DEADLINE STATE
# ------------------------------
def _vx_phase11_utc(value):
    if value is None or (isinstance(value, float) and np.isnan(value)):
        return None
    try:
        ts = pd.Timestamp(value)
        if ts.tzinfo is None:
            ts = ts.tz_localize("UTC")
        else:
            ts = ts.tz_convert("UTC")
        return ts
    except Exception:
        return None

_phase11_now = _vx_phase11_utc(PHASE11_TIMESTAMP)
_phase11_deadline = None
_events = globals().get("events")
if isinstance(_events, pd.DataFrame) and not _events.empty and {"id", "deadline_time"}.issubset(_events.columns):
    _evt = _events.loc[pd.to_numeric(_events["id"], errors="coerce") == PHASE11_CURRENT_GW]
    if not _evt.empty:
        _phase11_deadline = _vx_phase11_utc(_evt.iloc[0].get("deadline_time"))
PHASE11_BEFORE_DEADLINE = bool(_phase11_deadline is not None and _phase11_now < _phase11_deadline)

# ------------------------------
# PLAYER DECISION POOL
# ------------------------------
_h = player_expected_points_horizon.copy()
_gw = player_expected_points_gameweek.copy()
_fx = player_expected_points_fixture.copy()

for _frame in (_h, _gw, _fx):
    _frame["player_id"] = pd.to_numeric(_frame["player_id"], errors="coerce").astype("Int64")

_h = _h.dropna(subset=["player_id"]).copy()
_h["player_id"] = _h["player_id"].astype(int)
_gw = _gw.dropna(subset=["player_id"]).copy()
_gw["player_id"] = _gw["player_id"].astype(int)
_fx = _fx.dropna(subset=["player_id"]).copy()
_fx["player_id"] = _fx["player_id"].astype(int)

_future_gws = sorted(
    int(x) for x in pd.to_numeric(_gw["gameweek"], errors="coerce").dropna().astype(int).unique()
    if int(x) >= PHASE11_CURRENT_GW
)[:5]
if not _future_gws:
    raise RuntimeError("Phase 11 found no future player-Gameweek expected-points rows.")
_gw_weights = {gw: PHASE11_LINEUP_WEIGHTS[i] for i, gw in enumerate(_future_gws)}

# Weighted, non-overlapping horizon score from individual GW forecasts.
_weight_rows = []
for pid, grp in _gw.groupby("player_id", sort=False):
    mp = {
        int(r["gameweek"]): float(r["gameweek_expected_points"])
        for _, r in grp.iterrows()
        if pd.notna(r.get("gameweek")) and np.isfinite(pd.to_numeric(r.get("gameweek_expected_points"), errors="coerce"))
    }
    vals = []
    complete = True
    for gw in _future_gws:
        val = mp.get(int(gw), np.nan)
        if not np.isfinite(val):
            complete = False
        vals.append(float(val) if np.isfinite(val) else 0.0)
    weighted = float(sum(_gw_weights[gw] * vals[i] for i, gw in enumerate(_future_gws)))
    _weight_rows.append({
        "player_id": int(pid),
        "phase11_weighted_horizon_xpts": weighted if complete else np.nan,
        "phase11_horizon_complete": bool(complete),
    })
_weight_df = pd.DataFrame(_weight_rows)
_h = _h.merge(_weight_df, on="player_id", how="left")

# Current official player metadata. Ownership is a decision overlay only.
_pmeta = globals().get("players")
if not isinstance(_pmeta, pd.DataFrame) or _pmeta.empty:
    raise RuntimeError("Phase 11 requires the current official FPL player table.")
_pmeta = _pmeta.copy()
_pmeta["player_id"] = pd.to_numeric(_pmeta["id"], errors="coerce").astype("Int64")
_keep_meta = [c for c in (
    "player_id", "selected_by_percent", "status", "chance_of_playing_next_round",
    "chance_of_playing_this_round", "web_name", "now_cost"
) if c in _pmeta.columns]
_pmeta = _pmeta[_keep_meta].dropna(subset=["player_id"]).copy()
_pmeta["player_id"] = _pmeta["player_id"].astype(int)
_h = _h.merge(_pmeta, on="player_id", how="left", suffixes=("", "_official"))
_h["overall_selected_by_percent"] = pd.to_numeric(_h.get("selected_by_percent"), errors="coerce")
_h["availability_status"] = _h.get("status", "a").fillna("a").astype(str)
_h["chance_next"] = pd.to_numeric(_h.get("chance_of_playing_next_round"), errors="coerce")
_h["decision_available"] = ~_h["availability_status"].str.lower().eq("u")
_h.loc[_h["chance_next"].eq(0), "decision_available"] = False
for _c in (
    "final_gameweek_xpts", "next_3_xpts", "next_5_xpts", "price",
    "player_model_confidence", "phase11_weighted_horizon_xpts",
    "return_mc_mean", "return_mc_std", "return_p10", "return_p25",
    "return_p50", "return_p75", "return_p90", "prob_10_plus", "prob_15_plus",
):
    if _c not in _h.columns:
        _h[_c] = np.nan
    _h[_c] = pd.to_numeric(_h[_c], errors="coerce")
_h["decision_pool_ready"] = (
    _h["decision_available"]
    & np.isfinite(_h["final_gameweek_xpts"])
    & np.isfinite(_h["next_5_xpts"])
    & np.isfinite(_h["price"])
)
phase11_player_decision_pool = _h.copy()

# Current-GW minutes/confidence view for captain safety diagnostics.
_cur_gw = _gw.loc[pd.to_numeric(_gw["gameweek"], errors="coerce") == PHASE11_CURRENT_GW].copy()
_cur_gw = _cur_gw[[c for c in (
    "player_id", "gameweek_expected_points", "expected_minutes", "fixture_count", "player_model_confidence"
) if c in _cur_gw.columns]].copy()
_cur_gw = _cur_gw.rename(columns={
    "gameweek_expected_points": "current_gw_xpts",
    "expected_minutes": "current_gw_expected_minutes",
    "player_model_confidence": "current_gw_player_confidence",
})
phase11_player_decision_pool = phase11_player_decision_pool.merge(_cur_gw, on="player_id", how="left")

# ------------------------------
# CURRENT SQUAD STATE CONTRACT
# ------------------------------
PHASE11_INPUT_ROOT = Path(PROJECT_ROOT) / "MODEL_INPUTS"
PHASE11_INPUT_ROOT.mkdir(parents=True, exist_ok=True)
PHASE11_MANUAL_SQUAD_PATH = PHASE11_INPUT_ROOT / "current_squad_state.json"
PHASE11_MANUAL_SQUAD_TEMPLATE = PHASE11_INPUT_ROOT / "current_squad_state_TEMPLATE.json"
PHASE11_ELITE_EO_MANUAL_CSV = PHASE11_INPUT_ROOT / "elite_eo_snapshot.csv"
PHASE11_ELITE_EO_MANUAL_JSON = PHASE11_INPUT_ROOT / "elite_eo_snapshot.json"

if not PHASE11_MANUAL_SQUAD_TEMPLATE.exists():
    PHASE11_MANUAL_SQUAD_TEMPLATE.write_text(json.dumps({
        "entry_id": int(globals().get("TEAM_ID", globals().get("ENTRY_ID", 0)) or 0),
        "gameweek": PHASE11_CURRENT_GW,
        "snapshot_timestamp": "YYYY-MM-DDTHH:MM:SS+00:00",
        "source_name": "MANUAL_CURRENT_SQUAD",
        "is_pre_deadline_data": True,
        "bank": 0.0,
        "free_transfers": 1,
        "players": [
            {"player_id": 0, "purchase_price": 0.0, "selling_price": 0.0, "squad_position": 1}
        ],
        "notes": "Replace the sample row with all 15 current players. Do not use stale post-transfer public picks as a pre-deadline current squad.",
    }, indent=2), encoding="utf-8")


def _vx_phase11_money(value):
    try:
        v = float(value)
    except Exception:
        return np.nan
    if not np.isfinite(v):
        return np.nan
    return float(v / 10.0) if v > 20.0 else float(v)


def _vx_phase11_validate_squad_rows(rows, *, bank, free_transfers, source_label):
    frame = pd.DataFrame(rows or [])
    if len(frame) != 15:
        raise ValueError(f"{source_label}: expected exactly 15 squad rows, found {len(frame)}")
    if "player_id" not in frame.columns:
        raise ValueError(f"{source_label}: player_id is required")
    frame["player_id"] = pd.to_numeric(frame["player_id"], errors="coerce").astype("Int64")
    if frame["player_id"].isna().any() or frame["player_id"].duplicated().any():
        raise ValueError(f"{source_label}: player IDs must be 15 unique valid integers")
    frame["player_id"] = frame["player_id"].astype(int)
    known = set(phase11_player_decision_pool["player_id"].astype(int))
    unknown = sorted(set(frame["player_id"]) - known)
    if unknown:
        raise ValueError(f"{source_label}: unknown current-season player IDs: {unknown}")

    meta = phase11_player_decision_pool[["player_id", "team_id", "position", "price", "player_name"]].copy()
    frame = frame.merge(meta, on="player_id", how="left", validate="one_to_one")
    frame["position"] = frame["position"].astype(str).str.upper().replace({"GKP": "GK"})
    required = {"GK": 2, "DEF": 5, "MID": 5, "FWD": 3}
    counts = frame["position"].value_counts().to_dict()
    if any(int(counts.get(pos, 0)) != n for pos, n in required.items()):
        raise ValueError(f"{source_label}: illegal 15-man position counts: {counts}")
    if frame.groupby("team_id").size().max() > 3:
        raise ValueError(f"{source_label}: more than three players from one club")

    for price_col in ("purchase_price", "selling_price"):
        if price_col not in frame.columns:
            frame[price_col] = frame["price"]
        frame[price_col] = frame[price_col].map(_vx_phase11_money)
        frame[price_col] = frame[price_col].where(np.isfinite(frame[price_col]), frame["price"])
        if (frame[price_col] <= 0).any():
            raise ValueError(f"{source_label}: invalid {price_col}")
    if "squad_position" not in frame.columns:
        frame["squad_position"] = np.arange(1, 16)
    frame["squad_position"] = pd.to_numeric(frame["squad_position"], errors="coerce").fillna(99).astype(int)

    bank = _vx_phase11_money(bank)
    if not np.isfinite(bank) or bank < 0:
        raise ValueError(f"{source_label}: bank must be non-negative")
    if free_transfers is not None and not (isinstance(free_transfers, float) and np.isnan(free_transfers)):
        free_transfers = int(free_transfers)
        if free_transfers < 0 or free_transfers > 5:
            raise ValueError(f"{source_label}: free_transfers must be in 0..5")
    return frame, float(bank), free_transfers


def _vx_phase11_manual_squad():
    if not PHASE11_MANUAL_SQUAD_PATH.exists():
        return None
    payload = json.loads(PHASE11_MANUAL_SQUAD_PATH.read_text(encoding="utf-8"))
    if not isinstance(payload, dict):
        raise ValueError("manual current_squad_state.json must contain one JSON object")
    rows, bank, ft = _vx_phase11_validate_squad_rows(
        payload.get("players"),
        bank=payload.get("bank", 0.0),
        free_transfers=payload.get("free_transfers"),
        source_label="MANUAL_CURRENT_SQUAD",
    )
    gw = int(payload.get("gameweek", PHASE11_CURRENT_GW))
    ts = _vx_phase11_utc(payload.get("snapshot_timestamp"))
    if gw != PHASE11_CURRENT_GW:
        raise ValueError(f"manual current squad is for GW{gw}, expected GW{PHASE11_CURRENT_GW}")
    if ts is None:
        raise ValueError("manual current squad requires snapshot_timestamp")
    if _phase11_now is not None and ts > _phase11_now:
        raise ValueError("manual current squad snapshot timestamp cannot be in the future")
    pre_deadline = bool(payload.get("is_pre_deadline_data", False))
    if _phase11_deadline is not None and ts >= _phase11_deadline:
        pre_deadline = False
    return {
        "rows": rows,
        "bank": bank,
        "free_transfers": ft,
        "gameweek": gw,
        "snapshot_timestamp": ts.isoformat(),
        "source_name": str(payload.get("source_name", "MANUAL_CURRENT_SQUAD")),
        "pre_deadline": bool(pre_deadline),
        "fresh_for_decisions": bool(pre_deadline and PHASE11_BEFORE_DEADLINE),
        "state_type": "MANUAL_CURRENT_SQUAD",
    }


def _vx_phase11_public_visible_squad():
    """Last public FPL picks are a baseline only; they may be stale after transfers."""
    try:
        from fpl_data import fetch_gw_review_snapshot
    except Exception:
        return None
    entry_id = int(globals().get("TEAM_ID", globals().get("ENTRY_ID", 0)) or 0)
    if entry_id <= 0:
        return None
    event_ids = []
    if isinstance(_events, pd.DataFrame) and "id" in _events.columns:
        event_ids = sorted(
            int(x) for x in pd.to_numeric(_events["id"], errors="coerce").dropna().astype(int).unique()
            if int(x) <= PHASE11_CURRENT_GW
        )
    if not event_ids:
        event_ids = list(range(1, PHASE11_CURRENT_GW + 1))
    fixtures_payload = None
    _data = globals().get("data")
    if isinstance(_data, dict) and isinstance(_data.get("fixtures"), pd.DataFrame):
        fixtures_payload = _data["fixtures"].to_dict("records")
    bootstrap_payload = _data.get("bootstrap") if isinstance(_data, dict) else None

    for gw in reversed(event_ids):
        try:
            snap = fetch_gw_review_snapshot(
                entry_id, gw,
                bootstrap=bootstrap_payload,
                fixtures=fixtures_payload,
            )
            picks_payload = snap.get("picks") or {}
            picks = list(picks_payload.get("picks") or [])
            if len(picks) != 15:
                continue
            rows = []
            for p in picks:
                rows.append({
                    "player_id": int(p.get("element")),
                    "purchase_price": _vx_phase11_money(p.get("purchase_price")),
                    "selling_price": _vx_phase11_money(p.get("selling_price")),
                    "squad_position": int(p.get("position", 99) or 99),
                })
            entry_history = dict(picks_payload.get("entry_history") or {})
            bank = _vx_phase11_money(entry_history.get("bank", 0.0))
            rows, bank, ft = _vx_phase11_validate_squad_rows(
                rows, bank=bank, free_transfers=None,
                source_label=f"PUBLIC_VISIBLE_GW_{gw}",
            )
            return {
                "rows": rows,
                "bank": bank,
                "free_transfers": None,
                "gameweek": int(gw),
                "snapshot_timestamp": PHASE11_TIMESTAMP,
                "source_name": f"FPL_PUBLIC_PICKS_GW_{gw}",
                "pre_deadline": False,
                "fresh_for_decisions": False,
                "state_type": "PUBLIC_LAST_VISIBLE_BASELINE",
            }
        except Exception:
            continue
    return None


_phase11_squad_state_error = None
try:
    _phase11_squad_state = _vx_phase11_manual_squad()
except Exception as exc:
    _phase11_squad_state = None
    _phase11_squad_state_error = f"MANUAL_SQUAD_INVALID: {type(exc).__name__}: {exc}"
if _phase11_squad_state is None:
    _phase11_squad_state = _vx_phase11_public_visible_squad()

if _phase11_squad_state is None:
    phase11_current_squad_state = pd.DataFrame(columns=[
        "player_id", "purchase_price", "selling_price", "squad_position",
        "team_id", "position", "price", "player_name",
    ])
    PHASE11_SQUAD_STATE_STATUS = _phase11_squad_state_error or "NO_CURRENT_SQUAD_STATE_AVAILABLE"
    PHASE11_BANK = np.nan
    PHASE11_FREE_TRANSFERS = None
    PHASE11_SQUAD_STATE_FRESH = False
else:
    phase11_current_squad_state = _phase11_squad_state["rows"].copy()
    PHASE11_BANK = float(_phase11_squad_state["bank"])
    PHASE11_FREE_TRANSFERS = _phase11_squad_state["free_transfers"]
    PHASE11_SQUAD_STATE_FRESH = bool(_phase11_squad_state["fresh_for_decisions"])
    PHASE11_SQUAD_STATE_STATUS = (
        "CURRENT_PRE_DEADLINE_MANUAL_STATE"
        if PHASE11_SQUAD_STATE_FRESH
        else f"{_phase11_squad_state['state_type']}_NOT_ACTIONABLE"
    )

# Merge xPts and decision features onto the squad.
if not phase11_current_squad_state.empty:
    _merge_cols = [c for c in (
        "player_id", "player_code", "player_name", "team_id", "team_name", "position", "price",
        "final_gameweek_xpts", "next_3_xpts", "next_5_xpts", "phase11_weighted_horizon_xpts",
        "return_mc_mean", "return_mc_std", "return_p10", "return_p25", "return_p50",
        "return_p75", "return_p90", "prob_10_plus", "prob_15_plus",
        "player_model_confidence", "overall_selected_by_percent", "decision_available"
    ) if c in phase11_player_decision_pool.columns]
    _base = phase11_current_squad_state.drop(columns=[c for c in ("team_id", "position", "price", "player_name") if c in phase11_current_squad_state.columns])
    phase11_current_squad_state = _base.merge(
        phase11_player_decision_pool[_merge_cols], on="player_id", how="left", validate="one_to_one"
    )

# ------------------------------
# ELITE EO — OPTIONAL, NEVER FABRICATED
# ------------------------------
PHASE11_ELITE_ROOT = Path(FPL_VORTEX_DATA_ROOT) / "03_ELITE_MANAGERS"
PHASE11_EO_SOURCE_FILES = []
PHASE11_EO_REJECTED_FILES = []


def _vx_phase11_read_any(path):
    try:
        suf = path.suffix.lower()
        if suf == ".csv":
            return pd.read_csv(path)
        if suf in {".json", ".jsonl"}:
            if suf == ".jsonl":
                return pd.read_json(path, lines=True)
            try:
                return pd.read_json(path)
            except Exception:
                payload = json.loads(path.read_text(encoding="utf-8"))
                if isinstance(payload, dict):
                    for key in ("rows", "data", "players", "records"):
                        if isinstance(payload.get(key), list):
                            return pd.DataFrame(payload[key])
                    return pd.DataFrame([payload])
                return pd.DataFrame(payload)
        if suf in {".parquet", ".pq"}:
            return pd.read_parquet(path)
    except Exception:
        return pd.DataFrame()
    return pd.DataFrame()


def _vx_phase11_tier(value, path_text=""):
    s = (str(value) + " " + str(path_text)).lower().replace("-", "_").replace(" ", "_")
    if re.search(r"top_?100k|100000", s): return "TOP_100K"
    if re.search(r"top_?10k|10000", s): return "TOP_10K"
    if re.search(r"top_?1k|1000", s): return "TOP_1K"
    if re.search(r"top_?100\b", s): return "TOP_100"
    return None


def _vx_phase11_normalize_eo(frame, source_path):
    if not isinstance(frame, pd.DataFrame) or frame.empty:
        return pd.DataFrame()
    lower = {str(c).lower(): c for c in frame.columns}
    def pick(*names):
        for n in names:
            if n in lower:
                return lower[n]
        return None
    pid_col = pick("player_id", "element", "element_id", "id")
    pcode_col = pick("player_code", "code")
    eo_col = pick("effective_ownership", "effective_ownership_percent", "eo", "eo_percent", "effectiveownership")
    tier_col = pick("tier", "rank_tier", "cohort", "manager_tier")
    gw_col = pick("gameweek", "gw", "event")
    ts_col = pick("snapshot_timestamp", "source_timestamp", "timestamp", "captured_at")
    if eo_col is None or (pid_col is None and pcode_col is None):
        return pd.DataFrame()
    out = pd.DataFrame()
    if pid_col is not None:
        out["player_id"] = pd.to_numeric(frame[pid_col], errors="coerce")
    else:
        code_map = phase11_player_decision_pool.dropna(subset=["player_code"]).set_index("player_code")["player_id"].to_dict()
        out["player_id"] = pd.to_numeric(frame[pcode_col], errors="coerce").map(code_map)
    out["effective_ownership"] = pd.to_numeric(frame[eo_col], errors="coerce")
    out["tier"] = (
        frame[tier_col].map(lambda x: _vx_phase11_tier(x, source_path))
        if tier_col is not None
        else _vx_phase11_tier("", source_path)
    )
    if not isinstance(out["tier"], pd.Series):
        out["tier"] = [out["tier"]] * len(out)
    out["gameweek"] = pd.to_numeric(frame[gw_col], errors="coerce") if gw_col is not None else PHASE11_CURRENT_GW
    out["snapshot_timestamp"] = frame[ts_col].astype(str) if ts_col is not None else None
    out["source_file"] = str(source_path)
    out = out.dropna(subset=["player_id", "effective_ownership", "tier"]).copy()
    out["player_id"] = out["player_id"].astype(int)
    out["gameweek"] = pd.to_numeric(out["gameweek"], errors="coerce").fillna(PHASE11_CURRENT_GW).astype(int)
    out = out.loc[out["gameweek"].eq(PHASE11_CURRENT_GW)].copy()
    # Deadline safety for decision use: timestamp must be known and no later than deadline.
    safe = []
    for _, r in out.iterrows():
        ts = _vx_phase11_utc(r.get("snapshot_timestamp"))
        safe.append(bool(ts is not None and _phase11_deadline is not None and ts < _phase11_deadline))
    out["deadline_safe_for_decisions"] = safe
    return out


_eo_files = []
for p in (PHASE11_ELITE_EO_MANUAL_CSV, PHASE11_ELITE_EO_MANUAL_JSON):
    if p.exists():
        _eo_files.append(p)
if PHASE11_ELITE_ROOT.exists():
    for ext in ("*.csv", "*.json", "*.jsonl", "*.parquet", "*.pq"):
        _eo_files.extend(PHASE11_ELITE_ROOT.rglob(ext))
# de-duplicate paths
_eo_files = list(dict.fromkeys(str(p) for p in _eo_files))
_eo_parts = []
for ptxt in _eo_files:
    p = Path(ptxt)
    raw = _vx_phase11_read_any(p)
    norm = _vx_phase11_normalize_eo(raw, p)
    if norm.empty:
        PHASE11_EO_REJECTED_FILES.append(str(p))
    else:
        PHASE11_EO_SOURCE_FILES.append(str(p))
        _eo_parts.append(norm)
phase11_elite_eo_raw = pd.concat(_eo_parts, ignore_index=True) if _eo_parts else pd.DataFrame(columns=[
    "player_id", "effective_ownership", "tier", "gameweek", "snapshot_timestamp", "source_file", "deadline_safe_for_decisions"
])

# Use only deadline-safe snapshots for pre-deadline recommendation overlays.
_eo_safe = phase11_elite_eo_raw.loc[phase11_elite_eo_raw.get("deadline_safe_for_decisions", False) == True].copy() if not phase11_elite_eo_raw.empty else pd.DataFrame()
if not _eo_safe.empty:
    _eo_safe["_ts"] = pd.to_datetime(_eo_safe["snapshot_timestamp"], utc=True, errors="coerce")
    _eo_safe = _eo_safe.sort_values("_ts").groupby(["player_id", "tier"], as_index=False).tail(1)
    _eo_pivot = _eo_safe.pivot_table(index="player_id", columns="tier", values="effective_ownership", aggfunc="last").reset_index()
    _eo_pivot = _eo_pivot.rename(columns={
        "TOP_100": "eo_top_100",
        "TOP_1K": "eo_top_1k",
        "TOP_10K": "eo_top_10k",
        "TOP_100K": "eo_top_100k",
    })
else:
    _eo_pivot = pd.DataFrame(columns=["player_id", "eo_top_100", "eo_top_1k", "eo_top_10k", "eo_top_100k"])
phase11_player_decision_pool = phase11_player_decision_pool.merge(_eo_pivot, on="player_id", how="left")
if not phase11_current_squad_state.empty:
    phase11_current_squad_state = phase11_current_squad_state.merge(_eo_pivot, on="player_id", how="left")

PHASE11_EO_STATUS = (
    "DEADLINE_SAFE_ELITE_EO_AVAILABLE"
    if not _eo_safe.empty
    else "WAITING_FOR_APPROVED_DEADLINE_SAFE_ELITE_EO_SNAPSHOT"
)

# ------------------------------
# EXACT CURRENT-SQUAD LINEUP OPTIMISER
# ------------------------------
def _vx_phase11_best_lineup(frame, score_col="final_gameweek_xpts"):
    if not isinstance(frame, pd.DataFrame) or frame.empty or score_col not in frame.columns:
        return pd.DataFrame(), pd.DataFrame()
    f = frame.copy()
    f[score_col] = pd.to_numeric(f[score_col], errors="coerce")
    f = f.dropna(subset=["player_id", "position", score_col]).copy()
    groups = {pos: f.loc[f["position"].astype(str).str.upper().eq(pos)].copy() for pos in ("GK", "DEF", "MID", "FWD")}
    if len(groups["GK"]) < 1 or len(groups["DEF"]) < 3 or len(groups["MID"]) < 2 or len(groups["FWD"]) < 1:
        return pd.DataFrame(), pd.DataFrame()
    best_ids = None
    best_score = -np.inf
    gk_ids = groups["GK"]["player_id"].astype(int).tolist()
    def ids(pos): return groups[pos]["player_id"].astype(int).tolist()
    score_map = f.set_index("player_id")[score_col].to_dict()
    for gk in gk_ids:
        for nd in range(3, 6):
            for nm in range(2, 6):
                nf = 10 - nd - nm
                if nf < 1 or nf > 3:
                    continue
                if nd > len(ids("DEF")) or nm > len(ids("MID")) or nf > len(ids("FWD")):
                    continue
                for dc in combinations(ids("DEF"), nd):
                    for mc in combinations(ids("MID"), nm):
                        for fc in combinations(ids("FWD"), nf):
                            chosen = (gk,) + dc + mc + fc
                            sc = float(sum(float(score_map[x]) for x in chosen))
                            if sc > best_score:
                                best_score = sc
                                best_ids = set(chosen)
    if not best_ids:
        return pd.DataFrame(), pd.DataFrame()
    xi = f.loc[f["player_id"].isin(best_ids)].copy()
    bench = f.loc[~f["player_id"].isin(best_ids)].copy()
    return xi, bench


phase11_recommended_lineup = pd.DataFrame()
phase11_bench_order = pd.DataFrame()
phase11_best_goalkeeper = pd.DataFrame()
if not phase11_current_squad_state.empty:
    phase11_recommended_lineup, _bench = _vx_phase11_best_lineup(phase11_current_squad_state, "final_gameweek_xpts")
    if not phase11_recommended_lineup.empty:
        phase11_recommended_lineup = phase11_recommended_lineup.sort_values(["position", "final_gameweek_xpts"], ascending=[True, False]).reset_index(drop=True)
        _bench_gk = _bench.loc[_bench["position"].eq("GK")].copy()
        _bench_out = _bench.loc[~_bench["position"].eq("GK")].sort_values("final_gameweek_xpts", ascending=False).copy()
        _bench_out["bench_order"] = np.arange(1, len(_bench_out) + 1)
        if not _bench_gk.empty:
            _bench_gk["bench_order"] = 4
        phase11_bench_order = pd.concat([_bench_out, _bench_gk], ignore_index=True)
        phase11_best_goalkeeper = phase11_recommended_lineup.loc[phase11_recommended_lineup["position"].eq("GK")].copy()

# ------------------------------
# CAPTAINCY
# ------------------------------
phase11_captain_rankings = pd.DataFrame()
PHASE11_BEST_CAPTAIN = None
PHASE11_BEST_VICE = None
PHASE11_SAFE_CAPTAIN = None
PHASE11_CEILING_CAPTAIN = None
PHASE11_DIFFERENTIAL_CAPTAIN = None
if not phase11_recommended_lineup.empty:
    _cap = phase11_recommended_lineup.copy()
    _cap["captain_ev"] = 2.0 * pd.to_numeric(_cap["final_gameweek_xpts"], errors="coerce")

    # Mean xPts remains the canonical captain EV. "Safe" and "ceiling" labels
    # now use the explicit Phase-10 return distribution rather than the
    # uncalibrated Phase-3 confidence index.
    for _c in ("return_p25", "return_p50", "return_p90", "prob_10_plus", "prob_15_plus"):
        if _c not in _cap.columns:
            _cap[_c] = np.nan
        _cap[_c] = pd.to_numeric(_cap[_c], errors="coerce")
    _cap["safe_captain_score"] = _cap["return_p50"].where(
        _cap["return_p50"].notna(), pd.to_numeric(_cap["final_gameweek_xpts"], errors="coerce")
    )
    _cap["ceiling_captain_score"] = _cap["prob_10_plus"].where(
        _cap["prob_10_plus"].notna(), 0.0
    )

    _cap = _cap.sort_values(
        ["final_gameweek_xpts", "prob_10_plus", "return_p90"],
        ascending=False, na_position="last"
    ).reset_index(drop=True)
    _cap["captain_rank"] = np.arange(1, len(_cap) + 1)
    phase11_captain_rankings = _cap.copy()
    PHASE11_BEST_CAPTAIN = _cap.iloc[0].to_dict()
    if len(_cap) > 1:
        PHASE11_BEST_VICE = _cap.iloc[1].to_dict()

    _safe = _cap.sort_values(
        ["safe_captain_score", "return_p25", "final_gameweek_xpts"],
        ascending=False, na_position="last"
    )
    PHASE11_SAFE_CAPTAIN = _safe.iloc[0].to_dict() if not _safe.empty else None

    _ceil = _cap.sort_values(
        ["prob_10_plus", "prob_15_plus", "return_p90", "final_gameweek_xpts"],
        ascending=False, na_position="last"
    )
    PHASE11_CEILING_CAPTAIN = _ceil.iloc[0].to_dict() if not _ceil.empty else None

    own_signal_col = "eo_top_1k" if "eo_top_1k" in _cap.columns and _cap["eo_top_1k"].notna().any() else "overall_selected_by_percent"
    _diff = _cap.loc[
        pd.to_numeric(_cap.get(own_signal_col), errors="coerce").fillna(100.0)
        <= PHASE11_DIFFERENTIAL_OWNERSHIP_MAX
    ].copy()
    if not _diff.empty:
        floor = float(_cap["final_gameweek_xpts"].max()) * 0.75
        _diff = _diff.loc[_diff["final_gameweek_xpts"] >= floor].sort_values(
            ["prob_10_plus", "return_p90", "final_gameweek_xpts"],
            ascending=False, na_position="last"
        )
        if not _diff.empty:
            row = _diff.iloc[0].to_dict()
            row["differential_source"] = "TOP_1K_EO" if own_signal_col == "eo_top_1k" else "OVERALL_OWNERSHIP_FALLBACK"
            PHASE11_DIFFERENTIAL_CAPTAIN = row

# ------------------------------
# TRANSFER ENGINE
# ------------------------------
def _vx_phase11_team_counts(frame):
    if not isinstance(frame, pd.DataFrame) or frame.empty:
        return {}
    return frame.groupby("team_id").size().astype(int).to_dict()


def _vx_phase11_legal_single_swap(squad, out_row, in_row, bank):
    if str(out_row["position"]) != str(in_row["position"]):
        return False
    available = float(bank) + float(out_row["selling_price"])
    if float(in_row["price"]) > available + 1e-9:
        return False
    counts = _vx_phase11_team_counts(squad)
    out_team = int(out_row["team_id"]); in_team = int(in_row["team_id"])
    counts[out_team] = counts.get(out_team, 0) - 1
    counts[in_team] = counts.get(in_team, 0) + 1
    return max(counts.values() or [0]) <= 3


_transfer_rows = []
if not phase11_current_squad_state.empty and np.isfinite(PHASE11_BANK):
    current_ids = set(phase11_current_squad_state["player_id"].astype(int))
    incoming_pool = phase11_player_decision_pool.loc[
        phase11_player_decision_pool["decision_pool_ready"]
        & ~phase11_player_decision_pool["player_id"].isin(current_ids)
    ].copy()
    for _, out_row in phase11_current_squad_state.iterrows():
        out_id = int(out_row["player_id"])
        for _, in_row in incoming_pool.loc[incoming_pool["position"].eq(out_row["position"])].iterrows():
            if not _vx_phase11_legal_single_swap(phase11_current_squad_state, out_row, in_row, PHASE11_BANK):
                continue
            g1 = float(in_row["final_gameweek_xpts"] - out_row["final_gameweek_xpts"])
            g3 = float(in_row["next_3_xpts"] - out_row["next_3_xpts"])
            g5 = float(in_row["next_5_xpts"] - out_row["next_5_xpts"])
            wh = float(in_row["phase11_weighted_horizon_xpts"] - out_row["phase11_weighted_horizon_xpts"])
            horizon_avg = wh / max(1e-9, sum(_gw_weights.values()))
            utility = 0.65 * g1 + 0.35 * horizon_avg
            conf = float(in_row.get("player_model_confidence", 0.0) or 0.0)
            # Confidence remains visible for diagnosis but does not alter transfer
            # utility until its relationship with forecast error is calibrated.
            risk_adjusted = utility
            _transfer_rows.append({
                "out_player_id": out_id,
                "out_player_name": out_row.get("player_name"),
                "out_team_id": int(out_row["team_id"]),
                "in_player_id": int(in_row["player_id"]),
                "in_player_name": in_row.get("player_name"),
                "in_team_id": int(in_row["team_id"]),
                "position": str(out_row["position"]),
                "out_selling_price": float(out_row["selling_price"]),
                "in_price": float(in_row["price"]),
                "bank_after": float(PHASE11_BANK + float(out_row["selling_price"]) - float(in_row["price"])),
                "gain_next_gw": g1,
                "gain_next_3": g3,
                "gain_next_5": g5,
                "weighted_horizon_gain": wh,
                "decision_utility": float(utility),
                "risk_adjusted_utility": float(risk_adjusted),
                "incoming_confidence": conf,
                "incoming_overall_ownership": float(in_row["overall_selected_by_percent"]) if np.isfinite(in_row["overall_selected_by_percent"]) else np.nan,
                "incoming_top_1k_eo": float(in_row.get("eo_top_1k")) if np.isfinite(pd.to_numeric(in_row.get("eo_top_1k"), errors="coerce")) else np.nan,
                "squad_state_status": PHASE11_SQUAD_STATE_STATUS,
                "actionable": bool(PHASE11_UPSTREAM_RELEASED and PHASE11_SQUAD_STATE_FRESH and PHASE11_BEFORE_DEADLINE),
            })
phase11_transfer_shortlist = pd.DataFrame(_transfer_rows)
if not phase11_transfer_shortlist.empty:
    phase11_transfer_shortlist = phase11_transfer_shortlist.sort_values(
        ["risk_adjusted_utility", "gain_next_5", "gain_next_gw"], ascending=False
    ).reset_index(drop=True)
    phase11_transfer_shortlist["transfer_rank"] = np.arange(1, len(phase11_transfer_shortlist) + 1)

PHASE11_BEST_FREE_TRANSFER = phase11_transfer_shortlist.iloc[0].to_dict() if not phase11_transfer_shortlist.empty else None

# Sell verdicts are team-specific: compare each owned player with their best legal replacement.
_sell_rows = []
if not phase11_current_squad_state.empty:
    for _, owned in phase11_current_squad_state.iterrows():
        pid = int(owned["player_id"])
        cand = phase11_transfer_shortlist.loc[phase11_transfer_shortlist["out_player_id"].eq(pid)].head(1) if not phase11_transfer_shortlist.empty else pd.DataFrame()
        if cand.empty:
            _sell_rows.append({
                "player_id": pid, "player_name": owned.get("player_name"), "verdict": "HOLD",
                "best_replacement_player_id": None, "best_replacement_name": None,
                "gain_next_5": np.nan, "decision_utility": np.nan,
                "reason": "NO_LEGAL_POSITIVE_REPLACEMENT_EVIDENCE",
            })
            continue
        c = cand.iloc[0]
        sell = bool(float(c["gain_next_5"]) >= PHASE11_SELL_MIN_NEXT5_GAIN and float(c["risk_adjusted_utility"]) >= PHASE11_SELL_MIN_UTILITY)
        _sell_rows.append({
            "player_id": pid, "player_name": owned.get("player_name"), "verdict": "SELL" if sell else "HOLD",
            "best_replacement_player_id": int(c["in_player_id"]), "best_replacement_name": c["in_player_name"],
            "gain_next_5": float(c["gain_next_5"]), "decision_utility": float(c["risk_adjusted_utility"]),
            "reason": "LEGAL_REPLACEMENT_EDGE" if sell else "EDGE_BELOW_SELL_POLICY_THRESHOLD",
        })
phase11_sell_candidates = pd.DataFrame(_sell_rows)
if not phase11_sell_candidates.empty:
    phase11_sell_candidates = phase11_sell_candidates.sort_values(["decision_utility", "gain_next_5"], ascending=False, na_position="last").reset_index(drop=True)

# Buy candidates aggregate the best team-specific route into each incoming player.
if not phase11_transfer_shortlist.empty:
    _buy = phase11_transfer_shortlist.sort_values("risk_adjusted_utility", ascending=False).groupby("in_player_id", as_index=False).first()
    _buy["verdict"] = np.where(_buy["risk_adjusted_utility"] >= PHASE11_BUY_MIN_UTILITY, "BUY", "WATCH")
    phase11_buy_candidates = _buy.sort_values(["risk_adjusted_utility", "gain_next_5"], ascending=False).reset_index(drop=True)
else:
    phase11_buy_candidates = pd.DataFrame()

# Two-transfer combinations. A -4 recommendation is only possible when FT count is known to be one.
_pair_rows = []
if not phase11_transfer_shortlist.empty:
    singles = phase11_transfer_shortlist.head(100).to_dict("records")
    current_counts = _vx_phase11_team_counts(phase11_current_squad_state)
    for a, b in combinations(singles, 2):
        if a["out_player_id"] == b["out_player_id"] or a["in_player_id"] == b["in_player_id"]:
            continue
        if a["in_player_id"] in {b["out_player_id"]} or b["in_player_id"] in {a["out_player_id"]}:
            continue
        funds = PHASE11_BANK + float(a["out_selling_price"]) + float(b["out_selling_price"]) - float(a["in_price"]) - float(b["in_price"])
        if funds < -1e-9:
            continue
        counts = dict(current_counts)
        for r in (a, b):
            counts[int(r["out_team_id"])] = counts.get(int(r["out_team_id"]), 0) - 1
        for r in (a, b):
            counts[int(r["in_team_id"])] = counts.get(int(r["in_team_id"]), 0) + 1
        if max(counts.values() or [0]) > 3:
            continue
        gross1 = float(a["gain_next_gw"] + b["gain_next_gw"])
        gross5 = float(a["gain_next_5"] + b["gain_next_5"])
        util = float(a["risk_adjusted_utility"] + b["risk_adjusted_utility"])
        hit_cost = 4.0 if PHASE11_FREE_TRANSFERS == 1 else (8.0 if PHASE11_FREE_TRANSFERS == 0 else 0.0 if PHASE11_FREE_TRANSFERS is not None else np.nan)
        net1 = gross1 - hit_cost if np.isfinite(hit_cost) else np.nan
        _pair_rows.append({
            "out_1": a["out_player_name"], "in_1": a["in_player_name"],
            "out_2": b["out_player_name"], "in_2": b["in_player_name"],
            "out_1_id": int(a["out_player_id"]), "in_1_id": int(a["in_player_id"]),
            "out_2_id": int(b["out_player_id"]), "in_2_id": int(b["in_player_id"]),
            "bank_after": float(funds), "gross_gain_next_gw": gross1,
            "gross_gain_next_5": gross5, "combined_utility": util,
            "expected_hit_cost": hit_cost, "net_gain_next_gw_after_hit": net1,
        })
phase11_transfer_pairs = pd.DataFrame(_pair_rows)
if not phase11_transfer_pairs.empty:
    phase11_transfer_pairs = phase11_transfer_pairs.sort_values(["net_gain_next_gw_after_hit", "combined_utility", "gross_gain_next_5"], ascending=False, na_position="last").reset_index(drop=True)

PHASE11_BEST_MINUS4 = None
if PHASE11_FREE_TRANSFERS == 1 and not phase11_transfer_pairs.empty:
    eligible = phase11_transfer_pairs.loc[
        (phase11_transfer_pairs["net_gain_next_gw_after_hit"] > 0.0)
        & (phase11_transfer_pairs["combined_utility"] > 0.0)
    ]
    if not eligible.empty:
        PHASE11_BEST_MINUS4 = eligible.iloc[0].to_dict()

# ------------------------------
# MILP SQUAD / MULTI-GW LINEUP OPTIMISER
# ------------------------------
def _vx_phase11_solver_pool():
    pool = phase11_player_decision_pool.loc[phase11_player_decision_pool["decision_pool_ready"]].copy()
    pool = pool.drop_duplicates("player_id").reset_index(drop=True)
    return pool


def _vx_phase11_milp_squad(pool, budget, gameweeks, weights, *, bench_weight=PHASE11_BENCH_MILP_WEIGHT):
    """Static legal 15-man squad + optimal XI for each requested GW using SciPy/HiGHS."""
    try:
        from scipy.optimize import milp, LinearConstraint, Bounds
        from scipy.sparse import coo_matrix
    except Exception as exc:
        return None, None, f"SCIPY_MILP_UNAVAILABLE: {type(exc).__name__}: {exc}"
    if not isinstance(pool, pd.DataFrame) or pool.empty:
        return None, None, "EMPTY_PLAYER_POOL"
    pool = pool.copy().reset_index(drop=True)
    N = len(pool); G = len(gameweeks)
    if N < 15 or G < 1:
        return None, None, "INSUFFICIENT_POOL_OR_GAMEWEEKS"

    xpts_map = {}
    for gw in gameweeks:
        rows = _gw.loc[pd.to_numeric(_gw["gameweek"], errors="coerce") == int(gw), ["player_id", "gameweek_expected_points"]].copy()
        xpts_map[int(gw)] = rows.set_index("player_id")["gameweek_expected_points"].to_dict()
    X = np.zeros((G, N), dtype=float)
    for gi, gw in enumerate(gameweeks):
        for i, r in pool.iterrows():
            val = pd.to_numeric(xpts_map[int(gw)].get(int(r["player_id"])), errors="coerce")
            X[gi, i] = float(val) if np.isfinite(val) else -1000.0
    if np.any(np.max(X, axis=0) <= -999):
        # Players missing every requested GW cannot be selected.
        valid = np.max(X, axis=0) > -999
        pool = pool.loc[valid].reset_index(drop=True)
        X = X[:, valid]
        N = len(pool)
        if N < 15:
            return None, None, "INSUFFICIENT_COMPLETE_XPTS_POOL"

    # Variables: x_i squad, then y_gi starting XI for each GW.
    nvar = N * (1 + G)
    c = np.zeros(nvar, dtype=float)
    weighted_horizon = np.zeros(N, dtype=float)
    for gi, gw in enumerate(gameweeks):
        w = float(weights[gi])
        weighted_horizon += w * X[gi]
        y0 = N * (1 + gi)
        c[y0:y0+N] = -w * X[gi]
    c[:N] = -float(bench_weight) * weighted_horizon

    rr = []; cc = []; vv = []; lb = []; ub = []
    row_no = 0
    def add(coeff, lo=-np.inf, hi=np.inf):
        nonlocal row_no
        for j, val in coeff.items():
            if val != 0:
                rr.append(row_no); cc.append(int(j)); vv.append(float(val))
        lb.append(float(lo)); ub.append(float(hi)); row_no += 1

    # Exact 15-man squad and FPL positional quotas.
    add({i: 1.0 for i in range(N)}, 15, 15)
    req = {"GK": 2, "DEF": 5, "MID": 5, "FWD": 3}
    pos = pool["position"].astype(str).str.upper().tolist()
    for p, n in req.items():
        add({i: 1.0 for i in range(N) if pos[i] == p}, n, n)
    # Max three per club.
    team_ids = pd.to_numeric(pool["team_id"], errors="coerce").astype(int).tolist()
    for tid in sorted(set(team_ids)):
        add({i: 1.0 for i in range(N) if team_ids[i] == tid}, -np.inf, 3)
    # Budget.
    prices = pd.to_numeric(pool["price"], errors="coerce").to_numpy(float)
    add({i: prices[i] for i in range(N)}, -np.inf, float(budget))

    for gi, gw in enumerate(gameweeks):
        y0 = N * (1 + gi)
        # XI = 11, and y_i <= x_i.
        add({y0+i: 1.0 for i in range(N)}, 11, 11)
        for i in range(N):
            add({y0+i: 1.0, i: -1.0}, -np.inf, 0.0)
        # Formation rules.
        add({y0+i: 1.0 for i in range(N) if pos[i] == "GK"}, 1, 1)
        add({y0+i: 1.0 for i in range(N) if pos[i] == "DEF"}, 3, 5)
        add({y0+i: 1.0 for i in range(N) if pos[i] == "MID"}, 2, 5)
        add({y0+i: 1.0 for i in range(N) if pos[i] == "FWD"}, 1, 3)

    A = coo_matrix((vv, (rr, cc)), shape=(row_no, nvar)).tocsr()
    constraints = LinearConstraint(A, np.asarray(lb), np.asarray(ub))
    bounds = Bounds(np.zeros(nvar), np.ones(nvar))
    try:
        res = milp(
            c=c,
            integrality=np.ones(nvar, dtype=int),
            bounds=bounds,
            constraints=constraints,
            options={"time_limit": float(PHASE11_MILP_TIME_LIMIT_SECONDS)},
        )
    except Exception as exc:
        return None, None, f"MILP_ERROR: {type(exc).__name__}: {exc}"
    if res is None or res.x is None or not bool(getattr(res, "success", False)):
        return None, None, f"MILP_NOT_OPTIMAL: {getattr(res, 'message', 'unknown')}"
    x = res.x[:N]
    selected = pool.loc[x > 0.5].copy()
    if len(selected) != 15:
        return None, None, f"MILP_ILLEGAL_SELECTED_COUNT_{len(selected)}"
    lineup_rows = []
    for gi, gw in enumerate(gameweeks):
        y0 = N * (1 + gi)
        y = res.x[y0:y0+N]
        ids = set(pool.loc[y > 0.5, "player_id"].astype(int))
        if len(ids) != 11:
            return None, None, f"MILP_ILLEGAL_XI_COUNT_GW{gw}_{len(ids)}"
        for pid in ids:
            lineup_rows.append({"gameweek": int(gw), "player_id": int(pid), "lineup": True})
    return selected.reset_index(drop=True), pd.DataFrame(lineup_rows), "OPTIMAL_HIGHS_MILP"


def _vx_phase11_greedy_legal_squad(pool, budget, score_col):
    """Fail-safe legal fallback; used only if the exact MILP is unavailable."""
    if pool.empty:
        return pd.DataFrame()
    req = {"GK": 2, "DEF": 5, "MID": 5, "FWD": 3}
    selected = []
    team_counts = {}
    spend = 0.0
    # Value-aware ordering prevents an expensive-first dead end.
    p = pool.copy()
    p["_value"] = pd.to_numeric(p[score_col], errors="coerce") / pd.to_numeric(p["price"], errors="coerce").clip(lower=0.1)
    for pos, n in req.items():
        group = p.loc[p["position"].eq(pos)].sort_values(["_value", score_col], ascending=False)
        for _, r in group.iterrows():
            tid = int(r["team_id"]); price = float(r["price"])
            if team_counts.get(tid, 0) >= 3 or spend + price > budget + 1e-9:
                continue
            selected.append(r); team_counts[tid] = team_counts.get(tid, 0) + 1; spend += price
            if sum(1 for x in selected if str(x["position"]) == pos) >= n:
                break
    out = pd.DataFrame(selected)
    if len(out) != 15:
        return pd.DataFrame()
    return out.drop(columns=["_value"], errors="ignore").reset_index(drop=True)


# Current squad budget at selling value + bank is the legal purchasing budget.
PHASE11_SQUAD_BUDGET = (
    float(phase11_current_squad_state["selling_price"].sum() + PHASE11_BANK)
    if not phase11_current_squad_state.empty and np.isfinite(PHASE11_BANK)
    else 100.0
)
_solver_pool = _vx_phase11_solver_pool()

# Free Hit: exact one-GW squad + XI, then captain from the selected XI.
phase11_free_hit_squad = pd.DataFrame()
phase11_free_hit_lineup = pd.DataFrame()
PHASE11_FREE_HIT_SOLVER_STATUS = "NOT_RUN"
if not _solver_pool.empty:
    _fh_sq, _fh_plan, _fh_status = _vx_phase11_milp_squad(
        _solver_pool, PHASE11_SQUAD_BUDGET, [PHASE11_CURRENT_GW], [1.0]
    )
    PHASE11_FREE_HIT_SOLVER_STATUS = _fh_status
    if _fh_sq is None:
        _fh_sq = _vx_phase11_greedy_legal_squad(_solver_pool, PHASE11_SQUAD_BUDGET, "final_gameweek_xpts")
        if not _fh_sq.empty:
            _fh_xi, _ = _vx_phase11_best_lineup(_fh_sq, "final_gameweek_xpts")
            _fh_plan = pd.DataFrame({"gameweek": PHASE11_CURRENT_GW, "player_id": _fh_xi["player_id"], "lineup": True}) if not _fh_xi.empty else pd.DataFrame()
            PHASE11_FREE_HIT_SOLVER_STATUS += "|GREEDY_LEGAL_FALLBACK"
    if _fh_sq is not None and not _fh_sq.empty:
        phase11_free_hit_squad = _fh_sq.copy()
        _fh_ids = set(_fh_plan.loc[_fh_plan["gameweek"].eq(PHASE11_CURRENT_GW), "player_id"].astype(int)) if isinstance(_fh_plan, pd.DataFrame) and not _fh_plan.empty else set()
        phase11_free_hit_squad["role"] = np.where(phase11_free_hit_squad["player_id"].isin(_fh_ids), "XI", "BENCH")
        phase11_free_hit_lineup = phase11_free_hit_squad.loc[phase11_free_hit_squad["role"].eq("XI")].copy()

# Wildcard: one static 15-man squad optimised over the next five GWs with a legal XI each GW.
phase11_wildcard_squad = pd.DataFrame()
phase11_wildcard_lineup_plan = pd.DataFrame()
PHASE11_WILDCARD_SOLVER_STATUS = "NOT_RUN"
if not _solver_pool.empty:
    _wc_weights = [PHASE11_LINEUP_WEIGHTS[i] for i in range(len(_future_gws))]
    _wc_sq, _wc_plan, _wc_status = _vx_phase11_milp_squad(
        _solver_pool, PHASE11_SQUAD_BUDGET, _future_gws, _wc_weights
    )
    PHASE11_WILDCARD_SOLVER_STATUS = _wc_status
    if _wc_sq is None:
        _wc_sq = _vx_phase11_greedy_legal_squad(_solver_pool, PHASE11_SQUAD_BUDGET, "phase11_weighted_horizon_xpts")
        if not _wc_sq.empty:
            plan_rows = []
            for gw in _future_gws:
                score_map = _gw.loc[pd.to_numeric(_gw["gameweek"], errors="coerce").eq(gw), ["player_id", "gameweek_expected_points"]].rename(columns={"gameweek_expected_points": "_gw_score"})
                tmp = _wc_sq.merge(score_map, on="player_id", how="left")
                xi, _ = _vx_phase11_best_lineup(tmp, "_gw_score")
                for pid in xi.get("player_id", []):
                    plan_rows.append({"gameweek": gw, "player_id": int(pid), "lineup": True})
            _wc_plan = pd.DataFrame(plan_rows)
            PHASE11_WILDCARD_SOLVER_STATUS += "|GREEDY_LEGAL_FALLBACK"
    if _wc_sq is not None and not _wc_sq.empty:
        phase11_wildcard_squad = _wc_sq.copy()
        phase11_wildcard_lineup_plan = _wc_plan.copy() if isinstance(_wc_plan, pd.DataFrame) else pd.DataFrame()
        current_ids = set(phase11_wildcard_lineup_plan.loc[phase11_wildcard_lineup_plan["gameweek"].eq(PHASE11_CURRENT_GW), "player_id"].astype(int)) if not phase11_wildcard_lineup_plan.empty else set()
        phase11_wildcard_squad["current_gw_role"] = np.where(phase11_wildcard_squad["player_id"].isin(current_ids), "XI", "BENCH")

# ------------------------------
# CURRENT / WILDCARD / FREE-HIT SCORE PLANS
# ------------------------------
def _vx_phase11_static_plan_score(squad, gameweeks):
    rows = []
    if not isinstance(squad, pd.DataFrame) or squad.empty:
        return pd.DataFrame()
    ids = set(squad["player_id"].astype(int))
    meta_cols = [c for c in ("player_id", "player_name", "position", "team_id") if c in squad.columns]
    meta = squad[meta_cols].drop_duplicates("player_id")
    for gw in gameweeks:
        scores = _gw.loc[pd.to_numeric(_gw["gameweek"], errors="coerce").eq(gw), ["player_id", "gameweek_expected_points"]].copy()
        tmp = meta.merge(scores, on="player_id", how="left")
        tmp = tmp.loc[tmp["player_id"].isin(ids)].rename(columns={"gameweek_expected_points": "score"})
        xi, bench = _vx_phase11_best_lineup(tmp, "score")
        if xi.empty:
            continue
        cap = xi.sort_values("score", ascending=False).iloc[0]
        rows.append({
            "gameweek": int(gw),
            "xi_xpts": float(pd.to_numeric(xi["score"], errors="coerce").sum()),
            "captain_player_id": int(cap["player_id"]),
            "captain_name": cap.get("player_name"),
            "captain_incremental_xpts": float(cap["score"]),
            "xi_plus_captain_xpts": float(pd.to_numeric(xi["score"], errors="coerce").sum() + float(cap["score"])),
            "bench_xpts": float(pd.to_numeric(bench["score"], errors="coerce").sum()) if not bench.empty else 0.0,
        })
    return pd.DataFrame(rows)

phase11_current_squad_plan = _vx_phase11_static_plan_score(phase11_current_squad_state, _future_gws)
phase11_wildcard_plan_scores = _vx_phase11_static_plan_score(phase11_wildcard_squad, _future_gws)
phase11_free_hit_plan_scores = _vx_phase11_static_plan_score(phase11_free_hit_squad, [PHASE11_CURRENT_GW])

# ------------------------------
# CHIP SCENARIOS
# ------------------------------
def _weighted_plan_total(frame):
    if not isinstance(frame, pd.DataFrame) or frame.empty:
        return np.nan
    mp = frame.set_index("gameweek")["xi_plus_captain_xpts"].to_dict()
    vals = []
    for i, gw in enumerate(_future_gws):
        if gw not in mp or not np.isfinite(mp[gw]):
            return np.nan
        vals.append(PHASE11_LINEUP_WEIGHTS[i] * float(mp[gw]))
    return float(sum(vals))

def _vx_phase11_gw_plan_value(frame, gw):
    if not isinstance(frame, pd.DataFrame) or frame.empty or not {"gameweek", "xi_plus_captain_xpts"}.issubset(frame.columns):
        return np.nan
    hit = frame.loc[pd.to_numeric(frame["gameweek"], errors="coerce").eq(int(gw)), "xi_plus_captain_xpts"]
    return float(hit.iloc[0]) if not hit.empty and np.isfinite(pd.to_numeric(hit.iloc[0], errors="coerce")) else np.nan

_current_gw_team_ev = _vx_phase11_gw_plan_value(phase11_current_squad_plan, PHASE11_CURRENT_GW)
_fh_gw_ev = _vx_phase11_gw_plan_value(phase11_free_hit_plan_scores, PHASE11_CURRENT_GW)
_current_weighted = _weighted_plan_total(phase11_current_squad_plan)
_wc_weighted = _weighted_plan_total(phase11_wildcard_plan_scores)
_tc_increment = float(PHASE11_BEST_CAPTAIN["final_gameweek_xpts"]) if PHASE11_BEST_CAPTAIN is not None else np.nan
_bb_gross = float(pd.to_numeric(phase11_bench_order["final_gameweek_xpts"], errors="coerce").sum()) if not phase11_bench_order.empty else np.nan

phase11_chip_scenarios = pd.DataFrame([
    {
        "chip": "TRIPLE_CAPTAIN",
        "candidate_player": PHASE11_BEST_CAPTAIN.get("player_name") if PHASE11_BEST_CAPTAIN else None,
        "incremental_expected_points": _tc_increment,
        "comparison_basis": "ONE_EXTRA_CAPTAIN_MULTIPLE",
        "automatic_play_recommendation": False,
    },
    {
        "chip": "BENCH_BOOST",
        "candidate_player": None,
        "incremental_expected_points": _bb_gross,
        "comparison_basis": "GROSS_BENCH_XPTS_NOT_NET_OF_AUTOSUB_BASELINE",
        "automatic_play_recommendation": False,
    },
    {
        "chip": "FREE_HIT",
        "candidate_player": None,
        "incremental_expected_points": float(_fh_gw_ev - _current_gw_team_ev) if np.isfinite(_fh_gw_ev) and np.isfinite(_current_gw_team_ev) else np.nan,
        "comparison_basis": "OPTIMISED_ONE_GW_XI_PLUS_CAPTAIN_MINUS_CURRENT_SQUAD",
        "automatic_play_recommendation": False,
    },
    {
        "chip": "WILDCARD",
        "candidate_player": None,
        "incremental_expected_points": float(_wc_weighted - _current_weighted) if np.isfinite(_wc_weighted) and np.isfinite(_current_weighted) else np.nan,
        "comparison_basis": "WEIGHTED_NEXT5_OPTIMISED_XI_PLUS_CAPTAIN_MINUS_CURRENT_STATIC_SQUAD",
        "automatic_play_recommendation": False,
    },
])

# ------------------------------
# EO DASHBOARDS / RANK-RISK FLAGS
# ------------------------------
_eo_cols = [c for c in ("eo_top_100", "eo_top_1k", "eo_top_10k", "eo_top_100k") if c in phase11_player_decision_pool.columns]
phase11_eo_dashboard = phase11_player_decision_pool[[c for c in (
    "player_id", "player_name", "team_name", "position", "price", "final_gameweek_xpts", "next_5_xpts",
    "overall_selected_by_percent", "eo_top_100", "eo_top_1k", "eo_top_10k", "eo_top_100k"
) if c in phase11_player_decision_pool.columns]].copy()
for c in ("eo_top_100", "eo_top_1k", "eo_top_10k", "eo_top_100k"):
    if c not in phase11_eo_dashboard.columns:
        phase11_eo_dashboard[c] = np.nan

PHASE11_EO_TIER_TABLES = {}
for tier_col, tier_label in (
    ("eo_top_100", "TOP_100"),
    ("eo_top_1k", "TOP_1K"),
    ("eo_top_10k", "TOP_10K"),
    ("eo_top_100k", "TOP_100K"),
):
    tab = phase11_eo_dashboard.loc[phase11_eo_dashboard[tier_col].notna()].sort_values([tier_col, "final_gameweek_xpts"], ascending=False).head(50)
    PHASE11_EO_TIER_TABLES[tier_label] = tab.to_dict("records")

phase11_rank_protection = pd.DataFrame()
phase11_fade_candidates = pd.DataFrame()
PHASE11_HIGHEST_TOP1K_EO_RISK = None
if "eo_top_1k" in phase11_eo_dashboard.columns and phase11_eo_dashboard["eo_top_1k"].notna().any():
    _eo1 = phase11_eo_dashboard.dropna(subset=["eo_top_1k", "final_gameweek_xpts"]).copy()
    _eo1["rank_risk_score"] = _eo1["eo_top_1k"] / 100.0 * _eo1["final_gameweek_xpts"]
    owned_ids = set(phase11_current_squad_state["player_id"].astype(int)) if not phase11_current_squad_state.empty else set()
    _eo1["owned"] = _eo1["player_id"].isin(owned_ids)
    protection = _eo1.loc[_eo1["owned"] & (_eo1["eo_top_1k"] >= PHASE11_PROTECTION_EO_MIN)].sort_values("rank_risk_score", ascending=False)
    phase11_rank_protection = protection.copy()
    _nonowned = _eo1.loc[~_eo1["owned"]].sort_values("rank_risk_score", ascending=False)
    if not _nonowned.empty:
        PHASE11_HIGHEST_TOP1K_EO_RISK = _nonowned.iloc[0].to_dict()
    # Fade = materially owned by elite managers but below the position median xPts.
    med = _eo1.groupby("position")["final_gameweek_xpts"].median().to_dict()
    _eo1["position_median_xpts"] = _eo1["position"].map(med)
    phase11_fade_candidates = _eo1.loc[
        (~_eo1["owned"])
        & (_eo1["eo_top_1k"] >= 20.0)
        & (_eo1["final_gameweek_xpts"] < _eo1["position_median_xpts"])
    ].sort_values("eo_top_1k", ascending=False)

# ------------------------------
# FINAL DECISION PACKAGE
# ------------------------------
PHASE11_ACTIONABLE = bool(
    PHASE11_UPSTREAM_RELEASED
    and PHASE11_SQUAD_STATE_FRESH
    and PHASE11_BEFORE_DEADLINE
    and not phase11_recommended_lineup.empty
)
PHASE11_OUTPUT_MODE = "ACTIONABLE_VALIDATED_MODEL" if PHASE11_ACTIONABLE else "RESEARCH_ONLY_NOT_ACTIONABLE"

_best_sell = None
if not phase11_sell_candidates.empty:
    sells = phase11_sell_candidates.loc[phase11_sell_candidates["verdict"].eq("SELL")]
    if not sells.empty:
        _best_sell = sells.iloc[0].to_dict()
_best_buy = None
if not phase11_buy_candidates.empty:
    buys = phase11_buy_candidates.loc[phase11_buy_candidates["verdict"].eq("BUY")]
    if not buys.empty:
        _best_buy = buys.iloc[0].to_dict()

PHASE11_RESEARCH_DECISIONS = {
    "schema_version": 11,
    "phase11_version": PHASE11_VERSION,
    "forecast_timestamp": PHASE11_TIMESTAMP,
    "gameweek": PHASE11_CURRENT_GW,
    "output_mode": PHASE11_OUTPUT_MODE,
    "actionable": PHASE11_ACTIONABLE,
    "upstream_phase10_released": PHASE11_UPSTREAM_RELEASED,
    "phase10_output_mode": PHASE11_PHASE10_OUTPUT_MODE,
    "before_deadline": PHASE11_BEFORE_DEADLINE,
    "deadline_time": _phase11_deadline.isoformat() if _phase11_deadline is not None else None,
    "squad_state_status": PHASE11_SQUAD_STATE_STATUS,
    "squad_state_fresh": PHASE11_SQUAD_STATE_FRESH,
    "squad_state_error": _phase11_squad_state_error,
    "bank": float(PHASE11_BANK) if np.isfinite(PHASE11_BANK) else None,
    "free_transfers": PHASE11_FREE_TRANSFERS,
    "best_player_to_buy": _best_buy,
    "best_player_to_sell": _best_sell,
    "best_free_transfer": PHASE11_BEST_FREE_TRANSFER,
    "best_minus4_hit": PHASE11_BEST_MINUS4,
    "best_captain": PHASE11_BEST_CAPTAIN,
    "best_vice_captain": PHASE11_BEST_VICE,
    "safe_captain": PHASE11_SAFE_CAPTAIN,
    "ceiling_captain": PHASE11_CEILING_CAPTAIN,
    "differential_captain": PHASE11_DIFFERENTIAL_CAPTAIN,
    "highest_top1k_eo_risk": PHASE11_HIGHEST_TOP1K_EO_RISK,
    "best_goalkeeper_choice": phase11_best_goalkeeper.to_dict("records"),
    "recommended_lineup": phase11_recommended_lineup.to_dict("records"),
    "bench_order": phase11_bench_order.to_dict("records"),
    "captain_rankings": phase11_captain_rankings.to_dict("records"),
    "transfer_shortlist": phase11_transfer_shortlist.head(50).to_dict("records") if not phase11_transfer_shortlist.empty else [],
    "sell_candidates": phase11_sell_candidates.to_dict("records") if not phase11_sell_candidates.empty else [],
    "buy_candidates": phase11_buy_candidates.head(50).to_dict("records") if not phase11_buy_candidates.empty else [],
    "players_to_protect": phase11_rank_protection.head(20).to_dict("records") if not phase11_rank_protection.empty else [],
    "players_to_fade": phase11_fade_candidates.head(20).to_dict("records") if not phase11_fade_candidates.empty else [],
    "wildcard_squad": phase11_wildcard_squad.to_dict("records") if not phase11_wildcard_squad.empty else [],
    "free_hit_squad": phase11_free_hit_squad.to_dict("records") if not phase11_free_hit_squad.empty else [],
    "chip_scenarios": phase11_chip_scenarios.to_dict("records"),
    "eo_status": PHASE11_EO_STATUS,
    "eo_source_files": PHASE11_EO_SOURCE_FILES,
    "solver_status": {
        "wildcard": PHASE11_WILDCARD_SOLVER_STATUS,
        "free_hit": PHASE11_FREE_HIT_SOLVER_STATUS,
    },
    "policy_notes": [
        "Player expected points come only from Phase 10; ownership and elite EO never feed xPts.",
        "Public last-visible picks are never treated as a fresh pre-deadline squad after unseen transfers.",
        "A fresh manual current-squad snapshot is required for actionable pre-deadline team-specific decisions.",
        "Top-tier EO dashboards remain unavailable unless deadline-safe EO snapshots actually exist.",
        "Wildcard and Free Hit use legal squad constraints with an exact HiGHS MILP when SciPy MILP is available.",
        "Bench Boost value is gross bench xPts and is not mislabeled as net value after autosubs.",
        "No chip is auto-played by Phase 11; chip scenarios are evidence for the manager.",
    ],
}

# Safe downstream contract: research package always exists, but CURRENT_SQUAD_DECISIONS
# is publishable only after both model and current-squad gates are genuinely open.
if PHASE11_ACTIONABLE:
    CURRENT_SQUAD_DECISIONS = dict(PHASE11_RESEARCH_DECISIONS)
else:
    CURRENT_SQUAD_DECISIONS = {
        "schema_version": 11,
        "phase11_version": PHASE11_VERSION,
        "gameweek": PHASE11_CURRENT_GW,
        "actionable": False,
        "output_mode": PHASE11_OUTPUT_MODE,
        "blocked_reason": (
            "PHASE10_VALIDATION_RELEASE_NOT_OPEN"
            if not PHASE11_UPSTREAM_RELEASED
            else "FRESH_PRE_DEADLINE_CURRENT_SQUAD_STATE_REQUIRED"
            if not PHASE11_SQUAD_STATE_FRESH
            else "FPL_DEADLINE_HAS_PASSED"
            if not PHASE11_BEFORE_DEADLINE
            else "DECISION_ENGINE_INCOMPLETE"
        ),
    }

# ------------------------------
# PERSISTENCE — OUTSIDE DISPOSABLE VIDEO OUTPUT
# ------------------------------
PHASE11_OUTPUT_ROOT = Path(PROJECT_ROOT) / "DECISION_ENGINE"
PHASE11_OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
PHASE11_LATEST_PATH = PHASE11_OUTPUT_ROOT / "phase11_latest_decision_package.json"
PHASE11_PLAYER_POOL_PATH = PHASE11_OUTPUT_ROOT / "phase11_player_decision_pool.json"
PHASE11_TRANSFERS_PATH = PHASE11_OUTPUT_ROOT / "phase11_transfer_shortlist.json"
PHASE11_EO_PATH = PHASE11_OUTPUT_ROOT / "phase11_eo_dashboard.json"
PHASE11_WC_PATH = PHASE11_OUTPUT_ROOT / "phase11_wildcard_squad.json"
PHASE11_CHIP_PATH = PHASE11_OUTPUT_ROOT / "phase11_chip_scenarios.json"

PHASE11_LATEST_PATH.write_text(json.dumps(PHASE11_RESEARCH_DECISIONS, indent=2, default=str), encoding="utf-8")
phase11_player_decision_pool.to_json(PHASE11_PLAYER_POOL_PATH, orient="records", date_format="iso", indent=2)
phase11_transfer_shortlist.to_json(PHASE11_TRANSFERS_PATH, orient="records", date_format="iso", indent=2)
phase11_eo_dashboard.to_json(PHASE11_EO_PATH, orient="records", date_format="iso", indent=2)
phase11_wildcard_squad.to_json(PHASE11_WC_PATH, orient="records", date_format="iso", indent=2)
phase11_chip_scenarios.to_json(PHASE11_CHIP_PATH, orient="records", date_format="iso", indent=2)

# ------------------------------
# FINAL INTEGRITY GATE
# ------------------------------
_PHASE11_STATE_AFTER = {k: vx_frame_digest(globals()[k]) for k in _PHASE11_IMMUTABLE}
if _PHASE11_STATE_BEFORE != _PHASE11_STATE_AFTER:
    raise RuntimeError("Phase 11 integrity failure: decision engine mutated an upstream forecast table.")
if float(globals().get("MARKET_BLEND_WEIGHT", 0.0)) != 0.0 or float(globals().get("ELO_BLEND_WEIGHT", 0.0)) != 0.0:
    raise RuntimeError("Phase 11 integrity failure: external benchmark weights changed.")

PHASE11_READY = True
PHASE11_READY_FOR_PRESENTATION = bool(PHASE11_ACTIONABLE)

print("✅ PHASE 11 — FPL DECISION ENGINE BUILT")
print(f"   Output mode: {PHASE11_OUTPUT_MODE}")
print(f"   Upstream Phase-10 release: {PHASE11_UPSTREAM_RELEASED}")
print(f"   Squad state: {PHASE11_SQUAD_STATE_STATUS}")
print(f"   Elite EO: {PHASE11_EO_STATUS}")
print(f"   Transfer rows: {len(phase11_transfer_shortlist)}")
print(f"   Wildcard solver: {PHASE11_WILDCARD_SOLVER_STATUS}")
print(f"   Free-Hit solver: {PHASE11_FREE_HIT_SOLVER_STATUS}")
print("🔒 Team lambdas / probabilities / CS / FDR and Phase-10 xPts remain immutable.")
print("🔒 Market and Team Elo live blend weights remain 0.00.")
if not PHASE11_ACTIONABLE:
    print("⛔ Research outputs exist, but actionable recommendations remain blocked until the release + fresh-squad gates are open.")
print("📦 Phase 11 package:", PHASE11_LATEST_PATH)
```

```python
# ============================================================
# CELL 13F — PHASE 12
# CONTINUOUS IMPROVEMENT / WEEKLY GOVERNANCE LOOP
# FREEZE → OUTCOMES → VALIDATE → ERROR REVIEW → VERSION CONTROL
# ============================================================
# POLICY
# 1) Forecasts are frozen only BEFORE the official FPL deadline.
# 2) Frozen forecast packages are immutable; actuals are attached separately.
# 3) Raw/current inputs are fingerprinted and deadline-sensitive inputs are copied.
# 4) Team, player and decision outputs are evaluated after results arrive.
# 5) Candidate changes are tested out-of-sample and never silently promoted.
# 6) A release needs BOTH objective improvement evidence and explicit approval.
# 7) Phase 12 is governance only: it must never mutate live lambdas, probabilities,
#    FDR, player xPts, transfer/captain/squad outputs or external blend weights.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import math
import os
import re
import shutil

import numpy as np
import pandas as pd

PHASE12_VERSION = "VORTEX-V1-PHASE12-CONTINUOUS-IMPROVEMENT"
PHASE12_TIMESTAMP = datetime.now(timezone.utc).isoformat()

# ----------------------------------------------------------------
# EXISTING DRIVE GOVERNANCE WORKSPACE
# Reuse the folders already present in FPL_VORTEX_DATA.
# ----------------------------------------------------------------
PHASE12_WORKSPACE_ROOT = Path(
    globals().get(
        "PHASE12_WORKSPACE_ROOT_OVERRIDE",
        FPL_VORTEX_DATA_ROOT,
    )
)
PHASE12_MODEL_HISTORY_DIR = PHASE12_WORKSPACE_ROOT / "05_MODEL_HISTORY"
PHASE12_SNAPSHOT_DIR = PHASE12_WORKSPACE_ROOT / "06_SNAPSHOTS"
PHASE12_LOG_DIR = PHASE12_WORKSPACE_ROOT / "07_LOGS"

if not PHASE12_WORKSPACE_ROOT.exists():
    raise RuntimeError(
        "Phase 12 governance workspace is missing: "
        f"{PHASE12_WORKSPACE_ROOT}. Refusing to create an untracked replacement root."
    )
for _p in (PHASE12_MODEL_HISTORY_DIR, PHASE12_SNAPSHOT_DIR, PHASE12_LOG_DIR):
    _p.mkdir(parents=True, exist_ok=True)

PHASE12_CANDIDATE_DIR = PHASE12_MODEL_HISTORY_DIR / "candidates"
PHASE12_RELEASE_DIR = PHASE12_MODEL_HISTORY_DIR / "releases"
PHASE12_CANDIDATE_DIR.mkdir(parents=True, exist_ok=True)
PHASE12_RELEASE_DIR.mkdir(parents=True, exist_ok=True)

PHASE12_RUN_LOG = PHASE12_LOG_DIR / "phase12_weekly_runs.jsonl"
PHASE12_ERROR_LOG = PHASE12_LOG_DIR / "phase12_refresh_errors.jsonl"
PHASE12_PROMOTION_LOG = PHASE12_LOG_DIR / "phase12_version_promotions.jsonl"
PHASE12_VERSION_REGISTRY_PATH = PHASE12_MODEL_HISTORY_DIR / "phase12_version_registry.json"
PHASE12_VERSION_CATALOG_PATH = PHASE12_MODEL_HISTORY_DIR / "phase12_version_catalog.json"

PHASE12_INPUT_ROOT = Path(globals().get("PHASE12_INPUT_ROOT_OVERRIDE", Path(PROJECT_ROOT) / "MODEL_INPUTS"))
PHASE12_INPUT_ROOT.mkdir(parents=True, exist_ok=True)
PHASE12_RELEASE_APPROVAL_PATH = PHASE12_INPUT_ROOT / "phase12_release_approval.json"
PHASE12_RELEASE_APPROVAL_TEMPLATE = PHASE12_INPUT_ROOT / "phase12_release_approval_TEMPLATE.json"

# ----------------------------------------------------------------
# VERSION CATALOG — labels requested for the permanent model lineage.
# These are lineage labels, not a claim that every layer is production-released.
# ----------------------------------------------------------------
PHASE12_VERSION_CATALOG = [
    {
        "version": "FDR_V1.0_MODEL_ONLY",
        "layer": "TEAM_FORECAST_BASELINE",
        "description": "Pure custom model projected goals, Poisson probabilities, CS and FDR.",
        "promotion_gate": "Phase-7 deadline-safe validation evidence.",
    },
    {
        "version": "FDR_V1.1_MARKET_BENCHMARK",
        "layer": "EXTERNAL_BENCHMARK",
        "description": "Market and Team Elo comparisons remain external; no live blend.",
        "promotion_gate": "Benchmark layer only; does not replace the baseline forecast.",
    },
    {
        "version": "FDR_V1.2_MARKET_CALIBRATED",
        "layer": "TEAM_FORECAST_CALIBRATION",
        "description": "A market-calibrated team forecast may exist only after OOS improvement.",
        "promotion_gate": "Phase-9 holdout qualification + explicit release approval.",
    },
    {
        "version": "FDR_V2.0_PLAYER_XPTS",
        "layer": "PLAYER_EXPECTED_POINTS",
        "description": "Validated player expected-points layer downstream of the team forecast.",
        "promotion_gate": "Player-GW validation history + explicit release approval.",
    },
    {
        "version": "FDR_V3.0_FPL_DECISION_ENGINE",
        "layer": "FPL_DECISIONS",
        "description": "Transfers, captaincy, XI, bench, squad and chip decision layer.",
        "promotion_gate": "Actionable Phase-11 gate + decision-history review + explicit release approval.",
    },
]

# ----------------------------------------------------------------
# SMALL SAFE HELPERS
# ----------------------------------------------------------------
def _vx_p12_jsonable(value):
    if value is None:
        return None
    if isinstance(value, (np.integer,)):
        return int(value)
    if isinstance(value, (np.floating,)):
        return None if not np.isfinite(value) else float(value)
    if isinstance(value, (np.bool_,)):
        return bool(value)
    if isinstance(value, (pd.Timestamp, datetime)):
        ts = pd.Timestamp(value)
        if ts.tzinfo is None:
            ts = ts.tz_localize("UTC")
        else:
            ts = ts.tz_convert("UTC")
        return ts.isoformat()
    if isinstance(value, Path):
        return str(value)
    if isinstance(value, pd.DataFrame):
        return [_vx_p12_jsonable(x) for x in value.to_dict("records")]
    if isinstance(value, pd.Series):
        return {str(k): _vx_p12_jsonable(v) for k, v in value.to_dict().items()}
    if isinstance(value, dict):
        return {str(k): _vx_p12_jsonable(v) for k, v in value.items()}
    if isinstance(value, (list, tuple, set)):
        return [_vx_p12_jsonable(x) for x in value]
    if isinstance(value, float):
        return None if not math.isfinite(value) else value
    if value is pd.NA:
        return None
    try:
        if pd.isna(value):
            return None
    except Exception:
        pass
    if isinstance(value, (str, int, bool)):
        return value
    return str(value)


def _vx_p12_canonical_bytes(value):
    payload = _vx_p12_jsonable(value)
    return json.dumps(
        payload,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
        default=str,
    ).encode("utf-8")


def _vx_p12_digest(value):
    return hashlib.sha256(_vx_p12_canonical_bytes(value)).hexdigest()


def _vx_p12_atomic_json(path, payload):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    tmp = path.with_name(path.name + ".tmp")
    tmp.write_text(
        json.dumps(_vx_p12_jsonable(payload), ensure_ascii=False, indent=2, sort_keys=True),
        encoding="utf-8",
    )
    os.replace(tmp, path)
    return path


def _vx_p12_append_jsonl(path, payload):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    line = json.dumps(_vx_p12_jsonable(payload), ensure_ascii=False, sort_keys=True)
    with path.open("a", encoding="utf-8") as fh:
        fh.write(line + "\n")


def _vx_p12_frame(value):
    if isinstance(value, pd.DataFrame):
        return value.copy()
    if isinstance(value, list):
        return pd.DataFrame(value)
    if isinstance(value, dict):
        try:
            return pd.DataFrame(value)
        except Exception:
            return pd.DataFrame([value])
    return pd.DataFrame()


def _vx_p12_safe_slug(text):
    return re.sub(r"[^A-Za-z0-9_.-]+", "_", str(text)).strip("_") or "snapshot"


def _vx_p12_file_fingerprint(path):
    path = Path(path)
    if not path.exists() or not path.is_file():
        return None
    h = hashlib.sha256()
    with path.open("rb") as fh:
        for chunk in iter(lambda: fh.read(1024 * 1024), b""):
            h.update(chunk)
    stat = path.stat()
    return {
        "path": str(path),
        "size_bytes": int(stat.st_size),
        "modified_time_utc": datetime.fromtimestamp(stat.st_mtime, tz=timezone.utc).isoformat(),
        "sha256": h.hexdigest(),
    }


def _vx_p12_deadline_for_gw(events_obj, gw):
    ev = _vx_p12_frame(events_obj)
    if ev.empty or "id" not in ev.columns or "deadline_time" not in ev.columns:
        return None
    x = ev.loc[pd.to_numeric(ev["id"], errors="coerce").eq(int(gw))]
    if x.empty:
        return None
    ts = pd.to_datetime(x.iloc[0]["deadline_time"], utc=True, errors="coerce")
    return None if pd.isna(ts) else pd.Timestamp(ts)


def _vx_p12_current_gw():
    for name in ("PHASE11_CURRENT_GW", "NEXT_GW"):
        raw = globals().get(name)
        try:
            if raw is not None and int(raw) > 0:
                return int(raw)
        except Exception:
            pass
    p = _vx_p12_frame(globals().get("player_expected_points_gameweek"))
    if not p.empty and "gameweek" in p.columns:
        vals = pd.to_numeric(p["gameweek"], errors="coerce").dropna()
        if not vals.empty:
            return int(vals.min())
    raise RuntimeError("Phase 12 cannot determine the current Gameweek.")


def _vx_p12_season_label(ts):
    ts = pd.Timestamp(ts)
    y = int(ts.year if ts.month >= 7 else ts.year - 1)
    return f"{y}-{str(y + 1)[-2:]}"


def _vx_p12_metric_direction_better(candidate, baseline, direction):
    if not (np.isfinite(candidate) and np.isfinite(baseline)):
        return False
    direction = str(direction).strip().lower()
    if direction in {"lower", "min", "minimize"}:
        return float(candidate) < float(baseline)
    if direction in {"higher", "max", "maximize"}:
        return float(candidate) > float(baseline)
    raise ValueError(f"Unknown metric direction: {direction}")


def _vx_p12_metric_nonworse(candidate, baseline, direction):
    if not (np.isfinite(candidate) and np.isfinite(baseline)):
        return False
    direction = str(direction).strip().lower()
    if direction in {"lower", "min", "minimize"}:
        return float(candidate) <= float(baseline)
    if direction in {"higher", "max", "maximize"}:
        return float(candidate) >= float(baseline)
    raise ValueError(f"Unknown metric direction: {direction}")


# ----------------------------------------------------------------
# IMMUTABILITY DIGEST — Phase 12 is not allowed to change forecasts.
# ----------------------------------------------------------------
_PHASE12_IMMUTABLE_KEYS = (
    "projected_goals_model_only",
    "fixture_probabilities_model_only",
    "fixture_fdr_model_only",
    "player_expected_points_fixture",
    "player_expected_points_gameweek",
    "player_expected_points_horizon",
    "PHASE11_RESEARCH_DECISIONS",
    "CURRENT_SQUAD_DECISIONS",
)
_PHASE12_STATE_BEFORE = {
    k: _vx_p12_digest(globals()[k])
    for k in _PHASE12_IMMUTABLE_KEYS
    if k in globals()
}
_PHASE12_MARKET_WEIGHT_BEFORE = float(globals().get("MARKET_BLEND_WEIGHT", 0.0))
_PHASE12_ELO_WEIGHT_BEFORE = float(globals().get("ELO_BLEND_WEIGHT", 0.0))

# ----------------------------------------------------------------
# CURRENT DEADLINE STATE
# ----------------------------------------------------------------
PHASE12_CURRENT_GW = _vx_p12_current_gw()
PHASE12_NOW = pd.Timestamp.now(tz="UTC")
PHASE12_SEASON = _vx_p12_season_label(PHASE12_NOW)
PHASE12_DEADLINE = _vx_p12_deadline_for_gw(globals().get("events"), PHASE12_CURRENT_GW)
PHASE12_BEFORE_DEADLINE = bool(
    PHASE12_DEADLINE is not None and PHASE12_NOW < PHASE12_DEADLINE
)

PHASE12_GW_ROOT = (
    PHASE12_SNAPSHOT_DIR
    / PHASE12_SEASON
    / f"GW{PHASE12_CURRENT_GW:02d}"
)
PHASE12_GW_ROOT.mkdir(parents=True, exist_ok=True)

# ----------------------------------------------------------------
# SOURCE + INPUT MANIFEST
# Historical source archives are fingerprinted by commit/catalogue.
# Deadline-sensitive current objects are copied into the immutable snapshot.
# ----------------------------------------------------------------
_PHASE12_INPUT_OBJECT_NAMES = (
    "fixtures_raw",
    "team_context_raw",
    "team_news_overrides",
    "team_strength_updated",
    "fixture_context_phase2",
    "market_odds_raw",
    "team_elo_benchmark",
    "phase10_player_features",
    "phase11_current_squad_state",
    "phase11_eo_dashboard",
)
_PHASE12_DYNAMIC_COPY_NAMES = (
    "fixtures_raw",
    "team_context_raw",
    "team_news_overrides",
    "team_strength_updated",
    "fixture_context_phase2",
    "market_odds_raw",
    "team_elo_benchmark",
    "phase10_player_features",
    "phase11_current_squad_state",
    "phase11_eo_dashboard",
)

_phase12_input_manifest = {}
for _name in _PHASE12_INPUT_OBJECT_NAMES:
    if _name not in globals():
        continue
    _obj = globals()[_name]
    _fr = _vx_p12_frame(_obj)
    _phase12_input_manifest[_name] = {
        "sha256": _vx_p12_digest(_obj),
        "object_type": type(_obj).__name__,
        "rows": int(len(_fr)) if not _fr.empty else 0,
        "columns": list(map(str, _fr.columns)) if not _fr.empty else [],
    }

# Existing source catalogues / source-sync manifests are part of reproducibility.
_phase12_source_files = []
for _pattern in (
    PHASE12_WORKSPACE_ROOT / "04_MASTER" / "latest_sources.json",
    PHASE12_WORKSPACE_ROOT / "04_MASTER" / "source_archive_manifest.csv",
    PHASE12_WORKSPACE_ROOT / "04_MASTER" / "historical_source_catalogue.csv",
):
    fp = _vx_p12_file_fingerprint(_pattern)
    if fp:
        _phase12_source_files.append(fp)

for _log in sorted(PHASE12_LOG_DIR.glob("source_sync_*.json"))[-5:]:
    fp = _vx_p12_file_fingerprint(_log)
    if fp:
        _phase12_source_files.append(fp)

# ----------------------------------------------------------------
# PHASE VERSION MANIFEST
# ----------------------------------------------------------------
_phase12_phase_versions = {
    "phase12": PHASE12_VERSION,
    "phase4_poisson": globals().get("PHASE4_POISSON_MODEL_VERSION"),
    "phase4_output": globals().get("PHASE4_OUTPUT_MODEL_VERSION"),
    "phase5": globals().get("PHASE5_MODEL_VERSION"),
    "phase6": globals().get("PHASE6_MODEL_VERSION"),
    "phase7": globals().get("PHASE7_MODEL_VERSION"),
    "phase8": globals().get("PHASE8_VERSION"),
    "phase9": globals().get("PHASE9_VERSION"),
    "phase10": globals().get("PHASE10_VERSION"),
    "phase11": globals().get("PHASE11_VERSION"),
    "market_live_weight": _PHASE12_MARKET_WEIGHT_BEFORE,
    "elo_live_weight": _PHASE12_ELO_WEIGHT_BEFORE,
}
# Phase-3 model identity is taken from the actual immutable output rows.
_pg = _vx_p12_frame(globals().get("projected_goals_model_only"))
if not _pg.empty and "model_version" in _pg.columns:
    _phase12_phase_versions["phase3_output_versions"] = sorted(
        set(map(str, _pg["model_version"].dropna().unique()))
    )

# ----------------------------------------------------------------
# IMMUTABLE PRE-DEADLINE FORECAST PACKAGE
# ----------------------------------------------------------------
PHASE12_SNAPSHOT_CREATED = False
PHASE12_SNAPSHOT_PATH = None
PHASE12_SNAPSHOT_ID = None

def _vx_p12_write_frame(path, frame):
    fr = _vx_p12_frame(frame)
    _vx_p12_atomic_json(path, fr.to_dict("records"))
    return path


if PHASE12_BEFORE_DEADLINE:
    _stamp = PHASE12_NOW.strftime("%Y%m%dT%H%M%S%fZ")
    _base_id = f"{PHASE12_SEASON}_GW{PHASE12_CURRENT_GW:02d}_{_stamp}"
    _run_signature = _vx_p12_digest({
        "phase_versions": _phase12_phase_versions,
        "inputs": _phase12_input_manifest,
        "forecast_timestamp": PHASE12_TIMESTAMP,
    })[:10]
    PHASE12_SNAPSHOT_ID = f"{_base_id}_{_run_signature}"
    PHASE12_SNAPSHOT_PATH = PHASE12_GW_ROOT / PHASE12_SNAPSHOT_ID

    if PHASE12_SNAPSHOT_PATH.exists():
        raise RuntimeError(
            "Phase 12 immutable snapshot ID collision; refusing to overwrite an existing freeze."
        )
    PHASE12_SNAPSHOT_PATH.mkdir(parents=True, exist_ok=False)

    # Forecast tables
    _team_goals = _vx_p12_frame(globals().get("projected_goals_model_only"))
    _team_probs = _vx_p12_frame(globals().get("fixture_probabilities_model_only"))
    _team_fdr = _vx_p12_frame(globals().get("fixture_fdr_model_only"))
    _player_gw = _vx_p12_frame(globals().get("player_expected_points_gameweek"))
    _player_horizon = _vx_p12_frame(globals().get("player_expected_points_horizon"))
    _player_fixture = _vx_p12_frame(globals().get("player_expected_points_fixture"))

    if not _player_gw.empty and "gameweek" in _player_gw.columns:
        _player_gw_current = _player_gw.loc[
            pd.to_numeric(_player_gw["gameweek"], errors="coerce").eq(PHASE12_CURRENT_GW)
        ].copy()
    else:
        _player_gw_current = pd.DataFrame()

    _vx_p12_write_frame(PHASE12_SNAPSHOT_PATH / "team_projected_goals.json", _team_goals)
    _vx_p12_write_frame(PHASE12_SNAPSHOT_PATH / "team_probabilities.json", _team_probs)
    _vx_p12_write_frame(PHASE12_SNAPSHOT_PATH / "team_fdr.json", _team_fdr)
    _vx_p12_write_frame(PHASE12_SNAPSHOT_PATH / "player_gameweek_forecast.json", _player_gw_current)
    _vx_p12_write_frame(PHASE12_SNAPSHOT_PATH / "player_horizon_forecast.json", _player_horizon)
    _vx_p12_write_frame(PHASE12_SNAPSHOT_PATH / "player_fixture_forecast.json", _player_fixture)

    _vx_p12_atomic_json(
        PHASE12_SNAPSHOT_PATH / "decision_forecast.json",
        globals().get("PHASE11_RESEARCH_DECISIONS", {}),
    )
    _vx_p12_atomic_json(
        PHASE12_SNAPSHOT_PATH / "current_squad_decisions_publishable.json",
        globals().get("CURRENT_SQUAD_DECISIONS", {}),
    )
    _vx_p12_write_frame(
        PHASE12_SNAPSHOT_PATH / "external_benchmark.json",
        globals().get("PHASE8_EXTERNAL_BENCHMARK", globals().get("phase8_external_benchmark")),
    )

    _dynamic_dir = PHASE12_SNAPSHOT_PATH / "deadline_sensitive_inputs"
    _dynamic_dir.mkdir(parents=True, exist_ok=True)
    for _name in _PHASE12_DYNAMIC_COPY_NAMES:
        if _name in globals():
            _obj = globals()[_name]
            _fr = _vx_p12_frame(_obj)
            if not _fr.empty:
                _vx_p12_write_frame(_dynamic_dir / f"{_name}.json", _fr)
            else:
                _vx_p12_atomic_json(_dynamic_dir / f"{_name}.json", _obj)

    _manifest = {
        "schema_version": 12,
        "phase12_version": PHASE12_VERSION,
        "snapshot_id": PHASE12_SNAPSHOT_ID,
        "season": PHASE12_SEASON,
        "gameweek": PHASE12_CURRENT_GW,
        "forecast_timestamp": PHASE12_TIMESTAMP,
        "deadline_time": PHASE12_DEADLINE.isoformat() if PHASE12_DEADLINE is not None else None,
        "is_pre_deadline": True,
        "immutable_forecast_package": True,
        "phase_versions": _phase12_phase_versions,
        "input_manifest": _phase12_input_manifest,
        "source_file_manifest": _phase12_source_files,
        "production_flags": {
            "phase7_validation_evidence_ready": bool(globals().get("PHASE7_VALIDATION_EVIDENCE_READY", False)),
            "phase10_decision_release_ready": bool(globals().get("PHASE10_READY_FOR_DECISION_ENGINE", False)),
            "phase11_actionable": bool(globals().get("PHASE11_ACTIONABLE", False)),
            "phase11_output_mode": globals().get("PHASE11_OUTPUT_MODE"),
            "phase9_live_blend_active": bool(globals().get("PHASE9_LIVE_BLEND_ACTIVE", False)),
        },
        "contents": sorted(p.name for p in PHASE12_SNAPSHOT_PATH.iterdir()),
    }
    _manifest["package_sha256"] = _vx_p12_digest(_manifest)
    _vx_p12_atomic_json(PHASE12_SNAPSHOT_PATH / "manifest.json", _manifest)

    # Latest pointer may move BEFORE deadline. It never edits a prior snapshot.
    _vx_p12_atomic_json(
        PHASE12_GW_ROOT / "latest_pre_deadline.json",
        {
            "snapshot_id": PHASE12_SNAPSHOT_ID,
            "snapshot_path": str(PHASE12_SNAPSHOT_PATH),
            "forecast_timestamp": PHASE12_TIMESTAMP,
            "deadline_time": PHASE12_DEADLINE.isoformat() if PHASE12_DEADLINE is not None else None,
            "package_sha256": _manifest["package_sha256"],
        },
    )
    PHASE12_SNAPSHOT_CREATED = True

# ----------------------------------------------------------------
# ACTUAL OUTCOME INGEST
# Attach outcomes to the GW root, never to the immutable forecast directory.
# Only completed official FPL events are eligible.
# ----------------------------------------------------------------
def _vx_p12_event_finished(events_obj, gw):
    ev = _vx_p12_frame(events_obj)
    if ev.empty or "id" not in ev.columns:
        return False
    row = ev.loc[pd.to_numeric(ev["id"], errors="coerce").eq(int(gw))]
    if row.empty:
        return False
    if "vx_review_ready" not in row.columns:
        raise RuntimeError("Phase 12 requires canonical events.vx_review_ready from Cell 3.")
    return bool(row.iloc[0].get("vx_review_ready", False))


def _vx_p12_team_actuals(fixtures_obj, gw):
    fx = _vx_p12_frame(fixtures_obj)
    if fx.empty or "event" not in fx.columns:
        return pd.DataFrame()
    if "vx_completed" not in fx.columns:
        raise RuntimeError("Phase 12 actuals require canonical fixtures.vx_completed from Cell 3.")
    x = fx.loc[pd.to_numeric(fx["event"], errors="coerce").eq(int(gw))].copy()
    x = x.loc[x["vx_completed"].fillna(False).astype(bool)].copy()
    if x.empty:
        return pd.DataFrame()
    h_col = "team_h_score" if "team_h_score" in x.columns else None
    a_col = "team_a_score" if "team_a_score" in x.columns else None
    if h_col is None or a_col is None:
        return pd.DataFrame()
    x[h_col] = pd.to_numeric(x[h_col], errors="coerce")
    x[a_col] = pd.to_numeric(x[a_col], errors="coerce")
    x = x.dropna(subset=[h_col, a_col])
    if x.empty:
        return pd.DataFrame()
    out = pd.DataFrame({
        "fixture_id": pd.to_numeric(x.get("id"), errors="coerce").astype("Int64"),
        "gameweek": int(gw),
        "home_team_id": pd.to_numeric(x.get("team_h"), errors="coerce").astype("Int64"),
        "away_team_id": pd.to_numeric(x.get("team_a"), errors="coerce").astype("Int64"),
        "actual_home_goals": x[h_col].astype(int),
        "actual_away_goals": x[a_col].astype(int),
    })
    out["final_result"] = np.where(
        out["actual_home_goals"] > out["actual_away_goals"], "H",
        np.where(out["actual_home_goals"] < out["actual_away_goals"], "A", "D"),
    )
    out["actual_home_clean_sheet"] = out["actual_away_goals"].eq(0).astype(int)
    out["actual_away_clean_sheet"] = out["actual_home_goals"].eq(0).astype(int)
    return out


def _vx_p12_fetch_player_actuals(gw):
    try:
        from fpl_data import fetch_event_live
    except Exception as exc:
        raise RuntimeError(f"Cannot import canonical FPL event-live loader: {exc}") from exc

    cache_dir = PHASE12_LOG_DIR / "event_live_cache"
    cache_dir.mkdir(parents=True, exist_ok=True)
    payload = fetch_event_live(
        int(gw),
        cache_path=str(cache_dir / f"gw_{int(gw):02d}_live.json"),
    )
    rows = []
    for item in payload.get("elements", []):
        pid = int(item.get("id") or 0)
        if pid <= 0:
            continue
        st = dict(item.get("stats") or {})
        rows.append({
            "player_id": pid,
            "gameweek": int(gw),
            "actual_fpl_points": pd.to_numeric(st.get("total_points"), errors="coerce"),
            "actual_minutes": pd.to_numeric(st.get("minutes"), errors="coerce"),
            "actual_goals": pd.to_numeric(st.get("goals_scored"), errors="coerce"),
            "actual_assists": pd.to_numeric(st.get("assists"), errors="coerce"),
            "actual_clean_sheets": pd.to_numeric(st.get("clean_sheets"), errors="coerce"),
            "actual_goals_conceded": pd.to_numeric(st.get("goals_conceded"), errors="coerce"),
            "actual_saves": pd.to_numeric(st.get("saves"), errors="coerce"),
            "actual_bonus": pd.to_numeric(st.get("bonus"), errors="coerce"),
            "actual_defensive_contribution": pd.to_numeric(
                st.get("defensive_contribution"), errors="coerce"
            ),
            "actual_yellow_cards": pd.to_numeric(st.get("yellow_cards"), errors="coerce"),
            "actual_red_cards": pd.to_numeric(st.get("red_cards"), errors="coerce"),
        })
    return pd.DataFrame(rows)


def _vx_p12_poisson_deviance(y, mu):
    y = np.asarray(y, dtype=float)
    mu = np.clip(np.asarray(mu, dtype=float), 1e-12, None)
    term = np.where(y > 0, y * np.log(y / mu) - (y - mu), mu)
    return float(2.0 * np.nanmean(term))


def _vx_p12_validation_from_snapshot(snapshot_dir, team_actuals, player_actuals):
    snapshot_dir = Path(snapshot_dir)
    result = {
        "snapshot_id": snapshot_dir.name,
        "validated_at": PHASE12_TIMESTAMP,
        "team": {},
        "player": {},
        "decision": {},
        "error_review": {},
    }

    # ---------------- Team ----------------
    try:
        goals = pd.DataFrame(json.loads((snapshot_dir / "team_projected_goals.json").read_text(encoding="utf-8")))
        probs = pd.DataFrame(json.loads((snapshot_dir / "team_probabilities.json").read_text(encoding="utf-8")))
    except Exception:
        goals, probs = pd.DataFrame(), pd.DataFrame()

    ta = _vx_p12_frame(team_actuals)
    if not goals.empty and not ta.empty and "fixture_id" in goals.columns:
        g = goals.merge(ta, on="fixture_id", how="inner")
        if not g.empty and {"model_lambda_home", "model_lambda_away"}.issubset(g.columns):
            pred = np.r_[pd.to_numeric(g["model_lambda_home"], errors="coerce"),
                         pd.to_numeric(g["model_lambda_away"], errors="coerce")]
            actual = np.r_[pd.to_numeric(g["actual_home_goals"], errors="coerce"),
                           pd.to_numeric(g["actual_away_goals"], errors="coerce")]
            mask = np.isfinite(pred) & np.isfinite(actual)
            if mask.any():
                result["team"]["completed_fixtures"] = int(len(g))
                result["team"]["goal_mae"] = float(np.mean(np.abs(actual[mask] - pred[mask])))
                result["team"]["poisson_deviance"] = _vx_p12_poisson_deviance(actual[mask], pred[mask])

        if not probs.empty:
            p = probs.merge(ta, on="fixture_id", how="inner")
            if not p.empty:
                if {"home_clean_sheet_probability", "away_clean_sheet_probability"}.issubset(p.columns):
                    cs_pred = np.r_[
                        pd.to_numeric(p["home_clean_sheet_probability"], errors="coerce"),
                        pd.to_numeric(p["away_clean_sheet_probability"], errors="coerce"),
                    ]
                    cs_y = np.r_[
                        pd.to_numeric(p["actual_home_clean_sheet"], errors="coerce"),
                        pd.to_numeric(p["actual_away_clean_sheet"], errors="coerce"),
                    ]
                    m = np.isfinite(cs_pred) & np.isfinite(cs_y)
                    if m.any():
                        result["team"]["clean_sheet_brier"] = float(np.mean((cs_pred[m] - cs_y[m]) ** 2))
                req = {"home_win_probability", "draw_probability", "away_win_probability", "final_result"}
                if req.issubset(p.columns):
                    parr = p[["home_win_probability", "draw_probability", "away_win_probability"]].apply(
                        pd.to_numeric, errors="coerce"
                    ).to_numpy(float)
                    yy = np.zeros_like(parr)
                    for i, r in enumerate(p["final_result"].astype(str)):
                        yy[i, {"H": 0, "D": 1, "A": 2}.get(r, 1)] = 1.0
                    m = np.isfinite(parr).all(axis=1)
                    if m.any():
                        result["team"]["wdl_multiclass_brier"] = float(
                            np.mean(np.sum((parr[m] - yy[m]) ** 2, axis=1))
                        )

    # ---------------- Player ----------------
    try:
        pf = pd.DataFrame(json.loads((snapshot_dir / "player_gameweek_forecast.json").read_text(encoding="utf-8")))
    except Exception:
        pf = pd.DataFrame()
    pa = _vx_p12_frame(player_actuals)

    merged_player = pd.DataFrame()
    if not pf.empty and not pa.empty and "player_id" in pf.columns:
        merged_player = pf.merge(pa, on="player_id", how="inner")
        if not merged_player.empty:
            pred = pd.to_numeric(merged_player.get("gameweek_expected_points"), errors="coerce")
            act = pd.to_numeric(merged_player.get("actual_fpl_points"), errors="coerce")
            m = pred.notna() & act.notna()
            if m.any():
                err = act[m].astype(float) - pred[m].astype(float)
                result["player"].update({
                    "player_gameweeks": int(m.sum()),
                    "xpts_mae": float(np.mean(np.abs(err))),
                    "xpts_rmse": float(np.sqrt(np.mean(err ** 2))),
                    "xpts_bias_actual_minus_predicted": float(np.mean(err)),
                    "xpts_spearman": (
                        float(pred[m].corr(act[m], method="spearman"))
                        if int(m.sum()) >= 3 else None
                    ),
                })

                # Honest calibration table: predicted xPts bands vs observed average.
                bands = [-np.inf, 2, 4, 6, 8, 10, np.inf]
                labels = ["<2", "2-4", "4-6", "6-8", "8-10", "10+"]
                cal = pd.DataFrame({
                    "pred": pred[m].astype(float),
                    "actual": act[m].astype(float),
                })
                cal["band"] = pd.cut(cal["pred"], bins=bands, labels=labels, right=False)
                cal_rows = []
                for b, z in cal.groupby("band", observed=True):
                    cal_rows.append({
                        "xpts_band": str(b),
                        "n": int(len(z)),
                        "predicted_mean": float(z["pred"].mean()),
                        "actual_mean": float(z["actual"].mean()),
                        "bias_actual_minus_predicted": float((z["actual"] - z["pred"]).mean()),
                    })
                result["player"]["xpts_calibration_bands"] = cal_rows

                # Error review by position and confidence. These are diagnostics only.
                diag = []
                if "position" in merged_player.columns:
                    z = merged_player.loc[m].copy()
                    z["_pred"] = pred[m].astype(float).values
                    z["_actual"] = act[m].astype(float).values
                    for pos, q in z.groupby("position"):
                        diag.append({
                            "segment_type": "position",
                            "segment": str(pos),
                            "n": int(len(q)),
                            "mae": float(np.mean(np.abs(q["_actual"] - q["_pred"]))),
                            "bias_actual_minus_predicted": float(np.mean(q["_actual"] - q["_pred"])),
                        })
                if "player_model_confidence" in merged_player.columns:
                    z = merged_player.loc[m].copy()
                    z["_pred"] = pred[m].astype(float).values
                    z["_actual"] = act[m].astype(float).values
                    conf = pd.to_numeric(z["player_model_confidence"], errors="coerce")
                    if conf.notna().sum() >= 10:
                        try:
                            z["_conf_band"] = pd.qcut(conf, q=4, duplicates="drop")
                            for band, q in z.groupby("_conf_band", observed=True):
                                diag.append({
                                    "segment_type": "player_model_confidence",
                                    "segment": str(band),
                                    "n": int(len(q)),
                                    "mae": float(np.mean(np.abs(q["_actual"] - q["_pred"]))),
                                    "bias_actual_minus_predicted": float(np.mean(q["_actual"] - q["_pred"])),
                                })
                        except Exception:
                            pass
                result["error_review"]["player_segments"] = sorted(
                    diag,
                    key=lambda x: (-int(x["n"]), -abs(float(x["bias_actual_minus_predicted"]))),
                )

    # ---------------- Decision ----------------
    try:
        decision = json.loads((snapshot_dir / "decision_forecast.json").read_text(encoding="utf-8"))
    except Exception:
        decision = {}

    if decision and not pa.empty:
        amap = {
            int(r["player_id"]): float(r["actual_fpl_points"])
            for _, r in pa.dropna(subset=["player_id", "actual_fpl_points"]).iterrows()
        }
        cap = decision.get("best_captain")
        if isinstance(cap, dict) and cap.get("player_id") is not None:
            pid = int(cap["player_id"])
            if pid in amap:
                chosen = float(amap[pid])
                result["decision"]["captain_actual_points"] = chosen
                # Hindsight regret is evaluation only and never used as a live forecast input.
                rankings = decision.get("captain_rankings") or []
                candidates = [
                    int(x["player_id"]) for x in rankings
                    if isinstance(x, dict) and x.get("player_id") is not None and int(x["player_id"]) in amap
                ]
                if candidates:
                    best_actual = max(amap[x] for x in candidates)
                    result["decision"]["captain_hindsight_regret"] = float(best_actual - chosen)

        lineup = decision.get("recommended_lineup") or []
        lineup_ids = [
            int(x["player_id"]) for x in lineup
            if isinstance(x, dict) and x.get("player_id") is not None and int(x["player_id"]) in amap
        ]
        if lineup_ids:
            result["decision"]["starting_xi_actual_points"] = float(sum(amap[x] for x in lineup_ids))
            result["decision"]["starting_xi_players_scored"] = int(len(lineup_ids))

    return result, merged_player


PHASE12_REFRESHED_GAMEWEEKS = []
PHASE12_REFRESH_ERRORS = []
_phase12_events = _vx_p12_frame(globals().get("events"))
_phase12_fixtures = _vx_p12_frame(globals().get("fixtures"))

# Sweep every existing GW snapshot for newly-completed outcomes.
for _season_dir in sorted(PHASE12_SNAPSHOT_DIR.glob("*")):
    if not _season_dir.is_dir():
        continue
    for _gw_dir in sorted(_season_dir.glob("GW*")):
        if not _gw_dir.is_dir():
            continue
        m = re.fullmatch(r"GW(\d{1,2})", _gw_dir.name)
        if not m:
            continue
        _gw = int(m.group(1))
        _pointer = _gw_dir / "latest_pre_deadline.json"
        if not _pointer.exists():
            continue
        if not _vx_p12_event_finished(_phase12_events, _gw):
            continue

        try:
            _ptr = json.loads(_pointer.read_text(encoding="utf-8"))
            _snap = Path(_ptr["snapshot_path"])
            if not _snap.exists():
                raise RuntimeError(f"Frozen snapshot missing: {_snap}")

            _team_actual = _vx_p12_team_actuals(_phase12_fixtures, _gw)
            if _team_actual.empty:
                raise RuntimeError(f"Official team results are not complete for GW{_gw}.")

            _player_actual = _vx_p12_fetch_player_actuals(_gw)
            if _player_actual.empty:
                raise RuntimeError(f"Official player event-live results are empty for GW{_gw}.")

            _vx_p12_write_frame(_gw_dir / "actual_team_results.json", _team_actual)
            _vx_p12_write_frame(_gw_dir / "actual_player_results.json", _player_actual)

            _validation, _merged = _vx_p12_validation_from_snapshot(
                _snap, _team_actual, _player_actual
            )
            _vx_p12_atomic_json(_gw_dir / "validation_report.json", _validation)
            if not _merged.empty:
                _vx_p12_write_frame(_gw_dir / "player_forecast_vs_actual.json", _merged)

            PHASE12_REFRESHED_GAMEWEEKS.append(_gw)
        except Exception as exc:
            _err = {
                "timestamp": PHASE12_TIMESTAMP,
                "season_directory": _season_dir.name,
                "gameweek": _gw,
                "error_type": type(exc).__name__,
                "error": str(exc),
            }
            PHASE12_REFRESH_ERRORS.append(_err)
            _vx_p12_append_jsonl(PHASE12_ERROR_LOG, _err)

# ----------------------------------------------------------------
# AGGREGATED PLAYER VALIDATION HISTORY
# Never backfill a forecast; only consume genuinely frozen packages.
# ----------------------------------------------------------------
_phase12_validation_reports = []
for _vr in sorted(PHASE12_SNAPSHOT_DIR.glob("*/GW*/validation_report.json")):
    try:
        payload = json.loads(_vr.read_text(encoding="utf-8"))
        payload["_path"] = str(_vr)
        _phase12_validation_reports.append(payload)
    except Exception:
        pass

_player_metric_rows = []
_team_metric_rows = []
_decision_metric_rows = []
for _rep in _phase12_validation_reports:
    _p = dict(_rep.get("player") or {})
    _t = dict(_rep.get("team") or {})
    _d = dict(_rep.get("decision") or {})
    if _p.get("player_gameweeks"):
        _player_metric_rows.append({
            "snapshot_id": _rep.get("snapshot_id"),
            **{k: _p.get(k) for k in (
                "player_gameweeks", "xpts_mae", "xpts_rmse",
                "xpts_bias_actual_minus_predicted", "xpts_spearman",
            )},
        })
    if _t:
        _team_metric_rows.append({"snapshot_id": _rep.get("snapshot_id"), **_t})
    if _d:
        _decision_metric_rows.append({"snapshot_id": _rep.get("snapshot_id"), **_d})

phase12_player_validation_history = pd.DataFrame(_player_metric_rows)
phase12_team_validation_history = pd.DataFrame(_team_metric_rows)
phase12_decision_validation_history = pd.DataFrame(_decision_metric_rows)

PHASE12_PLAYER_HISTORY_PATH = PHASE12_MODEL_HISTORY_DIR / "phase12_player_validation_history.json"
PHASE12_TEAM_HISTORY_PATH = PHASE12_MODEL_HISTORY_DIR / "phase12_team_validation_history.json"
PHASE12_DECISION_HISTORY_PATH = PHASE12_MODEL_HISTORY_DIR / "phase12_decision_validation_history.json"
_vx_p12_write_frame(PHASE12_PLAYER_HISTORY_PATH, phase12_player_validation_history)
_vx_p12_write_frame(PHASE12_TEAM_HISTORY_PATH, phase12_team_validation_history)
_vx_p12_write_frame(PHASE12_DECISION_HISTORY_PATH, phase12_decision_validation_history)

# Aggregate all player forecast-vs-actual rows into a permanent error-review queue.
_phase12_player_compare_frames = []
for _cmp_path in sorted(PHASE12_SNAPSHOT_DIR.glob("*/GW*/player_forecast_vs_actual.json")):
    try:
        _z = pd.DataFrame(json.loads(_cmp_path.read_text(encoding="utf-8")))
        if not _z.empty:
            _z["_source_path"] = str(_cmp_path)
            _phase12_player_compare_frames.append(_z)
    except Exception:
        pass

phase12_error_review_queue = pd.DataFrame()
if _phase12_player_compare_frames:
    _allp = pd.concat(_phase12_player_compare_frames, ignore_index=True, sort=False)
    _pred = pd.to_numeric(_allp.get("gameweek_expected_points"), errors="coerce")
    _act = pd.to_numeric(_allp.get("actual_fpl_points"), errors="coerce")
    _allp["_error"] = _act - _pred
    _allp["_abs_error"] = (_act - _pred).abs()

    _review_rows = []
    if "position" in _allp.columns:
        for _seg, _q in _allp.dropna(subset=["_error"]).groupby("position"):
            _review_rows.append({
                "segment_type": "position",
                "segment": str(_seg),
                "n": int(len(_q)),
                "mae": float(_q["_abs_error"].mean()),
                "bias_actual_minus_predicted": float(_q["_error"].mean()),
            })

    if "player_id" in _allp.columns:
        for _pid, _q in _allp.dropna(subset=["_error"]).groupby("player_id"):
            if len(_q) < 3:
                continue
            _name = (
                str(_q["player_name"].dropna().iloc[0])
                if "player_name" in _q.columns and _q["player_name"].notna().any()
                else str(_pid)
            )
            _review_rows.append({
                "segment_type": "player",
                "segment": _name,
                "player_id": int(_pid),
                "n": int(len(_q)),
                "mae": float(_q["_abs_error"].mean()),
                "bias_actual_minus_predicted": float(_q["_error"].mean()),
            })

    if "expected_minutes" in _allp.columns and "actual_minutes" in _allp.columns:
        _em = pd.to_numeric(_allp["expected_minutes"], errors="coerce")
        _am = pd.to_numeric(_allp["actual_minutes"], errors="coerce")
        _m = _em.notna() & _am.notna()
        if _m.any():
            _review_rows.append({
                "segment_type": "minutes_model",
                "segment": "all_players",
                "n": int(_m.sum()),
                "mae": float(np.mean(np.abs(_am[_m] - _em[_m]))),
                "bias_actual_minus_predicted": float(np.mean(_am[_m] - _em[_m])),
            })

    phase12_error_review_queue = pd.DataFrame(_review_rows)
    if not phase12_error_review_queue.empty:
        phase12_error_review_queue["review_priority"] = (
            phase12_error_review_queue["bias_actual_minus_predicted"].abs()
            * np.sqrt(pd.to_numeric(phase12_error_review_queue["n"], errors="coerce").fillna(0.0))
        )
        phase12_error_review_queue = phase12_error_review_queue.sort_values(
            ["review_priority", "n"], ascending=[False, False]
        ).reset_index(drop=True)

PHASE12_ERROR_REVIEW_PATH = PHASE12_MODEL_HISTORY_DIR / "phase12_error_review_queue.json"
_vx_p12_write_frame(PHASE12_ERROR_REVIEW_PATH, phase12_error_review_queue)

# ----------------------------------------------------------------
# CONTROLLED CANDIDATE TEST / PROMOTION CONTRACT
# Candidate evidence must be produced by an out-of-sample historical test.
# Phase 12 never accepts self-reported "better" without metric comparison.
# ----------------------------------------------------------------
PHASE12_EVALUATION_FLOORS = {
    "fixture": 80,
    "player_gameweek": 500,
    "decision_gameweek": 6,
}

PHASE12_CANDIDATE_TEMPLATE = {
    "schema_version": 12,
    "candidate_id": "EXAMPLE_ONLY",
    "candidate_version": "FDR_VX.Y_CANDIDATE",
    "parent_version": "FDR_V1.0_MODEL_ONLY",
    "component": "TEAM_FORECAST | PLAYER_XPTS | FPL_DECISION",
    "evaluation_unit": "fixture | player_gameweek | decision_gameweek",
    "evaluation_mode": "OUT_OF_SAMPLE_HOLDOUT",
    "deadline_safe": True,
    "sample_size": 0,
    "primary_metrics": ["metric_name"],
    "guardrail_metrics": ["metric_name_2"],
    "metric_directions": {"metric_name": "lower", "metric_name_2": "lower"},
    "baseline_metrics": {"metric_name": None, "metric_name_2": None},
    "candidate_metrics": {"metric_name": None, "metric_name_2": None},
    "notes": "Candidate is promotable only if all primary metrics improve and all guardrails are non-worse.",
}
PHASE12_CANDIDATE_TEMPLATE_PATH = PHASE12_CANDIDATE_DIR / "candidate_evidence_TEMPLATE.json"
if not PHASE12_CANDIDATE_TEMPLATE_PATH.exists():
    _vx_p12_atomic_json(PHASE12_CANDIDATE_TEMPLATE_PATH, PHASE12_CANDIDATE_TEMPLATE)

PHASE12_RELEASE_APPROVAL_TEMPLATE_PAYLOAD = {
    "schema_version": 12,
    "approval_active": False,
    "candidate_id": "",
    "candidate_version": "",
    "approved_by": "",
    "approval_timestamp": "",
    "reason": "",
}
if not PHASE12_RELEASE_APPROVAL_TEMPLATE.exists():
    _vx_p12_atomic_json(
        PHASE12_RELEASE_APPROVAL_TEMPLATE,
        PHASE12_RELEASE_APPROVAL_TEMPLATE_PAYLOAD,
    )

if not PHASE12_RELEASE_APPROVAL_PATH.exists():
    _phase12_release_approval = dict(PHASE12_RELEASE_APPROVAL_TEMPLATE_PAYLOAD)
else:
    try:
        _phase12_release_approval = json.loads(
            PHASE12_RELEASE_APPROVAL_PATH.read_text(encoding="utf-8")
        )
    except Exception as exc:
        raise RuntimeError(f"Invalid Phase-12 release approval file: {exc}") from exc


def _vx_p12_assess_candidate(payload):
    errors = []
    unit = str(payload.get("evaluation_unit", "")).strip()
    floor = int(PHASE12_EVALUATION_FLOORS.get(unit, 80))
    sample_size = int(payload.get("sample_size") or 0)

    if str(payload.get("evaluation_mode", "")).upper() != "OUT_OF_SAMPLE_HOLDOUT":
        errors.append("evaluation_mode_must_be_OUT_OF_SAMPLE_HOLDOUT")
    if not bool(payload.get("deadline_safe", False)):
        errors.append("forecast_history_must_be_deadline_safe")
    if sample_size < floor:
        errors.append(f"sample_size_below_floor_{floor}")

    baseline = dict(payload.get("baseline_metrics") or {})
    candidate = dict(payload.get("candidate_metrics") or {})
    directions = dict(payload.get("metric_directions") or {})
    primary = list(payload.get("primary_metrics") or [])
    guardrails = list(payload.get("guardrail_metrics") or [])

    if not primary:
        errors.append("at_least_one_primary_metric_required")

    primary_results = {}
    for metric in primary:
        try:
            b = float(baseline.get(metric))
            c = float(candidate.get(metric))
            d = directions.get(metric)
            ok = _vx_p12_metric_direction_better(c, b, d)
        except Exception:
            ok = False
        primary_results[str(metric)] = bool(ok)
        if not ok:
            errors.append(f"primary_metric_not_improved:{metric}")

    guardrail_results = {}
    for metric in guardrails:
        try:
            b = float(baseline.get(metric))
            c = float(candidate.get(metric))
            d = directions.get(metric)
            ok = _vx_p12_metric_nonworse(c, b, d)
        except Exception:
            ok = False
        guardrail_results[str(metric)] = bool(ok)
        if not ok:
            errors.append(f"guardrail_regressed:{metric}")

    return {
        "candidate_id": payload.get("candidate_id"),
        "candidate_version": payload.get("candidate_version"),
        "parent_version": payload.get("parent_version"),
        "component": payload.get("component"),
        "evaluation_unit": unit,
        "sample_size": sample_size,
        "qualified": len(errors) == 0,
        "primary_metric_results": primary_results,
        "guardrail_metric_results": guardrail_results,
        "errors": errors,
    }


PHASE12_CANDIDATE_ASSESSMENTS = []
for _candidate_path in sorted(PHASE12_CANDIDATE_DIR.glob("*.json")):
    if _candidate_path.name.endswith("_TEMPLATE.json"):
        continue
    try:
        _payload = json.loads(_candidate_path.read_text(encoding="utf-8"))
        _assessment = _vx_p12_assess_candidate(_payload)
        _assessment["evidence_path"] = str(_candidate_path)
        PHASE12_CANDIDATE_ASSESSMENTS.append(_assessment)
    except Exception as exc:
        PHASE12_CANDIDATE_ASSESSMENTS.append({
            "candidate_id": _candidate_path.stem,
            "qualified": False,
            "errors": [f"invalid_candidate_file:{type(exc).__name__}:{exc}"],
            "evidence_path": str(_candidate_path),
        })

# Phase-9 qualification is surfaced as evidence for review but remains unpromoted.
for _q in list(globals().get("PHASE9_QUALIFIED_CANDIDATES", []) or []):
    PHASE12_CANDIDATE_ASSESSMENTS.append({
        "candidate_id": f"PHASE9::{_vx_p12_safe_slug(str(_q))}",
        "candidate_version": "FDR_V1.2_MARKET_CALIBRATED",
        "parent_version": "FDR_V1.0_MODEL_ONLY",
        "component": "TEAM_FORECAST_CALIBRATION",
        "qualified": True,
        "source": "PHASE9_QUALIFIED_CANDIDATES",
        "note": "Phase-9 holdout qualification only; Phase 12 still requires explicit release approval.",
    })

# ----------------------------------------------------------------
# VERSION REGISTRY + PROMOTION
# Initial baseline is the live model-only forecast. A later release is append-only.
# ----------------------------------------------------------------
if PHASE12_VERSION_REGISTRY_PATH.exists():
    try:
        PHASE12_VERSION_REGISTRY = json.loads(
            PHASE12_VERSION_REGISTRY_PATH.read_text(encoding="utf-8")
        )
    except Exception as exc:
        raise RuntimeError(f"Invalid Phase-12 version registry: {exc}") from exc
else:
    PHASE12_VERSION_REGISTRY = {
        "schema_version": 12,
        "active_forecast_version": "FDR_V1.0_MODEL_ONLY",
        "active_player_version": None,
        "active_decision_version": None,
        "validation_status": (
            "EVIDENCE_READY_FOR_REVIEW"
            if bool(globals().get("PHASE7_VALIDATION_EVIDENCE_READY", False))
            else "RESEARCH_BASELINE_ACCUMULATING_HISTORY"
        ),
        "releases": [{
            "version": "FDR_V1.0_MODEL_ONLY",
            "registered_at": PHASE12_TIMESTAMP,
            "status": "BASELINE_REGISTERED",
            "reason": "Model-only team forecast is the immutable baseline for all comparisons.",
        }],
    }

PHASE12_PROMOTION_APPLIED = False
PHASE12_PROMOTION_STATUS = "NO_RELEASE_APPROVAL"

_approval_active = bool(_phase12_release_approval.get("approval_active", False))
if _approval_active:
    _approval_id = str(_phase12_release_approval.get("candidate_id", "")).strip()
    _approval_version = str(_phase12_release_approval.get("candidate_version", "")).strip()
    _matches = [
        x for x in PHASE12_CANDIDATE_ASSESSMENTS
        if bool(x.get("qualified", False))
        and str(x.get("candidate_id", "")).strip() == _approval_id
        and str(x.get("candidate_version", "")).strip() == _approval_version
    ]
    if not _matches:
        PHASE12_PROMOTION_STATUS = "APPROVAL_REJECTED_NO_QUALIFIED_MATCHING_EVIDENCE"
    else:
        _assess = _matches[0]
        _component = str(_assess.get("component", "")).upper()

        # Extra live gates for player/decision releases.
        if "PLAYER" in _component and not bool(globals().get("PHASE10_READY_FOR_DECISION_ENGINE", False)):
            PHASE12_PROMOTION_STATUS = "APPROVAL_REJECTED_PHASE10_RELEASE_GATE_CLOSED"
        elif "DECISION" in _component and not bool(globals().get("PHASE11_ACTIONABLE", False)):
            PHASE12_PROMOTION_STATUS = "APPROVAL_REJECTED_PHASE11_ACTIONABLE_GATE_CLOSED"
        else:
            release_record = {
                "version": _approval_version,
                "candidate_id": _approval_id,
                "released_at": PHASE12_TIMESTAMP,
                "approved_by": _phase12_release_approval.get("approved_by"),
                "approval_timestamp": _phase12_release_approval.get("approval_timestamp"),
                "reason": _phase12_release_approval.get("reason"),
                "evidence": _assess,
            }
            existing = {
                (str(x.get("version")), str(x.get("candidate_id", "")))
                for x in PHASE12_VERSION_REGISTRY.get("releases", [])
            }
            if (_approval_version, _approval_id) not in existing:
                PHASE12_VERSION_REGISTRY.setdefault("releases", []).append(release_record)

            # Approval is not permission to silently change live mathematics.
            # A version becomes active only when the corresponding implementation
            # is already active in the runtime and all of its release gates are open.
            _activated = False
            if _approval_version == "FDR_V1.2_MARKET_CALIBRATED":
                if bool(globals().get("PHASE9_LIVE_BLEND_ACTIVE", False)):
                    PHASE12_VERSION_REGISTRY["active_forecast_version"] = _approval_version
                    _activated = True
            elif _approval_version == "FDR_V2.0_PLAYER_XPTS":
                PHASE12_VERSION_REGISTRY["active_player_version"] = _approval_version
                _activated = True
            elif _approval_version == "FDR_V3.0_FPL_DECISION_ENGINE":
                PHASE12_VERSION_REGISTRY["active_decision_version"] = _approval_version
                _activated = True
            else:
                # Future versions are registered as approved evidence. They become
                # active only after their code is implemented in a later controlled edit.
                _activated = False

            release_record["activation_state"] = (
                "ACTIVE_IN_RUNTIME" if _activated else "APPROVED_AWAITING_IMPLEMENTATION"
            )
            PHASE12_PROMOTION_APPLIED = True
            PHASE12_PROMOTION_STATUS = (
                "QUALIFIED_APPROVED_AND_ACTIVE"
                if _activated
                else "QUALIFIED_APPROVED_AWAITING_IMPLEMENTATION"
            )
            _vx_p12_append_jsonl(PHASE12_PROMOTION_LOG, release_record)

_vx_p12_atomic_json(PHASE12_VERSION_CATALOG_PATH, PHASE12_VERSION_CATALOG)
_vx_p12_atomic_json(PHASE12_VERSION_REGISTRY_PATH, PHASE12_VERSION_REGISTRY)

# ----------------------------------------------------------------
# WEEKLY GOVERNANCE SUMMARY
# ----------------------------------------------------------------
PHASE12_WEEKLY_SUMMARY = {
    "schema_version": 12,
    "phase12_version": PHASE12_VERSION,
    "run_timestamp": PHASE12_TIMESTAMP,
    "season": PHASE12_SEASON,
    "gameweek": PHASE12_CURRENT_GW,
    "deadline_time": PHASE12_DEADLINE.isoformat() if PHASE12_DEADLINE is not None else None,
    "before_deadline": PHASE12_BEFORE_DEADLINE,
    "snapshot_created": PHASE12_SNAPSHOT_CREATED,
    "snapshot_id": PHASE12_SNAPSHOT_ID,
    "snapshot_path": str(PHASE12_SNAPSHOT_PATH) if PHASE12_SNAPSHOT_PATH else None,
    "refreshed_completed_gameweeks": sorted(set(PHASE12_REFRESHED_GAMEWEEKS)),
    "refresh_errors": PHASE12_REFRESH_ERRORS,
    "phase_versions": _phase12_phase_versions,
    "input_manifest": _phase12_input_manifest,
    "source_file_manifest": _phase12_source_files,
    "candidate_assessments": PHASE12_CANDIDATE_ASSESSMENTS,
    "promotion_applied": PHASE12_PROMOTION_APPLIED,
    "promotion_status": PHASE12_PROMOTION_STATUS,
    "version_registry": PHASE12_VERSION_REGISTRY,
    "workflow": [
        "Freeze forecast before FPL deadline",
        "Save model version and all consumed deadline-sensitive inputs",
        "Record actual match results",
        "Record actual player FPL points",
        "Compare forecast with reality",
        "Find systematic model errors",
        "Test candidate improvements on deadline-safe out-of-sample history",
        "Release a new version only after objective improvement AND explicit approval",
    ],
}
PHASE12_LATEST_SUMMARY_PATH = PHASE12_MODEL_HISTORY_DIR / "phase12_latest_weekly_summary.json"
_vx_p12_atomic_json(PHASE12_LATEST_SUMMARY_PATH, PHASE12_WEEKLY_SUMMARY)
_vx_p12_append_jsonl(PHASE12_RUN_LOG, PHASE12_WEEKLY_SUMMARY)

# ----------------------------------------------------------------
# FINAL IMMUTABILITY / SAFETY GATE
# ----------------------------------------------------------------
_PHASE12_STATE_AFTER = {
    k: _vx_p12_digest(globals()[k])
    for k in _PHASE12_IMMUTABLE_KEYS
    if k in globals()
}
if _PHASE12_STATE_BEFORE != _PHASE12_STATE_AFTER:
    raise RuntimeError(
        "Phase 12 integrity failure: continuous-improvement governance mutated a live forecast/decision object."
    )
if float(globals().get("MARKET_BLEND_WEIGHT", 0.0)) != _PHASE12_MARKET_WEIGHT_BEFORE:
    raise RuntimeError("Phase 12 integrity failure: market live blend weight changed.")
if float(globals().get("ELO_BLEND_WEIGHT", 0.0)) != _PHASE12_ELO_WEIGHT_BEFORE:
    raise RuntimeError("Phase 12 integrity failure: Team Elo live blend weight changed.")

PHASE12_READY = True
PHASE12_CONTINUOUS_IMPROVEMENT_ACTIVE = True
PHASE12_PRODUCTION_FORECAST_VERSION = PHASE12_VERSION_REGISTRY.get(
    "active_forecast_version", "FDR_V1.0_MODEL_ONLY"
)
PHASE12_PRODUCTION_PLAYER_VERSION = PHASE12_VERSION_REGISTRY.get("active_player_version")
PHASE12_PRODUCTION_DECISION_VERSION = PHASE12_VERSION_REGISTRY.get("active_decision_version")

print("✅ PHASE 12 — CONTINUOUS IMPROVEMENT GOVERNANCE BUILT")
print(f"   Season / GW: {PHASE12_SEASON} / {PHASE12_CURRENT_GW}")
print(f"   Before deadline: {PHASE12_BEFORE_DEADLINE}")
print(f"   Frozen snapshot created: {PHASE12_SNAPSHOT_CREATED}")
print(f"   Completed GWs refreshed: {sorted(set(PHASE12_REFRESHED_GAMEWEEKS))}")
print(f"   Candidate changes reviewed: {len(PHASE12_CANDIDATE_ASSESSMENTS)}")
print(f"   Promotion status: {PHASE12_PROMOTION_STATUS}")
print(f"   Active forecast lineage: {PHASE12_PRODUCTION_FORECAST_VERSION}")
print(f"   Active player lineage: {PHASE12_PRODUCTION_PLAYER_VERSION or 'NOT RELEASED'}")
print(f"   Active decision lineage: {PHASE12_PRODUCTION_DECISION_VERSION or 'NOT RELEASED'}")
print("🔒 No live model object was changed by Phase 12.")
print("📦 Governance history:", PHASE12_MODEL_HISTORY_DIR)
print("📸 Immutable snapshots:", PHASE12_SNAPSHOT_DIR)
print("🧾 Weekly logs:", PHASE12_LOG_DIR)
```

```python
# ============================================================
# CELL 14 — GW REVIEW • VIDEO OPENING SLIDE
# ============================================================

from pathlib import Path
from IPython.display import display, Image as IPImage
import json

# ============================================================
# CELL 14 — SELF-CONTAINED TRUE-4K HTML → PNG RENDERER
# ============================================================
# Rendering infrastructure only. No FDR/xG/clean-sheet calculation lives here.
import subprocess as _vx_subprocess, sys as _vx_sys
from PIL import Image as _VxPILImage

try:
    from playwright.async_api import async_playwright
except ImportError:
    _vx_subprocess.run([_vx_sys.executable, "-m", "pip", "install", "-q", "playwright"], check=True)
    from playwright.async_api import async_playwright

# Colab can retain Playwright while losing either its matching Chromium build
# or required Linux shared libraries. Provision BOTH browser and OS dependencies
# through Playwright's supported one-command path. This remains the sole renderer
# setup, so later slide cells never duplicate apt/browser installation code.
_VX_PLAYWRIGHT_CHROMIUM_READY = False

def _vx_ensure_playwright_chromium(force=False):
    global _VX_PLAYWRIGHT_CHROMIUM_READY
    if _VX_PLAYWRIGHT_CHROMIUM_READY and not force:
        return
    result = _vx_subprocess.run(
        [
            _vx_sys.executable,
            "-m",
            "playwright",
            "install",
            "--with-deps",
            "chromium",
        ],
        check=False,
        capture_output=True,
        text=True,
    )
    if result.returncode != 0:
        detail = (result.stderr or result.stdout or "unknown installer error").strip()
        raise RuntimeError(
            "Playwright Chromium + Linux dependency installation failed "
            "in the shared 4K renderer: " + detail[-3000:]
        )
    _VX_PLAYWRIGHT_CHROMIUM_READY = True
    print("✅ Playwright Chromium and Linux dependencies ready")

async def export_4k_png(html_path, png_path):
    html_path = Path(html_path)
    png_path = Path(png_path)
    png_path.parent.mkdir(parents=True, exist_ok=True)
    _vx_ensure_playwright_chromium()
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True, args=["--no-sandbox", "--disable-dev-shm-usage"])
        page = await browser.new_page(viewport={"width":3840, "height":2160})
        await page.goto(html_path.as_uri(), wait_until="networkidle")
        await page.screenshot(path=str(png_path), full_page=False)
        await browser.close()
    with _VxPILImage.open(png_path) as img:
        if img.size != (3840,2160):
            raise RuntimeError(f"Expected true 4K 3840x2160, got {img.size}")
    print("✅ TRUE 4K:", png_path)


# ============================================================
# PRESENTATION COMPATIBILITY BRIDGE — NEW MATH, ORIGINAL CONTRACT
# ============================================================
# This block aggregates the already-calculated Phase 3/4/10/11 outputs into
# the table names consumed by the proven weekly presentation. It does not fit,
# blend, recalibrate, or replace any forecast.
import math as _vx_math
import re as _vx_re
import numpy as np
import pandas as pd

for _vx_required_model in (
    "fixtures",
    "NEXT_GW",
    "team_meta_by_id",
    "fixture_goal_expectations_model",
    "fixture_fdr_model_only",
):
    if _vx_required_model not in globals():
        raise RuntimeError(
            f"Run the improved model cells before Cell 14 — missing {_vx_required_model}"
        )


def format_fdr_value(value):
    if value is None:
        return "—"
    if isinstance(value, (list, tuple)):
        return "—" if not value else " + ".join(str(x) for x in value)
    text = str(value).strip()
    return "—" if text.lower() in {"", "nan", "none", "nat"} else text


def _vx_numeric_fdr_values(value):
    raw = value if isinstance(value, (list, tuple)) else _vx_re.findall(
        r"(?<![\d.])(?:[1-4](?:\.\d+)?|5(?:\.0+)?)(?![\d.])", str(value)
    )
    values = []
    for item in raw:
        try:
            number = float(item)
        except (TypeError, ValueError):
            continue
        if 1.0 <= number <= 5.0:
            values.append(number)
    return values


def _vx_fdr_bucket(value):
    values = _vx_numeric_fdr_values(value)
    if not values:
        return None
    return max(1, min(5, int(_vx_math.floor(sum(values) / len(values) + 0.5))))


def fdr_class(value):
    bucket = _vx_fdr_bucket(value)
    return "fdr-none" if bucket is None else f"fdr-{bucket}"


_vx_fixtures = fixtures.copy() if isinstance(fixtures, pd.DataFrame) else pd.DataFrame(fixtures)
_vx_team_short = {
    int(team_id): str(meta.get("short") or meta.get("name") or team_id).upper()
    for team_id, meta in team_meta_by_id.items()
}
_vx_display_gws = list(range(int(NEXT_GW), int(NEXT_GW) + 6))

# ------------------------------------------------------------
# CANONICAL MODEL FDR PRESENTATION CONTRACT
# ------------------------------------------------------------
# fixture_fdr_model_only (Phase 4B) is the ONLY FDR math source.
# No official FPL team_h/team_a difficulty fields are read here.
# The compatibility table keeps attack and defence FDR separately; the legacy
# single `fdr` presentation field is an alias of attack FDR only (not a blend).
_vx_model_fdr_rows = []
for _, _vx_fdr in fixture_fdr_model_only.iterrows():
    for _vx_side in ("home", "away"):
        _vx_team_id = int(_vx_fdr[f"{_vx_side}_team_id"])
        _vx_opp_side = "away" if _vx_side == "home" else "home"
        _vx_opponent_id = int(_vx_fdr[f"{_vx_opp_side}_team_id"])
        _vx_attack = float(_vx_fdr[f"attack_fdr_{_vx_side}"])
        _vx_defence = float(_vx_fdr[f"defence_fdr_{_vx_side}"])
        if not (1.0 <= _vx_attack <= 5.0 and 1.0 <= _vx_defence <= 5.0):
            raise RuntimeError(
                f"Phase-4 model FDR outside 1..5 for fixture {int(_vx_fdr['fixture_id'])}"
            )
        _vx_model_fdr_rows.append({
            "fixture_id": int(_vx_fdr["fixture_id"]),
            "gw": int(_vx_fdr["gameweek"]),
            "team_id": _vx_team_id,
            "team": str(team_meta_by_id[_vx_team_id]["short"]).upper(),
            "opponent_id": _vx_opponent_id,
            "opponent": str(team_meta_by_id[_vx_opponent_id]["short"]).upper(),
            "venue": "H" if _vx_side == "home" else "A",
            "attack_fdr_model": _vx_attack,
            "defence_fdr_model": _vx_defence,
            "attack_fdr_colour_band": str(
                _vx_fdr.get(f"attack_fdr_{_vx_side}_colour_band", "")
            ),
            "defence_fdr_colour_band": str(
                _vx_fdr.get(f"defence_fdr_{_vx_side}_colour_band", "")
            ),
        })

model_fixture_fdr_df = pd.DataFrame(_vx_model_fdr_rows).sort_values(
    ["gw", "fixture_id", "team_id"], kind="stable"
).reset_index(drop=True)
if model_fixture_fdr_df.empty:
    raise RuntimeError("Phase-4 model FDR produced no presentation rows")

_vx_fdr_rows = []
for _vx_team_id, _vx_meta in sorted(team_meta_by_id.items()):
    _vx_fixture_slots = []
    _vx_attack_slots = []
    _vx_defence_slots = []
    _vx_exact_attack = []
    _vx_exact_defence = []

    for _vx_gw in _vx_display_gws:
        _vx_gw_model = model_fixture_fdr_df.loc[
            pd.to_numeric(model_fixture_fdr_df["gw"], errors="coerce").eq(_vx_gw)
            & pd.to_numeric(model_fixture_fdr_df["team_id"], errors="coerce").eq(int(_vx_team_id))
        ].sort_values("fixture_id", kind="stable")

        _vx_labels = []
        _vx_attack_values = []
        _vx_defence_values = []
        for _, _vx_model_row in _vx_gw_model.iterrows():
            _vx_labels.append(
                f"{str(_vx_model_row['opponent']).upper()} ({str(_vx_model_row['venue']).upper()})"
            )
            _vx_attack = float(_vx_model_row["attack_fdr_model"])
            _vx_defence = float(_vx_model_row["defence_fdr_model"])
            _vx_attack_values.append(_vx_attack)
            _vx_defence_values.append(_vx_defence)
            _vx_exact_attack.append(_vx_attack)
            _vx_exact_defence.append(_vx_defence)

        _vx_fixture_slots.append(" / ".join(_vx_labels) if _vx_labels else "BLANK")
        _vx_attack_slots.append(
            " + ".join(f"{value:.2f}" for value in _vx_attack_values)
            if _vx_attack_values else "—"
        )
        _vx_defence_slots.append(
            " + ".join(f"{value:.2f}" for value in _vx_defence_values)
            if _vx_defence_values else "—"
        )

    _vx_fdr_rows.append({
        "team_id": int(_vx_team_id),
        "team": str(_vx_meta.get("name") or _vx_team_id),
        "team_short": str(_vx_meta.get("short") or _vx_team_id).upper(),
        "fixtures": _vx_fixture_slots,
        "fdr": _vx_attack_slots,
        "attack_fdr": _vx_attack_slots,
        "defence_fdr": _vx_defence_slots,
        "average_fdr": float(np.mean(_vx_exact_attack)) if _vx_exact_attack else np.nan,
        "average_attack_fdr": float(np.mean(_vx_exact_attack)) if _vx_exact_attack else np.nan,
        "average_defence_fdr": float(np.mean(_vx_exact_defence)) if _vx_exact_defence else np.nan,
        "fixture_ease": float(np.mean([6.0 - x for x in _vx_exact_attack])) if _vx_exact_attack else np.nan,
        "fdr_source": "PHASE4_MODEL_ONLY",
    })

fdr_table = pd.DataFrame(_vx_fdr_rows).sort_values(
    ["average_fdr", "team_short"], ascending=[True, True], na_position="last", kind="stable"
).reset_index(drop=True)

# Projected Goals presentation uses the exact same Phase-4 fixture snapshot as FDR.
# Phase 4 carries the Phase-3 model lambdas unchanged; this audit blocks any drift.
_vx_phase3_goal_audit = fixture_goal_expectations_model[
    ["fixture_id", "model_lambda_home", "model_lambda_away"]
].copy()
_vx_phase4_goal_audit = fixture_fdr_model_only[
    ["fixture_id", "model_lambda_home", "model_lambda_away"]
].copy()
_vx_goal_audit = _vx_phase4_goal_audit.merge(
    _vx_phase3_goal_audit,
    on="fixture_id",
    how="outer",
    suffixes=("_phase4", "_phase3"),
    validate="one_to_one",
    indicator=True,
)
if not _vx_goal_audit["_merge"].eq("both").all():
    raise RuntimeError("Projected Goals / FDR fixture set mismatch between Phase 3 and Phase 4")
for _vx_side in ("home", "away"):
    _vx_delta = (
        pd.to_numeric(_vx_goal_audit[f"model_lambda_{_vx_side}_phase4"], errors="coerce")
        - pd.to_numeric(_vx_goal_audit[f"model_lambda_{_vx_side}_phase3"], errors="coerce")
    ).abs()
    if _vx_delta.isna().any() or (_vx_delta > 1e-12).any():
        raise RuntimeError(
            f"Projected Goals / FDR lambda drift detected for {_vx_side} fixtures"
        )

_vx_goal_frame = fixture_fdr_model_only.copy()
# Fix 11: Clean Sheet presentation reads the SAME Phase-4B fixture snapshot as FDR.
# No separate clean-sheet presentation table or second probability calculation.
_vx_probability_frame = fixture_fdr_model_only.copy()
_vx_probability_by_fixture = {
    int(row["fixture_id"]): row for _, row in _vx_probability_frame.iterrows()
}
_vx_fixture_projection_rows = []
for _, _vx_goal in _vx_goal_frame.iterrows():
    _vx_fixture_id = int(_vx_goal["fixture_id"])
    _vx_probability = _vx_probability_by_fixture.get(_vx_fixture_id)
    if _vx_probability is None:
        raise RuntimeError(f"Clean-sheet probability missing for fixture {_vx_fixture_id}")
    _vx_home_id = int(_vx_goal["home_team_id"])
    _vx_away_id = int(_vx_goal["away_team_id"])
    _vx_fixture_projection_rows.append({
        "fixture_id": _vx_fixture_id,
        "gw": int(_vx_goal["gameweek"]),
        "home_id": _vx_home_id,
        "away_id": _vx_away_id,
        "home": str(team_meta_by_id[_vx_home_id]["name"]),
        "away": str(team_meta_by_id[_vx_away_id]["name"]),
        "home_xg": float(_vx_goal["model_lambda_home"]),
        "away_xg": float(_vx_goal["model_lambda_away"]),
        "home_cs": 100.0 * float(_vx_probability["home_clean_sheet_probability"]),
        "away_cs": 100.0 * float(_vx_probability["away_clean_sheet_probability"]),
        "model_confidence": float(_vx_goal.get("model_confidence", np.nan)),
        "source": str(_vx_goal.get("source_name", "VORTEX MODEL")),
    })
fixture_projection_df = pd.DataFrame(_vx_fixture_projection_rows)

_vx_projection_rows = []
for _vx_team_id, _vx_meta in sorted(team_meta_by_id.items()):
    _vx_row = {
        "team_id": int(_vx_team_id),
        "team": str(_vx_meta.get("name") or _vx_team_id),
        "team_short": str(_vx_meta.get("short") or _vx_team_id).upper(),
        "badge": str(_vx_meta.get("badge") or ""),
    }
    _vx_xg_values = []
    _vx_cs_values = []
    for _vx_slot, _vx_gw in enumerate(_vx_display_gws, start=1):
        _vx_matches = fixture_projection_df.loc[
            fixture_projection_df["gw"].eq(_vx_gw)
            & (
                fixture_projection_df["home_id"].eq(int(_vx_team_id))
                | fixture_projection_df["away_id"].eq(int(_vx_team_id))
            )
        ]
        _vx_match_xg = []
        _vx_match_cs = []
        _vx_opponents = []
        _vx_opponent_ids = []
        for _, _vx_match in _vx_matches.iterrows():
            _vx_is_home = int(_vx_match["home_id"]) == int(_vx_team_id)
            _vx_opponent_id = int(_vx_match["away_id"] if _vx_is_home else _vx_match["home_id"])
            _vx_match_xg.append(float(_vx_match["home_xg"] if _vx_is_home else _vx_match["away_xg"]))
            _vx_match_cs.append(float(_vx_match["home_cs"] if _vx_is_home else _vx_match["away_cs"]) / 100.0)
            _vx_opponent_ids.append(_vx_opponent_id)
            _vx_opponents.append(
                f"{_vx_team_short.get(_vx_opponent_id, str(_vx_opponent_id))} "
                f"({'H' if _vx_is_home else 'A'})"
            )

        _vx_xg = float(sum(_vx_match_xg)) if _vx_match_xg else 0.0
        _vx_no_cs = float(np.prod([1.0 - p for p in _vx_match_cs])) if _vx_match_cs else 1.0
        _vx_cs = 100.0 * (1.0 - _vx_no_cs) if _vx_match_cs else 0.0
        _vx_row[f"gw{_vx_slot}_xg"] = _vx_xg
        _vx_row[f"gw{_vx_slot}_cs"] = _vx_cs
        _vx_row[f"gw{_vx_slot}_opponent"] = " / ".join(_vx_opponents) if _vx_opponents else "BLANK"
        _vx_row[f"gw{_vx_slot}_opponent_ids"] = _vx_opponent_ids
        _vx_xg_values.append(_vx_xg)
        _vx_cs_values.append(_vx_cs)

    _vx_row["avg_xg"] = float(np.mean(_vx_xg_values))
    _vx_row["avg_cs"] = float(np.mean(_vx_cs_values))
    _vx_projection_rows.append(_vx_row)
team_projection_df = pd.DataFrame(_vx_projection_rows)

goalkeeper_fixture_fdr_df = model_fixture_fdr_df[
    ["fixture_id", "gw", "team_id", "team", "defence_fdr_model"]
].rename(columns={"defence_fdr_model": "goalkeeper_fdr_model"})

_vx_player_fixture = globals().get("phase10_player_fixture_expected_points", pd.DataFrame())
player_fixture_xpts_df = _vx_player_fixture.copy() if isinstance(_vx_player_fixture, pd.DataFrame) else pd.DataFrame()
if not player_fixture_xpts_df.empty:
    player_fixture_xpts_df = player_fixture_xpts_df.rename(
        columns={"gameweek": "gw", "fixture_expected_points": "xpts"}
    )

_vx_player_horizon = globals().get("phase10_player_horizon_expected_points", pd.DataFrame())
player_horizon_xpts_df = _vx_player_horizon.copy() if isinstance(_vx_player_horizon, pd.DataFrame) else pd.DataFrame()
if not player_horizon_xpts_df.empty:
    player_horizon_xpts_df["next_gw_xpts"] = player_horizon_xpts_df["final_gameweek_xpts"]
    player_horizon_xpts_df["xpts_3gw"] = player_horizon_xpts_df["next_3_xpts"]
    player_horizon_xpts_df["xpts_5gw"] = player_horizon_xpts_df["next_5_xpts"]

    _vx_player_gameweek = globals().get("phase10_player_gameweek_expected_points", pd.DataFrame())
    if isinstance(_vx_player_gameweek, pd.DataFrame) and not _vx_player_gameweek.empty:
        _vx_next_eight = _vx_player_gameweek.loc[
            pd.to_numeric(_vx_player_gameweek["gameweek"], errors="coerce").between(
                int(NEXT_GW), int(NEXT_GW) + 7
            )
        ].groupby("player_id", as_index=False)["gameweek_expected_points"].sum(min_count=1)
        _vx_next_eight = _vx_next_eight.rename(columns={"gameweek_expected_points": "xpts_8gw"})
        player_horizon_xpts_df = player_horizon_xpts_df.merge(
            _vx_next_eight, on="player_id", how="left", validate="one_to_one"
        )
    else:
        player_horizon_xpts_df["xpts_8gw"] = np.nan

    if isinstance(_vx_player_fixture, pd.DataFrame) and not _vx_player_fixture.empty:
        _vx_next = _vx_player_fixture.loc[
            pd.to_numeric(_vx_player_fixture["gameweek"], errors="coerce").eq(int(NEXT_GW))
        ].copy()
        _vx_minutes_rows = []
        for _vx_player_id, _vx_group in _vx_next.groupby("player_id", sort=False):
            def _vx_at_least_one(column):
                probabilities = pd.to_numeric(_vx_group[column], errors="coerce").dropna().clip(0, 1)
                return float(1.0 - np.prod(1.0 - probabilities)) if len(probabilities) else np.nan

            _vx_minutes_rows.append({
                "player_id": int(_vx_player_id),
                "p_start_next_gw": _vx_at_least_one("start_probability"),
                "p_appear_next_gw": _vx_at_least_one("appearance_probability"),
                "p_60_next_gw": _vx_at_least_one("sixty_minute_probability"),
                "expected_minutes_next_gw": float(
                    pd.to_numeric(_vx_group["expected_minutes"], errors="coerce").sum(min_count=1)
                ),
            })
        player_horizon_xpts_df = player_horizon_xpts_df.merge(
            pd.DataFrame(_vx_minutes_rows), on="player_id", how="left", validate="one_to_one"
        )


def _vx_phase11_rows(name):
    frame = globals().get(name, pd.DataFrame())
    return frame.to_dict(orient="records") if isinstance(frame, pd.DataFrame) else []


_vx_transfer_rows = _vx_phase11_rows("phase11_transfer_shortlist")
_vx_transfer_rows = [row for row in _vx_transfer_rows if bool(row.get("actionable", False))]
for _vx_transfer in _vx_transfer_rows:
    _vx_transfer["out_name"] = _vx_transfer.get("out_player_name")
    _vx_transfer["in_name"] = _vx_transfer.get("in_player_name")
    _vx_transfer["xpts_gain_5gw"] = _vx_transfer.get("gain_next_5", 0.0)

CURRENT_SQUAD_DECISIONS = {
    "recommended_lineup": _vx_phase11_rows("phase11_recommended_lineup"),
    "bench_order": _vx_phase11_rows("phase11_bench_order"),
    "captain_rankings": _vx_phase11_rows("phase11_captain_rankings"),
    "transfer_shortlist": _vx_transfer_rows,
}

market_probabilities = {}


def vx_apply_pl_lockup(html_text, *, top_px=4, right_px=30, width_px=540):
    """Inject the single shared Premier League lockup; no slide owns a copy."""
    result = str(html_text)
    full_season = f"{int(EPL_SEASON)}/{int(EPL_SEASON) + 1}"
    result = result.replace("2026/27", full_season)
    result = _vx_re.sub(
        r'<div class="header-season">.*?</div>', "", result, count=1, flags=_vx_re.S
    )
    scoped_css = (
        f"\n.scene{{--pl-bug-top:{int(top_px)}px;--pl-bug-right:{int(right_px)}px;"
        f"--pl-bug-width:{int(width_px)}px;}}\n"
        + str(PL_BUG_CSS)
        + "\n"
    )
    if "</style>" not in result or '<div class="scene' not in result:
        raise RuntimeError("Cannot apply Premier League lockup to malformed slide HTML")
    result = result.replace("</style>", scoped_css + "</style>", 1)
    result = _vx_re.sub(
        r'(<div class="scene(?:\s+[^"]*)?">)',
        lambda match: match.group(1) + str(PL_BUG_HTML),
        result,
        count=1,
    )
    return result


print("✅ Presentation bridge: improved math mapped to the proven weekly contract")
print(f"✅ Official FDR teams: {len(fdr_table)} • model projection teams: {len(team_projection_df)}")

# ------------------------------------------------------------
# CHECK DATA FROM EARLIER CELLS
# ------------------------------------------------------------
if not globals().get("GW_REVIEW_AVAILABLE", False):
    print("⏭️ CELL 14 GW REVIEW slide skipped: no review-ready GW for this run.")
else:
    # GW Review uses only the verified review snapshot / live FPL data.
    if "NEXT_GW" not in globals():
        raise NameError("Run earlier data cells first: missing NEXT_GW")

    ROOT = Path(globals().get("PROJECT_ROOT", Path("output/MyDrive/FPL_VORTEX")))
    HTML_PATH = Path(globals().get("HTML_WORK_DIR", Path("output/.work/html"))) / "gw_review_scene_4k.html"
    PNG_PATH  = Path(globals().get("SLIDE_DIR", Path("output/.work/slide"))) / "gw_review_scene_4k.png"
    HTML_PATH.parent.mkdir(parents=True, exist_ok=True)
    PNG_PATH.parent.mkdir(parents=True, exist_ok=True)

    # ------------------------------------------------------------
    # LOGO — canonical asset from enrichment cell only
    # ------------------------------------------------------------
    for _brand_asset in ("vortex_logo_html", "PL_BUG_HTML", "PL_BUG_CSS"):
        if _brand_asset not in globals():
            raise RuntimeError(f"Run the enrichment/assets cell before GW Review: missing {_brand_asset}")

    # ------------------------------------------------------------
    # HTML
    # ------------------------------------------------------------

    # ------------------------------------------------------------
    # GW REVIEW — UPDATED SOCCER-PITCH DESIGN
    # ------------------------------------------------------------


    # Self-contained approved GW Review pitch template.
    # It consumes only the canonical Python snapshot built earlier.

    EMBEDDED_PITCH_TEMPLATE = '<!DOCTYPE html>\n<html lang="en">\n<head>\n<meta charset="UTF-8" />\n<meta name="viewport" content="width=device-width, initial-scale=1.0" />\n<title>FPL VORTEX — 4K Gameweek Review</title>\n<style>\n  :root{\n    --W:3840px; --H:2160px;\n    --gold:#ffd33f; --gold2:#f6b800; --gold3:#b87900;\n    --ink:#07111e; --ink2:#0d1f31; --ink3:#132a3f;\n    --white:#f7fbff; --muted:#9fb1c3; --lime:#39e05e; --green:#12b94b;\n    --red:#ff5148; --orange:#ff9f26; --cyan:#48d9ff;\n    --panelShadow: 0 26px 70px rgba(0,0,0,.42), inset 0 1px 0 rgba(255,255,255,.45);\n  }\n  *{box-sizing:border-box}\n  html,body{margin:0;width:100%;height:100%;overflow:hidden;background:#02070d;font-family:Inter,Arial,Helvetica,sans-serif;color:var(--white)}\n  body{display:flex;align-items:center;justify-content:center}\n  #scaleShell{width:3840px;height:2160px;transform-origin:center center;position:relative;flex:none}\n  #stage{width:3840px;height:2160px;position:relative;overflow:hidden;background:\n    radial-gradient(circle at 86% 2%,rgba(255,255,255,.82) 0 14px,rgba(182,220,255,.24) 16px 90px,transparent 220px),\n    radial-gradient(circle at 44% 10%,rgba(88,188,255,.20),transparent 20%),\n    linear-gradient(180deg,#071628 0 19%,#0c2234 31%,#153a2c 53%,#06190e 100%);\n  }\n  #stage:before{content:"";position:absolute;inset:0;background:\n    radial-gradient(ellipse at 56% 27%,rgba(124,207,255,.17),transparent 31%),\n    linear-gradient(104deg,transparent 0 9%,rgba(255,255,255,.08) 9.5% 10%,transparent 10.5% 21%,rgba(255,255,255,.06) 21.3% 21.8%,transparent 22.2% 100%);\n    pointer-events:none;mix-blend-mode:screen;opacity:.65}\n  .stadium{position:absolute;left:0;right:0;top:250px;height:900px;overflow:hidden;opacity:.82}\n  .stadium:before{content:"";position:absolute;inset:0;background:\n    repeating-linear-gradient(90deg,rgba(255,255,255,.07) 0 7px,transparent 7px 56px),\n    linear-gradient(180deg,rgba(2,9,18,.25),rgba(9,21,31,.62) 18%,rgba(55,73,83,.22) 48%,rgba(4,9,14,.84) 82%,transparent);\n    clip-path:polygon(0 15%,100% 0,100% 74%,0 88%);filter:blur(2px)}\n  .stadium:after{content:"";position:absolute;left:-5%;right:-5%;bottom:30px;height:245px;background:radial-gradient(ellipse at center,rgba(170,220,255,.2),transparent 60%);filter:blur(30px)}\n  .lights{position:absolute;left:0;right:0;top:0;height:320px;background:\n    radial-gradient(circle at 4% 28%,white 0 8px,#b8e6ff 10px 19px,transparent 48px),\n    radial-gradient(circle at 9% 31%,white 0 8px,#b8e6ff 10px 19px,transparent 48px),\n    radial-gradient(circle at 14% 34%,white 0 8px,#b8e6ff 10px 19px,transparent 48px),\n    radial-gradient(circle at 78% 18%,white 0 8px,#b8e6ff 10px 19px,transparent 48px),\n    radial-gradient(circle at 83% 14%,white 0 8px,#b8e6ff 10px 19px,transparent 48px),\n    radial-gradient(circle at 88% 11%,white 0 8px,#b8e6ff 10px 19px,transparent 48px),\n    radial-gradient(circle at 93% 9%,white 0 8px,#b8e6ff 10px 19px,transparent 48px),\n    radial-gradient(circle at 98% 7%,white 0 8px,#b8e6ff 10px 19px,transparent 48px);opacity:.92}\n\n  /* BRAND */\n  .brand{position:absolute;left:58px;top:46px;width:1185px;height:360px;display:flex;align-items:center;gap:44px;z-index:20}\n  .crestmark{width:288px;height:288px;position:relative;filter:drop-shadow(0 12px 18px rgba(0,0,0,.45))}\n  .crestmark .ring{position:absolute;inset:36px;border:17px solid #fff;border-radius:50%;box-shadow:0 0 28px rgba(255,255,255,.14)}\n  .crestmark .ring:before{content:"";position:absolute;width:92px;height:66px;left:36px;top:82px;border:15px solid #fff;border-left-color:transparent;border-bottom-color:transparent;transform:rotate(-20deg);border-radius:8px}\n  .crestmark .crown{position:absolute;left:58px;top:0;width:170px;height:105px;clip-path:polygon(0 78%,8% 11%,34% 42%,50% 0,67% 42%,91% 11%,100% 78%);background:#fff}\n  .crestmark .ball{position:absolute;left:104px;top:128px;width:78px;height:78px;border:13px solid #fff;border-radius:50%;}\n  .brandText{margin-top:-5px}\n  .brandTop{font-weight:1000;font-size:111px;letter-spacing:-5px;line-height:.85;font-style:italic;text-shadow:0 6px 12px rgba(0,0,0,.55)}\n  .brandMain{font-weight:1000;font-size:148px;letter-spacing:-8px;line-height:.94;font-style:italic;background:linear-gradient(180deg,#fff6a3 0,#ffd84b 33%,#e9a900 76%,#8b5300 100%);-webkit-background-clip:text;color:transparent;text-shadow:0 4px 0 rgba(0,0,0,.08);filter:drop-shadow(0 9px 3px rgba(0,0,0,.55))}\n  .brandSub{font-size:30px;font-weight:900;letter-spacing:8px;color:#d9e6f4;margin-top:12px}\n  .slash{display:inline-block;width:108px;height:13px;background:linear-gradient(90deg,#fff4a6,var(--gold));transform:skewX(-42deg) translateY(-22px);margin-left:9px;box-shadow:0 0 18px rgba(255,210,63,.65)}\n\n  /* LEFT PANELS */\n  .left{position:absolute;left:54px;top:428px;width:1202px;height:1642px;z-index:15;display:flex;flex-direction:column;gap:24px}\n  .panel{position:relative;border-radius:21px;overflow:hidden;background:linear-gradient(180deg,#ffe86f 0,#ffd33e 46%,#efaf13 100%);border:2px solid rgba(255,241,164,.9);box-shadow:var(--panelShadow);color:#09111a}\n  .panel:after{content:"";position:absolute;inset:0;background:linear-gradient(105deg,rgba(255,255,255,.32),transparent 23%,transparent 73%,rgba(122,72,0,.08));pointer-events:none}\n  .panelTitle{height:98px;padding:18px 39px 12px;font-size:52px;font-weight:1000;font-style:italic;letter-spacing:1px;display:flex;align-items:center;justify-content:space-between;background:linear-gradient(90deg,rgba(255,255,255,.47),rgba(255,255,255,.05));border-bottom:5px solid rgba(68,45,0,.65);text-shadow:0 2px 0 rgba(255,255,255,.5)}\n  .panelTitle .smallTitle{font-size:24px;font-style:normal;letter-spacing:3px;padding:9px 16px;background:#091523;color:#fff;border-radius:999px}\n  .reviewPanel{height:474px}\n  .reviewGrid{height:376px;display:grid;grid-template-columns:1fr 1fr 1fr;grid-template-rows:1fr 1fr;gap:2px;background:rgba(20,20,20,.18)}\n  .metric{padding:22px 24px 16px;background:linear-gradient(180deg,rgba(255,248,177,.45),rgba(255,215,61,.10));display:flex;flex-direction:column;justify-content:center}\n  .metricLabel{font-size:27px;font-weight:900;letter-spacing:.8px;opacity:.76}\n  .metricValue{font-size:61px;font-weight:1000;letter-spacing:-3px;line-height:1.08;margin-top:5px}\n  .metricValue.good{color:#086e2d}.metricValue.bad{color:#9d1616}.metricValue.neutral{color:#131c26}\n  .metricSub{font-size:20px;font-weight:850;margin-top:3px;opacity:.7}\n  .chipsPanel{height:272px}\n  .chipRow{height:174px;display:grid;grid-template-columns:repeat(4,1fr);gap:8px;padding:13px 17px 16px}\n  .chip{border-radius:16px;border:2px solid rgba(62,35,0,.18);background:linear-gradient(180deg,rgba(255,255,255,.38),rgba(255,231,115,.14));display:flex;flex-direction:column;justify-content:center;align-items:center;position:relative;text-align:center}\n  .chipCode{width:88px;height:61px;border-radius:13px;background:#071522;color:#fff;display:flex;align-items:center;justify-content:center;font-size:30px;font-weight:1000;box-shadow:0 5px 12px rgba(0,0,0,.22)}\n  .chipName{font-size:22px;font-weight:1000;margin-top:8px}.chipStatus{font-size:18px;font-weight:900;margin-top:2px;color:#116c31}.chipStatus.used{color:#a11b1b}\n  .chipPlan{position:absolute;right:9px;top:8px;font-size:14px;font-weight:1000;background:rgba(4,18,29,.92);color:#ffe263;padding:5px 7px;border-radius:8px}\n  .fixturesPanel{height:294px}\n  .fixtureRow{height:196px;display:grid;grid-template-columns:repeat(5,1fr);padding:12px 14px 15px;gap:8px}\n  .fx{border-right:2px solid rgba(46,34,4,.26);display:flex;flex-direction:column;align-items:center;justify-content:center;min-width:0}.fx:last-child{border-right:none}\n  .fxCrest{width:72px;height:72px;object-fit:contain;filter:drop-shadow(0 3px 3px rgba(0,0,0,.25))}.fxCrest.ph{border-radius:50%;background:#111d2c;color:#fff;font-weight:1000;display:flex;align-items:center;justify-content:center;font-size:22px}\n  .fxOpp{font-size:24px;font-weight:1000;margin-top:5px;white-space:nowrap}\n  .benchPanel{height:408px;flex:1}\n  .benchCards{height:calc(100% - 98px);display:grid;grid-template-columns:repeat(4,1fr);gap:15px;padding:14px 17px 17px}\n  .mini{position:relative;border-radius:15px;background:linear-gradient(180deg,#ffe85f,#dca216);border:2px solid rgba(92,55,0,.28);overflow:hidden;padding-top:39px;text-align:center}\n  .miniPos{position:absolute;left:0;right:0;top:5px;font-size:18px;font-weight:1000;letter-spacing:.7px}\n  .miniPhoto{height:128px;display:flex;align-items:flex-end;justify-content:center;overflow:hidden}.miniPhoto img{height:144px;max-width:100%;object-fit:contain;filter:drop-shadow(0 5px 5px rgba(0,0,0,.25))}.miniAvatar{width:98px;height:98px;border-radius:50%;background:linear-gradient(160deg,#15273a,#07111c);display:flex;align-items:center;justify-content:center;color:#fff;font-size:28px;font-weight:1000;margin-bottom:8px}\n  .miniName{height:45px;background:#07131f;color:#fff;font-size:21px;font-weight:1000;display:flex;align-items:center;justify-content:center;padding:0 5px;white-space:nowrap;overflow:hidden;text-overflow:ellipsis}\n  .miniMeta{height:39px;background:#0b1722;color:#fff;font-size:17px;font-weight:850;display:flex;align-items:center;justify-content:center}.miniPts{position:absolute;right:9px;bottom:8px;width:51px;height:37px;border-radius:7px;background:#2ee86c;color:#06120b;font-size:22px;font-weight:1000;display:flex;align-items:center;justify-content:center}\n\n  /* PITCH */\n  .pitchShell{position:absolute;left:1286px;top:105px;width:2512px;height:1978px;z-index:10;filter:drop-shadow(0 30px 40px rgba(0,0,0,.55))}\n  .pitch{position:absolute;inset:0;clip-path:polygon(6.7% 0,91.5% 0,100% 100%,0 100%);background:\n    repeating-linear-gradient(90deg,rgba(255,255,255,.026) 0 18px,rgba(0,0,0,.018) 18px 35px),\n    repeating-linear-gradient(0deg,rgba(22,115,27,.10) 0 125px,rgba(38,150,39,.08) 125px 250px),\n    radial-gradient(ellipse at 50% 25%,rgba(129,238,70,.34),transparent 37%),\n    linear-gradient(180deg,#3a9d28 0,#208425 27%,#197324 60%,#176622 100%);border:7px solid rgba(238,255,228,.8)}\n  .pitch:before{content:"";position:absolute;inset:0;background-image:\n    radial-gradient(circle at 20% 20%,rgba(255,255,255,.09) 0 1px,transparent 1.5px),\n    radial-gradient(circle at 70% 60%,rgba(0,0,0,.07) 0 1px,transparent 1.5px);background-size:12px 12px,15px 15px;opacity:.55}\n  .pitchSvg{position:absolute;inset:0;width:100%;height:100%;pointer-events:none}\n  .pitchGlow{position:absolute;inset:-20px;clip-path:polygon(6.7% 0,91.5% 0,100% 100%,0 100%);border:16px solid rgba(172,228,255,.26);filter:blur(7px);pointer-events:none}\n  .pitchHeader{position:absolute;top:24px;left:50%;transform:translateX(-50%);z-index:6;background:rgba(4,17,27,.82);border:2px solid rgba(255,221,77,.75);border-radius:18px;padding:10px 26px;display:flex;gap:17px;align-items:center;backdrop-filter:blur(8px);box-shadow:0 10px 24px rgba(0,0,0,.25)}\n  .pitchHeader .gw{font-size:28px;font-weight:1000;color:#ffe45f}.pitchHeader .team{font-size:23px;font-weight:900}.sourceDot{width:12px;height:12px;border-radius:50%;background:#2ee86c;box-shadow:0 0 16px #2ee86c}\n  .players{position:absolute;inset:0;z-index:8}\n  .playerCard{position:absolute;width:300px;height:350px;transform:translate(-50%,-50%);background:linear-gradient(155deg,#07111b 0,#102638 65%,#02070d 100%);border-radius:25px;border:4px solid rgba(236,249,255,.72);box-shadow:0 20px 35px rgba(0,0,0,.58),inset 0 0 0 2px rgba(255,255,255,.08);overflow:hidden}\n  .playerCard.captain{border-color:#ffdf52;box-shadow:0 20px 35px rgba(0,0,0,.58),0 0 25px rgba(255,217,66,.36),inset 0 0 0 2px rgba(255,255,255,.12)}\n  .playerCard.vice{border-color:#9ddcff}\n  .photoZone{position:absolute;left:0;right:0;top:0;height:226px;display:flex;align-items:flex-end;justify-content:center;background:radial-gradient(circle at 50% 42%,rgba(56,97,128,.6),transparent 62%);overflow:hidden}\n  .photoZone img{height:226px;max-width:255px;object-fit:contain;object-position:center bottom;filter:drop-shadow(0 7px 8px rgba(0,0,0,.55))}\n  .avatarFallback{width:165px;height:165px;border-radius:50%;display:flex;align-items:center;justify-content:center;margin-bottom:22px;background:linear-gradient(145deg,#274a65,#07131f);font-size:46px;font-weight:1000;color:#dcecff;border:3px solid rgba(255,255,255,.18)}\n  .teamCrest{position:absolute;right:14px;top:13px;width:72px;height:72px;object-fit:contain;filter:drop-shadow(0 4px 3px rgba(0,0,0,.65))}\n  .badge{position:absolute;left:13px;top:14px;min-width:58px;height:54px;padding:0 10px;border-radius:12px;background:#ffda43;color:#06111a;font-size:28px;font-weight:1000;display:flex;align-items:center;justify-content:center;box-shadow:0 4px 10px rgba(0,0,0,.35)}\n  .badge.vc{background:#b8e8ff}.playerName{position:absolute;left:0;right:0;top:221px;height:58px;background:linear-gradient(180deg,#03070b,#0b1117);border-top:3px solid rgba(230,243,255,.58);border-bottom:3px solid rgba(230,243,255,.58);display:flex;align-items:center;justify-content:center;padding:0 8px;font-size:25px;font-weight:1000;white-space:nowrap;overflow:hidden;text-overflow:ellipsis;text-shadow:0 2px 2px #000}\n  .playerMeta{position:absolute;left:0;right:0;bottom:0;height:69px;background:#07111a;display:flex;align-items:center;padding:0 84px 0 16px;font-size:23px;font-weight:900;white-space:nowrap;overflow:hidden;text-overflow:ellipsis}.points{position:absolute;right:0;bottom:0;width:81px;height:69px;display:flex;align-items:center;justify-content:center;font-size:34px;font-weight:1000;color:#04100b;border-left:3px solid rgba(255,255,255,.4)}\n  .points.p0{background:#6f7c87;color:#fff}.points.p1{background:#ef5548;color:#fff}.points.p2{background:#a4ec42}.points.p3{background:#5ee84d}.points.p4{background:#2fdb5b}.points.p5{background:#14be54}.points.p6{background:#06a950;color:#fff}.points.haul{background:linear-gradient(180deg,#ffe96e,#ffb21f);color:#241400}\n  .tinyScore{position:absolute;left:12px;bottom:8px;font-size:13px;font-weight:900;color:#8ca2b5;letter-spacing:1px}\n\n  /* review ribbons on pitch */\n  .analysisRibbon{position:absolute;right:46px;top:44px;width:510px;z-index:11;display:flex;flex-direction:column;gap:12px}\n  .insight{height:82px;border-radius:16px;background:linear-gradient(90deg,rgba(4,16,27,.95),rgba(8,27,42,.88));border:2px solid rgba(255,220,65,.48);box-shadow:0 10px 24px rgba(0,0,0,.28);padding:10px 18px;display:grid;grid-template-columns:134px 1fr;align-items:center;backdrop-filter:blur(7px)}\n  .insightKey{font-size:17px;color:#ffdb55;font-weight:1000;letter-spacing:1.1px}.insightVal{font-size:24px;font-weight:1000;white-space:nowrap;overflow:hidden;text-overflow:ellipsis}.insightVal.good{color:#5cff80}.insightVal.bad{color:#ff7e72}\n\n  __PL_BUG_CSS__\n  .footer{position:absolute;left:58px;bottom:22px;right:64px;height:47px;display:flex;align-items:center;justify-content:space-between;z-index:30;font-size:24px;font-weight:850;color:#b8c7d4;letter-spacing:1px;text-shadow:0 2px 4px #000}.footer strong{color:#ffe04a}\n  .loading{position:absolute;left:50%;top:50%;transform:translate(-50%,-50%);z-index:60;border-radius:26px;padding:28px 44px;background:rgba(3,13,23,.93);border:2px solid rgba(255,220,72,.6);font-size:29px;font-weight:900;box-shadow:0 20px 50px rgba(0,0,0,.45)}\n  .hide{display:none!important}\n</style>\n\n<style id="gw-review-v2-ui">\n.panel{background:linear-gradient(180deg,#fff2a6 0%,#ffe37a 48%,#f4c94f 100%) !important;border-color:#fff1b7 !important}\n.panel:after{background:linear-gradient(105deg,rgba(255,255,255,.48),transparent 26%,transparent 76%,rgba(110,65,0,.05)) !important}\n.panelTitle{background:linear-gradient(90deg,rgba(255,255,255,.72),rgba(255,244,177,.30)) !important;border-bottom-color:rgba(69,43,0,.52) !important}\n.metric{background:linear-gradient(180deg,rgba(255,255,255,.48),rgba(255,236,139,.22)) !important}\n.metricLabel{opacity:.92 !important;color:#17202a !important}\n.metricSub{opacity:.88 !important;color:#24303a !important;font-size:22px !important;font-weight:900 !important}\n.chip{background:linear-gradient(180deg,#fff7c7 0%,#ffe98c 100%) !important;border-color:rgba(71,43,0,.22) !important}\n.chipCode{width:96px !important;height:66px !important;border-radius:18px !important;font-size:32px !important;letter-spacing:1px !important;color:#fff !important;border:3px solid rgba(255,255,255,.72) !important;box-shadow:0 7px 16px rgba(0,0,0,.24),inset 0 0 0 2px rgba(255,255,255,.12) !important}\n.chip:nth-child(1) .chipCode{background:linear-gradient(135deg,#4b0e8f,#e61792) !important}\n.chip:nth-child(2) .chipCode{background:linear-gradient(135deg,#00c8ff,#5b2cff) !important}\n.chip:nth-child(3) .chipCode{background:linear-gradient(135deg,#29e56f,#08a7d8) !important}\n.chip:nth-child(4) .chipCode{background:linear-gradient(135deg,#ffd447,#ff4f91) !important;color:#151018 !important}\n.chipName{font-size:23px !important}.chipStatus{font-size:19px !important}\n.pitchHeader{top:30px !important;left:70px !important;right:auto !important;transform:none !important;justify-content:flex-start !important;min-width:710px !important;padding:15px 26px !important;border-radius:15px !important;background:rgba(4,17,27,.90) !important}\n.pitchHeader .gw{font-size:48px !important;line-height:1 !important;font-weight:1000 !important;letter-spacing:1.2px !important;color:#ffe681 !important;text-transform:uppercase !important}\n.pitchHeader .team{font-size:25px !important;font-weight:950 !important;color:#fff !important;text-transform:uppercase !important}\n.sourceDot{width:15px !important;height:15px !important}\n.analysisRibbon{display:none !important}\n.playerCard{width:310px !important;height:350px !important}\n.photoZone{height:210px !important}\n.photoZone img{height:210px !important;max-width:270px !important}\n.teamCrest{width:82px !important;height:82px !important;right:12px !important;top:10px !important}\n.playerName{top:206px !important;height:66px !important;background:linear-gradient(180deg,#fff2a3,#f1c64e) !important;color:#07111a !important;border-top:3px solid rgba(255,255,255,.88) !important;border-bottom:3px solid rgba(91,58,0,.32) !important;font-size:30px !important;font-weight:1000 !important;text-shadow:none !important;padding:0 10px !important}\n.playerMeta{height:78px !important;background:linear-gradient(180deg,#fff8cf,#f6df8c) !important;color:#101922 !important;padding:0 102px 0 17px !important;font-size:27px !important;font-weight:1000 !important;letter-spacing:.2px !important}\n.tinyScore{left:17px !important;bottom:7px !important;font-size:16px !important;color:#3b4650 !important;font-weight:1000 !important}\n.points{width:96px !important;height:78px !important;font-size:43px !important;font-weight:1000 !important;border-left:3px solid rgba(55,37,0,.25) !important}\n.mini{background:linear-gradient(180deg,#fff2a6,#f2c84e) !important}\n.miniName{background:#fff0a4 !important;color:#101820 !important;font-size:23px !important;font-weight:1000 !important;border-top:2px solid rgba(86,55,0,.22) !important}\n.miniMeta{background:#fff8ce !important;color:#202b34 !important;font-size:19px !important;font-weight:950 !important}\n.miniPts{font-size:24px !important}\n</style>\n\n<style id="vortex-gw-review-final-polish">\n.brand.brandImage{left:58px!important;top:42px!important;width:1185px!important;height:345px!important;display:flex!important;align-items:center!important;justify-content:flex-start!important}\n.brand.brandImage .vortex-logo{display:block!important;width:auto!important;max-width:1120px!important;height:auto!important;max-height:320px!important;object-fit:contain!important;object-position:left center!important;filter:drop-shadow(0 12px 18px rgba(0,0,0,.42))!important}\n.brand.brandImage .logo-fallback,.brand.brandImage .text-logo{font-size:118px!important;font-weight:1000!important;color:#fff!important}\n#reviewMode{display:none!important}\n.pitchHeader{top:30px!important;left:190px!important;right:auto!important;transform:none!important;min-width:0!important;width:auto!important;max-width:1180px!important;padding:15px 28px!important;justify-content:flex-start!important}\n.pitchHeader .gw{font-size:50px!important;line-height:1!important;font-weight:1000!important;letter-spacing:1.2px!important;white-space:nowrap!important}\n.pitchHeader .team{font-size:27px!important;font-weight:1000!important;white-space:nowrap!important}\n.chipRow{height:174px!important;grid-template-columns:repeat(4,1fr)!important;gap:12px!important;padding:10px 17px 13px!important}\n.chip{border-radius:16px!important;padding:6px 8px 8px!important;justify-content:center!important;background:linear-gradient(180deg,#fff8c8 0%,#f7dd80 100%)!important;border:2px solid rgba(69,43,0,.18)!important}\n.chipPlan{display:none!important}\n.chipIcon{width:58px!important;height:58px!important;object-fit:contain!important;display:block!important;margin:0 auto 2px!important;filter:brightness(.48) saturate(.92) contrast(1.15)!important}\n.chipName{font-size:23px!important;line-height:1.05!important;font-weight:1000!important;margin:1px 0 5px!important;color:#101722!important;text-transform:uppercase!important;white-space:nowrap!important}\n.chipStatus{width:92%!important;min-height:34px!important;padding:5px 8px!important;margin:0 auto!important;border-radius:7px!important;display:flex!important;align-items:center!important;justify-content:center!important;font-size:20px!important;line-height:1!important;font-weight:1000!important;letter-spacing:.7px!important;text-transform:uppercase!important;border:2px solid rgba(0,0,0,.22)!important}\n.chipStatus.available{background:#49e675!important;color:#07130b!important}\n.chipStatus.used{background:#e7353d!important;color:#fff!important;border-color:#a70d14!important}\n.benchCards{height:calc(100% - 98px)!important;gap:14px!important;padding:12px 17px 15px!important}\n.mini{padding-top:28px!important;background:linear-gradient(180deg,#fff4b0 0%,#f1ce62 100%)!important;border-radius:18px!important;overflow:hidden!important}\n.miniPos{top:4px!important;font-size:18px!important;line-height:1.05!important}\n.miniPhoto{height:188px!important;display:flex!important;align-items:flex-end!important;justify-content:center!important;overflow:hidden!important;padding-top:10px!important}\n.miniPhoto img{height:172px!important;max-width:94%!important;object-fit:contain!important;object-position:center bottom!important;filter:drop-shadow(0 6px 6px rgba(0,0,0,.28))!important}\n.miniAvatar{width:142px!important;height:142px!important;font-size:40px!important;margin-bottom:7px!important}\n.miniName{height:62px!important;background:linear-gradient(180deg,#fff2a3,#f1c64e)!important;color:#07111a!important;font-size:30px!important;line-height:1!important;font-weight:1000!important;padding:0 8px!important;border-top:3px solid rgba(255,255,255,.85)!important;border-bottom:3px solid rgba(91,58,0,.28)!important}\n.miniMeta{height:72px!important;background:linear-gradient(180deg,#fff8cf,#f6df8c)!important;color:#101922!important;font-size:27px!important;line-height:1!important;font-weight:1000!important;padding:0 88px 0 10px!important;justify-content:flex-start!important}\n.miniPts{right:0!important;bottom:0!important;width:80px!important;height:72px!important;border-radius:0!important;font-size:40px!important;font-weight:1000!important;border-left:3px solid rgba(55,37,0,.25)!important}\n.miniPts.p0{background:#6f7c87!important;color:#fff!important}.miniPts.p1{background:#ef5548!important;color:#fff!important}.miniPts.p2{background:#a4ec42!important;color:#07120b!important}.miniPts.p3{background:#5ee84d!important;color:#07120b!important}.miniPts.p4{background:#2fdb5b!important;color:#07120b!important}.miniPts.p5{background:#14be54!important;color:#07120b!important}.miniPts.p6{background:#06a950!important;color:#fff!important}.miniPts.haul{background:linear-gradient(180deg,#ffe96e,#ffb21f)!important;color:#241400!important}\n</style>\n\n\n<style id="gw-review-v3-final-request">\n/* Left data stack slightly higher for clean footer clearance */\n.left{top:398px!important;}\n\n/* Review metric typography: black, larger, fully opaque; GW points red */\n.metric{padding:19px 24px 14px!important;}\n.metricLabel{font-size:32px!important;line-height:1!important;font-weight:1000!important;color:#070b0f!important;opacity:1!important;letter-spacing:.5px!important;}\n.metricValue{font-size:68px!important;line-height:1!important;font-weight:1000!important;color:#070b0f!important;opacity:1!important;margin-top:8px!important;}\n.metricValue.good,.metricValue.bad,.metricValue.neutral{color:#070b0f!important;}\n#gwPoints{color:#d71920!important;}\n.metricSub{font-size:25px!important;line-height:1.05!important;font-weight:1000!important;color:#070b0f!important;opacity:1!important;margin-top:8px!important;}\n\n/* Chips: actual reference symbols, enlarged to fit the card, all dark blue */\n.chipRow{height:174px!important;gap:12px!important;padding:7px 17px 10px!important;}\n.chip{padding:3px 7px 7px!important;justify-content:center!important;overflow:hidden!important;}\n.chipIcon{width:82px!important;height:82px!important;max-width:82px!important;max-height:82px!important;object-fit:contain!important;margin:-2px auto 0!important;filter:none!important;}\n.chipName{font-size:22px!important;line-height:1!important;margin:-3px 0 5px!important;color:#071522!important;font-weight:1000!important;}\n.chipStatus{min-height:33px!important;height:33px!important;font-size:19px!important;line-height:1!important;padding:4px 7px!important;}\n\n/* Bench cards: team crest by player\'s head/right, no overlap */\n.miniPhoto{position:relative!important;height:188px!important;padding-top:10px!important;overflow:hidden!important;}\n.miniPlayerImg{height:172px!important;max-width:88%!important;object-fit:contain!important;object-position:center bottom!important;}\n.miniCrest{position:absolute!important;right:8px!important;top:14px!important;width:55px!important;height:55px!important;object-fit:contain!important;z-index:8!important;padding:4px!important;border-radius:50%!important;background:rgba(255,250,218,.92)!important;border:2px solid rgba(7,21,34,.24)!important;box-shadow:0 4px 10px rgba(0,0,0,.24)!important;}\n.miniName{font-size:30px!important;font-weight:1000!important;}\n.miniMeta{font-size:27px!important;font-weight:1000!important;}\n.miniPts{font-size:40px!important;font-weight:1000!important;}\n\n/* Footer social strip: large, centered, clear */\n.footer{left:58px!important;right:58px!important;bottom:10px!important;height:66px!important;font-size:24px!important;font-weight:950!important;color:#eef6ff!important;letter-spacing:.5px!important;text-shadow:0 2px 4px #000!important;display:flex!important;align-items:center!important;justify-content:space-between!important;z-index:50!important;}\n.footerLeft{white-space:nowrap!important;}\n.footerRight{white-space:nowrap!important;font-size:22px!important;color:#cbd8e5!important;}\n.footerSocial{position:absolute!important;left:50%!important;top:50%!important;transform:translate(-50%,-50%)!important;display:flex!important;align-items:center!important;justify-content:center!important;gap:32px!important;white-space:nowrap!important;font-size:29px!important;font-weight:1000!important;color:#ffffff!important;letter-spacing:.5px!important;}\n.socialPair{display:inline-flex!important;align-items:center!important;gap:10px!important;}\n.ytIcon{width:48px!important;height:34px!important;display:inline-flex!important;align-items:center!important;justify-content:center!important;filter:drop-shadow(0 2px 3px rgba(0,0,0,.45))!important;}\n.ytIcon svg{width:48px!important;height:34px!important;display:block!important;}\n.xIcon{width:38px!important;height:38px!important;border-radius:9px!important;background:#020406!important;color:#ffffff!important;border:2px solid rgba(255,255,255,.88)!important;display:inline-flex!important;align-items:center!important;justify-content:center!important;font-size:27px!important;line-height:1!important;font-family:Arial,Helvetica,sans-serif!important;font-weight:1000!important;box-shadow:0 2px 4px rgba(0,0,0,.5)!important;}\n</style>\n\n</head>\n<body>\n<div id="scaleShell">\n<div id="stage">\n  <div class="lights"></div><div class="stadium"></div>\n\n  <div class="brand brandImage">__VORTEX_LOGO_HTML__</div>\n\n  <section class="left">\n    <div class="panel reviewPanel">\n      <div class="panelTitle"><span>GAMEWEEK REVIEW</span></div>\n      <div class="reviewGrid">\n        <div class="metric"><div class="metricLabel">GW POINTS</div><div class="metricValue neutral" id="gwPoints">—</div><div class="metricSub" id="vsAvg">vs avg —</div></div>\n        <div class="metric"><div class="metricLabel">OVERALL RANK</div><div class="metricValue neutral" id="overallRank">—</div><div class="metricSub" id="rankDelta">movement —</div></div>\n        <div class="metric"><div class="metricLabel">GW RANK</div><div class="metricValue neutral" id="gwRank">—</div><div class="metricSub" id="gwPercentile">global finish</div></div>\n        <div class="metric"><div class="metricLabel">CAPTAIN RETURN</div><div class="metricValue neutral" id="captainReturn">—</div><div class="metricSub" id="captainName">captain</div></div>\n        <div class="metric"><div class="metricLabel">BENCH POINTS</div><div class="metricValue neutral" id="benchPoints">—</div><div class="metricSub">unused output</div></div>\n        <div class="metric"><div class="metricLabel">TRANSFER COST</div><div class="metricValue neutral" id="hitPoints">0</div><div class="metricSub" id="teamValue">team value —</div></div>\n      </div>\n    </div>\n\n    <div class="panel chipsPanel">\n      <div class="panelTitle"><span>CHIP INTELLIGENCE</span><span class="smallTitle">H1 • USE BY GW19</span></div>\n      <div class="chipRow" id="chipRow"></div>\n    </div>\n\n    <div class="panel fixturesPanel">\n      <div class="panelTitle"><span id="fixtureTitle">GW REVIEW • CAPTAIN</span><span class="smallTitle">FIXTURE</span></div>\n      <div class="fixtureRow" id="fixtureRow"></div>\n    </div>\n\n    <div class="panel benchPanel">\n      <div class="panelTitle"><span>SUBSTITUTES</span><span id="benchSum" class="smallTitle">0 PTS</span></div>\n      <div class="benchCards" id="benchCards"></div>\n    </div>\n  </section>\n\n  <section class="pitchShell">\n    <div class="pitch"></div><div class="pitchGlow"></div>\n    <svg class="pitchSvg" viewBox="0 0 2512 1978" preserveAspectRatio="none" aria-hidden="true">\n      <g fill="none" stroke="rgba(246,255,239,.90)" stroke-width="10">\n        <path d="M171,43 L2290,43 L2473,1870 L38,1870 Z" />\n        <line x1="104" y1="1000" x2="2385" y2="1000" />\n        <ellipse cx="1245" cy="1000" rx="252" ry="165" />\n        <circle cx="1245" cy="1000" r="11" fill="rgba(246,255,239,.9)" />\n        <path d="M672 43 L1811 43 L1848 377 L633 377 Z" />\n        <path d="M979 43 L1507 43 L1523 188 L962 188 Z" />\n        <path d="M466 1870 L2048 1870 L2017 1515 L492 1515 Z" />\n        <path d="M835 1870 L1679 1870 L1665 1688 L849 1688 Z" />\n        <path d="M1035 377 C1105 494 1385 494 1455 377" />\n        <path d="M920 1515 C1015 1375 1465 1375 1560 1515" />\n        <path d="M104 1000 C104 936 161 886 224 886" /><path d="M2385 1000 C2385 936 2328 886 2265 886" />\n        <path d="M38 1870 C38 1802 98 1748 168 1748" /><path d="M2473 1870 C2473 1802 2413 1748 2343 1748" />\n      </g>\n    </svg>\n    <div class="pitchHeader"><span class="sourceDot" id="sourceDot"></span><span class="gw" id="pitchGW">GW — REVIEW</span><span class="team" id="entryName">FPL VORTEX</span></div>\n    <div class="analysisRibbon" id="analysisRibbon"></div>\n    <div class="players" id="players"></div>\n  </section>\n\n  __PL_BUG_HTML__<div class="footer">\n    <span class="footerLeft">FPL VORTEX &nbsp;|&nbsp; <strong id="footerGW">GAMEWEEK REVIEW</strong> &nbsp;|&nbsp; __SEASON_FULL__ SEASON</span>\n    <span class="footerSocial">\n      <span class="socialPair"><span class="ytIcon" aria-label="YouTube"><svg viewBox="0 0 48 34" aria-hidden="true"><rect x="1" y="1" width="46" height="32" rx="9" fill="#ff0000"/><path d="M20 9 L34 17 L20 25 Z" fill="#ffffff"/></svg></span><span>@FPLVORTEX</span></span>\n      <span class="socialPair"><span class="xIcon" aria-label="X">𝕏</span><span>@FPLVORTEXM</span></span>\n    </span>\n    <span id="dataStamp" class="footerRight">PUBLIC FPL DATA • ENTRY 158256</span>\n  </div>\n  <div id="loading" class="loading">Loading FPL review data…</div>\n</div>\n</div>\n<script>\n/* ============================================================\n   FPL VORTEX — GAMEWEEK REVIEW 4K\n   Native canvas: 3840 × 2160 (16:9)\n   Live URL options:\n     fpl_vortex_gw_review_4k.html?team=158256\n     fpl_vortex_gw_review_4k.html?team=158256&gw=3\n   ============================================================ */\nconst CONFIG = { teamId: __TEAM_ID__ };\nconst $ = s => document.querySelector(s);\nconst nf = n => (n==null || Number.isNaN(Number(n))) ? \'—\' : Number(n).toLocaleString(\'en-GB\');\nconst initials = s => (s||\'?\').split(/\\s+/).map(x=>x[0]).join(\'\').slice(0,2).toUpperCase();\nconst safe = s => String(s ?? \'\').replace(/[&<>"]/g, c=>({\'&\':\'&amp;\',\'<\':\'&lt;\',\'>\':\'&gt;\',\'"\':\'&quot;\'}[c]));\nfunction fit(){ const scale=Math.min(innerWidth/3840,innerHeight/2160); $(\'#scaleShell\').style.transform=`scale(${scale})`; }\naddEventListener(\'resize\',fit); fit();\nfunction playerImg(code){ return code ? `https://resources.premierleague.com/premierleague/photos/players/250x250/p${code}.png` : \'\'; }\nfunction crestImg(code){ return code ? `https://resources.premierleague.com/premierleague/badges/70/t${code}.png` : \'\'; }\nfunction pointsClass(p){ p=Number(p||0); if(p>=8) return \'haul\'; if(p<=0)return \'p0\'; if(p===1)return \'p1\'; if(p===2)return \'p2\'; if(p===3)return \'p3\'; if(p===4)return \'p4\'; if(p===5)return \'p5\'; return \'p6\'; }\nfunction posLabel(type){ return ({1:\'GK\',2:\'DEF\',3:\'MID\',4:\'FWD\'})[type]||\'SUB\'; }\nfunction chipPretty(name){ return ({wildcard:[\'WC\',\'WILDCARD\'],freehit:[\'FH\',\'FREE HIT\'],bboost:[\'BB\',\'BENCH BOOST\'],\'3xc\':[\'TC\',\'TRIPLE CAPTAIN\']})[name] || [name.toUpperCase(),name.toUpperCase()]; }\nfunction deltaFmt(d){ if(d==null) return \'movement —\'; if(d>0) return `▲ ${nf(d)} places`; if(d<0) return `▼ ${nf(Math.abs(d))} places`; return \'— no movement\'; }\nfunction setMetric(id,val,mode=\'neutral\'){ const el=$(id); el.textContent=val; el.className=\'metricValue \'+mode; }\nfunction scoreMode(n,good=0,bad=0){ return n>=good?\'good\':n<=bad?\'bad\':\'neutral\'; }\n\nconst CHIP_ICONS={"bboost":"data:image/png;base64,iVBORw0KGgoAAAANSUhEUgAAAC4AAAA8CAYAAAAZp4inAAABhklEQVR4nO1ZSxbCIAykfS68i8fxnB7Hu7jTFb4aC2SSIf08ZifFzBCSJtApEXG93d+156/nY2JxzSxDLdHaOVq4PWAV4/W+y+MeD3q9bxbO2HaPDXi7mHG6BBo66sm9BEtoF9CchAouEbPsZBQfIkToNjNs/w32FMzk+vmhNcSsgFbe7+twK9GIzaXGWQ4wCLLNHmGXbaoLEGJ4KRhZAOIYlfAe4eHlonWHKdVDjl3AmsItieOZo+W8aAwxxMj5XdvayNhGuUc/Hm2jGOOtrVp7Hn5Y3jKWa1jTlceo7/FIDOHRKHaHloyv5Yolj2q63JVTIirRIeG9T/r0tnaPqAqPukuxcB/W41NKtlIt/2Pt262tA8XjSFKx3joqjzMJW9BqUXs8IlERDqg77CkePY/OcoBFgMByiD7HpSfLaAuMu3L6xf4aEbvTTOnMn1IkDvfxSoK1AGuCm0s+o5J6bLh6FQ+xd+G0/iP6UwytH9cI2uvF08DAwMDAwPlhPnJtXb4Pe3f4Ae1eCAMNvOM7AAAAAElFTkSuQmCC","3xc":"data:image/png;base64,iVBORw0KGgoAAAANSUhEUgAAADYAAAA8CAYAAAAt3km7AAABzElEQVR4nO2ZQW7EMAhF66qL3qXH6TlznNwlu3YVybKCDZ8P7lS85UyC+TbGBLe3RD6/vn9m/1/n0VhjvbMMrViJ0j6jhTZDEqiz3tULXTHPCnhXL0wYI6w8NuihyNwnPdbQpAmLEjSiFegWZhUkOcaycwMLszhiDSOGbbOwSEHMsUJmkllBoOOq0/0uURabvY/mF1AHGCFs8YMmbHSGmf5721pfKJXHODD7TOvtacMyrbr3Yp0st7CIZMEYk7piWWWVhg/Py+PMXefR0K9kpKSaveMShpD14UkXdjvACkt0D8N7jBVSGpAjRD0bo+EnYZ7qAikCZn5Rs6ImbK7zaE/PsY+NkD0mzb61NJLe1RBSeaycsIYsspoqYU+OSM6tNjqSXCSbs9/+fOWB2tySPKJs99D32M4v6B6XMCnGV4f3mBBmCUKqZFYhSk3392Cr5HE7q1kJdI9RWgOe88mCtg6ltQYy2JI8kJ6EBSQiUtpv2f19dfvNalRr04J1fHUoIt1YFsik1qUEc1Cmbcl++sXfkyMMGyN1Vavl312uj7xs+20F4zzz2AitFT2OeScm7aYku3JJq+4j2wZFURRFURR7adE9wV28TCfYyi+STGRoqyIsewAAAABJRU5ErkJggg==","wildcard":"data:image/png;base64,iVBORw0KGgoAAAANSUhEUgAAAC8AAAA8CAYAAAD2ZeOZAAABgElEQVR4nO1ZSxaDIBCDvi56lx6n5/Q43sVdu9KnVGFmEkB4ZOkniWHk6x0Zr/fnG7u/zJNnaT1YRM6ljUufkYKSgtUQ2gpw8kiSaCtA5hklgHCYmo1Zt3toy0j1cC7TIaQfIXpIa/pKnMWzInpTI6ZtcgY3nBDa3SFaf8JSMuZIadU9GECMM+pZq78RWI2jPZCVb5knrxqk2KWCaojNlzCu1aLOKktDZL5k6hrNvpOvkbpUu+/kESzz5HO2XNR8zZKRYDPPNrrnQ7jP3l2vPa2kUqGze6xFTdNbH/QfNmaOvYzczJ8Rl1qzxhDz1XQ/T/1h9wgTy9HtZkm+VAkmzdes+5R20zUvMl8jfYlm/8k7VzZ9qZYq+RIfoNHoY9PJSsACvN2nJQrJLKButLKIU2CUWZbDhTMxBkeI/o91QjR5oBaC9RHWnx6aHjC6S4QDntuwtzU0qD7QIKDOKiWm7r4LNzAwMDAwMOCdO85Jro5fpMO65R0rHleTqTscLKTwAxF8+OSxa/oLAAAAAElFTkSuQmCC","freehit":"data:image/png;base64,iVBORw0KGgoAAAANSUhEUgAAADQAAAA8CAYAAAApK5mGAAABu0lEQVR4nO2Zy3UDIQxFmTlZuBeX4zpTjnvJzllNDsFgPUlPYuJwlzbogxgkxFaIXK63x6v/v+6fG1Nfj50lSHIGHePFvWJWI6Oi5YqQZ8WjomV2iGFQhFPqsEetLGsLwkIyPuhS/I6Jk7WOjAxiyZEYTtIYoFUeKftpcJSyQ249J0LXr0GoAm9E2vlMvT/HdpYzvd9QmYiNOzpQo1hjgFa2JBNOrJHOWHX0gBzKcMaqq4VWbR9kJeARokOWo9mLJ0ofDANK0TszGh+65ZjHKUovAWtwf0MRR75ngdIueJfr7TG1pyApt6xiVAKv2S0CIg2r5/TmSzLVW067zTSy0yqFg8ikyfq+YIcQZ6xGMQ+LrRQ5yc2KjCX5ihH6C9ushl6cokzrnCKKtcZFJtiXDqF11Yyb6Qi42mb0207xnDLjjjOtOEXJiMwB5FBmj8C7I6iVQku2M6Uoq+0z9LvhatvqlKXEl2SOQOTSmvWWXkB4sz5amUW2Vj79watnQGZb+P89Sba8zaNxyxnavj3MpQ/DkNNd8DwGRdV3NKEZTUQEWrUdcbNdLBaLxWKx4PIWpUhdR057TmHRFsXfD9E72ChremIAAAAASUVORK5CYII="};\nfunction renderChips(chips, gw){\n  const names=[\'bboost\',\'3xc\',\'wildcard\',\'freehit\'];\n  const byName=Object.fromEntries((chips||[]).map(c=>[c.name,c.event]));\n  $(\'#chipRow\').innerHTML=names.map(name=>{\n    const [code,title]=chipPretty(name), used=byName[name];\n    const status=used?`USED GW${used}`:\'AVAILABLE\';\n    const statusClass=used?\'used\':\'available\';\n    const icon=CHIP_ICONS[name]||\'\';\n    return `<div class="chip"><img class="chipIcon" src="${icon}" alt="${safe(title)}"><div class="chipName">${title}</div><div class="chipStatus ${statusClass}">${status}</div></div>`;\n  }).join(\'\');\n}\nfunction renderFixtures(fixtures){\n  $(\'#fixtureRow\').innerHTML=(fixtures||[]).slice(0,5).map(f=>{\n    const crest=f.teamCode?`<img class="fxCrest" src="${crestImg(f.teamCode)}" onerror="this.replaceWith(Object.assign(document.createElement(\'div\'),{className:\'fxCrest ph\',textContent:\'${safe(initials(f.opp))}\'}))">`:`<div class="fxCrest ph">${safe(initials(f.opp))}</div>`;\n    return `<div class="fx">${crest}<div class="fxOpp">${safe(f.opp)} (${safe(f.side)})</div></div>`;\n  }).join(\'\');\n}\nfunction renderBench(bench){\n  const total=bench.reduce((s,p)=>s+Number(p.points||0),0); $(\'#benchSum\').textContent=`${total} PTS`;\n  $(\'#benchCards\').innerHTML=bench.map((p,i)=>{\n    const img=p.code?`<img class="miniPlayerImg" src="${playerImg(p.code)}" onerror="this.style.display=\'none\';this.nextElementSibling.style.display=\'flex\'">`:\'\';\n    const crest=p.teamCode?`<img class="miniCrest" src="${crestImg(p.teamCode)}" onerror="this.style.display=\'none\'">`:\'\';\n    return `<div class="mini"><div class="miniPos">${i+1}. ${posLabel(p.type)}</div><div class="miniPhoto">${img}<div class="miniAvatar" style="${p.code?\'display:none\':\'\'}">${safe(initials(p.name))}</div>${crest}</div><div class="miniName">${safe(p.name)}</div><div class="miniMeta">${safe(p.fixture||\'REVIEW FIXTURE\')}</div><div class="miniPts ${pointsClass(p.points)}">${p.points??0}</div></div>`;\n  }).join(\'\');\n}\nfunction cardHtml(p,x,y){\n  const badge=p.captain?`<div class="badge">C</div>`:p.vice?`<div class="badge vc">VC</div>`:\'\';\n  const photo=p.code?`<img src="${playerImg(p.code)}" onerror="this.style.display=\'none\';this.nextElementSibling.style.display=\'flex\'">`:\'\';\n  const crest=p.teamCode?`<img class="teamCrest" src="${crestImg(p.teamCode)}" onerror="this.style.display=\'none\'">`:\'\';\n  return `<div class="playerCard ${p.captain?\'captain\':\'\'} ${p.vice?\'vice\':\'\'}" style="left:${x}%;top:${y}%">${badge}<div class="photoZone">${photo}<div class="avatarFallback" style="${p.code?\'display:none\':\'\'}">${safe(initials(p.name))}</div>${crest}</div><div class="playerName">${safe(p.name)}</div><div class="playerMeta">${safe(p.fixture||p.teamName||\'FIXTURE\')}<span class="tinyScore">${posLabel(p.type)}</span></div><div class="points ${pointsClass(p.points)}">${p.points??0}</div></div>`;\n}\nfunction renderPlayers(starters){\n  const groups={1:[],2:[],3:[],4:[]}; starters.forEach(p=>(groups[p.type]||groups[3]).push(p));\n  const ys={1:13.5,2:35.5,3:58.2,4:82.0};\n  const html=[];\n  [1,2,3,4].forEach(type=>{\n    const arr=groups[type]; if(!arr.length)return;\n    const span = type===1 ? 0 : (type===4?47:56);\n    arr.forEach((p,i)=>{\n      const x= arr.length===1 ? 50 : 50-span/2 + (span*(i/(arr.length-1)));\n      html.push(cardHtml(p,x,ys[type]));\n    });\n  });\n  $(\'#players\').innerHTML=html.join(\'\');\n}\nfunction renderInsights(data){\n  const starters=data.players.filter(p=>!p.bench), bench=data.players.filter(p=>p.bench);\n  const top=[...starters].sort((a,b)=>b.points-a.points)[0];\n  const cap=starters.find(p=>p.captain)||starters[0];\n  const benchPts=bench.reduce((s,p)=>s+p.points,0);\n  const diff=Number(data.history.points||0)-Number(data.event.average_entry_score||0);\n  const hit=Number(data.history.event_transfers_cost||0);\n  const rows=[\n    [\'TOP RETURN\',`${top?.name||\'—\'} • ${top?.points||0} pts`,\'good\'],\n    [\'CAPTAIN\',`${cap?.name||\'—\'} • ${cap?.points||0} raw pts`,cap?.points>=6?\'good\':\'\'],\n    [\'VS AVERAGE\',`${diff>=0?\'+\':\'\'}${diff} pts`,diff>=0?\'good\':\'bad\'],\n    [\'BENCH LEAK\',`${benchPts} pts`,benchPts>=10?\'bad\':benchPts<=5?\'good\':\'\'],\n    [\'TRANSFER HIT\',hit?`-${hit} pts`:\'No hit\',hit?\'bad\':\'good\']\n  ];\n  $(\'#analysisRibbon\').innerHTML=rows.map(r=>`<div class="insight"><div class="insightKey">${r[0]}</div><div class="insightVal ${r[2]}">${safe(r[1])}</div></div>`).join(\'\');\n}\nfunction render(data){\n  const h=data.history,e=data.event;\n  $(\'#pitchGW\').textContent=`GAMEWEEK ${e.id} REVIEW`;\n  $(\'#footerGW\').textContent=`GAMEWEEK ${e.id} REVIEW`;\n  $(\'#entryName\').textContent=(data.entry.name||\'FPL VORTEX\').toUpperCase();\n  setMetric(\'#gwPoints\',nf(h.points),scoreMode(Number(h.points)-Number(e.average_entry_score),4,-4));\n  const diff=Number(h.points||0)-Number(e.average_entry_score||0); $(\'#vsAvg\').textContent=`${diff>=0?\'+\':\'\'}${diff} vs avg ${e.average_entry_score??\'—\'}`;\n  setMetric(\'#overallRank\',nf(h.overall_rank),\'neutral\');\n  const rankMove=data.previousRank?Number(data.previousRank)-Number(h.overall_rank):null; $(\'#rankDelta\').textContent=deltaFmt(rankMove);\n  setMetric(\'#gwRank\',nf(h.rank),\'neutral\'); $(\'#gwPercentile\').textContent=e.highest_score?`high score ${e.highest_score}`:\'global finish\';\n  const starters=data.players.filter(p=>!p.bench), cap=starters.find(p=>p.captain)||starters[0];\n  const capMult=cap?.multiplier||2, capReturn=(cap?.points||0)*capMult; setMetric(\'#captainReturn\',capReturn,scoreMode(capReturn,12,4)); $(\'#captainName\').textContent=cap?`${cap.name} ×${capMult}`:\'captain\';\n  const bench=data.players.filter(p=>p.bench), bp=bench.reduce((s,p)=>s+Number(p.points||0),0); setMetric(\'#benchPoints\',bp,scoreMode(-bp,-5,-10));\n  const hit=Number(h.event_transfers_cost||0); setMetric(\'#hitPoints\',hit?`-${hit}`:\'0\',hit?\'bad\':\'good\'); $(\'#teamValue\').textContent=`team £${((h.value??data.entry.last_deadline_value??1000)/10).toFixed(1)}m`;\n  renderChips(data.chips,e.id); renderFixtures(data.fixtures); renderBench(bench); renderPlayers(starters); renderInsights(data);\n  $(\'#dataStamp\').textContent=`LIVE PUBLIC FPL API • ENTRY ${CONFIG.teamId}`;\n  $(\'#loading\').classList.add(\'hide\');\n}\n\nconst PY_LIVE_DATA = __REVIEW_DATA_JSON__;\nrender(PY_LIVE_DATA);\n</script>\n</body>\n</html>\n'

    # Build the final HTML directly; no unused intermediate template file.
    html = EMBEDDED_PITCH_TEMPLATE
    html = html.replace("__VORTEX_LOGO_HTML__", str(vortex_logo_html))
    html = html.replace("__PL_BUG_HTML__", PL_BUG_HTML)
    html = html.replace("__PL_BUG_CSS__", PL_BUG_CSS)
    html = html.replace("__SEASON_FULL__", f"{int(EPL_SEASON)}/{int(EPL_SEASON) + 1}")

    # ------------------------------------------------------------
    # USE CURRENT FPL TEAM FROM NOTEBOOK
    # ------------------------------------------------------------

    TEAM_ID_LOCAL = int(
        globals().get("FPL_TEAM_ID", 0)
        or globals().get("TEAM_ID", 0)
        or globals().get("ENTRY_ID", 0)
        or 0
    )

    if not TEAM_ID_LOCAL:
        raise RuntimeError(
            "FPL_TEAM_ID is missing. Set it in Cell 0 (for example 145784)."
        )

    # Consume the ONE canonical GW_REVIEW snapshot created upstream.
    if "GW_REVIEW" not in globals():
        raise RuntimeError("Run GW REVIEW DATA SNAPSHOT before this slide.")

    try:
        _boot = GW_REVIEW["bootstrap"]
        _entry = GW_REVIEW["entry"]
        _hist = GW_REVIEW["history"]
        _picks = GW_REVIEW["picks"]
        _live = GW_REVIEW["live"]
        _fixtures_all = GW_REVIEW["fixtures"]
        if int(GW_REVIEW["gw"]) != int(REVIEW_GW):
            raise RuntimeError(
                f"GW_REVIEW snapshot GW{GW_REVIEW['gw']} does not match upstream REVIEW_GW{REVIEW_GW}."
            )

        _events = {int(e["id"]): e for e in _boot.get("events", [])}
        _event = _events.get(REVIEW_GW, {"id": REVIEW_GW})
        _elements = {int(x["id"]): x for x in _boot.get("elements", [])}
        _teams = {int(t["id"]): t for t in _boot.get("teams", [])}
        _live_by = {int(x["id"]): x for x in _live.get("elements", [])}
        _gw_fixtures = [f for f in _fixtures_all if int(f.get("event") or 0) == REVIEW_GW]

        def _fixture_for_team(team_id):
            return next((
                f for f in _gw_fixtures
                if int(f.get("team_h", 0)) == int(team_id)
                or int(f.get("team_a", 0)) == int(team_id)
            ), None)

        _players = []
        for pk in _picks.get("picks", []):
            pid = int(pk.get("element", 0))
            el = _elements.get(pid, {})
            team_id = int(el.get("team", 0) or 0)
            tm = _teams.get(team_id, {})
            lv = (_live_by.get(pid, {}) or {}).get("stats", {}) or {}
            fx = _fixture_for_team(team_id)
            fixture = "—"
            if fx:
                home = int(fx.get("team_h", 0)) == team_id
                opp = _teams.get(int(fx.get("team_a" if home else "team_h", 0)), {})
                fixture = f"{opp.get('short_name', 'OPP')} ({'H' if home else 'A'})"
            _players.append({
                "id": pid,
                "name": el.get("web_name") or el.get("second_name") or "Player",
                "type": int(el.get("element_type", 3) or 3),
                "teamName": tm.get("short_name", ""),
                "fixture": fixture,
                "points": int(lv.get("total_points", 0) or 0),
                "captain": bool(pk.get("is_captain")),
                "vice": bool(pk.get("is_vice_captain")),
                "multiplier": int(pk.get("multiplier", 0) or 0),
                "bench": int(pk.get("position", 99) or 99) > 11,
                "code": el.get("code"),
                "teamCode": tm.get("code"),
            })
        _cap = next((p for p in _players if p.get("captain")), None) or next((p for p in _players if not p.get("bench")), None)
        _review_cap_fixtures = []
        if _cap:
            _cap_team = int((_elements.get(int(_cap["id"]), {}) or {}).get("team", 0) or 0)
            for f in _gw_fixtures:
                if _cap_team not in (int(f.get("team_h", 0)), int(f.get("team_a", 0))):
                    continue
                home = int(f.get("team_h", 0)) == _cap_team
                opp_id = int(f.get("team_a" if home else "team_h", 0))
                opp = _teams.get(opp_id, {})
                _review_cap_fixtures.append({
                    "opp": opp.get("short_name", "OPP"),
                    "side": "H" if home else "A",
                    "teamCode": opp.get("code"),
                    "event": REVIEW_GW,
                    "score": f'{f.get("team_h_score","—")}–{f.get("team_a_score","—")}',
                })
                break

        _current = next((x for x in _hist.get("current", []) if int(x.get("event", 0)) == REVIEW_GW), {})
        _prev = next((x for x in _hist.get("current", []) if int(x.get("event", 0)) == REVIEW_GW - 1), None)
        LIVE_REVIEW_DATA = {
            "entry": _entry,
            "event": {
                "id": int(_event.get("id", REVIEW_GW)),
                "average_entry_score": _event.get("average_entry_score"),
                "highest_score": _event.get("highest_score"),
                "finished": bool(_event.get("finished")),
            },
            "history": _current,
            "previousRank": (_prev or {}).get("overall_rank"),
            "chips": _hist.get("chips", []),
            "players": _players,
            "fixtures": _review_cap_fixtures,
        }
        print(f"✅ LIVE FPL REVIEW DATA: entry {TEAM_ID_LOCAL} • GW{REVIEW_GW} • {len(_players)} players")
    except Exception as _gw_exc:
        raise RuntimeError(f"Could not build GW Review from canonical snapshot: {_gw_exc}") from _gw_exc


    # Inject the already-fetched canonical Python snapshot.
    _live_json = json.dumps(LIVE_REVIEW_DATA, ensure_ascii=False).replace("</", "<\\/")
    html = html.replace("__TEAM_ID__", str(int(TEAM_ID_LOCAL)), 1)
    html = html.replace("__REVIEW_DATA_JSON__", _live_json, 1)
    if "__TEAM_ID__" in html or "__REVIEW_DATA_JSON__" in html:
        raise RuntimeError("GW Review HTML snapshot placeholders were not fully resolved.")

    # ------------------------------------------------------------
    # FULL 4K PAGE LOCK
    # ------------------------------------------------------------

    FULL_PAGE_FIX = """
    <style id="weekly-full-page-fix">
    html,body,#scaleShell,#stage{
        width:3840px !important;
        height:2160px !important;
        min-width:3840px !important;
        min-height:2160px !important;
        margin:0 !important;
        padding:0 !important;
        overflow:hidden !important;
    }
    #scaleShell{
        position:absolute !important;
        inset:0 !important;
        transform:none !important;
        transform-origin:top left !important;
    }
    </style>
    """

    html = html.replace(
        "</head>",
        FULL_PAGE_FIX + "\n</head>"
    )

    html = vx_polish_branding(html)

    HTML_PATH.write_text(
        html,
        encoding="utf-8"
    )

    print("✅ GW REVIEW PITCH HTML:", HTML_PATH)
    print(f"✅ Reviewing GW {REVIEW_GW}")
    print("✅ 3840 × 2160 full-page pitch")

    # ------------------------------------------------------------
    # RENDER WITH CELL 14'S AUTHORITATIVE SELF-CONTAINED 4K EXPORTER
    # ------------------------------------------------------------

    await export_4k_png(HTML_PATH, PNG_PATH)

    print("✅ GW REVIEW PNG:", PNG_PATH)
    print("✅ 3840 × 2160")
    print("👇 GW REVIEW PREVIEW")

    display(IPImage(filename=str(PNG_PATH)))
```

```python
# ============================================================
# CELL 22 — FULL-WORD NARRATION
# NO SHORT CODES • NO INITIALS • NO ABBREVIATED TTS LABELS
# ============================================================

import html as _vx_html
import re as _vx_re

for _vx_required in ("teams", "players"):
    if _vx_required not in globals():
        raise RuntimeError(
            f"Run the canonical FPL data cells before FULL-WORD NARRATION — missing {_vx_required}"
        )


def _vx_narration_records(obj):
    if hasattr(obj, "to_dict"):
        try:
            return list(obj.to_dict(orient="records"))
        except TypeError:
            pass
    if isinstance(obj, list):
        return [dict(x) for x in obj if isinstance(x, dict)]
    if isinstance(obj, tuple):
        return [dict(x) for x in obj if isinstance(x, dict)]
    return []


def _vx_narration_key(value):
    return _vx_re.sub(r"[^a-z0-9]+", "", str(value or "").casefold())


def _vx_narration_spoken_player_name(value):
    text = str(value or "").strip()
    match = _vx_re.fullmatch(
        r"[A-Za-z]\.\s*([^\W\d_][\w'’\-]*(?:\s+[^\W\d_][\w'’\-]*)*)",
        text,
    )
    return match.group(1) if match else text


# ------------------------------------------------------------
# CANONICAL TEAM SPEECH REGISTRY
# Built dynamically from the official FPL team table.
# No club names or club-code replacements are hard-coded here.
# ------------------------------------------------------------

VX_NARRATION_TEAM_BY_ID = {}
VX_NARRATION_TEAM_BY_CODE = {}
VX_NARRATION_TEAM_BY_ALIAS = {}
VX_NARRATION_TEAM_SHORT_ALIASES = set()

for _vx_team in _vx_narration_records(teams):
    try:
        _vx_team_id = int(_vx_team.get("id"))
    except Exception:
        continue

    _vx_spoken = str(_vx_team.get("name") or "").strip()
    if not _vx_spoken:
        continue

    VX_NARRATION_TEAM_BY_ID[_vx_team_id] = _vx_spoken

    try:
        _vx_code = int(_vx_team.get("code"))
        VX_NARRATION_TEAM_BY_CODE[_vx_code] = _vx_spoken
    except Exception:
        pass

    for _vx_alias in (
        _vx_team.get("name"),
        _vx_team.get("short_name"),
    ):
        _vx_alias = str(_vx_alias or "").strip()
        if not _vx_alias:
            continue
        VX_NARRATION_TEAM_BY_ALIAS[_vx_narration_key(_vx_alias)] = _vx_spoken

    _vx_short = str(_vx_team.get("short_name") or "").strip()
    if _vx_short:
        VX_NARRATION_TEAM_SHORT_ALIASES.add(_vx_short.casefold())


# ------------------------------------------------------------
# CANONICAL PLAYER SPEECH REGISTRY
# Every player resolves to official FPL web_name.
# Full names and initial+surname forms are generated from FPL data,
# never from a manually-maintained player replacement table.
# ------------------------------------------------------------

VX_NARRATION_PLAYER_BY_ID = {}
VX_NARRATION_PLAYER_BY_CODE = {}
VX_NARRATION_PLAYER_BY_ALIAS = {}
VX_NARRATION_PLAYER_TEXT_ALIASES = []

_vx_second_name_candidates = {}

for _vx_player in _vx_narration_records(players):
    try:
        _vx_player_id = int(_vx_player.get("id"))
    except Exception:
        continue

    _vx_web = str(_vx_player.get("web_name") or "").strip()
    if not _vx_web:
        continue
    _vx_spoken = _vx_narration_spoken_player_name(_vx_web)

    VX_NARRATION_PLAYER_BY_ID[_vx_player_id] = _vx_web

    try:
        _vx_code = int(_vx_player.get("code"))
        VX_NARRATION_PLAYER_BY_CODE[_vx_code] = _vx_web
    except Exception:
        pass

    _vx_first = str(_vx_player.get("first_name") or "").strip()
    _vx_second = str(_vx_player.get("second_name") or "").strip()
    _vx_full = " ".join(x for x in (_vx_first, _vx_second) if x).strip()

    for _vx_alias in (_vx_web, _vx_full):
        if _vx_alias:
            VX_NARRATION_PLAYER_BY_ALIAS[_vx_narration_key(_vx_alias)] = _vx_web

    if _vx_second:
        _vx_second_name_candidates.setdefault(_vx_narration_key(_vx_second), set()).add(_vx_web)

    if _vx_full and _vx_narration_key(_vx_full) != _vx_narration_key(_vx_web):
        VX_NARRATION_PLAYER_TEXT_ALIASES.append((_vx_full, _vx_spoken))

    if _vx_spoken.casefold() != _vx_web.casefold():
        VX_NARRATION_PLAYER_TEXT_ALIASES.append((_vx_web, _vx_spoken))

    if _vx_first and _vx_second:
        _vx_initial = _vx_first[0]
        VX_NARRATION_PLAYER_TEXT_ALIASES.extend([
            (f"{_vx_initial}. {_vx_second}", _vx_spoken),
            (f"{_vx_initial} {_vx_second}", _vx_spoken),
        ])

# A surname-only alias is safe only when it points to one official web_name.
for _vx_alias_key, _vx_web_names in _vx_second_name_candidates.items():
    if len(_vx_web_names) == 1:
        _vx_web = next(iter(_vx_web_names))
        VX_NARRATION_PLAYER_BY_ALIAS.setdefault(_vx_alias_key, _vx_web)

# Longest aliases first prevents a short alias from breaking a longer name.
VX_NARRATION_PLAYER_TEXT_ALIASES.sort(
    key=lambda pair: len(pair[0]),
    reverse=True,
)

VX_NARRATION_TEAM_TEXT_ALIASES = sorted(
    [
        (str(_vx_team.get("short_name") or "").strip(), str(_vx_team.get("name") or "").strip())
        for _vx_team in _vx_narration_records(teams)
        if str(_vx_team.get("short_name") or "").strip()
        and str(_vx_team.get("name") or "").strip()
        and _vx_narration_key(_vx_team.get("short_name")) != _vx_narration_key(_vx_team.get("name"))
    ],
    key=lambda pair: len(pair[0]),
    reverse=True,
)


def vx_narration_team(value):
    """Return the canonical spoken team name from id/code/name/short-name input."""
    if isinstance(value, dict):
        for key in ("id", "team_id", "team"):
            if key in value:
                try:
                    team_id = int(value[key])
                    if team_id in VX_NARRATION_TEAM_BY_ID:
                        return VX_NARRATION_TEAM_BY_ID[team_id]
                except Exception:
                    pass
        value = value.get("name") or value.get("short_name") or value.get("team") or ""

    try:
        number = int(value)
        if number in VX_NARRATION_TEAM_BY_ID:
            return VX_NARRATION_TEAM_BY_ID[number]
        if number in VX_NARRATION_TEAM_BY_CODE:
            return VX_NARRATION_TEAM_BY_CODE[number]
    except Exception:
        pass

    text = str(value or "").strip()
    resolved = VX_NARRATION_TEAM_BY_ALIAS.get(_vx_narration_key(text))
    if resolved:
        return resolved
    return text


def vx_narration_player(value):
    """Return the natural spoken player name from id/code/name-like input."""
    if isinstance(value, dict):
        for key in ("id", "player_id", "element"):
            if key in value:
                try:
                    player_id = int(value[key])
                    if player_id in VX_NARRATION_PLAYER_BY_ID:
                        return _vx_narration_spoken_player_name(VX_NARRATION_PLAYER_BY_ID[player_id])
                except Exception:
                    pass
        value = (
            value.get("web_name")
            or value.get("name")
            or " ".join(
                x for x in (
                    str(value.get("first_name") or "").strip(),
                    str(value.get("second_name") or "").strip(),
                ) if x
            )
        )

    try:
        number = int(value)
        if number in VX_NARRATION_PLAYER_BY_ID:
            return _vx_narration_spoken_player_name(VX_NARRATION_PLAYER_BY_ID[number])
        if number in VX_NARRATION_PLAYER_BY_CODE:
            return _vx_narration_spoken_player_name(VX_NARRATION_PLAYER_BY_CODE[number])
    except Exception:
        pass

    text = str(value or "").strip()
    resolved = VX_NARRATION_PLAYER_BY_ALIAS.get(_vx_narration_key(text), text)
    return _vx_narration_spoken_player_name(resolved)


def vx_narration_fixture(opponent, is_home):
    """Natural fixture phrase; caller supplies a canonical team identifier."""
    club = vx_narration_team(opponent)
    return f"{club} at home" if bool(is_home) else f"{club} away"


def _vx_replace_narration_phrase(text, source, target):
    source = str(source or "").strip()
    target = str(target or "").strip()
    if not source or not target or source.casefold() == target.casefold():
        return text
    pattern = _vx_re.compile(
        r"(?<![A-Za-z0-9])" + _vx_re.escape(source) + r"(?![A-Za-z0-9])",
        flags=_vx_re.IGNORECASE,
    )
    return pattern.sub(lambda _m: target, text)


def vx_assert_full_word_narration(text, context="narration"):
    """Fail closed: abbreviated/compact tokens never reach Ryan Edge TTS."""
    value = str(text or "")
    problems = []

    # Player web_name is explicitly the canonical speech identity. Protect exact
    # official web_name spans from generic initial/acronym diagnostics.
    diagnostic_value = value
    for _vx_web in sorted(set(VX_NARRATION_PLAYER_BY_ID.values()), key=len, reverse=True):
        if not _vx_web:
            continue
        diagnostic_value = _vx_re.sub(
            r"(?<![A-Za-z0-9])" + _vx_re.escape(_vx_web) + r"(?![A-Za-z0-9])",
            " playerwebnameplaceholder ",
            diagnostic_value,
            flags=_vx_re.IGNORECASE,
        )

    # Official short team aliases surviving the dynamic replacement are blocked.
    words = [w.casefold() for w in _vx_re.findall(r"\b[A-Za-z][A-Za-z0-9'.-]*\b", value)]
    surviving_team_codes = sorted({w for w in words if w in VX_NARRATION_TEAM_SHORT_ALIASES})
    if surviving_team_codes:
        problems.append("team short names: " + ", ".join(surviving_team_codes))

    # API-style labels are not natural narration.
    api_tokens = sorted(set(_vx_re.findall(r"\b[A-Za-z][A-Za-z0-9]*_[A-Za-z0-9_]+\b", diagnostic_value)))
    if api_tokens:
        problems.append("API-style fields: " + ", ".join(api_tokens[:8]))

    # All-cap abbreviations / compact labels, including letter+digit forms.
    compact_caps = sorted(set(_vx_re.findall(r"\b[A-Z]{2,}[A-Z0-9]*\b", diagnostic_value)))
    if compact_caps:
        problems.append("capital abbreviations: " + ", ".join(compact_caps[:8]))

    # Mixed short forms such as lower-case prefix + capital letter are blocked.
    mixed_compact = sorted(set(_vx_re.findall(r"\b[a-z][A-Z][A-Za-z0-9]*\b", diagnostic_value)))
    if mixed_compact:
        problems.append("mixed abbreviations: " + ", ".join(mixed_compact[:8]))

    # Initial + surname forms must resolve to the official web_name first.
    initials = sorted(set(_vx_re.findall(r"\b[A-Z]\.\s+[A-Z][A-Za-z'’-]+\b", diagnostic_value)))
    if initials:
        problems.append("initial-form player names: " + ", ".join(initials[:8]))

    # Parenthesised one-letter fixture labels are compact display syntax.
    compact_fixture = sorted(set(_vx_re.findall(r"\([A-Za-z]\)", diagnostic_value)))
    if compact_fixture:
        problems.append("compact fixture labels: " + ", ".join(compact_fixture[:8]))

    if problems:
        raise RuntimeError(
            f"FULL-WORD NARRATION blocked {context}: " + " | ".join(problems)
        )
    return value


def vx_full_word_narration(text, context="narration"):
    """Canonical final text transform used immediately before every Ryan TTS call."""
    value = _vx_html.unescape(str(text or ""))

    # Resolve player shorthand from official FPL identities to web_name.
    for source, target in VX_NARRATION_PLAYER_TEXT_ALIASES:
        value = _vx_replace_narration_phrase(value, source, target)

    # Resolve official short team names dynamically to canonical team names.
    for source, target in VX_NARRATION_TEAM_TEXT_ALIASES:
        value = _vx_replace_narration_phrase(value, source, target)

    # Natural punctuation/number wording; no football-specific replacement table.
    value = _vx_re.sub(r"(?<=\d)\s*%", " percent", value)
    value = value.replace("&", " and ")

    # API-style field syntax is intentionally NOT guessed or expanded here.
    # The fail-closed validator blocks it so a future caller must supply real
    # natural-language narration rather than a machine field label.

    value = _vx_re.sub(r"[ \t]+", " ", value)
    value = _vx_re.sub(r" *\n *", "\n", value)
    value = value.strip()

    return vx_assert_full_word_narration(value, context=context)


def vx_full_word_display(text, context="narration display"):
    """Narration-facing on-screen text follows the exact same full-word contract."""
    return vx_full_word_narration(text, context=context)


print("✅ FULL-WORD NARRATION READY")
print(f"✅ Dynamic team speech names: {len(VX_NARRATION_TEAM_BY_ID)}")
print(f"✅ Dynamic player web names: {len(VX_NARRATION_PLAYER_BY_ID)}")
print("✅ Unknown abbreviations fail closed before Ryan TTS")
```

```python
# ============================================================
# CELL 15 — AUTO GW REVIEW DATA + RYAN NARRATION TIMING
# ============================================================

from pathlib import Path
import json, subprocess, sys, re, unicodedata

ROOT   = Path(globals().get("PROJECT_ROOT", Path("output/MyDrive/FPL_VORTEX")))
AUDIO  = Path(globals().get("AUDIO_WORK_DIR", Path("output/.work/audio"))) / "review_segments"
TIMING = Path(globals().get("DATA_DIR", Path("output/.work/data")))
AUDIO.mkdir(parents=True, exist_ok=True)
TIMING.mkdir(parents=True, exist_ok=True)

if "GW_REVIEW" not in globals() or "REVIEW_GW" not in globals():
    raise RuntimeError("Run GW REVIEW DATA SNAPSHOT before this cell.")
if int(GW_REVIEW["gw"]) != int(REVIEW_GW):
    raise RuntimeError(
        f"GW_REVIEW snapshot GW{GW_REVIEW['gw']} does not match upstream REVIEW_GW{REVIEW_GW}."
    )

boot  = GW_REVIEW["bootstrap"]
picks = GW_REVIEW["picks"]
hist  = GW_REVIEW["history"]
live  = GW_REVIEW["live"]
all_review_fixtures = list(GW_REVIEW["fixtures"])
review_fx = [
    f for f in all_review_fixtures
    if int(f.get("event") or 0) == REVIEW_GW
]

review_teams = {x["id"]: x for x in boot["teams"]}
review_elements = {x["id"]: x for x in boot["elements"]}
review_live_map = {x["id"]: x["stats"] for x in live["elements"]}

# previous overall rank → arrow direction
current_hist=picks["entry_history"]
previous=[
    x for x in hist.get("current",[])
    if int(x["event"]) < REVIEW_GW
]

prev_rank=previous[-1]["overall_rank"] if previous else None
cur_rank=current_hist.get("overall_rank")

rank_dir="flat"
if prev_rank and cur_rank:
    rank_dir="up" if cur_rank < prev_rank else "down" if cur_rank > prev_rank else "flat"

def fixture_for(team_id):
    for f in review_fx:
        if f["team_h"]==team_id or f["team_a"]==team_id:
            home=f["team_h"]==team_id
            opp=review_teams[f["team_a"] if home else f["team_h"]]["short_name"]
            score=f'{f.get("team_h_score","—")}–{f.get("team_a_score","—")}'
            return {
                "opp":opp,
                "ha":"H" if home else "A",
                "score":score
            }
    return {"opp":"—","ha":"","score":"—"}


def next_fixture_for(team_id):
    """First scheduled fixture after the reviewed GW from the canonical snapshot."""
    candidates=[]
    for f in all_review_fixtures:
        event=int(f.get("event") or 0)
        if event <= REVIEW_GW:
            continue
        if f.get("team_h") != team_id and f.get("team_a") != team_id:
            continue
        kickoff=str(f.get("kickoff_time") or "")
        candidates.append((event, kickoff, f))
    if not candidates:
        return {"gw":None,"opp":"—","opp_name":"","ha":"","is_home":None,"kickoff_time":""}
    _, _, f=sorted(candidates,key=lambda x:(x[0],x[1]))[0]
    home=f["team_h"]==team_id
    opp_team=review_teams[f["team_a"] if home else f["team_h"]]
    return {
        "gw":int(f.get("event") or 0),
        "opp":opp_team["short_name"],
        "opp_name":opp_team["name"],
        "ha":"H" if home else "A",
        "is_home":bool(home),
        "kickoff_time":str(f.get("kickoff_time") or "")
    }


# Reuse the one authoritative player xPts/minutes engine and the one team
# probability model. This cell only packages their already-computed outputs
# into natural narration; it does not recalculate either model.
_review_horizon_by_id={}
if "player_horizon_xpts_df" in globals():
    try:
        _review_horizon_by_id={
            int(r["player_id"]):r
            for _,r in player_horizon_xpts_df.iterrows()
        }
    except Exception:
        _review_horizon_by_id={}

_review_next_cs_by_team={}
if "fixture_projection_df" in globals():
    try:
        _cs_tmp={}
        for _,r in fixture_projection_df.iterrows():
            if int(r.get("gw") or 0) != int(NEXT_GW):
                continue
            _cs_tmp.setdefault(int(r["home_id"]),[]).append(float(r["home_cs"]))
            _cs_tmp.setdefault(int(r["away_id"]),[]).append(float(r["away_cs"]))
        _review_next_cs_by_team={
            tid:sum(vals)/len(vals)
            for tid,vals in _cs_tmp.items()
            if vals
        }
    except Exception:
        _review_next_cs_by_team={}

_review_decisions=globals().get("CURRENT_SQUAD_DECISIONS",{}) or {}
_review_lineup_ids={
    int(r.get("player_id"))
    for r in _review_decisions.get("recommended_lineup",[])
    if r.get("player_id") is not None
}
_review_bench_ids={
    int(r.get("player_id"))
    for r in _review_decisions.get("bench_order",[])
    if r.get("player_id") is not None
}

POS={1:"GK",2:"DEF",3:"MID",4:"FWD"}
review_players=[]

for pick in picks["picks"]:
    e=review_elements[pick["element"]]
    s=review_live_map.get(pick["element"],{})
    t=review_teams[e["team"]]
    fixture=fixture_for(e["team"])

    review_players.append({
        "id":e["id"],
        "name":e["web_name"],
        "first_name":e.get("first_name", ""),
        "second_name":e.get("second_name", ""),
        "full_name":f'{e["first_name"]} {e["second_name"]}'.strip(),
        "team":t["short_name"],
        "team_name":t["name"],
        "team_code":t.get("code"),
        "pos":POS[e["element_type"]],
        "season_total_points":int(e.get("total_points",0) or 0),
        "season_form":float(e.get("form",0) or 0),
        "season_xg":float(e.get("expected_goals",0) or 0),
        "season_xa":float(e.get("expected_assists",0) or 0),
        "season_xgi":float(e.get("expected_goal_involvements",0) or 0),
        "season_xgc":float(e.get("expected_goals_conceded",0) or 0),
        "season_ppg":float(e.get("points_per_game",0) or 0),
        "season_bonus":int(e.get("bonus",0) or 0),
        "season_bps":int(e.get("bps",0) or 0),
        "season_penalties_saved":int(e.get("penalties_saved",0) or 0),
        "bench":pick["position"]>11,
        "squad_position":pick["position"],
        "captain":pick["is_captain"],
        "vice":pick["is_vice_captain"],
        "multiplier":pick["multiplier"],

        "points":int(s.get("total_points",0) or 0),
        "minutes":int(s.get("minutes",0) or 0),
        "goals":int(s.get("goals_scored",0) or 0),
        "assists":int(s.get("assists",0) or 0),
        "clean_sheets":int(s.get("clean_sheets",0) or 0),
        "goals_conceded":int(s.get("goals_conceded",0) or 0),
        "saves":int(s.get("saves",0) or 0),
        "bonus":int(s.get("bonus",0) or 0),
        "bps":int(s.get("bps",0) or 0),
        "gw_xg":float(s.get("expected_goals",0) or 0),
        "gw_xa":float(s.get("expected_assists",0) or 0),
        "gw_xgi":float(s.get("expected_goal_involvements",0) or 0),
        "gw_xgc":float(s.get("expected_goals_conceded",0) or 0),

        "price":float(e["now_cost"])/10,
        "ownership":float(e.get("selected_by_percent",0) or 0),

        "photo": (
            resolve_player_photo(e)
            if "resolve_player_photo" in globals()
            else (
                f'https://resources.premierleague.com/premierleague/'
                f'photos/players/250x250/p{e["code"]}.png'
            )
        ),

        "fixture":fixture,
        "next_fixture":next_fixture_for(e["team"]),
        "next_cs_pct":_review_next_cs_by_team.get(int(e["team"])),
        "model_projection":(
            {
                "next_gw_xpts":float(_review_horizon_by_id[e["id"]].get("next_gw_xpts",0) or 0),
                "xpts_5gw":float(_review_horizon_by_id[e["id"]].get("xpts_5gw",0) or 0),
                "p_start_next_gw":float(_review_horizon_by_id[e["id"]].get("p_start_next_gw",0) or 0),
                "p_appear_next_gw":float(_review_horizon_by_id[e["id"]].get("p_appear_next_gw",0) or 0),
                "p_60_next_gw":float(_review_horizon_by_id[e["id"]].get("p_60_next_gw",0) or 0),
                "expected_minutes_next_gw":float(_review_horizon_by_id[e["id"]].get("expected_minutes_next_gw",0) or 0),
            }
            if e["id"] in _review_horizon_by_id else {}
        ),
        "next_decision":(
            "start"
            if e["id"] in _review_lineup_ids else
            "bench"
            if e["id"] in _review_bench_ids else
            "watch"
        )
    })

review_starters=sorted(
    [p for p in review_players if not p["bench"]],
    key=lambda x:(["GK","DEF","MID","FWD"].index(x["pos"]),x["squad_position"])
)

review_bench=sorted(
    [p for p in review_players if p["bench"]],
    key=lambda x:x["squad_position"]
)

# ------------------------------------------------------------

# NATURAL GAMEWEEK REVIEW — SQUAD REVEAL
# ------------------------------------------------------------
# One authoritative narration path only.
#
# Spoken structure:
#   player identity -> this-GW evidence -> what it means -> next decision
#
# Animation contract is consumed by Cell 16:
#   Ryan says name -> pitch/bench image + hero reveal together -> hero leaves
#   -> player image stays -> full review board appears after the final reserve.
# ------------------------------------------------------------

bank=float(current_hist.get("bank",0))/10
value=float(current_hist.get("value",0))/10
gw_points=int(current_hist.get("points",0) or 0)
gw_rank=current_hist.get("rank")
transfer_cost=int(current_hist.get("event_transfers_cost",0) or 0)


def vx_review_join(items):
    items = [str(x).strip() for x in items if str(x).strip()]
    if not items:
        return ""
    if len(items) == 1:
        return items[0]
    if len(items) == 2:
        return " and ".join(items)
    return ", ".join(items[:-1]) + ", and " + items[-1]


def vx_review_next_match(p):
    f=p.get("next_fixture") or {}
    opp=str(f.get("opp_name") or "").strip()
    if not opp and f.get("opp") not in (None, "", "—"):
        opp=vx_narration_team(f.get("opp"))
    if not opp:
        return "the next fixture is still to be confirmed"
    is_home=f.get("is_home")
    if is_home is True:
        return f"next up is {opp} at home"
    if is_home is False:
        return f"next up is {opp} away"
    return f"next up is {opp}"


def vx_review_sentence_case(text):
    text=str(text or "")
    return text[:1].upper()+text[1:] if text else text


def vx_review_action(p):
    """Simple player recommendation based on expected minutes and the current squad setup."""
    model=p.get("model_projection") or {}
    exp_min=float(model.get("expected_minutes_next_gw",0) or 0)
    p_start=float(model.get("p_start_next_gw",0) or 0)
    pid=int(p["id"])

    if exp_min and (exp_min < 45 or p_start < .55):
        return "watch"
    if pid in _review_lineup_ids:
        return "start"
    if pid in _review_bench_ids:
        return "bench"
    return "hold"


def vx_review_action_phrase(action):
    return {
        "start":"in my starting eleven",
        "bench":"a bench option for now",
        "watch":"one I want to monitor",
        "hold":"a comfortable hold",
    }.get(str(action),str(action))


def vx_review_model_line(p, defensive=False):
    model=p.get("model_projection") or {}
    exp_min=float(model.get("expected_minutes_next_gw",0) or 0)
    xpts=float(model.get("next_gw_xpts",0) or 0)
    cs=p.get("next_cs_pct")
    pieces=[]

    if exp_min > 0:
        pieces.append(f"around {exp_min:.0f} expected minutes")
    if xpts > 0:
        pieces.append(f"{xpts:.1f} expected points")
    if defensive and cs is not None:
        pieces.append(f"a {float(cs):.0f} percent clean-sheet chance")

    if not pieces:
        return ""
    return "For the next Gameweek, the model currently has him at " + vx_review_join(pieces) + "."


def vx_review_return_detail(p):
    """Highlights the main reason behind the player's return in the reviewed Gameweek."""
    if p["pos"]=="GK":
        if p["clean_sheets"] and p["saves"]>=3:
            return f"the clean sheet plus {p['saves']} saves"
        if p["clean_sheets"]:
            return "the clean sheet"
        if p["saves"]>=3:
            return f"the {p['saves']} saves he made despite conceding"
        if p["goals_conceded"]:
            return f"the {p['goals_conceded']} goals conceded, which limited the defensive return"
        return "the minutes he got on the pitch"

    if p["pos"]=="DEF":
        routes=[]
        if p["goals"]:
            routes.append(f"{p['goals']} goal" + ("s" if p["goals"]!=1 else ""))
        if p["assists"]:
            routes.append(f"{p['assists']} assist" + ("s" if p["assists"]!=1 else ""))
        if p["clean_sheets"]:
            routes.append("a clean sheet")
        if p["bonus"]:
            routes.append(f"{p['bonus']} bonus point" + ("s" if p["bonus"]!=1 else ""))
        if routes:
            return "the combination of " + vx_review_join(routes)
        if p["minutes"]<60:
            return f"just {p['minutes']} minutes, so the bigger question is really his role and security of starts"
        return "a full enough run-out, but no clean-sheet or attacking return to turn those minutes into points"

    routes=[]
    if p["goals"]:
        routes.append(f"{p['goals']} goal" + ("s" if p["goals"]!=1 else ""))
    if p["assists"]:
        routes.append(f"{p['assists']} assist" + ("s" if p["assists"]!=1 else ""))
    if p["bonus"]:
        routes.append(f"{p['bonus']} bonus point" + ("s" if p["bonus"]!=1 else ""))
    if routes:
        return "the combination of " + vx_review_join(routes)

    if p["minutes"]<60:
        return f"the {p['minutes']}-minute appearance, which matters more to me than the blank because it may point to a change in role"
    if p["gw_xgi"]>0:
        return f"a blank on the scoresheet, but {p['gw_xgi']:.2f} expected goal involvement tells us the attacking opportunity was still there"
    return "no attacking return, although the playing time itself was still encouraging"


def vx_review_attacking_process(p):
    xg=float(p.get("gw_xg",0) or 0)
    xa=float(p.get("gw_xa",0) or 0)

    if xg>0 and xa>0:
        return f"he produced {xg:.2f} expected goals and {xa:.2f} expected assists, so he was involved both as a finisher and as a creator"

    if xg>0:
        return f"he posted {xg:.2f} expected goals, which tells me there was genuine scoring threat behind the performance"

    if xa>0:
        return f"he generated {xa:.2f} expected assists, so the creativity was there even if the Fantasy return did not follow"

    if p["minutes"]>=60:
        return f"he played {p['minutes']} minutes, but the underlying numbers show very little attacking threat"

    return f"he was limited to {p['minutes']} minutes, so the lack of opportunity is more important here than the attacking numbers in isolation"


def vx_review_player_text(p, position_index=0, bench=False):
    name=str(p["name"])
    team=vx_narration_team(p.get("team_name") or p.get("team"))
    points=int(p["points"])
    minutes=int(p["minutes"])
    detail=vx_review_return_detail(p)
    next_match=vx_review_next_match(p)
    action=vx_review_action(p)
    action_phrase=vx_review_action_phrase(action)
    defensive=p["pos"] in ("GK","DEF")
    model_line=vx_review_model_line(p,defensive=defensive)

    if bench:
        templates=[
            (
                f"First off the bench, I have {name} from {team}. He finished the Gameweek with {points} points. "
                f"I am not too concerned with the raw score in this spot; what matters more is {detail}. "
                f"{vx_review_sentence_case(next_match)}. {model_line} "
                f"So going into the next deadline, I still see him as {action_phrase}. For a bench player, I mainly want dependable minutes and a role I can trust when he is called on."
            ),
            (
                f"Next on my bench is {name}. He gave me {points} points from {minutes} minutes. "
                f"The useful part of that performance is {detail}. {vx_review_sentence_case(next_match)}. "
                f"{model_line} That keeps him as {action_phrase} for me. I do not need fireworks from this position; I need reliable cover and enough minutes to avoid carrying a dead squad spot."
            ),
            (
                f"The third substitute is {name}, who ended the week on {points} points. "
                f"Rather than reacting to that total, I am looking at {detail}. {vx_review_sentence_case(next_match)}. "
                f"{model_line} On that evidence, he is {action_phrase}. His job in the squad is to give me usable depth, so consistency of starts matters more than one isolated return or blank."
            ),
            (
                f"And rounding out the squad is {name}, with {points} points this Gameweek. "
                f"For this final bench slot, the first thing I care about is whether the minutes and role are dependable, and this week the key detail was {detail}. "
                f"{vx_review_sentence_case(next_match)}. {model_line} Put together, that leaves him {action_phrase}. I am happy to keep the decision simple unless the role or minutes start to deteriorate."
            ),
        ]

        return re.sub(r"\s+"," ",templates[position_index % len(templates)]).strip()

    if p["pos"] == "GK":
        text=(
            f"Starting in goal, I have {name} from {team}, and he finished the Gameweek on {points} points. "
            f"The score is only the headline; the part I actually care about is {detail}. "
            f"{vx_review_sentence_case(next_match)}. {model_line} "
            f"Taking the latest return, the fixture and the projection together, I have him {action_phrase}. "
            "With goalkeepers, I really do not want to overreact to one week. I am looking for secure minutes, a healthy clean-sheet probability, "
            "and a fixture run that gives him enough save or clean-sheet upside before I even think about using a transfer there."
        )

    elif p["pos"]=="DEF":
        openers=[
            f"Moving across the back line, the first defender is {name} from {team}, who finished with {points} points.",
            f"Next in defence is {name}, and he came away with {points} points this week.",
            f"The third defender is {name}, who closed the Gameweek on {points} points.",
            f"Then I have {name} in the back line, with {points} points from this round.",
            f"And completing the defence is {name}, who returned {points} points.",
        ]
        middle=[
            f"The score itself is useful, but the more important part is how he got there: {detail}.",
            f"I do not want to judge the pick from the points alone. The part of the performance I am taking forward is {detail}.",
            f"For me, the return needs context, and the clearest piece of context this week is {detail}.",
            f"What I care about here is not just the final number; it is the route to those points, which was {detail}.",
            f"I am looking past the headline score and focusing on the repeatable part of the performance: {detail}.",
        ][position_index % 5]
        endings=[
            f"{vx_review_sentence_case(next_match)}. {model_line} That leaves him {action_phrase}. From here, I want the clean-sheet chance, expected minutes and overall role to keep supporting that position.",
            f"{vx_review_sentence_case(next_match)}. {model_line} On balance, I have him {action_phrase}. For a defender, reliable minutes and a good defensive fixture are still the foundation before I start chasing attacking upside.",
            f"{vx_review_sentence_case(next_match)}. {model_line} My current call is {action_phrase}. I want to see a stable starting role and a believable route to clean sheets before one good or bad Gameweek changes my view.",
            f"{vx_review_sentence_case(next_match)}. {model_line} So for now, he is {action_phrase}. The next check is whether the fixture gives him enough defensive floor to make this spot worth keeping.",
            f"{vx_review_sentence_case(next_match)}. {model_line} I am keeping him {action_phrase}. One result is not enough to force a move when the bigger decision still comes down to minutes, fixture quality and repeatable routes to points.",
        ][position_index % 5]
        text=f"{openers[position_index % len(openers)]} {middle} {endings}"

    elif p["pos"]=="MID":
        process=vx_review_attacking_process(p)
        openers=[
            f"Now into midfield, starting with {name} from {team}. He finished the Gameweek with {points} points.",
            f"The next midfielder is {name}, who came away with {points} points.",
            f"In the next midfield slot I have {name}, ending the week on {points} points.",
            f"Then there is {name}, whose final return was {points} points.",
            f"And completing the midfield is {name}, with {points} points on the board.",
        ]
        bridges=[
            f"That return was shaped by {detail}, but I am much more interested in whether the performance underneath it is repeatable. In the attacking numbers, {process}.",
            f"The points tell us what happened, but not necessarily what happens next. We had {detail}, and underneath that, {process}. That is much better evidence for the next decision.",
            f"This week gave us {detail}. The bigger question is whether the attacking opportunity was still there, and the underlying data says {process}. That is the signal I trust more than one Gameweek score.",
            f"I am separating the result from the process here. The result was influenced by {detail}; underneath it, {process}. That helps me judge whether this is sustainable or simply a one-week outcome.",
            f"The visible return came from {detail}, but I want to go one layer deeper before making any transfer call. The underlying numbers show that {process}, which gives a much clearer picture of his attacking role.",
        ]
        endings=[
            f"{vx_review_sentence_case(next_match)}. {model_line} Put the fixture, expected minutes and attacking involvement together, and I have him {action_phrase}.",
            f"{vx_review_sentence_case(next_match)}. {model_line} That combination leaves him {action_phrase} for me. I would rather follow the role and the quality of the chances than chase the latest points total.",
            f"{vx_review_sentence_case(next_match)}. {model_line} Based on the opportunity and what the next fixture offers, he is {action_phrase} going into the deadline.",
            f"{vx_review_sentence_case(next_match)}. {model_line} Right now, the evidence has him {action_phrase}. What matters next is whether those attacking opportunities continue, not whether the previous score happened to be good or bad.",
            f"{vx_review_sentence_case(next_match)}. {model_line} For me, he stays {action_phrase}. I am weighing expected involvement, minutes and fixture together rather than letting one return or one blank make the decision for me.",
        ]
        idx=position_index % 5
        text=f"{openers[idx]} {bridges[idx]} {endings[idx]}"

    else:  # FWD
        process=vx_review_attacking_process(p)
        openers=[
            f"Up front, the first forward is {name} from {team}, and he finished with {points} points.",
            f"The next forward is {name}, who returned {points} points this Gameweek.",
            f"And completing the forward line is {name}, ending the week on {points} points.",
        ]
        bodies=[
            f"The return was influenced by {detail}, but with a forward I really want to know whether he was getting the chances we paid for. Looking underneath the score, {process}. That gives me a much better read on whether more returns should follow.",
            f"The result itself was shaped by {detail}. More importantly for the next decision, {process}. That underlying involvement is what I want to weigh against his price, expected minutes and the opponent coming next.",
            f"The main story from the Gameweek was {detail}. But I do not want to make a forward decision from one score alone. In the attacking process, {process}. That tells me far more about whether he is still getting the kind of opportunities I want to invest in.",
        ]
        endings=[
            f"{vx_review_sentence_case(next_match)}. {model_line} With the chance quality, role and fixture all considered together, he remains {action_phrase}.",
            f"{vx_review_sentence_case(next_match)}. {model_line} For now, that makes him {action_phrase}. I would only start moving away if the underlying opportunity, minutes or fixture outlook begins to slide.",
            f"{vx_review_sentence_case(next_match)}. {model_line} My current read is {action_phrase}. The call comes from the combination of opportunity and future expectation, not from treating one return or one blank as the whole story.",
        ]
        idx=position_index % 3
        text=f"{openers[idx]} {bodies[idx]} {endings[idx]}"

    if p.get("captain"):
        text += (
            " He also had the captain's armband, so his result naturally had a much bigger effect on my Gameweek. "
            "Even then, I want to separate the quality of the captaincy decision from the final points. "
            "The real question is whether the fixture, expected minutes and attacking opportunity made him the right player to back before the deadline."
        )
    elif p.get("vice"):
        text += (
            " He was also my vice-captain. That does not change the way I judge his individual performance, "
            "but it still matters as protection in case the captain unexpectedly misses out."
        )

    return re.sub(r"\s+"," ",text).strip()


# ------------------------------------------------------------
# BUILD THE ONE GW-REVIEW SCRIPT
# ------------------------------------------------------------

segments=[{
    "id":"headline",
    "text":(
        f"Gameweek {REVIEW_GW} is done, so before I start looking at transfers and the fixtures ahead, I want to go through this team properly and see what actually happened. "
        "I am not just reading out the points here. I want to separate the result from the performance, look at the minutes and underlying numbers, and work out what this Gameweek is really telling me about the squad."
    )
}]

_position_seen={"GK":0,"DEF":0,"MID":0,"FWD":0}
for p in review_starters:
    idx=_position_seen.get(p["pos"],0)
    _position_seen[p["pos"]]=idx+1
    segments.append({
        "id":f'player_{p["id"]}',
        "player_id":p["id"],
        "name":p["name"],
        "kind":"starter",
        "text":vx_review_player_text(p,idx,bench=False),
    })

segments.append({
    "id":"bench_intro",
    "text":(
        "Now let me check the bench, because this part of the squad has a different job. "
        "I am not expecting huge hauls every week; I want low-cost players who get reliable minutes and can step in without creating another problem when the starting eleven needs cover."
    )
})

for idx,p in enumerate(review_bench):
    segments.append({
        "id":f'player_{p["id"]}',
        "player_id":p["id"],
        "name":p["name"],
        "kind":"bench",
        "text":vx_review_player_text(p,idx,bench=True),
    })


# ------------------------------------------------------------
# FULL DATA-BOARD REVIEW — WHOLE BOARD APPEARS TOGETHER
# ------------------------------------------------------------

captain_player=next(
    (p for p in review_players if p.get("captain")),
    review_starters[0] if review_starters else review_players[0]
)
captain_return=int(captain_player["points"])*int(captain_player.get("multiplier",1) or 1)
highest_player=max(review_players,key=lambda p:(int(p["points"]),int(p["minutes"])))

_disappointment_pool=[
    p for p in review_starters
    if int(p.get("minutes",0) or 0) >= 60
]
if not _disappointment_pool:
    _disappointment_pool=list(review_starters)
disappointment=min(
    _disappointment_pool,
    key=lambda p:(int(p["points"]),-float(p.get("gw_xgi",0) or 0),-int(p.get("minutes",0) or 0))
)

_attackers=[p for p in review_starters if p["pos"] in ("MID","FWD")]
attack_xg=sum(float(p.get("gw_xg",0) or 0) for p in _attackers)
attack_xa=sum(float(p.get("gw_xa",0) or 0) for p in _attackers)

_defenders=[p for p in review_starters if p["pos"] in ("GK","DEF")]
def_clean_returns=sum(int(p.get("clean_sheets",0) or 0) for p in _defenders)
gk_saves=sum(int(p.get("saves",0) or 0) for p in _defenders if p["pos"]=="GK")
bench_points=sum(int(p["points"]) for p in review_bench)

_dis_reason=(
    f"{disappointment['gw_xgi']:.2f} expected goal involvement"
    if disappointment["pos"] in ("MID","FWD") and float(disappointment.get("gw_xgi",0) or 0)>0
    else f"{disappointment['minutes']} secure minutes without enough Fantasy Premier League return"
    if int(disappointment.get("minutes",0) or 0)>=60
    else f"only {disappointment['minutes']} minutes"
)

# Existing Phase-4 transfer shortlist is used only as decision context.
_transfer_rows=list(_review_decisions.get("transfer_shortlist") or [])
_transfer_rows=[
    r for r in _transfer_rows
    if float(r.get("xpts_gain_5gw",0) or 0) > 0
]
if _transfer_rows:
    _best_transfer=max(_transfer_rows,key=lambda r:float(r.get("xpts_gain_5gw",0) or 0))
    transfer_priority=(
        f"the strongest five-Gameweek upgrade showing in the model is "
        f"{_best_transfer.get('out_name')} to {_best_transfer.get('in_name')}, so that is the move I want to keep on the shortlist"
    )
else:
    transfer_priority="the model is not giving me a positive five-Gameweek transfer to force right now, so I am happy to wait for more information"

_risk_candidates=[
    p for p in review_players
    if (p.get("model_projection") or {}).get("expected_minutes_next_gw") is not None
]
if _risk_candidates:
    risk_player=min(
        _risk_candidates,
        key=lambda p:(
            float((p.get("model_projection") or {}).get("p_start_next_gw",1) or 0),
            float((p.get("model_projection") or {}).get("expected_minutes_next_gw",90) or 0),
        )
    )
    risk_text=f"the selection news around {risk_player['name']}"
else:
    risk_text="the late team news before the deadline"

if bench_points >= 10:
    lesson="there were meaningful points left on the bench, so I need to look more carefully at the start-versus-bench calls next time"
elif captain_return <= 4 and highest_player["id"] != captain_player["id"]:
    lesson="the captaincy limited the ceiling this week, but the answer is to improve the decision process rather than simply chase whoever scored last time"
elif sum(1 for p in review_starters if int(p["minutes"])<60) >= 2:
    lesson="minutes security is the first issue I need to solve before I start chasing fixture upside"
elif def_clean_returns == 0:
    lesson="the defence gave me no clean-sheet return, so the upcoming fixtures and clean-sheet model now have to justify keeping the same structure"
else:
    lesson="the strongest signals came from secure minutes and repeatable involvement rather than from the points total on its own"

segments += [
    {
        "id":"metrics_intro",
        "text":(
            "Now I have the full Gameweek board in front of me, and this is where I want to zoom out from the individual players. "
            "The whole squad stays on screen while I pull together the numbers that actually matter for the next decision."
        )
    },
    {
        "id":"metric_points",
        "text":(
            f"The team finished on {gw_points} points, and the biggest individual return came from "
            f"{highest_player['name']} with {highest_player['points']} points. That gives me the headline result, but I still want to know how much of it came from repeatable performance rather than one-off events."
        )
    },
    {
        "id":"metric_captain",
        "text":(
            f"My captain was {captain_player['name']}, and after the multiplier he contributed {captain_return} points. "
            "That is the outcome, but I do not want to judge captaincy only after I know the result. The better test is whether the minutes, role, fixture and attacking opportunity made him the right pick before the deadline."
        )
    },
    {
        "id":"metric_disappointment",
        "text":(
            f"The player who hurt the most was {disappointment['name']}, with {disappointment['points']} points despite {_dis_reason}. "
            "That is exactly the kind of situation where I need context. A bad Fantasy return does not automatically mean the pick became bad, and it definitely does not automatically mean I need a transfer."
        )
    },
    {
        "id":"metric_attack",
        "text":(
            f"Across my starting midfielders and forwards, the attack generated {attack_xg:.2f} expected goals and {attack_xa:.2f} expected assists. "
            "That is a much cleaner way to judge the attacking process because it tells me whether this team was actually creating enough scoring and assist opportunity, even if the final Fantasy Premier League points did not fully reflect it."
        )
    },
    {
        "id":"metric_defence",
        "text":(
            f"Defensively, the starting goalkeeper and defenders produced {def_clean_returns} clean-sheet returns"
            + (f" and the goalkeeper added {gk_saves} saves." if gk_saves else ".")
            + " That gives me the baseline. From here, I can compare what just happened with the clean-sheet probabilities and defensive fixtures coming next."
        )
    },
    {
        "id":"metric_bench",
        "text":(
            f"There were {bench_points} points sitting on the bench. I am treating that as information about squad depth and my selection calls, not as a reason to chase whichever substitute happened to score last week."
        )
    },
    {
        "id":"metric_bank",
        "text":(
            f"Financially, I have {bank:.1f} million in the bank and the squad value is {value:.1f} million. "
            "That tells me how much flexibility I actually have. The money is there to solve a problem or improve the team, not to create a transfer just because there is cash available."
        )
    },
    {
        "id":"metric_gwrank",
        "text":(
            f"The Gameweek rank came in at {gw_rank:,}. I will take that as part of the result, but it is not going to drive the next move on its own."
            if gw_rank else
            "The Gameweek rank is still updating, so I am not going to use an incomplete number to judge the week."
        )
    }
]

if cur_rank:
    if rank_dir=="up":
        rank_text=(
            f"Overall rank improved to {cur_rank:,}, so that is a green arrow. I will take it, but the arrow itself does not change the process for the next transfer; I still want the move that improves the team going forward."
        )
    elif rank_dir=="down":
        rank_text=(
            f"Overall rank moved back to {cur_rank:,}, so this one is a red arrow. That is the result, but I do not want to respond emotionally. The next decision still needs to come from the repeatable data."
        )
    else:
        rank_text=(
            f"Overall rank stayed at {cur_rank:,}. With no meaningful movement either way, there is even less reason to manufacture a transfer just to feel like I have done something."
        )
    segments.append({"id":"metric_orank","text":rank_text})

segments.append({
    "id":"metric_hit",
    "text":(
        "I did not spend any transfer points this week, so the final score came without a hit. That makes the Gameweek easier to judge because there is no points deduction to separate from the squad's actual return."
        if transfer_cost==0 else
        f"I also spent {transfer_cost} points on transfers this week. I need to keep that separate from the raw squad score, because the team performance and the cost of getting to that team are two different parts of the decision."
    )
})

segments.append({
    "id":"review_close",
    "text":(
        f"So the main lesson I am taking from this Gameweek is that {lesson}. "
        f"Looking toward the next deadline, {transfer_priority}, and the main thing I am monitoring is {risk_text}. "
        "That is the review done. I have taken what is useful from the last Gameweek, and now I can move on to the fixtures, projections and decisions ahead without chasing last week's points."
    )
})
# ------------------------------------------------------------
# NARRATION QA — NATURAL VARIETY + ONE SEGMENT PER PLAYER
# ------------------------------------------------------------

_player_segments=[s for s in segments if s.get("player_id")]
_expected_ids=[int(p["id"]) for p in review_starters+review_bench]
_segment_ids=[int(s["player_id"]) for s in _player_segments]

if _segment_ids != _expected_ids:
    raise RuntimeError(
        "GW Review narration/player order mismatch. "
        f"Expected {_expected_ids}, got {_segment_ids}"
    )

for s in _player_segments:
    if str(s["name"]).lower() not in str(s["text"]).lower():
        raise RuntimeError(f"Player name missing from narration segment: {s['id']}")

_review_norm=[re.sub(r"\s+"," ",str(s["text"]).strip().lower()) for s in segments]
if len(_review_norm) != len(set(_review_norm)):
    raise RuntimeError("GW Review narration contains an exact repeated segment.")

# Catch the old robotic repeated constructions if they accidentally return.
_old_robotic_patterns=[
    r"\bnext is\b.*\bnext is\b",
    r"\bthe score was built on\b",
    r"\bthe useful evidence\b",
]
for pattern in _old_robotic_patterns:
    if sum(bool(re.search(pattern,s["text"].lower())) for s in _player_segments) > 2:
        raise RuntimeError(f"GW Review narration is becoming repetitive: {pattern}")

# EDGE-TTS — ONE SEGMENT AT A TIME = EXACT ANIMATION WINDOWS
# ------------------------------------------------------------

for _required_tts_name in ("VOICE", "MOODS", "vx_tts_profile", "vx_edge_tts_stream", "vx_full_word_narration", "vx_full_word_display", "vx_narration_team", "vx_narration_player"):
    if _required_tts_name not in globals():
        raise RuntimeError(
            "Run SHARED PRODUCTION HELPERS before narration — "
            f"missing {_required_tts_name}"
        )

def review_tts_mood(seg):
    sid = str(seg.get("id", ""))

    if sid == "headline":
        return "intro"

    if sid in {"metric_points", "metric_bank", "metric_gwrank", "metric_orank"}:
        return "key_stat"

    if sid == "metric_hit" and transfer_cost > 0:
        return "warning"

    if sid == "review_close":
        return "outro"

    return "analysis"


async def make_segment(seg):

    path=AUDIO/f'{seg["id"]}.mp3'
    audio=bytearray()
    words=[]

    mood = review_tts_mood(seg)
    profile = vx_tts_profile(mood)
    seg["tts_mood"] = mood
    seg["tts_profile"] = profile

    # CELL 22 is the final authority: no shorthand reaches Ryan or the
    # narration-facing display/package text.
    seg["text"] = vx_full_word_narration(
        seg["text"],
        context=f"GW Review segment {seg['id']}",
    )
    seg["display_text"] = vx_full_word_display(
        seg["text"],
        context=f"GW Review display {seg['id']}",
    )

    audio_bytes, words = await vx_edge_tts_stream(seg["text"], profile)
    audio.extend(audio_bytes)

    path.write_bytes(audio)

    seg["audio"]=str(path)
    seg["duration"]=vx_media_duration(path)
    seg["words"]=words

    # Exact moment Ryan says the player's canonical FPL web_name.
    #
    # Edge TTS does not guarantee that one written word == one WordBoundary
    # chunk. A name can arrive as one boundary ("Raya") or several adjacent
    # boundaries ("Ray", "a"). Match an EXACT normalized span of adjacent
    # Ryan boundaries; never fall back to segment start or an estimated time.
    seg["name_time"]=None

    if seg.get("name"):
        canonical_name = vx_narration_player(
            seg.get("player_id") if seg.get("player_id") is not None else seg["name"]
        )
        canonical_name = str(canonical_name or seg["name"]).strip()
        seg["name"] = canonical_name

        def _review_boundary_key(value):
            value = unicodedata.normalize("NFKD", str(value or "").casefold())
            value = "".join(ch for ch in value if not unicodedata.combining(ch))
            return "".join(ch for ch in value if ch.isalnum())

        target_key = _review_boundary_key(canonical_name)
        if not target_key:
            raise RuntimeError(
                f"Canonical player web name is empty for {seg['id']}; "
                "Ryan animation sync cannot be validated."
            )

        boundary_keys = [_review_boundary_key(w.get("text", "")) for w in words]

        # Search only adjacent spoken boundaries and accept only an exact
        # concatenated identity match. This handles TTS token splitting while
        # keeping timing sourced from Ryan's real WordBoundary timestamp.
        for j in range(len(boundary_keys)):
            combined = ""
            for k in range(j, min(len(boundary_keys), j + 8)):
                part = boundary_keys[k]
                if not part:
                    continue
                combined += part

                if combined == target_key:
                    seg["name_time"] = float(words[j]["start"])
                    seg["name_boundary_span"] = [int(j), int(k)]
                    break

                if len(combined) >= len(target_key):
                    break

            if seg["name_time"] is not None:
                break

        if seg["name_time"] is None:
            boundary_preview = [
                str(w.get("text", ""))
                for w in words[:16]
            ]
            raise RuntimeError(
                f"Ryan name boundary not found for canonical web name "
                f"{canonical_name!r} in {seg['id']}. "
                f"Received {len(words)} WordBoundary events; first boundaries: "
                f"{boundary_preview}. Animation sync is blocked rather than guessed."
            )
    else:
        seg["name_time"]=0.0

for seg in segments:
    await make_segment(seg)

# ------------------------------------------------------------
# SAVE REVIEW PACKAGE
# ------------------------------------------------------------

review_package={
    "gw":REVIEW_GW,
    "players":review_players,
    "starters":review_starters,
    "bench":review_bench,
    "metrics":{
        "gw_points":gw_points,
        "bank":bank,
        "value":value,
        "gw_rank":gw_rank,
        "overall_rank":cur_rank,
        "previous_rank":prev_rank,
        "rank_dir":rank_dir,
        "transfer_cost":transfer_cost
    },
    "segments":segments
}

# Canonical in-memory handoff for downstream review cells.
# If the DATA artifact disappears later in the same runtime,
# Cell 16 can restore this exact completed package without rebuilding it.
GW_REVIEW_PACKAGE = review_package

OUT=TIMING/"gw_review_package.json"

OUT.write_text(
    json.dumps(review_package,indent=2),
    encoding="utf-8"
)

print(f"✅ GW {REVIEW_GW} review data loaded")
print(f"✅ Starting XI: {len(review_starters)}")
print(f"✅ Bench: {len(review_bench)}")
print(f"✅ Ryan narration segments: {len(segments)}")
print("✅ Ryan TTS moods: shared VOICE + MOODS profile")
print(f"✅ Rank direction: {rank_dir.upper()}")
print("✅ Saved:",OUT)
```

```python
# ============================================================
# CELL 15A — PLAYER CARD DESIGN
# HERO / LEFT-SIDE PLAYER CARD ONLY
# ============================================================
# IMPORTANT:
# - This is the ONE authoritative HERO / PLAYER CARD design.
# - It does NOT style the GW soccer-pitch player tiles/cards.
# - It contains presentation only: HTML + CSS + responsive JS.
# - Player/FDR/stat calculations remain in the existing notebook cells.
# ============================================================

PLAYER_CARD_DESIGN_VERSION = "V1.1_AUTO_FIT"
PLAYER_CARD_RATIO = 0.66  # width / height — master design ratio

PLAYER_CARD_CSS = r"""
<!-- VX_PLAYER_CARD_DESIGN_START -->
<style id="vx-player-card-design">

/* ==========================================================
   FPL VORTEX — HERO / PLAYER CARD
   ONE responsive component. Internal scale follows card size.
   ========================================================== */

.vx-player-card{
    --pc-bg0:#04070d;
    --pc-bg1:#07101a;
    --pc-bg2:#0a1420;
    --pc-gold:#e5b94b;
    --pc-gold-bright:#ffd86a;
    --pc-gold-muted:rgba(229,185,75,.52);
    --pc-gold-grid:rgba(229,185,75,.34);
    --pc-gold-faint:rgba(229,185,75,.20);
    --pc-red:#e31937;
    --pc-red-dark:#7d0a18;
    --pc-cyan:#28beea;
    --pc-white:#f4f6f8;
    --pc-muted:#aab3bf;
    --pc-purple:#8758e8;

    --pc-pad-x:3.0cqw;
    --pc-pad-y:1.75cqh;
    --pc-gap-xs:.38cqh;
    --pc-gap-sm:.62cqh;
    --pc-gap-md:.95cqh;
    --pc-radius-sm:.65cqw;
    --pc-radius-md:1.05cqw;
    --pc-line:clamp(1px,.10cqw,3px);
    --pc-line-main:clamp(2px,.17cqw,4px);

    position:relative;
    isolation:isolate;
    container-type:size;
    aspect-ratio:.66 / 1;
    overflow:hidden;

    display:grid;
    grid-template-rows:40.5fr 12.5fr 15fr 18fr 11fr 3fr;

    color:var(--pc-white);
    background:
      radial-gradient(circle at 12% 20%,rgba(227,25,55,.24),transparent 27%),
      radial-gradient(circle at 82% 6%,rgba(255,216,106,.09),transparent 23%),
      linear-gradient(150deg,var(--pc-bg2),var(--pc-bg0) 48%,#020407 100%);

    border:var(--pc-line-main) solid rgba(255,216,106,.90);
    border-radius:2.1cqw;

    box-shadow:
      0 0 1.15cqw rgba(255,196,48,.18),
      0 1.25cqw 2.7cqw rgba(0,0,0,.68),
      inset 0 0 .55cqw rgba(255,216,106,.08);

    font-family:"Arial Narrow","Roboto Condensed","Inter Tight",Arial,sans-serif;
    font-variant-numeric:tabular-nums;
    box-sizing:border-box;
}

.vx-player-card *,
.vx-player-card *::before,
.vx-player-card *::after{box-sizing:border-box;min-width:0;}

.vx-player-card::before{
    content:"";
    position:absolute;
    inset:.75cqw;
    z-index:20;
    pointer-events:none;
    border:var(--pc-line) solid rgba(229,185,75,.48);
    border-radius:1.45cqw;
    clip-path:polygon(2.4cqw 0,calc(100% - 2.4cqw) 0,100% 2.4cqw,100% calc(100% - 2.4cqw),calc(100% - 2.4cqw) 100%,2.4cqw 100%,0 calc(100% - 2.4cqw),0 2.4cqw);
}

.vx-player-card::after{
    content:"";
    position:absolute;
    inset:0;
    z-index:18;
    pointer-events:none;
    opacity:.28;
    background:
      linear-gradient(115deg,transparent 0 19%,rgba(255,255,255,.04) 19.2% 19.6%,transparent 19.8% 58%,rgba(255,216,106,.035) 58.2% 58.6%,transparent 58.8%),
      radial-gradient(circle at 50% 50%,transparent 58%,rgba(0,0,0,.45) 100%);
}

/* ----------------------------------------------------------
   1. IDENTITY / PORTRAIT
   ---------------------------------------------------------- */
.pc-identity{
    position:relative;
    overflow:hidden;
    display:grid;
    grid-template-columns:42.5% 57.5%;
    border-bottom:var(--pc-line-main) solid var(--pc-gold-muted);
    background:
      linear-gradient(90deg,rgba(100,0,10,.25),transparent 44%),
      linear-gradient(165deg,rgba(255,255,255,.035),rgba(0,0,0,.16));
}

.pc-portrait-zone{
    position:relative;
    overflow:hidden;
    min-height:0;
    background:
      radial-gradient(circle at 52% 43%,rgba(226,29,48,.34),transparent 48%),
      linear-gradient(145deg,rgba(114,4,16,.22),transparent 55%);
}

.pc-portrait-zone::before{
    content:"";
    position:absolute;
    inset:0;
    opacity:.62;
    background:
      linear-gradient(135deg,transparent 0 52%,rgba(231,24,50,.35) 52.4% 53.1%,transparent 53.5%),
      linear-gradient(45deg,transparent 0 35%,rgba(231,24,50,.24) 35.4% 36.1%,transparent 36.5%);
}

.pc-portrait{
    position:absolute;
    inset:0;
    width:100%;
    height:100%;
    object-fit:contain;
    object-position:center bottom;
    z-index:3;
    filter:drop-shadow(0 1.0cqw .8cqw rgba(0,0,0,.72));
}

.pc-identity-copy{
    position:relative;
    z-index:4;
    padding:1.35cqh 3.15cqw 1.0cqh 1.7cqw;
    display:flex;
    flex-direction:column;
    justify-content:center;
    overflow:hidden;
}

.pc-topline{
    position:absolute;
    left:1.7cqw;
    right:2.2cqw;
    top:1.05cqh;
    height:3.9cqh;
    display:flex;
    align-items:stretch;
    justify-content:space-between;
    gap:.7cqw;
}

.pc-asset{
    flex:1 1 54%;
    max-width:54%;
    display:flex;
    align-items:center;
    white-space:nowrap;
    overflow:hidden;
    color:#fff1a8;
    font-size:2.35cqw;
    line-height:1;
    font-weight:950;
    letter-spacing:.025em;
    text-shadow:0 .18cqw .32cqw rgba(0,0,0,.78);
}

.pc-position-pill{
    flex:0 1 45%;
    min-width:0;
    display:flex;
    align-items:center;
    justify-content:center;
    padding:.25cqh .65cqw;
    text-align:center;
    white-space:nowrap;
    overflow:hidden;
    border:var(--pc-line) solid rgba(255,216,106,.92);
    border-radius:1.4cqw;
    color:#fff4bd;
    background:linear-gradient(180deg,rgba(68,8,14,.92),rgba(4,7,13,.90));
    font-size:2.45cqw;
    line-height:1;
    font-weight:1000;
    letter-spacing:.025em;
    text-shadow:0 .18cqw .32cqw rgba(0,0,0,.78);
}

.pc-club-badge{
    position:absolute;
    right:2.0cqw;
    top:5.8cqh;
    width:8.4cqw;
    height:8.4cqw;
    object-fit:contain;
    object-position:center;
    z-index:7;
    filter:drop-shadow(0 .45cqw .45cqw rgba(0,0,0,.72));
}

.pc-name-block{
    width:100%;
    height:22.8cqh;
    padding-right:8.0cqw;
    margin-top:4.5cqh;
    display:flex;
    flex-direction:column;
    justify-content:center;
    gap:.28cqh;
    overflow:hidden;
}

.pc-name-line{
    flex:1 1 0;
    display:flex;
    align-items:center;
    width:100%;
    min-height:0;
    white-space:nowrap;
    overflow:hidden;
    text-overflow:clip;
    text-transform:uppercase;
    line-height:1.08;
    letter-spacing:-.035em;
    font-weight:1000;
}

.pc-first-name{
    color:#ffffff;
    font-size:8.4cqw;
    text-shadow:0 .28cqw .42cqw rgba(0,0,0,.72);
}

.pc-surname{
    color:#ffd65f;
    font-size:6.85cqw;
    text-shadow:0 .18cqw .30cqw rgba(0,0,0,.72);
}

.pc-clubline{
    margin-top:.55cqh;
    height:3.25cqh;
    display:flex;
    align-items:center;
    gap:.8cqw;
    white-space:nowrap;
    overflow:hidden;
    font-size:2.25cqw;
    line-height:1;
    font-weight:900;
    letter-spacing:.025em;
}

.pc-pos-short{color:var(--pc-red);font-weight:1000;flex:none;}
.pc-sep{color:var(--pc-gold);flex:none;}
.pc-club-name{overflow:hidden;text-overflow:ellipsis;color:#f6f7f8;}

/* ----------------------------------------------------------
   2. PRIMARY KPI ROW
   ---------------------------------------------------------- */
.pc-kpis{
    position:relative;
    padding:.65cqh 2.0cqw .72cqh;
    display:grid;
    grid-template-columns:repeat(4,minmax(0,1fr));
    gap:.48cqw;
    border-bottom:var(--pc-line-main) solid var(--pc-gold-muted);
    background:linear-gradient(180deg,rgba(2,5,9,.78),rgba(3,8,13,.92));
}

.pc-kpi{
    position:relative;
    overflow:hidden;
    display:grid;
    grid-template-rows:27% 59% 14%;
    align-items:stretch;
    text-align:center;
    border:var(--pc-line) solid var(--pc-gold-grid);
    border-radius:var(--pc-radius-sm);
    background:linear-gradient(180deg,rgba(255,255,255,.045),rgba(255,255,255,.012));
}

.pc-kpi::after{content:"";position:absolute;left:12%;right:12%;bottom:8%;height:var(--pc-line);background:currentColor;opacity:.46;}
.pc-kpi.gold{color:#f0bd24}.pc-kpi.red{color:#ff394d}.pc-kpi.cyan{color:#32c7f2}.pc-kpi.purple{color:#9366ed}
.pc-kpi-label{display:flex;align-items:center;justify-content:center;font-size:1.85cqw;font-weight:950;letter-spacing:.02em;white-space:nowrap;overflow:hidden;}
.pc-kpi-value{display:flex;align-items:center;justify-content:center;color:#fff;font-size:5.35cqw;line-height:.92;font-weight:1000;white-space:nowrap;overflow:hidden;text-shadow:0 .2cqw .34cqw rgba(0,0,0,.66);}
.pc-kpi-mark{display:flex;align-items:center;justify-content:center;font-size:1.4cqw;font-weight:900;opacity:.82;}

/* ----------------------------------------------------------
   3. POSITION-AWARE SEASON STATS
   ---------------------------------------------------------- */
.pc-stats{
    min-height:0;
    display:grid;
    grid-template-rows:23% 77%;
    border-bottom:var(--pc-line-main) solid var(--pc-gold-muted);
    background:linear-gradient(180deg,rgba(1,4,8,.91),rgba(3,7,11,.96));
}

.pc-section-title{
    display:flex;
    align-items:center;
    justify-content:center;
    gap:.9cqw;
    padding:0 2cqw;
    color:var(--pc-gold-bright);
    font-size:2.15cqw;
    line-height:1;
    font-weight:900;
    letter-spacing:.035em;
    white-space:nowrap;
    overflow:hidden;
    text-overflow:ellipsis;
    border-bottom:var(--pc-line) solid var(--pc-gold-faint);
}

.pc-stat-grid{
    min-height:0;
    padding:0 1.55cqw .55cqh;
    display:grid;
    grid-template-columns:repeat(8,minmax(0,1fr));
}

.pc-stat{
    min-width:0;
    min-height:0;
    display:grid;
    grid-template-rows:45% 55%;
    align-items:stretch;
    justify-items:stretch;
    padding:.18cqh .28cqw;
    text-align:center;
    border-right:var(--pc-line) solid var(--pc-gold-grid);
}
.pc-stat:last-child{border-right:0;}
.pc-stat-label{display:flex;align-items:center;justify-content:center;color:#ffd75f;font-size:1.45cqw;line-height:.92;font-weight:950;white-space:normal;overflow:hidden;width:100%;text-align:center;}
.pc-stat-value{display:flex;align-items:center;justify-content:center;margin-top:0;color:#fff;font-size:3.15cqw;line-height:.92;font-weight:1000;white-space:nowrap;overflow:hidden;width:100%;text-shadow:0 .16cqw .28cqw rgba(0,0,0,.62);}

/* ----------------------------------------------------------
   4. NEXT FIVE FDR
   ---------------------------------------------------------- */
.pc-fixtures{
    min-height:0;
    display:grid;
    grid-template-rows:19% 67% 14%;
    border-bottom:var(--pc-line-main) solid var(--pc-gold-muted);
    background:linear-gradient(180deg,rgba(2,6,10,.95),rgba(1,4,7,.98));
}

.pc-fixture-grid{
    min-height:0;
    padding:.55cqh 1.55cqw .5cqh;
    display:grid;
    grid-template-columns:repeat(5,minmax(0,1fr));
    gap:.35cqw;
}

.pc-fixture{
    min-width:0;
    overflow:hidden;
    display:grid;
    grid-template-rows:22% 48% 30%;
    border:var(--pc-line) solid var(--pc-gold-grid);
    border-radius:var(--pc-radius-sm);
    background:linear-gradient(180deg,rgba(255,255,255,.035),rgba(255,255,255,.012));
}

.pc-gw{display:flex;align-items:center;justify-content:center;color:#ffd75f;font-size:1.55cqw;font-weight:950;white-space:nowrap;overflow:hidden;}
.pc-opponent-row{min-height:0;display:grid;grid-template-columns:37% 63%;align-items:center;padding:0 .35cqw;}
.pc-opponent-badges{height:100%;display:flex;align-items:center;justify-content:center;gap:.18cqw;overflow:hidden;}
.pc-opponent-badges img{max-width:4.8cqw;max-height:4.8cqw;width:auto;height:auto;object-fit:contain;object-position:center;filter:drop-shadow(0 .2cqw .2cqw rgba(0,0,0,.62));}
.pc-opptext{min-width:0;text-align:center;line-height:.95;}
.pc-opp{display:flex;align-items:center;justify-content:center;color:#fff;font-size:2.45cqw;font-weight:1000;white-space:nowrap;overflow:hidden;}
.pc-ha{display:flex;align-items:center;justify-content:center;margin-top:.15cqh;color:#ffd75f;font-size:1.55cqw;font-weight:950;white-space:nowrap;overflow:hidden;}
.pc-fdr{margin:.2cqh .55cqw .5cqh;border-radius:.55cqw;display:flex;align-items:center;justify-content:center;font-size:3.15cqw;line-height:.94;font-weight:1000;border:var(--pc-line) solid rgba(255,255,255,.30);overflow:hidden;text-shadow:0 .16cqw .24cqw rgba(0,0,0,.46);}

/* EXACT VORTEX FDR fills — mirrors the verified FDR slide classes. */
.pc-fdr.fdr-1{background:linear-gradient(145deg,#45ff77 0%,#12b94b 100%);border-color:rgba(154,255,178,.96);color:#06131c;text-shadow:0 1px 0 rgba(255,255,255,.58);}
.pc-fdr.fdr-2{background:linear-gradient(145deg,#69d9ff 0%,#188bc9 100%);border-color:rgba(188,237,255,.96);color:#06131c;text-shadow:0 1px 0 rgba(255,255,255,.58);}
.pc-fdr.fdr-3{background:linear-gradient(145deg,#ffe36e 0%,#d99b00 100%);border-color:rgba(255,244,174,.98);color:#06131c;text-shadow:0 1px 0 rgba(255,255,255,.58);}
.pc-fdr.fdr-4,.pc-fdr.fdr-5{background:linear-gradient(145deg,#ff6673 0%,#cf1735 100%);border-color:rgba(255,190,197,.97);color:#fff;text-shadow:0 .18cqw .35cqw rgba(0,0,0,.62);}
.pc-fdr.fdr-none{background:linear-gradient(145deg,#24384b 0%,#122130 100%);border-color:rgba(178,199,216,.28);color:#d6e0e8;}

.pc-fdr-legend{display:grid;grid-template-columns:repeat(4,1fr);align-items:center;margin:0 1.55cqw .35cqh;border-top:var(--pc-line) solid var(--pc-gold-faint);color:#e9eef3;font-size:1.12cqw;font-weight:950;text-align:center;white-space:nowrap;}
.pc-fdr-legend span:nth-child(1){color:#43e970}.pc-fdr-legend span:nth-child(2){color:#6ddcff}.pc-fdr-legend span:nth-child(3){color:#ffd85e}.pc-fdr-legend span:nth-child(4){color:#ff5c6b}

/* ----------------------------------------------------------
   5. VERDICT / STATUS
   ---------------------------------------------------------- */
.pc-verdict{
    min-height:0;
    padding:.7cqh 1.7cqw .6cqh;
    display:grid;
    grid-template-columns:31% 69%;
    grid-template-rows:54% 46%;
    border-bottom:var(--pc-line-main) solid var(--pc-gold-muted);
    background:linear-gradient(135deg,rgba(9,10,10,.98),rgba(2,5,8,.99));
}

.pc-verdict-label{
    min-height:0;
    display:grid;
    grid-template-rows:54% 46%;
    align-items:stretch;
    padding-right:1.15cqw;
    border-right:var(--pc-line) solid var(--pc-gold-grid);
    color:#d8a936;
}
.pc-verdict-kicker{display:flex;align-items:center;font-size:2.05cqw;font-weight:950;white-space:nowrap;overflow:hidden;}
.pc-stars{display:flex;align-items:center;margin-top:0;color:#ffd12e;font-size:2.05cqw;letter-spacing:.08em;white-space:nowrap;overflow:hidden;}
.pc-verdict-main{
    min-height:0;
    display:grid;
    grid-template-rows:58% 42%;
    align-items:stretch;
    padding-left:1.45cqw;
    overflow:hidden;
}
.pc-verdict-title{display:flex;align-items:center;color:#fff;font-size:3.35cqw;line-height:.92;font-weight:1000;white-space:nowrap;overflow:hidden;}
.pc-verdict-sub{display:flex;align-items:center;margin-top:0;color:#f0f3f6;font-size:1.35cqw;line-height:1;font-weight:850;letter-spacing:.01em;white-space:nowrap;overflow:hidden;}
.pc-tags{grid-column:1 / -1;min-height:0;display:grid;grid-template-columns:repeat(5,minmax(0,1fr));gap:.45cqw;align-items:center;padding-top:.45cqh;}
.pc-tag{min-width:0;min-height:0;display:flex;align-items:center;justify-content:center;padding:.32cqh .30cqw;border:var(--pc-line) solid rgba(229,185,75,.72);border-radius:.55cqw;color:#ffd15a;background:rgba(4,7,10,.88);font-size:1.25cqw;line-height:1;font-weight:950;text-align:center;white-space:nowrap;overflow:hidden;}

/* ----------------------------------------------------------
   6. FOOTER
   ---------------------------------------------------------- */
.pc-footer{
    padding:0 2.1cqw;
    display:grid;
    grid-template-columns:1fr 1.5fr 1fr;
    align-items:center;
    gap:1cqw;
    background:linear-gradient(180deg,#050708,#020304);
    font-size:1.05cqw;
    line-height:1;
    font-weight:850;
    letter-spacing:.035em;
    white-space:nowrap;
}
.pc-footer-brand{height:100%;display:flex;align-items:center;color:#ffd15a;font-size:1.35cqw;font-weight:1000;overflow:hidden;}
.pc-footer-copy{height:100%;display:flex;align-items:center;justify-content:center;color:#f0f3f6;text-align:center;overflow:hidden;}
.pc-footer-season{height:100%;display:flex;align-items:center;justify-content:flex-end;color:#f0f3f6;text-align:right;overflow:hidden;}

/* ----------------------------------------------------------
   Small-card priority: remove decorative breathing room before
   sacrificing primary typography. The ratio still never changes.
   ---------------------------------------------------------- */
@container (max-width:700px){
    .vx-player-card{--pc-gap-sm:.45cqh;--pc-gap-md:.72cqh;}
    .pc-identity-copy{padding-right:2.5cqw;}
    .pc-name-block{padding-right:7.3cqw;}
    .pc-kpi-mark{opacity:.60;}
    .pc-verdict-sub{letter-spacing:0;}
}

</style>
<!-- VX_PLAYER_CARD_DESIGN_END -->
"""

PLAYER_CARD_JS = r"""
<script id="vx-player-card-runtime">
(() => {
  const CARD_RATIO = 0.66;

  const esc = (v) => String(v ?? "")
      .replaceAll("&","&amp;")
      .replaceAll("<","&lt;")
      .replaceAll(">","&gt;")
      .replaceAll('"',"&quot;");

  const safeText = (v, fallback="—") => {
      const s = String(v ?? "").trim();
      return s ? s : fallback;
  };

  const statSchemas = {
      GK: [
          ["SAVES","saves"],
          ["CLEAN SHEETS","clean_sheets"],
          ["MINUTES","minutes"],
          ["GOALS CONCEDED","goals_conceded"],
          ["xGC","xgc"],
          ["PEN SAVED","penalties_saved"],
          ["PPG","points_per_game"],
          ["BONUS","bonus"]
      ],
      DEF: [
          ["GOALS","goals"],
          ["ASSISTS","assists"],
          ["CLEAN SHEETS","clean_sheets"],
          ["MINUTES","minutes"],
          ["xG","xg"],
          ["xA","xa"],
          ["xGI","xgi"],
          ["BONUS","bonus"]
      ],
      MID: [
          ["GOALS","goals"],
          ["ASSISTS","assists"],
          ["MINUTES","minutes"],
          ["xG","xg"],
          ["xA","xa"],
          ["xGI","xgi"],
          ["PPG","points_per_game"],
          ["BONUS","bonus"]
      ],
      FWD: [
          ["GOALS","goals"],
          ["ASSISTS","assists"],
          ["MINUTES","minutes"],
          ["xG","xg"],
          ["xA","xa"],
          ["xGI","xgi"],
          ["PPG","points_per_game"],
          ["BONUS","bonus"]
      ]
  };

  function balancedSurnameLines(value){
      const words = safeText(value,"").split(/\s+/).filter(Boolean);
      if(words.length <= 1) return words;
      if(words.length === 2 && words.join(" ").length <= 15) return [words.join(" ")];

      let best = null;
      for(let i=1;i<words.length;i++){
          const left = words.slice(0,i).join(" ");
          const right = words.slice(i).join(" ");
          const score = Math.abs(left.length - right.length);
          if(!best || score < best.score) best = {score,left,right};
      }
      return best ? [best.left,best.right] : [words.join(" ")];
  }

  function statHTML(data){
      const pos = safeText(data.position,"MID").toUpperCase();
      const schema = statSchemas[pos] || statSchemas.MID;
      const stats = data.stats || {};
      return schema.map(([label,key]) => `
        <div class="pc-stat">
          <div class="pc-stat-label" data-pc-fit data-pc-fit-max=".020" data-pc-fit-min=".007" title="${esc(label)}">${esc(label)}</div>
          <div class="pc-stat-value" data-pc-fit data-pc-fit-max=".040" data-pc-fit-min=".014">${esc(safeText(stats[key]))}</div>
        </div>`).join("");
  }

  function fixtureHTML(fx){
      fx = fx || {};
      const badges = Array.isArray(fx.badges) ? fx.badges.filter(Boolean).slice(0,2) : [];
      const badgeHTML = badges.map(src => `<img src="${esc(src)}" alt="">`).join("");
      const fdrClass = /^(fdr-[1-5]|fdr-none)$/.test(String(fx.fdr_class || ""))
          ? String(fx.fdr_class) : "fdr-none";
      return `
        <div class="pc-fixture">
          <div class="pc-gw" data-pc-fit data-pc-fit-max=".021" data-pc-fit-min=".008">GW${esc(safeText(fx.gw,"—"))}</div>
          <div class="pc-opponent-row">
            <div class="pc-opponent-badges">${badgeHTML}</div>
            <div class="pc-opptext">
              <span class="pc-opp" data-pc-fit data-pc-fit-max=".032" data-pc-fit-min=".012">${esc(safeText(fx.opponent,"BLANK"))}</span>
              <span class="pc-ha" data-pc-fit data-pc-fit-max=".020" data-pc-fit-min=".008">${esc(safeText(fx.venue,""))}</span>
            </div>
          </div>
          <div class="pc-fdr ${fdrClass}" data-pc-fit data-pc-fit-max=".043" data-pc-fit-min=".015">${esc(safeText(fx.fdr,"—"))}</div>
        </div>`;
  }

  function renderPlayerCard(playerData, options={}){
      const p = playerData || {};
      const first = safeText(p.first_name, safeText(p.name, "PLAYER"));
      const surname = safeText(p.surname, "");
      const surnameLines = balancedSurnameLines(surname);
      const positionLabel = safeText(p.position_label, safeText(p.position,"PLAYER"));
      const fixtures = (Array.isArray(p.fixtures) ? p.fixtures : []).slice(0,5);
      while(fixtures.length < 5) fixtures.push({});
      const tags = (Array.isArray(p.tags) ? p.tags : []).slice(0,5);
      while(tags.length < 5) tags.push("");

      const crest = p.club_badge
          ? `<img class="pc-club-badge" src="${esc(p.club_badge)}" alt="">` : "";
      const portrait = p.player_image
          ? `<img class="pc-portrait" src="${esc(p.player_image)}" alt="">` : "";

      return `
      <article class="vx-player-card" data-player-id="${esc(p.id ?? "")}" aria-label="${esc(first + " " + surname)} player card">
        <section class="pc-identity">
          <div class="pc-portrait-zone">${portrait}</div>
          <div class="pc-identity-copy">
            <div class="pc-topline">
              <div class="pc-asset" data-pc-fit data-pc-fit-max=".032" data-pc-fit-min=".011">♛ ${esc(safeText(p.asset_label,"PLAYER REVIEW"))}</div>
              <div class="pc-position-pill" data-pc-fit data-pc-fit-max=".035" data-pc-fit-min=".012">${esc(positionLabel)}</div>
            </div>
            ${crest}
            <div class="pc-name-block">
              <span class="pc-name-line pc-first-name" data-pc-fit data-pc-fit-max=".095" data-pc-fit-min=".038">${esc(first)}</span>
              ${surnameLines.map(line => `<span class="pc-name-line pc-surname" data-pc-fit data-pc-fit-max=".078" data-pc-fit-min=".030">${esc(line)}</span>`).join("")}
            </div>
            <div class="pc-clubline">
              <span class="pc-pos-short" data-pc-fit data-pc-fit-max=".030" data-pc-fit-min=".010">${esc(safeText(p.position,""))}</span>
              <span class="pc-sep">|</span>
              <span class="pc-club-name" data-pc-fit data-pc-fit-max=".030" data-pc-fit-min=".010">${esc(safeText(p.club,""))}</span>
            </div>
          </div>
        </section>

        <section class="pc-kpis">
          <div class="pc-kpi gold"><div class="pc-kpi-label" data-pc-fit data-pc-fit-max=".026" data-pc-fit-min=".009">PRICE</div><div class="pc-kpi-value" data-pc-fit data-pc-fit-max=".068" data-pc-fit-min=".022">${esc(safeText(p.price))}</div><div class="pc-kpi-mark" data-pc-fit data-pc-fit-max=".019" data-pc-fit-min=".007">▰</div></div>
          <div class="pc-kpi red"><div class="pc-kpi-label" data-pc-fit data-pc-fit-max=".026" data-pc-fit-min=".009">TOTAL PTS</div><div class="pc-kpi-value" data-pc-fit data-pc-fit-max=".068" data-pc-fit-min=".022">${esc(safeText(p.total_points))}</div><div class="pc-kpi-mark" data-pc-fit data-pc-fit-max=".019" data-pc-fit-min=".007">↗</div></div>
          <div class="pc-kpi cyan"><div class="pc-kpi-label" data-pc-fit data-pc-fit-max=".026" data-pc-fit-min=".009">FORM</div><div class="pc-kpi-value" data-pc-fit data-pc-fit-max=".068" data-pc-fit-min=".022">${esc(safeText(p.form))}</div><div class="pc-kpi-mark" data-pc-fit data-pc-fit-max=".019" data-pc-fit-min=".007">⌁</div></div>
          <div class="pc-kpi purple"><div class="pc-kpi-label" data-pc-fit data-pc-fit-max=".026" data-pc-fit-min=".009">OWNERSHIP</div><div class="pc-kpi-value" data-pc-fit data-pc-fit-max=".068" data-pc-fit-min=".022">${esc(safeText(p.ownership))}</div><div class="pc-kpi-mark" data-pc-fit data-pc-fit-max=".019" data-pc-fit-min=".007">●●</div></div>
        </section>

        <section class="pc-stats">
          <div class="pc-section-title" data-pc-fit data-pc-fit-max=".030" data-pc-fit-min=".010">${esc(safeText(p.stats_heading, positionLabel + " STATS"))}</div>
          <div class="pc-stat-grid">${statHTML(p)}</div>
        </section>

        <section class="pc-fixtures">
          <div class="pc-section-title" data-pc-fit data-pc-fit-max=".030" data-pc-fit-min=".010">NEXT 5 FIXTURES <span>|</span> FDR</div>
          <div class="pc-fixture-grid">${fixtures.map(fixtureHTML).join("")}</div>
          <div class="pc-fdr-legend">
            <span data-pc-fit data-pc-fit-max=".016" data-pc-fit-min=".006">1–2 EASY</span>
            <span data-pc-fit data-pc-fit-max=".016" data-pc-fit-min=".006">3 MEDIUM</span>
            <span data-pc-fit data-pc-fit-max=".016" data-pc-fit-min=".006">4 HARD</span>
            <span data-pc-fit data-pc-fit-max=".016" data-pc-fit-min=".006">5 TOUGH</span>
          </div>
        </section>

        <section class="pc-verdict">
          <div class="pc-verdict-label">
            <div class="pc-verdict-kicker" data-pc-fit data-pc-fit-max=".029" data-pc-fit-min=".010">FPL VERDICT</div>
            <div class="pc-stars" data-pc-fit data-pc-fit-max=".028" data-pc-fit-min=".010">★★★★★</div>
          </div>
          <div class="pc-verdict-main">
            <div class="pc-verdict-title" data-pc-fit data-pc-fit-max=".046" data-pc-fit-min=".016">${esc(safeText(p.verdict_main,"PLAYER REVIEW"))}</div>
            <div class="pc-verdict-sub" data-pc-fit data-pc-fit-max=".021" data-pc-fit-min=".008">${esc(safeText(p.verdict_sub,"DATA DRIVEN • POINTS DELIVERED"))}</div>
          </div>
          <div class="pc-tags">${tags.map(t => `<div class="pc-tag" data-pc-fit data-pc-fit-max=".019" data-pc-fit-min=".007" title="${esc(t)}">${esc(t || "•")}</div>`).join("")}</div>
        </section>

        <footer class="pc-footer">
          <div class="pc-footer-brand" data-pc-fit data-pc-fit-max=".019" data-pc-fit-min=".007">FPL VORTEX</div>
          <div class="pc-footer-copy" data-pc-fit data-pc-fit-max=".016" data-pc-fit-min=".006">DATA DRIVEN. POINTS DELIVERED.</div>
          <div class="pc-footer-season" data-pc-fit data-pc-fit-max=".016" data-pc-fit-min=".006">${esc(safeText(p.season_label,"SEASON"))}</div>
        </footer>
      </article>`;
  }

  function elementFits(el){
      return (
          el.scrollWidth <= el.clientWidth + 1 &&
          el.scrollHeight <= el.clientHeight + 1
      );
  }

  function fitElement(card, el){
      if(!card || !el) return;
      const cardW = Math.max(1, card.getBoundingClientRect().width);
      const minRatio = Number(el.dataset.pcFitMin || ".008");
      const maxRatio = Number(el.dataset.pcFitMax || ".050");
      const minPx = Math.max(6, cardW * minRatio);
      const maxPx = Math.max(minPx, cardW * maxRatio);

      let low = minPx;
      let high = maxPx;
      let best = minPx;

      el.style.fontSize = `${maxPx}px`;

      for(let i=0;i<16;i++){
          const mid = (low + high) / 2;
          el.style.fontSize = `${mid}px`;
          if(elementFits(el)){
              best = mid;
              low = mid;
          } else {
              high = mid;
          }
      }

      el.style.fontSize = `${Math.max(minPx,best)}px`;
  }

  function fitText(card){
      if(!card) return;
      card.querySelectorAll("[data-pc-fit]").forEach(el => {
          el.style.fontSize = "";
      });
      card.querySelectorAll("[data-pc-fit]").forEach(el => fitElement(card, el));
  }

  function trimTransparentPadding(img){
      if(!img || img.dataset.pcAlphaChecked === "1") return;
      img.dataset.pcAlphaChecked = "1";

      const run = () => {
          if(!img.naturalWidth || !img.naturalHeight) return;
          try{
              const canvas = document.createElement("canvas");
              canvas.width = img.naturalWidth;
              canvas.height = img.naturalHeight;
              const ctx = canvas.getContext("2d", {willReadFrequently:true});
              ctx.drawImage(img,0,0);
              const px = ctx.getImageData(0,0,canvas.width,canvas.height).data;

              let minX=canvas.width, minY=canvas.height, maxX=-1, maxY=-1;
              for(let y=0;y<canvas.height;y++){
                  for(let x=0;x<canvas.width;x++){
                      const a = px[(y*canvas.width+x)*4+3];
                      if(a > 8){
                          if(x<minX) minX=x;
                          if(x>maxX) maxX=x;
                          if(y<minY) minY=y;
                          if(y>maxY) maxY=y;
                      }
                  }
              }
              if(maxX < minX || maxY < minY) return;

              const pad = 2;
              minX = Math.max(0,minX-pad);
              minY = Math.max(0,minY-pad);
              maxX = Math.min(canvas.width-1,maxX+pad);
              maxY = Math.min(canvas.height-1,maxY+pad);

              const w = maxX-minX+1;
              const h = maxY-minY+1;
              const almostFull = w >= canvas.width*.985 && h >= canvas.height*.985;
              if(almostFull) return;

              const out = document.createElement("canvas");
              out.width = w;
              out.height = h;
              out.getContext("2d").drawImage(canvas,minX,minY,w,h,0,0,w,h);
              img.src = out.toDataURL("image/png");
          }catch(_err){
              // Cross-origin images may not expose pixel data. Keep the original
              // proportional image unchanged rather than risking a broken card.
          }
      };

      if(img.complete) run();
      else img.addEventListener("load",run,{once:true});
  }

  function fitSlot(slot){
      if(!slot) return;
      const card = slot.querySelector(":scope > .vx-player-card");
      if(!card) return;
      // Size from the slot's untransformed layout box. getBoundingClientRect()
      // includes the 4K->output render transform and can double-shrink the hero.
      const slotW = Number(slot.clientWidth || slot.offsetWidth || 0);
      const slotH = Number(slot.clientHeight || slot.offsetHeight || 0);
      if(slotW <= 0 || slotH <= 0) return;

      let cardW = slotW;
      let cardH = cardW / CARD_RATIO;
      if(cardH > slotH){
          cardH = slotH;
          cardW = cardH * CARD_RATIO;
      }

      card.style.width = `${Math.floor(cardW)}px`;
      card.style.height = `${Math.floor(cardH)}px`;

      requestAnimationFrame(() => fitText(card));
  }

  function mountPlayerCard(slot, playerData, options={}){
      if(!slot) return null;
      slot.innerHTML = renderPlayerCard(playerData, options);
      const card = slot.querySelector(":scope > .vx-player-card");

      const portrait = card?.querySelector(".pc-portrait");
      if(portrait) trimTransparentPadding(portrait);

      card?.querySelectorAll("img").forEach(img => {
          if(!img.complete){
              img.addEventListener("load",() => fitSlot(slot),{once:true});
          }
      });

      if(slot.__vxPlayerCardObserver) slot.__vxPlayerCardObserver.disconnect();
      slot.__vxPlayerCardObserver = new ResizeObserver(() => fitSlot(slot));
      slot.__vxPlayerCardObserver.observe(slot);

      requestAnimationFrame(() => {
          fitSlot(slot);
          if(document.fonts && document.fonts.ready){
              document.fonts.ready.then(() => fitSlot(slot)).catch(()=>{});
          }
      });
      return card;
  }

  window.renderPlayerCard = renderPlayerCard;
  window.VXPlayerCard = {
      ratio:CARD_RATIO,
      render:renderPlayerCard,
      mount:mountPlayerCard,
      fitSlot,
      fitText
  };
})();
</script>
"""

print(f"✅ PLAYER CARD DESIGN loaded — {PLAYER_CARD_DESIGN_VERSION}")
print("✅ Scope: HERO / LEFT-SIDE PLAYER CARD ONLY")
print("✅ Responsive: aspect-ratio + container units + ResizeObserver")
print("✅ Data/math: supplied by existing notebook cells; not duplicated here")
```

```python
# ============================================================
# CELL 16 — GW REVIEW • PLAYER-BY-PLAYER EXACT SYNC
# ============================================================

from pathlib import Path
import json, re, subprocess
import pandas as pd

ROOT   = Path(globals().get("PROJECT_ROOT", Path("output/MyDrive/FPL_VORTEX")))
HTML   = Path(globals().get("HTML_WORK_DIR", Path("output/.work/html")))
AUDIO  = Path(globals().get("AUDIO_WORK_DIR", Path("output/.work/audio"))) / "review_segments"
TIMING = Path(globals().get("DATA_DIR", Path("output/.work/data")))
MP3_OUT = Path(globals().get("MP3_DIR", Path("output/.work/mp3")))

SRC_HTML  = HTML/"gw_review_scene_4k.html"
ANIM_HTML = HTML/"gw_review_scene_4k_animated.html"
PACKAGE   = TIMING/"gw_review_package.json"
MASTER_MP3 = MP3_OUT/"gw_review.mp3"
MASTER_VTT = TIMING/"gw_review_master.vtt"

if not SRC_HTML.exists():
    raise FileNotFoundError("Run Cell 14 first.")

# Canonical Cell 15 package handoff.
_review_pkg_memory = None

if "GW_REVIEW_PACKAGE" in globals():
    _review_pkg_memory = GW_REVIEW_PACKAGE
elif "review_package" in globals():
    # Compatibility with a Cell 15 execution completed before this fix
    # was loaded into the current Colab runtime.
    _review_pkg_memory = review_package

if _review_pkg_memory is not None:
    pkg = _review_pkg_memory

    if not isinstance(pkg, dict) or "segments" not in pkg or "gw" not in pkg:
        raise RuntimeError(
            "The in-memory GW review package is invalid. "
            "Rerun CELL 15 — AUTO GW REVIEW DATA + RYAN NARRATION TIMING."
        )

    if "GW_REVIEW" in globals():
        _expected_review_gw = int(GW_REVIEW["gw"])
        if int(pkg["gw"]) != _expected_review_gw:
            raise RuntimeError(
                f"Stale GW review package in memory: package GW {pkg['gw']} "
                f"but current review GW is {_expected_review_gw}. "
                "Rerun CELL 15 before Cell 16."
            )

    GW_REVIEW_PACKAGE = pkg

    if not PACKAGE.exists():
        PACKAGE.parent.mkdir(parents=True, exist_ok=True)
        PACKAGE.write_text(
            json.dumps(pkg, indent=2, ensure_ascii=False),
            encoding="utf-8",
        )
        print("✅ Restored missing GW review package from Cell 15 memory:", PACKAGE)

elif PACKAGE.exists():
    pkg = json.loads(PACKAGE.read_text(encoding="utf-8"))
    GW_REVIEW_PACKAGE = pkg

else:
    raise FileNotFoundError(
        "Missing GW review package and no completed Cell 15 package is "
        "available in memory. Run CELL 15 — AUTO GW REVIEW DATA + "
        "RYAN NARRATION TIMING once, then continue with Cell 16."
    )


for _name in ["PLAYER_CARD_CSS", "PLAYER_CARD_JS"]:
    if _name not in globals():
        raise NameError(
            "Run CELL 15A — PLAYER CARD DESIGN before Cell 16. "
            f"Missing {_name}."
        )

# Cell 16 owns no FDR math and defines no duplicate FDR helpers.
# Cell 14 is the presentation handoff from the authoritative model cells.
for _name in ("fdr_table", "format_fdr_value", "fdr_class"):
    if _name not in globals():
        raise NameError(
            f"Run Cell 14 first: missing shared presentation object {_name!r}. "
            "Do not recreate FDR math or helper functions in Cell 16."
        )
if "team_meta_by_short" not in globals():
    raise NameError("Run LIVE FOOTBALL ENRICHMENT first: missing canonical team metadata.")

# ------------------------------------------------------------
# 1. ENRICH PLAYERS + PITCH POSITIONS
#    HERO CARD DATA ONLY — NO HERO CARD DESIGN LIVES HERE.
# ------------------------------------------------------------

_POSITION_LABEL = {
    "GK":"GOALKEEPER",
    "DEF":"DEFENDER",
    "MID":"MIDFIELDER",
    "FWD":"FORWARD",
}

# Reuse the season already established by the enrichment cell when present.
try:
    _pc_season_start = int(globals().get("EPL_SEASON"))
except Exception:
    from datetime import datetime
    _now = datetime.now()
    _pc_season_start = _now.year if _now.month >= 7 else _now.year - 1

_PC_SEASON_LABEL = f"{_pc_season_start}/{str(_pc_season_start + 1)[-2:]} SEASON"

# Existing verified FDR table is the single source of truth for the card.
_pc_fdr_rows = {
    str(r["team_short"]).upper(): r
    for _, r in fdr_table.iterrows()
}

_pc_horizon_by_id = {}
if isinstance(globals().get("player_horizon_xpts_df"), pd.DataFrame) and not player_horizon_xpts_df.empty:
    _pc_horizon_by_id = {
        int(r["player_id"]): r
        for _, r in player_horizon_xpts_df.iterrows()
        if pd.notna(r.get("player_id"))
    }

_pc_fixture_xpts_by_player_gw = {}
if isinstance(globals().get("player_fixture_xpts_df"), pd.DataFrame) and not player_fixture_xpts_df.empty:
    _pc_fx = player_fixture_xpts_df.copy()
    _pc_fx["player_id"] = pd.to_numeric(_pc_fx["player_id"], errors="coerce")
    _pc_fx["gw"] = pd.to_numeric(_pc_fx["gw"], errors="coerce")
    _pc_fx = _pc_fx.dropna(subset=["player_id", "gw"])
    for (_pid, _gw), _grp in _pc_fx.groupby(["player_id", "gw"]):
        _pc_fixture_xpts_by_player_gw[(int(_pid), int(_gw))] = float(
            pd.to_numeric(_grp["xpts"], errors="coerce").fillna(0.0).sum()
        )

_pc_model_fdr_by_team_gw = {}
if isinstance(globals().get("model_fixture_fdr_df"), pd.DataFrame) and not model_fixture_fdr_df.empty:
    for _, _r in model_fixture_fdr_df.iterrows():
        _pc_model_fdr_by_team_gw[(str(_r.get("team", "")).upper(), int(_r["gw"]))] = _r

_pc_gk_fdr_by_team_gw = {}
if isinstance(globals().get("goalkeeper_fixture_fdr_df"), pd.DataFrame) and not goalkeeper_fixture_fdr_df.empty:
    for _, _r in goalkeeper_fixture_fdr_df.iterrows():
        _pc_gk_fdr_by_team_gw[(str(_r.get("team", "")).upper(), int(_r["gw"]))] = _r


def _pc_as_list(value):
    if isinstance(value, list):
        return value
    if isinstance(value, tuple):
        return list(value)
    try:
        if hasattr(value, "tolist"):
            out = value.tolist()
            return out if isinstance(out, list) else [out]
    except Exception:
        pass
    return [] if value is None else [value]


def _pc_text_number(value, decimals=1):
    try:
        number = float(value)
        return f"{number:.{decimals}f}"
    except Exception:
        return "—"


def _pc_int(value):
    try:
        return int(float(value))
    except Exception:
        return 0


def _pc_fixture_pack(team_short):
    """Presentation payload built from the existing verified fdr_table."""
    row = _pc_fdr_rows.get(str(team_short).upper())
    if row is None:
        return []

    fixture_values = _pc_as_list(row.get("fixtures"))
    fdr_values = _pc_as_list(row.get("fdr"))
    out = []

    for i in range(5):
        fixture_raw = fixture_values[i] if i < len(fixture_values) else ""
        fdr_raw = fdr_values[i] if i < len(fdr_values) else None

        fixture_text = format_fdr_value(fixture_raw)
        fdr_text = format_fdr_value(fdr_raw)

        matches = re.findall(r"\b([A-Z]{3})\s*\(([HA])\)", fixture_text.upper())
        opponent = "+".join(x[0] for x in matches) if matches else ("BLANK" if "BLANK" in fixture_text.upper() else "—")
        venue = "+".join(x[1] for x in matches) if matches else ""

        badges = []
        if "team_meta_by_short" in globals():
            for short, _venue in matches[:2]:
                badge = (team_meta_by_short.get(short, {}) or {}).get("badge", "")
                if badge:
                    badges.append(badge)

        _gw = int(NEXT_GW) + i
        _model_row = _pc_model_fdr_by_team_gw.get((str(team_short).upper(), _gw))
        _gk_row = _pc_gk_fdr_by_team_gw.get((str(team_short).upper(), _gw))
        out.append({
            "gw": _gw,
            "opponent": opponent,
            "venue": f"({venue})" if venue else "",
            "fdr": fdr_text,
            "fdr_class": fdr_class(fdr_raw),
            "badges": badges,
            # Model fields are attached for data consumers but the approved
            # fixture-tile HTML still displays official FDR only.
            "attack_fdr_model": (
                float(_model_row["attack_fdr_model"])
                if _model_row is not None and pd.notna(_model_row.get("attack_fdr_model"))
                else None
            ),
            "defence_fdr_model": (
                float(_model_row["defence_fdr_model"])
                if _model_row is not None and pd.notna(_model_row.get("defence_fdr_model"))
                else None
            ),
            "goalkeeper_fdr_model": (
                float(_gk_row["goalkeeper_fdr_model"])
                if _gk_row is not None and pd.notna(_gk_row.get("goalkeeper_fdr_model"))
                else None
            ),
        })

    return out


for p in pkg["players"]:

    # All review-season/player fields were frozen in the single GW_REVIEW
    # package upstream. Do not read shadow globals or refetch player data here.
    p["team_code"] = p.get("team_code")

    team_badge = ""
    try:
        team_badge = (team_meta_by_short.get(str(p["team"]).upper(), {}) or {}).get("badge", "")
    except Exception:
        pass

    # If the local badge cache is unavailable, use the same official PL team code
    # already present in notebook data. This is a display fallback, not new math.
    if not team_badge and p.get("team_code"):
        team_badge = (
            "https://resources.premierleague.com/premierleague/"
            f"badges/100/t{p['team_code']}.png"
        )

    stats = {
        "goals": _pc_int(p.get("goals", 0)),
        "assists": _pc_int(p.get("assists", 0)),
        "minutes": f"{_pc_int(p.get('minutes', 0)):,}",
        "clean_sheets": _pc_int(p.get("clean_sheets", 0)),
        "goals_conceded": _pc_int(p.get("goals_conceded", 0)),
        "saves": _pc_int(p.get("saves", 0)),
        "penalties_saved": _pc_int(p.get("season_penalties_saved", 0)),
        "xg": _pc_text_number(p.get("season_xg"), 1),
        "xa": _pc_text_number(p.get("season_xa"), 1),
        "xgi": _pc_text_number(p.get("season_xgi"), 1),
        "xgc": _pc_text_number(p.get("season_xgc"), 1),
        "points_per_game": _pc_text_number(p.get("season_ppg"), 1),
        "bonus": _pc_int(p.get("season_bonus", p.get("bonus", 0))),
        "bps": _pc_int(p.get("season_bps", p.get("bps", 0))),
    }

    fixture = p.get("fixture") or {}
    review_bits = [
        f"GW {pkg['gw']}",
        f"{p.get('points', 0)} PTS",
        f"{p.get('minutes', 0)} MIN",
    ]
    if fixture.get("opp"):
        review_bits.append(f"{fixture.get('opp')} {fixture.get('ha','')} {fixture.get('score','')}".strip())

    _pid = int(p.get("id", 0) or 0)
    _horizon_row = _pc_horizon_by_id.get(_pid)
    if _horizon_row is not None:
        p["model_projection"] = {
            "next_gw_xpts": float(_horizon_row.get("next_gw_xpts", 0.0) or 0.0),
            "xpts_3gw": float(_horizon_row.get("xpts_3gw", 0.0) or 0.0),
            "xpts_5gw": float(_horizon_row.get("xpts_5gw", 0.0) or 0.0),
            "xpts_8gw": float(_horizon_row.get("xpts_8gw", 0.0) or 0.0),
            "p_start_next_gw": float(_horizon_row.get("p_start_next_gw", 0.0) or 0.0),
            "p_appear_next_gw": float(_horizon_row.get("p_appear_next_gw", 0.0) or 0.0),
            "p_60_next_gw": float(_horizon_row.get("p_60_next_gw", 0.0) or 0.0),
            "expected_minutes_next_gw": float(_horizon_row.get("expected_minutes_next_gw", 0.0) or 0.0),
            "fixtures_next_gw": int(_horizon_row.get("fixtures_next_gw", 0) or 0),
        }
    else:
        p["model_projection"] = {}

    duty = "BENCH" if p.get("bench") else "STARTER"
    if p.get("captain"):
        leadership = "CAPTAIN"
    elif p.get("vice"):
        leadership = "VICE CAPTAIN"
    else:
        leadership = f"GW {pkg['gw']}"

    p["card"] = {
        "id": p.get("id"),
        "name": p.get("name", ""),
        "first_name": p.get("first_name") or str(p.get("full_name", p.get("name", ""))).split(" ")[0],
        "surname": p.get("second_name") or " ".join(str(p.get("full_name", "")).split(" ")[1:]) or p.get("name", ""),
        "position": p.get("pos", ""),
        "position_label": _POSITION_LABEL.get(p.get("pos"), p.get("pos", "PLAYER")),
        "club": p.get("team_name", ""),
        "club_short": p.get("team", ""),
        "club_badge": team_badge,
        "player_image": p.get("photo", ""),
        "asset_label": f"GW {pkg['gw']} PLAYER REVIEW",
        "price": f"£{float(p.get('price', 0)):.1f}m",
        "total_points": _pc_int(p.get("season_total_points", 0)),
        "form": _pc_text_number(p.get("season_form"), 1),
        "ownership": f"{float(p.get('ownership', 0) or 0):.1f}%",
        "stats": stats,
        "stats_heading": f"{_PC_SEASON_LABEL} • {_POSITION_LABEL.get(p.get('pos'), p.get('pos','PLAYER'))} STATS",
        "fixtures": _pc_fixture_pack(p.get("team", "")),
        "verdict_main": f"{_POSITION_LABEL.get(p.get('pos'), p.get('pos','PLAYER'))} REVIEW",
        "verdict_sub": "  |  ".join(review_bits),
        "tags": [
            _POSITION_LABEL.get(p.get("pos"), p.get("pos", "PLAYER")),
            duty,
            leadership,
            f"BONUS {p.get('bonus', 0)}",
            f"BPS {p.get('bps', 0)}",
        ],
        "season_label": _PC_SEASON_LABEL,
        # Data only: PLAYER_CARD_JS/CSS is unchanged in Phase 4.
        "model_projection": p.get("model_projection", {}),
    }

# IMPORTANT: pkg["players"], pkg["starters"] and pkg["bench"] may
# contain separate dictionary objects after JSON serialization.
# Rebind starters/bench to the enriched player objects so card data,
# team_code and every later animation field are always present.
_player_by_id = {int(p["id"]): p for p in pkg["players"]}

pkg["starters"] = [
    _player_by_id.get(int(p["id"]), p)
    for p in pkg["starters"]
]

pkg["bench"] = [
    _player_by_id.get(int(p["id"]), p)
    for p in pkg["bench"]
]

# same approved formation geometry
groups = {
    pos:[p for p in pkg["starters"] if p["pos"] == pos]
    for pos in ("GK","DEF","MID","FWD")
}

ys = {"GK":13.5, "DEF":35.5, "MID":58.2, "FWD":82.0}

for pos, arr in groups.items():

    span = 0 if pos == "GK" else 47 if pos == "FWD" else 56

    for i,p in enumerate(arr):

        p["pitch_x"] = (
            50 if len(arr) == 1
            else 50-span/2 + span*(i/(len(arr)-1))
        )

        p["pitch_y"] = ys[pos]

# ------------------------------------------------------------
# 2. GLOBAL AUDIO TIMELINE
# ------------------------------------------------------------

cursor = 0.0

for s in pkg["segments"]:

    s["t0"] = cursor
    s["t1"] = cursor + float(s["duration"])

    if s.get("player_id"):
        s["reveal_at"] = s["t0"] + float(s.get("name_time",0))

        # Pitch/bench image and HERO begin together at the exact word
        # boundary where Ryan says the player's name.
        s["hero_on"] = s["reveal_at"]

        # The player remains on the pitch/bench; only the temporary hero leaves
        # at the end of that player's explanation.
        s["hero_off"] = min(
            s["t1"]-0.02,
            max(s["hero_on"]+0.35, s["t1"]-0.12)
        )

    cursor = s["t1"]

TOTAL = cursor

# ------------------------------------------------------------
# 2B. SQUAD-REVEAL SYNC QA
# ------------------------------------------------------------

_player_sync_segments=[s for s in pkg["segments"] if s.get("player_id")]
_expected_sync_ids=[
    int(p["id"])
    for p in (pkg["starters"]+pkg["bench"])
]
_actual_sync_ids=[int(s["player_id"]) for s in _player_sync_segments]

if _actual_sync_ids != _expected_sync_ids:
    raise RuntimeError(
        "GW Review sync order mismatch: "
        f"expected {_expected_sync_ids}, got {_actual_sync_ids}"
    )

for s in _player_sync_segments:
    if not (float(s["t0"]) <= float(s["reveal_at"]) < float(s["t1"])):
        raise RuntimeError(f"Reveal outside narration window: {s['id']}")
    if abs(float(s["hero_on"])-float(s["reveal_at"])) > 1e-6:
        raise RuntimeError(f"Hero/pitch reveal not aligned to Ryan name: {s['id']}")
    if not (float(s["hero_on"]) < float(s["hero_off"]) <= float(s["t1"])):
        raise RuntimeError(f"Hero window invalid: {s['id']}")

_last_player_t1=max(float(s["t1"]) for s in _player_sync_segments)
_metrics_seg=next(
    (s for s in pkg["segments"] if s["id"]=="metrics_intro"),
    None
)
if _metrics_seg is None or float(_metrics_seg["t0"]) < _last_player_t1-1e-6:
    raise RuntimeError("Full data board is scheduled before the final bench review ends.")

print("✅ GW Review sync QA: Ryan name = hero reveal = player reveal")
print("✅ Player images persist; only hero windows end")
print("✅ Full data board begins after the final reserve")

# ------------------------------------------------------------
# 3. CONCAT RYAN SEGMENTS → ONE MASTER AUDIO
# ------------------------------------------------------------

concat_file = Path(globals().get("AUDIO_WORK_DIR", Path("output/.work/audio")))/"gw_review_concat.txt"

concat_file.write_text(
    "\n".join(
        f"file '{Path(s['audio']).as_posix()}'"
        for s in pkg["segments"]
    ),
    encoding="utf-8"
)

subprocess.run(
    [
        "ffmpeg","-y","-v","error",
        "-f","concat","-safe","0",
        "-i",str(concat_file),
        "-c","copy",
        str(MASTER_MP3)
    ],
    check=True
)

# ------------------------------------------------------------
# 4. VTT
# ------------------------------------------------------------

def ts(t):
    ms = round(t*1000)
    h,ms = divmod(ms,3600000)
    m,ms = divmod(ms,60000)
    sec,ms = divmod(ms,1000)
    return f"{h:02}:{m:02}:{sec:02}.{ms:03}"

vtt = ["WEBVTT",""]

for i,s in enumerate(pkg["segments"],1):
    vtt += [
        str(i),
        f"{ts(s['t0'])} --> {ts(s['t1'])}",
        s["text"],
        ""
    ]

MASTER_VTT.write_text(
    "\n".join(vtt),
    encoding="utf-8"
)

# ------------------------------------------------------------
# 5. TIMING LOOKUPS
# ------------------------------------------------------------

def seg(name):
    return next(
        (x for x in pkg["segments"] if x["id"] == name),
        None
    )

headline_seg = seg("headline")
bench_intro  = seg("bench_intro")
metrics_intro = seg("metrics_intro")

metric_map = {}

for sid,key in [
    ("metric_points","points"),
    ("metric_captain","captain"),
    ("metric_bench","bench"),
    ("metric_bank","bank"),
    ("metric_gwrank","gwrank"),
    ("metric_orank","orank"),
    ("metric_hit","hit"),
]:
    s = seg(sid)
    if s:
        metric_map[key] = [s["t0"],s["t1"]]

# ------------------------------------------------------------
# 6. REVIEW DATA FOR JAVASCRIPT
# ------------------------------------------------------------

metrics = pkg["metrics"]

captain = next(
    (p for p in pkg["starters"] if p["captain"]),
    pkg["starters"][0]
)

bench_pts = sum(p["points"] for p in pkg["bench"])

JS_DATA = {
    "gw":pkg["gw"],
    "starters":pkg["starters"],
    "bench":pkg["bench"],
    "segments":pkg["segments"],
    "total":TOTAL,

    "headline":f"GAMEWEEK {pkg['gw']} REVIEW • PLAYER BY PLAYER",

    "headline_t0":headline_seg["t0"],
    "headline_t1":headline_seg["t1"],

    "bench_t":bench_intro["t0"] if bench_intro else TOTAL,
    "metrics_t":metrics_intro["t0"] if metrics_intro else TOTAL,

    "metric_times":metric_map,

    "metrics":{
        "points":metrics["gw_points"],
        "bank":f'£{metrics["bank"]:.1f}m',
        "value":f'£{metrics["value"]:.1f}m',
        "gwrank":metrics["gw_rank"],
        "orank":metrics["overall_rank"],
        "prev_rank":metrics["previous_rank"],
        "rank_dir":metrics["rank_dir"],
        "hit":metrics["transfer_cost"],
        "captain":captain["name"],
        "captain_return":captain["points"]*captain["multiplier"],
        "bench_points":bench_pts,
    }
}

# ------------------------------------------------------------
# 7. PREMIUM REVIEW CSS
# ------------------------------------------------------------

CSS = r"""
<!-- VX_REVIEW_START -->
<style id="vx-review-css">

/* Original left information panels stay blank initially */
.left > .panel{
    opacity:0 !important;
    pointer-events:none !important;
}

.analysisRibbon{
    display:none !important;
}

/* player pitch cards initially invisible */
.playerCard{
    opacity:0;
    will-change:transform,opacity,filter;
}

/* -------------------------------------------------------
   LEFT REVIEW STAGE
   ------------------------------------------------------- */

#vxReviewStage{
    position:absolute;
    inset:0;
    z-index:80;
    overflow:hidden;
    border-radius:23px;
}

#vxBlank{
    position:absolute;
    inset:0;
    border:2px solid rgba(255,225,94,.45);
    border-radius:23px;

    background:
      radial-gradient(circle at 50% 30%,
      rgba(255,218,66,.11),transparent 35%),

      linear-gradient(160deg,
      rgba(6,18,31,.98),
      rgba(3,9,17,.99));

    box-shadow:
      inset 0 0 55px rgba(255,213,54,.055),
      0 20px 45px rgba(0,0,0,.35);
}

#vxHeadline{
    position:absolute;
    left:70px;
    right:70px;
    top:400px;

    font-size:69px;
    line-height:1.08;
    font-weight:1000;
    font-style:italic;

    color:#fff;
    text-align:center;

    text-shadow:
      0 4px 8px #000,
      0 0 25px rgba(255,216,70,.26);
}

#vxCaret{
    color:#ffe04b;
}

/* -------------------------------------------------------
   HERO CARD SLOT ONLY
   Internal card design is owned by CELL 15A — PLAYER CARD DESIGN.
   This slide controls only placement / available space.
   ------------------------------------------------------- */

#vxHeroWrap{
    position:absolute;
    /* 4K safe hero region: stage-left -> pitch boundary,
       18px below the logo -> 18px above the footer.
       Parent .left is x=54,y=428,w=1202,h=1642; pitch starts x=1286. */
    left:-54px;
    right:-30px;
    top:-4px;
    bottom:-3px;
    z-index:95;
    display:flex;
    align-items:center;
    justify-content:center;
    overflow:visible;
    opacity:0;
    pointer-events:none;
    transform:translateY(45px);
    will-change:opacity,transform;
}

#vxHeroWrap > .vx-player-card{
    flex:none;
}

/* -------------------------------------------------------
   BENCH DOCK
   ------------------------------------------------------- */

#vxBenchDock{
    position:absolute;
    left:0;
    right:0;
    bottom:0;

    height:310px;

    padding:18px;

    opacity:0;

    border-radius:20px;
    border:2px solid rgba(255,220,74,.35);

    background:
      linear-gradient(180deg,#091827,#030a11);

    display:grid;
    grid-template-columns:repeat(4,1fr);
    gap:13px;
}

.vxb{
    position:relative;
    opacity:0;

    border-radius:14px;
    overflow:hidden;

    background:
      linear-gradient(180deg,#10283a,#06121c);

    border:2px solid rgba(255,255,255,.14);

    text-align:center;
}

.vxb img{
    height:180px;
    max-width:100%;
    object-fit:contain;
}

.vxb-name{
    height:44px;

    display:flex;
    align-items:center;
    justify-content:center;

    color:#fff;
    background:#02080d;

    font-size:20px;
    font-weight:1000;
}

.vxb-pts{
    position:absolute;
    right:8px;
    top:8px;

    width:48px;
    height:40px;

    display:flex;
    align-items:center;
    justify-content:center;

    border-radius:8px;

    background:#ffe04b;
    color:#07111a;

    font-size:22px;
    font-weight:1000;
}

/* -------------------------------------------------------
   FINAL METRIC BOARD
   ------------------------------------------------------- */

#vxMetrics{
    position:absolute;
    left:0;
    right:0;
    top:0;

    height:1305px;

    padding:22px;

    display:grid;
    grid-template-columns:1fr 1fr;
    grid-template-rows:repeat(4,1fr);
    gap:14px;

    opacity:0;

    background:
      linear-gradient(
        160deg,
        rgba(5,16,28,.985),
        rgba(2,8,15,.995)
      );

    border:2px solid rgba(255,220,70,.36);
    border-radius:22px;
}

.vxm{
    position:relative;

    padding:28px;

    display:flex;
    flex-direction:column;
    justify-content:center;

    border-radius:17px;

    background:
      linear-gradient(
        160deg,
        rgba(255,224,74,.14),
        rgba(255,255,255,.025)
      );

    border:1px solid rgba(255,224,74,.24);

    transition:none;
}

.vxm-label{
    color:#94a9ba;
    font-size:23px;
    font-weight:900;
    letter-spacing:2px;
}

.vxm-value{
    margin-top:8px;

    color:white;
    font-size:54px;
    line-height:1;
    font-weight:1000;
}

.vxm-sub{
    margin-top:9px;

    color:#ffe04b;
    font-size:20px;
    font-weight:900;
}

.rank-up{
    color:#36ef78 !important;
}

.rank-down{
    color:#ff5b55 !important;
}

.rank-arrow{
    margin-left:12px;
    font-size:46px;
}

</style>
"""

# ------------------------------------------------------------
# 8. DETERMINISTIC REVIEW ENGINE
# ------------------------------------------------------------

JS = r"""
<script id="vx-review-engine">

const VX = __VX_DATA__;
window.VX = VX;

function $v(q){ return document.querySelector(q); }

function escv(x){
    return String(x ?? "")
      .replaceAll("&","&amp;")
      .replaceAll("<","&lt;")
      .replaceAll(">","&gt;");
}

function clamp(x,a=0,b=1){
    return Math.max(a,Math.min(b,x));
}

function ease(x){
    x=clamp(x);
    return 1-Math.pow(1-x,3);
}

function fmtN(x){
    if(x===null || x===undefined) return "—";
    return Number(x).toLocaleString("en-GB");
}

/* -------------------------------------------------------
   BUILD PITCH XI
   ------------------------------------------------------- */

function pitchCard(p){

    const crest = p.team_code
      ? `<img class="teamCrest"
           src="https://resources.premierleague.com/premierleague/badges/70/t${p.team_code}.png">`
      : "";

    const badge = p.captain
      ? `<div class="badge">C</div>`
      : p.vice
      ? `<div class="badge vc">VC</div>`
      : "";

    return `
    <div
      id="vxp_${p.id}"
      class="playerCard ${p.captain?"captain":""} ${p.vice?"vice":""}"
      style="left:${p.pitch_x}%;top:${p.pitch_y}%;opacity:0">

      ${badge}

      <div class="photoZone">
        <img src="${p.photo}">
        ${crest}
      </div>

      <div class="playerName">${escv(p.name)}</div>

      <div class="playerMeta">
        ${escv(p.fixture.opp)} (${escv(p.fixture.ha)})
        <span class="tinyScore">${escv(p.pos)}</span>
      </div>

      <div class="points ${pointsClass(p.points)}">
        ${p.points}
      </div>

    </div>`;
}

/* -------------------------------------------------------
   LEFT SUMMARY METRICS
   ------------------------------------------------------- */

function metricsHTML(){

    const m=VX.metrics;

    let arrow="";
    let rankClass="";

    if(m.rank_dir==="up"){
        arrow=`<span class="rank-arrow">▲</span>`;
        rankClass="rank-up";
    }
    else if(m.rank_dir==="down"){
        arrow=`<span class="rank-arrow">▼</span>`;
        rankClass="rank-down";
    }

    const hit = Number(m.hit||0);

    return `
      <div class="vxm" data-key="points">
        <div class="vxm-label">GW POINTS</div>
        <div class="vxm-value">${fmtN(m.points)}</div>
        <div class="vxm-sub">GAMEWEEK ${VX.gw}</div>
      </div>

      <div class="vxm" data-key="bank">
        <div class="vxm-label">BANK</div>
        <div class="vxm-value">${m.bank}</div>
        <div class="vxm-sub">SQUAD ${m.value}</div>
      </div>

      <div class="vxm" data-key="gwrank">
        <div class="vxm-label">GW RANK</div>
        <div class="vxm-value">${fmtN(m.gwrank)}</div>
        <div class="vxm-sub">GLOBAL GAMEWEEK RANK</div>
      </div>

      <div class="vxm" data-key="orank">
        <div class="vxm-label">OVERALL RANK</div>
        <div class="vxm-value ${rankClass}">
          ${fmtN(m.orank)} ${arrow}
        </div>
        <div class="vxm-sub">
          PREVIOUS ${fmtN(m.prev_rank)}
        </div>
      </div>

      <div class="vxm" data-key="captain">
        <div class="vxm-label">CAPTAIN</div>
        <div class="vxm-value">${escv(m.captain)}</div>
        <div class="vxm-sub">${m.captain_return} RETURNED PTS</div>
      </div>

      <div class="vxm" data-key="bench">
        <div class="vxm-label">BENCH POINTS</div>
        <div class="vxm-value">${m.bench_points}</div>
        <div class="vxm-sub">UNUSED OUTPUT</div>
      </div>

      <div class="vxm" data-key="hit">
        <div class="vxm-label">TRANSFER COST</div>
        <div class="vxm-value">${hit ? "-"+hit : "0"}</div>
        <div class="vxm-sub">${hit ? "POINT HIT" : "NO HIT"}</div>
      </div>

      <div class="vxm">
        <div class="vxm-label">REVIEW STATUS</div>
        <div class="vxm-value">COMPLETE</div>
        <div class="vxm-sub">STARTING XI + BENCH</div>
      </div>
    `;
}

/* -------------------------------------------------------
   INITIAL BUILD
   ------------------------------------------------------- */

function buildReview(){

    const players=$v("#players");
    players.innerHTML=VX.starters.map(p=>pitchCard(p)).join("");

    const left=$v(".left");

    left.insertAdjacentHTML(
      "beforeend",
      `
      <div id="vxReviewStage">

        <div id="vxBlank">
          <div id="vxHeadline"></div>
        </div>

        <div id="vxMetrics">
          ${metricsHTML()}
        </div>

        <div id="vxBenchDock">
          ${VX.bench.map(p=>`
            <div class="vxb" id="vxb_${p.id}">
              <div class="vxb-pts">${p.points}</div>
              <img src="${p.photo}">
              <div class="vxb-name">${escv(p.name)}</div>
            </div>
          `).join("")}
        </div>

      </div>
      <div id="vxHeroWrap"></div>`
    );

    $v("#pitchGW").textContent=`GW ${VX.gw} REVIEW`;
    $v("#entryName").textContent="MY SQUAD";
    $v("#loading").classList.add("hide");

    window.__ready=true;
    window.__seek(0);
}

/* -------------------------------------------------------
   EXACT SEEK ENGINE
   ------------------------------------------------------- */

window.__seek=function(t){

    t=clamp(Number(t)||0,0,VX.total);

    /* ---------- TYPEWRITER ---------- */

    const headline=$v("#vxHeadline");

    const typeEnd =
      VX.headline_t0 +
      (VX.headline_t1-VX.headline_t0)*0.72;

    const tp=clamp(
      (t-VX.headline_t0) /
      Math.max(.01,typeEnd-VX.headline_t0)
    );

    const chars=Math.floor(VX.headline.length*tp);

    headline.innerHTML =
      escv(VX.headline.slice(0,chars)) +
      (t < typeEnd ? `<span id="vxCaret">▌</span>` : "");

    const firstPlayer =
      VX.segments.find(s=>s.player_id);

    $v("#vxBlank").style.opacity =
      firstPlayer && t >= firstPlayer.reveal_at ? 0 : 1;

    /* ---------- STARTING XI ---------- */

    VX.segments
      .filter(s=>s.player_id && s.kind==="starter")
      .forEach(s=>{

        const el=$v(`#vxp_${s.player_id}`);

        if(!el) return;

        const p=ease(
          (t-s.reveal_at)/.42
        );

        el.style.opacity=p;

        el.style.transform=
          `translate(-50%,-50%) scale(${.70+.30*p})`;

        el.style.filter=
          p>0 && p<1
          ? `brightness(${1+.45*(1-p)})`
          : "";
    });

    /* ---------- BENCH ---------- */

    const benchDock=$v("#vxBenchDock");

    benchDock.style.opacity =
      t >= VX.bench_t ? 1 : 0;

    VX.segments
      .filter(s=>s.player_id && s.kind==="bench")
      .forEach(s=>{

        const el=$v(`#vxb_${s.player_id}`);
        if(!el) return;

        const p=ease((t-s.reveal_at)/.38);

        el.style.opacity=p;
        el.style.transform=
          `translateY(${(1-p)*30}px) scale(${.85+.15*p})`;
    });

    /* ---------- TEMP HERO ---------- */

    const heroSeg =
      VX.segments.find(
        s =>
          s.player_id &&
          t >= s.hero_on &&
          t < s.hero_off
      );

    const hw=$v("#vxHeroWrap");

    if(heroSeg && t < VX.metrics_t){

        const all=[...VX.starters,...VX.bench];

        const p=all.find(
          x=>x.id===heroSeg.player_id
        );

        if(p){

            if(hw.dataset.pid != String(p.id)){
                if(!window.VXPlayerCard || typeof window.VXPlayerCard.mount !== "function"){
                    throw new Error("PLAYER CARD DESIGN runtime missing. Run Cell 15A before Cell 16.");
                }
                window.VXPlayerCard.mount(hw, p.card || p);
                hw.dataset.pid=String(p.id);
            }

            const ein=ease(
              (t-heroSeg.hero_on)/.28
            );

            const eout=ease(
              (heroSeg.hero_off-t)/.28
            );

            const vis=Math.min(ein,eout);

            hw.style.opacity=vis;

            hw.style.transform=
              `translateY(${(1-vis)*34}px)`;
        }
    }
    else{
        hw.style.opacity=0;
    }

    /* ---------- METRICS ALL AT ONCE ---------- */

    const metrics=$v("#vxMetrics");

    const mp=ease(
      (t-VX.metrics_t)/.55
    );

    metrics.style.opacity=mp;
    metrics.style.transform=
      `translateY(${(1-mp)*28}px)`;

    if(t >= VX.metrics_t){

        document.querySelectorAll(".vxm").forEach(el=>{
            el.style.boxShadow="";
            el.style.borderColor="rgba(255,224,74,.24)";
            el.style.transform="scale(1)";
        });

        Object.entries(VX.metric_times).forEach(([key,r])=>{

            if(t>=r[0] && t<r[1]){

                const el=
                  document.querySelector(`.vxm[data-key="${key}"]`);

                if(el){
                    el.style.boxShadow=
                      "0 0 36px rgba(255,224,75,.32)";
                    el.style.borderColor="#ffe04b";
                    el.style.transform="scale(1.025)";
                }
            }
        });
    }

    document.documentElement.dataset.vortexTime=t.toFixed(3);

    return t;
};

window.__auditLayout=function(t){

    window.__seek(t);

    return {
      time:t,
      viewport:[innerWidth,innerHeight],
      scroll:[
        document.documentElement.scrollWidth,
        document.documentElement.scrollHeight
      ],
      pitchPlayers:
        [...document.querySelectorAll(".playerCard")]
        .filter(x=>Number(x.style.opacity)>.5).length
    };
};

if(document.readyState==="loading")
    document.addEventListener("DOMContentLoaded",buildReview);
else
    buildReview();

</script>
<!-- VX_REVIEW_END -->
"""

# ------------------------------------------------------------
# 9. INJECT INTO CELL 14 PITCH HTML
# ------------------------------------------------------------

html = SRC_HTML.read_text(
    encoding="utf-8",
    errors="ignore"
)

# remove previous Cell16 injection if rerun
html = re.sub(
    r"<!-- VX_REVIEW_START -->.*?"
    r"<!-- VX_REVIEW_END -->",
    "",
    html,
    flags=re.S
)

# Disable original page's automatic live render.
# Cell 15 is now the authoritative review-data source.
html = re.sub(
    r"\(async\(\)=>\{\s*"
    r"if\(CONFIG\.forceDemo\).*?"
    r"\}\)\(\);",
    "",
    html,
    flags=re.S
)

html = html.replace(
    "</head>",
    PLAYER_CARD_CSS + "\n" + CSS + "\n</head>"
)

html = html.replace(
    "</body>",
    PLAYER_CARD_JS + "\n" +
    JS.replace(
        "__VX_DATA__",
        json.dumps(JS_DATA,ensure_ascii=False)
    )
    + "\n</body>"
)

html = vx_polish_branding(html)

ANIM_HTML.write_text(
    html,
    encoding="utf-8"
)

# ------------------------------------------------------------
# 10. SAVE ANIMATION MANIFEST
# ------------------------------------------------------------

manifest = {
    "html":str(ANIM_HTML),
    "audio":str(MASTER_MP3),
    "vtt":str(MASTER_VTT),
    "duration":TOTAL,
    "canvas":[3840,2160],
    "gw":pkg["gw"],
    "starting_xi":len(pkg["starters"]),
    "bench":len(pkg["bench"]),
    "segments":pkg["segments"]
}

(TIMING/"gw_review_animation_manifest.json").write_text(
    json.dumps(manifest,indent=2),
    encoding="utf-8"
)

print("✅ GW REVIEW ANIMATION READY")
print(f"✅ GW: {pkg['gw']}")
print("✅ Blank pitch at opening")
print("✅ Deterministic typewriter headline")
print("✅ Ryan-name → player pitch reveal")
print("✅ Player remains on pitch")
print("✅ Responsive CELL 15A hero card appears on the spoken name")
print("✅ Hero disappears after explanation")
print("✅ Bench builds independently")
print("✅ Metrics reveal together")
print(f"✅ Total review duration: {TOTAL:.2f}s")
print("✅ HTML:",ANIM_HTML)
print("✅ AUDIO:",MASTER_MP3)
```

```python
# ============================================================
# CELL 14A — INTRO + OUTRO • BROADCAST MASTER DESIGN
# ============================================================

from pathlib import Path
from string import Template
from html import escape
from IPython.display import display, Image as IPImage
import pandas as pd

ROOT = Path(globals().get("PROJECT_ROOT", Path("output/MyDrive/FPL_VORTEX")))
HTML_DIR = Path(globals().get("HTML_WORK_DIR", Path("output/.work/html")))
PNG_DIR  = Path(globals().get("SLIDE_DIR", Path("output/.work/slide")))
HTML_DIR.mkdir(parents=True,exist_ok=True)
PNG_DIR.mkdir(parents=True,exist_ok=True)

INTRO_HTML = HTML_DIR/"intro_scene_4k.html"
OUTRO_HTML = HTML_DIR/"outro_scene_4k.html"
INTRO_PNG  = PNG_DIR/"intro_scene_4k.png"
OUTRO_PNG  = PNG_DIR/"outro_scene_4k.png"

for x in ["team_projection_df","fdr_table","NEXT_GW","vortex_logo_html","VX_FOOTER_UPDATED_DATE","PL_BUG_HTML","PL_BUG_CSS","vx_apply_pl_lockup"]:
    if x not in globals():
        raise NameError(f"Run earlier cells first — missing {x}")

p = team_projection_df.copy()
att  = p.sort_values("avg_xg",ascending=False).head(3)
deff = p.sort_values("avg_cs",ascending=False).head(3)
easy = fdr_table.sort_values("average_fdr").head(3)
hard = fdr_table.sort_values("average_fdr",ascending=False).head(3)

badge_map = dict(zip(p["team_short"],p["badge"]))


# ------------------------------------------------------------
# AUTO PREVIOUS-GW RESULTS STRIP — INTRO + OUTRO
# Uses completed official FPL fixtures only. Presentation only.
# ------------------------------------------------------------
def _vx_previous_results_strip():
    review_gw = globals().get("REVIEW_GW", globals().get("LATEST_COMPLETED_GW"))
    results_source = globals().get("match_results_raw")
    team_meta = globals().get("team_meta_by_id", {})

    try:
        review_gw = int(review_gw)
    except Exception:
        return "", None, 0

    if results_source is None or not hasattr(results_source, "copy"):
        return "", review_gw, 0

    rows = results_source.copy()
    if "event" not in rows.columns:
        return "", review_gw, 0

    rows = rows.loc[
        pd.to_numeric(rows["event"], errors="coerce").eq(review_gw)
    ].copy()

    if "vx_completed" not in rows.columns:
        raise RuntimeError("Intro previous-results strip requires canonical results.vx_completed from Cell 3.")
    rows = rows.loc[rows["vx_completed"].fillna(False).astype(bool)].copy()

    if "kickoff_time" in rows.columns:
        rows["_vx_kickoff"] = pd.to_datetime(rows["kickoff_time"], errors="coerce", utc=True)
        sort_cols = ["_vx_kickoff"] + (["id"] if "id" in rows.columns else [])
        rows = rows.sort_values(sort_cols)
    elif "id" in rows.columns:
        rows = rows.sort_values("id")

    chips = []
    for _, row in rows.iterrows():
        try:
            home_id = int(row.get("team_h"))
            away_id = int(row.get("team_a"))
        except Exception:
            continue

        home_meta = dict(team_meta.get(home_id, {}))
        away_meta = dict(team_meta.get(away_id, {}))
        home_badge = str(home_meta.get("badge", "") or "")
        away_badge = str(away_meta.get("badge", "") or "")
        home_short = escape(str(home_meta.get("short", home_meta.get("short_name", "HOME"))).upper())
        away_short = escape(str(away_meta.get("short", away_meta.get("short_name", "AWAY"))).upper())

        try:
            hs = int(float(row.get("team_h_score")))
            a_s = int(float(row.get("team_a_score")))
        except Exception:
            continue

        home_logo = (
            f'<img class="prevResultLogo" src="{home_badge}" alt="{home_short}">'
            if home_badge else f'<span class="prevResultFallback">{home_short[:3]}</span>'
        )
        away_logo = (
            f'<img class="prevResultLogo" src="{away_badge}" alt="{away_short}">'
            if away_badge else f'<span class="prevResultFallback">{away_short[:3]}</span>'
        )

        chips.append(
            '<div class="prevResultChip">'
            f'{home_logo}<span class="prevResultScore">{hs}-{a_s}</span>{away_logo}'
            '</div>'
        )

    if not chips:
        return "", review_gw, 0

    strip = (
        f'<div class="prevResultsStrip" style="--vx-result-count:{len(chips)}">'
        f'<div class="prevResultsLabel">GW{review_gw} RESULTS</div>'
        + "".join(chips)
        + '</div>'
    )
    return strip, review_gw, len(chips)


PREVIOUS_RESULTS_STRIP, PREVIOUS_RESULTS_GW, PREVIOUS_RESULTS_COUNT = _vx_previous_results_strip()
def _vx_team_pair(df, column="team_short", n=2):
    vals = [str(v).upper() for v in df[column].dropna().astype(str).head(n).tolist()]
    return " • ".join(vals) if vals else "TBD • TBD"

TODAY_ATTACK_TEAMS = _vx_team_pair(att)
TODAY_DEF_TEAMS = _vx_team_pair(deff)
TODAY_WARN_TEAMS = _vx_team_pair(hard)

SOCIAL = r'''
<span class="socialPair"><span class="ytIcon"><svg viewBox="0 0 48 34" aria-hidden="true"><rect x="1" y="1" width="46" height="32" rx="9" fill="#ff0000"/><path d="M20 9 L34 17 L20 25 Z" fill="#ffffff"/></svg></span><span>@FPLVORTEX</span></span>
<span class="socialPair"><span class="xIcon">𝕏</span><span>@FPLVORTEXM</span></span>
'''

STYLE = r'''
*{box-sizing:border-box}
html,body{margin:0;width:3840px;height:2160px;overflow:hidden;background:#020711;color:#fff;font-family:Inter,Arial,Helvetica,sans-serif}
.scene{position:relative;width:3840px;height:2160px;overflow:hidden;background:
radial-gradient(circle at 82% 8%,rgba(0,218,255,.20),transparent 28%),
radial-gradient(circle at 12% 84%,rgba(0,245,160,.12),transparent 31%),
linear-gradient(145deg,#020713 0%,#07182b 52%,#020711 100%)}
.scene:before{content:"";position:absolute;inset:0;background:linear-gradient(90deg,rgba(255,255,255,.025) 1px,transparent 1px),linear-gradient(rgba(255,255,255,.018) 1px,transparent 1px);background-size:96px 96px;opacity:.28;pointer-events:none}
.topHeader{position:absolute;left:78px;right:78px;top:44px;height:300px;display:flex;align-items:center;justify-content:space-between;z-index:10}
.brandLock{height:280px;display:flex;align-items:center;gap:44px;min-width:1700px}
.logoBox{width:650px;height:260px;display:flex;align-items:center;justify-content:flex-start;overflow:visible}
.logoBox .vortex-logo{display:block;max-width:640px!important;max-height:245px!important;width:auto!important;height:auto!important;object-fit:contain!important;filter:drop-shadow(0 12px 28px rgba(0,0,0,.42))}
.logoBox .logo-fallback{font-size:85px;font-weight:1000;color:#fff}.logoBox .logo-fallback span{color:#ffd447}
.brandWords{display:flex;flex-direction:column;justify-content:center}
.channelName{font-size:84px;line-height:.95;font-weight:1000;letter-spacing:2px;color:#ffd447;text-shadow:0 6px 22px rgba(0,0,0,.55)}
.channelSub{margin-top:17px;font-size:28px;font-weight:900;letter-spacing:6px;color:#c6d4df}
.gwPill{font-size:36px;font-weight:1000;letter-spacing:3px;color:#06121f;background:linear-gradient(180deg,#fff0a3,#ffd247);padding:20px 34px;border-radius:18px;box-shadow:0 10px 30px rgba(0,0,0,.35)}
.mainHeading{position:absolute;left:300px;right:300px;top:345px;text-align:center;font-size:92px;line-height:1;font-weight:1000;letter-spacing:3px;color:#fff;text-shadow:0 8px 28px rgba(0,0,0,.52)}
.mainHeading span{color:#35e6ff}
.headingRule{position:absolute;left:1180px;right:1180px;top:465px;height:5px;border-radius:5px;background:linear-gradient(90deg,transparent,#ffd447 20%,#35e6ff 80%,transparent)}
.agendaWrap{position:absolute;left:665px;right:665px;top:540px;height:1335px;display:flex;flex-direction:column;justify-content:center;gap:42px}
.agenda-card{height:245px;border-radius:30px;display:grid;grid-template-columns:230px 1fr;align-items:center;padding:26px 55px;background:linear-gradient(110deg,rgba(10,32,54,.985),rgba(4,17,31,.99));border:2px solid rgba(255,212,71,.38);box-shadow:0 22px 55px rgba(0,0,0,.34),inset 0 1px 0 rgba(255,255,255,.08);overflow:hidden}
.agenda-card:nth-child(even){border-color:rgba(53,230,255,.42)}
.agendaNum{font-size:96px;font-weight:1000;color:#ffd447;opacity:.96;letter-spacing:-3px}
.agenda-title{min-height:76px;font-size:58px;line-height:1.05;font-weight:1000;letter-spacing:2px;color:#fff;white-space:nowrap}
.agenda-sub{margin-top:14px;font-size:29px;line-height:1.18;font-weight:900;letter-spacing:2px;color:#9fc2d5}
.verdictGrid{position:absolute;left:260px;right:260px;top:535px;height:860px;display:grid;grid-template-columns:1fr 1fr;grid-template-rows:1fr 1fr;gap:34px 38px}
.verdict-card{border-radius:28px;padding:30px 38px;background:linear-gradient(145deg,rgba(11,34,57,.985),rgba(3,15,28,.995));border:2px solid rgba(255,255,255,.12);box-shadow:0 22px 55px rgba(0,0,0,.34);overflow:hidden}
.verdictTitle{font-size:38px;font-weight:1000;letter-spacing:3px;color:#35e6ff;margin-bottom:14px}.verdictTitle.green{color:#52f28c}.verdictTitle.gold{color:#ffd447}.verdictTitle.red{color:#ff6475}
.summaryRow{height:96px;display:grid;grid-template-columns:82px 1fr 250px;align-items:center;gap:18px;border-top:1px solid rgba(255,255,255,.075)}
.summaryRow:first-of-type{border-top:0}.clubBadge{width:66px;height:66px;object-fit:contain}.clubBadgeFallback{width:66px;height:66px}.sumName{font-size:29px;font-weight:1000;letter-spacing:.6px;color:#fff;white-space:nowrap;overflow:hidden;text-overflow:ellipsis}.sumVal{font-size:31px;font-weight:1000;text-align:right;color:#ffd447}
.gameplan{position:absolute;left:260px;right:260px;top:1450px;height:470px;border-radius:32px;padding:32px 44px;background:linear-gradient(100deg,rgba(0,245,160,.10),rgba(0,217,245,.09));border:2px solid rgba(53,230,255,.28);box-shadow:0 22px 55px rgba(0,0,0,.28)}
.gameplanHead{text-align:center;font-size:42px;font-weight:1000;letter-spacing:5px;color:#ffd447;margin-bottom:30px}
.planRow{height:300px;display:grid;grid-template-columns:1fr 92px 1fr 92px 1fr 92px 1.15fr 92px 1.35fr;align-items:stretch;gap:0}
.plan-step{border-radius:24px;background:rgba(4,18,32,.96);border:2px solid rgba(255,255,255,.13);display:flex;flex-direction:column;align-items:center;justify-content:center;text-align:center;padding:18px 24px;box-shadow:0 15px 32px rgba(0,0,0,.28)}
.planCode{font-size:58px;font-weight:1000;color:#35e6ff;line-height:1}.planLabel{margin-top:18px;font-size:29px;font-weight:1000;letter-spacing:1.5px;color:#fff;line-height:1.1}.plan-step.final .planCode{color:#ffd447}.plan-step.final{border-color:rgba(255,212,71,.38)}.planArrow{display:flex;align-items:center;justify-content:center;font-size:70px;font-weight:1000;color:#ffd447;text-shadow:0 0 18px rgba(255,212,71,.3)}
.prevResultsStrip{
  position:absolute;
  left:78px;
  right:78px;
  bottom:154px;
  height:122px;
  z-index:19;
  display:grid;
  grid-template-columns:290px repeat(var(--vx-result-count),minmax(0,1fr));
  gap:12px;
  align-items:stretch;
}
.prevResultsLabel{
  border-radius:12px;
  display:flex;
  align-items:center;
  justify-content:center;
  padding:0 22px;
  background:linear-gradient(180deg,#f2273f 0%,#d5122a 56%,#ad0b20 100%);
  border:3px solid rgba(255,255,255,.92);
  box-shadow:0 9px 22px rgba(0,0,0,.24),inset 0 1px 0 rgba(255,255,255,.28);
  color:#ffffff;
  font-size:32px;
  font-weight:1000;
  letter-spacing:2px;
  white-space:nowrap;
}
.prevResultsLabel span{color:#ffffff}
.prevResultChip{
  min-width:0;
  border-radius:20px;
  display:flex;
  align-items:center;
  justify-content:center;
  gap:12px;
  padding:7px 9px;
  background:linear-gradient(180deg,#e9feff 0%,#b8f8ff 48%,#7eeefa 100%);
  border:2px solid rgba(250,255,255,.96);
  box-shadow:0 9px 20px rgba(0,0,0,.26),inset 0 1px 0 rgba(255,255,255,.72);
  color:#04121f;
}
.prevResultLogo{
  width:74px;
  height:74px;
  object-fit:contain;
  filter:drop-shadow(0 4px 7px rgba(0,0,0,.20));
}
.prevResultFallback{
  min-width:66px;
  font-size:21px;
  font-weight:1000;
  text-align:center;
}
.prevResultScore{
  min-width:62px;
  text-align:center;
  font-size:40px;
  line-height:1;
  font-weight:1000;
  letter-spacing:-1px;
  color:#04121f;
}
.footer{position:absolute;left:78px;right:78px;bottom:24px;height:118px;border-top:2px solid rgba(255,255,255,.12);display:grid;grid-template-columns:minmax(0,1fr) auto minmax(0,1fr);align-items:center;column-gap:42px;z-index:20;font-size:34px;font-weight:950;letter-spacing:1px;color:#b8cad7}
.footerLeft{justify-self:start;white-space:nowrap}.footerLeft strong{font-size:39px;color:#ffd447}.footerRight{justify-self:end;text-align:right;display:flex;flex-direction:column;gap:7px;white-space:nowrap}.footerRight .updated{font-size:32px;font-weight:1000;color:#fff}.footerRight .season{font-size:27px;color:#9fb4c4}.footerSocial{justify-self:center;display:flex;align-items:center;gap:54px;color:#fff;font-size:36px;font-weight:1000;letter-spacing:1px}.socialPair{display:inline-flex;align-items:center;gap:15px;white-space:nowrap}.ytIcon{display:inline-flex;width:58px;height:41px}.ytIcon svg{width:58px;height:41px}.xIcon{font-size:45px;line-height:1;color:#fff}
'''


OUTRO_STYLE = r'''
.outroScene{background:
radial-gradient(circle at 86% 8%,rgba(255,212,71,.16),transparent 24%),
radial-gradient(circle at 12% 88%,rgba(53,230,255,.13),transparent 30%),
linear-gradient(140deg,#04111e 0%,#091b2d 45%,#020711 100%)}
.outroScene .topHeader{top:54px;height:244px}
.outroScene .brandLock{min-width:unset}
.outroScene .logoBox{width:560px;height:220px}
.outroScene .logoBox .vortex-logo{max-width:540px!important;max-height:205px!important}
.outroScene .channelName{font-size:74px}
.outroScene .channelSub{font-size:24px;letter-spacing:5px}
.outroScene .gwPill{font-size:32px;padding:18px 28px}
.outroScene .summaryStage{position:absolute;inset:0}
.outroScene .outroGrid{position:absolute;left:140px;right:140px;top:355px;bottom:292px;display:grid;grid-template-columns:1.03fr .97fr;gap:42px}
.outroScene .summaryPanel{position:relative;border-radius:36px;padding:46px 48px 42px;background:linear-gradient(145deg,rgba(8,26,43,.96),rgba(2,13,24,.98));border:2px solid rgba(255,255,255,.11);box-shadow:0 24px 60px rgba(0,0,0,.36),inset 0 1px 0 rgba(255,255,255,.08);overflow:hidden}
.outroScene .summaryPanel:before{content:"";position:absolute;inset:0;background:linear-gradient(135deg,rgba(255,255,255,.03),transparent 36%,rgba(255,255,255,.02) 65%,transparent 100%);pointer-events:none}
.outroScene .todayPanel{border-color:rgba(255,212,71,.28)}
.outroScene .tomorrowPanel{border-color:rgba(53,230,255,.24)}
.outroScene .panelTag{font-size:28px;font-weight:1000;letter-spacing:6px;color:#9fc2d5}
.outroScene .panelHeadline{margin-top:18px;min-height:92px;font-size:74px;line-height:.96;font-weight:1000;letter-spacing:2px;color:#fff;text-transform:uppercase}
.outroScene .lineStack{margin-top:30px;display:flex;flex-direction:column;gap:22px}
.outroScene .typeLine{position:relative;min-height:122px;display:flex;align-items:center;padding:0 34px;border-radius:26px;background:linear-gradient(90deg,rgba(255,255,255,.06),rgba(255,255,255,.02));border:2px solid rgba(255,255,255,.07);box-shadow:inset 0 1px 0 rgba(255,255,255,.04)}
.outroScene .todayPanel .typeLine{border-color:rgba(255,212,71,.18)}
.outroScene .tomorrowPanel .typeLine{border-color:rgba(53,230,255,.17)}
.outroScene .lineText{display:block;font-size:58px;line-height:1.02;font-weight:1000;letter-spacing:1.2px;color:#fff;white-space:nowrap;text-transform:uppercase}
.outroScene .tomorrowPanel .lineText{font-size:58px}
.outroScene .lineText{text-shadow:0 5px 18px rgba(0,0,0,.34)}
.outroScene .todayPanel .typeLine:nth-child(1) .lineText{color:#ffd447}
.outroScene .todayPanel .typeLine:nth-child(2) .lineText{color:#35e6ff}
.outroScene .todayPanel .typeLine:nth-child(3) .lineText{color:#52f28c}
.outroScene .tomorrowPanel .typeLine:nth-child(1) .lineText{color:#52f28c}
.outroScene .tomorrowPanel .typeLine:nth-child(2) .lineText{color:#35e6ff}
.outroScene .tomorrowPanel .typeLine:nth-child(3) .lineText{color:#ff7b8a}
.outroScene .tomorrowPanel .typeLine:nth-child(4) .lineText{color:#ffd447}
.outroScene .tomorrowPanel .typeLine:nth-child(5) .lineText{color:#d7a8ff}
.outroScene .todayPanel .panelHeadline{color:#ffd447}
.outroScene .todayPanel .take-head{display:flex;flex-direction:column;align-items:flex-start;gap:8px;min-height:150px}
.outroScene .todayPanel .take-head-1{display:block;font-size:70px;line-height:.92;font-weight:1000;letter-spacing:2px;color:#ffd447;text-transform:uppercase}
.outroScene .todayPanel .take-head-2{display:block;font-size:66px;line-height:.92;font-weight:1000;letter-spacing:2px;color:#ffffff;text-transform:uppercase}
.outroScene .todayPanel .typeLine{min-height:152px;align-items:center;padding:22px 34px}
.outroScene .todayPanel .lineBlock{display:flex;flex-direction:column;align-items:flex-start;justify-content:center;gap:10px;width:100%}
.outroScene .todayPanel .lineText{font-size:42px;line-height:1;font-weight:1000;letter-spacing:1.3px;white-space:normal;text-transform:uppercase}
.outroScene .todayPanel .lineSub{display:block;font-size:38px;line-height:1;font-weight:1000;letter-spacing:1.1px;color:#ffffff;white-space:normal;text-transform:uppercase;text-shadow:0 5px 18px rgba(0,0,0,.34)}
.outroScene .todayPanel .take-1 .lineText{color:#ffd447}
.outroScene .todayPanel .take-2 .lineText{color:#35e6ff}
.outroScene .todayPanel .take-3 .lineText{color:#ff9f43}
.outroScene .todayPanel .take-3 .lineSub{color:#ffbf7a}
.outroScene .todayPanel .takeawayFooter{margin-top:28px;padding:22px 28px 24px;border-radius:24px;background:linear-gradient(145deg,rgba(255,255,255,.06),rgba(255,255,255,.02));border:2px solid rgba(255,212,71,.24);box-shadow:inset 0 1px 0 rgba(255,255,255,.05)}
.outroScene .todayPanel .takeawayCall{font-size:46px;line-height:1;font-weight:1000;letter-spacing:1.6px;color:#ffd447;text-transform:uppercase;text-shadow:0 5px 18px rgba(0,0,0,.34)}
.outroScene .todayPanel .takeawaySub{margin-top:10px;font-size:34px;line-height:1.06;font-weight:1000;letter-spacing:1.2px;color:#ffffff;text-transform:uppercase;text-shadow:0 5px 18px rgba(0,0,0,.34)}
.outroScene .tomorrowPanel .panelHeadline{color:#35e6ff}
.outroScene .tomorrowPanel .tom-head{display:flex;flex-direction:column;align-items:flex-start;gap:10px;min-height:176px}
.outroScene .tomorrowPanel .tom-head-1{display:block;font-size:52px;line-height:.94;font-weight:1000;letter-spacing:2px;color:#35e6ff;text-transform:uppercase}
.outroScene .tomorrowPanel .tom-head-2{display:block;font-size:60px;line-height:.92;font-weight:1000;letter-spacing:1.6px;color:#ffffff;text-transform:uppercase;max-width:100%}
.outroScene .tomorrowPanel .tom-sub{margin-top:18px;font-size:28px;line-height:1.08;font-weight:1000;letter-spacing:3px;color:#9fefff;text-transform:uppercase;white-space:normal}
.outroScene .tomorrowPanel .lineStack{margin-top:24px}
.outroScene .tomorrowPanel .typeLine{min-height:106px;padding:0 30px}
.outroScene .tomorrowPanel .lineText{font-size:48px;letter-spacing:1px;white-space:normal}
.outroScene .footer{bottom:28px}
.outroScene .ctaStage{position:absolute;inset:0;display:flex;flex-direction:column;align-items:center;justify-content:center;gap:28px;opacity:0;pointer-events:none;z-index:30}
.outroScene .ctaCopy{font-size:76px;line-height:.98;font-weight:1000;letter-spacing:1.8px;text-align:center;color:#fff;text-shadow:0 12px 26px rgba(0,0,0,.38)}
.outroScene .ctaButtons{display:flex;gap:34px;align-items:center;justify-content:center}
.outroScene .ctaBtn{min-width:530px;height:138px;border-radius:30px;display:flex;align-items:center;justify-content:center;padding:0 34px;font-size:54px;font-weight:1000;letter-spacing:1.8px;color:#fff;box-shadow:0 18px 40px rgba(0,0,0,.38),inset 0 1px 0 rgba(255,255,255,.14)}
.outroScene .ctaBtn.subscribe{background:linear-gradient(180deg,#ff304b,#c70f25);border:3px solid rgba(255,255,255,.88)}
.outroScene .ctaBtn.follow{background:linear-gradient(180deg,#0e3a54,#0a2235);border:3px solid rgba(53,230,255,.58)}
.outroScene .outroHandle{font-size:50px;font-weight:1000;letter-spacing:2px;color:#ffd447;text-shadow:0 8px 22px rgba(0,0,0,.42)}
.outroScene .logoStage{position:absolute;inset:0;display:flex;flex-direction:column;align-items:center;justify-content:center;gap:40px;opacity:0;pointer-events:none;z-index:40}
.outroScene .logoHolder{width:1450px;height:720px;display:flex;align-items:center;justify-content:center}
.outroScene .logoHolder .vortex-logo{display:block;max-width:100%!important;max-height:100%!important;width:auto!important;height:auto!important;object-fit:contain!important;filter:drop-shadow(0 0 26px rgba(53,230,255,.22)) drop-shadow(0 0 50px rgba(255,212,71,.14)) drop-shadow(0 24px 45px rgba(0,0,0,.52))}
.outroScene .logoHolder .logo-fallback{font-size:164px;font-weight:1000;letter-spacing:2px;color:#fff}
.outroScene .logoHolder .logo-fallback span{color:#ffd447}
.outroScene .bgFade{position:absolute;inset:0;background:#000;opacity:0;pointer-events:none;z-index:20}
'''

INTRO = Template(r'''<!DOCTYPE html><html><head><meta charset="UTF-8"><style>$style</style></head><body><div class="scene introScene">
<header class="topHeader"><div class="brandLock"><div class="logoBox">$logo</div><div class="brandWords"><div class="channelName">FPL VORTEX</div><div class="channelSub">GAMEWEEK DATA • MODEL • DECISIONS</div></div></div><div class="gwPill">GAMEWEEK $gw</div></header>
<div class="mainHeading">GAMEWEEK $gw <span>• VORTEX AGENDA</span></div><div class="headingRule"></div>
<main class="agendaWrap">
<div class="agenda-card agenda-1"><div class="agendaNum">01</div><div><div class="agenda-title" data-vx-type="GAMEWEEK REVIEW">GAMEWEEK REVIEW</div><div class="agenda-sub">MY PREVIOUS GAMEWEEK SQUAD • PLAYER BY PLAYER</div></div></div>
<div class="agenda-card agenda-2"><div class="agendaNum">02</div><div><div class="agenda-title" data-vx-type="FIXTURE DIFFICULTY">FIXTURE DIFFICULTY</div><div class="agenda-sub">ALL 20 TEAMS • NEXT 6 GAMEWEEKS</div></div></div>
<div class="agenda-card agenda-3"><div class="agendaNum">03</div><div><div class="agenda-title" data-vx-type="PROJECTED GOALS">PROJECTED GOALS</div><div class="agenda-sub">ATTACKING OUTLOOK • NEXT 6 GAMEWEEKS</div></div></div>
<div class="agenda-card agenda-4"><div class="agendaNum">04</div><div><div class="agenda-title" data-vx-type="CLEAN SHEET ODDS">CLEAN SHEET ODDS</div><div class="agenda-sub">DEFENSIVE OUTLOOK • NEXT 6 GAMEWEEKS</div></div></div>
</main>$results_strip<footer class="footer"><div class="footerLeft"><strong>FPL VORTEX</strong> &nbsp;|&nbsp; GAMEWEEK $gw</div><div class="footerSocial">$social</div><div class="footerRight"><span class="updated">UPDATED: $updated</span><span class="season">2026/27 SEASON</span></div></footer></div></body></html>''').substitute(style=STYLE,logo=vortex_logo_html,gw=int(NEXT_GW),social=SOCIAL,updated=str(VX_FOOTER_UPDATED_DATE),results_strip=PREVIOUS_RESULTS_STRIP)

OUTRO = Template(r'''<!DOCTYPE html><html><head><meta charset="UTF-8"><style>$style</style></head><body><div class="scene outroScene">
<div class="bgFade" id="bgFade"></div>
<div class="summaryStage" id="summaryStage">
<header class="topHeader"><div class="brandLock"><div class="logoBox">$logo</div><div class="brandWords"><div class="channelName">FPL VORTEX</div><div class="channelSub">DAILY DATA • PLAYER DECISIONS • RANK CLIMB</div></div></div><div class="gwPill">GAMEWEEK $gw</div></header>
<div class="outroGrid">
<section class="summaryPanel todayPanel" id="todayPanel"><div class="panelTag">DAY 1 RECAP</div><div class="panelHeadline take-head"><span class="take-head-1" data-vx-type="GAMEWEEK $gw">GAMEWEEK $gw</span><span class="take-head-2" data-vx-type="DATA SUMMARY">DATA SUMMARY</span></div><div class="lineStack"><div class="typeLine take-1"><div class="lineBlock"><span class="lineText" data-vx-type="BEST ATTACKING FIXTURES">BEST ATTACKING FIXTURES</span><span class="lineSub">$today_attack_teams</span></div></div><div class="typeLine take-2"><div class="lineBlock"><span class="lineText" data-vx-type="BEST DEFENSIVE FIXTURES">BEST DEFENSIVE FIXTURES</span><span class="lineSub">$today_def_teams</span></div></div><div class="typeLine take-3"><div class="lineBlock"><span class="lineText" data-vx-type="FIXTURE WARNING">FIXTURE WARNING</span><span class="lineSub">$today_warn_teams</span></div></div></div><div class="takeawayFooter"><div class="takeawayCall" data-vx-type="DATA OVER HYPE">DATA OVER HYPE</div><div class="takeawaySub" data-vx-type="SUBSCRIBE FOR WEEKLY FPL DATA">SUBSCRIBE FOR WEEKLY FPL DATA</div></div></section>
<section class="summaryPanel tomorrowPanel" id="tomorrowPanel"><div class="panelTag">TOMORROW VIDEO</div><div class="panelHeadline tom-head"><span class="tom-head-1" data-vx-type="TOMORROW:">TOMORROW:</span><span class="tom-head-2" data-vx-type="FPL GW $gw PLAYER DATA AND RANK IMPACT">FPL GW$gw PLAYER DATA &amp; RANK IMPACT</span></div><div class="tom-sub" data-vx-type="MINUTES TOP 1K EO DIFFERENTIALS PLAYER TRAPS">MINUTES • TOP 1K EO • DIFFERENTIALS • PLAYER TRAPS</div><div class="lineStack"><div class="typeLine tom-1"><span class="lineText" data-vx-type="EXPECTED MINUTES">EXPECTED MINUTES</span></div><div class="typeLine tom-2"><span class="lineText" data-vx-type="TOP 1K EO">TOP 1K EO</span></div><div class="typeLine tom-3"><span class="lineText" data-vx-type="RANK GAIN OR LOSS">RANK GAIN OR LOSS</span></div><div class="typeLine tom-4"><span class="lineText" data-vx-type="DIFFERENTIALS">DIFFERENTIALS</span></div><div class="typeLine tom-5"><span class="lineText" data-vx-type="PLAYER TRAPS">PLAYER TRAPS</span></div></div></section>
</div>
$results_strip<footer class="footer outroFooter"><div class="footerLeft"><strong>FPL VORTEX</strong> &nbsp;|&nbsp; GAMEWEEK $gw</div><div class="footerSocial">$social</div><div class="footerRight"><span class="updated">UPDATED: $updated</span><span class="season">2026/27 SEASON</span></div></footer>
</div>
<section class="ctaStage" id="ctaStage"><div class="ctaCopy cta-copy-1"><span data-vx-type="SUBSCRIBE FOR DAY 2">SUBSCRIBE FOR DAY 2</span></div><div class="ctaButtons"><div class="ctaBtn subscribe cta-btn-sub">SUBSCRIBE</div><div class="ctaBtn follow cta-btn-follow">FOLLOW</div></div><div class="ctaCopy cta-copy-2"><span data-vx-type="PLAYER DATA AND RANK IMPACT">PLAYER DATA & RANK IMPACT</span></div><div class="outroHandle cta-handle">$handle</div></section>
<section class="logoStage" id="logoStage"><div class="logoHolder">$logo</div><div class="outroHandle logo-handle">$handle</div></section>
</div></body></html>''').substitute(style=STYLE+"\n"+OUTRO_STYLE,logo=vortex_logo_html,gw=int(NEXT_GW),social=SOCIAL,updated=str(VX_FOOTER_UPDATED_DATE),handle=str(globals().get("CHANNEL_HANDLE", "@FPLVORTEX")).strip(),today_attack_teams=TODAY_ATTACK_TEAMS,today_def_teams=TODAY_DEF_TEAMS,today_warn_teams=TODAY_WARN_TEAMS,results_strip=PREVIOUS_RESULTS_STRIP)

INTRO = vx_apply_pl_lockup(INTRO)
OUTRO = vx_apply_pl_lockup(OUTRO)
INTRO = vx_polish_branding(INTRO)
OUTRO = vx_polish_branding(OUTRO)
if PREVIOUS_RESULTS_COUNT:
    print(
        f"✅ Previous-GW results strip: GW{PREVIOUS_RESULTS_GW} • "
        f"{PREVIOUS_RESULTS_COUNT} completed fixtures • INTRO + OUTRO"
    )
else:
    print("ℹ️ Previous-GW results strip: no completed fixture results available yet")
INTRO_HTML.write_text(INTRO,encoding="utf-8")
OUTRO_HTML.write_text(OUTRO,encoding="utf-8")
print("✅ INTRO HTML:",INTRO_HTML)
print("✅ OUTRO HTML:",OUTRO_HTML)
if "export_4k_png" not in globals(): raise NameError("Run Cell 13 first — export_4k_png missing.")
await export_4k_png(INTRO_HTML,INTRO_PNG)
await export_4k_png(OUTRO_HTML,OUTRO_PNG)
print("✅ INTRO + OUTRO TRUE 4K")
display(IPImage(filename=str(INTRO_PNG)))
display(IPImage(filename=str(OUTRO_PNG)))
```

```python
# ============================================================
# CELL 14B — ORIGINAL WEEKLY FDR / PROJECTED GOALS / CLEAN SHEET DESIGN
# IMPROVED MODEL OUTPUTS ONLY • PRESENTATION RESTORED IN PLACE
# ============================================================
from pathlib import Path
from string import Template
from html import escape
import ast
import re
import numpy as np
import pandas as pd

for _vx_required_presentation in (
    "fdr_table", "team_projection_df", "fixture_projection_df",
    "NEXT_GW", "team_meta_by_id", "team_meta_by_short",
    "vortex_logo_html", "VX_FOOTER_UPDATED_DATE", "PL_BUG_HTML",
    "PL_BUG_CSS", "vx_apply_pl_lockup", "export_4k_png",
    "vx_phase4_fdr_colour_band", "_vx_numeric_fdr_values",
):
    if _vx_required_presentation not in globals():
        raise RuntimeError(
            f"Run Cell 14 before Cell 14B — missing {_vx_required_presentation}"
        )

HTML_DIR = Path(HTML_WORK_DIR)
PNG_DIR = Path(SLIDE_DIR)
HTML_DIR.mkdir(parents=True, exist_ok=True)
PNG_DIR.mkdir(parents=True, exist_ok=True)
XG_HTML_PATH = HTML_DIR / "projected_goals_scene_4k.html"
CS_HTML_PATH = HTML_DIR / "clean_sheet_scene_4k.html"
FDR_HTML_PATH = HTML_DIR / "fdr_scene_4k.html"
XG_PNG_PATH = PNG_DIR / "projected_goals_scene_4k.png"
CS_PNG_PATH = PNG_DIR / "clean_sheet_scene_4k.png"
FDR_PNG_PATH = PNG_DIR / "fdr_scene_4k.png"
MODEL_NEXT_GW = int(NEXT_GW)
team_meta = {int(team_id): dict(meta) for team_id, meta in team_meta_by_id.items()}

def finite(value, default=0.0):
    try:
        number = float(value)
        return number if np.isfinite(number) else float(default)
    except (TypeError, ValueError):
        return float(default)


# ============================================================
# 2. REQUIRE CANONICAL UPSTREAM DATA
# ============================================================
for required_name in ["fdr_table","NEXT_GW","team_meta_by_id","team_meta_by_short","vortex_logo_html","HTML_WORK_DIR","VX_FOOTER_UPDATED_DATE"]:
    if required_name not in globals():
        raise NameError(f"{required_name} is not loaded. Run Cells 0–5 in order first.")
required_columns={
    "team_id","team","team_short","fixtures",
    "attack_fdr","defence_fdr","average_attack_fdr","average_defence_fdr","fdr_source"
}
missing_columns=required_columns-set(fdr_table.columns)
if missing_columns: raise ValueError("fdr_table is missing required model-FDR columns: "+str(sorted(missing_columns)))
if not fdr_table["fdr_source"].astype(str).eq("PHASE4_MODEL_ONLY").all():
    raise RuntimeError("FDR slide refuses any non-Phase4 FDR source")
df=(fdr_table.copy().sort_values(["average_attack_fdr","team_short"],ascending=[True,True],na_position="last").reset_index(drop=True))
GW_HEADERS=[f"GW{int(NEXT_GW)+offset}" for offset in range(6)]
print("✅ FDR source: Cell 9 / Phase 4B model-only attack + defence FDR")
print("✅ No official FPL difficulty and no new blend used")
print("✅ Next GW:",NEXT_GW); print("✅ Horizon: 6 GWs"); print("✅ Teams:",len(df)); print("✅ Sorted by ATTACK FDR: EASY → HARD")


# ============================================================
# 4. PROJECT PATHS
# ============================================================

HTML_DIR = Path(HTML_WORK_DIR)
HTML_DIR.mkdir(parents=True, exist_ok=True)


HTML_PATH = (
    HTML_DIR
    / "fdr_scene_4k.html"
)


PREVIEW_PATH = (
    HTML_DIR
    / "fdr_preview_fit.html"
)


# ============================================================
# 5. CANONICAL VISUAL ASSETS
# ============================================================
print("✅ Reusing canonical VORTEX logo and club crests")


# ============================================================
# 9. DATA HELPERS
# ============================================================

def to_list(
    value
):

    if isinstance(
        value,
        list
    ):

        return value


    if isinstance(
        value,
        tuple
    ):

        return list(
            value
        )


    if value is None:

        return []


    if isinstance(
        value,
        str
    ):

        text = (
            value.strip()
        )


        if not text:

            return []


        try:

            parsed = (
                ast.literal_eval(
                    text
                )
            )


            if isinstance(
                parsed,
                (
                    list,
                    tuple,
                )
            ):

                return list(
                    parsed
                )


        except Exception:

            pass


        return [
            text
        ]


    try:

        if pd.isna(
            value
        ):

            return []


    except Exception:

        pass


    return [
        value
    ]


def model_fdr_visual_class(value):
    """Presentation colour only; decision math remains the exact Phase-4 FDR values."""
    values = _vx_numeric_fdr_values(value)
    if not values:
        return "fdr-none"
    # DGW colour is presentation-only; Attack and Defence values are never blended together.
    visual_value = float(np.mean(values))
    band = str(vx_phase4_fdr_colour_band(visual_value))
    return {
        "Dark green": "fdr-1",
        "Green": "fdr-2",
        "Grey": "fdr-3",
        "Orange": "fdr-4",
        "Red": "fdr-5",
    }.get(band, "fdr-none")








# ============================================================
# 10. BUILD 20 DISPLAY ROWS
#     6 GWS + AVG
# ============================================================

render_rows = []


for _, row in df.iterrows():

    fixtures_list = (
        to_list(
            row["fixtures"]
        )
    )


    attack_fdr_values = to_list(row["attack_fdr"])
    defence_fdr_values = to_list(row["defence_fdr"])


    cells = []


    for i in range(6):

        cells.append(
            {

                "fixture":

                    (
                        fixtures_list[i]

                        if i < len(
                            fixtures_list
                        )

                        else ""
                    ),

                "attack_fdr":
                    attack_fdr_values[i] if i < len(attack_fdr_values) else "",

                "defence_fdr":
                    defence_fdr_values[i] if i < len(defence_fdr_values) else "",
            }
        )


    team_id = int(
        row["team_id"]
    )


    team_short = (

        str(
            row["team_short"]
        )

        .upper()
        .strip()
    )


    team_meta = (
        team_meta_by_id
        .get(
            team_id
        )
    )


    if team_meta is None:

        team_meta = (

            team_meta_by_short
            .get(
                team_short,
                {}
            )
        )


    render_rows.append(
        {

            "team_id":
                team_id,

            "team":
                str(
                    row["team"]
                ),

            "team_short":
                team_short,

            "team_badge":

                (
                    team_meta.get(
                        "badge",
                        ""
                    )

                    if team_meta

                    else ""
                ),

            "average_attack_fdr": row["average_attack_fdr"],
            "average_defence_fdr": row["average_defence_fdr"],

            "cells":
                cells,
        }
    )


print(
    "✅ Display rows:",
    len(render_rows)
)


# ============================================================
# 11. FIND FIXTURE OPPONENTS
# ============================================================

def find_fixture_opponents(
    fixture_value
):

    fixture_text = (

        format_fdr_value(
            fixture_value
        )

        .upper()
        .strip()
    )


    if (
        not fixture_text
        or fixture_text == "BLANK"
    ):
        return []


    matches = []


    for short_name, meta in team_meta_by_short.items():

        short_name = (
            str(short_name)
            .upper()
            .strip()
        )

        full_name = (
            str(meta.get("name", ""))
            .upper()
            .strip()
        )


        labels = [short_name]

        if full_name:
            labels.append(full_name)


        for label in labels:

            match = re.search(
                rf"(?<![A-Z0-9]){re.escape(label)}\s*[\(\[]([HA])[\)\]]",
                fixture_text,
            )


            if match:

                matches.append(
                    (
                        match.start(),
                        short_name,
                        match.group(1),
                        meta,
                    )
                )

                break


    matches.sort(
        key=lambda item: item[0]
    )


    results = []


    for _, short_name, venue, meta in matches:

        results.append(
            {
                "short": short_name,
                "venue": venue,
                "badge": meta.get(
                    "badge",
                    ""
                ),
            }
        )


    return results


# ============================================================
# 12. BUILD EACH TEAM ROW
# ============================================================

def build_team_row(
    item
):

    team_name = escape(
        str(item["team_short"]).upper().strip()
    )


    team_badge = (
        item.get(
            "team_badge",
            ""
        )
    )


    if team_badge:

        own_badge_html = (

            '<img '
            'class="team-badge" '
            'src="'

            + team_badge

            + '" '
            'alt="">'
        )


    else:

        own_badge_html = (

            '<div '
            'class="team-badge-placeholder">'
            '</div>'
        )


    fixture_cards = []


    for cell in item["cells"]:

        attack_text = escape(format_fdr_value(cell["attack_fdr"]))
        defence_text = escape(format_fdr_value(cell["defence_fdr"]))
        attack_cls = model_fdr_visual_class(cell["attack_fdr"])
        defence_cls = model_fdr_visual_class(cell["defence_fdr"])


        opponents = (
            find_fixture_opponents(
                cell["fixture"]
            )
        )

        fixture_text = (
            escape(
                " / ".join(
                    f'{opponent["short"]}[{opponent["venue"]}]'
                    for opponent in opponents[:2]
                )
            )
            if opponents
            else "—"
        )


        badges_html = []


        for opponent in opponents[:2]:

            badge = (
                opponent.get(
                    "badge",
                    ""
                )
            )


            if badge:

                badges_html.append(

                    '<img '
                    'class="opponent-badge" '
                    'src="'

                    + badge

                    + '" '
                    'alt="">'
                )


        fixture_cards.append(
            f'''

            <div class="fixture-card">

                <div class="fixture-top">
                    <div class="opponent-badges">{''.join(badges_html)}</div>
                    <div class="fixture-name">{fixture_text}</div>
                </div>

                <div class="fdr-dual">
                    <div class="fdr-pill {attack_cls}">
                        <span class="fdr-label">A</span><span class="fdr-value">{attack_text}</span>
                    </div>
                    <div class="fdr-pill {defence_cls}">
                        <span class="fdr-label">D</span><span class="fdr-value">{defence_text}</span>
                    </div>
                </div>

            </div>

            '''
        )


    # --------------------------------------------------------
    # SIX-GW ATTACK + DEFENCE AVERAGES — KEPT SEPARATE
    # --------------------------------------------------------
    avg_attack = item.get("average_attack_fdr")
    avg_defence = item.get("average_defence_fdr")

    if avg_attack is None or pd.isna(avg_attack):
        avg_attack_text, avg_attack_class = "—", "fdr-none"
    else:
        avg_attack_text = f"{float(avg_attack):.1f}"
        avg_attack_class = model_fdr_visual_class(float(avg_attack))

    if avg_defence is None or pd.isna(avg_defence):
        avg_defence_text, avg_defence_class = "—", "fdr-none"
    else:
        avg_defence_text = f"{float(avg_defence):.1f}"
        avg_defence_class = model_fdr_visual_class(float(avg_defence))

    return f'''

    <div class="team-row">


        <div class="team-cell">

            {own_badge_html}

            <div class="team-name">

                {team_name}

            </div>

        </div>


        {''.join(fixture_cards)}


        <div class="avg-cell">
            <div class="fdr-pill {avg_attack_class}">
                <span class="fdr-label">A</span><span class="fdr-value">{avg_attack_text}</span>
            </div>
            <div class="fdr-pill {avg_defence_class}">
                <span class="fdr-label">D</span><span class="fdr-value">{avg_defence_text}</span>
            </div>
        </div>


    </div>

    '''


# ============================================================
# 13. TWO PANELS
#
# LEFT  = 10 EASIEST
# RIGHT = NEXT 10 / HARDEST
# ============================================================

def build_fdr_panel(
    rows
):
    rows_html = "".join(
        build_team_row(row)
        for row in rows
    )

    gw_header_html = "".join(
        f"<div>{escape(label)}</div>"
        for label in GW_HEADERS
    )

    return f"""
    <section class="table-panel">
        <div class="column-header">
            <div class="team-header">TEAM</div>
            {gw_header_html}
            <div>AVG A/D</div>
        </div>
        <div class="table-body">
            {rows_html}
        </div>
    </section>
    """


left_panel_html = (
    build_fdr_panel(
        render_rows[:10]
    )
)


right_panel_html = (
    build_fdr_panel(
        render_rows[10:20]
    )
)


# ============================================================
# 14. DATE
# ============================================================

data_updated = str(VX_FOOTER_UPDATED_DATE)


# ============================================================
# 15. FOOTER SOCIAL ICONS
# ============================================================

youtube_icon = '''

<svg
    class="social-icon youtube-icon"
    viewBox="0 0 48 34"
    aria-hidden="true"
>

    <rect
        x="0"
        y="0"
        width="48"
        height="34"
        rx="8"
        fill="#FF0033"
    ></rect>

    <polygon
        points="20,9 20,25 33,17"
        fill="#FFFFFF"
    ></polygon>

</svg>

'''


x_icon = '''

<svg
    class="social-icon x-icon"
    viewBox="0 0 32 32"
    aria-hidden="true"
>

    <path

        fill="#FFFFFF"

        d="
        M4 4
        L12.9 16.1
        L4.5 28
        H8.4
        L14.8 18.9
        L21.5 28
        H28
        L18.6 15.2
        L26.5 4
        H22.6
        L16.7 12.4
        L10.5 4
        Z

        M9.5 6.7
        H12
        L22.5 25.3
        H20
        Z
        "

    ></path>

</svg>

'''


# ============================================================
# 16. COMPLETE 3840 × 2160 HTML
#
# string.Template means CSS uses normal:
#
# {
# }
#
# NOT:
#
# {{
# }}
#
# This removes the old f-string brace bug.
# ============================================================

html_template = Template(
r'''

<!DOCTYPE html>

<html lang="en">


<head>


<meta charset="UTF-8">


<meta
    name="viewport"
    content="
        width=3840,
        height=2160,
        initial-scale=1.0
    "
>


<title>
FPL VORTEX — Fixture Difficulty
</title>


<style>


/* ========================================================
   RESET
======================================================== */

* {
    box-sizing: border-box;
}


html,
body {

    margin: 0;
    padding: 0;

    width: 3840px;
    height: 2160px;

    overflow: hidden;

    background: #020711;
}


body {

    font-family:
        "Arial Black",
        "Helvetica Neue",
        Arial,
        sans-serif;

    color: #ffffff;
}


/* ========================================================
   SCENE
======================================================== */

.scene {

    position: relative;

    width: 3840px;
    height: 2160px;

    overflow: hidden;

    padding:
        38px
        52px
        28px;

    display: flex;

    flex-direction: column;

    background:

        radial-gradient(
            circle at 84% 7%,
            rgba(0,226,255,.14),
            transparent 31%
        ),

        radial-gradient(
            circle at 2% 93%,
            rgba(0,245,160,.09),
            transparent 28%
        ),

        linear-gradient(
            145deg,
            #020713 0%,
            #07182b 49%,
            #020711 100%
        );
}


/* ========================================================
   TOP ACCENT
======================================================== */

.top-accent {

    position: absolute;

    top: 0;
    left: 0;

    width: 100%;
    height: 10px;

    background:

        linear-gradient(
            90deg,
            #00f5a0 0%,
            #1ac77e 32%,
            #19d5f2 59%,
            transparent 90%
        );

    box-shadow:

        0 0 28px
        rgba(0,245,160,.18);
}


/* ========================================================
   HEADER
======================================================== */

.header {

    height: 150px;

    flex: 0 0 150px;

    display: flex;

    align-items: center;

    justify-content: space-between;
}


/* ========================================================
   LARGE VORTEX LOGO
======================================================== */

.logo-area {

    width: 760px;
    height: 150px;

    display: flex;

    align-items: center;

    justify-content: flex-start;

    overflow: visible;
}


.vortex-logo {

    display: block;

    width: auto;
    height: auto;

    max-width: 660px;
    max-height: 150px;

    object-fit: contain;

    object-position:
        left center;

    filter:

        drop-shadow(
            0 7px 18px
            rgba(0,0,0,.38)
        )

        drop-shadow(
            0 0 24px
            rgba(0,245,160,.14)
        );
}


.logo-fallback {

    font-size: 70px;

    font-weight: 950;

    letter-spacing: -2px;
}


.logo-fallback span {

    color: #23d98d;
}


/* ========================================================
   HEADER RIGHT
======================================================== */

.header-meta {

    text-align: right;

    /* Keep matrix header text clear of the shared 540 px PL lockup. */
    padding-right: 620px;
}


.header-kicker {

    font-family:
        Arial,
        sans-serif;

    font-size: 36px;

    font-weight: 950;

    letter-spacing: 4px;

    color: #e8fbff;

    text-shadow: 0 0 18px rgba(25,213,242,.30);
}


.header-season {

    margin-top: 10px;

    font-family:
        Arial,
        sans-serif;

    font-size: 30px;

    font-weight: 950;

    letter-spacing: 3px;

    color: #55ffd2;
}


/* ========================================================
   HERO
======================================================== */

.hero {

    height: 185px;

    flex: 0 0 185px;

    display: flex;

    align-items: center;

    justify-content: space-between;

    margin-bottom: 16px;
}


.hero-copy {

    min-width: 0;
}


.main-title {

    margin: 0;

    font-size: 112px;

    line-height: .91;

    font-weight: 950;

    letter-spacing: -4px;

    white-space: nowrap;

    color: #ffffff;
}


.main-title span {

    color: #20d9f1;

    text-shadow:

        0 0 26px
        rgba(32,217,241,.13);
}


.subtitle {

    margin-top: 20px;

    font-family:
        Arial,
        sans-serif;

    display: inline-block;

    margin-top: 16px;

    padding: 11px 20px 12px;

    border-radius: 14px;

    border: 1px solid rgba(94,207,255,.52);

    background: linear-gradient(90deg, rgba(28,131,204,.98), rgba(10,74,132,.96));

    font-size: 46px;

    line-height: 1;

    font-weight: 950;

    letter-spacing: 1.8px;

    color: #ffffff;

    text-shadow: 0 2px 10px rgba(0,0,0,.45);
}


/* ========================================================
   MODEL FDR SCALE
======================================================== */

.scale-card {

    min-width: 575px;

    padding:
        22px
        28px;

    border-radius: 18px;

    border:

        1px solid
        rgba(255,255,255,.13);

    background:

        linear-gradient(
            145deg,
            rgba(8,23,40,.97),
            rgba(3,14,27,.97)
        );

    box-shadow:

        0 15px 44px
        rgba(0,0,0,.24);
}


.scale-heading {

    margin-bottom: 13px;

    font-family:
        Arial,
        sans-serif;

    font-size: 29px;

    line-height: 1;

    font-weight: 950;

    letter-spacing: 2px;

    color: #ffffff;
}


.scale-row {

    display: flex;

    align-items: center;

    gap: 9px;
}


.scale-box {

    width: 58px;
    height: 23px;

    border-radius: 20px;
}




.scale-copy {

    margin-left: 11px;

    font-family:
        Arial,
        sans-serif;

    font-size: 27px;

    font-weight: 950;

    color: #ffffff;
}


/* ========================================================
   TWO TABLES
======================================================== */

.tables {

    flex: 1 1 auto;

    min-height: 0;

    display: grid;

    grid-template-columns:
        minmax(0,1fr)
        minmax(0,1fr);

    gap: 24px;
}


.table-panel {

    min-width: 0;
    min-height: 0;

    overflow: hidden;

    border-radius: 25px;

    border:

        1px solid
        rgba(255,255,255,.10);

    background:

        linear-gradient(
            180deg,
            rgba(9,27,47,.985),
            rgba(4,16,30,.995)
        );

    box-shadow:

        0 24px 72px
        rgba(0,0,0,.28);
}


/* ========================================================
   HEADER:
   TEAM + GW1-GW6 + AVG
======================================================== */

.column-header {

    height: 84px;

    padding:
        0
        11px;

    display: grid;

    grid-template-columns:
        300px
        repeat(7,minmax(0,1fr));

    align-items: center;

    text-align: center;

    border-bottom:

        1px solid
        rgba(255,255,255,.11);

    background:

        linear-gradient(
            90deg,
            rgba(1,9,18,.99),
            rgba(7,23,39,.98)
        );

    font-family:
        Arial,
        sans-serif;

    font-size: 28px;

    line-height: 1;

    font-weight: 950;

    letter-spacing: .6px;

    color: #ffffff;

    text-shadow: 0 2px 8px rgba(0,0,0,.45);
}


.team-header {

    padding-left: 15px;

    text-align: left;
}


/* ========================================================
   ROWS
======================================================== */

.table-body {

    padding:
        7px
        11px
        8px;
}


.team-row {

    height: 151px;

    display: grid;

    grid-template-columns:
        300px
        repeat(7,minmax(0,1fr));

    align-items: center;

    border-bottom:

        1px solid
        rgba(255,255,255,.055);
}


.team-row:last-child {

    border-bottom: none;
}


/* ========================================================
   TEAM
======================================================== */

.team-cell {

    min-width: 0;

    height: 100%;

    padding:
        0
        10px;

    display: flex;

    align-items: center;

    gap: 14px;
}


.team-badge {

    width: 88px;
    height: 88px;

    flex:
        0 0 88px;

    object-fit: contain;

    filter:

        drop-shadow(
            0 6px 9px
            rgba(0,0,0,.38)
        );
}


.team-badge-placeholder {

    width: 88px;
    height: 88px;

    flex:
        0 0 88px;

    border-radius: 50%;

    background:

        rgba(255,255,255,.05);
}


.team-name {

    min-width: 0;

    overflow: hidden;

    text-overflow: ellipsis;

    white-space: nowrap;

    font-family:
        "Arial Black",
        Arial,
        sans-serif;

    font-size: 30px;

    line-height: 1;

    font-weight: 950;

    letter-spacing: -.7px;

    color: #ffffff;

    text-shadow:

        0 3px 10px
        rgba(0,0,0,.25);
}


/* ========================================================
   FIXTURE + AVG CARDS
======================================================== */

.fixture-card,
.avg-cell {

    height: 116px;

    margin:
        8px
        4px;

    padding:
        8px
        6px;

    border-radius: 18px;

    display: flex;

    flex-direction: column;

    align-items: center;

    justify-content: center;

    text-align: center;

    border:

        2px solid
        rgba(255,255,255,.12);

    box-shadow:

        inset
        0 1px 0
        rgba(255,255,255,.12);
}


.fixture-top {

    width: 100%;

    min-width: 0;

    display: flex;

    align-items: center;

    justify-content: center;

    gap: 4px;

    padding: 0 2px;
}


.opponent-badges {

    display: flex;

    align-items: center;

    justify-content: center;

    gap: 1px;

    flex:
        0 0 auto;
}


.opponent-badge {

    width: 52px;
    height: 52px;

    object-fit: contain;

    flex: 0 0 52px;

    filter:

        drop-shadow(
            0 3px 5px
            rgba(0,0,0,.35)
        );
}


.fixture-name {

    min-width: 0;

    overflow: hidden;

    text-overflow: ellipsis;

    white-space: nowrap;

    font-family:
        "Arial Black",
        Arial,
        sans-serif;

    flex: 1 1 auto;

    font-size: 30px;

    line-height: 1;

    font-weight: 1000;

    letter-spacing: -.7px;

    color: #ffffff;

    text-shadow: 0 2px 7px rgba(0,0,0,.52);
}


.fdr-dual,
.avg-cell {
    display:flex;
    align-items:center;
    justify-content:center;
    gap:6px;
    width:100%;
}

.fdr-dual { margin-top:7px; }

.fdr-pill {
    min-width:0;
    flex:1 1 0;
    min-height:39px;
    padding:4px 5px;
    border-radius:10px;
    display:flex;
    align-items:center;
    justify-content:center;
    gap:5px;
    border:2px solid rgba(255,255,255,.18);
    font-family:"Arial Black",Arial,sans-serif;
    font-weight:1000;
    line-height:1;
    white-space:nowrap;
}
.fdr-label { font-size:20px; opacity:.86; }
.fdr-value { font-size:28px; letter-spacing:-.7px; }

.fixture-card {
    background:linear-gradient(145deg,#13283a 0%,#0a1a28 100%);
}




/* ========================================================
   BRIGHT FDR BANDS
   1–2 GREEN | >2–3 SKY | >3–4 GOLD | >4–5 RED
======================================================== */

.fdr-1,
.fdr-2 {
    background: linear-gradient(145deg, #45ff77 0%, #12b94b 100%);
    border-color: rgba(154,255,178,.96);
    box-shadow: 0 0 24px rgba(57,255,106,.24), inset 0 1px 0 rgba(255,255,255,.35);
}

.fdr-3 {
    background: linear-gradient(145deg, #69d9ff 0%, #188bc9 100%);
    border-color: rgba(188,237,255,.96);
    box-shadow: 0 0 24px rgba(94,207,255,.23), inset 0 1px 0 rgba(255,255,255,.34);
}

.fdr-4 {
    background: linear-gradient(145deg, #ffe36e 0%, #d99b00 100%);
    border-color: rgba(255,244,174,.98);
    box-shadow: 0 0 24px rgba(255,211,77,.23), inset 0 1px 0 rgba(255,255,255,.35);
}

.fdr-5 {
    background: linear-gradient(145deg, #ff6673 0%, #cf1735 100%);
    border-color: rgba(255,190,197,.97);
    box-shadow: 0 0 24px rgba(255,79,94,.24), inset 0 1px 0 rgba(255,255,255,.32);
}

.fdr-none {
    background: linear-gradient(145deg, #24384b 0%, #122130 100%);
    border-color: rgba(178,199,216,.28);
}

/* High contrast inside the five FDR pills. */
.fdr-pill.fdr-1,.fdr-pill.fdr-2,.fdr-pill.fdr-3,.fdr-pill.fdr-4{color:#06131c;text-shadow:0 1px 0 rgba(255,255,255,.35)}
.fdr-pill.fdr-5{color:#ffffff;text-shadow:0 2px 6px rgba(0,0,0,.58)}


/* ========================================================
   LARGE FOOTER
======================================================== */

.vortex-footer {

    height: 126px;

    flex:
        0 0 126px;

    margin-top: 14px;

    padding:
        0
        34px;

    display: grid;

    grid-template-columns:
        1.15fr
        1.30fr
        1.15fr;

    align-items: center;

    border-radius: 15px;

    border:

        1px solid
        rgba(255,255,255,.09);

    background:

        linear-gradient(
            90deg,
            rgba(1,9,18,.995),
            rgba(8,24,41,.995),
            rgba(1,9,18,.995)
        );

    box-shadow:

        0 -10px 28px
        rgba(0,0,0,.16);

    font-family:
        Arial,
        sans-serif;
}


.footer-section {

    min-width: 0;

    display: flex;

    align-items: center;

    gap: 16px;

    font-size: 31px;

    line-height: 1;

    font-weight: 900;

    letter-spacing: .9px;
}


.footer-left {

    justify-content:
        flex-start;
}


.footer-center {

    justify-content:
        center;

    gap: 48px;
}


.footer-right {

    justify-content:
        flex-end;
}


.footer-label {

    color: #7790a5;

    letter-spacing: 1.4px;
}


.footer-value {

    color: #ffffff;
}


/* ========================================================
   SOCIAL
======================================================== */

.social-group {

    display: flex;

    align-items: center;

    gap: 12px;

    white-space: nowrap;

    color: #ffffff;
}


.social-group.x-social {

    color: #28e3bd;
}


.social-icon {

    display: block;

    flex:
        0 0 auto;
}


.youtube-icon {

    width: 64px;
    height: 46px;
}


.x-icon {

    width: 48px;
    height: 48px;
}


.footer-separator {

    width: 2px;
    height: 48px;

    background:

        linear-gradient(
            transparent,
            rgba(36,225,192,.48),
            transparent
        );
}


</style>


</head>


<body>


<div class="scene">


    <div class="top-accent">
    </div>


    <!-- ====================================================
         HEADER
    ===================================================== -->

    <header class="header">


        <div class="logo-area">

            $vortex_logo_html

        </div>


        <div class="header-meta">

            <div class="header-kicker">

                FIXTURE INTELLIGENCE

            </div>


            <div class="header-season">

                FPL VORTEX • 2026/27

            </div>

        </div>


    </header>


    <!-- ====================================================
         TITLE
    ===================================================== -->

    <section class="hero">


        <div class="hero-copy">


            <h1 class="main-title">

                FIXTURE

                <span>
                    DIFFICULTY
                </span>

            </h1>


            <div class="subtitle">

                $gw_start–$gw_end
                • MODEL ATTACK / DEFENCE FDR
                • A = ATTACK • D = DEFENCE • SORTED ATTACK EASY → HARD

            </div>


        </div>


        <div class="scale-card">


            <div class="scale-heading">

                DIFFICULTY SCALE

            </div>


            <div class="scale-row">


                <div class="scale-box fdr-1">
                </div>


                <div class="scale-box fdr-2">
                </div>


                <div class="scale-box fdr-3">
                </div>


                <div class="scale-box fdr-4">
                </div>


                <div class="scale-box fdr-5">
                </div>


                <div class="scale-copy">

                    EASY → HARD

                </div>


            </div>


        </div>


    </section>


    <!-- ====================================================
         TABLES
    ===================================================== -->

    <main class="tables">


        $left_panel_html


        $right_panel_html


    </main>


    <!-- ====================================================
         FOOTER
    ===================================================== -->

    <footer class="vortex-footer">


        <div class="footer-section footer-left">


            <span class="footer-label">

                SOURCE:

            </span>


            <span class="footer-value">

                VORTEX PHASE 4 MODEL

            </span>


        </div>


        <div class="footer-section footer-center">


            <div class="social-group">


                $youtube_icon


                <span>

                    FPL VORTEX

                </span>


            </div>


            <div class="footer-separator">
            </div>


            <div class="social-group x-social">


                $x_icon


                <span>

                    @FplVortexM

                </span>


            </div>


        </div>


        <div class="footer-section footer-right">


            <span class="footer-label">

                UPDATED:

            </span>


            <span class="footer-value">

                $data_updated

            </span>


        </div>


    </footer>


</div>


</body>


</html>

'''
)


# ============================================================
# 17. INSERT DATA INTO HTML
# ============================================================

html_content = (
    html_template.substitute(

        vortex_logo_html=
            vortex_logo_html,

        left_panel_html=
            left_panel_html,

        right_panel_html=
            right_panel_html,

        youtube_icon=
            youtube_icon,

        x_icon=
            x_icon,

        data_updated=
            data_updated,

        gw_start=
            GW_HEADERS[0],

        gw_end=
            GW_HEADERS[-1],
    )
)


html_content = vx_apply_pl_lockup(html_content)
html_content = vx_polish_branding(html_content)

# ============================================================
# 18. SAVE STANDALONE 4K HTML
# ============================================================

HTML_PATH.write_text(

    html_content,

    encoding="utf-8",
)


print()

print(
    "=============================================="
)

print(
    "✅ FPL VORTEX 4K FDR HTML CREATED"
)

print(
    "=============================================="
)

print(
    "HTML:",
    HTML_PATH
)

print(
    "Canvas: 3840 × 2160"
)

print(
    "Columns: TEAM + " + "–".join(GW_HEADERS) + " + AVG"
)

print(
    "Order: EASY → HARD"
)

print(
    "FDR palette: 1–2 GREEN | >2–3 SKY | >3–4 GOLD | >4–5 RED"
)

print(
    "Logo: transparent padding trimmed + enlarged"
)

print(
    "Footer: enlarged icons + text"
)



# ------------------------------------------------------------
# REAL GAMEWEEK COLUMN LABELS
# Example when MODEL_NEXT_GW == 2: GW2, GW3, GW4, GW5, GW6, GW7
# ------------------------------------------------------------
GW_HEADERS = [
    f"GW{MODEL_NEXT_GW + offset}"
    for offset in range(6)
]


# ============================================================
# 16. COLOR SYSTEM
#
# SAME VORTEX DIRECTION AS FDR:
#
# BEST     = MINT
# GOOD     = GREEN
# NEUTRAL  = SLATE
# WEAK     = AMBER
# POOR     = RED
# ============================================================

def xg_class(
    value
):
    value = finite(value, 0)

    # 0–1.00 RED | 1.01–1.20 GOLD | 1.21–1.50 SKY | >=1.51 GREEN
    if value <= 1.00:
        return "tier-5"
    if value <= 1.20:
        return "tier-4"
    if value <= 1.50:
        return "tier-3"
    return "tier-1"


def cs_class(
    value
):
    """Unified with xg_class: same fixed thresholds/tier scheme as Projected Goals."""
    value = finite(value, 0)
    p_cs = float(np.clip(value / 100.0, 0.0, 1.0)) * 100.0

    # 0-30% RED | 30.01-45% GOLD | 45.01-60% SKY | >=60.01% GREEN
    if p_cs <= 30.0:
        return "tier-5"
    if p_cs <= 45.0:
        return "tier-4"
    if p_cs <= 60.0:
        return "tier-3"
    return "tier-1"


# ============================================================
# 17. HTML ROW BUILDER
# ============================================================

def opponent_badges_html(
    opponent_ids
):

    badges = []


    if not isinstance(
        opponent_ids,
        list
    ):

        return ""


    for opponent_id in (
        opponent_ids[:2]
    ):

        try:

            opponent_id = int(
                opponent_id
            )

        except Exception:

            continue


        badge = (
            team_meta
            .get(
                opponent_id,
                {}
            )
            .get(
                "badge",
                ""
            )
        )


        if badge:

            badges.append(

                '<img '
                'class="opponent-badge" '
                'src="'

                +
                badge

                +
                '" alt="">'
            )


    return "".join(
        badges
    )


def build_metric_row(
    item,
    metric
):

    team_name = escape(

        str(
            item["team"]
        ).upper()
    )


    badge = str(
        item.get(
            "badge",
            ""
        )
    )


    if badge:

        team_badge_html = (

            '<img class="team-badge" src="'

            +
            badge

            +
            '" alt="">'
        )


    else:

        team_badge_html = (

            '<div class="team-badge-placeholder">'
            '</div>'
        )


    cards = []


    for index in range(
        1,
        7
    ):

        opponent = escape(

            str(
                item[
                    f"gw{index}_opponent"
                ]
            )
        )


        opponent_ids = (
            item[
                f"gw{index}_opponent_ids"
            ]
        )


        badges_html = (
            opponent_badges_html(
                opponent_ids
            )
        )


        if metric == "xg":

            value = finite(

                item[
                    f"gw{index}_xg"
                ],

                0
            )


            display_value = (
                "—"
                if opponent == "BLANK"
                else
                f"{value:.2f}"
            )


            tier = (
                xg_class(
                    value
                )
            )


        else:

            value = finite(

                item[
                    f"gw{index}_cs"
                ],

                0
            )


            display_value = (
                "—"
                if opponent == "BLANK"
                else
                f"{value:.0f}%"
            )


            tier = (
                "fdr-none"
                if opponent == "BLANK"
                else cs_class(value)
            )


        cards.append(
            f"""

            <div class="metric-card {tier}">

                <div class="fixture-top">

                    <div class="opponent-badges">
                        {badges_html}
                    </div>

                    <div class="fixture-name">
                        {opponent}
                    </div>

                </div>

                <div class="metric-number">
                    {display_value}
                </div>

            </div>

            """
        )


    if metric == "xg":

        average = finite(
            item[
                "avg_xg"
            ],
            0
        )


        average_text = (
            f"{average:.2f}"
        )


        average_tier = (
            xg_class(
                average
            )
        )


    else:

        average = finite(
            item[
                "avg_cs"
            ],
            0
        )


        average_text = (
            f"{average:.0f}%"
        )


        average_tier = (
            cs_class(
                average
            )
        )


    return f"""

    <div class="team-row">

        <div class="team-cell">

            {team_badge_html}

            <div class="team-name">
                {team_name}
            </div>

        </div>

        {''.join(cards)}

        <div class="avg-cell {average_tier}">

            <span>
                {average_text}
            </span>

        </div>

    </div>

    """


# ============================================================
# 18. PANEL BUILDER
# ============================================================

def build_projection_panel(
    records,
    metric
):
    rows = "".join(
        build_metric_row(record, metric)
        for record in records
    )

    gw_header_html = "".join(
        f"<div>{escape(label)}</div>"
        for label in GW_HEADERS
    )

    return f"""
    <section class="table-panel">
        <div class="column-header">
            <div class="team-header">TEAM</div>
            {gw_header_html}
            <div>AVG</div>
        </div>
        <div class="table-body">
            {rows}
        </div>
    </section>
    """


# ============================================================
# 19. SOCIAL ICONS
# ============================================================

youtube_icon = """

<svg
    class="social-icon youtube-icon"
    viewBox="0 0 48 34"
>

    <rect
        x="0"
        y="0"
        width="48"
        height="34"
        rx="8"
        fill="#FF0033">
    </rect>

    <polygon
        points="20,9 20,25 33,17"
        fill="#FFFFFF">
    </polygon>

</svg>

"""


x_icon = """

<svg
    class="social-icon x-icon"
    viewBox="0 0 32 32"
>

    <path
        fill="#FFFFFF"
        d="
        M4 4
        L12.9 16.1
        L4.5 28
        H8.4
        L14.8 18.9
        L21.5 28
        H28
        L18.6 15.2
        L26.5 4
        H22.6
        L16.7 12.4
        L10.5 4
        Z
        M9.5 6.7
        H12
        L22.5 25.3
        H20
        Z
        ">
    </path>

</svg>

"""


# ============================================================
# 20. SHARED 4K HTML TEMPLATE
#
# string.Template = normal CSS braces.
# No f-string CSS brace bug.
# ============================================================

scene_template = Template(
r"""

<!DOCTYPE html>

<html lang="en">

<head>

<meta charset="UTF-8">

<meta
    name="viewport"
    content="
        width=3840,
        height=2160,
        initial-scale=1
    "
>

<title>
$document_title
</title>

<style>


* {
    box-sizing: border-box;
}


html,
body {

    margin: 0;
    padding: 0;

    width: 3840px;
    height: 2160px;

    overflow: hidden;

    background: #020711;
}


body {

    font-family:
        "Arial Black",
        "Helvetica Neue",
        Arial,
        sans-serif;

    color: #ffffff;
}


/* ========================================================
   FULL 4K SCENE
======================================================== */

.scene {

    position: relative;

    width: 3840px;
    height: 2160px;

    overflow: hidden;

    padding:
        38px
        52px
        28px;

    display: flex;

    flex-direction: column;

    background:

        radial-gradient(
            circle at 84% 7%,
            rgba(0,226,255,.14),
            transparent 31%
        ),

        radial-gradient(
            circle at 2% 93%,
            rgba(0,245,160,.09),
            transparent 28%
        ),

        linear-gradient(
            145deg,
            #020713 0%,
            #07182b 49%,
            #020711 100%
        );
}


.top-accent {

    position: absolute;

    top: 0;
    left: 0;

    width: 100%;
    height: 10px;

    background:

        linear-gradient(
            90deg,
            #00f5a0,
            #1ac77e,
            #19d5f2,
            transparent
        );
}


/* ========================================================
   HEADER
======================================================== */

.header {

    height: 150px;

    flex:
        0 0 150px;

    display: flex;

    align-items: center;

    justify-content:
        space-between;
}


.logo-area {

    width: 760px;
    height: 150px;

    display: flex;

    align-items: center;

    justify-content:
        flex-start;
}


.vortex-logo {

    display: block;

    width: auto;
    height: auto;

    max-width: 660px;
    max-height: 150px;

    object-fit: contain;

    filter:

        drop-shadow(
            0 7px 18px
            rgba(0,0,0,.38)
        )

        drop-shadow(
            0 0 24px
            rgba(0,245,160,.14)
        );
}


.logo-fallback {

    font-size: 70px;

    font-weight: 950;
}


.logo-fallback span {

    color: #22e7c0;
}


.header-meta {

    text-align: right;
}


.header-kicker {

    font-family:
        Arial,
        sans-serif;

    font-size: 36px;

    font-weight: 950;

    letter-spacing: 4px;

    color: #e8fbff;

    text-shadow: 0 0 18px rgba(25,213,242,.30);
}


.header-season {

    margin-top: 10px;

    font-family:
        Arial,
        sans-serif;

    font-size: 30px;

    font-weight: 950;

    letter-spacing: 3px;

    color: #55ffd2;
}


/* ========================================================
   HERO
======================================================== */

.hero {

    height: 185px;

    flex:
        0 0 185px;

    display: flex;

    align-items: center;

    justify-content:
        space-between;

    margin-bottom: 16px;
}


.main-title {

    margin: 0;

    font-size: 108px;

    line-height: .91;

    font-weight: 950;

    letter-spacing: -4px;

    white-space: nowrap;
}


.main-title span {

    color: #20d9f1;
}


.subtitle {

    margin-top: 20px;

    font-family:
        Arial,
        sans-serif;

    display: inline-block;

    margin-top: 16px;

    padding: 11px 20px 12px;

    border-radius: 14px;

    border: 1px solid rgba(94,207,255,.52);

    background: linear-gradient(90deg, rgba(28,131,204,.98), rgba(10,74,132,.96));

    font-size: 46px;

    line-height: 1;

    font-weight: 950;

    letter-spacing: 1.7px;

    color: #ffffff;

    text-shadow: 0 2px 10px rgba(0,0,0,.45);
}


/* ========================================================
   SCALE
======================================================== */

.scale-card {

    min-width: 575px;

    padding:
        22px
        28px;

    border-radius: 18px;

    border:
        1px solid
        rgba(255,255,255,.13);

    background:
        rgba(3,14,27,.97);
}


.scale-heading {

    margin-bottom: 13px;

    font-family:
        Arial,
        sans-serif;

    font-size: 36px;

    font-weight: 950;

    letter-spacing: 2px;

    color: #ffffff;
}


.scale-row {

    display: flex;

    align-items: center;

    gap: 9px;
}


.scale-box {

    width: 58px;
    height: 23px;

    border-radius: 20px;
}


.scale5 { background: #ff4f5e; }
.scale4 { background: #ffd34d; }
.scale3 { background: #5ecfff; }
.scale2 { background: #39ff6a; }
.scale1 { background: #39ff6a; }


.scale-copy {

    margin-left: 11px;

    font-family:
        Arial,
        sans-serif;

    font-size: 27px;

    font-weight: 950;

    color: #ffffff;
}


/* ========================================================
   TABLES
======================================================== */

.tables {

    flex:
        1 1 auto;

    min-height: 0;

    display: grid;

    grid-template-columns:
        minmax(0,1fr)
        minmax(0,1fr);

    gap: 24px;
}


.table-panel {

    min-width: 0;

    overflow: hidden;

    border-radius: 25px;

    border:
        1px solid
        rgba(255,255,255,.10);

    background:

        linear-gradient(
            180deg,
            rgba(9,27,47,.985),
            rgba(4,16,30,.995)
        );
}


.column-header {

    height: 84px;

    padding:
        0
        11px;

    display: grid;

    grid-template-columns:
        300px
        repeat(7,minmax(0,1fr));

    align-items: center;

    text-align: center;

    border-bottom:
        1px solid
        rgba(255,255,255,.11);

    background:
        rgba(1,9,18,.98);

    font-family:
        Arial,
        sans-serif;

    font-size: 28px;

    font-weight: 950;

    letter-spacing: .6px;

    color: #ffffff;

    text-shadow: 0 2px 8px rgba(0,0,0,.45);
}


.team-header {

    padding-left: 15px;

    text-align: left;
}


.table-body {

    padding:
        7px
        11px
        8px;
}


.team-row {

    height: 151px;

    display: grid;

    grid-template-columns:
        300px
        repeat(7,minmax(0,1fr));

    align-items: center;

    border-bottom:
        1px solid
        rgba(255,255,255,.055);
}


.team-row:last-child {

    border-bottom: none;
}


/* ========================================================
   TEAM
======================================================== */

.team-cell {

    min-width: 0;

    height: 100%;

    padding:
        0
        10px;

    display: flex;

    align-items: center;

    gap: 14px;
}


.team-badge {

    width: 76px;
    height: 76px;

    flex:
        0 0 76px;

    object-fit: contain;
}


.team-badge-placeholder {

    width: 76px;
    height: 76px;

    flex:
        0 0 76px;
}


.team-name {

    min-width: 0;

    overflow: hidden;

    text-overflow:
        ellipsis;

    white-space:
        nowrap;

    font-size: 30px;

    font-weight: 950;

    letter-spacing: -.7px;
}


/* ========================================================
   METRIC CARDS
======================================================== */

.metric-card,
.avg-cell {

    height: 116px;

    margin:
        8px
        4px;

    padding:
        8px
        6px;

    border-radius: 18px;

    display: flex;

    flex-direction: column;

    align-items: center;

    justify-content: center;

    text-align: center;

    border:
        2px solid
        rgba(255,255,255,.12);

    box-shadow:
        inset
        0 1px 0
        rgba(255,255,255,.12);
}


.fixture-top {

    width: 100%;

    min-width: 0;

    display: flex;

    align-items: center;

    justify-content: center;

    gap: 4px;

    padding: 0 2px;
}


.opponent-badges {

    display: flex;

    align-items: center;

    gap: 1px;
}


.opponent-badge {

    width: 48px;
    height: 48px;

    object-fit: contain;

    flex: 0 0 42px;
}


.fixture-name {

    min-width: 0;

    overflow: hidden;

    text-overflow:
        ellipsis;

    white-space:
        nowrap;

    flex: 1 1 auto;

    font-size: 31px;

    line-height: 1;

    font-weight: 1000;

    letter-spacing: -.7px;

    color: #ffffff;

    text-shadow: 0 2px 7px rgba(0,0,0,.52);
}


.metric-number,
.avg-cell span {

    margin-top: 8px;

    font-size: 42px;

    line-height: .9;

    font-weight: 950;

    letter-spacing: -1px;

    color: #ffffff;

    text-shadow:
        0 3px 8px
        rgba(0,0,0,.34);
}


.avg-cell span {

    margin-top: 0;

    font-size: 43px;
}


/* ========================================================
   UNIFIED PROJECTED-GOALS FILL PALETTE
   PROJECTED GOALS + CLEAN SHEET ODDS
   GREEN | SKY BLUE | GOLD | RED
======================================================== */

.tier-1,
.tier-2,
.metric-card.fdr-1,
.avg-cell.fdr-1,
.metric-card.fdr-2,
.avg-cell.fdr-2 {
    background: linear-gradient(145deg, #45ff77 0%, #12b94b 100%);
    border-color: rgba(154,255,178,.96);
    box-shadow: 0 0 24px rgba(57,255,106,.24), inset 0 1px 0 rgba(255,255,255,.35);
}

.tier-3,
.metric-card.fdr-3,
.avg-cell.fdr-3 {
    background: linear-gradient(145deg, #69d9ff 0%, #188bc9 100%);
    border-color: rgba(188,237,255,.96);
    box-shadow: 0 0 24px rgba(94,207,255,.23), inset 0 1px 0 rgba(255,255,255,.34);
}

.tier-4,
.metric-card.fdr-4,
.avg-cell.fdr-4 {
    background: linear-gradient(145deg, #ffe36e 0%, #d99b00 100%);
    border-color: rgba(255,244,174,.98);
    box-shadow: 0 0 24px rgba(255,211,77,.23), inset 0 1px 0 rgba(255,255,255,.35);
}

.tier-5,
.metric-card.fdr-5,
.avg-cell.fdr-5 {
    background: linear-gradient(145deg, #ff6673 0%, #cf1735 100%);
    border-color: rgba(255,190,197,.97);
    box-shadow: 0 0 24px rgba(255,79,94,.24), inset 0 1px 0 rgba(255,255,255,.32);
}

/* High-contrast labels/numbers on bright metric fills. */
.metric-card.tier-1 .fixture-name,
.metric-card.tier-2 .fixture-name,
.metric-card.tier-3 .fixture-name,
.metric-card.tier-4 .fixture-name,
.metric-card.tier-1 .metric-number,
.metric-card.tier-2 .metric-number,
.metric-card.tier-3 .metric-number,
.metric-card.tier-4 .metric-number,
.avg-cell.tier-1 span,
.avg-cell.tier-2 span,
.avg-cell.tier-3 span,
.avg-cell.tier-4 span { color:#06131c; text-shadow:0 1px 0 rgba(255,255,255,.58); }
.metric-card.tier-5 .fixture-name,
.metric-card.tier-5 .metric-number,
.avg-cell.tier-5 span { color:#ffffff; text-shadow:0 2px 7px rgba(0,0,0,.62); }

.metric-card.fdr-none {
    background: linear-gradient(145deg, #24384b 0%, #122130 100%);
    border-color: rgba(178,199,216,.28);
}

/* Fix 11: Clean Sheet cards reuse the exact FDR five-band fills. */
.metric-card.fdr-1 .fixture-name,
.metric-card.fdr-2 .fixture-name,
.metric-card.fdr-3 .fixture-name,
.metric-card.fdr-4 .fixture-name,
.metric-card.fdr-1 .metric-number,
.metric-card.fdr-2 .metric-number,
.metric-card.fdr-3 .metric-number,
.metric-card.fdr-4 .metric-number,
.avg-cell.fdr-1 span,
.avg-cell.fdr-2 span,
.avg-cell.fdr-3 span,
.avg-cell.fdr-4 span { color:#06131c; text-shadow:0 1px 0 rgba(255,255,255,.35); }
.metric-card.fdr-5 .fixture-name,
.metric-card.fdr-5 .metric-number,
.avg-cell.fdr-5 span { color:#ffffff; text-shadow:0 2px 7px rgba(0,0,0,.62); }


/* ========================================================
   FOOTER
======================================================== */

.vortex-footer {

    height: 126px;

    flex:
        0 0 126px;

    margin-top: 14px;

    padding:
        0
        34px;

    display: grid;

    grid-template-columns:
        1.2fr
        1.3fr
        1.2fr;

    align-items: center;

    border-radius: 15px;

    border:
        1px solid
        rgba(255,255,255,.09);

    background:
        rgba(1,9,18,.98);

    font-family:
        Arial,
        sans-serif;
}


.footer-section {

    min-width: 0;

    display: flex;

    align-items: center;

    gap: 16px;

    font-size: 29px;

    font-weight: 900;

    letter-spacing: .8px;
}


.footer-left {

    justify-content:
        flex-start;
}


.footer-center {

    justify-content:
        center;

    gap: 48px;
}


.footer-right {

    justify-content:
        flex-end;
}


.footer-label {

    color: #7790a5;
}


.footer-value {

    color: #ffffff;
}


.social-group {

    display: flex;

    align-items: center;

    gap: 12px;

    white-space: nowrap;
}


.x-social {

    color: #28e3bd;
}


.youtube-icon {

    width: 64px;
    height: 46px;
}


.x-icon {

    width: 42px;
    height: 48px;
}


.footer-separator {

    width: 2px;
    height: 42px;

    background:
        rgba(36,225,192,.34);
}


</style>

</head>


<body>


<div class="scene">


    <div class="top-accent">
    </div>


    <header class="header">


        <div class="logo-area">

            $vortex_logo

        </div>


        <div class="header-meta">

            <div class="header-kicker">

                $header_kicker

            </div>


            <div class="header-season">

                FPL VORTEX • 2026/27

            </div>

        </div>


    </header>


    <section class="hero">


        <div>


            <h1 class="main-title">

                $title_first

                <span>
                    $title_second
                </span>

            </h1>


            <div class="subtitle">

                $subtitle

            </div>


        </div>


        <div class="scale-card">


            <div class="scale-heading">

                $scale_heading

            </div>


            <div class="scale-row">


                <div class="scale-box scale5">
                </div>

                <div class="scale-box scale4">
                </div>

                <div class="scale-box scale3">
                </div>

                <div class="scale-box scale2">
                </div>

                <div class="scale-box scale1">
                </div>


                <div class="scale-copy">

                    LOW → HIGH

                </div>


            </div>


        </div>


    </section>


    <main class="tables">

        $left_panel

        $right_panel

    </main>


    <footer class="vortex-footer">


        <div class="footer-section footer-left">

            <span class="footer-label">
                SOURCE:
            </span>

            <span class="footer-value">
                $source_text
            </span>

        </div>


        <div class="footer-section footer-center">


            <div class="social-group">

                $youtube_icon

                <span>
                    FPL VORTEX
                </span>

            </div>


            <div class="footer-separator">
            </div>


            <div class="social-group x-social">

                $x_icon

                <span>
                    @FplVortexM
                </span>

            </div>


        </div>


        <div class="footer-section footer-right">

            <span class="footer-label">
                UPDATED:
            </span>

            <span class="footer-value">
                $data_updated
            </span>

        </div>


    </footer>


</div>


</body>

</html>

"""
)


# ============================================================
# 21. BUILD PROJECTED GOALS HTML
# ============================================================

xg_sorted = (

    team_projection_df
    .sort_values(

        [
            "avg_xg",
            "team_short",
        ],

        ascending=[
            False,
            True,
        ]
    )

    .reset_index(
        drop=True
    )
)


xg_records = (
    xg_sorted
    .to_dict(
        "records"
    )
)


xg_left = (
    build_projection_panel(
        xg_records[:10],
        "xg"
    )
)


xg_right = (
    build_projection_panel(
        xg_records[10:],
        "xg"
    )
)


# ============================================================
# 22. BUILD CLEAN SHEET HTML
# ============================================================

cs_sorted = (

    team_projection_df
    .sort_values(

        [
            "avg_cs",
            "team_short",
        ],

        ascending=[
            False,
            True,
        ]
    )

    .reset_index(
        drop=True
    )
)


cs_records = (
    cs_sorted
    .to_dict(
        "records"
    )
)


cs_left = (
    build_projection_panel(
        cs_records[:10],
        "cs"
    )
)


cs_right = (
    build_projection_panel(
        cs_records[10:],
        "cs"
    )
)


# ============================================================
# 23. SOURCE LABEL
# ============================================================

# Projected Goals and Clean Sheet Odds are locked to the exact Phase-4B fixture
# snapshot used by FDR. These are labels only; no extra forecast is calculated.
xg_source_text = "VORTEX PHASE 4 MODEL • SAME GOAL LAMBDAS AS FDR"
cs_source_text = "VORTEX PHASE 4 MODEL • SAME CS PROBABILITY AS DEFENCE FDR"


data_updated = str(VX_FOOTER_UPDATED_DATE)


# ============================================================
# 24. WRITE BOTH STANDALONE HTML FILES
# ============================================================

xg_html = scene_template.substitute(

    document_title=
        "FPL VORTEX — Projected Goals",

    vortex_logo=
        vortex_logo_html,

    header_kicker=
        '<span style="position:relative;left:-560px;">ATTACK INTELLIGENCE</span>',

    title_first=
        "PROJECTED",

    title_second=
        "GOALS",

    subtitle=
        (
            f"{GW_HEADERS[0]}–{GW_HEADERS[-1]} • "
            "TEAM EXPECTED GOALS • "
            "SORTED HIGH → LOW"
        ),

    scale_heading=
        "ATTACKING OUTLOOK",

    left_panel=
        xg_left,

    right_panel=
        xg_right,

    source_text=
        xg_source_text,

    youtube_icon=
        youtube_icon,

    x_icon=
        x_icon,

    data_updated=
        data_updated,
)


print("✅ PROJECTED GOALS source: Phase-4 model lambdas shared exactly with FDR")
print("✅ No second xG formula, no market blend, no official FPL difficulty used")

print("✅ CLEAN SHEET source: Phase-4 model probabilities shared exactly with Defence FDR")
_vx_cs_required_css = (
    ".metric-card.fdr-1,", ".metric-card.fdr-2,", ".metric-card.fdr-3,",
    ".metric-card.fdr-4,", ".metric-card.fdr-5,"
)
if not all(token in scene_template.template for token in _vx_cs_required_css):
    raise RuntimeError("Clean Sheet CSS guard failed: one or more Phase-4 FDR fill classes are missing")
print("✅ CLEAN SHEET fill: exact Projected Goals green/sky/gold/red palette")
print("✅ No second CS formula, no market blend, no official FPL difficulty used")

cs_html = scene_template.substitute(

    document_title=
        "FPL VORTEX — Clean Sheet Odds",

    vortex_logo=
        vortex_logo_html,

    header_kicker=
        '<span style="position:relative;left:-560px;">DEFENCE INTELLIGENCE</span>',

    title_first=
        "CLEAN SHEET",

    title_second=
        "ODDS",

    subtitle=
        (
            f"{GW_HEADERS[0]}–{GW_HEADERS[-1]} • "
            "POISSON CLEAN-SHEET PROBABILITY • "
            "SORTED HIGH → LOW"
        ),

    scale_heading=
        "CLEAN-SHEET OUTLOOK",

    left_panel=
        cs_left,

    right_panel=
        cs_right,

    source_text=
        cs_source_text,

    youtube_icon=
        youtube_icon,

    x_icon=
        x_icon,

    data_updated=
        data_updated,
)


xg_html = vx_apply_pl_lockup(xg_html)
cs_html = vx_apply_pl_lockup(cs_html)
xg_html = vx_polish_branding(xg_html)
cs_html = vx_polish_branding(cs_html)

XG_HTML_PATH.write_text(
    xg_html,
    encoding="utf-8"
)


CS_HTML_PATH.write_text(
    cs_html,
    encoding="utf-8"
)


print()
print(
    "✅ PROJECTED GOALS HTML"
)

print(
    XG_HTML_PATH
)


print()
print(
    "✅ CLEAN SHEET ODDS HTML"
)

print(
    CS_HTML_PATH
)



if "export_4k_png" not in globals():
    raise RuntimeError("Run Cell 14 first — shared 4K renderer is missing.")
print("⏳ Rendering restored original matrix slides...")
await export_4k_png(FDR_HTML_PATH, FDR_PNG_PATH)
await export_4k_png(XG_HTML_PATH, XG_PNG_PATH)
await export_4k_png(CS_HTML_PATH, CS_PNG_PATH)
print("✅ ORIGINAL WEEKLY FDR / GOALS / CLEAN-SHEET HTML + PNG RESTORED")
```

```python
# ============================================================
# CELL 17 — GW REVIEW • MASTER-CONTROLLED MP4 RENDER
# Cell 0 controls Draft 480p5 / Final 1080p24

if not globals().get("SLIDE_SELECTION", {}).get("gw_review", True):
    print("⏭️ GW REVIEW skipped by Cell 0.")
else:
    # ============================================================

    from pathlib import Path
    import json, subprocess, sys, math
    from IPython.display import Video, display

    ROOT   = Path(globals().get("PROJECT_ROOT", Path("output/MyDrive/FPL_VORTEX")))
    HTML   = Path(globals().get("HTML_WORK_DIR", Path("output/.work/html")))
    VIDEO  = Path(globals().get("MP4_DIR", Path("output/.work/mp4")))
    PREV   = Path(globals().get("PREVIEW_WORK_DIR", Path("output/.work/preview")))
    TIMING = Path(globals().get("DATA_DIR", Path("output/.work/data")))
    SILENT_DIR = Path(globals().get("VIDEO_WORK_DIR", Path("output/.work/video")))

    VIDEO.mkdir(parents=True, exist_ok=True)
    PREV.mkdir(parents=True, exist_ok=True)
    SILENT_DIR.mkdir(parents=True, exist_ok=True)

    ANIM_HTML = HTML/"gw_review_scene_4k_animated.html"
    MANIFEST  = TIMING/"gw_review_animation_manifest.json"

    SILENT_MP4 = SILENT_DIR/"gw_review_silent.mp4"
    FINAL_MP4  = VIDEO/"gw_review.mp4"

    # ------------------------------------------------------------
    # OUTPUT
    # ------------------------------------------------------------

    DESIGN_W, DESIGN_H = 3840, 2160
    W = int(globals().get("VIDEO_WIDTH", 3840))
    H = int(globals().get("VIDEO_HEIGHT", 2160))
    FPS = int(globals().get("VIDEO_FPS", 30))
    SCALE_X = W / DESIGN_W
    SCALE_Y = H / DESIGN_H
    SCALE = (SCALE_X, SCALE_Y)

    if not ANIM_HTML.exists():
        raise FileNotFoundError("Run Cell 16 first.")

    if not MANIFEST.exists():
        raise FileNotFoundError("GW review animation manifest missing.")

    M = json.loads(
        MANIFEST.read_text(encoding="utf-8")
    )

    AUDIO = Path(M["audio"])

    if not AUDIO.exists():
        raise FileNotFoundError(AUDIO)


    # ------------------------------------------------------------
    # HELPERS
    # ------------------------------------------------------------





    AUDIO_DURATION = vx_media_duration(AUDIO)
    if "vx_frame_plan" not in globals():
        raise RuntimeError("Run Cell 9 first — shared vx_frame_plan is missing.")
    FRAME_PLAN = vx_frame_plan(AUDIO_DURATION, FPS)
    N_FRAMES = int(FRAME_PLAN["frames"])
    VIDEO_GRID_DURATION = float(FRAME_PLAN["grid_duration"])

    print(f"🎨 Design canvas : {DESIGN_W}×{DESIGN_H}")
    print(f"📺 Video output  : {W}×{H}")
    print(f"🔎 Scale         : {SCALE}")
    print(f"🎞 FPS           : {FPS}")
    print(f"🎙 Duration      : {AUDIO_DURATION:.2f}s")
    print(f"🎬 Frames        : {N_FRAMES:,}")
    print(f"⏱ Frame grid    : {VIDEO_GRID_DURATION:.3f}s")


    # ------------------------------------------------------------
    # SHARED PLAYWRIGHT RUNTIME FROM CELL 14
    # ------------------------------------------------------------
    if "_vx_ensure_playwright_chromium" not in globals() or "async_playwright" not in globals():
        raise RuntimeError("Run Cell 14 first — shared Playwright renderer is missing.")
    _vx_ensure_playwright_chromium()

    # ------------------------------------------------------------
    # EXACT HALF-SCALE
    # ------------------------------------------------------------



    # ------------------------------------------------------------
    # RENDER
    # ------------------------------------------------------------

    async def render_review():

        async with async_playwright() as pw:

            try:

                browser = await pw.chromium.launch(
                    headless=True,
                    args=[
                        "--no-sandbox",
                        "--disable-dev-shm-usage",
                        "--disable-gpu"
                    ]
                )

            except Exception:

                subprocess.run(
                    [
                        sys.executable,
                        "-m","playwright",
                        "install","chromium"
                    ],
                    check=True
                )

                browser = await pw.chromium.launch(
                    headless=True,
                    args=[
                        "--no-sandbox",
                        "--disable-dev-shm-usage",
                        "--disable-gpu"
                    ]
                )

            context = await browser.new_context(
                viewport={
                    "width":W,
                    "height":H
                },
                device_scale_factor=1
            )

            page = await context.new_page()

            await page.goto(
                ANIM_HTML.resolve().as_uri(),
                wait_until="load",
                timeout=120000
            )

            await page.wait_for_function(
                "window.__ready === true",
                timeout=60000
            )

            # ----------------------------------------------------
            # PLAYER CARD DATA / COMPONENT GUARD
            # ----------------------------------------------------
            # No fallback card calculations are allowed here. Cell 16 must
            # supply the authoritative card payload and Cell 15A the design.
            guard = await page.evaluate(r"""
            () => {
                const VX = window.VX || {};
                const starters = Array.isArray(VX.starters) ? VX.starters : [];
                const bench = Array.isArray(VX.bench) ? VX.bench : [];
                const missing = [...starters, ...bench]
                    .filter(p => !p || !p.card || !Array.isArray(p.card.fixtures))
                    .map(p => p?.id ?? "?");

                return {
                    starters: starters.length,
                    bench: bench.length,
                    segments: Array.isArray(VX.segments) ? VX.segments.length : 0,
                    missing,
                    component: !!(window.VXPlayerCard && typeof window.VXPlayerCard.mount === "function")
                };
            }
            """)

            if not guard.get("component"):
                raise RuntimeError(
                    "PLAYER CARD DESIGN runtime missing. Run Cell 15A then Cell 16."
                )

            if guard.get("missing"):
                raise RuntimeError(
                    "Missing authoritative player-card payload for player IDs: "
                    + ", ".join(map(str, guard["missing"]))
                )

            # Explicit responsive hero-card smoke test before the expensive render.
            hero_qa = await page.evaluate(r"""
            () => {
                const VX = window.VX || {};
                const segs = Array.isArray(VX.segments) ? VX.segments : [];
                const hero = segs.find(s =>
                    s && s.player_id &&
                    Number.isFinite(Number(s.hero_on)) &&
                    Number.isFinite(Number(s.hero_off))
                );

                if (hero && typeof window.__seek === "function") {
                    const t = (Number(hero.hero_on) + Number(hero.hero_off)) / 2;
                    window.__seek(t);
                }

                const slot = document.querySelector("#vxHeroWrap");
                const card = slot?.querySelector(":scope > .vx-player-card");
                if (!slot || !card) return {ok:false, reason:"hero card did not mount"};

                if (window.VXPlayerCard) { window.VXPlayerCard.fitSlot(slot); window.VXPlayerCard.fitText(card); }
                const s = slot.getBoundingClientRect();
                const c = card.getBoundingClientRect();
                const ratio = c.height ? c.width / c.height : 0;
                const overflow = c.width > s.width + 2 || c.height > s.height + 2;
                const textOverflow = [...card.querySelectorAll("[data-pc-fit]")]
                    .some(el => el.scrollWidth > el.clientWidth + 2);
                const cardOverflow = card.scrollWidth > card.clientWidth + 2 || card.scrollHeight > card.clientHeight + 2;

                return {
                    ok: Math.abs(ratio - .66) < .015 && !overflow && !cardOverflow && !textOverflow,
                    ratio,
                    overflow,
                    cardOverflow,
                    textOverflow,
                    card:[Math.round(c.width),Math.round(c.height)],
                    slot:[Math.round(s.width),Math.round(s.height)]
                };
            }
            """)

            if not hero_qa.get("ok"):
                raise RuntimeError(f"Responsive hero-card QA failed: {hero_qa}")

            print(
                "✅ Hero card responsive QA: "
                f"{hero_qa['card'][0]}×{hero_qa['card'][1]} "
                f"ratio={hero_qa['ratio']:.3f}"
            )

            # ----------------------------------------------------
            # PLAYER-BY-PLAYER REVEAL / PERSISTENCE QA
            # ----------------------------------------------------
            sync_qa = await page.evaluate(r"""
            () => {
                const VX = window.VX || {};
                const segs = (VX.segments || []).filter(s => s && s.player_id);
                const failures = [];
                let expectedStarters = 0;
                let expectedBench = 0;

                const opacityOf = el => {
                    if (!el) return 0;
                    const raw = Number(el.style.opacity);
                    return Number.isFinite(raw) ? raw : Number(getComputedStyle(el).opacity || 0);
                };

                for (const s of segs) {
                    const isStarter = s.kind === "starter";
                    if (isStarter) expectedStarters += 1;
                    else expectedBench += 1;

                    const probe = Math.min(
                        Number(s.t1) - 0.06,
                        Number(s.reveal_at) + 0.50
                    );
                    window.__seek(Math.max(Number(s.reveal_at) + 0.08, probe));

                    const playerEl = document.querySelector(
                        isStarter ? `#vxp_${s.player_id}` : `#vxb_${s.player_id}`
                    );
                    const hero = document.querySelector("#vxHeroWrap");
                    const heroPid = hero?.dataset?.pid || "";

                    if (opacityOf(playerEl) <= 0.45) {
                        failures.push(`${s.id}: player not revealed with spoken name`);
                    }
                    if (heroPid !== String(s.player_id) || opacityOf(hero) <= 0.05) {
                        failures.push(`${s.id}: hero not synced to spoken name`);
                    }

                    const visibleStarters = [...document.querySelectorAll(".playerCard")]
                        .filter(el => opacityOf(el) > 0.45).length;
                    const visibleBench = [...document.querySelectorAll(".vxb")]
                        .filter(el => opacityOf(el) > 0.45).length;

                    if (visibleStarters < expectedStarters) {
                        failures.push(`${s.id}: earlier starter disappeared`);
                    }
                    if (visibleBench < expectedBench) {
                        failures.push(`${s.id}: earlier bench player disappeared`);
                    }

                    const after = Math.min(
                        Number(VX.total) - 0.01,
                        Number(s.t1) + 0.03
                    );
                    window.__seek(after);
                    if (opacityOf(playerEl) <= 0.45) {
                        failures.push(`${s.id}: player did not persist after hero`);
                    }
                }

                const metricsProbe = Math.min(
                    Number(VX.total) - 0.01,
                    Number(VX.metrics_t) + 0.70
                );
                window.__seek(metricsProbe);
                const metrics = document.querySelector("#vxMetrics");
                const finalStarters = [...document.querySelectorAll(".playerCard")]
                    .filter(el => opacityOf(el) > 0.45).length;
                const finalBench = [...document.querySelectorAll(".vxb")]
                    .filter(el => opacityOf(el) > 0.45).length;

                if (opacityOf(metrics) <= 0.80) failures.push("full data board did not reveal together");
                if (finalStarters !== (VX.starters || []).length) failures.push("final starting XI incomplete");
                if (finalBench !== (VX.bench || []).length) failures.push("final bench incomplete");

                return {
                    ok: failures.length === 0,
                    failures,
                    starters: finalStarters,
                    bench: finalBench,
                    segments: segs.length
                };
            }
            """)

            if not sync_qa.get("ok"):
                raise RuntimeError(f"GW Review animation sync QA failed: {sync_qa}")

            print(
                "✅ GW Review animation sync QA: "
                f"{sync_qa['starters']} starters + {sync_qa['bench']} bench persistent, "
                "hero synced to Ryan name, full board last"
            )

            await vx_apply_render_scale(page, W, H, DESIGN_W, DESIGN_H)

            try:
                await page.evaluate(
                    "() => document.fonts.ready"
                )
            except:
                pass

            await page.wait_for_timeout(1000)

            # ----------------------------------------------------
            # FULL-SCREEN CHECK
            # ----------------------------------------------------

            size = await page.evaluate("""
            () => ({
                w: innerWidth,
                h: innerHeight,
                body: document.body.getBoundingClientRect().toJSON()
            })
            """)

            if size["w"] != W or size["h"] != H:
                raise RuntimeError(
                    f"Viewport failed: {size}"
                )

            body = size["body"]

            # transformed body must visually cover entire 1920×1080
            if (
                abs(body["width"] - W) > 2
                or abs(body["height"] - H) > 2
            ):
                raise RuntimeError(
                    "Full-screen scaling failed: "
                    f"{body['width']}×{body['height']}"
                )

            print("✅ Full-bleed 16:9")
            print("✅ No left/right bars")
            print("✅ No top/bottom bars")
            print("✅ No cropping")

            # ----------------------------------------------------
            # PREVIEW
            # ----------------------------------------------------

            preview_t = min(
                AUDIO_DURATION - .05,
                max(.05, AUDIO_DURATION * .70)
            )

            await page.evaluate(
                "(t)=>window.__seek(t)",
                preview_t
            )

            await page.screenshot(
                path=str(PREV/f"gw_review_{H}p_preview.png"),
                type="png"
            )

            print(f"✅ {H}p preview saved")

            # ----------------------------------------------------
            # DIRECT FRAME PIPE
            # ----------------------------------------------------

            ff = subprocess.Popen(
                [
                    "ffmpeg",
                    "-y",
                    "-v","error",

                    "-f","image2pipe",
                    "-framerate",str(FPS),
                    "-vcodec","png",
                    "-i","-",

                    "-an",

                    "-c:v","libx264",
                    "-preset","veryfast",
                    "-crf","18",

                    "-pix_fmt","yuv420p",

                    "-movflags","+faststart",

                    str(SILENT_MP4)
                ],
                stdin=subprocess.PIPE,
                stderr=subprocess.PIPE
            )

            report_every=max(
                1,
                N_FRAMES//20
            )

            try:

                for i in range(N_FRAMES):

                    t=min(
                        i/FPS,
                        max(
                            0,
                            AUDIO_DURATION-(1/FPS)
                        )
                    )

                    await page.evaluate(
                        "(t)=>window.__seek(t)",
                        float(t)
                    )

                    frame=await page.screenshot(
                        type="png"
                    )

                    ff.stdin.write(frame)

                    if (
                        i % report_every == 0
                        or i == N_FRAMES-1
                    ):

                        pct=(i+1)/N_FRAMES*100

                        print(
                            f"\r🎬 Rendering "
                            f"{pct:5.1f}% • "
                            f"{i+1:,}/{N_FRAMES:,}",
                            end=""
                        )

                ff.stdin.close()

                rc=ff.wait()

                if rc != 0:

                    err=ff.stderr.read().decode(
                        "utf-8",
                        errors="ignore"
                    )

                    raise RuntimeError(
                        err[-3000:]
                    )

            finally:

                if ff.stdin and not ff.stdin.closed:
                    ff.stdin.close()

            print(f"\n✅ {W}×{H} silent video rendered")

            await context.close()
            await browser.close()


    await render_review()


    # ------------------------------------------------------------
    # ADD RYAN AUDIO
    # ------------------------------------------------------------

    subprocess.run(
        [
            "ffmpeg",
            "-y",
            "-v","error",

            "-i",str(SILENT_MP4),
            "-i",str(AUDIO),

            "-map","0:v:0",
            "-map","1:a:0",

            "-c:v","copy",

            "-c:a","aac",
            "-b:a","192k",

            "-shortest",

            "-movflags","+faststart",

            str(FINAL_MP4)
        ],
        check=True
    )


    # ------------------------------------------------------------
    # FINAL QA
    # ------------------------------------------------------------

    info = vx_video_info(FINAL_MP4)
    final_dur = vx_media_duration(FINAL_MP4)

    if (
        int(info["width"]) != W
        or int(info["height"]) != H
    ):
        raise RuntimeError(
            f"Resolution failed: "
            f"{info['width']}×{info['height']}"
        )

    # Strong frame-grid QA: the encoded video must match the exact integer-frame
    # duration, while the audio-to-grid delta may be no more than half a frame.
    grid_delta = abs(final_dur - VIDEO_GRID_DURATION)
    mux_tolerance = max(0.03, 0.10 / max(FPS, 1))
    audio_delta = abs(final_dur - AUDIO_DURATION)
    audio_tolerance = float(FRAME_PLAN["max_quantization_error"]) + mux_tolerance

    if grid_delta > mux_tolerance:
        raise RuntimeError(
            f"Encoded frame-grid mismatch: {final_dur:.3f}s vs "
            f"{VIDEO_GRID_DURATION:.3f}s (delta={grid_delta:.3f}s)"
        )
    if audio_delta > audio_tolerance:
        raise RuntimeError(
            f"Duration mismatch: {final_dur:.3f}s vs {AUDIO_DURATION:.3f}s "
            f"(delta={audio_delta:.3f}s, allowed={audio_tolerance:.3f}s at {FPS}fps)"
        )

    print(
        f"✅ Duration QA: audio {AUDIO_DURATION:.3f}s → "
        f"frame grid {VIDEO_GRID_DURATION:.3f}s → encoded {final_dur:.3f}s"
    )

    print("\n" + "="*65)
    print("✅ GW REVIEW — FINAL RENDER PASSED")
    print("="*65)
    print("✅ Design   : 3840×2160")
    print(f"✅ Output   : {W}×{H}")
    print("✅ Aspect   : 16:9 FULL SCREEN")
    print(f"✅ FPS      : {FPS}")
    print("✅ H.264    : CRF 18")
    print("✅ Audio    : Ryan AAC")
    print("✅ No crop")
    print("✅ No black borders")
    print("✅ FILE:",FINAL_MP4)

    display(
        Video(
            str(FINAL_MP4),
            embed=False,
            width=1200
        )
    )
```

```python
# ============================================================
# CELL 18 — INTRO + FDR + GOALS + CS + OUTRO NARRATION
# ============================================================

from pathlib import Path
import json, re, subprocess, sys

ROOT   = Path(globals().get("PROJECT_ROOT", Path("output/MyDrive/FPL_VORTEX")))
AUDIO  = Path(globals().get("MP3_DIR", Path("output/.work/mp3")))
TIMING = Path(globals().get("DATA_DIR", Path("output/.work/data")))
AUDIO.mkdir(parents=True, exist_ok=True)
TIMING.mkdir(parents=True, exist_ok=True)

for x in ["team_projection_df","fdr_table","NEXT_GW"]:
    if x not in globals():
        raise NameError(f"Missing {x}")

for _required_tts_name in ("VOICE", "MOODS", "vx_tts_profile", "vx_edge_tts_stream", "vx_full_word_narration", "vx_full_word_display", "vx_narration_team", "vx_narration_fixture"):
    if _required_tts_name not in globals():
        raise RuntimeError(
            "Run SHARED PRODUCTION HELPERS before narration — "
            f"missing {_required_tts_name}"
        )


SCENE_MOODS = {
    "00_intro": "intro",
    "02_fdr": "analysis",
    "03_projected_goals": "analysis",
    "04_clean_sheets": "key_stat",
    "05_outro": "outro",
}

narr_projection = team_projection_df.copy()
narr_fdr = fdr_table.copy()

# PROJECTED GOALS narration uses the exact same six-Gameweek team model
# and ranking as the approved Projected Goals slide. No second ranking/model.
_GOALS_6GW_COLUMNS = [f"gw{i}_xg" for i in range(1, 7)]
_missing_goal_columns = [c for c in _GOALS_6GW_COLUMNS if c not in narr_projection.columns]
if _missing_goal_columns:
    raise RuntimeError(
        "Projected-goals narration is missing model columns: "
        + ", ".join(_missing_goal_columns)
    )

narr_projection["projected_goals_6gw"] = (
    narr_projection[_GOALS_6GW_COLUMNS]
    .apply(pd.to_numeric, errors="coerce")
    .fillna(0.0)
    .sum(axis=1)
)

# Match CELL 14B Projected Goals exactly: avg_xg DESC, team_short ASC.
narr_goal_ranked = (
    narr_projection
    .sort_values(["avg_xg", "team_short"], ascending=[False, True])
    .reset_index(drop=True)
)
narr_goal_target = narr_goal_ranked.head(5).reset_index(drop=True)
narr_goal_low = (
    narr_goal_ranked
    .tail(3)
    .sort_values(["avg_xg", "team_short"], ascending=[True, True])
    .reset_index(drop=True)
)

if len(narr_goal_target) < 5 or len(narr_goal_low) < 3:
    raise RuntimeError("Projected-goals narration requires at least 8 ranked teams.")

_goal_spoken_teams = (
    narr_goal_target["team"].astype(str).tolist()
    + narr_goal_low["team"].astype(str).tolist()
)
if len(set(_goal_spoken_teams)) != 8:
    raise RuntimeError("Projected-goals top-five/bottom-three narration overlaps teams.")

# CLEAN SHEET narration uses the exact same six-Gameweek team model
# and ranking as the approved Clean Sheet Odds slide. No second ranking/model.
narr_cs_ranked = (
    narr_projection
    .sort_values(["avg_cs", "team_short"], ascending=[False, True])
    .reset_index(drop=True)
)
narr_cs_target = narr_cs_ranked.head(5).reset_index(drop=True)
narr_cs_low = (
    narr_cs_ranked
    .tail(3)
    .sort_values(["avg_cs", "team_short"], ascending=[True, True])
    .reset_index(drop=True)
)
narr_defence = narr_cs_ranked.head(3).reset_index(drop=True)

if len(narr_cs_target) < 5 or len(narr_cs_low) < 3:
    raise RuntimeError("Clean-sheet narration requires at least 8 ranked teams.")

_cs_spoken_teams = (
    narr_cs_target["team"].astype(str).tolist()
    + narr_cs_low["team"].astype(str).tolist()
)
if len(set(_cs_spoken_teams)) != 8:
    raise RuntimeError("Clean-sheet top-five/bottom-three narration overlaps teams.")

# FDR narration has its own decision-focused ranking: five easiest to target,
# then the three hardest to avoid for fresh transfers. Official FPL FDR remains
# the sole source; this does not create a second fixture model.
narr_fdr_ranked = narr_fdr.sort_values(["average_fdr", "team"], ascending=[True, True]).reset_index(drop=True)
narr_target = narr_fdr_ranked.head(5).reset_index(drop=True)
narr_avoid = narr_fdr_ranked.tail(3).sort_values(["average_fdr", "team"], ascending=[False, True]).reset_index(drop=True)

# Compatibility name retained for downstream summary wording only.
narr_hard = narr_avoid.copy()

if len(narr_target) < 5 or len(narr_avoid) < 3:
    raise RuntimeError("FDR narration requires at least 8 ranked Premier League teams.")

_fdr_spoken_teams = narr_target["team"].astype(str).tolist() + narr_avoid["team"].astype(str).tolist()
if len(set(_fdr_spoken_teams)) != 8:
    raise RuntimeError("FDR target/avoid narration contains overlapping teams.")


def vx_narr_team(frame, i):
    row=frame.iloc[i]
    return vx_narration_team(row.get("team_id") if "team_id" in row.index else row.get("team"))


def vx_narr_fdr_value(frame, i):
    return float(frame.iloc[i]["average_fdr"])


def vx_narr_good_fixtures(frame, i, max_items=2):
    """Natural fixture descriptions built from canonical ids, never compact FDR labels."""
    row=frame.iloc[i]
    team_id=int(row["team_id"])
    fdr_raw=row.get("fdr", [])
    fdr_list=list(fdr_raw) if isinstance(fdr_raw, (list, tuple)) else [fdr_raw]

    fixture_table=fixtures.copy() if isinstance(fixtures,pd.DataFrame) else pd.DataFrame(fixtures)
    candidates=[]
    for slot in range(6):
        gw=int(NEXT_GW)+slot
        rating=fdr_list[slot] if slot < len(fdr_list) else None
        try:
            rating_value=float(rating)
        except Exception:
            rating_value=9.0

        gw_fixtures=fixture_table.loc[
            (pd.to_numeric(fixture_table["event"],errors="coerce")==gw)
            & (
                (pd.to_numeric(fixture_table["team_h"],errors="coerce")==team_id)
                | (pd.to_numeric(fixture_table["team_a"],errors="coerce")==team_id)
            )
        ].copy()
        if gw_fixtures.empty:
            continue

        descriptions=[]
        for _,fx in gw_fixtures.iterrows():
            home=int(fx["team_h"])==team_id
            opponent_id=int(fx["team_a"] if home else fx["team_h"])
            descriptions.append(vx_narration_fixture(opponent_id,home))

        if descriptions:
            label=vx_review_join(descriptions) if "vx_review_join" in globals() else (
                descriptions[0] if len(descriptions)==1 else " and ".join(descriptions)
            )
            candidates.append((rating_value,slot,label))

    candidates.sort(key=lambda x:(x[0],x[1]))
    labels=[label for _,_,label in candidates[:max_items]]
    if not labels:
        return "the softer fixtures in that six-Gameweek window"
    if len(labels)==1:
        return labels[0]
    return f"{labels[0]} and {labels[1]}"


def vx_narr_goals_6gw(frame, i):
    return float(frame.iloc[i]["projected_goals_6gw"])


def vx_narr_cs_6gw(frame, i):
    value=float(frame.iloc[i]["avg_cs"])
    if value <= 1.000001:
        value *= 100.0
    return value


def vx_narr_cross_table():
    left=narr_projection[["team","avg_xg","avg_cs"]].copy()
    right=narr_fdr[["team","average_fdr"]].copy()
    cross=left.merge(right,on="team",how="left")
    cross["xg_rank"]=pd.to_numeric(cross["avg_xg"],errors="coerce").rank(method="min",ascending=False)
    cross["cs_rank"]=pd.to_numeric(cross["avg_cs"],errors="coerce").rank(method="min",ascending=False)
    cross["fdr_rank"]=pd.to_numeric(cross["average_fdr"],errors="coerce").rank(method="min",ascending=True)
    fallback=float(len(cross)+5)
    cross["xg_rank"]=cross["xg_rank"].fillna(fallback)
    cross["cs_rank"]=cross["cs_rank"].fillna(fallback)
    cross["fdr_rank"]=cross["fdr_rank"].fillna(fallback)
    cross["attack_combo"]=cross["xg_rank"] + cross["fdr_rank"]
    cross["defence_combo"]=cross["cs_rank"] + cross["fdr_rank"]
    return cross


narr_cross = vx_narr_cross_table()
narr_attack_pick = narr_cross.sort_values(["attack_combo","xg_rank","fdr_rank"]).iloc[0]
narr_defence_pick = narr_cross.sort_values(["defence_combo","cs_rank","fdr_rank"]).iloc[0]

attack_pick_team=str(narr_attack_pick["team"])
defence_pick_team=str(narr_defence_pick["team"])
defence_pick_cs=float(narr_defence_pick["avg_cs"])
if defence_pick_cs <= 1.000001:
    defence_pick_cs *= 100.0
defence_pick_fdr=float(narr_defence_pick["average_fdr"])

defence_top_team=vx_narr_team(narr_defence,0)
hardest_team=vx_narr_team(narr_hard,0)

if defence_pick_team == defence_top_team:
    defence_crossover_line=(
        f"That leading defence also has an average fixture difficulty rating of {defence_pick_fdr:.1f}. "
        "When the schedule and clean-sheet model agree like that, their secure starters deserve serious consideration."
    )
else:
    defence_crossover_line=(
        f"For the schedule-plus-model combination, {defence_pick_team} stand out: {defence_pick_cs:.0f} percent clean-sheet projection "
        f"with an average fixture difficulty rating of {defence_pick_fdr:.1f}. That is the sort of defensive profile I would rather buy than a green fixture box on its own."
    )

scripts = {

"00_intro":

f"""

Welcome to Fantasy Premier League Vortex analysis.

Gameweek {NEXT_GW} is not about chasing last week's points. It is about spotting what could happen next.

Some teams are heading into excellent fixtures. Others may look tempting on paper, but the numbers tell a different story.

We will start with a quick Gameweek review — the points, the minutes, and any role changes worth paying attention to.

Then we move into fixture difficulty, looking at the six-Gameweek runs that could create real Fantasy Premier League value, and the ones that deserve caution.

After that, we will check the projected goals to see which attacks are most likely to produce chances.

Then we turn to clean sheet odds and identify the defences that have both fixtures and data on their side.

And at the end, I will bring everything together into clear transfer priorities for Gameweek {NEXT_GW}.

The goal is simple: spot the opportunity early, before everyone else does.

""".strip(),


"02_fdr":
f"""
Over the next six Gameweeks, the fixture schedule can shape the transfer plan.

I am focusing on the five teams with the most favourable runs, then the three clubs facing the toughest stretch. These are not automatic buys or instant sells. Fixture difficulty rating shows where opportunity may sit; minutes, role, attacking output and clean-sheet potential still decide the move.

Let us count down the target list.

Number five: {vx_narr_team(narr_target,4)}, with an average fixture difficulty rating of {vx_narr_fdr_value(narr_target,4):.1f}.
Their schedule is good enough to keep reliable starters on the radar without forcing a transfer. If you already own one, holding may be sensible. For a new move, I want secure minutes and a clear route to returns. The fixtures create the platform; the player data must confirm it.

Number four: {vx_narr_team(narr_target,3)}, averaging {vx_narr_fdr_value(narr_target,3):.1f}.
The useful part of the run includes {vx_narr_good_fixtures(narr_target,3)}. That can open a value route into the side instead of pushing us toward the most expensive name. A regular defender, budget attacker or set-piece option could be the better entry point. Check the starting role before buying.

Number three is {vx_narr_team(narr_target,2)}, on {vx_narr_fdr_value(narr_target,2):.1f}.
This stretch gives them room to build momentum. The important question is which player benefits most: the central forward, a midfielder arriving in the box, or a defender with attacking involvement. The club is well placed; the edge comes from choosing the right role, not simply the badge.

At number two, {vx_narr_team(narr_target,1)} come in at {vx_narr_fdr_value(narr_target,1):.1f}.
This is a run for managers planning ahead. You do not have to move immediately. Saving a free transfer, keeping budget available, or avoiding a short-term punt can leave the route open when the best option becomes clear. The fixtures are a reason to prepare early.

And number one: {vx_narr_team(narr_target,0)}. They have the easiest six-Gameweek schedule, with an average fixture difficulty rating of {vx_narr_fdr_value(narr_target,0):.1f}.
This is the first club I would investigate for a fresh transfer. Start with secure starters and clear roles. If the attacking model supports the run, their forwards and midfielders gain appeal. If the defensive numbers agree, the goalkeeper and defenders become more useful too. The lowest fixture difficulty rating starts the case; the underlying data finishes it.

Now to the warning side.

Third-worst are {vx_narr_team(narr_avoid,2)}, with an average fixture difficulty rating of {vx_narr_fdr_value(narr_avoid,2):.1f}.
I would hesitate before adding a new player from this run. A proven asset can still justify a hold, but a fresh transfer needs a stronger reason when easier routes to points exist elsewhere.

Second-worst: {vx_narr_team(narr_avoid,1)}, averaging {vx_narr_fdr_value(narr_avoid,1):.1f}.
The next block reduces their margin for error. Do not chase a recent goal, assist or clean sheet without checking what follows. Last week's return does not automatically make a good six-Gameweek transfer.

The toughest schedule belongs to {vx_narr_team(narr_avoid,0)}, at {vx_narr_fdr_value(narr_avoid,0):.1f}.
For new transfers, this is the clearest caution flag. Premium assets may still be good enough to keep, but budget and mid-price options must compete with alternatives carrying much friendlier runs. I would look elsewhere unless the individual upside is compelling.

So the five teams on the target shortlist are {vx_narr_team(narr_target,0)}, {vx_narr_team(narr_target,1)}, {vx_narr_team(narr_target,2)}, {vx_narr_team(narr_target,3)}, and {vx_narr_team(narr_target,4)}.
The three warning teams are {vx_narr_team(narr_avoid,0)}, {vx_narr_team(narr_avoid,1)}, and {vx_narr_team(narr_avoid,2)}.

Use the top five to focus your transfer research. Use the bottom three as a reason to demand stronger evidence before chasing recent points. The best Fantasy Premier League move combines a favourable schedule with secure minutes, a useful role and numbers that support future returns.
""".strip(),

"03_projected_goals":

f"""

Fixtures show where opportunity exists. Projected goals show which teams are most likely to convert it into Fantasy returns.

For this ranking, I am looking at the next six Gameweeks using the existing Fantasy Premier League Vortex model. We will count down the five strongest projected attacks, then the three weakest.

Starting in fifth place, {vx_narr_team(narr_goal_target,4)}.

The model projects {vx_narr_goals_6gw(narr_goal_target,4):.2f} goals across the next six Gameweeks.

That puts them on the attacking watchlist. The team environment looks promising, but the best route still depends on secure minutes, position and goal involvement.

At number four, {vx_narr_team(narr_goal_target,3)} are projected for {vx_narr_goals_6gw(narr_goal_target,3):.2f} goals over the same six-Gameweek window.

This is more than a good fixture run. The model sees enough attacking strength to support investment.

Look closely at the main striker, advanced midfielders and set-piece takers. Reliable minutes can turn a strong team projection into a useful transfer.

Third are {vx_narr_team(narr_goal_target,2)}, with {vx_narr_goals_6gw(narr_goal_target,2):.2f} projected goals.

Now the attacking signal gets stronger. But a high team total does not make every player equally valuable.

The priority is the player combining minutes, advanced role and direct goal involvement. Price then decides whether the premium or value route makes more sense.

At number two, {vx_narr_team(narr_goal_target,1)} come in at {vx_narr_goals_6gw(narr_goal_target,1):.2f} projected goals.

That puts them among the strongest attacks for the next six Gameweeks.

If you have no coverage, their key attackers deserve serious consideration. The question is whether you want short-term upside or a player you can comfortably hold through the full run.

At the top of the attacking projection is {vx_narr_team(narr_goal_target,0)}.

Fantasy Premier League Vortex projects {vx_narr_goals_6gw(narr_goal_target,0):.2f} goals for them across the next six Gameweeks, the highest total in the league.

That makes them the first attack to investigate.

Focus on secure starters with central roles, penalties, set pieces or consistently high goal involvement. That is where the strongest Fantasy value is most likely to sit.

The useful way to read this is as a team-level signal, not a guarantee for any one asset. Before making a move, compare how much of the attack runs through the player you are considering. Penalties, set pieces, expected minutes, starting position and substitution risk can separate a strong Fantasy pick from a player who simply belongs to a strong attack.

Now to the low-projection side of the table.

Third-lowest are {vx_narr_team(narr_goal_low,2)}, on {vx_narr_goals_6gw(narr_goal_low,2):.2f} projected goals.

Returns are still possible, but the overall attacking environment is weaker. A new transfer needs a strong individual reason.

Second-lowest, {vx_narr_team(narr_goal_low,1)}, are projected for {vx_narr_goals_6gw(narr_goal_low,1):.2f} goals.

That lowers their appeal for fresh investment. Existing assets may still be holds, but stronger attacking environments exist elsewhere.

The lowest projected attack is {vx_narr_team(narr_goal_low,0)}, with {vx_narr_goals_6gw(narr_goal_low,0):.2f} goals across the six-Gameweek period.

One player can always produce a return, but the wider numbers make this attack difficult to prioritise.

So the five highest projected attacks are {vx_narr_team(narr_goal_target,0)}, {vx_narr_team(narr_goal_target,1)}, {vx_narr_team(narr_goal_target,2)}, {vx_narr_team(narr_goal_target,3)}, and {vx_narr_team(narr_goal_target,4)}.

The three lowest are {vx_narr_team(narr_goal_low,0)}, {vx_narr_team(narr_goal_low,1)}, and {vx_narr_team(narr_goal_low,2)}.

Use the projections to narrow the shortlist, then let minutes, role, price and goal involvement decide the player. The strongest transfer is where all four signals align.

""".strip(),

"04_clean_sheets":

f"""

Fixtures can look favourable, but clean-sheet probability tells us which defences are most likely to turn them into Fantasy points.

For this section, I am ranking teams by their average clean-sheet probability across the next six Gameweeks.

This is especially useful for goalkeeper and defender decisions. Nothing is guaranteed, but stronger probabilities give those picks a better base for returns.

Let us start with the five teams that stand out.

Number five is {vx_narr_team(narr_cs_target,4)}.

Their average clean-sheet probability over the next six Gameweeks is {vx_narr_cs_6gw(narr_cs_target,4):.0f} percent.

That puts them on the shortlist.

First, check minutes. Then look for extra upside through attacking threat, set pieces, saves, or bonus potential. The team projection creates the opportunity; the player profile decides the pick.

At number four, {vx_narr_team(narr_cs_target,3)} come in at {vx_narr_cs_6gw(narr_cs_target,3):.0f} percent.

This looks like a solid defensive environment without necessarily paying a premium.

A cheaper regular starter or goalkeeper could offer the better route. If the clean-sheet potential is similar, price and minutes can make the difference.

Third are {vx_narr_team(narr_cs_target,2)}, with an average clean-sheet chance of {vx_narr_cs_6gw(narr_cs_target,2):.0f} percent.

Now the defensive signal becomes stronger.

My order is simple: secure minutes first, clean-sheet probability second, then attacking, save, or bonus upside.

If a regular starter offers more than one route to points, that is where the value improves.

Number two is {vx_narr_team(narr_cs_target,1)}.

The model gives them an average clean-sheet probability of {vx_narr_cs_6gw(narr_cs_target,1):.0f} percent over the next six Gameweeks.

That puts them among the strongest defensive environments.

If you already own a secure starter, this supports a hold. If you need a new defender or goalkeeper, they deserve serious consideration.

The strongest clean-sheet projection belongs to {vx_narr_team(narr_cs_target,0)}.

Their average clean-sheet probability is {vx_narr_cs_6gw(narr_cs_target,0):.0f} percent across the next six Gameweeks, the highest in the model.

This is the first defence I would investigate.

Start with secure starters, then compare price, attacking involvement, save potential, and bonus routes.

The best option is not always the biggest name. Often, the value comes from the reliable player who gives you access to the strongest defensive environment.

Now let us move to the weakest clean-sheet outlooks.

Third-lowest are {vx_narr_team(narr_cs_low,2)}, averaging {vx_narr_cs_6gw(narr_cs_low,2):.0f} percent.

A clean sheet can still happen, but the six-Gameweek outlook is not strong enough to make them a priority.

For a new transfer, I would want extra attacking or save upside before investing.

Second-lowest are {vx_narr_team(narr_cs_low,1)}, with a clean-sheet projection of {vx_narr_cs_6gw(narr_cs_low,1):.0f} percent.

That creates a weak base for defensive returns.

A cheap price alone is not enough. If the player offers little attacking, save, or bonus potential, stronger options exist elsewhere.

The lowest clean-sheet probability belongs to {vx_narr_team(narr_cs_low,0)}.

Their six-Gameweek average is only {vx_narr_cs_6gw(narr_cs_low,0):.0f} percent.

For a new defensive transfer, this is the clearest caution.

A cheap defender can still have a role, but as a pure clean-sheet pick, the model does not support the move.

So the five best defensive projections are {vx_narr_team(narr_cs_target,0)}, {vx_narr_team(narr_cs_target,1)}, {vx_narr_team(narr_cs_target,2)}, {vx_narr_team(narr_cs_target,3)}, and {vx_narr_team(narr_cs_target,4)}.

The three weakest clean-sheet outlooks are {vx_narr_team(narr_cs_low,0)}, {vx_narr_team(narr_cs_low,1)}, and {vx_narr_team(narr_cs_low,2)}.

Start with the strongest defensive environments, then let minutes, price, attacking threat, saves, and bonus potential decide the player.

""".strip(),

"05_outro":

f"""

Today, we found the teams with the best fixture runs, strongest projected output, and highest clean-sheet potential.

Tomorrow, we turn that data into player decisions: safe core picks, rank-climb options, player traps, captaincy, and the final transfer plan.

Subscribe and follow for the next video. Let us make smarter decisions, stay ahead, and climb the ranks.

""".strip(),

}

# FDR pacing target: with Ryan at the Professor/Calm -10% rate, this wording is
# intentionally built for roughly a 3.5–4 minute spoken section. Actual duration
# remains authoritative only after Edge TTS generates the MP3.
_fdr_word_count=len(re.findall(r"\b[\w’'-]+\b", scripts["02_fdr"]))
if not 470 <= _fdr_word_count <= 620:
    raise RuntimeError(f"FDR narration pacing drifted outside target range: {_fdr_word_count} words")
print(f"✅ FDR narration: {_fdr_word_count} words • top 5 target • worst 3 avoid")

_goals_word_count=len(re.findall(r"\b[\w’'-]+\b", scripts["03_projected_goals"]))
if not 500 <= _goals_word_count <= 760:
    raise RuntimeError(
        f"Projected-goals narration pacing drifted outside target range: {_goals_word_count} words"
    )
print(
    f"✅ Projected Goals narration: {_goals_word_count} words • "
    "next 5 GWs • top 5 target • bottom 3 caution"
)

_cs_word_count=len(re.findall(r"\b[\w’'-]+\b", scripts["04_clean_sheets"]))
if not 520 <= _cs_word_count <= 820:
    raise RuntimeError(
        f"Clean-sheet narration pacing drifted outside target range: {_cs_word_count} words"
    )
print(
    f"✅ Clean Sheet narration: {_cs_word_count} words • "
    "next 5 GWs • top 5 target • bottom 3 avoid"
)

_outro_word_count=len(re.findall(r"\b[\w’'-]+\b", scripts["05_outro"]))
if not 35 <= _outro_word_count <= 65:
    raise RuntimeError(
        f"Outro narration pacing drifted outside target range: {_outro_word_count} words"
    )
print(
    f"✅ Outro narration: {_outro_word_count} words • target 20–25 seconds • recap → tomorrow → subscribe"
)

# Narration QA: accidental repetition INSIDE one scene is a hard failure.
# The same short sentence appearing in two different scenes is not a runtime
# dependency error, so it is reported rather than blocking the whole notebook.
_scene_sentence_sets={}
for _scene_key,_scene_text in scripts.items():
    _sentences=[
        re.sub(r"\s+"," ",s.strip().lower())
        for s in re.split(r"(?<=[.!?])\s+", _scene_text)
        if s.strip()
    ]
    _duplicates=sorted({s for s in _sentences if _sentences.count(s) > 1})
    if _duplicates:
        raise RuntimeError(
            f"{_scene_key} narration repeats full sentences: "
            + " | ".join(_duplicates[:5])
        )
    _scene_sentence_sets[_scene_key]=set(_sentences)

_cross_scene_repeats=[]
_scene_keys=list(_scene_sentence_sets)
for _i,_left in enumerate(_scene_keys):
    for _right in _scene_keys[_i+1:]:
        for _sentence in sorted(_scene_sentence_sets[_left] & _scene_sentence_sets[_right]):
            _cross_scene_repeats.append((_left,_right,_sentence))
if _cross_scene_repeats:
    print(
        "ℹ️ Cross-scene sentence reuse detected but not blocked: "
        + " | ".join(
            f"{a}/{b}: {sentence}"
            for a,b,sentence in _cross_scene_repeats[:3]
        )
    )

# Cell 0 controls which shared scenes actually continue to narration/render.
_selected_shared = set(globals().get("SELECTED_SCENES", list(scripts)))
scripts = {
    key: vx_full_word_narration(
        text,
        context=f"shared narration {key}",
    )
    for key, text in scripts.items()
    if key in _selected_shared
}


async def make_audio(key,text,rate_override=None):

    out=AUDIO/f"{key}.mp3"
    words=[]
    raw=bytearray()

    mood = SCENE_MOODS.get(key, "analysis")
    profile = vx_tts_profile(mood)
    if rate_override is not None:
        profile["rate"]=str(rate_override)

    spoken_text = vx_full_word_narration(
        text,
        context=f"Ryan shared scene {key}",
    )

    audio_bytes, words = await vx_edge_tts_stream(spoken_text, profile)
    raw.extend(audio_bytes)
    if not raw:
        raise RuntimeError(f"Ryan returned no audio bytes for {key}.")
    if not words:
        raise RuntimeError(
            f"Ryan returned 0 WordBoundary events for {key}. "
            "Exact animation sync cannot continue."
        )

    out.write_bytes(raw)

    return {
        "audio":str(out),
        "duration":vx_media_duration(out),
        "words":words,
        "script":spoken_text,
        "display_text":vx_full_word_display(spoken_text, context=f"shared display {key}"),
        "tts_mood":mood,
        "tts_profile":profile,
    }

manifest={}

def _vx_rate_percent(rate_text):
    m=re.fullmatch(r"\s*([+-]?\d+(?:\.\d+)?)%\s*",str(rate_text))
    if not m:
        raise ValueError(f"Unsupported Edge TTS rate: {rate_text!r}")
    return float(m.group(1))

def _vx_rate_text(value):
    value=max(-45.0,min(50.0,float(value)))
    return f"{value:+.0f}%"

for key,text in scripts.items():
    manifest[key]=await make_audio(key,text)

    # Keep the approved outro script, but automatically correct Ryan's speaking
    # rate if Edge TTS drifts outside the 20–25 second production window.
    if key=="05_outro":
        for _attempt in range(2):
            _d=float(manifest[key]["duration"])
            if 20.0 <= _d <= 25.8:
                break
            _current_rate=_vx_rate_percent(manifest[key]["tts_profile"]["rate"])
            _current_speed=max(0.20,1.0+_current_rate/100.0)
            _target_seconds=23.5
            _needed_speed=_current_speed*(_d/_target_seconds)
            _new_rate=_vx_rate_text((_needed_speed-1.0)*100.0)
            print(
                f"↻ Outro Ryan pacing correction: {_d:.2f}s at "
                f"{manifest[key]['tts_profile']['rate']} → {_new_rate}"
            )
            manifest[key]=await make_audio(key,text,rate_override=_new_rate)

    print(
        f"✅ {key:<20}"
        f"{manifest[key]['duration']:>6.2f}s • "
        f"{len(manifest[key]['words'])} words • "
        f"{manifest[key]['tts_mood']} {manifest[key]['tts_profile']['rate']} / {manifest[key]['tts_profile']['pitch']}"
    )

if "05_outro" in manifest:
    _outro_duration = float(manifest["05_outro"]["duration"])
    if not 20.0 <= _outro_duration <= 25.8:
        raise RuntimeError(
            f"Outro duration must land between 20 and 25 seconds after automatic Ryan pacing correction. "
            f"Current audio: {_outro_duration:.2f}s"
        )
    print(f"✅ Outro duration locked: {_outro_duration:.2f}s")

OUT=TIMING/"other_scenes_narration.json"

OUT.write_text(
    json.dumps(manifest,indent=2),
    encoding="utf-8"
)

print(f"\n✅ SELECTED NARRATION READY — {len(manifest)} scene(s)")
print(f"✅ Ryan voice: {VOICE}")
print("✅ Shared TTS moods active")
print("✅ Intro")
print("✅ FDR")
print("✅ Projected Goals")
print("✅ Clean Sheet %")
print("✅ Updated short outro")
print("✅ Next video = SAFE CORE PICKS • RANK-CLIMB OPTIONS • PLAYER TRAPS • CAPTAINCY • TRANSFER PLAN")
```

```python
# ============================================================
# CELL 19 — SHARED ANIMATION ENGINE
# INTRO • FDR • GOALS • CLEAN SHEET • OUTRO
# ============================================================

from pathlib import Path
import json, re, base64, mimetypes

ROOT   = Path(globals().get("PROJECT_ROOT", Path("output/MyDrive/FPL_VORTEX")))
HTML   = Path(globals().get("HTML_WORK_DIR", Path("output/.work/html")))
ANIM   = HTML/"animated"
TIMING = Path(globals().get("DATA_DIR", Path("output/.work/data")))
ANIM.mkdir(parents=True, exist_ok=True)

NARR = TIMING/"other_scenes_narration.json"

if not NARR.exists():
    raise FileNotFoundError("Run Cell 18 first.")

narr = json.loads(NARR.read_text(encoding="utf-8"))

SCENES = {
    "00_intro": (
        "intro_scene_4k.html",
        "intro_scene_4k_animated.html"
    ),
    "02_fdr": (
        "fdr_scene_4k.html",
        "fdr_scene_4k_animated.html"
    ),
    "03_projected_goals": (
        "projected_goals_scene_4k.html",
        "projected_goals_scene_4k_animated.html"
    ),
    "04_clean_sheets": (
        "clean_sheet_scene_4k.html",
        "clean_sheet_scene_4k_animated.html"
    ),
    "05_outro": (
        "outro_scene_4k.html",
        "outro_scene_4k_animated.html"
    ),
}

_selected_shared = set(globals().get("SELECTED_SCENES", list(SCENES)))
SCENES = {
    key: value for key, value in SCENES.items()
    if key in _selected_shared and key in narr
}

# ------------------------------------------------------------
# WORD TIMING
# ------------------------------------------------------------



# ------------------------------------------------------------
# CURRENT TEAM NAMES FOR SPOKEN ROW HIGHLIGHTS
# ------------------------------------------------------------

TEAM_NAMES=[]

if "team_projection_df" in globals():
    TEAM_NAMES += (
        team_projection_df["team"]
        .dropna().astype(str).tolist()
    )

if "fdr_table" in globals():
    TEAM_NAMES += (
        fdr_table["team"]
        .dropna().astype(str).tolist()
    )

TEAM_NAMES=list(dict.fromkeys(TEAM_NAMES))


# ------------------------------------------------------------
# DATA-DRIVEN DEAD-AIR FACTS
# Only shown when the body would otherwise stay blank for >= 5s.
# These are visual facts; they never replace or alter Ryan narration.
# ------------------------------------------------------------

_vx_easy = fdr_table.sort_values("average_fdr").iloc[0]
_vx_hard = fdr_table.sort_values("average_fdr", ascending=False).iloc[0]
_vx_attack_top = team_projection_df.sort_values("avg_xg", ascending=False).iloc[0]
_vx_attack_low = team_projection_df.sort_values("avg_xg", ascending=True).iloc[0]
_vx_cs_top = team_projection_df.sort_values("avg_cs", ascending=False).iloc[0]
_vx_cs_low = team_projection_df.sort_values("avg_cs", ascending=True).iloc[0]

SCENE_DEAD_AIR_FACTS = {
    "00_intro": [
        f"FIXTURE WATCH • {_vx_easy['team']} owns the easiest upcoming run at {_vx_easy['average_fdr']:.1f} average FDR.",
        f"ATTACK WATCH • {_vx_attack_top['team']} leads the model at {_vx_attack_top['avg_xg']:.2f} projected goals per fixture.",
    ],
    "02_fdr": [
        f"EASIEST RUN • {_vx_easy['team']} • {_vx_easy['average_fdr']:.1f} average FDR.",
        f"TOUGH RUN • {_vx_hard['team']} • {_vx_hard['average_fdr']:.1f} average FDR.",
    ],
    "03_projected_goals": [
        f"ATTACK LEADER • {_vx_attack_top['team']} • {_vx_attack_top['avg_xg']:.2f} projected goals per fixture.",
        f"ATTACK CAUTION • {_vx_attack_low['team']} • {_vx_attack_low['avg_xg']:.2f} projected goals per fixture.",
    ],
    "04_clean_sheets": [
        f"CLEAN-SHEET LEADER • {_vx_cs_top['team']} • {_vx_cs_top['avg_cs']:.0f}% average chance.",
        f"DEFENSIVE CAUTION • {_vx_cs_low['team']} • {_vx_cs_low['avg_cs']:.0f}% average chance.",
    ],
    "05_outro": [
        "NEXT UP • Team data becomes player decisions: core picks, rank-climb options, traps and captaincy.",
    ],
}

# ------------------------------------------------------------
# ORIGINAL CHANNEL LOGO FOR OUTRO FINALE
# ------------------------------------------------------------

ELEMENTS=Path(globals().get("ELEMENTS_ROOT", Path(MY_DRIVE) / "elements"))

logo_candidates=[
    ELEMENTS / "logo.png",
    ROOT / "elements" / "logo.png",
    ROOT / "logo.png",
]

OUTRO_LOGO = next((candidate for candidate in logo_candidates if candidate.exists()), None)
if OUTRO_LOGO is None and "_vx_logo_path" in globals():
    try:
        OUTRO_LOGO = Path(globals()["_vx_logo_path"])
    except Exception:
        OUTRO_LOGO = None

def data_uri(path):

    if not path or not path.exists():
        return ""

    mime=mimetypes.guess_type(path.name)[0] or "image/png"

    return (
        f"data:{mime};base64,"
        + base64.b64encode(path.read_bytes()).decode()
    )

OUTRO_LOGO_URI=data_uri(OUTRO_LOGO) if OUTRO_LOGO else ""

print("✅ Outro original logo:", OUTRO_LOGO)

# ------------------------------------------------------------
# MOTION PLAN
# ------------------------------------------------------------

def events_for(key, info):

    words=info["words"]
    dur=float(info["duration"])

    def _sync(phrase, min_duration=0.28):
        start_time, spoken_span=vx_phrase_span(
            words,
            phrase,
            required=True,
            context=f"{key} animation",
        )
        return round(float(start_time),3), max(float(min_duration),float(spoken_span))

    if key=="00_intro":

        welcome_at,_=_sync("Welcome",0.35)
        agenda_1_at,agenda_1_dur=_sync("Gameweek review",0.45)
        agenda_2_at,agenda_2_dur=_sync("fixture difficulty",0.45)
        agenda_3_at,agenda_3_dur=_sync("projected goals",0.45)
        agenda_4_at,agenda_4_dur=_sync("clean sheet odds",0.45)

        return [
            {"sel":".topHeader,.brandLock,.logoBox,.brandWords,.gwPill","at":welcome_at,"dur":0.65,"stagger":0.05,"fx":"left"},
            {"sel":".mainHeading,.headingRule","at":welcome_at,"dur":0.70,"stagger":0.08,"fx":"zoom"},
            {"sel":".agenda-1","at":agenda_1_at,"dur":agenda_1_dur,"stagger":0,"fx":"type"},
            {"sel":".agenda-2","at":agenda_2_at,"dur":agenda_2_dur,"stagger":0,"fx":"type"},
            {"sel":".agenda-3","at":agenda_3_at,"dur":agenda_3_dur,"stagger":0,"fx":"type"},
            {"sel":".agenda-4","at":agenda_4_at,"dur":agenda_4_dur,"stagger":0,"fx":"type"},
            {"sel":".prevResultsStrip","at":welcome_at,"dur":0.48,"stagger":0,"fx":"rise"},
            {"sel":".footer","at":welcome_at,"dur":0.45,"stagger":0,"fx":"fade"},
        ]

    if key=="02_fdr":

        return [
            {
                "sel":".header,.titlebar,.headline",
                "at":0,"dur":0.65,"stagger":0.05,"fx":"left"
            },
            {
                "sel":".panel,.table-panel,.board",
                "at":vx_phrase_time(words,"five teams with the most favourable runs",1.1),
                "dur":0.65,"stagger":0.15,"fx":"rise"
            },
            {
                "sel":"tbody tr,.team-row",
                "at":vx_phrase_time(words,"Number five",2.0),
                "dur":0.38,"stagger":0.055,"fx":"left"
            },
        ]

    if key=="03_projected_goals":

        return [
            {
                "sel":".header,.titlebar,.headline",
                "at":0,"dur":0.65,"stagger":0.05,"fx":"left"
            },
            {
                "sel":".panel,.table-panel,.board",
                "at":vx_phrase_time(words,"Projected goals",1.3),
                "dur":0.60,"stagger":0.14,"fx":"rise"
            },
            {
                "sel":"tbody tr,.team-row",
                "at":vx_phrase_time(words,"Starting in fifth place",2.0),
                "dur":0.38,"stagger":0.055,"fx":"left"
            },
        ]

    if key=="04_clean_sheets":

        return [
            {
                "sel":".header,.titlebar,.headline",
                "at":0,"dur":0.65,"stagger":0.05,"fx":"left"
            },
            {
                "sel":".panel,.table-panel,.board",
                "at":vx_phrase_time(words,"five teams that stand out",1.1),
                "dur":0.60,"stagger":0.14,"fx":"rise"
            },
            {
                "sel":"tbody tr,.team-row",
                "at":vx_phrase_time(words,"Number five",2.0),
                "dur":0.38,"stagger":0.055,"fx":"left"
            },
        ]

    # OUTRO — every narrative transition is anchored to Ryan WordBoundary data.
    today_at,_=_sync("Today",0.35)
    take_1_at,take_1_dur=_sync("best fixture runs",0.40)
    take_2_at,take_2_dur=_sync("strongest projected output",0.40)
    take_3_at,take_3_dur=_sync("highest clean-sheet potential",0.40)
    tomorrow_at,_=_sync("Tomorrow",0.35)
    tomorrow_head_at,tomorrow_head_dur=_sync("player decisions",0.45)
    tom_1_at,tom_1_dur=_sync("safe core picks",0.35)
    tom_2_at,tom_2_dur=_sync("rank-climb options",0.35)
    tom_3_at,tom_3_dur=_sync("player traps",0.35)
    tom_4_at,tom_4_dur=_sync("captaincy",0.35)
    tom_5_at,tom_5_dur=_sync("transfer plan",0.35)
    cta_at,cta_dur=_sync("Subscribe and follow",0.45)
    next_video_at,next_video_dur=_sync("next video",0.35)
    climb_at,_=_sync("climb the ranks",0.35)

    return [
        {"sel":"#summaryStage .topHeader","at":today_at,"dur":0.55,"stagger":0.05,"fx":"left"},
        {"sel":"#todayPanel","at":today_at,"dur":0.65,"stagger":0.0,"fx":"rise"},
        {"sel":".take-head","at":today_at,"dur":0.70,"stagger":0.0,"fx":"type"},
        {"sel":".take-1","at":take_1_at,"dur":take_1_dur,"stagger":0.0,"fx":"type"},
        {"sel":".take-2","at":take_2_at,"dur":take_2_dur,"stagger":0.0,"fx":"type"},
        {"sel":".take-3","at":take_3_at,"dur":take_3_dur,"stagger":0.0,"fx":"type"},
        {"sel":"#tomorrowPanel","at":tomorrow_at,"dur":0.58,"stagger":0.0,"fx":"rise"},
        {"sel":".tom-head","at":tomorrow_head_at,"dur":tomorrow_head_dur,"stagger":0.0,"fx":"type"},
        {"sel":".tom-1","at":tom_1_at,"dur":tom_1_dur,"stagger":0.0,"fx":"type"},
        {"sel":".tom-2","at":tom_2_at,"dur":tom_2_dur,"stagger":0.0,"fx":"type"},
        {"sel":".tom-3","at":tom_3_at,"dur":tom_3_dur,"stagger":0.0,"fx":"type"},
        {"sel":".tom-4","at":tom_4_at,"dur":tom_4_dur,"stagger":0.0,"fx":"type"},
        {"sel":".tom-5","at":tom_5_at,"dur":tom_5_dur,"stagger":0.0,"fx":"type"},
        {"sel":".prevResultsStrip","at":today_at,"dur":0.48,"stagger":0.0,"fx":"rise"},
        {"sel":".outroFooter","at":today_at,"dur":0.45,"stagger":0.0,"fx":"fade"},
        {"sel":".cta-copy-1","at":cta_at,"dur":cta_dur,"stagger":0.0,"fx":"type"},
        {"sel":".cta-btn-sub","at":cta_at,"dur":0.42,"stagger":0.0,"fx":"zoom"},
        {"sel":".cta-copy-2","at":next_video_at,"dur":next_video_dur,"stagger":0.0,"fx":"type"},
        {"sel":".cta-btn-follow","at":next_video_at,"dur":0.42,"stagger":0.0,"fx":"zoom"},
        {"sel":".cta-handle","at":climb_at,"dur":0.45,"stagger":0.0,"fx":"fade"},
    ]

# ------------------------------------------------------------
# COMMON 4K CSS
# ------------------------------------------------------------

CSS = r"""
<!-- VX_SHARED_START -->
<meta name="viewport"
      content="width=3840,initial-scale=1,maximum-scale=1">

<style id="vx-shared-css">

html,body{
    width:3840px !important;
    height:2160px !important;
    min-width:3840px !important;
    min-height:2160px !important;
    max-width:3840px !important;
    max-height:2160px !important;

    margin:0 !important;
    padding:0 !important;

    overflow:hidden !important;
    zoom:1 !important;
}

body{
    position:relative !important;
}

/* no wall-clock animation */
*,
*::before,
*::after{
    animation:none !important;
    transition:none !important;
}

.vx-managed{
    will-change:opacity,transform,filter;
}

/* spoken team highlight */
.vx-spoken{
    position:relative;
    z-index:10;
}

/* --------------------------------------------------------
   DEAD-AIR FACT / TEASER
   -------------------------------------------------------- */
#vxDeadAirTeaser{
    position:absolute;left:560px;right:560px;top:790px;min-height:280px;
    z-index:18;display:flex;flex-direction:column;align-items:center;
    justify-content:center;gap:22px;padding:44px 70px;border-radius:34px;
    border:2px solid rgba(53,230,255,.24);
    background:linear-gradient(135deg,rgba(4,17,31,.94),rgba(7,29,48,.92));
    box-shadow:0 24px 70px rgba(0,0,0,.34),inset 0 1px 0 rgba(255,255,255,.08);
    text-align:center;pointer-events:none;opacity:0;
    transform:translateY(18px) scale(.985);
}
#vxDeadAirTeaser .vxFactKicker{
    font-size:28px;line-height:1;font-weight:1000;letter-spacing:7px;color:#ffd447;
}
#vxDeadAirTeaser .vxFactText{
    min-height:142px;display:flex;align-items:center;justify-content:center;
    font-size:68px;line-height:1.03;font-weight:1000;letter-spacing:1px;
    color:#f7fbff;text-shadow:0 8px 24px rgba(0,0,0,.42);
}
#vxDeadAirTeaser .vxFactRule{
    width:310px;height:5px;border-radius:999px;
    background:linear-gradient(90deg,#ffd447,#35e6ff,#52f28c);
}

/* --------------------------------------------------------
   OUTRO OVERLAY HELPERS
   -------------------------------------------------------- */

#summaryStage,
#ctaStage,
#logoStage,
#bgFade{
    will-change:opacity,transform,filter;
}

#summaryStage{opacity:1;}
#ctaStage,#logoStage{opacity:0;}

</style>
"""

# ------------------------------------------------------------
# SHARED DETERMINISTIC JS
# ------------------------------------------------------------

JS = r"""
<script id="vx-shared-engine">

const VX=__VX__;

function clamp(x,a=0,b=1){
    return Math.max(a,Math.min(b,x));
}

function ease(x){
    x=clamp(x);
    return 1-Math.pow(1-x,3);
}

function pulse(x){
    return Math.sin(clamp(x)*Math.PI);
}

function normalize(x){
    return String(x||"")
      .toLowerCase()
      .replace(/[^a-z0-9]+/g," ")
      .trim();
}

function bestTextElement(txt){

    const q=normalize(txt);
    let best=null, size=1e9;

    document.querySelectorAll(
      "tr,.team-row,.club,.panel,td,div,span"
    ).forEach(el=>{

        const s=normalize(el.innerText);

        if(
            s &&
            s.includes(q) &&
            s.length<size
        ){
            best=el;
            size=s.length;
        }
    });

    return best;
}

let bindings=[];
let highlights=[];

function initVX(){

    const seen=new Set();

    VX.events.forEach(e=>{

        let els=[];

        try{
            els=[...document.querySelectorAll(e.sel)];
        }catch(_){}

        els.forEach((el,i)=>{

            if(seen.has(el)) return;

            seen.add(el);
            el.classList.add("vx-managed");

            bindings.push({
                el,
                at:e.at+i*e.stagger,
                dur:e.dur,
                fx:e.fx
            });
        });
    });

    VX.highlights.forEach(h=>{

        const el=bestTextElement(h.text);

        if(el){
            el.classList.add("vx-spoken");

            highlights.push({
                el,
                at:h.at,
                dur:1.15
            });
        }
    });

    if(Array.isArray(VX.teasers) && VX.teasers.length){
        const teaser=document.createElement("div");
        teaser.id="vxDeadAirTeaser";
        teaser.innerHTML=
          '<div class="vxFactKicker">VORTEX QUICK READ</div>'+
          '<div class="vxFactText"></div>'+
          '<div class="vxFactRule"></div>';
        document.body.appendChild(teaser);
    }

    window.__ready=true;
    window.__seek(0);
}

/* --------------------------------------------------------
   SEEK
   -------------------------------------------------------- */

window.__seek=function(t){

    t=clamp(Number(t)||0,0,VX.duration);

    /* standard entrances */
    bindings.forEach(b=>{

        const p=ease((t-b.at)/b.d);

        b.el.style.opacity=p;

        if(b.fx==="left")
            b.el.style.transform=
              `translateX(${(1-p)*-55}px)`;

        else if(b.fx==="rise")
            b.el.style.transform=
              `translateY(${(1-p)*42}px)
               scale(${.98+.02*p})`;

        else if(b.fx==="zoom")
            b.el.style.transform=
              `scale(${.88+.12*p})`;

        else if(b.fx==="type"){
            b.el.style.transform=`translateY(${(1-p)*28}px) scale(${.985+.015*p})`;
            const title=b.el.querySelector("[data-vx-type]");
            if(title){
                const full=title.dataset.vxType || "";
                if(t < b.at) title.textContent="";
                else{
                    const tp=clamp((t-b.at)/b.dur);
                    const n=Math.floor(full.length*tp);
                    title.textContent=full.slice(0,n) + (tp<1 ? "▌" : "");
                }
            }
        }

        else
            b.el.style.transform="none";
    });

    /* narrated team pulse */
    highlights.forEach(h=>{

        const p=pulse((t-h.at)/h.d);

        h.el.style.filter=
          p>0
          ? `brightness(${1+.25*p})
             saturate(${1+.20*p})`
          : "";

        h.el.style.boxShadow=
          p>0
          ? `0 0 ${20+35*p}px
             rgba(0,229,220,${.12+.28*p})`
          : "";
    });

    /* data-driven dead-air teaser */
    const teaser=document.querySelector("#vxDeadAirTeaser");
    if(teaser && Array.isArray(VX.teasers)){
        const active=VX.teasers.find(x=>t>=x.at && t<x.end);
        if(active){
            const age=t-active.at;
            const remain=active.end-t;
            const fadeIn=ease(age/0.45);
            const fadeOut=ease(remain/0.45);
            const vis=Math.min(fadeIn,fadeOut);
            teaser.style.opacity=String(vis);
            teaser.style.transform=`translateY(${(1-vis)*18}px) scale(${.985+.015*vis})`;

            const textEl=teaser.querySelector(".vxFactText");
            if(textEl){
                const full=String(active.text||"");
                const typeP=clamp(age/0.95);
                const n=Math.floor(full.length*typeP);
                textEl.textContent=full.slice(0,n)+(typeP<1 ? "▌" : "");
            }
        }else{
            teaser.style.opacity="0";
            teaser.style.transform="translateY(18px) scale(.985)";
            const textEl=teaser.querySelector(".vxFactText");
            if(textEl) textEl.textContent="";
        }
    }

    /* ----------------------------------------------------
       OUTRO STAGING
       ---------------------------------------------------- */

    if(VX.key==="05_outro" && VX.outro){

        const summary=document.querySelector("#summaryStage");
        const today=document.querySelector("#todayPanel");
        const cta=document.querySelector("#ctaStage");
        const logo=document.querySelector("#logoStage");
        const bgFade=document.querySelector("#bgFade");
        const logoHolder=document.querySelector("#logoStage .logoHolder");
        const logoImg=document.querySelector("#logoStage .vortex-logo, #logoStage img");

        const summaryOut=ease((t-VX.outro.ctaStart)/0.55);
        const ctaIn=ease((t-VX.outro.ctaStart)/0.55);
        const ctaOut=ease((t-VX.outro.logoStart)/0.42);
        const logoIn=ease((t-VX.outro.logoStart)/0.72);
        const dark=ease((t-VX.outro.tomorrowStart)/0.45);

        if(summary){
            summary.style.opacity=String(1-summaryOut);
            summary.style.transform=`translateY(${summaryOut*16}px)`;
        }
        if(today){
            today.style.filter=`brightness(${1-0.18*dark}) saturate(${1-0.08*dark})`;
        }
        if(cta){
            const ctaOpacity=Math.max(0, ctaIn*(1-ctaOut));
            cta.style.opacity=String(ctaOpacity);
            cta.style.transform=`translateY(${(1-ctaIn)*24 - ctaOut*8}px)`;
        }
        if(bgFade){bgFade.style.opacity=String(0.96*logoIn);}
        if(logo){
            logo.style.opacity=String(logoIn);
            logo.style.transform=`translateY(${(1-logoIn)*16}px)`;
        }
        if(logoHolder){
            const settle=clamp((t-VX.outro.logoStart)/1.25);
            const scale=1.05 - 0.05*settle;
            logoHolder.style.transform=`scale(${scale})`;
        }
        if(logoImg){
            const age=Math.max(0,t-VX.outro.logoStart);
            const shake=age<1.15 ? Math.sin(age*28)*(1-age/1.15)*8 : 0;
            logoImg.style.transform=`translateX(${shake}px)`;
        }
    }

    document.documentElement.dataset.vortexTime=
      t.toFixed(3);

    return t;
};

window.__auditLayout=function(t){

    window.__seek(t);

    return {
        time:t,
        viewport:[innerWidth,innerHeight],
        scroll:[
          document.documentElement.scrollWidth,
          document.documentElement.scrollHeight
        ]
    };
};

if(document.readyState==="loading")
    document.addEventListener("DOMContentLoaded",initVX);
else
    initVX();

</script>
<!-- VX_SHARED_END -->
"""

# ------------------------------------------------------------
# BUILD FIVE ANIMATED SCENES
# ------------------------------------------------------------

manifest={}

for key,(src_name,out_name) in SCENES.items():

    src=HTML/src_name
    out=ANIM/out_name

    if not src.exists():
        raise FileNotFoundError(f"Missing {src}")

    info=narr[key]
    words=info["words"]
    script=info["script"]
    duration=float(info["duration"])

    events=events_for(key,info)

    # Fill genuine dead-air: if meaningful body content is more than ~3s away, show a data teaser.
    if key=="00_intro":
        _body_candidates=[
            e for e in events
            if ".agenda-1" in e["sel"] or ".prevResultsStrip" in e["sel"]
        ]
        _body_event=min(_body_candidates, key=lambda e: float(e["at"])) if _body_candidates else None
    elif key in {"02_fdr","03_projected_goals","04_clean_sheets"}:
        _body_event=next((e for e in events if ".panel" in e["sel"] or ".board" in e["sel"]), None)
    else:
        _body_event=next((e for e in events if "#todayPanel" in e["sel"]), None)

    _body_start=float(_body_event["at"]) if _body_event else 0.0
    teasers=[]

    if _body_start >= 3.2:
        _facts=list(SCENE_DEAD_AIR_FACTS.get(key, []))
        _start=0.65
        _end=max(_start, _body_start-1.0)
        _available=_end-_start

        if _facts and _available >= 2.1:
            _count=2 if len(_facts) >= 2 and _available >= 5.8 else 1
            _slot=_available/_count
            for _i in range(_count):
                _at=_start+(_i*_slot)
                _teaser_end=min(_end, _at+_slot-0.18)
                if _teaser_end-_at >= 1.8:
                    teasers.append({
                        "text":str(_facts[_i]),
                        "at":round(_at,3),
                        "end":round(_teaser_end,3),
                    })

    # Team highlight occurs exactly when Ryan says name
    highlights=[]

    for team in TEAM_NAMES:

        if vx_norm_text(team) not in vx_norm_text(script):
            continue

        t=vx_phrase_time(words,team,-1)

        if t>=0:
            highlights.append({
                "text":team,
                "at":round(t,3)
            })

    vx={
        "key":key,
        "duration":duration,
        "events":events,
        "highlights":highlights,
        "teasers":teasers,
        "logo":OUTRO_LOGO_URI
    }
    # Outro timings are strict Ryan WordBoundary anchors — no guessed fallbacks.
    if key=="05_outro":

        today_1=vx_phrase_time(words,"best fixture runs",required=True,context="05_outro recap")
        today_2=vx_phrase_time(words,"strongest projected output",required=True,context="05_outro recap")
        today_3=vx_phrase_time(words,"highest clean-sheet potential",required=True,context="05_outro recap")
        tomorrow=vx_phrase_time(words,"Tomorrow",required=True,context="05_outro tomorrow panel")
        cta_start=vx_phrase_time(words,"Subscribe and follow",required=True,context="05_outro CTA")
        logo_start=vx_phrase_time(words,"Let us make smarter",required=True,context="05_outro logo finale")

        vx["outro"]={
            "today1":today_1,
            "today2":today_2,
            "today3":today_3,
            "tomorrowStart":tomorrow,
            "ctaStart":cta_start,
            "logoStart":logo_start,
        }

        highlights += [
            {"text":"Best Fixtures","at":round(today_1,3)},
            {"text":"Best Projected Attack","at":round(today_2,3)},
            {"text":"Best Clean-Sheet Odds","at":round(today_3,3)},
            {"text":"Safe Core Picks","at":round(vx_phrase_time(words,"safe core picks",required=True,context="05_outro agenda"),3)},
            {"text":"Rank-Climb Options","at":round(vx_phrase_time(words,"rank-climb options",required=True,context="05_outro agenda"),3)},
            {"text":"Player Traps","at":round(vx_phrase_time(words,"player traps",required=True,context="05_outro agenda"),3)},
            {"text":"Captaincy","at":round(vx_phrase_time(words,"captaincy",required=True,context="05_outro agenda"),3)},
            {"text":"Transfer Plan","at":round(vx_phrase_time(words,"transfer plan",required=True,context="05_outro agenda"),3)},
        ]

    page=src.read_text(
        encoding="utf-8",
        errors="ignore"
    )

    # clean old shared injection
    page=re.sub(
        r"<!-- VX_SHARED_START -->.*?"
        r"<!-- VX_SHARED_END -->",
        "",
        page,
        flags=re.S
    )

    page=page.replace(
        "</head>",
        CSS+"\n</head>"
    )

    page=page.replace(
        "</body>",
        JS.replace(
            "__VX__",
            json.dumps(vx,ensure_ascii=False)
        )+"\n</body>"
    )

    out.write_text(
        page,
        encoding="utf-8"
    )

    manifest[key]={
        "html":str(out),
        "audio":info["audio"],
        "duration":duration,
        "canvas":[3840,2160],
        "highlights":highlights,
        "events":events,
        "teasers":teasers
    }

    print(
        f"✅ {key:<20}"
        f"{duration:>6.2f}s • "
        f"{len(highlights)} spoken highlights • "
        f"{len(teasers)} dead-air facts"
    )

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

OUT=TIMING/"other_scenes_animation.json"

OUT.write_text(
    json.dumps(manifest,indent=2),
    encoding="utf-8"
)

print("\n" + "="*62)
print("✅ CELL 19 COMPLETE")
print("="*62)
print("✅ Intro animated")
print("✅ FDR animated")
print("✅ Projected Goals animated")
print("✅ Clean Sheet Odds animated")
print("✅ Outro animated")
print("✅ Intro + Outro exact Ryan WordBoundary sync")
print("✅ All locked 3840 × 2160")
print("✅ Spoken-team highlighting")
print("✅ Outro footer matches Intro")
print("✅ Subscribe + follow CTA")
print("✅ Original-logo growl finale")
print("✅ Blank/dead-air guard: data teaser inserted automatically when meaningful body content is delayed")
```

```python
# ============================================================
# CELL 20 — SHARED MASTER-CONTROLLED MP4 RENDERER
# Cell 0 controls selected scenes + Draft 480p5 / Final 1080p24
# ============================================================

from pathlib import Path
import json, subprocess, sys, math
from IPython.display import display, Image as IPImage

ROOT     = Path(globals().get("PROJECT_ROOT", Path("output/MyDrive/FPL_VORTEX")))
TIMING   = Path(globals().get("DATA_DIR", Path("output/.work/data")))
SECTIONS = Path(globals().get("MP4_DIR", Path("output/.work/mp4")))
PREVIEWS = Path(globals().get("PREVIEW_WORK_DIR", Path("output/.work/preview")))
SILENT_DIR = Path(globals().get("VIDEO_WORK_DIR", Path("output/.work/video"))) / "silent"

SECTIONS.mkdir(parents=True, exist_ok=True)
PREVIEWS.mkdir(parents=True, exist_ok=True)
SILENT_DIR.mkdir(parents=True, exist_ok=True)

MANIFEST = TIMING/"other_scenes_animation.json"

if not MANIFEST.exists():
    raise FileNotFoundError("Run Cell 19 first.")

SCENES = json.loads(
    MANIFEST.read_text(encoding="utf-8")
)

ORDER = [
    key for key in [
        "00_intro",
        "02_fdr",
        "03_projected_goals",
        "04_clean_sheets",
        "05_outro",
    ]
    if globals().get("SLIDE_SELECTION", {}).get(key, True)
]

# ------------------------------------------------------------
# DESIGN vs VIDEO
# ------------------------------------------------------------

DESIGN_W, DESIGN_H = 3840, 2160

W = int(globals().get("VIDEO_WIDTH", 3840))
H = int(globals().get("VIDEO_HEIGHT", 2160))
FPS = int(globals().get("VIDEO_FPS", 30))
SCALE_X = W / DESIGN_W
SCALE_Y = H / DESIGN_H
SCALE = (SCALE_X, SCALE_Y)


# ------------------------------------------------------------
# HELPERS
# ------------------------------------------------------------





# ------------------------------------------------------------
# SHARED PLAYWRIGHT RUNTIME FROM CELL 14
# ------------------------------------------------------------
if "_vx_ensure_playwright_chromium" not in globals() or "async_playwright" not in globals():
    raise RuntimeError("Run Cell 14 first — shared Playwright renderer is missing.")
_vx_ensure_playwright_chromium()

# ------------------------------------------------------------
# EXACT 50% FULL-PAGE SCALE
# ------------------------------------------------------------



# ------------------------------------------------------------
# ONE REUSABLE RENDERER
# ------------------------------------------------------------

async def render_scene(page, key, cfg):

    html  = Path(cfg["html"])
    audio = Path(cfg["audio"])

    if not html.exists():
        raise FileNotFoundError(html)

    if not audio.exists():
        raise FileNotFoundError(audio)

    duration = vx_media_duration(audio)

    expected = float(cfg["duration"])

    if abs(duration-expected) > .30:
        raise RuntimeError(
            f"{key}: audio/timeline mismatch "
            f"{duration:.3f}s vs {expected:.3f}s"
        )

    if "vx_frame_plan" not in globals():
        raise RuntimeError("Run Cell 9 first — shared vx_frame_plan is missing.")
    frame_plan = vx_frame_plan(duration, FPS)
    frames = int(frame_plan["frames"])
    frame_grid_duration = float(frame_plan["grid_duration"])

    silent = SILENT_DIR/f"{key}_silent.mp4"
    final  = SECTIONS/f"{key}.mp4"
    preview = PREVIEWS/f"{key}_{H}p.png"

    print("\n" + "="*66)
    print(f"🎬 {key}")
    print(f"🎨 Design : {DESIGN_W}×{DESIGN_H}")
    print(f"📺 Output : {W}×{H}")
    print(f"🎙 Audio  : {duration:.2f}s")
    print(f"🎞 Frames : {frames:,}")
    print(f"⏱ Frame grid: {frame_grid_duration:.3f}s")
    print("="*66)

    # --------------------------------------------------------
    # LOAD
    # --------------------------------------------------------

    await page.goto(
        html.resolve().as_uri(),
        wait_until="load",
        timeout=120000
    )

    await page.wait_for_function(
        "window.__ready === true",
        timeout=60000
    )

    await vx_apply_render_scale(page, W, H, DESIGN_W, DESIGN_H)

    try:
        await page.evaluate(
            "() => document.fonts.ready"
        )
    except:
        pass

    try:
        await page.wait_for_function(
            """
            () => Array.from(document.images)
            .every(img => img.complete)
            """,
            timeout=15000
        )
    except:
        print("⚠️ Continuing after image wait timeout.")

    # --------------------------------------------------------
    # FULL-BLEED AUDIT
    # --------------------------------------------------------

    audit = await page.evaluate("""
    () => {
        const b = document.body.getBoundingClientRect();

        return {
            viewport:[innerWidth,innerHeight],
            body:[
                Math.round(b.left),
                Math.round(b.top),
                Math.round(b.width),
                Math.round(b.height)
            ]
        };
    }
    """)

    if audit["viewport"] != [W,H]:
        raise RuntimeError(
            f"{key}: viewport failed {audit['viewport']}"
        )

    if audit["body"] != [0,0,W,H]:
        raise RuntimeError(
            f"{key}: scaled body failed {audit['body']}"
        )

    # check several animation states
    for t in [
        .05,
        duration*.25,
        duration*.50,
        duration*.75,
        max(.05,duration-.05)
    ]:

        await page.evaluate(
            "(t)=>window.__seek(t)",
            float(t)
        )

    print("✅ Full-screen 16:9")
    print("✅ No crop")
    print("✅ No outside black bars")

    # --------------------------------------------------------
    # PREVIEW
    # --------------------------------------------------------

    preview_t=max(
        .05,
        duration-.20
    )

    await page.evaluate(
        "(t)=>window.__seek(t)",
        preview_t
    )

    await page.screenshot(
        path=str(preview),
        type="png"
    )

    # --------------------------------------------------------
    # FRAME STREAM → H264
    # --------------------------------------------------------

    ff = subprocess.Popen(
        [
            "ffmpeg",
            "-y",
            "-v","error",

            "-f","image2pipe",
            "-framerate",str(FPS),
            "-vcodec","png",
            "-i","-",

            "-an",

            "-c:v","libx264",
            "-preset","veryfast",
            "-crf","18",

            "-pix_fmt","yuv420p",

            "-movflags","+faststart",

            str(silent)
        ],
        stdin=subprocess.PIPE,
        stderr=subprocess.PIPE
    )

    report_every=max(
        1,
        frames//20
    )

    try:

        for i in range(frames):

            t=min(
                i/FPS,
                max(
                    0,
                    duration-(1/FPS)
                )
            )

            await page.evaluate(
                "(t)=>window.__seek(t)",
                float(t)
            )

            frame=await page.screenshot(
                type="png"
            )

            ff.stdin.write(frame)

            if (
                i % report_every == 0
                or i == frames-1
            ):

                pct=(i+1)/frames*100

                print(
                    f"\rRendering "
                    f"{pct:5.1f}% • "
                    f"{i+1:,}/{frames:,}",
                    end=""
                )

        ff.stdin.close()

        rc=ff.wait()

        if rc != 0:

            err=ff.stderr.read().decode(
                "utf-8",
                errors="ignore"
            )

            raise RuntimeError(
                f"{key} render failed:\n"
                + err[-3000:]
            )

    finally:

        if ff.stdin and not ff.stdin.closed:
            ff.stdin.close()

    print("\n✅ Video frames encoded")

    # --------------------------------------------------------
    # ADD RYAN
    # --------------------------------------------------------

    subprocess.run(
        [
            "ffmpeg",
            "-y",
            "-v","error",

            "-i",str(silent),
            "-i",str(audio),

            "-map","0:v:0",
            "-map","1:a:0",

            "-c:v","copy",

            "-c:a","aac",
            "-b:a","192k",

            "-shortest",

            "-movflags","+faststart",

            str(final)
        ],
        check=True
    )

    # --------------------------------------------------------
    # QA
    # --------------------------------------------------------

    vi = vx_video_info(final)
    vd = vx_media_duration(final)

    if (
        int(vi["width"]) != W
        or int(vi["height"]) != H
    ):
        raise RuntimeError(
            f"{key}: resolution failed "
            f"{vi['width']}×{vi['height']}"
        )

    grid_delta = abs(vd - min(frame_grid_duration, duration))
    mux_tolerance = max(0.03, 0.10 / max(FPS, 1))
    audio_delta = abs(vd - duration)
    audio_tolerance = float(frame_plan["max_quantization_error"]) + mux_tolerance

    if grid_delta > mux_tolerance:
        raise RuntimeError(
            f"{key}: encoded frame-grid mismatch "
            f"{vd:.3f}s vs {frame_grid_duration:.3f}s"
        )
    if audio_delta > audio_tolerance:
        raise RuntimeError(
            f"{key}: duration mismatch "
            f"{vd:.3f}s vs {duration:.3f}s "
            f"(allowed {audio_tolerance:.3f}s at {FPS}fps)"
        )

    print(
        f"✅ PASSED • {W}×{H} • "
        f"{FPS}fps • {vd:.2f}s"
    )

    return {
        "key":key,
        "final":str(final),
        "silent":str(silent),
        "preview":str(preview),
        "duration":vd,
        "width":W,
        "height":H,
        "fps":FPS
    }


# ------------------------------------------------------------
# RENDER ALL FIVE — ONE BROWSER
# ------------------------------------------------------------

async def render_all_remaining():

    results={}

    if not ORDER:
        print("⏭️ No shared scenes selected in Cell 0.")
        return results

    async with async_playwright() as pw:

        try:

            browser=await pw.chromium.launch(
                headless=True,
                args=[
                    "--no-sandbox",
                    "--disable-dev-shm-usage",
                    "--disable-gpu"
                ]
            )

        except Exception:

            print("⏳ Installing Chromium once...")

            subprocess.run(
                [
                    sys.executable,
                    "-m","playwright",
                    "install","chromium"
                ],
                check=True
            )

            browser=await pw.chromium.launch(
                headless=True,
                args=[
                    "--no-sandbox",
                    "--disable-dev-shm-usage",
                    "--disable-gpu"
                ]
            )

        context=await browser.new_context(
            viewport={
                "width":W,
                "height":H
            },
            device_scale_factor=1
        )

        page=await context.new_page()

        for key in ORDER:

            if key not in SCENES:
                raise KeyError(
                    f"Cell 19 manifest missing {key}"
                )

            results[key]=await render_scene(
                page,
                key,
                SCENES[key]
            )

        await context.close()
        await browser.close()

    return results


RENDERED = await render_all_remaining()


# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

OUT=TIMING/"other_scenes_render.json"

OUT.write_text(
    json.dumps(
        RENDERED,
        indent=2
    ),
    encoding="utf-8"
)


# ------------------------------------------------------------
# REPORT
# ------------------------------------------------------------

print("\n")
print("="*70)
print("✅ CELL 20 — ALL FIVE SCENES PASSED")
print("="*70)

for key in ORDER:

    r=RENDERED[key]

    print(
        f"✅ {key:<22}"
        f"{r['duration']:>7.2f}s • "
        f"{W}×{H}"
    )

print("\n✅ 4K design retained")
print(f"✅ {W}×{H} @ {FPS}fps MP4 output")
print(f"✅ Dynamic 4K→video scale: {SCALE_X:.6f}×{SCALE_Y:.6f}")
print("✅ Full-width / full-height 16:9")
print("✅ No crop")
print("✅ No external black bars")
print("✅ Ryan audio attached")

if "05_outro" in RENDERED:
    print("\n👇 OUTRO PREVIEW")
    display(
        IPImage(
            filename=RENDERED["05_outro"]["preview"],
            width=1200
        )
    )
```

```python
# ============================================================
# CELL 21 — FINAL MASTER-CONTROLLED MP4 ASSEMBLY + PUBLISH QA
# ============================================================

from pathlib import Path
import subprocess, json
from IPython.display import Video, display

ROOT  = Path(globals().get("PROJECT_ROOT", Path("output/MyDrive/FPL_VORTEX")))
VIDEO = Path(globals().get("MP4_DIR", Path("output/.work/mp4")))
SECT  = VIDEO
TIME  = Path(globals().get("DATA_DIR", Path("output/.work/data")))
MP3_OUT = Path(globals().get("MP3_DIR", Path("output/.work/mp3")))
WORK = Path(globals().get("WORK_ROOT", Path("output/.work")))

TIME.mkdir(parents=True, exist_ok=True)
VIDEO.mkdir(parents=True, exist_ok=True)
MP3_OUT.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# FINAL LOCKED VIDEO ORDER
# ------------------------------------------------------------

ALL_ORDER = [
    ("00_intro",            "INTRO",             SECT/"00_intro.mp4"),
    ("gw_review",           "GW REVIEW",         SECT/"gw_review.mp4"),
    ("02_fdr",              "FDR",               SECT/"02_fdr.mp4"),
    ("03_projected_goals",  "PROJECTED GOALS",   SECT/"03_projected_goals.mp4"),
    ("04_clean_sheets",     "CLEAN SHEET ODDS",  SECT/"04_clean_sheets.mp4"),
    ("05_outro",            "OUTRO",             SECT/"05_outro.mp4"),
]

_selection = globals().get("SLIDE_SELECTION", {})
ORDER = [
    (label, path)
    for key, label, path in ALL_ORDER
    if _selection.get(key, True)
]

if not ORDER:
    raise ValueError("Cell 0 has no slides selected for MP4 assembly.")

NEXT_GW_LOCAL = globals().get("NEXT_GW")

MP4_QUALITY_LOCAL = str(globals().get("MP4_QUALITY", "FINAL")).upper()
W = int(globals().get("VIDEO_WIDTH", 3840))
H = int(globals().get("VIDEO_HEIGHT", 2160))
FPS = int(globals().get("VIDEO_FPS", 30))

if NEXT_GW_LOCAL is not None:
    FINAL_MP4 = VIDEO/f"FPL_VORTEX_GW{int(NEXT_GW_LOCAL)}_COMBINED_{MP4_QUALITY_LOCAL}_{H}P_{FPS}FPS.mp4"
else:
    FINAL_MP4 = VIDEO/f"FPL_VORTEX_WEEKLY_COMBINED_{MP4_QUALITY_LOCAL}_{H}P_{FPS}FPS.mp4"

CONCAT_FILE = WORK/"final_scene_concat.txt"
QA_FILE     = TIME/"final_video_qa.json"



# ------------------------------------------------------------
# FFPROBE
# ------------------------------------------------------------

def probe(path):

    r = subprocess.run(
        [
            "ffprobe",
            "-v","quiet",
            "-print_format","json",
            "-show_format",
            "-show_streams",
            str(path)
        ],
        capture_output=True,
        text=True,
        check=True
    )

    return json.loads(r.stdout)


def fps_float(value):

    if not value or value == "0/0":
        return 0.0

    n,d = value.split("/")

    return float(n) / float(d)


# ------------------------------------------------------------
# CHECK CURRENT TRUSTED SECTIONS FIRST
# ------------------------------------------------------------

reports=[]
signatures=[]
expected_duration=0.0

print("="*72)
print(f"🔍 FINAL SELECTED-SCENE QA — {len(ORDER)} scene(s)")
print("="*72)

for number,(label,path) in enumerate(ORDER,1):

    if not path.exists():
        raise FileNotFoundError(
            f"{label} missing:\n{path}"
        )

    info=probe(path)

    streams=info.get("streams",[])

    video_stream=next(
        (
            s for s in streams
            if s.get("codec_type")=="video"
        ),
        None
    )

    audio_stream=next(
        (
            s for s in streams
            if s.get("codec_type")=="audio"
        ),
        None
    )

    if video_stream is None:
        raise RuntimeError(
            f"{label}: video stream missing"
        )

    if audio_stream is None:
        raise RuntimeError(
            f"{label}: audio stream missing"
        )

    width=int(video_stream.get("width",0))
    height=int(video_stream.get("height",0))

    fps=fps_float(
        video_stream.get("avg_frame_rate")
        or video_stream.get("r_frame_rate")
    )

    duration=float(
        info["format"]["duration"]
    )

    if (width,height)!=(W,H):

        raise RuntimeError(
            f"{label}: {width}×{height} "
            f"instead of {W}×{H}"
        )

    if abs(fps-FPS) > .05:

        raise RuntimeError(
            f"{label}: {fps:.3f}fps "
            f"instead of {FPS}fps"
        )

    expected_duration += duration

    signature=(
        video_stream.get("codec_name"),
        video_stream.get("pix_fmt"),
        video_stream.get("width"),
        video_stream.get("height"),
        video_stream.get("avg_frame_rate"),
        video_stream.get("time_base"),

        audio_stream.get("codec_name"),
        audio_stream.get("sample_rate"),
        audio_stream.get("channels"),
        audio_stream.get("channel_layout"),
    )

    signatures.append(signature)

    reports.append({
        "order":number,
        "scene":label,
        "file":str(path),
        "duration":duration,
        "resolution":[width,height],
        "fps":fps,
        "video_codec":video_stream.get("codec_name"),
        "audio_codec":audio_stream.get("codec_name"),
    })

    print(
        f"✅ {number}. "
        f"{label:<18} "
        f"{duration:>7.2f}s • "
        f"{width}×{height} • "
        f"{fps:.0f}fps"
    )


# ------------------------------------------------------------
# CONCAT MANIFEST
# ------------------------------------------------------------

CONCAT_FILE.write_text(
    "ffconcat version 1.0\n"
    +
    "\n".join(
        f"file '{path.resolve().as_posix()}'"
        for _,path in ORDER
    )
    +
    "\n",
    encoding="utf-8"
)


# ------------------------------------------------------------
# FAST PATH:
# STREAM COPY WHEN ALL SECTION STREAMS MATCH
# ------------------------------------------------------------

streams_match = len(set(signatures)) == 1

if streams_match:

    print("\n✅ All stream settings match.")
    print("⚡ Joining without another video encode...")

    subprocess.run(
        [
            "ffmpeg",
            "-y",
            "-v","error",

            "-f","concat",
            "-safe","0",
            "-i",str(CONCAT_FILE),

            "-map","0:v:0",
            "-map","0:a:0",

            "-c","copy",

            "-movflags","+faststart",

            str(FINAL_MP4)
        ],
        check=True
    )

    ASSEMBLY_MODE="STREAM COPY"


# ------------------------------------------------------------
# SAFE FALLBACK:
# NORMALIZE ONLY IF STREAMS DON'T MATCH
# ------------------------------------------------------------

else:

    print("\n⚠️ Stream settings differ.")
    print(f"🔧 Performing one controlled {W}×{H} final encode...")

    inputs=[]

    for _,path in ORDER:
        inputs += ["-i",str(path)]

    filters=[]

    for i in range(len(ORDER)):

        filters.append(
            f"[{i}:v:0]"
            f"scale={W}:{H},"
            f"fps={FPS},"
            f"format=yuv420p,"
            f"setpts=PTS-STARTPTS"
            f"[v{i}]"
        )

        filters.append(
            f"[{i}:a:0]"
            f"aresample=48000,"
            f"asetpts=PTS-STARTPTS"
            f"[a{i}]"
        )

    joined="".join(
        f"[v{i}][a{i}]"
        for i in range(len(ORDER))
    )

    filters.append(
        joined +
        f"concat=n={len(ORDER)}:v=1:a=1[v][a]"
    )

    subprocess.run(
        [
            "ffmpeg",
            "-y",
            "-v","error",

            *inputs,

            "-filter_complex",
            ";".join(filters),

            "-map","[v]",
            "-map","[a]",

            "-c:v","libx264",
            "-preset","fast",
            "-crf","18",
            "-pix_fmt","yuv420p",

            "-c:a","aac",
            "-b:a","192k",
            "-ar","48000",

            "-movflags","+faststart",

            str(FINAL_MP4)
        ],
        check=True
    )

    ASSEMBLY_MODE="NORMALIZED FINAL ENCODE"


# ------------------------------------------------------------
# FINAL MASTER QA
# ------------------------------------------------------------

master=probe(FINAL_MP4)

streams=master["streams"]

v=next(
    s for s in streams
    if s.get("codec_type")=="video"
)

a=next(
    s for s in streams
    if s.get("codec_type")=="audio"
)

final_w=int(v.get("width",0))
final_h=int(v.get("height",0))

final_fps=fps_float(
    v.get("avg_frame_rate")
    or v.get("r_frame_rate")
)

final_duration=float(
    master["format"]["duration"]
)

duration_delta=abs(
    final_duration-expected_duration
)


# ------------------------------------------------------------
# HARD PUBLISH GATES
# ------------------------------------------------------------

if (final_w,final_h)!=(W,H):

    raise RuntimeError(
        f"FINAL resolution failed: "
        f"{final_w}×{final_h}"
    )

if abs(final_fps-FPS)>.05:

    raise RuntimeError(
        f"FINAL fps failed: "
        f"{final_fps:.3f}"
    )

if v.get("codec_name")!="h264":

    raise RuntimeError(
        f"FINAL video codec is "
        f"{v.get('codec_name')}"
    )

if a.get("codec_name")!="aac":

    raise RuntimeError(
        f"FINAL audio codec is "
        f"{a.get('codec_name')}"
    )

# Tiny concat timestamp differences are okay.
if duration_delta > .75:

    raise RuntimeError(
        "FINAL duration gate failed:\n"
        f"Selected scenes = {expected_duration:.3f}s\n"
        f"Master     = {final_duration:.3f}s\n"
        f"Difference = {duration_delta:.3f}s"
    )


# ------------------------------------------------------------
# PHASE 4 — OUTPUT CONTRACT / NO-TEMP-IN-DRIVE QA
# ------------------------------------------------------------

OUTPUT_ROOT_LOCAL = Path(globals().get("OUTPUT_ROOT", VIDEO.parent))
SLIDE_OUT = Path(globals().get("SLIDE_DIR", OUTPUT_ROOT_LOCAL / "SLIDE"))
_ALLOWED_OUTPUT_DIRS = {
    "MP4": VIDEO,
    "MP3": MP3_OUT,
    "SLIDE": SLIDE_OUT,
    "DATA": TIME,
}

_output_contract_violations = []

# P0 is the single folder owner: no extra top-level output directories/files.
if OUTPUT_ROOT_LOCAL.exists():
    for child in OUTPUT_ROOT_LOCAL.iterdir():
        if child.is_dir() and child.name not in _ALLOWED_OUTPUT_DIRS:
            _output_contract_violations.append(f"unexpected_output_directory:{child.name}")
        elif child.is_file():
            _output_contract_violations.append(f"unexpected_root_file:{child.name}")

# DATA is durable metadata/timing only; temporary production formats stay in WORK.
if TIME.exists():
    for p in TIME.rglob("*"):
        if p.is_file() and p.suffix.lower() not in {".json", ".vtt"}:
            _output_contract_violations.append(f"invalid_data_artifact:{p.name}")

for p in VIDEO.rglob("*") if VIDEO.exists() else []:
    if p.is_file() and p.suffix.lower() != ".mp4":
        _output_contract_violations.append(f"invalid_mp4_artifact:{p.name}")
    if p.is_file() and ("silent" in p.name.lower() or "preview" in p.name.lower()):
        _output_contract_violations.append(f"temporary_video_in_drive:{p.name}")

for p in MP3_OUT.rglob("*") if MP3_OUT.exists() else []:
    if p.is_file() and p.suffix.lower() != ".mp3":
        _output_contract_violations.append(f"invalid_mp3_artifact:{p.name}")
    if p.is_file() and ("segment" in p.name.lower() or "temp" in p.name.lower()):
        _output_contract_violations.append(f"temporary_audio_in_drive:{p.name}")

for p in SLIDE_OUT.rglob("*") if SLIDE_OUT.exists() else []:
    if p.is_file() and p.suffix.lower() not in {".png", ".jpg", ".jpeg", ".webp"}:
        _output_contract_violations.append(f"invalid_slide_artifact:{p.name}")

_required_phase0_data = ["phase0_audit.json"]
_missing_phase0_data = [name for name in _required_phase0_data if not (TIME / name).exists()]
if _missing_phase0_data:
    _output_contract_violations.append(
        "missing_required_phase0_data:" + ",".join(_missing_phase0_data)
    )

# Remove transient GW-review renderer HTML before the Phase 0 publish gate.
# These files are temporary inputs; the rendered MP4/PNG outputs are already complete.
_vx_review_html_root = Path(globals().get("HTML_WORK_DIR", Path("output/.work/html")))
for _vx_review_html in _vx_review_html_root.glob("gw_review_*.html"):
    _vx_review_html.unlink(missing_ok=True)

# Final publish is blocked if stale model state reappears anywhere.
if "vx_phase0_assert_cleanroom" not in globals():
    _output_contract_violations.append("phase0_cleanroom_guard_missing")
else:
    try:
        vx_phase0_assert_cleanroom("pre-publish")
    except Exception as _phase0_exc:
        _output_contract_violations.append("phase0_cleanroom_fail:" + str(_phase0_exc))

if _output_contract_violations:
    raise RuntimeError(
        "PHASE 0 OUTPUT CONTRACT FAILED:\n- "
        + "\n- ".join(_output_contract_violations)
    )

# ------------------------------------------------------------
# COMBINED MP3 — audio exactly matching the final selected video
# ------------------------------------------------------------

if NEXT_GW_LOCAL is not None:
    COMBINED_MP3 = MP3_OUT / f"FPL_VORTEX_GW{int(NEXT_GW_LOCAL)}_COMBINED.mp3"
else:
    COMBINED_MP3 = MP3_OUT / "FPL_VORTEX_WEEKLY_COMBINED.mp3"

subprocess.run(
    [
        "ffmpeg", "-y", "-v", "error",
        "-i", str(FINAL_MP4),
        "-vn",
        "-c:a", "libmp3lame",
        "-b:a", "192k",
        str(COMBINED_MP3),
    ],
    check=True,
)

print("✅ Combined MP3:", COMBINED_MP3)


# ------------------------------------------------------------
# SAVE QA REPORT
# ------------------------------------------------------------

qa={
    "passed":True,

    "scene_order":[
        label for label,_ in ORDER
    ],

    "scene_count":len(ORDER),

    "design_canvas":[3840,2160],

    "video_output":[
        final_w,
        final_h
    ],

    "aspect_ratio":"16:9",

    "fps":final_fps,

    "video_codec":
        v.get("codec_name"),

    "audio_codec":
        a.get("codec_name"),

    "assembly_mode":
        ASSEMBLY_MODE,

    "expected_duration":
        expected_duration,

    "final_duration":
        final_duration,

    "duration_delta":
        duration_delta,

    "sections":
        reports,

    "final_file":
        str(FINAL_MP4),

    "combined_mp3":
        str(COMBINED_MP3),

    "output_contract":{
        "passed": True,
        "output_root": str(OUTPUT_ROOT_LOCAL),
        "allowed_directories": list(_ALLOWED_OUTPUT_DIRS.keys()),
        "required_phase0_data": _required_phase0_data,
        "violations": _output_contract_violations,
    },
}

QA_FILE.write_text(
    json.dumps(
        qa,
        indent=2
    ),
    encoding="utf-8"
)


# ------------------------------------------------------------
# FINISHED
# ------------------------------------------------------------

minutes=int(final_duration//60)
seconds=final_duration-minutes*60

print("\n")
print("="*72)
print("🏆 FPL VORTEX — FINAL VIDEO PASSED")
print("="*72)

print("\nFINAL ORDER")
for i,(label,_) in enumerate(ORDER,1):
    print(f"{i}. {label}")

print("\nMASTER")
print("✅ Design canvas : 3840×2160")
print(f"✅ Video output  : {W}×{H}")
print("✅ Aspect ratio  : 16:9 FULL SCREEN")
print(f"✅ Frame rate    : {FPS} fps")
print("✅ Video codec   : H.264")
print("✅ Audio codec   : AAC")
print(f"✅ Duration      : {minutes}:{seconds:05.2f}")
print(f"✅ Assembly      : {ASSEMBLY_MODE}")
print(f"✅ Mode          : {MP4_QUALITY_LOCAL}")
print("✅ QA:",QA_FILE)
print("✅ Phase-4 output contract: no temp HTML/PY/TXT/silent/segment artifacts in Drive output")

print("\n🎬 FINAL PUBLISHABLE VIDEO")
print(FINAL_MP4)
print("🎧 FINAL COMBINED MP3")
print(COMBINED_MP3)

display(
    Video(
        str(FINAL_MP4),
        embed=False,
        width=1200
    )
)
```
